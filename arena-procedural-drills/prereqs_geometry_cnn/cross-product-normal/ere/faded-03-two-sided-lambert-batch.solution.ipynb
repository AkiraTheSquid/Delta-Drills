## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Two-sided Lambertian shading lights a face regardless of which way its normal points: brightness = `|normal · light_dir|` rather than `max(0, ·)`. Over a batch of `(N, 3)` unit normals and one unit light direction, the dot products come from a single matrix-vector product, and the absolute value handles back-facing triangles.

## Faded exercise 3

Implement `two_sided_shade(normals, light)` where `normals` is `(N, 3)` unit normals and `light` is a `(3,)` light vector (not necessarily unit). Return an `(N,)` tensor of two-sided diffuse brightness in `[0, 1]`. Complete the line that computes the per-face brightness from the dot products.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def two_sided_shade(normals: Tensor, light: Tensor) -> Tensor:
    light_dir = light / light.norm()                # (3,) unit
    dots = normals @ light_dir                      # (N,)
    brightness = dots.abs()                         # (N,)
    return brightness


def _test():
    normals = t.tensor([
        [0.0, 0.0, 1.0],    # facing light
        [0.0, 0.0, -1.0],   # facing away -> two-sided still bright
        [1.0, 0.0, 0.0],    # perpendicular -> dark
    ])
    light = t.tensor([0.0, 0.0, 4.0])
    out = two_sided_shade(normals, light)
    light_dir = light / light.norm()
    expected = (normals @ light_dir).abs()
    assert out.shape == (3,)
    assert t.allclose(out, expected, atol=1e-6)
    assert t.allclose(out, t.tensor([1.0, 1.0, 0.0]), atol=1e-6)
    assert (out >= 0).all()


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
