The load-bearing first step of manual padding is allocating a zero-filled destination sized to fit the padded input, using `new_zeros` so dtype and device inherit from the source. For a symmetric pad of `p` on every side of a `(B, IC, H, W)` input, that size is `(B, IC, H + 2*p, W + 2*p)`.

## Faded exercise 2

Implement `pad2d_sym(x, p)` for `x` of shape `(B, IC, H, W)`, returning shape `(B, IC, H + 2*p, W + 2*p)` with a `p`-thick zero frame and `x` in the center. The interior assignment is given — complete the allocation of the zero buffer (correct size, dtype, and device inherited from `x`).

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def pad2d_sym(x: Tensor, p: int) -> Tensor:
    B, IC, H, W = x.shape
    out = x.new_zeros(B, IC, H + 2 * p, W + 2 * p)
    out[..., p : p + H, p : p + W] = x
    return out


def _test():
    t.manual_seed(0)
    for p in [0, 1, 3]:
        x = t.randn(2, 4, 6, 5)
        y = pad2d_sym(x, p)
        H, W = x.shape[2], x.shape[3]
        assert tuple(y.shape) == (2, 4, H + 2 * p, W + 2 * p)
        assert y.dtype == x.dtype and y.device == x.device
        ref = t.nn.functional.pad(x, (p, p, p, p))
        assert t.allclose(y, ref)
        assert t.allclose(y[..., p:p + H, p:p + W], x)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
