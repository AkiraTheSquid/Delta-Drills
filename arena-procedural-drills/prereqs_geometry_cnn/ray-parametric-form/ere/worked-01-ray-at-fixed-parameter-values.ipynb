## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

A ray is defined by an origin point `O` and a direction vector `D`. Any point on the ray is given by `R(u) = O + u * D` for scalar parameter `u`. When `u = 0`, you get the origin itself; positive `u` moves forward along the direction; the direction does not need to be normalized. This parametric formula is the fundamental building block of all ray-tracing intersection tests.

## Worked solution

**Step 1 — Unpack origin and direction.** A ray stored as shape `(2, 3)` has `O = ray[0]` and `D = ray[1]`, each of shape `(3,)`.

**Step 2 — Choose three parameter values.** We pick `u = 0.0` (the origin), `u = 1.0` (one step forward), and `u = 2.5` (two-and-a-half steps forward).

**Step 3 — Compute each point.** `R(u) = O + u * D`. For each scalar `u`, this is a simple multiply-and-add on 3D vectors.

**Step 4 — Verify.** At `u=0`, the result must equal `O`. At `u=1`, it equals `O + D`. The points should be collinear and evenly spaced along the direction.

```
import torch as t

t.manual_seed(61)

# Define a ray: origin at (1, 2, 0), direction (0, 1, 0) — pointing up Y axis
O = t.tensor([1.0, 2.0, 0.0])
D = t.tensor([0.0, 1.0, 0.0])
ray = t.stack([O, D])   # shape (2, 3)

def eval_ray_at(ray: t.Tensor, u: float) -> t.Tensor:
    """Evaluate R(u) = O + u * D."""
    O_, D_ = ray[0], ray[1]
    return O_ + u * D_

u_values = [0.0, 1.0, 2.5]
for u in u_values:
    point = eval_ray_at(ray, u)
    print(f'u={u}: R(u) = {point.tolist()}')

# u=0 should be exactly the origin
print(f'u=0 equals O: {t.equal(eval_ray_at(ray, 0.0), O)}')
# u=1 should be O + D
print(f'u=1 equals O+D: {t.equal(eval_ray_at(ray, 1.0), O + D)}')
```
