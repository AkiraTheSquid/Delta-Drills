## Concept

For a batch of rays stored as `(B, 2, 3)`, origins are `rays[:, 0]` (shape `(B, 3)`) and directions are `rays[:, 1]` (shape `(B, 3)`). Slicing correctly is the first step before any evaluation. After slicing, the point batch `O + u * D` follows by simple scalar-tensor broadcasting.

## Faded exercise 2

### Exercise — Slice origins and directions from batch rays

Complete `eval_batch_rays(rays, u)`. Given `rays` of shape `(B, 2, 3)` and scalar `u`, return the `(B, 3)` endpoints.

Fill in the slicing and the parametric evaluation.

**Fill in:** Slice rays[:, 0] for origins and rays[:, 1] for directions, then return O + u * D.

````
import torch as t

def eval_batch_rays(rays: t.Tensor, u: float) -> t.Tensor:
    O = rays[:, 0]
    D = rays[:, 1]
    return O + u * D

t.manual_seed(71)
B = 6
origins = t.randn(B, 3)
dirs = t.randn(B, 3)
rays = t.stack([origins, dirs], dim=1)
pts = eval_batch_rays(rays, 2.0)
print(pts.shape)


def _test():
    import torch as t
    t.manual_seed(71)
    B = 6
    origins = t.randn(B, 3)
    dirs = t.randn(B, 3)
    rays = t.stack([origins, dirs], dim=1)
    pts = eval_batch_rays(rays, 2.0)
    assert pts.shape == (B, 3)
    # u=0 -> origins
    pts0 = eval_batch_rays(rays, 0.0)
    assert t.equal(pts0, origins)
    # u=1 -> origins + dirs
    pts1 = eval_batch_rays(rays, 1.0)
    assert t.allclose(pts1, origins + dirs)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
