## Concept

When evaluating a single ray `(2, 3)` at a 1D array of `M` parameter values `us` of shape `(M,)`, you need to reshape `us` to `(M, 1)` before multiplying by `D` (shape `(3,)`). This broadcast produces shape `(M, 3)` — one point per parameter value. The `unsqueeze(-1)` call adds the trailing dimension needed for the broadcast.

## Faded exercise 1

### Exercise — Evaluate ray at M parameter values

Complete `eval_ray(ray, us)`. Given `ray` of shape `(2, 3)` and `us` of shape `(M,)`, return the `(M, 3)` batch of points `O + us[m] * D` for each m.

Fill in the return statement that evaluates R(u) for all values in us.

**Fill in:** Compute O + unsqueeze(us) * D to broadcast (M,1) * (3,) -> (M, 3), then add origin.

````
import torch as t

def eval_ray(ray: t.Tensor, us: t.Tensor) -> t.Tensor:
    O, D = ray[0], ray[1]
    return O + us.unsqueeze(-1) * D

t.manual_seed(70)
ray = t.stack([t.tensor([1.0, 0.0, 0.0]), t.tensor([0.0, 1.0, 0.0])])
us = t.linspace(0, 3, 4)
points = eval_ray(ray, us)
print(points.shape, points[0].tolist())


def _test():
    import torch as t
    t.manual_seed(70)
    O = t.tensor([1.0, 2.0, 3.0])
    D = t.tensor([0.0, 0.0, 1.0])
    ray = t.stack([O, D])
    us = t.tensor([0.0, 1.0, 2.0, 5.0])
    pts = eval_ray(ray, us)
    assert pts.shape == (4, 3), f'expected (4,3), got {pts.shape}'
    # u=0 -> origin
    assert t.allclose(pts[0], O)
    # u=1 -> O + D
    assert t.allclose(pts[1], O + D)
    # u=5 -> O + 5*D
    assert t.allclose(pts[3], O + 5.0 * D)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
