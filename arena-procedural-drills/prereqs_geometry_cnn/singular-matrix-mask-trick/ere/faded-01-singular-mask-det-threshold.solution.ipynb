## Concept

The first step of the singular mask trick is computing the determinant of each slice with `torch.linalg.det`, then building a boolean mask where the absolute determinant is below a small threshold eps. This threshold is necessary because floating-point arithmetic can produce near-zero but non-zero determinants for what should be exactly singular matrices.
