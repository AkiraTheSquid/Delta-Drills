## Concept

Dilation `D` expands a `K`-tap kernel to an effective span `K_eff = D*(K-1) + 1`, and the output length is `(L + 2P - K_eff)//S + 1`. With `D=1`, `K_eff = K` and the formula reduces to the standard one.
