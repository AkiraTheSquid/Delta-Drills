## Concept

`torch.linalg.solve` raises `RuntimeError` on a singular matrix. The graceful unbatched pattern is a `try/except RuntimeError` that returns `None` from the except branch, so the caller gets a clean sentinel instead of a crash.

## Faded exercise 1

### Return None when the solve fails

Implement `safe_solve(A, b)`: return `t.linalg.solve(A, b)` on success, and `None` if it raises because `A` is singular. The `try` is wired for you; complete the `except` clause so it catches the right exception type and returns `None`.

**Fill in:** the except RuntimeError clause that returns None on a singular matrix

````
import torch as t
from typing import Optional

def safe_solve(A: t.Tensor, b: t.Tensor) -> Optional[t.Tensor]:
    try:
        return t.linalg.solve(A, b)
    except ZeroDivisionError:
        raise NotImplementedError()  # TODO: the except RuntimeError clause that returns None on a singular matrix

print(safe_solve(t.tensor([[1.0, 2.0], [2.0, 4.0]]), t.tensor([1.0, 2.0])))


def _test():
    # invertible: independent ground truth via hand solve of a diagonal system
    A = t.tensor([[2.0, 0.0], [0.0, 5.0]])
    b = t.tensor([4.0, 10.0])
    x = safe_solve(A, b)
    assert x is not None
    assert t.allclose(x, t.tensor([2.0, 2.0]), atol=1e-5), x
    # singular: dependent rows -> must return None
    A_sing = t.tensor([[1.0, 2.0], [2.0, 4.0]])
    assert safe_solve(A_sing, t.tensor([1.0, 2.0])) is None


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
