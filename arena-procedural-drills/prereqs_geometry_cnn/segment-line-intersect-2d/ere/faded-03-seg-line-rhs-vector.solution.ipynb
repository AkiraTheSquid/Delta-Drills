## Concept

In the parametric intersection system, the right-hand side of the equation is `b = L0 - S0`, the vector from the segment's start point to a point on the line. This makes sense geometrically: both the t*d term and the s*e term are measured from S0, so the displacement to reach the line must also be measured from S0. Getting this sign wrong (e.g. writing `S0 - L0`) gives a reflected solution.

## Faded exercise 3

Given specific 2D points, set up the complete solve for segment (S0, S1) vs line through (L0, L1) and verify the intersection point.

1. Compute d, e.
2. Build A = stack([d, -e], dim=1).
3. Compute the right-hand side b.
4. Solve and return the t_seg parameter.

The blank step is forming the right-hand side b correctly.

**Fill in:** Compute the right-hand side vector b as the displacement from the segment start S0 to the line point L0.

````
import torch as t

t.manual_seed(0)

S0 = t.tensor([1.0, 1.0])
S1 = t.tensor([5.0, 3.0])
L0 = t.tensor([3.0, 0.0])
L1 = t.tensor([3.0, 5.0])

d = S1 - S0
e = L1 - L0
A = t.stack([d, -e], dim=1)
b = L0 - S0
ts = t.linalg.solve(A, b)
t_seg = ts[0].item()
hit = (0.0 <= t_seg <= 1.0)

print(f't_seg={t_seg:.4f}  hit={hit}')
pt = S0 + t_seg * d
print(f'Intersection point: {pt.tolist()}')


def _test():
    import torch as t

    S0 = t.tensor([1.0, 1.0])
    S1 = t.tensor([5.0, 3.0])
    L0 = t.tensor([3.0, 0.0])
    L1 = t.tensor([3.0, 5.0])

    d = S1 - S0
    e = L1 - L0
    A = t.stack([d, -e], dim=1)
    b_ref = L0 - S0
    assert t.allclose(b, b_ref, atol=1e-5), f'b={b} expected={b_ref}'
    # the resulting t_seg should place intersection at x=3
    ts_ref = t.linalg.solve(A, b_ref)
    t_ref = ts_ref[0].item()
    assert abs(t_seg - t_ref) < 1e-5
    pt = S0 + t_seg * d
    assert abs(pt[0].item() - 3.0) < 1e-4, f'x coord: {pt[0].item()}'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
