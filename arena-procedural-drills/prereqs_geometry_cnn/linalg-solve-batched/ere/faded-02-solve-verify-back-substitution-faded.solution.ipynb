## Concept

After computing `x = t.linalg.solve(A, b)`, you can verify correctness by checking that `A @ x` reproduces `b`. With batch dimensions, the matmul is `(A: (K,n,n)) @ (x.unsqueeze(-1): (K,n,1))` to get `(K,n,1)`, then squeeze to `(K,n)` for comparison.

## Faded exercise 2

The solution `x` of shape `(K, 3)` is already computed. Fill in the verification step that computes `A @ x` and checks it against `b`.

**Fill in:** Compute Ax by matrix-multiplying A (K,n,n) against x unsqueezed to (K,n,1) and squeeze the result back to (K,n).

````
import torch as t

t.manual_seed(2)
K, n = 6, 3
A = t.randn(K, n, n) + 2 * t.eye(n)
b = t.randn(K, n)
x = t.linalg.solve(A, b)

Ax = (A @ x.unsqueeze(-1)).squeeze(-1)
assert t.allclose(Ax, b, atol=1e-4)


def _test():
    import torch as t

    assert Ax is not None
    assert Ax.shape == b.shape, f'expected {b.shape} got {Ax.shape}'
    assert t.allclose(Ax, b, atol=1e-4), f'max residual: {(Ax-b).abs().max().item()}'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
