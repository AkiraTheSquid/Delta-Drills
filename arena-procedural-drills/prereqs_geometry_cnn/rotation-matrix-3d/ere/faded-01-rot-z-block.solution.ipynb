## Concept

A z-axis rotation fixes z and rotates the x-y plane: `R_z = [[c,-s,0],[s,c,0],[0,0,1]]` with `c=cos(theta)`, `s=sin(theta)`. It is orthogonal with determinant +1.

## Faded exercise 1

`c` and `s` are computed. Complete the matrix literal so it is the correct z-axis rotation (planar rotation in the x-y block, z fixed).

**Fill in:** return the 3x3 R_z matrix [[c,-s,0],[s,c,0],[0,0,1]]

````
import math
Tensor = t.Tensor


def rot_z(theta: float) -> Tensor:
    c, s = math.cos(theta), math.sin(theta)
    return t.tensor([
        [c, -s, 0.0],
        [s,  c, 0.0],
        [0.0, 0.0, 1.0],
    ])


R = rot_z(0.5)

def _test():
    import math
    R = rot_z(0.5)
    assert tuple(R.shape) == (3, 3)
    assert bool(t.allclose(R @ R.T, t.eye(3), atol=1e-6)), 'must be orthogonal'
    assert abs(t.linalg.det(R).item() - 1.0) < 1e-6, 'det must be +1'
    # independent ground truth: rotating [1,0,0] by 0.5 about z gives [cos,sin,0]
    rotated = R @ t.tensor([1.0, 0.0, 0.0])
    expected = t.tensor([math.cos(0.5), math.sin(0.5), 0.0])
    assert t.allclose(rotated, expected, atol=1e-6), 'x-axis must map to [cos,sin,0]'

try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
