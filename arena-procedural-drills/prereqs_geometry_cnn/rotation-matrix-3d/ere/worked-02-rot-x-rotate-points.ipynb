## Concept

Rotation about the x-axis fixes the x-coordinate: `R_x = [[1,0,0],[0,c,-s],[0,s,c]]`. To rotate a batch of points stored as rows in a `(N, 3)` matrix, apply `points @ R_x.T`, which is the row-vector convention equivalent to rotating each point by `R_x`.

## Worked solution

We build `R_x(theta)` with the x-coordinate fixed and the y-z plane rotated. We make a small `(3, 3)` set of points (the three standard basis vectors as rows) and rotate them with `points @ R_x.T`. Because the x-axis is fixed, the first basis vector `[1,0,0]` is unchanged, while `[0,1,0]` and `[0,0,1]` rotate within the y-z plane. We verify that distances from the origin are preserved (rotation is an isometry) by comparing row norms before and after, and print the rotated points.

```
import math
Tensor = t.Tensor


def rot_x(theta: float) -> Tensor:
    c, s = math.cos(theta), math.sin(theta)
    return t.tensor([
        [1.0, 0.0, 0.0],
        [0.0, c, -s],
        [0.0, s,  c],
    ])


R = rot_x(math.pi / 2)
points = t.eye(3)              # rows are e_x, e_y, e_z
rotated = points @ R.T
print('rotated points:', [[round(v, 4) for v in row] for row in rotated.tolist()])
before = points.norm(dim=1)
after = rotated.norm(dim=1)
print('norms preserved:', bool(t.allclose(before, after, atol=1e-6)))
```
