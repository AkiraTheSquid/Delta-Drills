## Concept

A rotation about the z-axis by angle theta leaves the z-coordinate fixed and rotates the x-y plane: `R_z = [[c, -s, 0], [s, c, 0], [0, 0, 1]]` where `c = cos(theta)`, `s = sin(theta)`. Applying it as `p @ R_z.T` (or `R_z @ p`) rotates points. It is orthogonal with determinant +1, the hallmark of a proper rotation.

## Worked solution

We compute `c` and `s` from `theta`, then assemble the 3x3 matrix row by row with `t.tensor`. The top-left 2x2 block is the standard planar rotation; the bottom-right entry is 1 so z is untouched. We verify the rotation properties: `R @ R.T` is the identity (orthogonality) and `det(R)` is +1. We also rotate the unit z-vector and confirm it is unchanged, since the z-axis is the rotation axis. We print the determinant and the rotated z-vector.

```
import math
Tensor = t.Tensor


def rot_z(theta: float) -> Tensor:
    c, s = math.cos(theta), math.sin(theta)
    return t.tensor([
        [c, -s, 0.0],
        [s,  c, 0.0],
        [0.0, 0.0, 1.0],
    ])


R = rot_z(math.pi / 2)
print('det:', round(t.linalg.det(R).item(), 5))
print('orthogonal:', bool(t.allclose(R @ R.T, t.eye(3), atol=1e-6)))
z = t.tensor([0.0, 0.0, 1.0])
print('z fixed:', bool(t.allclose(R @ z, z, atol=1e-6)))
```
