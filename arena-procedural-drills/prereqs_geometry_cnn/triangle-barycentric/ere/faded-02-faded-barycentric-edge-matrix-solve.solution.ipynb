## Concept

To convert a Cartesian point `P` to barycentric coordinates `(u, v)` with respect to triangle `ABC`, you solve the 2×2 system `[B-A | C-A] @ [u, v] = P - A`. The matrix columns are the edge vectors `B-A` and `C-A`. One call to `t.linalg.solve` handles this.

## Faded exercise 2

Implement `faded2_to_barycentric(A, B, C, P)` where all arguments are `(2,)` float tensors. Build the edge matrix `M = t.stack([B - A, C - A], dim=1)` (edges as columns). Return `t.linalg.solve(M, P - A)`. The blank is the `t.linalg.solve` call.

**Fill in:** The t.linalg.solve(M, P - A) call that solves for barycentric coordinates (u, v).

````
import torch as t

def faded2_to_barycentric(A, B, C, P):
    M = t.stack([B - A, C - A], dim=1)  # (2, 2)
    return t.linalg.solve(M, P - A)


def _test():
    import torch as t
    A = t.tensor([0.0, 0.0])
    B = t.tensor([1.0, 0.0])
    C = t.tensor([0.0, 1.0])
    # Centroid should be (1/3, 1/3)
    centroid = (A + B + C) / 3
    uv = faded2_to_barycentric(A, B, C, centroid)
    assert uv.shape == (2,), f'wrong shape: {uv.shape}'
    assert abs(uv[0].item() - 1/3) < 1e-5, f'u={uv[0].item()}'
    assert abs(uv[1].item() - 1/3) < 1e-5, f'v={uv[1].item()}'
    # Vertex A should be (0, 0)
    uv_a = faded2_to_barycentric(A, B, C, A)
    assert abs(uv_a[0].item()) < 1e-5 and abs(uv_a[1].item()) < 1e-5
    # Vertex B should be (1, 0)
    uv_b = faded2_to_barycentric(A, B, C, B)
    assert abs(uv_b[0].item() - 1.0) < 1e-5 and abs(uv_b[1].item()) < 1e-5


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
