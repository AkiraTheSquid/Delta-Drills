## Concept

Given barycentric coordinates `(u, v)` with respect to triangle `ABC`, the Cartesian position is `P = A + u*(B-A) + v*(C-A)`. This is the defining equation of barycentric coordinates — the solve in the previous step is its inverse. You can use it to verify a solve or to generate points at known positions within the triangle.

## Faded exercise 3

Implement `faded3_reconstruct(A, B, C, uv)` where `A`, `B`, `C` are `(2,)` vertices and `uv` is a `(2,)` barycentric coordinate tensor. Return `A + uv[0] * (B - A) + uv[1] * (C - A)`. The blank is the complete return expression.

**Fill in:** The return statement A + uv[0] * (B - A) + uv[1] * (C - A) that reconstructs the Cartesian point from barycentric coordinates.

````
import torch as t

def faded3_reconstruct(A, B, C, uv):
    return A + uv[0] * (B - A) + uv[1] * (C - A)


def _test():
    import torch as t
    A = t.tensor([1.0, 2.0])
    B = t.tensor([4.0, 2.0])
    C = t.tensor([1.0, 5.0])
    # Edge midpoint of AB: u=0.5, v=0 => should reconstruct to midpoint
    uv_ab_mid = t.tensor([0.5, 0.0])
    P = faded3_reconstruct(A, B, C, uv_ab_mid)
    expected = (A + B) / 2
    assert t.allclose(P, expected, atol=1e-5), f'P={P}, expected={expected}'
    # Vertex C: u=0, v=1
    uv_c = t.tensor([0.0, 1.0])
    P_c = faded3_reconstruct(A, B, C, uv_c)
    assert t.allclose(P_c, C, atol=1e-5), f'P_c={P_c}, expected C={C}'
    # Centroid: u=v=1/3
    uv_cent = t.tensor([1/3, 1/3])
    P_cent = faded3_reconstruct(A, B, C, uv_cent)
    expected_cent = (A + B + C) / 3
    assert t.allclose(P_cent, expected_cent, atol=1e-5)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
