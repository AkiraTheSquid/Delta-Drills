## Concept

A point expressed in barycentric coordinates `(u, v)` lies inside the triangle if all three conditions hold simultaneously: `u >= 0`, `v >= 0`, and `u + v <= 1`. You can apply all three with a single chained `&` expression over 1D slices of the `(N, 2)` input tensor.

## Faded exercise 1

Implement `faded1_inside(uvs)` where `uvs` has shape `(N, 2)`. Return a boolean tensor of shape `(N,)` that is True for points inside (or on the boundary of) the canonical triangle. Extract `u = uvs[:, 0]` and `v = uvs[:, 1]`, then return the three-predicate conjunction. The blank is the return statement with the combined predicate.

**Fill in:** The return statement computing (u >= 0) & (v >= 0) & (u + v <= 1) as the inside-triangle predicate.

````
import torch as t

def faded1_inside(uvs):
    u = uvs[:, 0]
    v = uvs[:, 1]
    return (u >= 0) & (v >= 0) & (u + v <= 1)


def _test():
    import torch as t
    # Ground-truth points
    inside_pts = t.tensor([[0.0, 0.0], [0.5, 0.0], [0.0, 0.5], [1/3, 1/3], [1.0, 0.0], [0.0, 1.0]])
    outside_pts = t.tensor([[1.1, 0.0], [-0.1, 0.5], [0.6, 0.6], [0.5, 0.6]])
    all_pts = t.cat([inside_pts, outside_pts])
    result = faded1_inside(all_pts)
    expected = t.tensor([True]*len(inside_pts) + [False]*len(outside_pts))
    assert (result == expected).all(), f'mismatch: {result}'


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
