import torch as t

def kl_gaussian(mu: t.Tensor, logsigma: t.Tensor):
    per_elem = 0.5 * (mu.pow(2) + (2 * logsigma).exp() - 1 - 2 * logsigma)
    per_sample = per_elem.sum(dim=-1)
    scalar = per_sample.mean()
    return per_sample, scalar
