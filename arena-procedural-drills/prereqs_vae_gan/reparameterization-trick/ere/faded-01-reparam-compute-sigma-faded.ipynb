## Concept

In the reparameterization trick, the encoder outputs `(mu, logsigma)` where `logsigma` is the log of the variance. To sample `z ~ N(mu, sigma^2)`, first recover `sigma = exp(0.5 * logsigma)` — this converts log-variance to standard deviation. Then compute `z = mu + sigma * eps` where `eps ~ N(0, 1)`.
