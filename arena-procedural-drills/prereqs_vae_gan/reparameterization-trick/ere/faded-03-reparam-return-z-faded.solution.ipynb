## Concept

Once `sigma` and `eps` are computed, the reparameterized sample is `z = mu + sigma * eps`. This combines the mean `mu` (shift) with the scaled noise `sigma * eps` (spread). The key insight is that `eps` is treated as a non-learnable constant, so gradients flow through the multiplication `sigma * eps` and the addition `mu + ...` back to the learnable parameters.

## Faded exercise 3

The `sigma` computation and `eps` draw are provided. Your task is to **compute and return `z = mu + sigma * eps`**.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def reparameterize(mu: t.Tensor, logsigma: t.Tensor) -> t.Tensor:
    sigma = (0.5 * logsigma).exp()
    eps = t.randn_like(mu)
    return mu + sigma * eps


def _test():
    import torch as t
    t.manual_seed(11)
    mu       = t.tensor([[1.0, 2.0, 3.0]])
    logsigma = t.tensor([[0.0, 0.0, 0.0]])  # sigma = 1
    t.manual_seed(11)
    z = reparameterize(mu, logsigma)
    assert z.shape == mu.shape
    # Reconstruct expected: same seed, sigma=1 so z = mu + randn
    t.manual_seed(11)
    sigma = (0.5 * logsigma).exp()
    eps = t.randn_like(mu)
    expected = mu + sigma * eps
    assert t.allclose(z, expected, atol=1e-6), f'Expected {expected}, got {z}'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
