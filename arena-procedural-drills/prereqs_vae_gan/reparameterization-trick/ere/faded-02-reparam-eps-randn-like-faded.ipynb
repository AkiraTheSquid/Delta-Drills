## Concept

The noise `eps` in the reparameterization trick must match the shape, dtype, and device of `mu`. Using `t.randn_like(mu)` achieves this automatically — it creates a tensor of the same shape filled with standard normal draws. This is preferable to `t.randn(*mu.shape)` because it correctly handles GPU tensors and non-float32 dtypes.
