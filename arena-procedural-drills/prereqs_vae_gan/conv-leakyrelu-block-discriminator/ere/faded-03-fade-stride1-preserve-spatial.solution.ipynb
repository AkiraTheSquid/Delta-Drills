## Concept

A spatial-preserving discriminator refinement block uses `Conv2d(kernel_size=3, stride=1, padding=1, bias=False)` so output H/W equal input H/W, followed by BatchNorm and LeakyReLU(0.2). With kernel 3 and padding 1, stride 1 keeps the spatial size unchanged.

### Faded — complete the spatial-preserving conv

Build a block `Conv2d -> BatchNorm2d -> LeakyReLU(0.2)` whose output spatial dims EQUAL the input dims (no downsampling). The BatchNorm and activation are filled in. Complete the `nn.Conv2d` so that, for kernel size 3 and stride 1, the spatial size is preserved and the conv carries no bias.

**Fill in:** the nn.Conv2d layer with kernel_size=3, stride=1, padding=1, bias=False so spatial dims are preserved.

````
import torch.nn as nn

def build_stride1_block(in_ch, out_ch):
    conv = nn.Conv2d(in_ch, out_ch, kernel_size=3, stride=1, padding=1, bias=False)
    return nn.Sequential(
        conv,
        nn.BatchNorm2d(out_ch),
        nn.LeakyReLU(0.2, inplace=True),
    )


def _test():
    import torch.nn as nn
    t.manual_seed(0)
    block = build_stride1_block(6, 12)
    for H, W in [(7, 7), (16, 24), (5, 9)]:
        x = t.randn(2, 6, H, W)
        out = block(x)
        assert tuple(out.shape) == (2, 12, H, W), (tuple(out.shape), H, W)
    conv = block[0]
    assert isinstance(conv, nn.Conv2d)
    assert conv.kernel_size == (3, 3)
    assert conv.stride == (1, 1)
    assert conv.padding == (1, 1)
    assert conv.bias is None


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
