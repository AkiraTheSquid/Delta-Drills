## Concept

The VAE training loss is `reconstruction + beta * kl`, where only the KL term carries the `beta` weight. Getting the weighting on the correct term (and leaving reconstruction unscaled) is the whole point.

## Faded exercise 1

Implement `elbo(recon, kl, beta)`. `recon` and `kl` are 0-D scalar tensors and `beta` is a Python float. Return the beta-weighted ELBO loss as a scalar tensor. Complete the blanked combination line.

**Fill in:** the line that returns reconstruction plus beta times the KL term

````
import torch as t

t.manual_seed(3)
recon = t.tensor(3.0)
kl = t.tensor(1.5)

def elbo(recon, kl, beta):
    loss = recon + beta * kl
    return loss

print(float(elbo(recon, kl, 2.0)))


def _test():
    # independent ground truth computed with plain python floats
    for beta in (0.0, 1.0, 2.5, 4.0):
        out = elbo(t.tensor(3.0), t.tensor(1.5), beta)
        expected = 3.0 + beta * 1.5
        assert abs(float(out) - expected) < 1e-6, (beta, float(out), expected)
    # beta=0 must leave reconstruction untouched
    assert float(elbo(t.tensor(3.0), t.tensor(1.5), 0.0)) == 3.0


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
