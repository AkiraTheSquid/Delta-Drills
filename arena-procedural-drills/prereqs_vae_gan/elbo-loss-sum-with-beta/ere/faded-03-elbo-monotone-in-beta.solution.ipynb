Holding `(recon, kl)` fixed with `kl > 0`, the composite loss `recon + beta * kl` is strictly increasing in `beta`. Sweeping beta produces a straight line whose slope is `kl` and whose intercept is `recon`.

## Faded exercise 3

Implement `elbo_sweep(recon, kl, betas)`. `recon` and `kl` are scalar tensors, `betas` is a list of floats. Return a 1-D tensor of `recon + beta * kl` for each beta, in order. Complete the blanked per-beta loss expression.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

t.manual_seed(5)
recon = t.tensor(0.9)
kl = t.tensor(0.4)

def elbo_sweep(recon, kl, betas):
    out = []
    for beta in betas:
        loss = recon + beta * kl
        out.append(loss)
    return t.stack(out)

print(elbo_sweep(recon, kl, [0.0, 1.0, 2.0, 3.0]))


def _test():
    betas = [0.0, 1.0, 2.0, 3.0]
    out = elbo_sweep(t.tensor(0.9), t.tensor(0.4), betas)
    assert out.shape == (4,), out.shape
    # independent ground truth per element
    expected = t.tensor([0.9 + b * 0.4 for b in betas])
    assert t.allclose(out, expected, atol=1e-5), (out, expected)
    # strictly increasing because kl>0
    diffs = out[1:] - out[:-1]
    assert bool((diffs > 0).all())


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
