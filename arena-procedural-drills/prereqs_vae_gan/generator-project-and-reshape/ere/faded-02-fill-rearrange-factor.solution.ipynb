## Concept

`einops.rearrange('b (c h w) -> b c h w', c=C, h=H, w=W)` splits the flat projection into channel and spatial axes, with `c` slowest and `w` fastest — matching row-major reshape order while documenting each axis.

Implement `project_rearrange(z, weight, bias, C, H, W)`. Project `z`, then reshape the flat output to `(B, C, H, W)` using `einops.rearrange`. Complete the blanked rearrange pattern string.

**Fill in:** the rearrange pattern splitting the flat feature axis into c, h, w

````
import torch as t
import einops
from einops import rearrange

t.manual_seed(4)
latent_dim, C, H, W = 64, 8, 4, 4
z = t.randn(3, latent_dim)
weight = t.randn(C * H * W, latent_dim) * 0.02
bias = t.zeros(C * H * W)

def project_rearrange(z, weight, bias, C, H, W):
    flat = z @ weight.T + bias
    pattern = 'b (c h w) -> b c h w'
    return rearrange(flat, pattern, c=C, h=H, w=W)

print(project_rearrange(z, weight, bias, C, H, W).shape)


def _test():
    out = project_rearrange(z, weight, bias, C, H, W)
    assert out.shape == (3, C, H, W), out.shape
    # independent ground truth: plain view must agree with the einops factorization order
    flat = z @ weight.T + bias
    expected = flat.view(3, C, H, W)
    assert t.equal(out, expected)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
