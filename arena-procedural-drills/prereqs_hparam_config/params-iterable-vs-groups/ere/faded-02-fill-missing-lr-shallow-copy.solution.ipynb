````
import torch as t

def fill_lr_in_groups(group_list, default_lr):
    out = []
    for group in group_list:
        g = dict(group)  # TODO: Create a shallow copy of each group dict before modifying it.
        if 'lr' not in g:
            g['lr'] = default_lr
        out.append(g)
    return out

# Test
t.manual_seed(5)
groups = [{'params': [t.randn(2)], 'lr': 0.1}, {'params': [t.randn(3)]}]
filled = fill_lr_in_groups(groups, default_lr=0.01)
print(filled[1]['lr'], 'lr' in groups[1])


def _test():
    import torch as t
    t.manual_seed(5)
    p1, p2 = t.randn(2), t.randn(3)
    original_groups = [{'params': [p1], 'lr': 0.1}, {'params': [p2]}]
    filled = fill_lr_in_groups(original_groups, default_lr=0.01)
    assert filled[0]['lr'] == 0.1, 'existing lr should be preserved'
    assert filled[1]['lr'] == 0.01, 'missing lr should be filled with default'
    assert 'lr' not in original_groups[1], 'caller dict must NOT be mutated'
    assert filled[0] is not original_groups[0], 'must return a copy, not the original'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
