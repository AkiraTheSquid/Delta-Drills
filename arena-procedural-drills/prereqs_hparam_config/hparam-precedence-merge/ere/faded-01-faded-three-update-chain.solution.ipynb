## Concept

The three-call `dict.update` pattern is an explicit way to build a merged config without mutating any input. Start with an empty dict, then call `.update(defaults)`, `.update(file_cfg)`, `.update(cli_args)` in order of increasing priority. Each `.update` call only adds or overwrites — it never deletes — so every key from every layer ends up in the result with the correct priority winner.

## Faded exercise 1

**Fill in:** The two remaining update calls (file_cfg then cli_args) followed by the return statement.

````
def merge_config(defaults, file_cfg, cli_args):
    out = {}
    out.update(defaults)
    out.update(file_cfg)
    out.update(cli_args)
    return out

# Exercise it
defaults  = {'lr': 0.01, 'batch_size': 32, 'dropout': 0.1}
file_cfg  = {'lr': 0.001, 'batch_size': 64}
cli_args  = {'lr': 0.005}
result = merge_config(defaults, file_cfg, cli_args)
print(result)  # lr=0.005, batch_size=64, dropout=0.1


def _test():
    defaults  = {'lr': 0.01, 'bs': 32, 'wd': 0.0, 'ep': 100}
    file_cfg  = {'lr': 0.001, 'bs': 64}
    cli_args  = {'lr': 0.005}

    result = merge_config(defaults, file_cfg, cli_args)

    assert result['lr'] == 0.005,  "CLI must win"
    assert result['bs'] == 64,     "file must win over defaults"
    assert result['wd'] == 0.0,    "defaults must fill missing keys"
    assert result['ep'] == 100,    "defaults must fill missing keys"

    # No mutation
    assert defaults == {'lr': 0.01, 'bs': 32, 'wd': 0.0, 'ep': 100}
    assert file_cfg == {'lr': 0.001, 'bs': 64}
    assert cli_args == {'lr': 0.005}


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
