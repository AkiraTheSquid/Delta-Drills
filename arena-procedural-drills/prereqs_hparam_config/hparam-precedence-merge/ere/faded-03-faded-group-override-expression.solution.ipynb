## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Resolving a parameter group's hyperparameters follows the pattern `{**defaults, **kwargs, **group}`. This single expression handles all three precedence levels in one go: defaults supply fallbacks, optimizer-level kwargs override those, and the per-group dict wins over everything. Any key absent from `group` and `kwargs` still appears in the result via `defaults`.

## Faded exercise 3

Complete `resolve_group` by filling in the return expression. The blank is the dict unpacking expression that applies group > kwargs > defaults precedence.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def resolve_group(defaults, kwargs, group):
    return {**defaults, **kwargs, **group}

# Exercise it
defaults = {'lr': 1e-2, 'weight_decay': 0.0, 'eps': 1e-8}
kwargs   = {'lr': 1e-3}
group    = {'lr': 1e-4, 'weight_decay': 1e-4}
resolved = resolve_group(defaults, kwargs, group)
print(resolved)  # lr=1e-4, weight_decay=1e-4, eps=1e-8


def _test():
    defaults = {'lr': 1e-2, 'wd': 0.0, 'eps': 1e-8, 'momentum': 0.9}
    kwargs   = {'lr': 1e-3, 'wd': 1e-5}
    group    = {'lr': 1e-4}  # only overrides lr

    resolved = resolve_group(defaults, kwargs, group)

    assert resolved['lr'] == 1e-4,       "group must win for lr"
    assert resolved['wd'] == 1e-5,       "kwargs must win for wd (group absent)"
    assert resolved['eps'] == 1e-8,      "defaults must supply eps"
    assert resolved['momentum'] == 0.9,  "defaults must supply momentum"

    # No mutation
    assert defaults['lr'] == 1e-2
    assert kwargs['lr'] == 1e-3
    assert group == {'lr': 1e-4}


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
