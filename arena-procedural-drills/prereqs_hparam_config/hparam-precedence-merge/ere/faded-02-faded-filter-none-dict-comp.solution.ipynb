## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

When argparse returns `None` for unset flags, those `None` values must be filtered out before the config merge. The clean approach is a dict comprehension: `{k: v for k, v in cli_args.items() if v is not None}`. This produces a new dict with only the explicitly set flags, leaving the original `cli_args` unchanged. The filtered dict is then used as the CLI layer in the merge.

## Faded exercise 2

Complete `merge_filter_none` by filling in the `cli_filtered` expression. The blank is the dict comprehension that removes None-valued keys from `cli_args`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def merge_filter_none(defaults, file_cfg, cli_args):
    cli_filtered = {k: v for k, v in cli_args.items() if v is not None}
    out = {}
    out.update(defaults)
    out.update(file_cfg)
    out.update(cli_filtered)
    return out

# Exercise it
defaults = {'lr': 0.001, 'bs': 32, 'wd': 0.0}
file_cfg = {'lr': 0.005, 'bs': 128}
cli_args = {'lr': 0.01, 'bs': None}  # bs was not set by user
result = merge_filter_none(defaults, file_cfg, cli_args)
print(result)  # lr=0.01, bs=128, wd=0.0


def _test():
    defaults = {'lr': 0.001, 'bs': 32, 'wd': 0.0, 'ep': 50}
    file_cfg = {'lr': 0.005, 'bs': 128}
    cli_args = {'lr': 0.01, 'bs': None, 'ep': None}

    result = merge_filter_none(defaults, file_cfg, cli_args)

    assert result['lr'] == 0.01,  "Explicitly set CLI value must win"
    assert result['bs'] == 128,   "None in CLI must NOT overwrite file value"
    assert result['ep'] == 50,    "None in CLI must NOT overwrite default"
    assert result['wd'] == 0.0,   "Missing-from-CLI key must come from defaults"

    # No mutation
    assert cli_args == {'lr': 0.01, 'bs': None, 'ep': None}


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
