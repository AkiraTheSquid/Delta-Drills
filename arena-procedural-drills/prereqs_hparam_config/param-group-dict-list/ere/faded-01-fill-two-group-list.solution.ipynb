## Concept

Differential learning rates are built as a `list[dict]` where each dict pairs a materialized `'params'` list with its `'lr'`. Wrapping `.parameters()` in `list(...)` guards against single-use-generator bugs.

## Faded exercise 1

Complete `make_param_groups` by building the head group dict. Fill in the head group entry.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

t.manual_seed(3)

def make_param_groups(encoder, head, encoder_lr, head_lr):
    head_group = {'params': list(head.parameters()), 'lr': head_lr}
    return [
        {'params': list(encoder.parameters()), 'lr': encoder_lr},
        head_group,
    ]

enc, head = nn.Linear(4, 4), nn.Linear(4, 2)
opt = t.optim.SGD(make_param_groups(enc, head, 1e-4, 1e-2))
print([g['lr'] for g in opt.param_groups])


def _test():
    enc, head = nn.Linear(4, 4), nn.Linear(4, 2)
    groups = make_param_groups(enc, head, 1e-4, 1e-2)
    assert len(groups) == 2
    assert groups[0]['lr'] == 1e-4
    assert groups[1]['lr'] == 1e-2
    # head group must contain exactly the head's params
    assert len(groups[1]['params']) == len(list(head.parameters()))
    opt = t.optim.SGD(groups)
    assert [g['lr'] for g in opt.param_groups] == [1e-4, 1e-2]


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
