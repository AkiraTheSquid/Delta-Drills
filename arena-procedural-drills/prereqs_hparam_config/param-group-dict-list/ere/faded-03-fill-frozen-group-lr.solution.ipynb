Setting a param group's `lr` to `0.0` freezes that subset while keeping it registered with the optimizer. The head group keeps a normal learning rate.

## Faded exercise 3

Complete `make_frozen_backbone_groups` so the backbone group uses a zero learning rate. Fill in the backbone group's lr.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

t.manual_seed(8)

def make_frozen_backbone_groups(backbone, head, head_lr):
    frozen_lr = 0.0
    return [
        {'params': list(backbone.parameters()), 'lr': frozen_lr},
        {'params': list(head.parameters()), 'lr': head_lr},
    ]

bb, head = nn.Linear(4, 4), nn.Linear(4, 1)
opt = t.optim.SGD(make_frozen_backbone_groups(bb, head, 0.1))
print([g['lr'] for g in opt.param_groups])


def _test():
    t.manual_seed(80)
    bb, head = nn.Linear(4, 4), nn.Linear(4, 1)
    opt = t.optim.SGD(make_frozen_backbone_groups(bb, head, 0.1))
    assert opt.param_groups[0]['lr'] == 0.0
    assert opt.param_groups[1]['lr'] == 0.1
    for p in list(bb.parameters()) + list(head.parameters()):
        p.grad = t.ones_like(p)
    bb0 = bb.weight.data.clone()
    hd0 = head.weight.data.clone()
    opt.step()
    # frozen backbone unchanged; head moved
    assert t.allclose(bb.weight.data, bb0)
    assert not t.allclose(head.weight.data, hd0)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
