The manual SGD step is an outer loop over `optimizer.param_groups` and an inner loop over each `group['params']`. Per parameter, skip `grad is None`, then apply `p -= lr * grad` in place using the group's own learning rate.

## Faded exercise 1

Complete `manual_sgd_step` so each parameter is updated in place using its group's learning rate. Fill in the in-place update line.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

t.manual_seed(3)

a = t.nn.Parameter(t.randn(3))
b = t.nn.Parameter(t.randn(3))
opt = t.optim.SGD([
    {'params': [a], 'lr': 0.1},
    {'params': [b], 'lr': 0.5},
])
a.grad = t.ones(3)
b.grad = t.full((3,), 2.0)

def manual_sgd_step(optimizer):
    for group in optimizer.param_groups:
        lr = group['lr']
        for p in group['params']:
            if p.grad is None:
                continue
            p.data.add_(p.grad, alpha=-lr)

manual_sgd_step(opt)
print(a.data, b.data)


def _test():
    t.manual_seed(50)
    x = t.nn.Parameter(t.randn(4))
    y = t.nn.Parameter(t.randn(4))
    optimizer = t.optim.SGD([
        {'params': [x], 'lr': 0.1},
        {'params': [y], 'lr': 0.25},
    ])
    gx = t.randn(4)
    gy = t.randn(4)
    x.grad = gx.clone()
    y.grad = gy.clone()
    x0 = x.data.clone()
    y0 = y.data.clone()
    manual_sgd_step(optimizer)
    # independent ground truth
    assert t.allclose(x.data, x0 - 0.1 * gx)
    assert t.allclose(y.data, y0 - 0.25 * gy)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
