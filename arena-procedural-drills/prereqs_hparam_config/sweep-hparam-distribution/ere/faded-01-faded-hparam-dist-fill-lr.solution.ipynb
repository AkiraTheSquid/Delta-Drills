## Concept

Learning rate typically needs to be explored over multiple orders of magnitude — the difference between 0.0001 and 0.001 matters as much as between 0.001 and 0.01. The correct wandb distribution for this is `'log_uniform_values'`, which samples uniformly on the log scale between `min` and `max` (specified in value space).
