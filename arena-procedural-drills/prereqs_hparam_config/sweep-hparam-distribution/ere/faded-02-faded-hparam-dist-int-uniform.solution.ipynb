## Concept

Integer hyperparameters like number of layers require `'int_uniform'` rather than `'uniform'` to ensure wandb samples only whole numbers. The `min` and `max` values should be Python integers. Using `'uniform'` for integer counts would produce floats that could cause errors or unexpected truncation in model code.
