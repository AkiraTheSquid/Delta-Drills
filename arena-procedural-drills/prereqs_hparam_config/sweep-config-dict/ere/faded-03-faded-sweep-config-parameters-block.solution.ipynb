## Concept

The `'parameters'` dict in a sweep config maps each hyperparameter name to a spec dict. The spec must contain exactly one of: `'value'` (fixed), `'values'` (discrete list), or `'distribution'` with `'min'`/`'max'` (continuous). Mixing more than one specifier in the same entry is a schema error.

**Fill in:** Replace `None` with a dict containing all four hyperparameter specs as described.

````
def build_full_config() -> dict:
    return {
        'method': 'bayes',
        'metric': {'name': 'val_loss', 'goal': 'minimize'},
        'parameters': {
            'lr': {'distribution': 'log_uniform_values', 'min': 1e-5, 'max': 1e-1},
            'batch_size': {'values': [32, 64, 128]},
            'optimizer': {'value': 'adamw'},
            'weight_decay': {'distribution': 'log_uniform_values', 'min': 1e-6, 'max': 1e-3},
        },
    }

cfg = build_full_config()
if cfg['parameters'] is not None:
    for k, v in cfg['parameters'].items():
        print(f'  {k}: {v}')


def _test():
    cfg = build_full_config()
    p = cfg['parameters']
    assert isinstance(p, dict), 'parameters must be a dict'
    assert set(p.keys()) == {'lr', 'batch_size', 'optimizer', 'weight_decay'}
    # lr
    assert p['lr'].get('distribution') == 'log_uniform_values'
    assert p['lr']['min'] == 1e-5 and p['lr']['max'] == 1e-1
    # batch_size
    assert p['batch_size'].get('values') == [32, 64, 128]
    # optimizer
    assert p['optimizer'].get('value') == 'adamw'
    assert 'values' not in p['optimizer']
    # weight_decay
    assert p['weight_decay'].get('distribution') == 'log_uniform_values'
    assert p['weight_decay']['min'] == 1e-6 and p['weight_decay']['max'] == 1e-3


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
