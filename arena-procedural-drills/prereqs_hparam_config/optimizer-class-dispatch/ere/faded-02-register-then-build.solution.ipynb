## Concept

Adding a custom optimizer to the dispatch table at runtime requires only a single dict assignment after validating the input class. Once registered, the same `build_optimizer` factory function can construct it by name — no code changes needed in the factory itself.

## Faded exercise 2

Complete `register_optimizer(name, cls)`. The two validation checks are already written. Fill in the registration step: insert `cls` into the `OPTIM_MAP` dict under `name`.

**Fill in:** The dict assignment that stores cls in OPTIM_MAP under the given name key.

````
import torch as t
import torch.nn as nn

OPTIM_MAP = {
    'sgd':   t.optim.SGD,
    'adam':  t.optim.Adam,
    'adamw': t.optim.AdamW,
}

def register_optimizer(name: str, cls) -> None:
    if not isinstance(cls, type):
        raise TypeError(f'Expected a class, got {type(cls).__name__}')
    if not issubclass(cls, t.optim.Optimizer):
        raise TypeError(f'{cls.__name__} must subclass torch.optim.Optimizer')
    OPTIM_MAP[name] = cls

def build_optimizer(name: str, params, lr: float) -> t.optim.Optimizer:
    return OPTIM_MAP[name](params, lr=lr)

# --- run it ---
register_optimizer('rmsprop', t.optim.RMSprop)
print('rmsprop' in OPTIM_MAP)  # True


def _test():
    import torch as t
    import torch.nn as nn
    # Reset registry to known state
    OPTIM_MAP.clear()
    OPTIM_MAP.update({'sgd': t.optim.SGD, 'adam': t.optim.Adam, 'adamw': t.optim.AdamW})
    register_optimizer('rmsprop', t.optim.RMSprop)
    assert 'rmsprop' in OPTIM_MAP
    assert OPTIM_MAP['rmsprop'] is t.optim.RMSprop
    # Should reject non-optimizer
    try:
        register_optimizer('bad', object)
        assert False, 'Should raise TypeError'
    except TypeError:
        pass
    # Build the newly registered optimizer
    model = nn.Linear(4, 2)
    opt = build_optimizer('rmsprop', list(model.parameters()), lr=1e-3)
    assert isinstance(opt, t.optim.RMSprop)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
