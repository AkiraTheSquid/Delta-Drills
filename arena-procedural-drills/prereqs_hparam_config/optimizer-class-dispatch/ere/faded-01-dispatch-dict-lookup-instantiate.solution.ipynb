## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

A dispatch dictionary maps string names to optimizer classes. The factory function simply indexes into the dict to retrieve the class, then calls it with `(params, lr=lr)`. This separates 'which optimizer' (config concern) from 'how to instantiate it' (plumbing concern).

The `OPTIM_MAP` dictionary is already defined. Complete `build_optimizer(name, params, lr)`: look up the class by `name` in `OPTIM_MAP` and call it with `params` and `lr` to return an optimizer instance.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

OPTIM_MAP = {
    'sgd':   t.optim.SGD,
    'adam':  t.optim.Adam,
    'adamw': t.optim.AdamW,
}

def build_optimizer(name: str, params, lr: float) -> t.optim.Optimizer:
    return OPTIM_MAP[name](params, lr=lr)

# --- run it ---
t.manual_seed(0)
model = nn.Linear(8, 4)
opt = build_optimizer('adam', model.parameters(), lr=1e-3)
print(type(opt).__name__)  # Adam


def _test():
    import torch as t
    import torch.nn as nn
    t.manual_seed(0)
    model = nn.Linear(8, 4)
    for name, cls in [('sgd', t.optim.SGD), ('adam', t.optim.Adam), ('adamw', t.optim.AdamW)]:
        opt = build_optimizer(name, list(model.parameters()), lr=1e-3)
        assert isinstance(opt, cls), f'Expected {cls.__name__}, got {type(opt).__name__}'
    try:
        build_optimizer('unknown', list(model.parameters()), lr=1e-3)
        assert False, 'Should have raised KeyError'
    except KeyError:
        pass


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
