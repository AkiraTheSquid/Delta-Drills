## Concept

`dataclasses.asdict(obj)` recursively converts a dataclass instance into a plain dict of its fields, which is JSON-serializable when the fields are simple scalars. You can then reconstruct the original by splatting the dict back into the constructor with `Cls(**d)`, giving a clean save/load round-trip for configs.

Complete `to_payload`, which must return the JSON-serializable dict form of a `CheckpointArgs` instance. Use `dataclasses.asdict`. The dataclass definition and the `from_payload` reconstructor are already written for you; your blank is only the body of `to_payload`.

**Fill in:** the conversion of the dataclass instance to a plain dict via dataclasses.asdict(args).

````
import json
from dataclasses import dataclass, asdict

@dataclass
class CheckpointArgs:
    lr: float = 1e-3
    epochs: int = 5
    save_dir: str = 'ckpts'

def to_payload(args):
    return asdict(args)

def from_payload(d):
    return CheckpointArgs(**d)

orig = CheckpointArgs(lr=2e-4, epochs=8, save_dir='runs/exp1')
blob = json.dumps(to_payload(orig))
back = from_payload(json.loads(blob))
print(back)
print(back == orig)


def _test():
    import json
    orig = CheckpointArgs(lr=2e-4, epochs=8, save_dir='runs/exp1')
    payload = to_payload(orig)
    assert isinstance(payload, dict), type(payload)
    assert payload == {'lr': 2e-4, 'epochs': 8, 'save_dir': 'runs/exp1'}, payload
    blob = json.dumps(payload)
    back = from_payload(json.loads(blob))
    assert back == orig, (back, orig)
    assert isinstance(back, CheckpointArgs), type(back)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
