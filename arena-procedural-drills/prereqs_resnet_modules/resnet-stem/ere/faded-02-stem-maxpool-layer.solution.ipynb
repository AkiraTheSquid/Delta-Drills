## Faded exercise 2

The conv, BatchNorm, and ReLU are present. Complete the final layer: the 3x3 stride-2 maxpool with padding 1 that performs the second 2x downsample.

**Fill in:** construct the 3x3 stride-2 padding-1 MaxPool2d

````
import torch.nn as nn


def build_stem():
    pool = nn.MaxPool2d(kernel_size=3, stride=2, padding=1)
    return nn.Sequential(
        nn.Conv2d(3, 64, kernel_size=7, stride=2, padding=3, bias=False),
        nn.BatchNorm2d(64),
        nn.ReLU(inplace=True),
        pool,
    )


stem = build_stem().eval()
x = t.zeros(1, 3, 224, 224)
out = stem(x)

def _test():
    stem = build_stem().eval()
    pool = stem[3]
    assert isinstance(pool, nn.MaxPool2d), 'last layer must be MaxPool2d'
    assert pool.kernel_size == 3 and pool.stride == 2 and pool.padding == 1
    out = stem(t.zeros(1, 3, 224, 224))
    assert tuple(out.shape) == (1, 64, 56, 56), 'full stem must produce (1,64,56,56)'
    # without the pool the conv alone would give 112; pool must halve to 56
    assert out.shape[-1] == 56

try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
