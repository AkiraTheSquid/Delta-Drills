## Concept

When the classifier is a `nn.Sequential` stored at `model.classifier`, the final head is accessed via `model.classifier[-1]`. You can replace it with index assignment: `model.classifier[-1] = nn.Linear(in_features, new_num_classes)`. PyTorch's Sequential supports index-based assignment and updates the module registry automatically.
