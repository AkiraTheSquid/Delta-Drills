To pool with window `p`, the einops pattern must factor each spatial axis into a kept index and a window index: `(h p1)` for height, `(w p2)` for width, reducing to `b c h w`.

## Faded exercise 2

Complete `maxpool(x, p)`. The reducer and factor sizes are given; fill in the einops pattern string that factors both spatial axes for pooling.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import einops

def maxpool(x, p):
    pattern = 'b c (h p1) (w p2) -> b c h w'
    return einops.reduce(x, pattern, 'max', p1=p, p2=p)


def _test():
    import torch.nn.functional as F
    t.manual_seed(32)
    x = t.randn(1, 4, 8, 8)
    out = maxpool(x, 4)
    assert tuple(out.shape) == (1, 4, 2, 2)
    ref = F.max_pool2d(x, kernel_size=4)
    assert t.allclose(out, ref)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
