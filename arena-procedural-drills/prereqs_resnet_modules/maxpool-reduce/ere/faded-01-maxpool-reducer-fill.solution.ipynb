## Concept

`einops.reduce` with the factored pattern `'b c (h p1) (w p2) -> b c h w'` performs pooling; the reducer string chooses the operation. For MaxPool2d the reducer is `'max'`.

## Faded exercise 1

Complete `maxpool(x, p)` so it reproduces `nn.MaxPool2d(p)`. The einops pattern and factor sizes are given; fill in the reducer string argument.

**Fill in:** the reducer string that selects the maximum over each window

````
import torch as t
import einops

def maxpool(x, p):
    reducer = 'max'
    return einops.reduce(x, 'b c (h p1) (w p2) -> b c h w', reducer, p1=p, p2=p)


def _test():
    import torch.nn.functional as F
    t.manual_seed(31)
    x = t.randn(2, 3, 6, 8)
    out = maxpool(x, 2)
    assert tuple(out.shape) == (2, 3, 3, 4)
    # independent ground truth: torch's max_pool2d
    ref = F.max_pool2d(x, kernel_size=2)
    assert t.allclose(out, ref)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
