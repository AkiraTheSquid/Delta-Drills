## BatchNorm `.eval()` uses running stats, not batch stats

Ex1 updated `running_mean` and `running_var` via the EMA in train mode only. The deepening move toggles `.train()` vs `.eval()` on the same input and observes that the OUTPUT itself changes — eval mode uses the FIXED running stats, while train mode uses the batch's own mean/var.

```
train mode:  y = (x − batch_mean) / √(batch_var + ε) · γ + β
             [running_mean, running_var update via EMA]
eval  mode:  y = (x − running_mean) / √(running_var + ε) · γ + β
             [no update to running stats]
```

**Why this matters at inference.** A model trained with BN sees batch stats during training; at deployment the batch may be size 1 (single image), and batch stats become meaningless or undefined (var = 0 for B=1). Eval mode swaps to the running stats accumulated over training, giving stable, batch-independent outputs.

**Same module, two outputs.** Calling `bn.eval()` then `bn.train()` on the same `bn` and feeding the same `x` returns DIFFERENT tensors. The difference is the canonical 'why does my model behave weirdly at inference' debugging step for any new BN user.

Steps:
1. `bn.train()` → call `y_train = bn(x)`. This uses the batch's OWN mean/var AND updates `bn.running_mean` / `bn.running_var` via the EMA (PyTorch handles this internally).
2. `bn.eval()` → call `y_eval = bn(x)`. This uses the (now-updated) `running_mean` / `running_var` and does NOT update them.
3. Return tuple `(y_train, y_eval)`.

- `bn`: an `nn.BatchNorm2d` or `nn.BatchNorm1d` instance with pre-existing `running_mean` / `running_var` (defaults are 0 / 1).

Output: `(y_train, y_eval)` — both Tensors of shape `x.shape`.

````python
def ex2_bn_train_vs_eval_forward(x: Tensor, bn) -> tuple:
    """Run bn(x) in train() and eval() modes; return (y_train, y_eval)."""
    raise NotImplementedError()


def _test_ex2():
    import torch.nn as nn

    # === Train vs eval outputs DIFFER (the whole point) ===
    t.manual_seed(0)
    bn = nn.BatchNorm2d(4)
    x = t.randn(8, 4, 5, 6) * 3.0 + 2.0   # non-unit stats so the two modes disagree
    y_train, y_eval = ex2_bn_train_vs_eval_forward(x, bn)
    assert y_train.shape == x.shape and y_eval.shape == x.shape
    diff = (y_train - y_eval).abs()
    assert diff.max() > 1e-3, (
        f'train and eval outputs must differ on non-trivial input; got max diff {diff.max().item():.4e}'
    )

    # === Train mode: output's per-channel batch mean ≈ 0 (after centering) ===
    # In train mode, the batch is centered by its own mean; per-channel
    # spatial+batch mean of the output should be ≈ bn.bias (init=0) and var
    # scaled by bn.weight (init=1).
    ch_mean = y_train.mean(dim=(0, 2, 3))   # (C,)
    assert t.allclose(ch_mean, t.zeros(4), atol=1e-4), (
        f'train-mode per-channel batch mean must be ≈0, got {ch_mean}'
    )
    ch_var = y_train.var(dim=(0, 2, 3), unbiased=False)
    assert t.allclose(ch_var, t.ones(4), atol=1e-3), (
        f'train-mode per-channel batch var must be ≈1, got {ch_var}'
    )

    # === Eval mode: running stats actually used ===
    # After the train pass, running_mean ≠ 0 and running_var ≠ 1.
    rm = bn.running_mean.clone()
    rv = bn.running_var.clone()
    assert not t.allclose(rm, t.zeros(4)), f'running_mean must have moved off 0 after train pass, got {rm}'
    assert not t.allclose(rv, t.ones(4)), f'running_var must have moved off 1 after train pass, got {rv}'
    # y_eval should match a manual computation using the running stats.
    manual_eval = (x - rm[None, :, None, None]) / t.sqrt(rv[None, :, None, None] + bn.eps)
    manual_eval = manual_eval * bn.weight[None, :, None, None] + bn.bias[None, :, None, None]
    assert t.allclose(y_eval, manual_eval, atol=1e-5), (
        f'eval output must use running stats; mismatch max {(y_eval - manual_eval).abs().max().item():.2e}'
    )

    # === Eval mode does NOT update running stats ===
    t.manual_seed(1)
    bn2 = nn.BatchNorm2d(3)
    x2 = t.randn(4, 3, 2, 2) * 5.0
    bn2.train()
    bn2(x2)                                   # one train pass to move running stats
    rm_before = bn2.running_mean.clone()
    rv_before = bn2.running_var.clone()
    bn2.eval()
    for _ in range(3):
        bn2(x2)                               # multiple eval passes — must not change running stats
    assert t.allclose(bn2.running_mean, rm_before), 'eval mode must NOT update running_mean'
    assert t.allclose(bn2.running_var, rv_before), 'eval mode must NOT update running_var'

    # === Mode is left in eval() after the function returns ===
    # We don't strictly require this, but it's the natural last step.
    # Test instead that the function doesn't break with consecutive calls.
    y_t2, y_e2 = ex2_bn_train_vs_eval_forward(x2, bn2)
    assert y_t2.shape == x2.shape and y_e2.shape == x2.shape

    # === Return type is a 2-tuple of Tensors ===
    ret = ex2_bn_train_vs_eval_forward(x, bn)
    assert isinstance(ret, tuple) and len(ret) == 2
    assert isinstance(ret[0], t.Tensor) and isinstance(ret[1], t.Tensor)
    _dd_passed.add('ex2')
    print("ex2 ✓")

_test_ex2()
````

<details><summary>Solution</summary>

```python
def ex2_bn_train_vs_eval_forward(x, bn):
    bn.train()
    y_train = bn(x)
    bn.eval()
    y_eval = bn(x)
    return y_train, y_eval
```

**Mode toggling is the whole API.** PyTorch's BN does the stat-swap internally based on `self.training`. The Module-level `.train()` / `.eval()` flip flips this flag on every child (including BNs) — no need to thread a `training=True` arg through.

**Train pass UPDATES running stats; eval pass does NOT.** The second test confirms this by capturing `running_mean`/`var` after a train pass, then running multiple eval passes and checking the stats haven't moved.

**Why this is the canonical deploy bug.** A user who forgets to call `model.eval()` before inference gets the BN-in-train-mode behaviour: each inference batch's own stats are used. For a single-image inference call, batch_var ≈ 0 → division by ε → garbled outputs. The two-line `model.eval()` fix is invisible from the loss curves during training.
