## How these two atoms compose

The ray-parametric form says any point on a ray is `P(s) = O + s D`. The barycentric form says any point on a triangle is `Q(u, v) = A + u (B - A) + v (C - A)`. Setting them EQUAL and rearranging is what gives you the 3x3 linear system in `(s, u, v)`.

This drill exercises the DERIVATION explicitly: given `(s, u, v)`, you must reconstruct the 3D intersection point TWO ways (once via the ray, once via the triangle) and assert they agree. That's the contract the linear-system formulation is enforcing under the hood.

### Composite Exercise — derive (s, u, v) from ray-parametric + triangle-plane equality

**Atoms exercised together**: `triangle-barycentric`, `ray-parametric-form`

Implement `cx20_suv_round_trip(O, D, A, B, C, s, u, v)` that returns the 3D point computed BOTH ways from a known `(s, u, v)` solution:

- `ray_point = O + s * D`  (ray-parametric form)
- `tri_point = A + u * (B - A) + v * (C - A)`  (barycentric form)

Return a tuple `(ray_point, tri_point)`. Each is a `(3,)` tensor.

The test asserts `t.allclose(ray_point, tri_point)` for a valid `(s, u, v)` solution and that the two diverge when `(s, u, v)` is perturbed.

````python
def cx20_suv_round_trip(O, D, A, B, C, s, u, v):
    # Atom A (ray-parametric-form): P(s) = O + s D.
    ray_point = O + s * D
    # Atom B (triangle-barycentric): Q(u, v) = A + u (B - A) + v (C - A).
    tri_point = A + u * (B - A) + v * (C - A)
    return ray_point, tri_point
````
