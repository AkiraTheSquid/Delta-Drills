## `log_back` as the canonical elementwise chain rule

1. **`chain-rule-elementwise`** — for any elementwise op `out = f(x)`, the
   Jacobian is diagonal, so `dL/dx[i] = dL/dout[i] * f'(x[i])` — no matrix
   materialised, just a per-position product.
2. **`log-back`** — instantiate that rule for `f = log`, where `f'(x) = 1/x`.
   The whole back fn collapses to `grad_out / x`.

Composition: this is the shortest possible non-trivial back fn — the chain
rule (atom 1) is what TELLS you to multiply by the local derivative; the
specific `1/x` (atom 2) is what TELLS you the derivative of log.


### Composite Exercise — log_back — one-line elementwise chain rule grad_out / x

**Atoms exercised together**: `log-back`, `chain-rule-elementwise`

Implement `cx5_log_back(grad_out, out, x)` for `out = log(x)`.

One line. Use the elementwise chain rule: `grad_in = grad_out * f'(x)` with
`f'(x) = 1/x`, which collapses to `grad_out / x`.

Constraints:
- Do NOT materialize the diagonal Jacobian (no `torch.diag`, no `eye`).
- Return shape must equal `x.shape`.
- `out` is part of the signature but unused (log doesn't need cached out).


````python
def cx5_log_back(grad_out, out, x):
    # Elementwise chain rule: grad_in = grad_out * f'(x), with f'(x) = 1/x.
    return grad_out / x
````
