For `out = x - y` the arg1 partial is `-1`, so the backward returns `-g`. For `out = x / y` the arg1 partial is `-x/y**2`, so the backward returns `-g * x / (y*y)`. These two are the only non-trivial entries among sub/div arg1 lambdas; arg0 entries pass the gradient through (possibly negated).

## Faded exercise 1

### Complete BACK for sub and div

Build the `BACK` dict for `sub` and `div` (four entries). The `add`-style passthrough entries (`('sub', 0)` returns `g`, `('div', 0)` returns `g / y`) are filled in. You must complete the two arg1 lambdas: `('sub', 1)` and `('div', 1)`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
BACK = {
    ('sub', 0): lambda g, o, x, y: g,
    ('sub', 1): lambda g, o, x, y: -g,
    ('div', 0): lambda g, o, x, y: g / y,
    ('div', 1): lambda g, o, x, y: -g * x / (y * y),
}


def _test():
    t.manual_seed(1)
    x = t.randn(6, requires_grad=True)
    y = (t.randn(6).abs() + 0.5).requires_grad_(True)
    g = t.randn(6)
    # sub reference
    s = x - y
    s.backward(g, retain_graph=True)
    assert t.allclose(BACK[('sub', 0)](g, s.detach(), x.detach(), y.detach()), x.grad, atol=1e-5)
    assert t.allclose(BACK[('sub', 1)](g, s.detach(), x.detach(), y.detach()), y.grad, atol=1e-5)
    # div reference (fresh grads)
    x.grad = None; y.grad = None
    d = x / y
    d.backward(g)
    assert t.allclose(BACK[('div', 0)](g, d.detach(), x.detach(), y.detach()), x.grad, atol=1e-5)
    assert t.allclose(BACK[('div', 1)](g, d.detach(), x.detach(), y.detach()), y.grad, atol=1e-5)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
