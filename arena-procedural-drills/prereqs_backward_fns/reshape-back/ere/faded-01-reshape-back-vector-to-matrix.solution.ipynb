Reshape is a pure permutation of storage, so its backward is the inverse reshape: send `grad_out` back to the original `x.shape`. The forward's target shape is irrelevant in backward — `x.shape` is what you restore.

## Faded exercise 1

The forward is `out = x.reshape(3, 4)` for a `(12,)` input `x`. Complete `reshape_back(grad_out, out, x, new_shape)` by returning the gradient reshaped back to the input's original shape.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
Tensor = t.Tensor


def reshape_back(grad_out: Tensor, out: Tensor, x: Tensor, new_shape: tuple) -> Tensor:
    return grad_out.reshape(x.shape)


t.manual_seed(0)
x = t.randn(12)
out = x.reshape(3, 4)
grad_out = t.arange(12, dtype=t.float32).reshape(3, 4)
grad_x = reshape_back(grad_out, out, x, (3, 4))

def _test():
    t.manual_seed(0)
    x = t.randn(12)
    out = x.reshape(3, 4)
    grad_out = t.arange(12, dtype=t.float32).reshape(3, 4)
    grad_x = reshape_back(grad_out, out, x, (3, 4))
    assert tuple(grad_x.shape) == (12,), 'grad must match x.shape (12,)'
    # independent ground truth via autograd
    xg = x.clone().requires_grad_(True)
    xg.reshape(3, 4).backward(grad_out)
    assert t.allclose(grad_x, xg.grad), 'must match autograd gradient'

try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
