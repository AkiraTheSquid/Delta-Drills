## Faded exercise 2

The forward squeezes `x` of shape `(5, 1)` to `(5,)`. Complete the body of `reshape_back` so the gradient is restored to the input's `(5, 1)` shape.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
Tensor = t.Tensor


def reshape_back(grad_out: Tensor, out: Tensor, x: Tensor, new_shape: tuple) -> Tensor:
    result = grad_out.reshape(x.shape)
    return result


t.manual_seed(3)
x = t.randn(5, 1)
out = x.reshape(5)
grad_out = t.randn(5)
grad_x = reshape_back(grad_out, out, x, (5,))

def _test():
    t.manual_seed(3)
    x = t.randn(5, 1)
    grad_out = t.randn(5)
    grad_x = reshape_back(grad_out, x.reshape(5), x, (5,))
    assert tuple(grad_x.shape) == (5, 1), 'must restore the squeezed axis'
    xg = x.clone().requires_grad_(True)
    xg.reshape(5).backward(grad_out)
    assert t.allclose(grad_x, xg.grad), 'must match autograd'

try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
