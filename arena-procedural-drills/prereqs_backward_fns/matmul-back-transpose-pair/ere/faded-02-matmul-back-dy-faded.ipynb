## Concept

For `out = x @ y` with `x: (m, k)` and `y: (k, n)`, the gradient w.r.t. `y` is `dL/dy = x.T @ grad_out`. The shape reasoning: `dL/dy` must have shape `(k, n)`, and `x.T: (k, m)` times `grad_out: (m, n)` gives `(k, n)`. The transpose lives on the input you are NOT differentiating w.r.t. — here `x` is transposed because we want the gradient w.r.t. `y`.
