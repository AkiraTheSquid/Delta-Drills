## Concept

For batched matmul `out = x @ y` with `x: (B, m, k)` and `y: (B, k, n)`, the gradient w.r.t. `x` is `dL/dx = grad_out @ y.transpose(-1, -2)`. Using `.transpose(-1, -2)` instead of `.T` is essential: `.T` on a 3-D tensor reverses all axes, destroying the batch dimension, while `.transpose(-1, -2)` swaps only the last two axes.
