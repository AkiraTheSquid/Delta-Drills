## Concept

For `out = x @ y` with `x: (m, k)` and `y: (k, n)`, the gradient w.r.t. `x` is `dL/dx = grad_out @ y.T`. This is derived from the shape constraint: `dL/dx` must have shape `(m, k)`, and the only valid matmul of `grad_out: (m, n)` with a transformation of `y` that produces `(m, k)` is `(m, n) @ (n, k) = grad_out @ y.T`.

## Faded exercise 1

Implement `matmul_back_x` below. Fill in the single blanked line that computes `dL/dx`.

**Fill in:** Compute dL/dx as grad_out @ y.T using the matmul transpose-pair rule.

````
import torch as t

def matmul_back_x(grad_out, x, y):
    dL_dx = grad_out @ y.T
    return dL_dx

# Exercise it
t.manual_seed(57)
m, k, n = 3, 4, 5
x = t.randn(m, k)
y = t.randn(k, n)
grad_out = t.randn(m, n)
dL_dx = matmul_back_x(grad_out, x, y)
print(f"dL/dx shape: {dL_dx.shape}  (expected ({m},{k}))")


import torch as t

def _test():
    t.manual_seed(57)
    m, k, n = 3, 4, 5
    x = t.randn(m, k)
    y = t.randn(k, n)
    grad_out = t.randn(m, n)

    dL_dx = matmul_back_x(grad_out, x, y)

    assert dL_dx.shape == (m, k), f"Expected ({m},{k}), got {dL_dx.shape}"

    # Verify against autograd
    x_ag = x.clone().requires_grad_(True)
    (x_ag @ y).backward(grad_out)
    assert t.allclose(dL_dx, x_ag.grad, atol=1e-5)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
