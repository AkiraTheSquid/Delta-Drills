import torch as t
import numpy as np

def permute_back(grad_out, out, x, dims):
    inverse = tuple(int(i) for i in np.argsort(dims))
    return grad_out.permute(*inverse)

def assert_round_trip(x: t.Tensor, dims: tuple):
    y = x.permute(*dims)
    recovered = permute_back(y, y, x, dims)
    assert recovered.shape == x.shape, f'shape mismatch: {recovered.shape} != {x.shape}'
    assert t.equal(recovered, x), f'value mismatch on dims={dims}'

t.manual_seed(4)
x = t.randn(2, 5, 3)
assert_round_trip(x, (1, 2, 0))
print('Round-trip passed.')
