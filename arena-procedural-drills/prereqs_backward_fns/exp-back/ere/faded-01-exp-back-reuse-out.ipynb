For the operation `y = exp(x)`, the gradient with respect to `x` is `exp(x)` multiplied by the upstream gradient. Because `exp(x) = out` (the forward output), the backward function uses the cached `out` directly rather than recomputing the exponential. This is one of the simplest and most important examples of caching forward values for efficient backpropagation.
