Subtraction `out = a - b` can be seen as `a + (-b)`. The gradient to `a` is `grad_out`; the gradient to `b` reuses the negate rule, giving `-grad_out`. This reuses `negative_back` as the building block for the second operand.

## Faded exercise 3

Implement `sub_back_b(grad_out, a, b)` returning the gradient with respect to `b` in `out = a - b`, by applying the negate backward rule. Fill in the gradient expression.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

t.manual_seed(5)

def negative_back(grad_out, out, x):
    return -grad_out

def sub_back_b(grad_out, a, b):
    out = a - b
    grad_b = negative_back(grad_out, out, b)
    return grad_b

a = t.randn(2, 3)
b = t.randn(2, 3)
grad_out = t.randn(2, 3)
print(sub_back_b(grad_out, a, b))


def _test():
    t.manual_seed(11)
    a = t.randn(3, 4, requires_grad=True)
    b = t.randn(3, 4, requires_grad=True)
    out = a - b
    g = t.randn(3, 4)
    out.backward(g)
    # independent ground truth from autograd
    manual = sub_back_b(g, a.detach(), b.detach())
    assert manual.shape == b.shape
    assert t.allclose(manual, b.grad)
    # and a is the un-flipped grad
    assert t.allclose(a.grad, g)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
