## Concept

For the forward op `out = -x`, the local derivative is `-1` at every element. By the elementwise chain rule, the incoming gradient is simply sign-flipped: `dL/dx = grad_out * (-1) = -grad_out`. The result has exactly the same shape and dtype as `x`.

## Worked solution

We implement `negative_back(grad_out, out, x)` for `out = -x`. Differentiating `-x` with respect to `x` gives `-1` everywhere, so the chain rule collapses to multiplying the upstream gradient by `-1`. That is just `-grad_out`, returned as a fresh tensor with no in-place mutation. To exercise it, we seed, draw a `(2, 3)` matrix `x`, compute the forward `out = -x`, draw an upstream gradient of the same shape, and call `negative_back`. We then verify against PyTorch autograd by recomputing `(-x).sum()` style gradients implicitly: the hand-computed `grad_in` should equal `-grad_out` exactly, which we print.

```
import torch as t

t.manual_seed(0)

def negative_back(grad_out, out, x):
    # d/dx(-x) = -1 elementwise -> grad_in = -grad_out
    return -grad_out

x = t.randn(2, 3)
out = -x
grad_out = t.randn(2, 3)
grad_in = negative_back(grad_out, out, x)
print(grad_in)
print('matches -grad_out:', t.equal(grad_in, -grad_out))
```
