## Concept

Interval logging uses `step % log_every == 0` checked AFTER the counter tick, so logged step values are clean multiples of `log_every` (10, 20, 30...) rather than off-by-one (9, 19, 29...). The tick-before-check order is what makes the boundary land correctly.

## Worked solution

We tick the counter every batch but only append to the log when `step % log_every == 0`. With `log_every=3` over 7 losses, the gate fires at steps 3 and 6, so the log has exactly two entries with step values 3 and 6. The 7th step does not log because 7 is not a multiple of 3. We print the log to confirm the step values are multiples of `log_every`, demonstrating the tick-then-check ordering.

```
def train_log_every(losses, log_every):
    step = 0
    log = []
    for loss in losses:
        step += 1
        if step % log_every == 0:
            log.append((step, loss))
    return step, log


final_step, log = train_log_every([0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3], log_every=3)
print('final step:', final_step)
print('logged at steps:', [s for s, _ in log])
```
