## Faded exercise 1

Complete `train_epoch`. The loss is recorded and the log append happens after. Fill in the step increment so the counter ticks once per batch before the log append.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def train_epoch(losses, start_step):
    step = start_step
    log = []
    for loss in losses:
        step += 1
        log.append((step, loss))
    return step, log


final_step, log = train_epoch([0.9, 0.7, 0.5], start_step=0)

def _test():
    final_step, log = train_epoch([0.9, 0.7, 0.5], start_step=0)
    assert final_step == 3, f'3 batches from 0 -> final step 3, got {final_step}'
    assert [s for s, _ in log] == [1, 2, 3], f'steps must be 1,2,3 got {[s for s,_ in log]}'
    # accumulation: starting at 10 over 2 batches -> 12
    fs2, log2 = train_epoch([0.1, 0.2], start_step=10)
    assert fs2 == 12 and [s for s, _ in log2] == [11, 12]

try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
