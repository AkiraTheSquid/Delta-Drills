Passing the counter back in as `start_step` for the next epoch makes the step monotonically increasing across the whole run, not resetting each epoch. The final step after E epochs of B batches each is `E*B`.

## Worked solution

We run two epochs by threading the returned `final_step` of epoch 1 into the `start_step` of epoch 2. Epoch 1 has 2 batches (steps 1, 2); epoch 2 starts at 2 and produces steps 3, 4. The combined log shows a continuous step axis across epoch boundaries. We print the per-epoch logs and the global final step (4), demonstrating accumulation rather than per-epoch reset.

```
def train_epoch(losses, start_step):
    step = start_step
    log = []
    for loss in losses:
        step += 1
        log.append((step, loss))
    return step, log


step = 0
all_log = []
for epoch_losses in [[0.8, 0.6], [0.5, 0.4]]:
    step, log = train_epoch(epoch_losses, start_step=step)
    all_log.extend(log)
print('global final step:', step)
print('all log:', all_log)
```
