## Best-val-loss checkpoint — track best, snapshot state_dict

Ex1 implemented `fit`+`validate`+`_step`. Ex2 added a callback list. The deepening move adds the SINGLE most-common Trainer extension: remember the best validation loss seen so far, and snapshot the model's `state_dict()` when it improves.

```python
self.best_val_loss = float('inf')
self.best_state    = None    # snapshot dict (or None if never improved)

def validate(self):
    # ... compute val_loss ...
    self.history['val_loss'].append(val_loss)
    if val_loss < self.best_val_loss:
        self.best_val_loss = val_loss
        # DEEP-COPY: state_dict() returns LIVE references; we need a snapshot.
        self.best_state = {k: v.detach().clone() for k, v in self.model.state_dict().items()}
```

**Why deep-copy.** `model.state_dict()` returns a dict of LIVE tensors — the same tensors that `optimizer.step()` is about to mutate. Saving the dict directly would mean `best_state` updates IN LOCKSTEP with the training, defeating the whole point. `.detach().clone()` per tensor produces an independent snapshot.

**`best_state = None` sentinel.** Before the first validate call, no snapshot has been taken. Returning `None` is cleaner than 'snapshot of the random init' — callers can check `if trainer.best_state is None: ...`.

**Strict `<`, not `<=`.** Use strict less-than for 'improved'. With `<=`, ties would overwrite the snapshot needlessly and you'd snapshot the LATER tied model rather than the EARLIER one (which is usually preferred — earlier means less overfit).

````python
class Ex3TrainerWithCheckpoint:
    def __init__(self, model, optimizer, train_loader, val_loader, loss_fn):
        raise NotImplementedError()

    def _step(self, x, y):
        raise NotImplementedError()

    def fit(self, n_epochs: int):
        raise NotImplementedError()

    def validate(self):
        raise NotImplementedError()


def _test_ex3():
    from torch.utils.data import TensorDataset, DataLoader
    import math

    # === Tiny regression task: y = 2x + 1 ===
    t.manual_seed(0)
    N = 64
    x_train = t.randn(N, 1)
    y_train = 2.0 * x_train + 1.0 + 0.05 * t.randn(N, 1)
    x_val = t.randn(16, 1)
    y_val = 2.0 * x_val + 1.0

    train_loader = DataLoader(TensorDataset(x_train, y_train), batch_size=8, shuffle=True)
    val_loader = DataLoader(TensorDataset(x_val, y_val), batch_size=8, shuffle=False)

    model = nn.Linear(1, 1)
    opt = t.optim.SGD(model.parameters(), lr=0.1)
    loss_fn = nn.MSELoss()

    trainer = Ex3TrainerWithCheckpoint(model, opt, train_loader, val_loader, loss_fn)

    # === Attributes ===
    assert trainer.best_val_loss == float('inf'), (
        f'best_val_loss must init to inf; got {trainer.best_val_loss}'
    )
    assert trainer.best_state is None, (
        f'best_state must init to None; got {trainer.best_state}'
    )
    assert trainer.step == 0
    assert trainer.history == {'train_loss': [], 'val_loss': []}

    # === Fit 4 epochs — best_state must populate ===
    trainer.fit(n_epochs=4)
    assert trainer.best_state is not None, 'after fit, best_state must be a dict snapshot'
    assert isinstance(trainer.best_state, dict)
    assert 'weight' in trainer.best_state and 'bias' in trainer.best_state, (
        f'best_state should mirror state_dict keys; got keys {list(trainer.best_state.keys())}'
    )

    # === best_val_loss is the min of the recorded val losses ===
    val_losses = trainer.history['val_loss']
    assert len(val_losses) == 4, f'4 epochs → 4 val losses; got {len(val_losses)}'
    assert abs(trainer.best_val_loss - min(val_losses)) < 1e-9, (
        f'best_val_loss must equal min(val_loss history): {trainer.best_val_loss} vs {min(val_losses)}'
    )

    # === best_state corresponds to the epoch where val_loss was minimal ===
    # For a converging regression run, the best epoch should be the LAST one.
    # But the snapshot must match the model state AT that epoch — independent of
    # further training. We test independence below.

    # === Snapshot is INDEPENDENT: mutate model post-fit, snapshot unchanged ===
    snapshot_w = trainer.best_state['weight'].clone()
    snapshot_b = trainer.best_state['bias'].clone()
    # Mutate the model:
    with t.no_grad():
        model.weight.fill_(999.0)
        model.bias.fill_(-999.0)
    # best_state should NOT have changed:
    assert t.equal(trainer.best_state['weight'], snapshot_w), (
        f'snapshot must be independent of post-fit mutations'
    )
    assert t.equal(trainer.best_state['bias'], snapshot_b)

    # === Snapshot tensors are DETACHED ===
    for k, v in trainer.best_state.items():
        assert isinstance(v, t.Tensor), f'snapshot entry {k} must be tensor; got {type(v)}'
        assert not v.requires_grad, f'snapshot tensor {k} must be detached (no grad-track)'

    # === Strict-improvement: a worse val loss does NOT overwrite best_state ===
    # Manually trigger a worse epoch: synthetic best_state should remain.
    t.manual_seed(123)
    model2 = nn.Linear(1, 1)
    # Make model2 fit poorly: huge initial weight.
    with t.no_grad():
        model2.weight.fill_(50.0)
        model2.bias.fill_(0.0)
    opt2 = t.optim.SGD(model2.parameters(), lr=0.0)  # lr=0 → no learning → val loss flat
    trainer2 = Ex3TrainerWithCheckpoint(model2, opt2, train_loader, val_loader, loss_fn)
    trainer2.fit(n_epochs=3)
    vls = trainer2.history['val_loss']
    # All val losses should be (essentially) identical since lr=0.
    assert max(vls) - min(vls) < 1e-3, f'lr=0 → val loss constant; got {vls}'
    # best_state was set on epoch 0 and never overwritten (subsequent epochs are NOT strictly less).
    assert trainer2.best_state is not None
    assert abs(trainer2.best_val_loss - vls[0]) < 1e-6, (
        f'best_val_loss should equal epoch-0 val_loss (first improvement); '
        f'got {trainer2.best_val_loss} vs {vls[0]}'
    )

    # === Sanity: validate is run via fit ===
    model3 = nn.Linear(1, 1)
    opt3 = t.optim.SGD(model3.parameters(), lr=0.1)
    trainer3 = Ex3TrainerWithCheckpoint(model3, opt3, train_loader, val_loader, loss_fn)
    assert trainer3.best_state is None
    trainer3.fit(n_epochs=1)
    assert trainer3.best_state is not None, 'after 1 epoch validate must have run'
    assert len(trainer3.history['val_loss']) == 1
    _dd_passed.add('ex3')
    print("ex3 ✓")

_test_ex3()
````

```python
class Ex3TrainerWithCheckpoint:
    def __init__(self, model, optimizer, train_loader, val_loader, loss_fn):
        self.model = model
        self.optimizer = optimizer
        self.train_loader = train_loader
        self.val_loader = val_loader
        self.loss_fn = loss_fn
        self.step = 0
        self.history = {'train_loss': [], 'val_loss': []}
        self.best_val_loss = float('inf')
        self.best_state = None

    def _step(self, x, y):
        return self.loss_fn(self.model(x), y)

    def fit(self, n_epochs):
        for _epoch in range(n_epochs):
            self.model.train()
            for x, y in self.train_loader:
                loss = self._step(x, y)
                loss.backward()
                self.optimizer.step()
                self.optimizer.zero_grad()
                self.step += 1
                self.history['train_loss'].append(loss.item())
            self.validate()

    def validate(self):
        self.model.eval()
        total = 0.0
        count = 0
        with t.inference_mode():
            for x, y in self.val_loader:
                loss = self.loss_fn(self.model(x), y)
                total += loss.item() * x.shape[0]
                count += x.shape[0]
        val_loss = total / count
        self.history['val_loss'].append(val_loss)
        if val_loss < self.best_val_loss:
            self.best_val_loss = val_loss
            self.best_state = {
                k: v.detach().clone() for k, v in self.model.state_dict().items()
            }
```

**`.detach().clone()` is the canonical snapshot idiom.** `clone()` alone preserves the autograd graph link; `detach()` alone shares storage with the live model. The combo gives an independent tensor that is NOT mutated by subsequent `optimizer.step()` calls.

**Strict `<` not `<=`.** With `<=`, ties on val loss overwrite the snapshot. The convention is 'keep the EARLIER best' — earlier means less overfit. Strict less-than enforces that.

**Sentinel `None` is cleaner than 'init from random init'.** Before the first `validate()` call, no real evaluation has happened. Storing a random-init snapshot would lie about 'best'; `None` lets callers detect 'no validation has run yet'.
