## Faded exercise 2

Implement `train_and_log(losses, batch_size, wandb)`. For each loss in `losses` (enumerate from step=1), compute `examples_seen = step * batch_size` and call `wandb.log({'loss': loss, 'examples_seen': examples_seen})`. Return the total number of log calls.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def train_and_log(losses, batch_size: int, wandb):
    n_calls = 0
    for step, loss in enumerate(losses, start=1):
        examples_seen = step * batch_size
        wandb.log({'loss': loss, 'examples_seen': examples_seen})
        n_calls += 1
    return n_calls


def _test():
    class MockWandb:
        def __init__(self):
            self.calls = []
        def log(self, payload):
            self.calls.append(dict(payload))
    w = MockWandb()
    losses = [1.0, 0.8, 0.6]
    n = train_and_log(losses, batch_size=16, wandb=w)
    assert n == 3
    assert len(w.calls) == 3
    for i, c in enumerate(w.calls):
        assert 'loss' in c
        assert 'examples_seen' in c
        assert c['examples_seen'] == (i + 1) * 16


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
