Training logs recorded as `(step, loss)` pairs are not directly comparable across runs with different batch sizes. Converting them to `(examples_seen, loss)` where `examples_seen = step * batch_size` makes curves from different batch-size configurations overlap correctly on the same plot.

## Faded exercise 1

Implement `convert_to_examples_seen(step_loss_history, batch_size)` that takes a list of `(step, loss)` tuples and returns a list of `(examples_seen, loss)` tuples where `examples_seen = step * batch_size`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def convert_to_examples_seen(step_loss_history, batch_size: int):
    return [(step * batch_size, loss) for step, loss in step_loss_history]


def _test():
    history = [(1, 2.0), (2, 1.5), (3, 1.2)]
    result = convert_to_examples_seen(history, batch_size=32)
    assert result == [(32, 2.0), (64, 1.5), (96, 1.2)]
    result2 = convert_to_examples_seen(history, batch_size=64)
    assert result2 == [(64, 2.0), (128, 1.5), (192, 1.2)]
    # empty
    assert convert_to_examples_seen([], 16) == []


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
