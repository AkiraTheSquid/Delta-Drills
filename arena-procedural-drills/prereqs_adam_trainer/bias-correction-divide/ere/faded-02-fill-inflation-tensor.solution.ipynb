The bias-correction inflation factor `1/(1 - beta**t)` for steps `t=1..n` can be built as one tensor. With step indices in a float tensor, `beta ** steps` exponentiates elementwise, and the reciprocal of `1 - beta**steps` gives the monotonically decreasing schedule that asymptotes to 1.

## Faded exercise 2

### Faded — vectorized inflation factors

Implement `inflation_factors(beta, n, dtype)` returning a length-`n` tensor whose `i`-th entry is `1/(1 - beta**(i+1))`. The 1-based step indices are already created for you; complete the line that turns them into the inflation-factor tensor.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def inflation_factors(beta, n, dtype=t.float64):
    steps = t.arange(1, n + 1, dtype=dtype)
    factors = 1.0 / (1.0 - beta ** steps)
    return factors


def _test():
    beta, n = 0.9, 6
    out = inflation_factors(beta, n)
    steps = t.arange(1, n + 1, dtype=t.float64)
    expected = 1.0 / (1.0 - beta ** steps)
    assert out.shape == (n,)
    assert t.allclose(out, expected, atol=1e-10), (out, expected)
    # monotonically decreasing toward 1
    assert bool(t.all(out[1:] < out[:-1]))
    assert out[-1] > 1.0


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
