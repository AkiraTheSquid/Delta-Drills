## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Adam's second moment `v` is a zero-initialized EMA with decay `beta2`, so it is biased low early in training. The unbiased estimate is `v_hat = v / (1 - beta2 ** t)`, where `t` is the 1-based step. The denominator is a Python scalar and grows toward 1 as `t` increases.

## Faded exercise 1

### Faded — second-moment bias correction

Implement `bias_correct_v(v, beta2, t_step)` which returns the bias-corrected second moment `v_hat`. The surrounding function and the test harness are already written; you only need to fill in the correction-factor computation.

Return a NEW tensor — do not mutate `v` in place.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def bias_correct_v(v, beta2, t_step):
    correction = 1 - beta2 ** t_step
    return v / correction


def _test():
    v = t.tensor([0.002, 0.004, 0.006])
    beta2, t_step = 0.999, 2
    out = bias_correct_v(v, beta2, t_step)
    expected = v / (1 - beta2 ** t_step)
    assert out.shape == v.shape
    assert t.allclose(out, expected, atol=1e-8), (out, expected)
    # raw v unchanged (no in-place mutation)
    assert t.allclose(v, t.tensor([0.002, 0.004, 0.006]))
    # corrected magnitude exceeds raw (inflation)
    assert bool(t.all(out.abs() >= v.abs()))


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
