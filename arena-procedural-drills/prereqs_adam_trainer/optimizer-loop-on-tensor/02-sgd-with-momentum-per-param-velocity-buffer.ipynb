## SGD with momentum — explicit per-param loop + per-param buffer — quick refresher

Plain SGD's `p -= lr * p.grad` becomes SGD-with-momentum by introducing a per-parameter velocity buffer:
```
for p, b in zip(self.params, self.buffers):
    if p.grad is None:
        continue
    b.copy_(mu * b + p.grad)            # accumulate velocity in place
    p -= self.lr * b                     # step along velocity
```
Three differences from plain SGD:
1. Each parameter gets a `zeros_like(p)` buffer at init.
2. The loop carries one extra `copy_` per param (velocity recurrence).
3. The update reads from `b`, not directly from `p.grad`.

Default momentum `mu = 0.9` matches PyTorch's `torch.optim.SGD` default. The `None`-grad guard and `@t.inference_mode()` decorator carry over unchanged from plain SGD.

Implement `Ex2MomentumSGD`. A hand-rolled SGD with momentum.

1. `__init__(self, params, lr, momentum=0.9)`:
   - Materialize `self.params = list(params)`.
   - Store `self.lr = lr`, `self.momentum = momentum`.
   - Allocate `self.bufs = [t.zeros_like(p) for p in self.params]`.
2. `step(self)` decorated with `@t.inference_mode()`. For each `(p, b)` in `zip(self.params, self.bufs)`:
   - If `p.grad is None`, SKIP.
   - Else: `b.copy_(self.momentum * b + p.grad)`, then `p -= self.lr * b`.
3. `zero_grad(self)`: `p.grad = None` for every param.

````python
class Ex2MomentumSGD:
    """Hand-rolled SGD with momentum — per-param loop + velocity buffer."""

    def __init__(self, params, lr: float, momentum: float = 0.9):
        raise NotImplementedError()

    def step(self):
        raise NotImplementedError()

    def zero_grad(self):
        raise NotImplementedError()


def _test_ex2():
    # === Single-param sanity ===
    p = t.nn.Parameter(t.tensor([10.0, 20.0, 30.0]))
    p.grad = t.tensor([1.0, 1.0, 1.0])
    opt = Ex2MomentumSGD([p], lr=0.1, momentum=0.9)

    # Buffer init checks.
    assert isinstance(opt.bufs, list) and len(opt.bufs) == 1
    assert opt.bufs[0].shape == p.shape
    assert t.all(opt.bufs[0] == 0)
    buf_id_before = id(opt.bufs[0])
    buf_ptr_before = opt.bufs[0].data_ptr()

    # Step 1: b = 0.9*0 + g = [1,1,1]; p -= 0.1 * [1,1,1] = -0.1 each.
    opt.step()
    assert t.allclose(p.detach(), t.tensor([9.9, 19.9, 29.9])), p.detach()
    assert t.allclose(opt.bufs[0], t.tensor([1.0, 1.0, 1.0]))
    assert id(opt.bufs[0]) == buf_id_before, 'buf must not be rebound'
    assert opt.bufs[0].data_ptr() == buf_ptr_before, 'buf storage must not be reallocated'

    # Step 2: same grad, b = 0.9 * 1 + 1 = 1.9; p -= 0.1 * 1.9 = -0.19.
    p.grad = t.tensor([1.0, 1.0, 1.0])
    opt.step()
    assert t.allclose(opt.bufs[0], t.tensor([1.9, 1.9, 1.9]))
    assert t.allclose(p.detach(), t.tensor([9.71, 19.71, 29.71]), atol=1e-6), p.detach()

    # === None-grad guard ===
    p_frozen = t.nn.Parameter(t.tensor([7.0, 7.0]))
    p_active = t.nn.Parameter(t.tensor([1.0, 2.0]))
    p_active.grad = t.tensor([10.0, 10.0])
    assert p_frozen.grad is None
    opt2 = Ex2MomentumSGD([p_frozen, p_active], lr=0.01, momentum=0.5)
    opt2.step()         # must not raise on None-grad
    # p_frozen should be untouched AND its buf should still be zero.
    assert t.allclose(p_frozen.detach(), t.tensor([7.0, 7.0]))
    assert t.all(opt2.bufs[0] == 0), f'frozen param buf should stay zero; got {opt2.bufs[0]}'
    # p_active moved.
    expected_buf_a = 0.5 * 0 + t.tensor([10.0, 10.0])
    assert t.allclose(opt2.bufs[1], expected_buf_a)
    assert t.allclose(p_active.detach(), t.tensor([0.9, 1.9]), atol=1e-6)

    # === zero_grad clears all grads ===
    opt2.zero_grad()
    for q in opt2.params:
        assert q.grad is None

    # === Match torch.optim.SGD with momentum over 10 steps ===
    t.manual_seed(0)
    ref_model = t.nn.Linear(3, 1)
    our_model = t.nn.Linear(3, 1)
    # Copy initial weights from ref to ours (so they start identical).
    with t.no_grad():
        our_model.weight.copy_(ref_model.weight)
        our_model.bias.copy_(ref_model.bias)

    ref_opt = t.optim.SGD(ref_model.parameters(), lr=0.05, momentum=0.9)
    our_opt = Ex2MomentumSGD(our_model.parameters(), lr=0.05, momentum=0.9)

    X = t.randn(16, 3, generator=t.Generator().manual_seed(7))
    Y = t.randn(16, 1, generator=t.Generator().manual_seed(8))
    for step_i in range(10):
        # Reference step.
        ref_opt.zero_grad()
        loss_ref = ((ref_model(X) - Y) ** 2).mean()
        loss_ref.backward()
        ref_opt.step()
        # Our step.
        our_opt.zero_grad()
        loss_our = ((our_model(X) - Y) ** 2).mean()
        loss_our.backward()
        our_opt.step()
        # Parameters must match to within float tolerance.
        assert t.allclose(ref_model.weight.detach(), our_model.weight.detach(), atol=1e-5), (
            f'step {step_i}: weight mismatch vs torch.optim.SGD'
        )
        assert t.allclose(ref_model.bias.detach(), our_model.bias.detach(), atol=1e-5), (
            f'step {step_i}: bias mismatch'
        )
    _dd_passed.add('ex2')
    print("ex2 ✓")

_test_ex2()
````

<details><summary>Solution</summary>

```python
class Ex2MomentumSGD:
    def __init__(self, params, lr, momentum=0.9):
        self.params = list(params)
        self.lr = lr
        self.momentum = momentum
        self.bufs = [t.zeros_like(p) for p in self.params]

    @t.inference_mode()
    def step(self):
        for p, b in zip(self.params, self.bufs):
            if p.grad is None:
                continue
            b.copy_(self.momentum * b + p.grad)
            p -= self.lr * b

    def zero_grad(self):
        for p in self.params:
            p.grad = None
```

**Why a buffer LIST parallel to params.** Each Parameter has a different shape, so the buffers can't be one big contiguous tensor (without flattening). The parallel-list pattern is what PyTorch's `torch.optim.SGD` does internally too — `state[p]` is a dict keyed by Parameter that stores the velocity buffer.

**Why `b.copy_(...)` not `b = ...`.** Identical issue to EMA-first-moment ex1: rebinding the local loop variable doesn't update the list entry. The next step would read a stale zero buffer and behave like plain SGD with no momentum.

**Match to torch.optim.SGD.** PyTorch's default formula is `v_new = momentum*v + g` (no `(1-mu)` factor — this is the classical-momentum form, not the EMA form). Our impl matches exactly, which is why the trajectories agree to 1e-5 over 10 steps. If you wanted the EMA-style `v_new = momentum*v + (1-momentum)*g` you'd need a different lr scale — that's Adam's first moment, not SGD.
