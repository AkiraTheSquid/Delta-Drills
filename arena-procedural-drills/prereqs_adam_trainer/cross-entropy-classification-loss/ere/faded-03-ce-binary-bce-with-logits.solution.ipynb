## Concept

For binary classification you can either run 2-class `F.cross_entropy` on `(B, 2)` logits, or collapse to a single logit (the difference of the two class logits) and use `F.binary_cross_entropy_with_logits`. Both are numerically stable because they work in logit space. The single-logit form needs the float label of the positive class.

## Faded exercise 3

### Faded — two-class cross-entropy as BCE-with-logits

Implement `binary_ce(logits2, labels)` where `logits2` is `(B, 2)`. Collapse it to a single logit (`logit_class1 - logit_class0`) and score it with `F.binary_cross_entropy_with_logits` so the result matches 2-class `F.cross_entropy(logits2, labels)`. The collapse and the float-label conversion are given. Complete the ONE step that computes the BCE-with-logits loss.

Inputs:
- `logits2`: `(B, 2)` raw scores.
- `labels`: `(B,)` int64 in {0, 1}.

Output: scalar loss matching `F.cross_entropy(logits2, labels)` to 1e-5.

**Fill in:** computes the binary-cross-entropy-with-logits loss from the single collapsed logit and the float positive-class label

````
import torch.nn.functional as F

def binary_ce(logits2, labels):
    single_logit = logits2[:, 1] - logits2[:, 0]   # (B,)
    float_labels = labels.float()                  # (B,)
    loss = F.binary_cross_entropy_with_logits(single_logit, float_labels)
    return loss


import torch.nn.functional as F

def _test():
    t.manual_seed(0)
    logits2 = t.randn(9, 2)
    labels = t.randint(0, 2, (9,))
    mine = binary_ce(logits2, labels)
    ref = F.cross_entropy(logits2, labels)
    assert mine.shape == (), f"expected scalar, got {mine.shape}"
    assert t.allclose(mine, ref, atol=1e-5), f"mine={mine.item()} ref={ref.item()}"


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
