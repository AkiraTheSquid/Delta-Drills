## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Unboxing replaces each `MiniTensor` arg with its `.array` and passes everything else through. The decision must use `isinstance(a, MiniTensor)`, not duck-typing on `.array`, because unrelated objects (e.g. numpy arrays) may expose an `.array` attribute and would be wrongly unwrapped.

## Faded exercise 1

### Unbox positional args

Implement `unbox_args(args)` returning a new tuple where each `MiniTensor` is replaced by its `.array` and all other values pass through unchanged, in order. Complete the per-element expression inside the comprehension.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
class MiniTensor:
    def __init__(self, array):
        self.array = array

def unbox_args(args: tuple) -> tuple:
    return tuple(a.array if isinstance(a, MiniTensor) else a for a in args)

m = MiniTensor([1, 2])
print(unbox_args((m, 5, 'x')))


def _test():
    m1 = MiniTensor([1, 2, 3])
    m2 = MiniTensor('payload')
    out = unbox_args((m1, 3.0, m2, 'x'))
    assert isinstance(out, tuple)
    assert len(out) == 4
    # tensors unboxed to the exact .array object
    assert out[0] is m1.array
    assert out[2] is m2.array
    # non-tensors pass through unchanged
    assert out[1] == 3.0
    assert out[3] == 'x'
    # empty input -> empty tuple
    assert unbox_args(()) == ()


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
