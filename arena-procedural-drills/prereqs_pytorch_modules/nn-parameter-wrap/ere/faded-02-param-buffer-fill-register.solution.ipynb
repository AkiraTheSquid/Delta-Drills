A non-learnable running statistic should be registered with `register_buffer`, not as a Parameter. Buffers appear in `state_dict` and move with `.to(device)`, but are not optimized.

## Faded exercise 2

Complete `RunningScaler` so `running_mean` is registered as a buffer of zeros. Fill in the buffer registration call.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

class RunningScaler(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.weight = nn.Parameter(t.ones(dim))
        self.register_buffer('running_mean', t.zeros(dim))
    def forward(self, x):
        return (x - self.running_mean) * self.weight

mod = RunningScaler(3)
print(list(mod.state_dict().keys()))


def _test():
    mod = RunningScaler(3)
    param_names = {n for n, _ in mod.named_parameters()}
    buffer_names = {n for n, _ in mod.named_buffers()}
    assert 'weight' in param_names
    assert 'running_mean' in buffer_names
    assert 'running_mean' not in param_names  # buffer, not parameter
    assert 'running_mean' in mod.state_dict()  # but in state_dict


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
