## Concept

Passing `requires_grad=False` to `nn.Parameter` registers a frozen weight: visible in `.parameters()` and `.state_dict()`, but autograd leaves its `.grad` as `None`.

## Faded exercise 3

Complete `FrozenBias` so `bias` is a frozen Parameter (requires_grad False). Fill in the bias assignment.

**Fill in:** the bias assignment as a frozen nn.Parameter with requires_grad=False

````
import torch as t
import torch.nn as nn

class FrozenBias(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.gain = nn.Parameter(t.ones(dim))
        self.bias = nn.Parameter(t.zeros(dim), requires_grad=False)
    def forward(self, x):
        return x * self.gain + self.bias

mod = FrozenBias(3)
mod(t.randn(3, requires_grad=True)).sum().backward()
print('bias.grad is None:', mod.bias.grad is None)


def _test():
    mod = FrozenBias(3)
    assert mod.bias.requires_grad is False
    assert mod.gain.requires_grad is True
    # both visible despite freezing
    names = {n for n, _ in mod.named_parameters()}
    assert names == {'gain', 'bias'}
    x = t.randn(3, requires_grad=True)
    mod(x).sum().backward()
    assert mod.bias.grad is None  # frozen -> never touched
    assert mod.gain.grad is not None


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
