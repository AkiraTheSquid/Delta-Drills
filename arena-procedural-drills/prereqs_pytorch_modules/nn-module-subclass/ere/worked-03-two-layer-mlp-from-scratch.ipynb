## Concept

A typical Module wires children and weights in `__init__` (after `super().__init__()`) and composes them in `forward`. Passing the weights in lets us check the forward output by hand rather than worrying about initialization.

## Worked solution

We build `TwoLayerNet` whose `__init__` takes two weight tensors and two bias tensors. After `super().__init__()` we wrap each as an `nn.Parameter` so they register. The `forward` applies the first affine map, a ReLU nonlinearity, then the second affine map, following the `nn.Linear` convention `x @ W.T + b`. We seed, construct small explicit weights, run a single input through, and print the output shape and values, confirming the two-layer composition produces the expected `(batch, out)` shape.

```
import torch as t
import torch.nn as nn

t.manual_seed(0)

class TwoLayerNet(nn.Module):
    def __init__(self, w1, b1, w2, b2):
        super().__init__()
        self.w1 = nn.Parameter(w1)
        self.b1 = nn.Parameter(b1)
        self.w2 = nn.Parameter(w2)
        self.b2 = nn.Parameter(b2)
    def forward(self, x):
        h = t.relu(x @ self.w1.T + self.b1)
        return h @ self.w2.T + self.b2

w1, b1 = t.randn(5, 3), t.randn(5)
w2, b2 = t.randn(2, 5), t.randn(2)
net = TwoLayerNet(w1, b1, w2, b2)
x = t.randn(4, 3)
out = net(x)
print('output shape:', tuple(out.shape))
```
