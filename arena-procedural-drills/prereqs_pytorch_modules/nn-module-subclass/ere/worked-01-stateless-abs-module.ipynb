## Concept

An `nn.Module` with no learnable state needs only a `forward` method; you can omit `__init__` entirely and the base `nn.Module.__init__` runs implicitly. Calling the instance with `module(x)` routes through `__call__` to `forward` without any extra wiring.

## Worked solution

We want the simplest possible Module: one that returns the elementwise absolute value of its input. We subclass `t.nn.Module` and write only `forward(self, x)` returning `x.abs()`. Because there is no state to register, we deliberately skip `__init__`, relying on the base class to initialize the internal bookkeeping. We instantiate `AbsLayer`, then invoke it as `layer(x)` (not `layer.forward(x)`) to confirm the `__call__` plumbing works. We print the output to show negatives became positive.

```
import torch as t

class AbsLayer(t.nn.Module):
    def forward(self, x):
        return x.abs()

layer = AbsLayer()
x = t.tensor([-2.0, 1.0, -3.0])
print(layer(x))
```
