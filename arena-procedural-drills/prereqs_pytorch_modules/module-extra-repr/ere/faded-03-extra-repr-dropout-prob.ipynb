## Concept

Built-in modules like `nn.Dropout` show their key hyperparameter via `extra_repr` (e.g. `Dropout(p=0.5)`). Replicating this means returning a short f-string naming the hyperparameter; PyTorch wraps it in the class name automatically.

## Faded exercise 3

Complete `MyDropout` so `print(mod)` renders as `MyDropout(p=<value>)`. Fill in the `extra_repr` method.

**Fill in:** the extra_repr return string reporting the dropout probability p

````
import torch as t

class MyDropout(t.nn.Module):
    def __init__(self, p=0.5):
        super().__init__()
        self.p = float(p)

    def extra_repr(self):
        raise NotImplementedError()  # TODO: the extra_repr return string reporting the dropout probability p

    def forward(self, x):
        if not self.training or self.p == 0.0:
            return x
        mask = (t.rand_like(x) > self.p).float()
        return x * mask / (1.0 - self.p)

mod = MyDropout(0.25)
print(repr(mod))


def _test():
    mod = MyDropout(0.25)
    r = repr(mod)
    assert r.startswith('MyDropout(')
    assert 'p=0.25' in r
    assert mod.extra_repr() == 'p=0.25'
    assert isinstance(mod.extra_repr(), str)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
