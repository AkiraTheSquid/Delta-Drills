With `drop_last=False` (the default) a `DataLoader` keeps the final partial batch, so iterating one epoch yields `ceil(N / batch_size)` batches and covers all `N` items. Keeping the tail is the right default when you cannot afford to silently drop data, e.g. at evaluation time.

### Build a loader that keeps the partial last batch

Implement `make_keep_last_loader(dataset, batch_size)` returning a `DataLoader` with `num_workers=0`, `pin_memory=False`, `shuffle=False`, and `drop_last=False` so the final partial batch is retained.

Complete the blanked line that constructs and returns the `DataLoader`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from torch.utils.data import DataLoader, TensorDataset

def make_keep_last_loader(dataset, batch_size):
    loader = DataLoader(
        dataset,
        batch_size=batch_size,
        num_workers=0,
        pin_memory=False,
        shuffle=False,
        drop_last=False,
    )
    return loader


import math
from torch.utils.data import TensorDataset

def _test():
    t.manual_seed(0)
    N, bs = 17, 5
    X = t.arange(N * 2, dtype=t.float32).reshape(N, 2)
    ds = TensorDataset(X)
    loader = make_keep_last_loader(ds, bs)
    assert loader.drop_last is False
    sizes = [b[0].shape[0] for b in loader]
    assert len(sizes) == math.ceil(N / bs), (len(sizes), math.ceil(N / bs))
    assert sum(sizes) == N, (sum(sizes), N)
    assert sizes[-1] == N % bs, (sizes[-1], N % bs)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
