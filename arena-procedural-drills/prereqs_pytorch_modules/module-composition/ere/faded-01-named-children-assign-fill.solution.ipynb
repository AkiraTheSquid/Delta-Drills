## Concept

Assigning child Modules as named attributes inside `__init__` (after `super().__init__()`) auto-registers them, making their parameters visible to the parent. Forgetting these assignments leaves the parent with no learnable layers.

## Faded exercise 1

Complete `TwoLayerMLP.__init__` by assigning the two child Linear layers as the named attributes `self.fc1` and `self.fc2`. The forward is given. Fill in the two assignments.

**Fill in:** assigning the two child Linear layers to self.fc1 and self.fc2

````
import torch as t
import torch.nn as nn

class TwoLayerMLP(nn.Module):
    def __init__(self, in_f, hid, out_f):
        super().__init__()
        self.fc1 = nn.Linear(in_f, hid)
        self.fc2 = nn.Linear(hid, out_f)
    def forward(self, x):
        return self.fc2(t.relu(self.fc1(x)))


def _test():
    t.manual_seed(61)
    model = TwoLayerMLP(4, 8, 3)
    # independent expectations
    names = [n for n, _ in model.named_children()]
    assert sorted(names) == ['fc1', 'fc2']
    assert sum(1 for _ in model.parameters()) == 4
    out = model(t.randn(5, 4))
    assert tuple(out.shape) == (5, 3)
    pnames = {n for n, _ in model.named_parameters()}
    assert 'fc1.weight' in pnames and 'fc2.bias' in pnames


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
