A variable-depth model stores its layers in `nn.ModuleList` so they all register. Using a plain Python list would leave the layers unregistered and invisible to `.parameters()`.

## Faded exercise 3

Complete `DeepMLP.__init__` so `self.layers` is an `nn.ModuleList` of `num_layers` Linear(dim, dim) modules. The forward is given. Fill in the ModuleList construction.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

class DeepMLP(nn.Module):
    def __init__(self, dim, num_layers):
        super().__init__()
        self.layers = nn.ModuleList([nn.Linear(dim, dim) for _ in range(num_layers)])
    def forward(self, x):
        for i, layer in enumerate(self.layers):
            x = layer(x)
            if i < len(self.layers) - 1:
                x = t.relu(x)
        return x


def _test():
    t.manual_seed(63)
    model = DeepMLP(4, 3)
    assert isinstance(model.layers, nn.ModuleList)
    assert len(model.layers) == 3
    # 3 Linears -> 6 params, all registered
    assert sum(1 for _ in model.parameters()) == 6
    out = model(t.randn(5, 4))
    assert tuple(out.shape) == (5, 4)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
