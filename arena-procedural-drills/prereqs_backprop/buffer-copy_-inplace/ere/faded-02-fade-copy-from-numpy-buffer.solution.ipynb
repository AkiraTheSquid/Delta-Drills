## Concept

`tensor.copy_(src)` accepts any tensor source and writes it into the destination's existing storage, keeping the destination's dtype and identity. A common pattern is to compute statistics elsewhere (here with NumPy), wrap the result as a torch tensor, and `copy_` it into the registered buffer so the buffer link survives.

## Faded exercise 2

Implement `load_stats_into_buffer(buf, np_values)` where `np_values` is a NumPy array of the same length as `buf`. Convert the NumPy array to a torch tensor and copy it into `buf` IN PLACE so `buf`'s `data_ptr()` is preserved. The conversion line is given; complete the in-place copy.

**Fill in:** Copies the torch-converted source tensor into buf's existing storage in place.

````
def load_stats_into_buffer(buf: Tensor, np_values) -> None:
    src = t.from_numpy(np_values).to(buf.dtype)
    buf.copy_(src)


def _test():
    np.random.seed(0)
    vals = np.random.rand(6).astype(np.float64)
    buf = t.zeros(6, dtype=t.float32)
    ptr_before = buf.data_ptr()
    expected = t.from_numpy(vals).to(t.float32)
    ret = load_stats_into_buffer(buf, vals)
    assert ret is None, "function must return None (mutate in place)"
    assert buf.data_ptr() == ptr_before, "buffer storage must be preserved"
    assert buf.dtype == t.float32, "destination dtype must stay float32"
    assert t.allclose(buf, expected), f"got {buf}, expected {expected}"


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
