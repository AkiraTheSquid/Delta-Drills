## Concept

For `out = sin(x)` the elementwise derivative is `d(out)/dx = cos(x)`. The chain rule gives `dL/dx = grad_out * cos(x)`. This op cannot reuse `out` (you'd need `sqrt(1-out**2)` and lose the sign), so it reads the original input `x` — still under the same `(grad_out, out, x)` signature.

## Faded exercise 2

Implement `sin_back(grad_out, out, x)`, the backward fn for `out = torch.sin(x)`, using the canonical `(grad_out, out, x)` signature. Complete the line that computes the local derivative `d(out)/dx`. The final return then applies the chain rule. Return a tensor matching `x`'s shape.

**Fill in:** The local elementwise derivative d(out)/dx of out = sin(x), namely torch.cos(x).

````
def sin_back(grad_out, out, x):
    # out = sin(x); d(out)/dx = cos(x).
    local = t.cos(x)
    return grad_out * local

t.manual_seed(0)
x = t.randn(5)
out = t.sin(x)
grad_out = t.randn(5)
print(sin_back(grad_out, out, x))


def _test():
    t.manual_seed(2)
    x = t.randn(5, requires_grad=True)
    out = t.sin(x)
    grad_out = t.randn(5)
    out.backward(grad_out)
    ref = x.grad
    got = sin_back(grad_out, out.detach(), x.detach())
    assert got.shape == x.shape, (got.shape, x.shape)
    assert t.allclose(got, ref, atol=1e-6), (got, ref)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
