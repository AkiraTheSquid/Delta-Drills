## Iterating `parents.items()` to dispatch back fns by argnum

Ex1 built `{argnum: tensor}` for positional Tensors; ex2 extended with kwarg-name keys. The deepening move USES the dict at reverse time: iterate `parents.items()`, look up `BACK_FUNCS[(fn, argnum)]`, compute that parent's grad contribution.

```python
for argnum, parent in recipe.parents.items():
    if not isinstance(argnum, int):
        continue  # kwarg parents handled separately
    back_fn = BACK_FUNCS[(recipe.func, argnum)]
    grad_for_parent = back_fn(grad_out, out_arr, *recipe.args)
    accumulate(parent, grad_for_parent)
```

**Why int-only filtering.** ex2 mixed int keys (positional) and str keys (kwarg names). The dispatcher for POSITIONAL back fns takes the int keys. Kwarg-positioned Tensors get a SEPARATE back-fn registry keyed by `(fn, kwarg_name: str)` — different table, same call shape.

**`*recipe.args` un-splats stored positional args.** The back fn signature is `(grad_out, out, *forward_args)`. Storing args as a tuple and splatting at call time is the cleanest way to feed the back fn whatever the forward had — `add` has 2 args, `clamp` has 1 (plus min/max kwargs), `sum` has 1 (plus dim kwarg). Splat handles all of them uniformly.

- `back_funcs`: dict keyed by `(fwd_fn, argnum: int)`.

- DO NOT compute grads for str-keyed parents (kwarg Tensors). Filter them out via `isinstance(k, int)`.
- DO NOT mutate `recipe`.

````python
def ex3_dispatch_back_fns(recipe, grad_out, out_arr, back_funcs):
    """Iterate parents.items(), dispatch back fn per argnum, return {argnum: grad}."""
    raise NotImplementedError()


def _test_ex3():
    from dataclasses import dataclass
    from typing import Callable

    @dataclass
    class Recipe:
        func: Callable
        args: tuple
        kwargs: dict
        parents: dict

    class MiniTensor:
        def __init__(self, array):
            self.array = array

    # === Binary add: 2 positional parents ===
    def add_back0(grad_out, out, x, y, **_): return grad_out * t.ones_like(x)
    def add_back1(grad_out, out, x, y, **_): return grad_out * t.ones_like(y)
    back_funcs = {(t.add, 0): add_back0, (t.add, 1): add_back1}

    x = MiniTensor(t.tensor([1.0, 2.0]))
    y = MiniTensor(t.tensor([3.0, 4.0]))
    out_arr = x.array + y.array
    recipe = Recipe(t.add, (x.array, y.array), {}, {0: x, 1: y})
    grad_out = t.ones_like(out_arr)

    result = ex3_dispatch_back_fns(recipe, grad_out, out_arr, back_funcs)
    assert isinstance(result, dict)
    assert set(result.keys()) == {0, 1}, f'expected argnums 0,1; got {list(result.keys())}'
    assert t.allclose(result[0], t.ones_like(x.array))
    assert t.allclose(result[1], t.ones_like(y.array))

    # === Unary log: 1 parent ===
    def log_back0(grad_out, out, x, **_): return grad_out / x
    back_funcs = {(t.log, 0): log_back0}
    x = MiniTensor(t.tensor([1.0, 2.0, 4.0]))
    out_arr = t.log(x.array)
    recipe = Recipe(t.log, (x.array,), {}, {0: x})
    grad_out = t.ones_like(out_arr)
    result = ex3_dispatch_back_fns(recipe, grad_out, out_arr, back_funcs)
    assert set(result.keys()) == {0}
    assert t.allclose(result[0], 1.0 / x.array)

    # === Skip str-keyed parents (kwarg Tensors) ===
    def add_back0_skip(grad_out, out, x, **_): return grad_out * t.ones_like(x)
    back_funcs = {(t.add, 0): add_back0_skip}
    x = MiniTensor(t.tensor([1.0]))
    kwarg_t = MiniTensor(t.tensor([99.0]))  # str-keyed parent — must be skipped
    out_arr = t.tensor([1.0])
    recipe = Recipe(t.add, (x.array,), {'mask': kwarg_t.array}, {0: x, 'mask': kwarg_t})
    result = ex3_dispatch_back_fns(recipe, t.ones_like(out_arr), out_arr, back_funcs)
    assert set(result.keys()) == {0}, f'must skip str-keyed; got {list(result.keys())}'
    assert 'mask' not in result

    # === Empty parents (no Tensor inputs — degenerate) ===
    recipe = Recipe(t.add, (), {}, {})
    result = ex3_dispatch_back_fns(recipe, t.tensor(1.0), t.tensor(0.0), {})
    assert result == {}, f'empty parents must yield empty dict; got {result}'

    # === Missing back fn raises KeyError ===
    x = MiniTensor(t.tensor([1.0]))
    recipe = Recipe(t.add, (x.array,), {}, {0: x})  # no entry registered
    raised = False
    try:
        ex3_dispatch_back_fns(recipe, t.ones_like(x.array), x.array, {})
    except KeyError:
        raised = True
    assert raised, 'missing back fn must raise KeyError'

    # === Kwargs splat to back fn ===
    # Define a back fn that DEPENDS on a kwarg to verify the splat reaches it.
    def sum_back0(grad_out, out, x, dim=None, **_):
        if dim is None:
            return grad_out * t.ones_like(x)
        return grad_out.unsqueeze(dim).expand_as(x)
    back_funcs = {(t.sum, 0): sum_back0}
    x = MiniTensor(t.arange(6.0).reshape(2, 3))
    out_arr = x.array.sum(dim=1)
    recipe = Recipe(t.sum, (x.array,), {'dim': 1}, {0: x})
    grad_out = t.tensor([1.0, 2.0])
    result = ex3_dispatch_back_fns(recipe, grad_out, out_arr, back_funcs)
    expected = t.tensor([[1., 1., 1.], [2., 2., 2.]])
    assert t.allclose(result[0], expected), f'kwargs splat broken: {result[0]}'

    # === Recipe is not mutated ===
    x = MiniTensor(t.tensor([1.0]))
    back_funcs = {(t.log, 0): lambda g, o, x, **_: g / x}
    recipe = Recipe(t.log, (x.array,), {}, {0: x})
    before_parents = dict(recipe.parents)
    before_kwargs = dict(recipe.kwargs)
    _ = ex3_dispatch_back_fns(recipe, t.ones(1), t.zeros(1), back_funcs)
    assert recipe.parents == before_parents, 'parents was mutated'
    assert recipe.kwargs == before_kwargs, 'kwargs was mutated'
    _dd_passed.add('ex3')
    print("ex3 ✓")

_test_ex3()
````

<details><summary>Solution</summary>

```python
def ex3_dispatch_back_fns(recipe, grad_out, out_arr, back_funcs):
    grads = {}
    for argnum, parent in recipe.parents.items():
        if not isinstance(argnum, int):
            continue  # kwarg-named parents handled by a different dispatcher
        back_fn = back_funcs[(recipe.func, argnum)]
        grads[argnum] = back_fn(
            grad_out, out_arr, *recipe.args, **recipe.kwargs
        )
    return grads
```

**`isinstance(argnum, int)` filter is critical.** ex2 mixed int and str keys in parents. The positional back-fn dispatcher must skip the str keys — they belong to a parallel kwarg-back-fn table that this drill doesn't model. Without the filter, you'd do `back_funcs[(recipe.func, 'mask')]` which crashes.

**`*recipe.args, **recipe.kwargs` is the universal call shape.** Every back fn's signature is `(grad_out, out, *forward_args, **forward_kwargs)`. Splatting works for unary, binary, ternary, and parameterized ops uniformly — that's the whole point of the Recipe abstraction.

**The dispatcher is intentionally minimal.** No grad accumulation (that's the leaf-accumulator's job); no graph traversal (that's topological sort's job). Just: 'given a Recipe and an incoming grad, hand back the per-argnum grad contributions'. Single responsibility.
