## Concept

Building the parents dict requires iterating over the positional args with `enumerate`, keeping only those entries where the argument is a Tensor (by `isinstance`), and using the enumeration index as the dict key. This preserves the original argument positions so the backward dispatch can find the right back-function for each input.

### Exercise — Dict comprehension filtering Tensor args

Complete `build_parents(args)`. Return a dict mapping each original argument index to the MiniTensor at that position, skipping all non-MiniTensor args.

Fill in the dict comprehension body.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from dataclasses import dataclass
from typing import Any
import torch as t

@dataclass
class MiniTensor:
    array: Any
    grad: Any = None

def build_parents(args: tuple) -> dict:
    # Keep only tensor-like args (duck-typed: they carry an `array` payload),
    # keyed by their original position in args.
    return {idx: a for idx, a in enumerate(args) if hasattr(a, "array")}

# Quick check
t.manual_seed(20)
ta = MiniTensor(t.randn(3))
tb = MiniTensor(t.randn(3))
result = build_parents((ta, 9.0, tb))
print(result)


def _test():
    import torch as t
    from dataclasses import dataclass
    from typing import Any
    @dataclass
    class MiniTensor:
        array: Any
        grad: Any = None
    t.manual_seed(20)
    ta = MiniTensor(t.randn(3))
    tb = MiniTensor(t.randn(3))
    # Two Tensors with scalar in between
    r1 = build_parents((ta, 9.0, tb))
    assert r1 == {0: ta, 2: tb}, f'got {r1}'
    # All scalars -> empty
    r2 = build_parents((1, 2.0, True))
    assert r2 == {}, f'got {r2}'
    # Single tensor
    r3 = build_parents((ta,))
    assert r3 == {0: ta}


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
