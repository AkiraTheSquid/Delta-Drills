## Concept

The `torch.bmm` function requires both inputs to be rank-3 tensors. To apply it to a batched matrix `A: (B, M, N)` and a batched vector `x: (B, N)`, you must first unsqueeze `x` to `(B, N, 1)`, apply `bmm`, then squeeze the trailing dimension back to get `(B, M)`. This idiom converts a batched matvec into the rank-3 form that `bmm` expects.
