The forward pass of a single linear layer (no activation) is `out = W @ x + b`, where `W: (out_dim, in_dim)`, `x: (in_dim,)`, and `b: (out_dim,)`. The matvec `W @ x` produces `(out_dim,)`, and adding the bias `b` of the same shape is elementwise. The output is 1-D with shape `(out_dim,)` — not `(out_dim, 1)`.
