## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

When you apply `@` between a matrix `W: (M, N)` and a 1-D vector `x: (N,)`, PyTorch returns a 1-D result of shape `(M,)`. This is different from a column-vector result of shape `(M, 1)`. Keeping the output 1-D is important when the result feeds into operations that expect a 1-D vector, such as loss computation over a single sample.

## Faded exercise 1

Complete the function below. Given weight matrix `W` and 1-D input vector `x`, compute the matvec and return the result. Fill in the single blanked line.

Do not use `unsqueeze` — pass `x` directly to `@`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def matvec_1d(W, x):
    out = W @ x
    return out

# Exercise it
t.manual_seed(15)
W = t.randn(4, 6)
x = t.randn(6)
result = matvec_1d(W, x)
print(f"result shape: {result.shape}  (expected (4,), NOT (4,1))")
print(f"result: {result.round(decimals=3)}")


import torch as t

def _test():
    t.manual_seed(15)
    W = t.randn(4, 6)
    x = t.randn(6)

    result = matvec_1d(W, x)

    assert result.shape == (4,), f"Expected (4,), got {result.shape}"

    # Verify numerically against reference
    expected = W @ x
    assert t.allclose(result, expected, atol=1e-6)

    # Verify 1-D, not 2-D
    assert result.dim() == 1, "Output must be 1-D, not 2-D"


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
