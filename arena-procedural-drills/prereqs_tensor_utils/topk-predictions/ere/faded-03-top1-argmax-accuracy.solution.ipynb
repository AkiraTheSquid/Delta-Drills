## Concept

Top-1 accuracy is the special case k=1: the single highest logit per row is the prediction, obtained with `argmax(dim=-1)`. Accuracy is the mean of the boolean match against the labels. This is the baseline the top-k metrics are compared against.

## Faded exercise 3

### Compute top-1 accuracy

Implement `top1_accuracy(logits, labels)` returning a scalar tensor: the fraction of rows whose argmax prediction equals the true label. Complete the line that produces the predicted class per row.

**Fill in:** the argmax over the class axis giving the predicted class per row

````
import torch as t

def top1_accuracy(logits: t.Tensor, labels: t.Tensor) -> t.Tensor:
    preds = logits.argmax(dim=-1)
    return (preds == labels).float().mean()

t.manual_seed(0)
print(top1_accuracy(t.randn(5, 4), t.randint(0, 4, (5,))).item())


def _test():
    logits = t.tensor([
        [3.0, 0.0, 1.0],   # argmax 0
        [0.0, 5.0, 1.0],   # argmax 1
        [1.0, 0.0, 9.0],   # argmax 2
        [2.0, 1.0, 0.0],   # argmax 0
    ])
    labels = t.tensor([0, 1, 0, 0])  # 3 of 4 correct
    acc = top1_accuracy(logits, labels)
    assert acc.shape == (), acc.shape
    # independent truth: matches are rows 0,1,3 -> 3/4 = 0.75
    assert abs(acc.item() - 0.75) < 1e-6, acc.item()


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
