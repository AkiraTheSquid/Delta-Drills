`topk` returns a named tuple `(values, indices)`. The ranked predicted class ids live on the `.indices` attribute, sorted highest-score first. Reading the wrong attribute (`.values`) gives scores, not labels.

## Faded exercise 2

### Pull the ranked label ids out of topk

Implement `topk_labels(logits, k)` returning the `(B, k)` tensor of ranked predicted class ids for each row. Complete the line that calls topk and selects the indices attribute.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def topk_labels(logits: t.Tensor, k: int) -> t.Tensor:
    idx = logits.topk(k, dim=-1).indices
    return idx

print(topk_labels(t.tensor([[0.1, 5.0, 2.0, 1.0]]), 2))


def _test():
    logits = t.tensor([
        [0.1, 5.0, 2.0, 1.0],
        [9.0, 0.0, 8.0, 1.0],
    ])
    idx = topk_labels(logits, 2)
    assert idx.shape == (2, 2), idx.shape
    assert idx.dtype == t.long, idx.dtype
    # independent truth: row0 top2 ids = [1, 2]; row1 top2 ids = [0, 2]
    assert idx[0].tolist() == [1, 2], idx[0].tolist()
    assert idx[1].tolist() == [0, 2], idx[1].tolist()


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
