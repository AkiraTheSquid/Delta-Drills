## Concept

In Lambertian shading, surface intensity is the dot product of a surface normal and the light direction, but back-facing surfaces produce negative dots that must be clamped to zero (a surface can't be lit by a light behind it). `t.where(dots < 0, 0, dots)` performs exactly this clip.

## Worked solution

We compute per-surface light intensity and clip the back-facing values.

1. We have a vector of normal-light dot products, some negative (back-facing).
2. `t.where(dots < 0, t.zeros_like(dots), dots)` zeroes the negatives while keeping the positive intensities.
3. This is the physically correct floor: unlit surfaces contribute zero, not negative, light.

We print the raw dots and the clipped intensities to show the negatives were truncated.

```
import torch as t

def lambertian_intensity(dots: t.Tensor) -> t.Tensor:
    return t.where(dots < 0, t.zeros_like(dots), dots)

dots = t.tensor([0.8, -0.3, 0.0, -1.0, 0.5])
print('dots:     ', dots.tolist())
print('intensity:', lambertian_intensity(dots).tolist())
```
