## Concept

`t.where(cond, a, b)` selects elementwise: where `cond` is True it takes `a`, otherwise `b`. ReLU is `max(0, x)`, which expresses as `t.where(x < 0, zeros, x)` — clip negatives to zero and pass non-negatives through.

## Worked solution

We implement ReLU as a branching selection.

1. We build the condition `x < 0`, a boolean tensor flagging the negative entries.
2. `t.where(x < 0, t.zeros_like(x), x)` picks `0.0` where the condition holds and the original value everywhere else.
3. Using `zeros_like(x)` keeps the dtype and shape aligned, so broadcasting is trivially correct.

We print a mixed-sign input and its ReLU, confirming negatives became zero and positives survived.

```
import torch as t

def relu_via_where(x: t.Tensor) -> t.Tensor:
    return t.where(x < 0, t.zeros_like(x), x)

x = t.tensor([-3.0, -0.5, 0.0, 1.5, 4.0])
print('input: ', x.tolist())
print('relu:  ', relu_via_where(x).tolist())
```
