## Concept

`t.where(cond, a, b)` takes `a` where `cond` is True else `b`. ReLU clips negatives to zero, so the select is `t.where(x < 0, zeros_like(x), x)`.

## Faded exercise 1

### Express ReLU as a where

Implement `relu(x)` using a single `t.where` call: zero where `x` is negative, `x` otherwise. Complete the `t.where(...)` expression.

**Fill in:** the t.where(x < 0, t.zeros_like(x), x) selection

````
import torch as t

def relu(x: t.Tensor) -> t.Tensor:
    out = t.where(x < 0, t.zeros_like(x), x)
    return out

print(relu(t.tensor([-1.0, 2.0, -3.0, 4.0])).tolist())


def _test():
    x = t.tensor([-2.0, -0.1, 0.0, 0.5, 3.0])
    out = relu(x)
    assert out.shape == x.shape
    # independent ground truth via clamp
    expected = x.clamp(min=0.0)
    assert t.allclose(out, expected), out
    # negatives strictly zeroed, positives preserved
    assert float(out[0]) == 0.0 and float(out[1]) == 0.0
    assert float(out[4]) == 3.0


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
