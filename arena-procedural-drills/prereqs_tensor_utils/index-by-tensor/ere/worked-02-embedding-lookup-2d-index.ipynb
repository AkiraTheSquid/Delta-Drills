When the index tensor is 2-D `(B, T)`, indexing an embedding table `(V, D)` yields `(B, T, D)`: the full index shape becomes the leading dimensions and the embedding dimension trails. This is the canonical token-embedding lookup.

## Worked solution

We look up embeddings for a batch of token-id sequences.

1. `embed` is the `(V, D)` table; `idx` is a `(B, T)` `LongTensor` of token ids.
2. `embed[idx]` replaces every scalar id in `idx` with its `D`-vector from the table.
3. The result shape is `idx.shape + (D,)` = `(B, T, D)`. The 2-D index shape leads, the embedding dim trails — exactly what a transformer expects after token embedding.
4. We confirm a specific `out[b, t]` equals `embed[idx[b, t]]` to verify the per-position lookup.

```
import torch as t

t.manual_seed(1)
V, D = 10, 5
embed = t.randn(V, D)
idx = t.randint(0, V, (3, 4))

def embedding_lookup(embed, idx):
    return embed[idx]

out = embedding_lookup(embed, idx)
print(out.shape)
b, t_pos = 1, 2
print('lookup matches:', bool(t.equal(out[b, t_pos], embed[idx[b, t_pos]])))
```
