## Concept

Unit-normalizing embeddings before cosine similarity is standard in dense retrieval. The pattern `.norm(dim=-1, keepdim=True)` computes per-vector L2 norms while preserving the trailing dimension for broadcast division. After normalization, dot product equals cosine similarity.
