## Concept

Gradient clipping and direction-only updates require normalizing gradient vectors to unit length. The `.norm(dim=-1, keepdim=True)` pattern computes each vector's magnitude while preserving tensor rank, so the subsequent division broadcasts element-wise across all dimensions.
