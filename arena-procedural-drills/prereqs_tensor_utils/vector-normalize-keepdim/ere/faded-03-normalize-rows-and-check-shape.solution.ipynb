## Concept

When using `.norm(dim=-1, keepdim=True)`, the result has the same number of dimensions as the input but with the last axis collapsed to size 1. This is what enables correct broadcasting: a `(B, D)` tensor divided by a `(B, 1)` norm tensor yields a `(B, D)` result with each row on the unit sphere.
