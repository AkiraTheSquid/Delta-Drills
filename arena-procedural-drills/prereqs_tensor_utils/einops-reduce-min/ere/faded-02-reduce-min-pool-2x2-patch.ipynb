## Concept

The `'b c (h p1) (w p2) -> b c h w'` pattern with `p1=2, p2=2` partitions the spatial grid into non-overlapping `2x2` patches and takes the minimum within each. The patch axes `p1` and `p2` appear on the left but not the right, so they are reduced. The block axes `h` and `w` are preserved.
