## Faded exercise 1

Implement `faded1_collect_mode_flags(model)`. Starting from the model's current state, record `flag_initial = model.training`. Then call `model.train()` and record `flag_after_train`. Then call `model.eval()` and record `flag_after_eval`. Return a dict with all three flags under those exact keys.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch.nn as nn

def faded1_collect_mode_flags(model):
    flag_initial = model.training
    model.train()
    flag_after_train = model.training
    model.eval()
    flag_after_eval = model.training
    return {
        'flag_initial': flag_initial,
        'flag_after_train': flag_after_train,
        'flag_after_eval': flag_after_eval,
    }


def _test():
    import torch.nn as nn
    model = nn.Linear(3, 2)
    # Start from known state: train
    model.train()
    result = faded1_collect_mode_flags(model)
    assert result['flag_initial'] == True
    assert result['flag_after_train'] == True
    assert result['flag_after_eval'] == False
    # Also test starting from eval
    model.eval()
    result2 = faded1_collect_mode_flags(model)
    assert result2['flag_initial'] == False
    assert result2['flag_after_train'] == True
    assert result2['flag_after_eval'] == False


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
