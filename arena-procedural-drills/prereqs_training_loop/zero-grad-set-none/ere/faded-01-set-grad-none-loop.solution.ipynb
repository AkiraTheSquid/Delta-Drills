## Concept

`zero_grad` with the `set_to_none` convention iterates the parameter list and sets each `.grad = None`. None is treated as zero on next access and is cheaper than a zeroed buffer.

## Faded exercise 1

### Clear gradients with the None convention

Implement `zero_grad(params)` that sets every parameter's gradient to `None` (no return value). Complete the loop body that clears one parameter's grad.

**Fill in:** setting p.grad = None for each parameter

````
import torch as t
import torch.nn as nn

def zero_grad(params):
    for p in params:
        p.grad = None

t.manual_seed(0)
m = nn.Linear(3, 2)
for p in m.parameters():
    p.grad = t.ones_like(p)
zero_grad(list(m.parameters()))
print(all(p.grad is None for p in m.parameters()))


def _test():
    t.manual_seed(0)
    m = nn.Linear(3, 2)
    params = list(m.parameters())
    for p in params:
        p.grad = t.ones_like(p)
    # precondition: grads present
    assert all(p.grad is not None for p in params)
    ret = zero_grad(params)
    assert ret is None, 'zero_grad mutates in place, returns nothing'
    # all grads cleared to None
    assert all(p.grad is None for p in params)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
