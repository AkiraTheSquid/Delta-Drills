## Concept

When you create a noise tensor with `torch.randn_like(template)`, the resulting tensor lives on the same device as `template`. This means you never need to call `.to(device)` on the noise — it is already in the right place. For CPU testing, this is trivially true; for GPU training, it prevents cross-device errors during the forward pass.

**Fill in:** Return standard-normal noise with the same properties as template using a single randn_like call.

````
import torch as t

def sample_noise(template: t.Tensor) -> t.Tensor:
    return t.randn_like(template)

t.manual_seed(25)
template = t.zeros(5, 10, dtype=t.float32)
noise = sample_noise(template)
print(noise.shape, noise.dtype, noise.device)


def _test():
    import torch as t
    t.manual_seed(25)
    for dtype in [t.float32, t.float64]:
        template = t.zeros(5, 10, dtype=dtype)
        noise = sample_noise(template)
        assert noise.shape == template.shape
        assert noise.dtype == template.dtype
        assert noise.device == template.device
        # Sanity: noise should be roughly standard normal
        assert abs(noise.mean().item()) < 0.5


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
