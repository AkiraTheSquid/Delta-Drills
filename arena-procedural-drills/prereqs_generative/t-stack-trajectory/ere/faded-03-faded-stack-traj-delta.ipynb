Given a `(T, D)` trajectory tensor, the per-step velocity (delta) is computed by subtracting consecutive timesteps: `traj[1:] - traj[:-1]`. This produces a `(T-1, D)` tensor where each row is the change from one step to the next. Slicing `traj[t0:t1]` first lets you compute deltas within any time window.
