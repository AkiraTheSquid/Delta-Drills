import torch as t

def collect_trajectory(initial: t.Tensor, n_steps: int) -> t.Tensor:
    z = initial.clone()
    snapshots = []
    for _ in range(n_steps):
        z = z * 0.9  # fake update
        snapshots.append(z.detach().clone())
    return t.stack(snapshots, dim=0)
