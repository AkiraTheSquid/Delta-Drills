## Concept

`torch.stack(tensors, dim=d)` inserts the new stacked axis at position `d`. When the input tensors have shape `(B, D)` and you want the output to be `(B, T, D)` (batch-first), you must stack along `dim=1` so that the new time axis ends up in the middle.

**Fill in:** Replace `None` with the integer axis index that places the new time dimension between batch and feature.

````
import torch as t

def stack_trajectory_batch_first(snapshots: list) -> t.Tensor:
    """snapshots: list of T tensors each shape (B, D).
    Returns: (B, T, D) tensor."""
    return t.stack(snapshots, dim=1)

# Test
t.manual_seed(2)
B, D, T = 3, 4, 5
snaps = [t.randn(B, D) for _ in range(T)]
result = stack_trajectory_batch_first(snaps)
print('result.shape:', result.shape)   # expect (3, 5, 4)


def _test():
    import torch as t
    B, D, T = 3, 4, 5
    snaps = [t.arange(B * D, dtype=t.float).reshape(B, D) + i for i in range(T)]
    result = stack_trajectory_batch_first(snaps)
    assert result.shape == (B, T, D), f'expected (3,5,4), got {result.shape}'
    # Verify content: result[b, t, :] == snaps[t][b, :]
    for ti in range(T):
        assert t.allclose(result[:, ti, :], snaps[ti])


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
