## Concept

A `(K, D)` codebook is decoded into a `(B, L, D)` tensor by indexing it with a `(B, L)` grid of integer code ids. The advanced-index `codebook[codes]` fans each code id out to its full D-vector, preserving the leading `(B, L)` shape and appending `D`. Repeated ids select the same source row.

## Faded exercise 2

Complete `decode_codes(codebook, codes)` so it returns the `(B, L, D)` tensor of decoded vectors, where output position `(b, l)` is `codebook[codes[b, l]]`. The function signature and the exercising harness are given; fill in the single fan-out indexing expression.

**Fill in:** The advanced-integer-indexing expression that fans the (K, D) codebook out to (B, L, D) using the (B, L) code ids.

````
import torch as t
from torch import Tensor

def decode_codes(codebook: Tensor, codes: Tensor) -> Tensor:
    out = codebook[codes]
    return out


def _test():
    import torch as t
    t.manual_seed(0)
    K, D, B, L = 5, 4, 2, 3
    codebook = t.randn(K, D)
    codes = t.randint(0, K, (B, L))
    out = decode_codes(codebook, codes)
    assert tuple(out.shape) == (B, L, D)
    ref = t.stack([t.stack([codebook[codes[b, l]] for l in range(L)]) for b in range(B)])
    assert t.allclose(out, ref)
    assert out.dtype == codebook.dtype


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
