## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

In the generator step the fake is NOT detached so gradients flow into `G`, and the loss is `-D(fake).mean()`: the generator minimizes the negative critic score, i.e. it pushes the critic to rate fakes highly. The sign is what distinguishes the G objective from the D objective.

## Faded exercise 2

### Write the generator's loss

Implement `g_step(G, D, G_opt, z)`: one generator update returning the G loss as a float. The fake (un-detached) is built for you; complete the loss expression so the generator is rewarded for fooling the critic.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

def g_step(G, D, G_opt, z):
    G_opt.zero_grad()
    fake = G(z)
    loss_G = -D(fake).mean()
    loss_G.backward()
    G_opt.step()
    return loss_G.item()

t.manual_seed(0)
G = nn.Linear(3, 4); D = nn.Linear(4, 1)
G_opt = t.optim.SGD(G.parameters(), lr=0.01)
print(g_step(G, D, G_opt, t.randn(5, 3)))


def _test():
    t.manual_seed(0)
    G = nn.Linear(3, 4); D = nn.Linear(4, 1)
    G_opt = t.optim.SGD(G.parameters(), lr=0.01)
    z = t.randn(5, 3)
    # independent ground truth of the loss value before the step
    with t.no_grad():
        expected = (-D(G(z)).mean()).item()
    G_before = G.weight.detach().clone()
    loss = g_step(G, D, G_opt, z)
    assert abs(loss - expected) < 1e-5, (loss, expected)
    # generator must have moved
    assert not t.equal(G_before, G.weight.detach())


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
