## Concept

`.detach()` returns a tensor that shares storage with the original but is cut off from autograd: same values, but `requires_grad=False` and `grad_fn=None`. This lets you prove that detaching changes only the backward graph, never the forward numbers.

## Faded exercise 3

Implement `detach_report(G, z)` that returns a tuple `(values_equal, detached_requires_grad, detached_has_no_grad_fn)`. Compute `with_graph = G(z)` and a detached copy of the same forward; then compare their values with `torch.allclose`, and inspect the detached tensor's `requires_grad` and whether its `grad_fn` is None. Complete the blanked line that builds the detached tensor.

**Fill in:** Builds the detached version of G(z): identical values but disconnected from the autograd graph.

````
import torch.nn as nn

t.manual_seed(0)
G = nn.Linear(5, 3)

def detach_report(G, z):
    with_graph = G(z)
    detached = G(z).detach()
    values_equal = bool(t.allclose(with_graph, detached))
    return values_equal, detached.requires_grad, detached.grad_fn is None


def _test():
    import torch.nn as nn
    t.manual_seed(0)
    Gt = nn.Linear(5, 3)
    zt = t.randn(6, 5)
    ve, drg, dno = detach_report(Gt, zt)
    assert ve is True, 'detach must not change the forward values'
    assert drg is False, 'detached tensor must have requires_grad=False'
    assert dno is True, 'detached tensor must have grad_fn=None'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
