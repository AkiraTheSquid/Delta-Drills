## Concept

A non-leaf tensor is the output of a computation, not a directly-created variable. Optimizers silently skip non-leaf tensors because they have no stored gradient slot. Asserting `p.is_leaf` for each parameter before building the optimizer catches this class of bug immediately, with a message that includes the tensor's shape for easy identification.

## Faded exercise 1

Implement `check_all_leaves(params)` which iterates `params` and asserts `p.is_leaf` for each. On failure, raise `AssertionError` with a message including `tuple(p.shape)`. Return `True` if all pass.

The `requires_grad` check is NOT needed here — only the leaf check.

**Fill in:** Assert p.is_leaf for each parameter; on failure, raise AssertionError with the tensor's shape in the message.

````
import torch as t
import torch.nn as nn

def check_all_leaves(params) -> bool:
    for p in params:
        assert p.is_leaf, f'shape={tuple(p.shape)} is non-leaf — optimizer will silently skip it'
    return True


def _test():
    import torch as t
    import torch.nn as nn
    # All leaves should pass
    good = [nn.Parameter(t.randn(3)), nn.Parameter(t.randn(4, 2))]
    assert check_all_leaves(good) is True
    # A non-leaf should raise
    src = t.randn(5, requires_grad=True)
    non_leaf = src + 1  # non-leaf
    try:
        check_all_leaves([non_leaf])
        assert False, 'Should have raised'
    except AssertionError as e:
        msg = str(e)
        assert '(5,)' in msg or 'shape' in msg, f'Message should include shape: {msg}'


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
