## Concept

After collecting one representative per class into a Python list, `t.stack(per_class, dim=0)` assembles them into a single tensor. Each element of the list has shape `(*sample_shape)`, and stacking along `dim=0` adds a new leading dimension of size `num_classes`, producing `(num_classes, *sample_shape)`. Without the stack, the loop produces a Python list which cannot be used directly as a tensor.

## Faded exercise 2

Complete `one_per_class` by filling in the return statement. The blank assembles the `per_class` list into a single tensor with the classes indexed along the first dimension.

**Fill in:** The t.stack call that combines the per_class list into a (num_classes, *sample_shape) tensor.

````
import torch as t

t.manual_seed(0)

def one_per_class(data: t.Tensor, labels: t.Tensor, num_classes: int) -> t.Tensor:
    per_class = []
    for c in range(num_classes):
        mask = labels == c
        per_class.append(data[mask][0])
    return t.stack(per_class, dim=0)

# Exercise it
t.manual_seed(0)
N, num_classes, D = 30, 3, 6
data = t.randn(N, D)
base = t.arange(N) % num_classes
perm = t.randperm(N)
data, labels = data[perm], base[perm]
holdout = one_per_class(data, labels, num_classes)
print(holdout.shape)  # (3, 6)


def _test():
    import torch as t
    t.manual_seed(55)
    N, C, D = 50, 5, 4
    data = t.randn(N, D)
    base = t.arange(N) % C
    perm = t.randperm(N)
    data, labels = data[perm], base[perm]

    result = one_per_class(data, labels, C)

    assert isinstance(result, t.Tensor), "Return type must be Tensor, not list"
    assert result.shape == (C, D), f"Expected ({C}, {D}), got {result.shape}"

    # Verify the holdout is a proper subset (each row in original data at correct label)
    for c in range(C):
        class_data = data[labels == c]
        assert t.allclose(result[c], class_data[0])


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
