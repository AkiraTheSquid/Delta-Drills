## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

The core of the one-per-class pattern is building a boolean selector: `mask = labels == c` gives a 1-D boolean tensor that is `True` exactly at positions where the label equals the target class. Applied to data as `data[mask]`, it returns all samples belonging to class `c`. Taking `[0]` then picks the first one.

## Faded exercise 1

Complete `one_per_class` by filling in the mask expression inside the loop. The blank selects all samples belonging to class `c`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

t.manual_seed(0)

def one_per_class(data: t.Tensor, labels: t.Tensor, num_classes: int) -> t.Tensor:
    per_class = []
    for c in range(num_classes):
        mask = labels == c
        per_class.append(data[mask][0])
    return t.stack(per_class, dim=0)

# Exercise it
t.manual_seed(0)
N, num_classes, D = 30, 3, 4
data = t.randn(N, D)
base = t.arange(N) % num_classes
perm = t.randperm(N)
data, labels = data[perm], base[perm]
holdout = one_per_class(data, labels, num_classes)
print(holdout.shape)  # (3, 4)


def _test():
    import torch as t
    t.manual_seed(42)
    N, C, D = 40, 4, 5
    data = t.randn(N, D)
    base = t.arange(N) % C
    perm = t.randperm(N)
    data, labels = data[perm], base[perm]

    holdout = one_per_class(data, labels, C)

    assert holdout.shape == (C, D), f"Shape {holdout.shape} != ({C}, {D})"
    for c in range(C):
        # The holdout for class c must be the first sample in the class-c group
        class_data = data[labels == c]
        assert t.allclose(holdout[c], class_data[0]), \
            f"Class {c}: holdout doesn't match first class-c sample"


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
