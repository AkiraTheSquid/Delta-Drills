`DistributedSampler(dataset, num_replicas=W, rank=r, shuffle=True, seed=s)` constructs the sampler but does not yet determine the sample order. The shuffled order is computed when you iterate, using the epoch seed set by `set_epoch(epoch)`. Forgetting `set_epoch` means every epoch has the same shuffle — each rank sees the same samples in the same order, which reduces effective data diversity.

```python
from torch.utils.data.distributed import DistributedSampler

def collect_one_shard(dataset, world_size, rank, seed, epoch):
    sampler = DistributedSampler(
        dataset, num_replicas=world_size, rank=rank,
        shuffle=True, seed=seed
    )
    sampler.set_epoch(epoch)
    indices = list(sampler)
    return indices
```
