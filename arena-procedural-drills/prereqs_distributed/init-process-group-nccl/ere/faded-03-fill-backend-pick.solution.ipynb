````
CUDA_OK = False  # stand-in for t.cuda.is_available() so the drill runs on CPU

def pick_backend():
    backend = 'nccl' if CUDA_OK else 'gloo'
    return backend

print(pick_backend())


def _test():
    global CUDA_OK
    # with CUDA unavailable -> gloo
    CUDA_OK = False
    assert pick_backend() == 'gloo'
    # with CUDA available -> nccl
    CUDA_OK = True
    assert pick_backend() == 'nccl'
    # restore
    CUDA_OK = False


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
