````
import torch as t

def build_rank_ctx(rank: int, world_size: int) -> dict:
    device = t.device(f'cuda:{rank}')
    return {
        'rank': rank,
        'world_size': world_size,
        'device': device,
        'is_master': rank == 0,
        'device_str': f'cuda:{rank}',
    }

ctx = build_rank_ctx(0, 4)
print(ctx)


def _test():
    import torch as t
    for rank in range(4):
        ctx = build_rank_ctx(rank, 4)
        assert ctx['rank'] == rank
        assert ctx['world_size'] == 4
        assert isinstance(ctx['device'], t.device)
        assert ctx['device'].index == rank
        assert ctx['is_master'] == (rank == 0)
        assert ctx['device_str'] == f'cuda:{rank}'
    # Uniqueness invariant
    ctxs = [build_rank_ctx(r, 4) for r in range(4)]
    device_indices = {c['device'].index for c in ctxs}
    assert len(device_indices) == 4


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
