## Concept

`t.save(model.state_dict(), path)` persists the parameter dictionary (not the class). Restoring later uses `model.load_state_dict(t.load(path))`. Saving the model object directly is discouraged; the state_dict is the portable form.

## Faded exercise 3

Complete `save_checkpoint(model, path)` so it persists the model's parameters to `path` in the recommended portable form. Fill in the save call.

**Fill in:** the t.save call writing model.state_dict() to path

````
import torch as t
import torch.nn as nn

def save_checkpoint(model, path):
    t.save(model.state_dict(), path)
    return path


def _test():
    import tempfile, os
    t.manual_seed(43)
    src = nn.Linear(4, 2)
    with t.no_grad():
        src.weight.fill_(1.5)
    tmp = tempfile.mkdtemp()
    path = os.path.join(tmp, 'sd.pt')
    save_checkpoint(src, path)
    loaded = t.load(path, weights_only=True)
    # independent check: loaded is a state_dict (plain dict of tensors), restorable
    assert isinstance(loaded, dict)
    dst = nn.Linear(4, 2)
    dst.load_state_dict(loaded)
    assert t.allclose(dst.weight, src.weight)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
