```python
import torch as t

def broadcast_protocol(tensor, rank: int, world_size: int, src: int = 0) -> list:
    if rank == src:
        return [('send', other) for other in range(world_size) if other != src]
    return [('recv', src)]

# Quick check
for r in range(3):
    print(r, broadcast_protocol(None, r, 3))
```
