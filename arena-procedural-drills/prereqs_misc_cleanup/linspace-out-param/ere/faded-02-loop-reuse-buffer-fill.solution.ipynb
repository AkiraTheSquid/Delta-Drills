## Concept

In a hot loop you allocate the buffer ONCE before the loop and fill it with `linspace(..., out=buf)` each iteration, so memory stays flat. Allocating inside the loop (`buf = linspace(...)`) would churn a new tensor every step.

## Faded exercise 2

Complete `schedule_sums(N, ends)` which builds ONE buffer and refills it for each endpoint, collecting `buf.sum().item()` per fill. Fill in the line that allocates the single reusable buffer before the loop.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def schedule_sums(N, ends):
    buf = t.zeros(N)
    sums = []
    for e in ends:
        t.linspace(0.0, e, N, out=buf)
        sums.append(buf.sum().item())
    return sums, buf


def _test():
    N = 5
    ends = [4.0, 8.0]
    sums, buf = schedule_sums(N, ends)
    assert buf.numel() == N
    # independent truth: for end e, linspace(0,e,5) = [0, e/4, e/2, 3e/4, e]; sum = 2.5*e
    assert abs(sums[0] - 2.5 * 4.0) < 1e-5
    assert abs(sums[1] - 2.5 * 8.0) < 1e-5
    # buffer holds the LAST fill (end=8): [0,2,4,6,8]
    assert t.allclose(buf, t.tensor([0.0, 2.0, 4.0, 6.0, 8.0]))


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
