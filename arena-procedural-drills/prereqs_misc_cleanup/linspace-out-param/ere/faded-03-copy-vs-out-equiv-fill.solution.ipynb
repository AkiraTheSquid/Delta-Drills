## Concept

`buf.copy_(torch.linspace(...))` allocates a temp ramp then copies it into `buf` in place — same end values and preserved `data_ptr()` as `linspace(..., out=buf)`, just with one extra temporary. Matching the temp's dtype avoids a precision round-trip.

## Faded exercise 3

Complete `fill_via_copy(buf, lo, hi)` so it fills `buf` in place using the temp-then-copy idiom (NOT the out= kwarg). Fill in the line that copies a freshly built linspace into buf.

**Fill in:** copying a dtype-matched linspace temp into buf in place

````
import torch as t

def fill_via_copy(buf, lo, hi):
    buf.copy_(t.linspace(lo, hi, buf.numel(), dtype=buf.dtype))
    return buf


def _test():
    buf = t.zeros(5)
    ptr0 = buf.data_ptr()
    out = fill_via_copy(buf, 0.0, 8.0)
    assert out is buf
    assert buf.data_ptr() == ptr0, 'copy_ must keep storage'
    # equivalence to the out= idiom
    ref = t.zeros(5)
    t.linspace(0.0, 8.0, 5, out=ref)
    assert t.equal(buf, ref)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
