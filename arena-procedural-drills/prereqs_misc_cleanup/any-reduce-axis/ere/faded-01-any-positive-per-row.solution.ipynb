## Concept

`.any(dim=1)` collapses the column axis of a 2-D tensor, producing one boolean per row. The `dim` you pass is the axis that disappears, so `dim=1` keeps the row axis and answers a per-row question.

## Faded exercise 1

Implement `row_has_positive(x)`. Given an `(N, M)` float tensor, return an `(N,)` bool tensor whose entry `i` is `True` iff row `i` contains at least one strictly positive value. Use `.any(dim=...)` with the correct axis — no Python loop, output dtype `torch.bool`, output shape `(N,)`.

**Fill in:** Reduce the per-cell positivity mask along the column axis (dim=1) to get one bool per row.

````
def row_has_positive(x):
    pos = (x > 0)  # (N, M) bool
    out = pos.any(dim=1)  # (N,) bool
    return out

t.manual_seed(0)
x = t.randn(3, 4)
print(row_has_positive(x))


def _test():
    t.manual_seed(0)
    x = t.randn(3, 4)
    out = row_has_positive(x)
    ref = (x > 0).any(dim=1)
    assert out.shape == (3,), out.shape
    assert out.dtype == t.bool, out.dtype
    assert t.equal(out, ref)
    # explicit construction
    x2 = t.tensor([[-1.0, -2.0, -3.0], [0.0, 0.5, -1.0]])
    out2 = row_has_positive(x2)
    assert t.equal(out2, t.tensor([False, True]))


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
