PyTorch's optimizer repr iterates each param group's non-`'params'` keys in alphabetical order and prints them as `    key: value` lines. Sorting ensures the repr is deterministic even if the dict was built in a different insertion order.

## Faded exercise 2

The header and footer lines are provided. Fill in the inner per-group loop: for each group, write the `'Parameter Group <i>'` line and then one `'    key: value'` line per non-`'params'` key, sorted alphabetically.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

def format_optimizer_repr(name: str, param_groups: list) -> str:
    lines = [f'{name} (']
    for i, group in enumerate(param_groups):
        lines.append(f'Parameter Group {i}')
        for k in sorted(k for k in group if k != 'params'):
            lines.append(f'    {k}: {group[k]}')
    lines.append(')')
    return '\n'.join(lines)

# --- run it ---
groups = [
    {'params': [], 'lr': 0.001, 'weight_decay': 0.01},
    {'params': [], 'lr': 0.1},
]
print(format_optimizer_repr('SGD', groups))


def _test():
    groups = [
        {'params': [], 'lr': 0.001, 'weight_decay': 0.01},
        {'params': [], 'lr': 0.1},
    ]
    result = format_optimizer_repr('SGD', groups)
    expected = 'SGD (\nParameter Group 0\n    lr: 0.001\n    weight_decay: 0.01\nParameter Group 1\n    lr: 0.1\n)'
    assert result == expected, f'Got:\n{result}\n\nExpected:\n{expected}'
    # Verify keys are alphabetically sorted
    groups2 = [{'params': [], 'z_param': 1, 'a_param': 2}]
    res2 = format_optimizer_repr('Test', groups2)
    lines2 = res2.split('\n')
    # a_param should appear before z_param
    a_idx = next(i for i, l in enumerate(lines2) if 'a_param' in l)
    z_idx = next(i for i, l in enumerate(lines2) if 'z_param' in l)
    assert a_idx < z_idx, 'Keys should be alphabetically sorted'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
