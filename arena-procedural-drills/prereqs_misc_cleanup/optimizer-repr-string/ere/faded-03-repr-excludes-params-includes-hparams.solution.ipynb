## Concept

The `__repr__` of an optimizer must include all scalar hyperparameters and must exclude the `params` list (and any buffer tensors). This distinction matters because the hyperparameters are small, serializable, and meaningful; the param tensors are large and non-serializable in that context.

## Faded exercise 3

Complete `AdamW.__repr__`: return a string in the format `'AdamW(lr=..., betas=..., eps=..., weight_decay=...)'`. Use the instance attributes `self.lr`, `self.betas`, `self.eps`, `self.weight_decay`. Do NOT include `self.params`.

**Fill in:** The f-string return in __repr__ that includes lr, betas, eps, and weight_decay but excludes self.params.

````
import torch as t

class AdamW:
    def __init__(self, params, lr=1e-3, betas=(0.9, 0.999), eps=1e-8, weight_decay=0.01):
        self.params       = list(params)
        self.lr           = lr
        self.betas        = betas
        self.eps          = eps
        self.weight_decay = weight_decay

    def __repr__(self):
        return f'AdamW(lr={self.lr}, betas={self.betas}, eps={self.eps}, weight_decay={self.weight_decay})'

# --- run it ---
opt = AdamW([t.tensor([1.0])], lr=3e-4, betas=(0.9, 0.95), eps=1e-6, weight_decay=0.1)
print(repr(opt))
# AdamW(lr=0.0003, betas=(0.9, 0.95), eps=1e-06, weight_decay=0.1)


def _test():
    import torch as t
    opt = AdamW([t.tensor([1.0])], lr=3e-4, betas=(0.9, 0.95), eps=1e-6, weight_decay=0.1)
    rep = repr(opt)
    assert rep == 'AdamW(lr=0.0003, betas=(0.9, 0.95), eps=1e-06, weight_decay=0.1)', f'Got: {rep}'
    assert 'tensor' not in rep.lower(), 'repr must not include tensor data'
    # Default construction
    opt2 = AdamW([t.tensor([1.0])])
    rep2 = repr(opt2)
    assert 'lr=0.001' in rep2
    assert 'weight_decay=0.01' in rep2


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
