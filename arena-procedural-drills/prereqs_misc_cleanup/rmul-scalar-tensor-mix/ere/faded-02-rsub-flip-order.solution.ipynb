## Concept

Subtraction is not commutative. For `5 - box`, Python calls `box.__rsub__(5)`, which must return `5 - box.value` — the left operand minus self, NOT self minus left.

## Faded exercise 2

The `Num` wrapper has `__sub__` defined. Complete `__rsub__` so that `5 - Num(3)` evaluates to `Num(2)`, preserving the operand order (other minus self).

**Fill in:** return a Num of (other - self.value), with other on the left

````
class Num:
    def __init__(self, value):
        self.value = float(value)
    def __sub__(self, other):
        ov = other.value if isinstance(other, Num) else other
        return Num(self.value - ov)
    def __rsub__(self, other):
        result = Num(other - self.value)
        return result
    def __repr__(self):
        return f'Num({self.value})'


out = 5 - Num(3)

def _test():
    out = 5 - Num(3)
    assert isinstance(out, Num), '5 - Num must return a Num via __rsub__'
    assert abs(out.value - 2.0) < 1e-9, '5 - Num(3) must be Num(2.0), not Num(-2.0)'
    # the non-reflected direction still works and differs
    assert abs((Num(3) - 5).value - (-2.0)) < 1e-9

try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
