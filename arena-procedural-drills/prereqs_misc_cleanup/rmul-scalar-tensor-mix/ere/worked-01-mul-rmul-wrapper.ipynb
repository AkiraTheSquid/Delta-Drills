## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

When Python evaluates `2 * w`, it first asks `int.__mul__(2, w)`, which returns `NotImplemented` because int doesn't know the wrapper type. Python then tries the reflected operation `w.__rmul__(2)`. So a wrapper must define BOTH `__mul__` (for `w * 2`) and `__rmul__` (for `2 * w`). Because scalar multiplication is commutative, `__rmul__` can just delegate to `__mul__`.

## Worked solution

We define `Vec` holding a `.data` tensor. `__mul__(self, other)` returns a new `Vec` of `self.data * other`. `__rmul__(self, other)` delegates to `__mul__` since `scalar * tensor == tensor * scalar`. We then check both `3 * v` and `v * 3` produce the same values, proving the reflected op fills the left-scalar gap. We print both results to show they match.

```
class Vec:
    def __init__(self, data):
        self.data = data
    def __mul__(self, other):
        return Vec(self.data * other)
    def __rmul__(self, other):
        return self.__mul__(other)
    def __repr__(self):
        return f'Vec({self.data.tolist()})'


v = Vec(t.tensor([1.0, 2.0, 3.0]))
left = 3 * v
right = v * 3
print('3 * v =', left)
print('v * 3 =', right)
print('equal:', t.equal(left.data, right.data))
```
