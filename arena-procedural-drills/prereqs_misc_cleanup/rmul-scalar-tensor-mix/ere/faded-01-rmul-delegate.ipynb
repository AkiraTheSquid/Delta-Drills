`2 * w` first tries `int.__mul__`, which returns `NotImplemented`, so Python falls back to `w.__rmul__(2)`. Because scalar multiplication is commutative, `__rmul__` can simply delegate to `__mul__`.

class Vec:
    def __init__(self, data):
        self.data = data
    def __mul__(self, other):
        return Vec(self.data * other)
    def __rmul__(self, other):
        return self.__mul__(other)
    def __repr__(self):
        return f'Vec({self.data.tolist()})'


v = Vec(t.tensor([2.0, 4.0]))
result = 3 * v
