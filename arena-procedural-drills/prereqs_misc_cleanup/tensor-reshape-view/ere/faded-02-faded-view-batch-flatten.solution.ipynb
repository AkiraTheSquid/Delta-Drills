## Concept

`.view(B, -1)` keeps the first (batch) dimension and collapses all remaining dimensions into a single flat feature vector. The `-1` tells PyTorch to infer that dimension automatically from the total number of elements. This is the standard pattern for connecting convolutional layers to linear layers in a neural network.
