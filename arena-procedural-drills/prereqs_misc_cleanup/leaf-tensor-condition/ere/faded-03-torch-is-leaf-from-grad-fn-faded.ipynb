## Concept

In PyTorch's autograd, `tensor.is_leaf` is True when `grad_fn is None`. This includes both trainable parameters (requires_grad=True) and plain constants (requires_grad=False). The `grad_fn` attribute plays the same role as ARENA's `recipe` attribute — it is populated for computed tensors and absent for user-created ones.

import torch as t

def classify_pytorch(tensors):
    out = []
    for x in tensors:
        if x.grad_fn is not None:
            out.append('non-leaf')
        elif x.requires_grad:
