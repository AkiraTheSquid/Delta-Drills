## Concept

A MiniTensor is a leaf when it was created by the user (no Recipe) and the user wants gradients on it (requires_grad=True). Both conditions must hold: a frozen tensor (recipe=None, rg=False) is NOT a training leaf, and an interior tensor (recipe present, rg=True) is NOT a leaf even though it is tracked.
