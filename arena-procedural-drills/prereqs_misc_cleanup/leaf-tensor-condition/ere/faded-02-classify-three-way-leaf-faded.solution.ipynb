Tensors fall into three categories based on `(recipe, requires_grad)`: leaf-trainable (recipe=None, rg=True), leaf-frozen (recipe=None, rg=False), and non-leaf (recipe present). The three-way split is more informative than a binary leaf check because it distinguishes trainable parameters from constants.

## Faded exercise 2

The `classify` function handles `'non-leaf'` and `'leaf-frozen'` already. Fill in the branch that identifies and labels leaf-trainable tensors.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from dataclasses import dataclass
from typing import Any, Optional
import numpy as np

@dataclass
class Recipe:
    func: Any
    args: tuple
    kwargs: dict
    parents: dict

class MiniTensor:
    def __init__(self, array, requires_grad=False):
        self.array = array
        self.requires_grad = requires_grad
        self.recipe: Optional[Recipe] = None

def classify(tensor: MiniTensor) -> str:
    if tensor.recipe is not None:
        return 'non-leaf'
    elif not tensor.requires_grad:
        return 'leaf-frozen'
    else:
        return 'leaf-trainable'


def _test():
    import numpy as np

    arr = np.array([1.0])
    fake = Recipe(np.exp, (arr,), {}, {})

    t1 = MiniTensor(arr.copy(), requires_grad=True)
    t2 = MiniTensor(arr.copy(), requires_grad=False)
    t3 = MiniTensor(arr.copy(), requires_grad=True);  t3.recipe = fake
    t4 = MiniTensor(arr.copy(), requires_grad=False); t4.recipe = fake

    assert classify(t1) == 'leaf-trainable'
    assert classify(t2) == 'leaf-frozen'
    assert classify(t3) == 'non-leaf'
    assert classify(t4) == 'non-leaf'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
