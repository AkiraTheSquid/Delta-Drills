### Accumulate elapsed seconds for one stage

Implement `time_stage(n_iters, sleep_each)`. Loop `n_iters` times; each iteration sleeps `sleep_each` seconds, and you accumulate the measured elapsed time of that sleep into `total`. Use `time.perf_counter()` at both endpoints. Complete the line that adds this iteration's measured elapsed time to `total`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import time

def time_stage(n_iters, sleep_each):
    total = 0.0
    for _ in range(n_iters):
        t0 = time.perf_counter()
        time.sleep(sleep_each)
        total += time.perf_counter() - t0
    return total

print(round(time_stage(3, 0.01), 4))


def _test():
    n, s = 4, 0.01
    total = time_stage(n, s)
    # measured time must be at least the nominal sleep total (sleep overshoots, never undershoots)
    assert total >= n * s, (total, n * s)
    # generous upper bound to catch wildly wrong accumulation (e.g. counting once)
    assert total < n * s + 1.0, total
    # zero iterations -> zero seconds
    assert time_stage(0, s) == 0.0


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
