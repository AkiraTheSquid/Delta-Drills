## Concept

Calling `pbar.set_postfix(**metrics_dict)` updates the live metric display of a tqdm bar. You wrap your iterable with `tqdm(iterable, desc='...')` and record each call in a sidecar list. Any Python-serializable value (string, int, float) can appear as a postfix field.
