## Concept

A single `tqdm(total=N)` bar can span multiple named phases by calling `pbar.set_description('phase_name')` before each phase and `pbar.update(1)` per step. This gives a single progress bar that shows both phases end-to-end, with the label updating as the phase changes.

Implement `faded2_manual_phases(pretrain_losses, finetune_losses)`. Create `pbar = tqdm(total=len(pretrain_losses) + len(finetune_losses))`. In the pretrain phase, call `pbar.set_description('pretrain')`, then for each loss call `pbar.set_postfix(loss=f'{loss:.3f}')` and `pbar.update(1)`. Repeat for the finetune phase with description `'finetune'`. Close the bar and return a dict `{'final_n': pbar.n, 'final_desc': <phase_name_stripped>}` where the description is rstripped of `': '`.

**Fill in:** The pbar.update(1) call inside the finetune loop that advances the bar by one step.

````
from tqdm import tqdm

def faded2_manual_phases(pretrain_losses, finetune_losses):
    pbar = tqdm(total=len(pretrain_losses) + len(finetune_losses))
    pbar.set_description('pretrain')
    for loss in pretrain_losses:
        pbar.set_postfix(loss=f'{loss:.3f}')
        pbar.update(1)
    pbar.set_description('finetune')
    for loss in finetune_losses:
        pbar.set_postfix(loss=f'{loss:.3f}')
        raise NotImplementedError()  # TODO: The pbar.update(1) call inside the finetune loop that advances the bar by one step.
    desc = pbar.desc.rstrip(': ') if pbar.desc else ''
    out = {'final_n': pbar.n, 'final_desc': desc}
    pbar.close()
    return out


def _test():
    pretrain = [0.9, 0.7, 0.5]
    finetune = [0.4, 0.3]
    out = faded2_manual_phases(pretrain, finetune)
    assert out['final_n'] == 5, f"expected 5, got {out['final_n']}"
    assert out['final_desc'] == 'finetune', f"expected 'finetune', got {out['final_desc']}"


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
