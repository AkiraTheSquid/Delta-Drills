## Concept

An offset callback skips all steps before `start_step` (the warm-up window) and only then applies its cadence. The skip is `if step < start_step: continue`.

## Faded exercise 2

Complete `run_offset(n_steps, eval_every, start_step, n_eval, sink)`. The offset-aligned cadence is given; fill in the warm-up gate that skips early steps.

**Fill in:** the guard that skips steps before start_step

````
def run_offset(n_steps, eval_every, start_step, n_eval, sink):
    n_fires = 0
    for step in range(n_steps):
        if step < start_step:
            continue
        if (step - start_step) % eval_every != 0:
            continue
        sink.append({'step': step, 'samples': [f'step={step}-sample={i}' for i in range(n_eval)]})
        n_fires += 1
    return n_fires


def _test():
    sink = []
    fires = run_offset(n_steps=12, eval_every=3, start_step=4, n_eval=1, sink=sink)
    # independent truth: steps >=4 with (step-4)%3==0 -> 4,7,10
    expected = [s for s in range(12) if s >= 4 and (s - 4) % 3 == 0]
    assert expected == [4, 7, 10]
    assert [d['step'] for d in sink] == expected
    assert fires == 3
    # nothing before start_step
    assert all(d['step'] >= 4 for d in sink)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
