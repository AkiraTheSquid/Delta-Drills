## Concept

The 'log every K steps' callback fires whenever the step number is a multiple of `eval_every`, i.e. `step % eval_every == 0`. Step 0 satisfies this, giving a baseline dump before training.

## Faded exercise 1

Complete `run_callback(n_steps, eval_every, n_eval, sink)`. Everything is filled except the condition that decides whether to fire on this step. Fill in that boolean.

**Fill in:** the modulo-K condition that fires on multiples of eval_every

````
def run_callback(n_steps, eval_every, n_eval, sink):
    n_fires = 0
    for step in range(n_steps):
        should_fire = step % eval_every == 0
        if should_fire:
            sink.append({'step': step, 'samples': [f'step={step}-sample={i}' for i in range(n_eval)]})
            n_fires += 1
    return n_fires


def _test():
    sink = []
    fires = run_callback(n_steps=10, eval_every=3, n_eval=2, sink=sink)
    # independent ground truth: multiples of 3 below 10 -> 0,3,6,9
    expected_steps = [s for s in range(10) if s % 3 == 0]
    assert fires == len(expected_steps) == 4
    assert [d['step'] for d in sink] == expected_steps
    assert all(len(d['samples']) == 2 for d in sink)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
