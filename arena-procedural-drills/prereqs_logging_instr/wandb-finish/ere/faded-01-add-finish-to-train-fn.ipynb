## Concept

A complete wandb-instrumented training function must call `wandb.finish()` after the training loop ends. This flushes remaining metrics to wandb's servers and marks the run as complete. Without `finish()`, the run stays open and subsequent `wandb.init()` calls in a sweep can behave unexpectedly.
