## Concept

The examples_seen counter makes runs with different batch sizes comparable on wandb charts. It increments by `batch_size` before each log call, and is passed as the `step=` argument to `wandb.log`. The metric dict and step argument must be consistent: the counter reflects examples processed up to and including the current batch.
