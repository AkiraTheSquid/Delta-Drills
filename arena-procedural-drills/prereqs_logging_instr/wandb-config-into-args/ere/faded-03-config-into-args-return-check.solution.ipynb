## Concept

The `update_args` sweep pattern mutates the dataclass args in place and returns the same object. Callers depend on this mutation contract — they pass `args` to `train()` which calls `update_args` inside, and the caller's reference to `args` immediately reflects the sweep-sampled values.

Complete the blank that performs the entire hasattr/setattr update loop.

**Fill in:** Loop over config.items() and use hasattr/setattr to apply each key-value pair onto args, skipping unknown keys.

````
from dataclasses import dataclass

@dataclass
class Args:
    lr: float = 1e-3
    wd: float = 0.01
    wandb_project: str = 'test'

def update_and_check(args, config):
    for k, v in config.items():
        if hasattr(args, k):
            setattr(args, k, v)
    result = args
    assert result is args, 'must return same object'
    return result

args = Args()
outp = update_and_check(args, {'lr': 9e-5, 'wd': 0.001})
print('lr:', outp.lr, 'wd:', outp.wd)


from dataclasses import dataclass

@dataclass
class Args:
    lr: float = 1e-3
    wd: float = 0.01
    wandb_project: str = 'test'

def update_and_check(args, config):
    for k, v in config.items():
        if hasattr(args, k):
            setattr(args, k, v)
    result = args
    assert result is args
    return result

def _test():
    args = Args()
    result = update_and_check(args, {'lr': 9e-5, 'wd': 0.0005, '_meta': 42})
    assert result is args
    assert result.lr == 9e-5
    assert result.wd == 0.0005
    assert result.wandb_project == 'test'
    assert not hasattr(args, '_meta')


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
