Applying a wandb sweep config to a dataclass requires iterating the config dict and writing each sampled value onto the args object with `setattr`. The key guard — `if hasattr(args, k)` — ensures that wandb-internal metadata keys (like `_wandb`) are silently skipped rather than causing `AttributeError`.
