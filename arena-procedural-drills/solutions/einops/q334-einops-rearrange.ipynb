## Problem

Write solve(img_a, img_b) for two identical-shape (H, W, C) images: produce a single (2·H, W, C) image whose rows INTERLEAVE the two inputs — row 0 of a, row 0 of b, row 1 of a, row 1 of b, … Pattern: einops.rearrange([img_a, img_b], 'b h w c -> (h b) w c').


<details><summary>💡 Hint (click to reveal)</summary>

'b h w c -> (h b) w c' with the images as a list — (h b) interleaves rows.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
img_a = arr[1]
img_b = arr[1]

def solve(img_a, img_b):
    return einops.rearrange([img_a, img_b], 'b h w c -> (h b) w c')

print(solve(img_a, img_b))
```

## Why this works

einops treats a list of arrays as a new leading axis. In the output group (h b), b varies FASTEST — so rows alternate a, b, a, b down the image. '(b h)' would instead stack all of a above all of b; inside-group order is the difference between interleave and concatenate.
