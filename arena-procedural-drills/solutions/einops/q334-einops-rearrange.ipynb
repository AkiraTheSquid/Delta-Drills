## Problem

You are given a list [img, img] of two identical 3D images, each of shape (height, width, channels) = (150, 150, 3). Produce a single 3D array of shape (2*height, width, channels) = (300, 150, 3) by interleaving the two images row-by-row: row 0 of the output is row 0 of the first image, row 1 is row 0 of the second image, row 2 is row 1 of the first image, and so on.


<details><summary>💡 Hint (click to reveal)</summary>

Treat the list as a batch axis, then merge height with batch in that order so rows alternate between the two images.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
img = arr[0]
b = 2

def solve():
    return einops.rearrange([img, img], 'b h w c -> (h b) w c')

print(solve())
```

## Why this works

Stacking the two images gives a (b, h, w, c) tensor. Merging height with batch as '(h b)' interleaves them: for each original row, both images' copies of that row appear consecutively, producing the row-by-row interleaved (300, 150, 3) result.
