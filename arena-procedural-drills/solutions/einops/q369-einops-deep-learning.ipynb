## Problem

Rearrange the channels of a tensor where the channels are initially grouped into two sets. The first group has a size of four, and the second group also has a size of four. The output should have the groups swapped while maintaining the same height and width dimensions.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the channel axis into named groups, then transpose just the two group axes in the output pattern.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(2 * 16 * 8 * 8).reshape(2, 16, 8, 8)
b = 2

def solve():
    return einops.rearrange(x, 'b (g1 g2 c) h w -> b (g2 g1 c) h w', g1=4, g2=4)

print(solve())
```

## Why this works

The 16 channels are decomposed into two grouping factors g1 and g2 (each 4) plus an inner channel axis c. Swapping g1 and g2 in the output pattern reorders the channel blocks while leaving the per-block ordering and spatial dims untouched.
