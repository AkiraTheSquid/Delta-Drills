## Problem

Write a function solve(x, g1, g2) that takes a tensor of shape (b, g1*g2*c, h, w) whose channel axis factors as (g1, g2, c) with g1 slowest, and SWAPS the two group levels: return the same shape with channels reordered as (g2, g1, c) — 'b (g1 g2 c) h w -> b (g2 g1 c) h w'.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the channel axis three ways, swap the two outer factors in the output group, merge back.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(x, g1, g2):
    return einops.rearrange(x, 'b (g1 g2 c) h w -> b (g2 g1 c) h w', g1=g1, g2=g2)


print(solve(t.arange(8.0).reshape(1, 4, 1, 2), 2, 2))
```

## Why this works

'(g1 g2 c)' to '(g2 g1 c)' transposes the group hierarchy without touching the per-group channels — the channel-shuffle operation of ShuffleNet, which lets grouped convolutions exchange information.
