```python
import torch as t
import einops

arr = t.movedim(t.tensor(np.load('/delta_numbers.npy')), 1, -1)
img = arr[0]

def solve(img):
    return einops.repeat(img, 'h w c -> (h 2) (w 2) c')

print(solve(img))
```

## Why this works

Expanding both spatial axes with a trailing factor duplicates each pixel into a 2×2 block — nearest-neighbor upscaling. The factor position matters: (h 2) keeps copies adjacent (blocky upscale); (2 h) would tile whole copies of the image.
