## Problem

For an image with height, width, and color channels, increase both the height and width dimensions by a factor of two by repeating each pixel in a nearest-neighbor manner.


<details><summary>💡 Hint (click to reveal)</summary>

Use repeat with a literal 2 merged into both the height and width output groups.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.repeat(img, 'h w c -> (h 2) (w 2) c')

print(solve())
```

## Why this works

Repeating with a literal factor 2 folded into both spatial composites duplicates each pixel along height and width, giving a 2x nearest-neighbor upsample. Channels are untouched.
