## Problem

Write solve(img) for a single CHW image: repeat it along its HEIGHT axis by a factor of three — three times as tall, width and channels unchanged. Use einops.repeat; the target image shows the canonical fixture's result.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve(img):
    return einops.repeat(img, 'c h w -> c (h 3) w')

print(solve(img))
```

## Why this works

`(h 3)` in the OUTPUT of a repeat expands every height index into 3 consecutive copies: the image gets three times as tall, and channels/width pass through untouched. Group order matters — '(3 h)' would instead stack three full copies of the image.
