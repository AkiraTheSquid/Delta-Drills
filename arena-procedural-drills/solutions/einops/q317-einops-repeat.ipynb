## Problem

Repeat a single CHW image along its height axis by a factor of three, resulting in an output that is three times as tall while preserving the width and color channels.


<details><summary>💡 Hint (click to reveal)</summary>

Use repeat (not rearrange) and place the literal tiling factor next to the height axis inside a grouped (h 3) on the output side.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.repeat(img, 'c h w -> c (h 3) w')

print(solve())
```

## Why this works

einops.repeat with the literal factor 3 on the height axis tiles each row three times, producing an output three times as tall. Channels and width are passed through unchanged, so the image is stacked vertically three times.
