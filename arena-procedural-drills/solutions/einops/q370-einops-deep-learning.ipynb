## Problem

Write a function solve(x) that takes a batch (b, c, h, w) and centers each CHANNEL across the WHOLE BATCH and all pixels: subtract from every value its channel's mean over (b, h, w). Reduce to '1 c 1 1' so the means broadcast — BatchNorm-style centering (contrast per-image centering, which keeps b).


<details><summary>💡 Hint (click to reveal)</summary>

Reduce b, h, AND w to singletons, keeping only c at full size — the statistic is shared across the batch.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x):
    return x - einops.reduce(x, 'b c h w -> 1 c 1 1', 'mean')


print(solve(np.arange(8.0).reshape(2, 1, 2, 2)))
```

## Why this works

reduce(x, 'b c h w -> 1 c 1 1', 'mean') averages each channel over every image and pixel — the BatchNorm mean. Per-instance centering ('b c 1 1') keeps separate stats per image; which axes you collapse defines which normalization you built.
