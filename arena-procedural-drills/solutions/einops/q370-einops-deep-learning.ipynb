## Problem

You are given x, a 4D float32 array of shape (batch, channels, height, width) = (2, 12, 8, 8). Compute, for each channel, a single mean value taken over the batch and both spatial axes (so 12 scalar means in total), then subtract that per-channel mean from every entry of the corresponding channel. The result has the same shape (2, 12, 8, 8) as x.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce to a shape that keeps the channel axis but has singletons elsewhere, so it broadcasts back over x on subtraction.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(2 * 12 * 8 * 8).reshape(2, 12, 8, 8).astype(np.float32)

def solve():
    return x - einops.reduce(x, 'b c h w -> 1 c 1 1', 'mean')

print(solve())
```

## Why this works

reduce to '1 c 1 1' averages over batch and both spatial axes, producing one mean per channel while keeping broadcastable singleton dimensions. Subtracting that from x broadcasts the per-channel mean across every entry of the matching channel.
