## Problem

Write solve(x, h1, w1) for a (B, C, H, W) batch: space-to-depth — carve each spatial map into h1×w1 blocks and fold the block-position indices INTO the channel axis, output (B, h1·w1·C, H/h1, W/w1) with channel order (h1, w1, c). Pattern: 'b c (h h1) (w w1) -> b (h1 w1 c) h w'.

## Reference solution


```python
import torch as t
import einops

x = t.arange(2 * 12 * 8 * 8).reshape(2, 12, 8, 8)
h1 = 4
w1 = 2

def solve(x, h1, w1):
    return einops.rearrange(x, 'b c (h h1) (w w1) -> b (h1 w1 c) h w', h1=h1, w1=w1)

print(solve(x, h1, w1))
```

## Why this works

Space-to-depth: each h1×w1 block position becomes extra channel depth, ordered (h1, w1, c) — channel varies fastest within a block position. Spatial dims shrink by the block factors while channel count multiplies by h1·w1; the inverse pattern is pixel-shuffle.
