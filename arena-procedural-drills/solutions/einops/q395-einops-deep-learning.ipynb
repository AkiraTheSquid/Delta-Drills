## Problem

For a batch of images with multiple channels, rearrange the spatial dimensions by grouping pixels into larger blocks, where the height is divided into segments of four and the width into segments of two, resulting in a new arrangement of channels and spatial dimensions.


<details><summary>💡 Hint (click to reveal)</summary>

Split each spatial axis into an outer index and a block factor, then compose those block factors together with channels into the new channel dimension.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
x = np.arange(2 * 12 * 8 * 8).reshape(2, 12, 8, 8)
b = 2

def solve():
    return einops.rearrange(x, 'b c (h h1) (w w1) -> b (h1 w1 c) h w', h1=4, w1=2)

print(solve())
```

## Why this works

This is a space-to-depth: height splits into (h, h1=4) and width into (w, w1=2), and the block factors h1, w1 are folded into the channel axis, trading spatial resolution for channel depth while batch stays fixed.
