## Problem

For a batch of sequences, divide each sequence into smaller segments, where each segment has a fixed length of four, while maintaining the original depth of each segment.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the sequence-length axis into (n p) and fix the inner factor p to the segment length so einops infers the number of segments.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy')
seq = np.arange(2 * 12 * 3).reshape(2, 12, 3)
b = 2

def solve():
    return einops.rearrange(seq, 'b (n p) d -> b n p d', p=4)

print(solve())
```

## Why this works

Splitting the length axis as (n p) with p=4 factors the 12-long sequence into n=3 segments of 4, adding a new explicit segment axis while keeping batch and depth intact. No data moves; it is a pure reshape of one axis into two.
