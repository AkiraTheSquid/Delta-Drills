## Problem

Write a function solve(seq, p) that takes a batch of sequences of shape (b, t, d) — t divisible by p — and splits each sequence into segments of length p: return shape (b, t//p, p, d), the einops pattern 'b (n p) d -> b n p d'.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(seq, p):
    return einops.rearrange(seq, 'b (n p) d -> b n p d', p=p)


print(solve(np.arange(8.0).reshape(1, 4, 2), 2))
```

## Why this works

'(n p)' with n outer and p inner cuts the sequence into consecutive length-p chunks, so seq.reshape(b, t//p, p, d) is identical. Windowed attention and patch embedding both start with exactly this segmentation.
