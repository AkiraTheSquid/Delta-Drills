## Problem

Write a function solve(tensors) that takes a PYTHON LIST of channels-first images (c, h, w) and returns a single array of shape (h, w, c, len(list)) — stacked AND rearranged so the list axis lands LAST: rearrange(list, 'b c h w -> h w c b').


<details><summary>💡 Hint (click to reveal)</summary>

List-as-batch plus a permutation that sends the new axis to the back — one call.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(tensors):
    return einops.rearrange(tensors, 'b c h w -> h w c b')


print(solve([np.ones((1, 2, 2)), np.zeros((1, 2, 2))]))
```

## Why this works

rearrange(list, 'b c h w -> h w c b') stacks the list and transposes in one step, producing the batch-last layout some visualization and broadcasting patterns expect.
