## Problem

Write a function solve(x_seq) that takes a temporal tensor of shape (b, c, 2*t) and DOWNSAMPLES time by averaging non-overlapping PAIRS: return shape (b, c, t) via 'b c (t two) -> b c t' with 'mean'.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x_seq):
    return einops.reduce(x_seq, 'b c (t two) -> b c t', 'mean', two=2)


print(solve(np.arange(8.0).reshape(1, 1, 8)))
```

## Why this works

'(t two)' splits the sequence into consecutive pairs; 'mean' collapses each pair — 1-D average pooling with window 2, halving the temporal resolution.
