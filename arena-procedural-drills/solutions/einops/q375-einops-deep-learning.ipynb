## Problem

Downsample a temporal sequence of shape (b, c, t*2) by averaging non-overlapping pairs along the time axis. Output shape (b, c, t).


<details><summary>💡 Hint (click to reveal)</summary>

Split the time axis into output-length times a window of 2, then mean-reduce the window.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x_seq = np.arange(2 * 3 * 8).reshape(2, 3, 8).astype(float)

def solve():
    return einops.reduce(x_seq, 'b c (t two) -> b c t', 'mean', two=2)

print(solve())
```

## Why this works

Factoring the time axis as (t two) with two=2 groups consecutive timesteps into non-overlapping pairs. reduce with 'mean' averages each pair, halving the temporal length.
