## Problem

Given img, a 3D array of shape (height, width, channels) representing a single color image in channels-last layout, return a 3D array of shape (channels, height, width) holding the same pixel data in channels-first layout: the channel axis is moved from the trailing position to the leading position, while height and width keep their relative order and follow it.


<details><summary>💡 Hint (click to reveal)</summary>

One rearrange that lists the channel name first and keeps height then width after it.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(img, 'h w c -> c h w')

print(solve())
```

## Why this works

This is a single-image HWC-to-CHW permutation. The channel axis moves from last to first while height and width retain their relative order behind it.
