## Problem

Write solve(imgs) for a (6, H, W, C) channels-last batch: split the batch into a 3-row × 2-column grid, row-major, producing one ((3·H), (2·W), C) image. Pattern: '(r nc) h w ch -> (r h) (nc w) ch' with r=3.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
imgs = arr

def solve(imgs):
    return einops.rearrange(imgs, '(r nc) h w ch -> (r h) (nc w) ch', r=3)

print(solve(imgs))
```

## Why this works

Splitting the 6-image batch as (r nc) with r=3 gives 3 grid rows × 2 columns, row-major. The same two-group output as every montage: rows stack under (r h), columns join under (nc w).
