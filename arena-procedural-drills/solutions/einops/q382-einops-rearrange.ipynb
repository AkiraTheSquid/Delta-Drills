## Problem

You are given arr, a 4D channels-last batch of shape (batch, height, width, channels) = (6, 150, 150, 3) holding 6 digit images. Split the batch axis 6 into r = 3 rows and nc = 2 columns of images, then tile them into a single 3D channels-last canvas of shape (r*height, nc*width, channels) = (450, 300, 3), where the i-th batch image lands in row i // 2 and column i % 2 of the 3x2 grid.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = np.moveaxis(_load_numbers(), 1, -1)

def solve():
    r = 3
    return einops.rearrange(arr, '(r nc) h w ch -> (r h) (nc w) ch', r=r)

print(solve())
```

## Why this works

Factoring the batch of 6 as (r nc) with r=3 infers nc=2, giving a 3x2 layout where image i lands at row i//2, column i%2. Merging (r h) into height and (nc w) into width tiles them into a single (450, 300, 3) canvas.
