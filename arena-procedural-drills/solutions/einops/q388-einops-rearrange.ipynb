## Problem

For a collection of sequences, interchange the axes representing time and batch size, so that each time step is grouped by its corresponding sequence.


<details><summary>💡 Hint (click to reveal)</summary>

Just reorder the named axes in the output pattern so time leads batch; no splitting or merging is needed.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
seq = np.arange(2 * 12 * 3).reshape(2, 12, 3)
b = 2

def solve():
    return einops.rearrange(seq, 'b t d -> t b d')

print(solve())
```

## Why this works

This is a pure axis transpose: naming the input b t d and the output t b d swaps the batch and time axes while leaving the feature dimension d untouched, so each time step is now indexed first.
