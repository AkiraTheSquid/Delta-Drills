## Problem

Write a function solve(weights, d) that takes per-token scalar weights of shape (b, t) and a feature size d, and EXPANDS them to shape (b, t, d) — each scalar copied across all d feature positions, ready to multiply elementwise against a (b, t, d) activation tensor.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(weights, d):
    return einops.repeat(weights, 'b t -> b t d', d=d)


print(solve(np.array([[1.0, 2.0]]), 2))
```

## Why this works

einops.repeat(weights, 'b t -> b t d', d=d) materializes the broadcast that weights[:, :, None] * x would do implicitly — useful when the copied tensor itself is needed, e.g. for masking or concatenation.
