## Problem

You are given weights, a 2D integer array of shape (2, 4) representing per-token scalar weights for a batch of 2 sequences of 4 tokens each, and d=3, the feature dimension. Return a 3D array of shape (2, 4, 3) in (batch, tokens, features) layout where each weight is broadcast across all 3 feature slots so that output[b, t, :] equals weights[b, t] for every feature index.


<details><summary>💡 Hint (click to reveal)</summary>

einops.repeat with a new trailing axis whose size is the feature count broadcasts each weight across features.

</details>

## Reference solution


```python
import numpy as np
import einops

b = 2
t = 4
d = 3
weights = np.arange(b * t).reshape(b, t)

def solve():
    return einops.repeat(weights, 'b t -> b t d', d=d)

print(solve())
```

## Why this works

repeat adds a trailing feature axis 'd' that the input lacks, copying each scalar weight across all d feature slots so output[b, t, :] is constant and equal to weights[b, t].
