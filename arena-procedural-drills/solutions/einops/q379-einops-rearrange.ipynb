## Problem

Write solve(imgs) for a (B, C, H, W) batch: swap the HEIGHT and WIDTH dimensions of every image (spatial transpose) — output (B, C, W, H). Pattern: 'b c h w -> b c w h'.

## Reference solution


```python
import torch as t
import einops

arr = t.tensor(np.load('/delta_numbers.npy'))
imgs = arr

def solve(imgs):
    return einops.rearrange(imgs, 'b c h w -> b c w h')

print(solve(imgs))
```

## Why this works

Swapping the last two axes transposes every image's spatial layout — same operation as `.swapaxes(2, 3)`, but the einops pattern documents intent. Note the output SHAPE only changes when H ≠ W; the content always transposes.
