## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers()
b = 2

def solve():
    return einops.rearrange(arr, 'b c h w -> b c w h')

print(solve())
```

## Why this works

Swapping h and w in the output pattern transposes the two spatial axes of every image while leaving batch and channel axes in place. It is a pure axis permutation, not a reshape.
