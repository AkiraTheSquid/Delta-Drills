## Problem

Write a function solve(arr) that takes a channels-last batch of shape (b, h, w, c) and concatenates the images SIDE BY SIDE: return shape (h, b*w, c) where image n occupies columns n*w through (n+1)*w - 1 — the einops pattern 'b h w c -> h (b w) c'.


<details><summary>💡 Hint (click to reveal)</summary>

The batch index must vary SLOWER than width inside the merged axis — '(b w)' puts each whole image before the next.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b h w c -> h (b w) c')


print(solve(np.arange(8.0).reshape(2, 1, 2, 2)))
```

## Why this works

rearrange(arr, 'b h w c -> h (b w) c') merges batch into width with b outermost, laying the images left-to-right. In numpy: transpose the batch axis next to width, then reshape — the '(b w)' order is what keeps each image contiguous instead of interleaving columns.
