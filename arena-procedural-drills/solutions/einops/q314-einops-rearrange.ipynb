## Problem

Rearrange a batch of images by first moving the batch dimension into the width axis, effectively combining them, and then returning to the original structure with the batch dimension restored.


<details><summary>💡 Hint (click to reveal)</summary>

Fold batch into width with (b w), then undo it by splitting the width axis back into (b w) while supplying b so einops knows the split sizes.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    tmp = einops.rearrange(arr, 'b h w c -> h (b w) c')
    return einops.rearrange(tmp, 'h (b w) c -> b h w c', b=b)

print(solve())
```

## Why this works

The first rearrange folds the batch into the width axis to lay images side by side; the second splits that combined axis back out using the known batch size b. Because the operations are exact inverses, the round trip reconstructs the original array.
