## Problem

1D average pooling over time with non-overlapping window of length 4 on a tensor of shape (b, c, t*4). Output shape (b, c, t).


<details><summary>💡 Hint (click to reveal)</summary>

Factor time into output-length times a window of 4 and average over the window.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x_seq = np.arange(2 * 3 * 12).reshape(2, 3, 12).astype(float)

def solve():
    return einops.reduce(x_seq, 'b c (t w) -> b c t', 'mean', w=4)

print(solve())
```

## Why this works

The time axis factors as (t w) with w=4, grouping every four consecutive timesteps into a non-overlapping window. reduce with 'mean' averages each window, shrinking the temporal length by a factor of four.
