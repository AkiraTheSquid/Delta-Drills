## Problem

Write a function solve(x_seq, w) that takes a temporal tensor (b, c, t*w) and average-pools time with non-overlapping windows of length w: return shape (b, c, t) via 'b c (t w) -> b c t' with 'mean'.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x_seq, w):
    return einops.reduce(x_seq, 'b c (t w) -> b c t', 'mean', w=w)


print(solve(np.arange(8.0).reshape(1, 1, 8), 4))
```

## Why this works

'(t w)' with w as the pooled factor averages each length-w block — 1-D average pooling with an arbitrary stride-equals-window size, w=1 degenerating to the identity.
