## Problem

Write a function solve(arr) that takes a channels-last batch (b, h, w, c) and returns the channels-FIRST batch (b, c, h, w).


<details><summary>💡 Hint (click to reveal)</summary>

Move the trailing channel letter to slot two — the inverse of the bchw-to-bhwc conversion.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b h w c -> b c h w')


print(solve(t.arange(6.0).reshape(1, 1, 2, 3)))
```
