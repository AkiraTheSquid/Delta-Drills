## Problem

You are given arr, a 4D array of shape (batch, height, width, channels) = (6, 150, 150, 3) representing a batch of channels-last digit images. Reorder the axes so the channel axis comes immediately after the batch axis, producing a 4D channels-first array of shape (batch, channels, height, width) = (6, 3, 150, 150) with all pixel values preserved.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    return einops.rearrange(arr, 'b h w c -> b c h w')

print(solve())
```

## Why this works

Converting a batch from NHWC to BCHW is a pure permutation: the channel name moves to right after batch while height and width keep their order, with all pixel values preserved.
