## Problem

You are given img, a 3D channels-first image of shape (channels, height, width) = (3, 150, 150) holding a single digit. Produce a 3D array of shape (channels, 2*height, width) = (3, 300, 150) whose top half along the height axis is a copy of img and whose bottom half is another copy of img, so the digit appears twice stacked vertically with channels and width preserved.


<details><summary>💡 Hint (click to reveal)</summary>

einops.repeat with a literal multiplier fused into the height axis stacks two vertical copies.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]
ws = 2

def solve():
    return einops.repeat(arr[0], 'c h w -> c (2 h) w')

print(solve())
```

## Why this works

Using a literal factor in the output pattern, '(2 h)' repeats the image twice along the height axis, stacking two copies of the digit vertically while channels and width are preserved, giving (3, 300, 150).
