## Problem

Write a function solve(img, k) that takes a channels-first image (c, h, w) and returns shape (c, k*h, w) with the WHOLE image stacked k times vertically — copy after copy, via repeat's '(k h)' grouping where the new index varies SLOWEST. (A companion drill stretches rows instead — order inside the parentheses is the whole difference.)


<details><summary>💡 Hint (click to reveal)</summary>

'(k h)' with k OUTER means complete copies follow each other — tiling, not stretching.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(img, k):
    return einops.repeat(img, 'c h w -> c (k h) w', k=k)


print(solve(t.arange(4.0).reshape(1, 2, 2), 2))
```

## Why this works

In '(k h)', k varies slower than h, so the output runs through the whole image before repeating it: t.tile along height. The mirror pattern '(h k)' interleaves copies row by row — stretching. The parenthesis order IS the semantics.
