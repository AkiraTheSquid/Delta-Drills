## Problem

You are given seq, a 3D array of shape (batch, tokens, heads*embedding_dim) = (2, 12, 8) representing a transformer sequence whose trailing axis is the concatenation of 8 attention heads. With heads = 8, split that trailing axis into a heads axis and a per-head embedding_dim axis (so embedding_dim = 1 here) and move the heads axis next to the batch, producing a 4D array of shape (batch, heads, tokens, embedding_dim) = (2, 8, 12, 1).

Factor the last axis into heads and per-head dim, then reorder so heads sits right after batch.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
seq = np.arange(2 * 12 * 8).reshape(2, 12, 8)
b = 2

def solve():
    return einops.rearrange(seq, 'b t (nh d) -> b nh t d', nh=8)

print(solve())
```

## Why this works

The trailing concatenated axis is split into (nh d) with nh=8 heads (so d=1), and the heads axis is moved next to batch ahead of tokens, giving the standard multi-head attention layout (batch, heads, tokens, head_dim).
