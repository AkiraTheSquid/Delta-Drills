## Problem

Write a function solve(seq, nh) that takes a transformer sequence of shape (b, t, nh*d) — the head outputs concatenated per token, head index slowest — and SPLITS the heads back out: return shape (b, nh, t, d) via 'b t (nh d) -> b nh t d'.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(seq, nh):
    return einops.rearrange(seq, 'b t (nh d) -> b nh t d', nh=nh)


print(solve(np.arange(8.0).reshape(1, 2, 4), 2))
```

## Why this works

'b t (nh d) -> b nh t d' recovers the per-head layout attention math needs: reshape to expose the head factor, then transpose it before time. Together with 'b nh t d -> b t (nh d)' it round-trips a transformer block's shape choreography.
