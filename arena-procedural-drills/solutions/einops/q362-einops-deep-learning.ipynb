## Problem

Write a function solve(x, split) that takes a tensor of shape (b, c*split, h, w) whose channels INTERLEAVE `split` groups (group index FASTEST: channel order is g0, g1, g0, g1, ...) and returns shape (split, b, c, h, w) via 'b (c split) h w -> split b c h w'. (Contrast the consecutive-groups drill, where the group index is slowest.)


<details><summary>💡 Hint (click to reveal)</summary>

'(c split)' puts the group index INNERMOST — every other channel belongs to the other group.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(x, split):
    return einops.rearrange(x, 'b (c split) h w -> split b c h w', split=split)


print(solve(t.arange(8.0).reshape(1, 4, 1, 2), 2))
```

## Why this works

With split innermost, channels alternate between groups (deinterleaving), whereas '(split c)' would cut consecutive halves. Same shapes, different data — parenthesis order inside a factored axis is a real semantic choice, and this pair of drills is exactly that contrast.
