## Problem

You are given x, a 4D array of shape (batch, channels, height, width) whose channels axis has an even length and is interpreted as interleaved pairs (channels = c_in * 2, with the two members of each pair sitting at consecutive channel positions). Return a 5D array of shape (2, batch, c_in, height, width) whose leading axis selects between the two interleaved sub-streams: index 0 collects every even-indexed channel and index 1 collects every odd-indexed channel.


<details><summary>💡 Hint (click to reveal)</summary>

Split the channel axis so the size-2 factor is the inner (fastest) index, then promote that factor to the front.

</details>

## Reference solution


```python
import numpy as np
import einops

x = np.arange(2 * 12 * 8 * 8).reshape(2, 12, 8, 8)
b = 2

def solve():
    return einops.rearrange(x, 'b (c_in split) h w -> split b c_in h w', split=2)

print(solve())
```

## Why this works

Factoring the channel axis as `(c_in split)` puts the size-2 `split` as the fastest-varying inner index, so it collects the interleaved (even/odd) channel members. Moving `split` to the front gives a leading axis selecting between the two interleaved sub-streams.
