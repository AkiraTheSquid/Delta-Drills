## Problem

For an image represented by its height, width, and color channels, average the color channels while collapsing the height dimension, resulting in a structure that retains the width and color channels.


<details><summary>💡 Hint (click to reveal)</summary>

Use einops.reduce with a pattern that keeps width and channels while omitting height, and pick the averaging reduction.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy').astype(np.float32)
img = arr[0]

def solve():
    return einops.reduce(img, 'h w c -> w c', 'mean')

print(solve())
```

## Why this works

einops.reduce with the 'mean' operation collapses any axis that appears in the input pattern but not the output pattern. Here 'h' is dropped, so each (w, c) cell becomes the mean over all height rows, yielding a (width, channels) array.
