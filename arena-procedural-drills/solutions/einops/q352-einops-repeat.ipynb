## Problem

Given a 3D image array of shape (channels, height, width) representing a single color image in channels-first layout, return a 3D array of shape (channels, height * 2, width) in which each original row of pixels is duplicated into two consecutive identical rows, stretching the image vertically to twice its original height while leaving channels and width unchanged.


<details><summary>💡 Hint (click to reveal)</summary>

Use repeat on the channels-first image with a literal 2 merged into the height axis only.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.repeat(arr[0], 'c h w -> c (h 2) w')

print(solve())
```

## Why this works

Folding a literal 2 into the height composite `(h 2)` duplicates each row into two consecutive identical rows, doubling vertical extent. Channels and width are preserved unchanged.
