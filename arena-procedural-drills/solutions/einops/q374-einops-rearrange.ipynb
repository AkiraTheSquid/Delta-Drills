## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(imgs):
    return einops.rearrange(imgs, 'b h w c -> h (b w) c')


print(solve([t.zeros((2, 1, 1)), t.ones((2, 1, 1))]))
```
