## Problem

Write a function solve(imgs) that takes a PYTHON LIST of channels-last images (h, w, c) and concatenates them SIDE BY SIDE: return shape (h, len(list)*w, c) — the list index merging straight into width via 'b h w c -> h (b w) c'.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(imgs):
    return einops.rearrange(imgs, 'b h w c -> h (b w) c')


print(solve([np.zeros((2, 1, 1)), np.ones((2, 1, 1))]))
```

## Why this works

rearrange(list, 'b h w c -> h (b w) c') is np.concatenate(list, axis=1): each image occupies its own contiguous span of columns, in list order.
