## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = np.moveaxis(_load_numbers(), 1, -1)
b = 2

def solve():
    return einops.rearrange(arr, 'b h w c -> h (b w) c')

print(solve())
```

## Why this works

Folding (b w) on the output lays the batch images out left-to-right along a single widened width axis, while height and channels are preserved. The result is one tall strip whose total width is the sum of the individual image widths.
