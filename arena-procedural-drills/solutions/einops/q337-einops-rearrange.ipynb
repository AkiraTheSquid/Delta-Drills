## Problem

Write a function solve(merged, b) that takes an image strip of shape (h, b*w, c) — b images concatenated side by side — and SPLITS it back into the batch: return shape (b, h, w, c) via the inverse pattern 'h (b w) c -> b h w c'.


<details><summary>💡 Hint (click to reveal)</summary>

Factoring '(b w)' on the INPUT side is the undo of merging it on the output side — you must tell einops b (or w) since the strip alone is ambiguous.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(merged, b):
    return einops.rearrange(merged, 'h (b w) c -> b h w c', b=b)


print(solve(np.arange(8.0).reshape(1, 4, 2), 2))
```

## Why this works

'h (b w) c -> b h w c' cuts the wide strip into b equal segments (b outermost matches how they were joined) and lifts them back into a batch axis. Merges lose the factorization, so the inverse pattern needs b passed explicitly.
