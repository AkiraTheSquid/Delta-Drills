## Problem

Transform a batch of images by first rearranging the dimensions to group the batch and width together, then return to the original arrangement while ensuring the transformation is correct.


<details><summary>💡 Hint (click to reveal)</summary>

Compose batch into width, then invert with the same merged pattern split apart, supplying the batch size so einops can factor it.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    tmp = einops.rearrange(arr, 'b h w c -> h (b w) c')
    return einops.rearrange(tmp, 'h (b w) c -> b h w c', b=b)

print(solve())
```

## Why this works

The first rearrange folds batch into the width axis to form a wide composite; the second reverses it by splitting '(b w)' back into separate batch and width axes (binding b so the split is unambiguous), recovering the original tensor exactly.
