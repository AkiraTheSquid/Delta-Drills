## Problem

Given a (batch, dim) class-embedding tensor 'cls' and an integer t, broadcast each embedding across t time steps to produce shape (batch, t, dim).


<details><summary>💡 Hint (click to reveal)</summary>

Use einops.repeat to insert a new time axis between batch and dim, giving its size as a keyword.

</details>

## Reference solution


```python
import numpy as np
import einops

b = 2
t = 4
cls = np.arange(b * 3).reshape(b, 3)

def solve():
    return einops.repeat(cls, 'b d -> b t d', t=t)

print(solve())
```

## Why this works

einops.repeat introduces a new axis 't' (size given by t=t) that is not present in the input, broadcasting each (b, d) embedding identically across all t time steps to produce (b, t, d).
