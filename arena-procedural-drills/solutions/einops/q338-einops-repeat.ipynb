## Problem

Write a function solve(cls, t) that takes a class-embedding tensor of shape (b, d) and an integer t, and BROADCASTS each embedding across t time steps: return shape (b, t, d) where every time slice of item n equals cls[n].

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(cls, t):
    return einops.repeat(cls, 'b d -> b t d', t=t)


print(solve(np.array([[1.0, 2.0]]), 3))
```

## Why this works

einops.repeat(cls, 'b d -> b t d', t=t) inserts a new time axis and copies each embedding along it — np.repeat(cls[:, None, :], t, axis=1), or a broadcast view. Prepending a class token to every sequence position starts exactly like this.
