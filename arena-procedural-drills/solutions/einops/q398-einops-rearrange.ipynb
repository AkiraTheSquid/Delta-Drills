## Problem

Reconstruct an image from a batch of color channels organized into 4x4 patches by rearranging the patches into a complete image format, specifying the height and width of the final image.


<details><summary>💡 Hint (click to reveal)</summary>

Split the patch axis into (h w) using the given grid sizes, then compose each grid index with its corresponding patch axis to rebuild the spatial dims.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
h, w = 15, 15
patches = einops.rearrange(arr[0], 'c (h p1) (w p2) -> (h w) c p1 p2', p1=10, p2=10)

def solve():
    return einops.rearrange(patches, '(h w) c p1 p2 -> c (h p1) (w p2)', h=h, w=w)

print(solve())
```

## Why this works

The patch index (h w) is split back into grid coordinates using the supplied h and w, then each grid coordinate is interleaved with its in-patch offset via (h p1) and (w p2) to stitch the patches back into the full channels-first image.
