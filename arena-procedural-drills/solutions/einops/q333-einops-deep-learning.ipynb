## Problem

Write a function solve(tensors) that takes a PYTHON LIST of channels-first images, each of shape (c, h, w), and returns a single channels-LAST batch tensor of shape (len(list), h, w, c) — einops.rearrange applied directly to the list stacks it as the new leading axis.


<details><summary>💡 Hint (click to reveal)</summary>

einops accepts a LIST of tensors: the list index becomes the pattern's first axis, so stacking and permuting happen in one call.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(tensors):
    return einops.rearrange(tensors, 'b c h w -> b h w c')


print(solve([t.ones((1, 2, 2)), t.zeros((1, 2, 2))]))
```

## Why this works

rearrange(list, 'b c h w -> b h w c') treats the list as the b axis — t.stack(list) fused with the channels-last transpose. One call replaces the stack-then-transpose two-step.
