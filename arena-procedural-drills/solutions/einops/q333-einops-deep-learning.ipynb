## Problem

You are given list_of_tensors, a 4D array of shape (batch, channels, height, width) holding several channels-first images stacked along the leading axis. Return a 4D array of shape (batch, height, width, channels) in which the channels axis has been moved from position 1 to the trailing position, converting each image from channels-first to channels-last layout while keeping the batch ordering.


<details><summary>💡 Hint (click to reveal)</summary>

Permute the four named axes so the channel name moves to the end, leaving batch first.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
b = 2

list_of_tensors = arr[:b]

def solve():
    return einops.rearrange(list_of_tensors, 'b c h w -> b h w c')

print(solve())
```

## Why this works

Converting channels-first to channels-last is a pure permutation: the channel axis named 'c' is moved from position 1 to the trailing slot while batch, height, and width keep their order.
