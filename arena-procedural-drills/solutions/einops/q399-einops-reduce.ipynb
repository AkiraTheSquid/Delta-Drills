## Problem

Write solve(imgs) for a float (B, H, W, C) batch: grayscale each image by AVERAGING across the color channels, keeping a trailing singleton channel — output (B, H, W, 1). Pattern: einops.reduce 'b h w c -> b h w ()' with 'mean'.


<details><summary>💡 Hint (click to reveal)</summary>

einops.reduce 'b h w c -> b h w ()', 'mean' — average channels, keep the axis.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy').astype(np.float32), 1, -1)
imgs = arr

def solve(imgs):
    return einops.reduce(imgs, 'b h w c -> b h w ()', 'mean')

print(solve(imgs))
```

## Why this works

Reducing c with 'mean' averages the color channels into luminance; writing () in the output keeps a singleton channel axis so the result stays 4-D — downstream image code (and the renderer) expects that trailing dimension.
