## Problem

For a batch of images with color channels, compute the grayscale version of each image by averaging the values across the color channels, resulting in a single channel output for each image while maintaining the height and width.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce across channels with a mean, but write a () in the output so the channel axis collapses to size 1 rather than disappearing.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy').astype(np.float32), 1, -1)
b = 2

def solve():
    return einops.reduce(arr, 'b h w c -> b h w ()', 'mean')

print(solve())
```

## Why this works

Reducing over the channel axis with 'mean' and writing () in its place averages the color channels into a single retained singleton channel, yielding a per-pixel grayscale value while keeping batch, height, and width.
