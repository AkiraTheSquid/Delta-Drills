## Problem

Write solve(imgs) for a (12, C, H, W) batch of channels-first images: tile them into a single CHW image laid out as a 4-row × 3-column grid (row-major: image i lands at row i//3, column i%3). Pattern: '(g1 g2) c h w -> c (g1 h) (g2 w)' with g1=4.

## Reference solution


```python
import numpy as np
import einops

_base = np.load('/delta_numbers.npy')
imgs = np.concatenate([_base] * (-(-12 // _base.shape[0])))[:12]

def solve(imgs):
    return einops.rearrange(imgs, '(g1 g2) c h w -> c (g1 h) (g2 w)', g1=4)

print(solve(imgs))
```

## Why this works

Reading the batch axis as (g1 g2) with g1=4 makes image i land at grid row i//3, column i%3 (row-major). The output groups (g1 h) and (g2 w) then concatenate the grid rows vertically and columns horizontally into one big CHW image.
