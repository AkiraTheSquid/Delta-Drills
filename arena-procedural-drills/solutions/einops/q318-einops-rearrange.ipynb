## Problem

You are given arr, a 4D array of shape (batch, channels, height, width) = (12, 3, 150, 150) holding 12 channels-first digit images. Tile these 12 images into a single composite image laid out as a 4-row by 3-column grid (rows fill first), producing a 3D channels-first array of shape (channels, 4 * height, 3 * width) = (3, 600, 450).

Factor the batch axis into grid rows and columns, then fuse the row factor with height and the column factor with width in the output.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

_base_arr = np.load('/delta_numbers.npy')
_arr_repeats = -(-12 // _base_arr.shape[0])
arr = np.concatenate([_base_arr] * _arr_repeats, axis=0)
arr = arr[:12]

def solve():
    return einops.rearrange(arr[:12], '(g1 g2) c h w -> c (g1 h) (g2 w)', g1=4)

print(solve())
```

## Why this works

Splitting the batch of 12 into a grid (g1 g2) with g1=4 gives 4 rows and 3 columns (row-major), then merging g1 into height and g2 into width tiles the images into one composite channels-first canvas of shape (c, 4h, 3w).
