## Problem

For a single image, interchange the color channels and width dimensions, resulting in a new arrangement where the color information is positioned between the height and width.


<details><summary>💡 Hint (click to reveal)</summary>

Just reorder the axis names on the output side to put channels between height and width; no splitting or merging is involved.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(img, 'h w c -> h c w')

print(solve())
```

## Why this works

This is a simple transpose that swaps the width and channel axes, placing the color axis between height and width. einops.rearrange performs the permutation by reordering the named axes in the output pattern.
