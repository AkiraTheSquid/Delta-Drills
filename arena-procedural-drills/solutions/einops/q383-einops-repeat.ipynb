## Problem

For a single CHW image, duplicate each color channel so that the output has twice as many channels, while keeping the height and width dimensions unchanged.


<details><summary>💡 Hint (click to reveal)</summary>

Use repeat with a new factor r=2 merged in front of the channel axis to double the channels.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers()
img = arr[0]

def solve():
    return einops.repeat(img, 'c h w -> (r c) h w', r=2)

print(solve())
```

## Why this works

einops.repeat introduces a new axis r=2 and tiles the channel axis, doubling the channel count. Writing (r c) places the repeat factor outside c, so the whole channel block is duplicated as a contiguous copy while height and width are unchanged.
