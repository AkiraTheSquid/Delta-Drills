## Problem

You are given x, a 4D array of shape (batch, channels, height, width) of channels-first images, plus integer subgrid factors hs and ws (here both 2) that evenly divide height and width. Produce the packed array of shape (hs * ws * batch, channels, height / hs, width / ws) by splitting each image into an hs-by-ws spatial grid and folding both subgrid axes plus batch into the leading axis. Return the packed array.


<details><summary>💡 Hint (click to reveal)</summary>

Fold the spatial subgrid factors and batch into one leading axis, then run the mirror-image pattern to undo it, keeping hs and ws fixed.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy')
x = arr[:2]
hs = 2
ws = 2
b = 2

def solve():
    y = einops.rearrange(x, 'b c (h hs) (w ws) -> (hs ws b) c h w', hs=hs, ws=ws)
    return einops.rearrange(y, '(hs ws b) c h w -> b c (h hs) (w ws)', hs=hs, ws=ws)

print(solve())
```

## Why this works

The first rearrange tiles each image into an hs-by-ws spatial grid and folds those subgrid factors plus batch into the leading axis. The second rearrange applies the exact inverse pattern, so the composition is the identity and recovers the original BCHW array.
