## Problem

Write a function solve(x) that takes a 4-D tensor (b, c, h, w) and returns the SUM of every element as a 0-D result — einops.reduce with the empty output pattern '->'.


<details><summary>💡 Hint (click to reveal)</summary>

Every letter vanishes — the empty right side is the grand total.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(x):
    return einops.reduce(x, 'b c h w -> ', 'sum')


print(solve(t.ones((2, 12, 8, 8))))
```

## Why this works

reduce(x, 'b c h w -> ', 'sum') collapses all four axes to a single number, x.sum() in pattern form — the einops spelling of a global reduction.
