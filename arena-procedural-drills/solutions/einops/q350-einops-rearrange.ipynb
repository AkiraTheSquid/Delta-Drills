## Problem

Write solve(img) for an (H, W, C) image with H and W even: within every non-overlapping 2×2 patch, TRANSPOSE the patch (swap its within-patch row and column). Pattern: '(h p1) (w p2) c -> (h p2) (w p1) c' with p1=2, p2=2. Output shape is unchanged; content moves.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
img = arr[0]

def solve(img):
    return einops.rearrange(img, '(h p1) (w p2) c -> (h p2) (w p1) c', p1=2, p2=2)

print(solve(img))
```

## Why this works

The input groups carve the image into 2×2 patches with within-patch coordinates (p1, p2). Swapping them in the output transposes each patch in place — the patch GRID stays put, but every patch's contents flip across their diagonal. Shape is unchanged; only content moves.
