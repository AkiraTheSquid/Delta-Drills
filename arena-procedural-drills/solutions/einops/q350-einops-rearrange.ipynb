## Problem

For an image with spatial dimensions and color channels, rearrange each 2x2 patch so that the width and height of the patches are swapped while keeping the color channels in place.


<details><summary>💡 Hint (click to reveal)</summary>

Factor each spatial axis into an outer and a size-2 inner index, then swap the two inner indices on the output.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
img = arr[0]

def solve():
    return einops.rearrange(img, '(h p1) (w p2) c -> (h p2) (w p1) c', p1=2, p2=2)

print(solve())
```

## Why this works

Splitting height and width into 2x2 patches via `(h p1)` and `(w p2)` exposes the intra-patch indices p1 and p2. Swapping them on the output side `(h p2) (w p1)` transposes each patch's row and column while channels stay put.
