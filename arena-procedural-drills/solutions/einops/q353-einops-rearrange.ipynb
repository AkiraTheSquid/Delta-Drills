## Problem

For a batch of sequences divided into chunks, combine the chunk and sequence dimensions into a single dimension while preserving the depth, resulting in a structure where the sequence length is the product of the number of chunks and the length of each chunk.


<details><summary>💡 Hint (click to reveal)</summary>

Use rearrange and combine the two middle axes into one composite, leaving batch and depth alone.

</details>

## Reference solution


```python
import numpy as np
import einops

seq_chunks = np.arange(2 * 3 * 4 * 5).reshape(2, 3, 4, 5)

def solve():
    return einops.rearrange(seq_chunks, 'b n p d -> b (n p) d')

print(solve())
```

## Why this works

Merging the chunk axis n and per-chunk length p into a single composite `(n p)` flattens the two into one sequence dimension of length n*p, while batch and depth are preserved. The C-order merge keeps chunks contiguous.
