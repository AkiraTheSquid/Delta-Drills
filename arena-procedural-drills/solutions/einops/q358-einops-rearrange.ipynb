## Problem

For a batch of images with height, width, and color channels, first remove the color channels dimension by squeezing it out, and then add it back as a singleton dimension, returning to the original shape.


<details><summary>💡 Hint (click to reveal)</summary>

Use the () token in the rearrange pattern to add a singleton axis, then a second rearrange that matches and drops it.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    tmp = einops.rearrange(arr, 'b h w c -> b () h w c')
    return einops.rearrange(tmp, 'b () h w c -> b h w c')

print(solve())
```

## Why this works

The `()` token inserts a size-1 axis (unsqueeze) and, when matched on the input side, removes it (squeeze). Inserting then removing returns the array to its original shape, demonstrating both directions of singleton-axis handling.
