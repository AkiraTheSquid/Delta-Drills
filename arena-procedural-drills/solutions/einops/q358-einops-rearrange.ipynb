## Problem

Write a function solve(arr) that takes a channels-last batch (b, h, w, c) and INSERTS a singleton axis right after the batch axis: return shape (b, 1, h, w, c) via the '()' output group — 'b h w c -> b () h w c'.


<details><summary>💡 Hint (click to reveal)</summary>

An empty '()' group on the output side mints a length-1 axis at that position.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b h w c -> b () h w c')


print(solve(np.ones((2, 2, 2, 1))))
```

## Why this works

rearrange(arr, 'b h w c -> b () h w c') is arr[:, None] / np.expand_dims(arr, 1): a singleton axis for broadcasting or interface compatibility. The mirror pattern with '()' on the INPUT side removes it again.
