## Problem

Split a tensor of shape (b, 4*bbox, h, w) into four tensors (one per coordinate) each of shape (b, bbox, h, w). Return them stacked along a new leading axis with shape (4, b, bbox, h, w).


<details><summary>💡 Hint (click to reveal)</summary>

Split the 4*bbox channel axis into a leading coord=4 factor, hoist it to the front, then unpack the result into the four coordinate tensors.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(2 * 8 * 3 * 3).reshape(2, 8, 3, 3)

def solve():
    bbox_x, bbox_y, bbox_w, bbox_h = einops.rearrange(
        x, 'b (coord bbox) h w -> coord b bbox h w', coord=4)
    return bbox_x, bbox_y, bbox_w, bbox_h

print(solve())
```

## Why this works

The channel axis 8 is split into (coord bbox) with coord=4, then coord is moved to the front so the array can be unpacked into four (b, bbox, h, w) tensors. Returning them as a tuple is equivalent to the stacked (4, b, bbox, h, w) array since stacking the four pieces reproduces it.
