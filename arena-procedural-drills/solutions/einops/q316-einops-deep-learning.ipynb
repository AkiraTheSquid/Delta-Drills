## Problem

Write a function solve(x, coord) that takes a tensor of shape (b, coord*k, h, w) whose channel axis interleaves `coord` groups of k channels (group index varying SLOWEST), and returns shape (coord, b, k, h, w) with the groups split out and moved to the front — 'b (coord k) h w -> coord b k h w'.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the channel axis with the group index OUTERMOST, then transpose that new axis to the front.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x, coord):
    return einops.rearrange(x, 'b (coord k) h w -> coord b k h w', coord=coord)


print(solve(np.arange(8.0).reshape(1, 4, 1, 2), 4))
```

## Why this works

'(coord k)' splits the channels into coord consecutive groups of k; naming coord first in the output moves it ahead of batch. Unpacking the result as x0, x1, ... then destructures a stacked prediction head (like bbox coordinates) in one line.
