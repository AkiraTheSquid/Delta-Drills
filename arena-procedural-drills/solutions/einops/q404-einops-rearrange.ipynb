## Problem

For an image with color channels and spatial dimensions, extract non-overlapping patches of size 4 by 4, resulting in a new arrangement where the image is represented as a collection of these patches along with the color channels.


<details><summary>💡 Hint (click to reveal)</summary>

Factor height and width into (grid, patch) pairs, then compose the two grid indices into one leading axis while keeping channel and patch axes separate.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(img, 'c (h p1) (w p2) -> (h w) c p1 p2', p1=10, p2=10)

print(solve())
```

## Why this works

Each spatial axis is split into a patch-grid index and an in-patch offset (p1, p2 = 10), and the two grid indices are composed into a single leading patch-count axis, listing all non-overlapping patches with channels and patch offsets retained.
