## Problem

You are given arr, a 4D array of shape (batch, height, width, channels) = (2, 3, 8, 5) in BHWC layout. Reinterpret the width axis as a (remaining_width, factor) pair where factor = 4 (so remaining_width = 2), then merge that factor of 4 into the height axis as its outer index, producing a 4D array of shape (batch, factor * height, remaining_width, channels) = (2, 12, 2, 5).

Split width into a remaining-width and a factor of 4, then compose that factor as the outer part of a new height axis.

## Reference solution


```python
import numpy as np
import einops

arr = np.arange(2 * 3 * 8 * 5).reshape(2, 3, 8, 5)

def solve():
    return einops.rearrange(arr, 'b h (w w4) c -> b (w4 h) w c', w4=4)

print(solve())
```

## Why this works

The width axis is factored into (w, w4) with w4=4, then w4 is folded into height as its outer index via (w4 h); this moves the factor-of-4 block from width into height, giving shape (2, 12, 2, 5).
