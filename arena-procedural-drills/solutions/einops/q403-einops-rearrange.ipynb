## Problem

Write a function solve(arr, w4) that takes a BHWC batch (b, h, w*w4, c) and folds width chunks into height: factor the width axis as (w, w4) with w4 INNER, then merge w4 into height AHEAD of h — return shape (b, w4*h, w, c) via 'b h (w w4) c -> b (w4 h) w c'.


<details><summary>💡 Hint (click to reveal)</summary>

Factor width, then merge the inner factor into height with it OUTERMOST in '(w4 h)' — the folded strips stack above one another.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(arr, w4):
    return einops.rearrange(arr, 'b h (w w4) c -> b (w4 h) w c', w4=w4)


print(solve(np.arange(16.0).reshape(1, 1, 8, 2), 4))
```

## Why this works

'b h (w w4) c -> b (w4 h) w c' slices the width into w4-wide phases and stacks each phase's full image strip vertically (w4 outer in the merge). An exotic fold — but exactly the kind of pattern einops makes checkable at a glance versus a blind reshape chain.
