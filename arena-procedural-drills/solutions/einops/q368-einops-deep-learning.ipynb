## Problem

Write a function solve(x3d) that takes a 5-D volume batch (b, c, x, y, z) with all three spatial dims EVEN and max-pools it with non-overlapping 2x2x2 windows: return shape (b, c, x//2, y//2, z//2) via 'b c (x 2) (y 2) (z 2) -> b c x y z' with 'max'. (Note the anonymous literal 2s in the pattern.)


<details><summary>💡 Hint (click to reveal)</summary>

Factor each spatial axis into (kept, 2) and reduce the three pooled factors away with max.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x3d):
    return einops.reduce(x3d, 'b c (x two1) (y two2) (z two3) -> b c x y z', 'max', two1=2, two2=2, two3=2)


print(solve(np.arange(16.0).reshape(1, 2, 2, 2, 2)))
```

## Why this works

The pattern splits every spatial axis into output position and within-window offset; 'max' consumes the three offsets simultaneously — 3-D max pooling, the volumetric big sibling of the 2-D reshape-pool trick.
