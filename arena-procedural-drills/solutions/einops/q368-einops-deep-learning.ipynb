## Problem

Given a 5D array x_3d of shape (batch, channels, X, Y, Z) where each of the three spatial dimensions X, Y, Z is even, return a 5D array of shape (batch, channels, X/2, Y/2, Z/2) where each output entry is the maximum over a non-overlapping 2x2x2 spatial block of the input at the corresponding batch and channel index (3D max pooling with window 2 and stride 2 along every spatial axis).


<details><summary>💡 Hint (click to reveal)</summary>

Split each spatial axis into an output-size axis times a literal window of 2, then reduce with 'max'.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

np.random.seed(0)
b = 2
c = 3
x_3d = np.random.rand(b, c, 4, 4, 4)

def solve():
    return einops.reduce(x_3d, 'b c (x2 2) (y2 2) (z2 2) -> b c x2 y2 z2', 'max')

print(solve())
```

## Why this works

Each spatial axis is factored as (out 2): the literal 2 is the pooling window and the named axis is the number of output positions. reduce with 'max' then takes the maximum within each non-overlapping 2x2x2 block, giving stride-2 window-2 3D max pooling.
