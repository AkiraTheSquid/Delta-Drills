## Problem

Write solve(imgs) for a (B, H, W, C) batch: make each image three times as TALL by repeating each ROW three times consecutively (row 0 three times, then row 1 three times, …). Pattern: einops.repeat 'b h w c -> b (h r) w c' with r=3.

einops.repeat 'b h w c -> b (h r) w c' with r=3 — r after h repeats rows in place.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
imgs = arr

def solve(imgs):
    return einops.repeat(imgs, 'b h w c -> b (h r) w c', r=3)

print(solve(imgs))
```

## Why this works

In the output group (h r), r varies fastest: each source row appears 3 times consecutively before the next row starts. Compare '(r h)': that would stack three full copies of each image instead. Same letters, different order, different picture.
