## Problem

For a batch of images with height, width, and color channels, increase the height of each image by a factor of three by repeating each row three times, while keeping the width and color channels unchanged.


<details><summary>💡 Hint (click to reveal)</summary>

Use repeat with a new factor axis folded into the height group, leaving the other axes identical on both sides.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    return einops.repeat(arr, 'b h w c -> b (h r) w c', r=3)

print(solve())
```

## Why this works

Repeat introduces a new named axis `r=3` inside the height composite `(h r)`, so every original row is emitted three consecutive times. Width, channels, and batch pass through unchanged, tripling the height.
