## Problem

Write solve(img) for a (C, H, W) image whose C is EVEN: treat the channel axis as (p × two) pairs and split OUT the pair-member axis to the front — output (2, C/2, H, W) where result[0] holds the even-pair members. Pattern: '(p two) h w -> two p h w' with two=2.

'(p two) h w -> two p h w', two=2 — pair-member axis splits out to the front.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = np.repeat(arr[5], 2, axis=0)

def solve(img):
    return einops.rearrange(img, '(p two) h w -> two p h w', two=2)

print(solve(img))
```

## Why this works

Reading channels as (p two) makes `two` the FASTEST-varying index: channels 0,2,4 are pair-member 0 and channels 1,3,5 are pair-member 1. Splitting `two` to the front yields (2, C/2, H, W) — result[0] holds the even channels. '(two p)' would split first-half/second-half instead.
