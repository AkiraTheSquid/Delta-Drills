## Problem

You are given img, a 3D array of shape (channels, height, width) = (6, 150, 150) in channels-first layout where channels is even. Treat the channels axis as consecutive pairs of length 2, split it into two halves such that within every pair the first member goes into the first half and the second member goes into the second half, and promote that within-pair index to a new leading axis. Return a 4D array of shape (2, channels/2, height, width) = (2, 3, 150, 150).


<details><summary>💡 Hint (click to reveal)</summary>

Factor channels into (pair_index, within_pair) with the inner factor 2, then move the within-pair index to the front.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = np.repeat(arr[0], 2, axis=0)

def solve():
    return einops.rearrange(img, '(p two) h w -> two p h w', two=2)

print(solve())
```

## Why this works

The channel axis is read as consecutive pairs via (p two) with two=2; promoting the within-pair index 'two' to the leading axis splits the channels so the first member of each pair lands in slice 0 and the second in slice 1.
