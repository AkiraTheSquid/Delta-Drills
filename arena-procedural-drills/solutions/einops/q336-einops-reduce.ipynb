## Problem

You are given arr, a 4D array of shape (batch, channels, height, width) = (6, 3, 150, 150) holding 6 channels-first digit images. First downsample each image by taking the maximum value over every non-overlapping 2x2 spatial block, halving both height and width. Then tile the 6 downsampled images into a 2-row by 3-column composite (rows fill first), producing a 3D channels-first array of shape (channels, 2 * 75, 3 * 75) = (3, 150, 225).


<details><summary>💡 Hint (click to reveal)</summary>

In one reduce, decompose batch into grid factors and each spatial axis into output-size times a 2-wide pool factor, reduce with max, and fuse grid factors into the spatial output.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.reduce(arr, '(b1 b2) c (h h2) (w w2) -> c (b1 h) (b2 w)', 'max', h2=2, w2=2, b1=2)

print(solve())
```

## Why this works

The same einops.reduce call does pooling and tiling at once: height and width are each split into an output block and a 2-wide pooling factor (h2, w2) that gets reduced by max, while the batch is split into grid factors b1, b2 that merge into the spatial layout to tile the six downsampled images into a (3, 150, 225) grid.
