## Problem

Write a function solve(arr, r) that takes a channels-first batch (b, c, h, w), keeps only the EVEN-indexed images arr[0], arr[2], ... (assume that count is divisible by r), and tiles them into an r-row grid of shape (c, r*h, (count//r)*w) — a slice composed with the grid rearrange '(r n) c h w -> c (r h) (n w)'.


<details><summary>💡 Hint (click to reveal)</summary>

Slice first (every second image), THEN grid — rearrange only reorganizes whatever you hand it.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr, r):
    return einops.rearrange(arr[::2], '(r n) c h w -> c (r h) (n w)', r=r)


print(solve(t.arange(16.0).reshape(4, 1, 1, 4), 2))
```

## Why this works

arr[::2] selects the even-indexed images; the grid pattern then tiles that subset exactly like the full-batch montage. Composing plain indexing with einops patterns is the everyday idiom — selection is indexing's job, layout is einops'.
