## Problem

You are given arr, a 4D uint8 array of shape (6, 3, 150, 150) holding 6 RGB digit images in (batch, channels, height, width) layout. Arrange the 6 images into a 2-row by 3-column grid (rows fixed at 2, columns inferred as 3) so each cell is one full image, returning a 3D array of shape (3, 2*150, 3*150) in (channels, grid_rows*height, grid_cols*width) layout.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the batch into a fixed number of rows and an inferred number of columns, then fuse rows into height and columns into width.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
ws = 2

def solve():
    return einops.rearrange(arr, '(r n) c h w -> c (r h) (n w)', r=2)

print(solve())
```

## Why this works

The batch axis is decomposed into r rows and n columns; fixing r=2 lets einops infer n=3. Merging r with height and n with width tiles the images into a 2x3 channels-first grid of shape (3, 300, 450).
