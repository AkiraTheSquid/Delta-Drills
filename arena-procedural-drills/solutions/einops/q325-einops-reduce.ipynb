For a batch of CHW images, reduce each image to a single scalar by averaging over all channels and spatial dimensions, returning one value per image in the batch.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy').astype(np.float32)
b = 2

def solve():
    return einops.reduce(arr, 'b c h w -> b', 'mean')

print(solve())
```

## Why this works

Keeping only b in the output while dropping c, h, and w tells einops.reduce to average over all channel and spatial elements, collapsing each image to a single scalar. The result is one mean value per batch element.
