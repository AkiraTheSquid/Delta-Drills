## Problem

Write a function solve(arr) that takes a channels-first batch of shape (b, c, h, w) and reduces each image to ONE scalar — the MEAN over its channels and all pixels — returning a length-b vector.


<details><summary>💡 Hint (click to reveal)</summary>

reduce with a pattern that keeps only b and the reduction name 'mean' — the dropped axes are what gets averaged.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.reduce(arr, 'b c h w -> b', 'mean')


print(solve(t.ones((2, 3, 2, 2))))
```

## Why this works

einops.reduce(arr, 'b c h w -> b', 'mean') averages every value belonging to each batch item — arr.mean(axis=(1, 2, 3)). Global average pooling collapsed all the way to one number per image.
