## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(img, 'c h w -> (c h) w')

print(solve())
```

## Why this works

Merging the channel and height axes into `(c h)` stacks the per-channel planes vertically into one tall strip, leaving width unchanged. The C-order merge keeps each channel's rows contiguous.
