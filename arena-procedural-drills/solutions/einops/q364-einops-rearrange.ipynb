## Problem

You are given arr, a 4D channels-first batch of shape (batch, channels, height, width) = (6, 3, 150, 150) holding 6 digit images. Split the batch axis 6 into b1 = 2 rows and b2 = 3 columns of images, then tile them into a single 3D channels-first canvas of shape (channels, b1*height, b2*width) = (3, 300, 450), where the i-th batch image lands in row i // 3 and column i % 3 of the 2x3 grid.

Split the batch axis into row and column factors, then merge the row factor into height and the column factor into width.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]
ws = 2

def solve():
    return einops.rearrange(arr, '(b1 b2) c h w -> c (b1 h) (b2 w)', b1=2)

print(solve())
```

## Why this works

Splitting the batch of 6 as `(b1 b2)` with b1=2 gives a 2x3 logical grid, and merging b1 into height and b2 into width tiles the images into a single canvas. Channels move to the front for channels-first output; image i lands at row i//3, column i%3.
