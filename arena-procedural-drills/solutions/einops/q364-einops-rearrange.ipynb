## Problem

Write a function solve(arr, b1) that takes a channels-LAST batch (b, h, w, c) — b divisible by b1 — and tiles the images into a b1-row montage: return shape (b1*h, (b//b1)*w, c) via '(b1 b2) h w c -> (b1 h) (b2 w) c'. (The companion drill does this for channels-first input.)


<details><summary>💡 Hint (click to reveal)</summary>

Factor the batch into grid rows and columns, then merge each with its matching spatial axis — channels ride along at the end.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr, b1):
    return einops.rearrange(arr, '(b1 b2) h w c -> (b1 h) (b2 w) c', b1=b1)


print(solve(t.arange(8.0).reshape(4, 1, 2, 1), 2))
```

## Why this works

'(b1 b2) h w c -> (b1 h) (b2 w) c' is the channels-last montage: row-major batch order fills the grid left-to-right, top-to-bottom, giving one big displayable image.
