## Problem

Write a function solve(arr) that takes a channels-first batch (b, c, h, w) and stacks the images VERTICALLY in channels-first layout: return shape (c, b*h, w) via 'b c h w -> c (b h) w'.


<details><summary>💡 Hint (click to reveal)</summary>

Batch merges into height with b outermost — channels move in front first so the merge is contiguous.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b c h w -> c (b h) w')


print(solve(t.arange(8.0).reshape(2, 1, 2, 2)))
```

## Why this works

'b c h w -> c (b h) w' brings channels forward and stacks the images top-to-bottom — the vertical filmstrip montage in conv layout, one transpose plus reshape in PyTorch.
