## Problem

You are given arr, a 4D uint8 array of shape (6, 3, 150, 150) holding 6 RGB digit images in (batch, channels, height, width) layout. Stack the 6 images on top of each other along the height axis, returning a 3D array of shape (3, 6*150, 150) in (channels, batch*height, width) layout so all 6 images form a single tall column with image 0 on top and image 5 on the bottom.


<details><summary>💡 Hint (click to reveal)</summary>

Bring channels to the front and compose the batch axis together with height so the images stack along the height dimension.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]
b = 2

def solve():
    return einops.rearrange(arr, 'b c h w -> c (b h) w')

print(solve())
```

## Why this works

Moving the channel axis to the front and merging batch into height stacks every image vertically: the (b h) composition lays image 0's rows first, then image 1's, and so on, forming one tall column.
