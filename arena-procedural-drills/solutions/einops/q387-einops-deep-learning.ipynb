## Problem

You are given x, a 4D array of shape (batch, channels, height, width) = (2, 3, 150, 150) holding channels-first digit images, with hs = 2 and ws = 2. Split each image's height into 2 stripes and its width into 2 stripes (forming a 2x2 grid of half-size sub-images per input image), then promote those sub-grid indices to the leading batch axis, producing a 4D array of shape (hs * ws * batch, channels, height / hs, width / ws) = (8, 3, 75, 75).


<details><summary>💡 Hint (click to reveal)</summary>

Decompose height and width into (stripe, size) pairs, then compose the stripe indices together with batch as the new leading dimension.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
x = arr[:2]
hs = 2
ws = 2
b = 2

def solve():
    return einops.rearrange(x, 'b c (h hs) (w ws) -> (hs ws b) c h w', hs=2, ws=2)

print(solve())
```

## Why this works

Each spatial axis is split into a stripe count (hs, ws) and a per-stripe size (h, w); placing hs and ws ahead of b in the output merges the 2x2 grid of sub-images into the leading batch axis, yielding hs*ws*batch half-size tiles.
