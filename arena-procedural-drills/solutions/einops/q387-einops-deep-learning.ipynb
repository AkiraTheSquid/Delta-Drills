## Problem

Write solve(x, hs, ws) for a (B, C, H, W) batch: split each image into an hs × ws subgrid and move the subgrid indices out into the batch axis — output (hs·ws·B, C, H/hs, W/ws), subgrid-major then batch. Pattern: 'b c (h hs) (w ws) -> (hs ws b) c h w'.


<details><summary>💡 Hint (click to reveal)</summary>

'b c (h hs) (w ws) -> (hs ws b) c h w' — subgrid positions become batch entries.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
x = arr[2:4]
hs = 2
ws = 2

def solve(x, hs, ws):
    return einops.rearrange(x, 'b c (h hs) (w ws) -> (hs ws b) c h w', hs=hs, ws=ws)

print(solve(x, hs, ws))
```

## Why this works

Identical mechanics to the other subgrid drill: (h hs) names hs vertical strips, (w ws) names ws horizontal ones, and (hs ws b) orders the output batch subgrid-major — every image's top-left crop, then top-right, and so on.
