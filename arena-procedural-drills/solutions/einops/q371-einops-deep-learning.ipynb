## Problem

Inverse of packing 2x2 spatial subgrids into the batch axis: take a tensor of shape (hs*ws*b, c, h, w) and unpack it back to (b, c, h*hs, w*ws) with hs=ws=2.


<details><summary>💡 Hint (click to reveal)</summary>

Read off the forward pattern and write its mirror image: split the packed batch axis and fold the factors back into height and width.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers()
y = einops.rearrange(arr[:2], 'b c (h hs) (w ws) -> (hs ws b) c h w', hs=2, ws=2)

def solve():
    return einops.rearrange(y, '(hs ws b) c h w -> b c (h hs) (w ws)', hs=2, ws=2)

print(solve())
```

## Why this works

The packing folded the 2x2 spatial subgrid factors hs and ws into the batch axis as (hs ws b). The inverse decomposes that batch axis back into the same named factors and re-interleaves them into the spatial axes as (h hs) and (w ws), exactly undoing the forward rearrange.
