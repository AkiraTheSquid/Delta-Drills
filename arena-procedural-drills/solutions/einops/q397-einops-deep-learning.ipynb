## Problem

For a batch of tensors organized in channel-first format, combine the channels and batch dimensions into a single dimension while keeping the height and width dimensions intact.


<details><summary>💡 Hint (click to reveal)</summary>

Put height and width first, then merge batch and channels into one parenthesized trailing axis.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
b = 2

def solve():
    return einops.rearrange(arr, 'b c h w -> h w (b c)')

print(solve())
```

## Why this works

Height and width are moved to the front and the batch and channel axes are composed into a single trailing (b c) axis, collapsing the two leading dimensions into one while preserving the spatial grid.
