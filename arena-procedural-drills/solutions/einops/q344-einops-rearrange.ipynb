## Problem

Given a 3D image array of shape (channels, height, width) representing a single color image in channels-first layout, return a 3D array of shape (channels, width, height) that holds the same image with its two spatial axes exchanged, so the result looks like the original reflected across its main diagonal.


<details><summary>💡 Hint (click to reveal)</summary>

One rearrange that exchanges the height and width names, keeping channels first.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(arr[1], 'c h w -> c w h')

print(solve())
```

## Why this works

Swapping the named height and width axes in the output pattern transposes the two spatial dimensions of a channels-first image, reflecting it across the main diagonal while channels stay leading.
