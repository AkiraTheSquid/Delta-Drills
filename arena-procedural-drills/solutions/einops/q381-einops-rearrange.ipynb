## Problem

You are given four channels-last images img_a, img_b, img_c, img_d, each a 3D array of shape (height, width, channels) with identical shape. Combine them into a single 3D array of shape (2 * height, 2 * width, channels) arranged as a 2-by-2 spatial grid: img_a and img_b occupy the top row (left and right), and img_c and img_d occupy the bottom row (left and right).

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = np.moveaxis(_load_numbers(), 1, -1)
img = arr[0]
img_a = img
img_b = img
img_c = img
img_d = img

def solve():
    return einops.rearrange([img_a, img_b, img_c, img_d], '(r nc) h w ch -> (r h) (nc w) ch', r=2)

print(solve())
```

## Why this works

The list of four images becomes a batch of 4 that is factored as (r nc) with r=2 rows and nc=2 columns. Merging (r h) stacks rows vertically and (nc w) stacks columns horizontally, tiling the images into a 2x2 spatial canvas.
