## Problem

You are given x_heads, a 4D array of shape (batch, heads, tokens, embedding_dim) representing multi-head attention output. Concatenate the per-head embeddings into a single embedding per token, returning a 3D array of shape (batch, tokens, heads * embedding_dim) where the head dimension is folded into the trailing embedding axis.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers()
b = 2
x_heads = arr.reshape(b, -1, arr.shape[1], arr.shape[2])

def solve():
    return einops.rearrange(x_heads, 'b nh t d -> b t (nh d)')

print(solve())
```

## Why this works

Multi-head attention output has a separate heads axis nh; folding (nh d) on the output concatenates each head's embedding into one longer per-token vector. Moving t before the merged axis gives the standard (batch, tokens, heads*dim) shape.
