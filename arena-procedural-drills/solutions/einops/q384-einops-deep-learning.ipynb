## Problem

Write a function solve(x_heads) that takes multi-head attention output of shape (b, nh, t, d) and CONCATENATES the heads back together per token: return shape (b, t, nh*d) via 'b nh t d -> b t (nh d)'.


<details><summary>💡 Hint (click to reveal)</summary>

Bring heads next to depth first, then merge them — head index outer so each head's features stay contiguous.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(x_heads):
    return einops.rearrange(x_heads, 'b nh t d -> b t (nh d)')


print(solve(t.arange(8.0).reshape(1, 2, 2, 2)))
```

## Why this works

'b nh t d -> b t (nh d)' transposes heads beside the embedding and merges — the concat-heads step that precedes the output projection in every transformer block. In PyTorch: transpose(0, 2, 1, 3) then reshape.
