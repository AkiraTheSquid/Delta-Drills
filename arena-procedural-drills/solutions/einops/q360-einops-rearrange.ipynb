## Problem

For a 2D image represented by its height and width, add a singleton channel dimension at the beginning, and then remove that channel dimension, returning to the original height and width.


<details><summary>💡 Hint (click to reveal)</summary>

Add a size-1 leading axis with rearrange, then a second rearrange that consumes that axis to drop it.

</details>

## Reference solution


```python
import numpy as np
import einops

x2d = np.arange(3 * 4).reshape(3, 4)

def solve():
    y = einops.rearrange(x2d, 'h w -> 1 h w')
    return einops.rearrange(y, '1 h w -> h w')

print(solve())
```

## Why this works

Writing a literal 1 (or `()`) on the output adds a singleton channel axis, and matching it on the input of a second rearrange removes it. The round trip returns the array to its original (h, w) shape.
