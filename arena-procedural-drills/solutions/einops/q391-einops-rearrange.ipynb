## Problem

Write a function solve(img) that takes a channels-first image (c, h, w) and FLATTENS its spatial axes per channel: return shape (c, h*w) via 'c h w -> c (h w)'.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(img):
    return einops.rearrange(img, 'c h w -> c (h w)')


print(solve(t.arange(8.0).reshape(2, 2, 2)))
```

## Why this works

'c h w -> c (h w)' concatenates each channel's rows into one long vector — img.reshape(c, -1), the per-channel half of a full flatten and the inverse of the h-given unflatten.
