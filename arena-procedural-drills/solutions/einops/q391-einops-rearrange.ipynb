## Problem

For a single CHW image, combine the height and width dimensions into a single dimension while keeping the channels separate.


<details><summary>💡 Hint (click to reveal)</summary>

Keep the channel axis as-is and merge height and width into one parenthesized axis in the output.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(img, 'c h w -> c (h w)')

print(solve())
```

## Why this works

Composing h and w into a single (h w) axis flattens the spatial grid into one dimension while the channel axis stays separate, producing a (c, h*w) matrix.
