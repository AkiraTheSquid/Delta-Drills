## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers().astype(np.float32)
b = 2

def solve():
    return einops.reduce(arr, 'b c h w -> b c', 'mean')

print(solve())
```

## Why this works

einops.reduce collapses the axes that disappear from the output pattern. Mapping 'b c h w -> b c' with 'mean' keeps batch and channel while averaging over every (h, w) spatial position, yielding one scalar per channel per image.
