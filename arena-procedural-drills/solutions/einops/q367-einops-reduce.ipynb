## Problem

Write a function solve(arr) that takes a channels-first batch (b, c, h, w) and returns the (b, c) tensor of PER-CHANNEL spatial means: each value averages one channel's h x w map.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.reduce(arr, 'b c h w -> b c', 'mean')


print(solve(t.arange(8.0).reshape(1, 2, 2, 2)))
```

## Why this works

reduce(arr, 'b c h w -> b c', 'mean') is global average pooling — arr.mean(axis=(2, 3)) — the layer that turns conv feature maps into one descriptor per channel before a classifier.
