## Problem

Rearrange the channels of a batch of images by grouping them into sets and then reorganizing the output so that the number of channels and groups are swapped.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers()
x = np.arange(2 * 12 * 8 * 8).reshape(2, 12, 8, 8)
b = 2

def solve():
    return einops.rearrange(x, 'b (g c_in) h w -> b (c_in g) h w', g=4)

print(solve())
```

## Why this works

The channel axis factors into g groups of c_in channels each. Re-emitting it as (c_in g) instead of (g c_in) transposes the two factors, which is exactly the ShuffleNet channel-shuffle permutation, while spatial dims stay fixed.
