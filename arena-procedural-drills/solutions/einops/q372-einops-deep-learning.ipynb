## Problem

Write a function solve(x, g) that takes a tensor of shape (b, g*c, h, w) with g channel groups (group index slowest) and TRANSPOSES groups against within-group channels: return the same shape reordered by 'b (g c) h w -> b (c g) h w' — after which consecutive channels come from DIFFERENT groups.


<details><summary>💡 Hint (click to reveal)</summary>

Factor as (g, c), emit as (c, g) — a transpose hidden inside one axis.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x, g):
    return einops.rearrange(x, 'b (g c) h w -> b (c g) h w', g=g)


print(solve(np.arange(4.0).reshape(1, 4, 1, 1), 2))
```

## Why this works

'(g c)' to '(c g)' interleaves the groups: channel k of every group becomes adjacent. This is the classic channel shuffle written as a factored-axis transpose — reshape, swapaxes, reshape in numpy.
