## Problem

You are given arr, a 4D array of shape (batch, channels, height, width) = (6, 3, 150, 150) holding 6 channels-first digit images. Tile these 6 images into a single composite image arranged as a 2-row by 3-column grid (rows fill first), producing a 3D channels-first array of shape (channels, 2 * height, 3 * width) = (3, 300, 450).


<details><summary>💡 Hint (click to reveal)</summary>

Split the batch axis into two grid factors with the row count fixed, then merge one factor into height and the other into width.

## Reference solution


```python
import numpy as np
import einops

_base_arr = np.load('/delta_numbers.npy')
_arr_repeats = -(-6 // _base_arr.shape[0])
arr = np.concatenate([_base_arr] * _arr_repeats, axis=0)
arr = arr[:6]

def solve():
    return einops.rearrange(arr[:6], '(g1 g2) c h w -> c (g1 h) (g2 w)', g1=2)

print(solve())
```

## Why this works

The batch axis of 6 is factored into a 2x3 grid (g1 rows, g2 cols) by binding g1=2. Stacking g1 with h builds the vertical tiling and g2 with w builds the horizontal tiling, while channels stay leading, giving a (3, 300, 450) composite.
