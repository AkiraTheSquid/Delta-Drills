## Problem

Write a function solve(arr, g1) that takes a channels-first batch of shape (b, c, h, w) — b divisible by g1 — and tiles the images into a g1-row grid: return shape (c, g1*h, (b//g1)*w) via '(g1 g2) c h w -> c (g1 h) (g2 w)', batch order filling each row left to right.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(arr, g1):
    return einops.rearrange(arr, '(g1 g2) c h w -> c (g1 h) (g2 w)', g1=g1)


print(solve(np.arange(8.0).reshape(4, 1, 1, 2), 2))
```

## Why this works

'(g1 g2)' reads the batch as a grid row-major; '(g1 h)' stacks image rows vertically and '(g2 w)' lays each row's images side by side — the montage/contact-sheet rearrange, all layout and no arithmetic.
