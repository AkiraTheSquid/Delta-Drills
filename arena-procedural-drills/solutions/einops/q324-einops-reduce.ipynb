## Problem

2x2 non-overlapping average pooling on a BCHW batch — output H and W are half the input. Use named factors h2 and w2 for the pooling window.


<details><summary>💡 Hint (click to reveal)</summary>

Factor both spatial axes into output-size and window-size parts, drop the window parts from the output, and use the 'mean' reduction.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy').astype(np.float32)
b = 2

def solve():
    return einops.reduce(arr, 'b c (h h2) (w w2) -> b c h w', 'mean', h2=2, w2=2)

print(solve())
```

## Why this works

einops.reduce splits each spatial axis into a coarse index and a 2-wide window factor (h2, w2), then averages over the window factors because they are absent from the output. This is exactly 2x2 non-overlapping average pooling, halving H and W.
