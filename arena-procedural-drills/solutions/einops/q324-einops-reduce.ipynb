## Reference solution


```python
import torch as t
import einops

arr = t.tensor(np.load('/delta_numbers.npy')).to(t.float32)
x = arr

def solve(x):
    return einops.reduce(x, 'b c (h h2) (w w2) -> b c h w', 'mean', h2=2, w2=2)

print(solve(x))
```

## Why this works

Grouping the input axes as (h h2)/(w w2) and dropping h2/w2 from the output tells reduce to apply 'mean' over each 2×2 window — average pooling in one expression. Output spatial dims are exactly half the input.
