## Problem

You are given hwcs, a 4D uint8 array of shape (2, 150, 150, 3) holding 2 RGB digit images in (batch, height, width, channels) layout. Concatenate the 2 images side by side along the width axis, returning a 3D array of shape (150, 2*150, 3) in (height, batch*width, channels) layout so the second image appears immediately to the right of the first.


<details><summary>💡 Hint (click to reveal)</summary>

Fold the batch axis into width so the two images sit horizontally adjacent, keeping height and channels.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2
hwcs = arr[:b]

def solve():
    return einops.rearrange(hwcs, 'b h w c -> h (b w) c')

print(solve())
```

## Why this works

Merging batch into the width axis as '(b w)' lays the images out left-to-right: for each shared height row, the first image's pixels are followed by the second image's, producing a (150, 300, 3) side-by-side concatenation.
