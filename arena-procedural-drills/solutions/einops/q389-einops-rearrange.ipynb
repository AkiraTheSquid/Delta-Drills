## Problem

Arrange the first six HWC images into a grid with three rows and two columns, keeping the color channels intact while adjusting the layout of the images.

## Reference solution


```python
import numpy as np
import einops

_base_arr = np.load('/delta_numbers.npy')
_arr_repeats = -(-6 // _base_arr.shape[0])
arr = np.concatenate([_base_arr] * _arr_repeats, axis=0)
arr = arr[:6]
arr = np.moveaxis(arr, 1, -1)
b = 2

def solve():
    return einops.rearrange(arr[:6], '(g1 g2) h w c -> (g1 h) (g2 w) c', g1=3)

print(solve())
```

## Why this works

The 6-image batch is factored into a grid (g1 g2) with g1=3 rows and g2=2 columns; merging g1 into height and g2 into width tiles the images into a single 3x2 montage, with the channel axis preserved.
