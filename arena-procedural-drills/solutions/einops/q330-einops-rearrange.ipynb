## Problem

Rearrange a batch of images by moving the batch dimension to the end, so that the output has the color channels and spatial dimensions first, followed by the batch.


<details><summary>💡 Hint (click to reveal)</summary>

A single rearrange with the same axis names reordered so batch lands last does the whole job.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
b = 2

def solve():
    return einops.rearrange(arr, 'b c h w -> c h w b')

print(solve())
```

## Why this works

rearrange is a pure axis permutation when input and output patterns contain the same named axes. Moving 'b' to the end of the output pattern transposes the batch dimension to the trailing position.
