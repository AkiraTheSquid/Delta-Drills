## Problem

Repeat a single CHW image along its width axis so the output is four times as wide, with each column of the original appearing four times in sequence.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers()
img = arr[0]

def solve():
    return einops.repeat(img, 'c h w -> c h (w 4)')

print(solve())
```

## Why this works

einops.repeat with the output group (w 4) tiles the width axis four times. Because the literal 4 sits to the right of w, each original column is repeated as a block, making the image four times as wide.
