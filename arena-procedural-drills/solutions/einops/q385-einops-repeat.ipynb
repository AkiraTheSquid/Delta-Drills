## Reference solution


```python
import torch as t
import einops

arr = t.tensor(np.load('/delta_numbers.npy'))
img = arr[0]

def solve(img):
    return einops.repeat(img, 'c h w -> c h (w 4)')

print(solve(img))
```

## Why this works

In the output group (w 4), the factor varies fastest: column 0 four times, then column 1 four times — horizontal stretching. '(4 w)' would place four whole copies of the image side by side instead.
