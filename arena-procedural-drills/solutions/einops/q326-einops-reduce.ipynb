## Problem

Write a function solve(arr) that takes a channels-last batch of shape (b, h, w, c) and returns the (b, h, w) tensor where each output pixel is the MAXIMUM across that pixel's channels.

Drop only the channel letter and pick 'max' as the reduction.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.reduce(arr, 'b h w c -> b h w', 'max')


print(solve(t.arange(8.0).reshape(1, 2, 2, 2)))
```

## Why this works

einops.reduce(arr, 'b h w c -> b h w', 'max') collapses channels with a per-pixel maximum — arr.max(axis=-1), a brightest-channel projection to one grayscale-style map per image.
