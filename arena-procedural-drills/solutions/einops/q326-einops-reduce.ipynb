## Problem

Given a 4D array arr of shape (batch, height, width, channels) representing a batch of color images in channels-last format, return a 3D array of shape (batch, height, width) where each output pixel holds the maximum value taken across the channel axis at that batch index and spatial location (one grayscale-style image per batch element, formed by collapsing channels via the per-pixel maximum).


<details><summary>💡 Hint (click to reveal)</summary>

Reduce with 'max' and omit only the channel axis from the output pattern so the maximum is taken across channels.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    return einops.reduce(arr, 'b h w c -> b h w', 'max')

print(solve())
```

## Why this works

Dropping the channel axis c from the output while keeping b, h, and w makes einops.reduce take the maximum across channels at each pixel. This collapses the color channels into a single per-pixel value per image.
