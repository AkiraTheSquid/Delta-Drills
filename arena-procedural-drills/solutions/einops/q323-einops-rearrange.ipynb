## Problem

You are given patches, a 4D array of shape (h*w, p1, p2, channels) = (225, 10, 10, 3) containing the 225 non-overlapping 10x10 RGB tiles of a 150x150 image, listed in row-major order with h = 15 tile-rows and w = 15 tile-columns. Reassemble the original image as a 3D array of shape (h*p1, w*p2, channels) = (150, 150, 3) by placing each tile back at its (row, column) position in the grid.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
h, w = 15, 15
patches = einops.rearrange(arr[0], '(h p1) (w p2) c -> (h w) p1 p2 c', p1=10, p2=10)

def solve():
    return einops.rearrange(patches, '(h w) p1 p2 c -> (h p1) (w p2) c', h=h, w=w)

print(solve())
```

## Why this works

The patch axis of length 225 is split into tile-row and tile-column indices (h w) with h=15, then each tile's pixels (p1, p2) are interleaved with those grid indices via (h p1) and (w p2) to rebuild the full 150x150 image. Supplying h (and thus w) tells einops how to factor the patch axis.
