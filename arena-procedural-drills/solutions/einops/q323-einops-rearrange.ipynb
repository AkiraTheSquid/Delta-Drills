## Problem

Write a function solve(patches, h, w) that takes a tile stack of shape (h*w, p1, p2, c) — the non-overlapping p1 x p2 tiles of an image, listed row-major — and REASSEMBLES the image: return shape (h*p1, w*p2, c) via '(h w) p1 p2 c -> (h p1) (w p2) c'.


<details><summary>💡 Hint (click to reveal)</summary>

Split the tile index back into its (row, col) grid coordinates, then interleave rows-of-tiles with rows-within-tiles.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(patches, h, w):
    return einops.rearrange(patches, '(h w) p1 p2 c -> (h p1) (w p2) c', h=h, w=w)


print(solve(t.arange(16.0).reshape(4, 2, 2, 1), 2, 2))
```

## Why this works

'(h w)' recovers each tile's grid position from its list index; merging '(h p1)' and '(w p2)' stitches tile rows and pixel rows back together. This inverts the patch-extraction rearrange used by vision transformers.
