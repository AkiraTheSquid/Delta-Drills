## Problem

Write a function solve(arr) that takes a channels-first batch (b, c, h, w) and returns shape (b*h, w, c): batch merged into height AND the layout converted to channels-last in the same pattern — 'b c h w -> (b h) w c'.


<details><summary>💡 Hint (click to reveal)</summary>

Permute to channels-last first (conceptually), then the (b h) merge is contiguous — einops fuses both.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b c h w -> (b h) w c')


print(solve(np.arange(8.0).reshape(2, 1, 2, 2)))
```

## Why this works

'b c h w -> (b h) w c' both moves channels to the end and stacks images vertically — transpose to (b, h, w, c) followed by a reshape. A vertical filmstrip in displayable layout from conv-layout input.
