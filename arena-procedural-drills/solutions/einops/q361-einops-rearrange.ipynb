## Problem

Write a function solve(imgs) that takes a PYTHON LIST of channels-last images, each (h, w, c), and STACKS it into one batch tensor of shape (len(imgs), h, w, c) — einops.rearrange on the list with the identity pattern 'b h w c -> b h w c'. The return value must be an actual PyTorch tensor, not the list.


<details><summary>💡 Hint (click to reveal)</summary>

Handing einops a list makes the list index the first pattern axis — an identity pattern is t.stack.

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(imgs):
    return einops.rearrange(imgs, 'b h w c -> b h w c')


print(solve([t.ones((2, 2, 1)), t.zeros((2, 2, 1))]))
```

## Why this works

rearrange(list, 'b h w c -> b h w c') validates that every list element shares a shape and stacks them along the new b axis — t.stack with shape checking built in.
