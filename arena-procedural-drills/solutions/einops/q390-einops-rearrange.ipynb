## Problem

Write a function solve(flat, h) that takes a channel-wise FLATTENED image of shape (c, h*w) and the original height h, and restores the spatial layout: return shape (c, h, w) via 'c (h w) -> c h w' — the inverse flatten needs h (or w) passed in.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(flat, h):
    return einops.rearrange(flat, 'c (h w) -> c h w', h=h)


print(solve(np.arange(8.0).reshape(2, 4), 2))
```

## Why this works

'c (h w) -> c h w' with h given splits each flat vector back into rows of the original width (w deduced from the total). flat.reshape(c, h, -1) is the numpy twin, -1 doing the same inference.
