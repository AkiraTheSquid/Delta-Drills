## Problem

Transform a tensor with color channels and flattened spatial dimensions back into a 2D format by separating the height and width, ensuring that each color channel retains its original spatial arrangement.


<details><summary>💡 Hint (click to reveal)</summary>

Split the single flattened axis into (h w) and pass the known height and width sizes so einops can unflatten it.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
h, w = 150, 150
flat = arr[0].reshape(3, h * w)

def solve():
    return einops.rearrange(flat, 'c (h w) -> c h w', h=h, w=w)

print(solve())
```

## Why this works

The flattened spatial axis is decomposed back into separate h and w axes by supplying their sizes; because the flattening was row-major, this exactly reconstructs the original 2D spatial layout per channel.
