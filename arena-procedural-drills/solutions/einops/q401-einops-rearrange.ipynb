## Problem

Write solve(img_hwc) for an (H, W, C) image with H and W divisible by 3: extract all non-overlapping 3×3 patches into a ((H/3 · W/3), 3, 3, C) tensor, patches ordered row-major. Pattern: '(h p1) (w p2) c -> (h w) p1 p2 c' with p1=3, p2=3.

## Reference solution


```python
import torch as t
import einops

arr = t.movedim(t.tensor(np.load('/delta_numbers.npy')), 1, -1)
img_hwc = arr[0]

def solve(img_hwc):
    return einops.rearrange(img_hwc, '(h p1) (w p2) c -> (h w) p1 p2 c', p1=3, p2=3)

print(solve(img_hwc))
```

## Why this works

The input groups carve the image into 3×3 patches; putting (h w) FIRST in the output flattens the patch-grid coordinates into one patch index (row-major), with each patch's own 3×3×C content following. This is exactly the patchify step in ViT-style models.
