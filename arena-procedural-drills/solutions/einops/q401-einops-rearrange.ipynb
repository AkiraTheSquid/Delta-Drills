## Problem

From a batch of images in HWC format, extract non-overlapping patches of size 3 by 3, resulting in a new structure that includes the number of patches along with the color channels.


<details><summary>💡 Hint (click to reveal)</summary>

Decompose height and width into (grid, patch) pairs with patch size 3, then compose the two grid indices into a single patch-count axis.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
img_hwc = arr[0]

def solve():
    return einops.rearrange(img_hwc, '(h p1) (w p2) c -> (h w) p1 p2 c', p1=3, p2=3)

print(solve())
```

## Why this works

Each spatial axis is split into a patch-grid index and an in-patch offset (p1, p2 = 3); composing the two grid indices into (h w) lists all non-overlapping 3x3 patches along a new leading axis, with channels preserved.
