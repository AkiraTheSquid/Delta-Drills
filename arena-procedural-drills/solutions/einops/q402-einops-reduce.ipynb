## Problem

Write solve(img) for an (H, W, C) image: compute each COLUMN's per-channel maximum (reduce the height axis to a singleton: 'h w c -> () w c' with 'max') and subtract it from the image. Return img - mx. (uint8 wraparound on underflow is expected — same on both sides.)


<details><summary>💡 Hint (click to reveal)</summary>

Column max: einops.reduce(img, 'h w c -> () w c', 'max'), then subtract.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
img = arr[3]

def solve(img):
    mx = einops.reduce(img, 'h w c -> () w c', 'max')
    return img - mx

print(solve(img))
```

## Why this works

Reducing height to a singleton gives each column's per-channel max as a (1, W, C) array that broadcasts back over the rows for the subtraction. The mirror image of the row-max drill — which axis you reduce is the entire difference.
