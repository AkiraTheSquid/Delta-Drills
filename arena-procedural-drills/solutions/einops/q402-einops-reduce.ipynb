## Problem

For an image with height, width, and color channels, compute the maximum value for each column across all color channels and then subtract this maximum from the original image, maintaining the same dimensions.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce over the height axis with max but retain a singleton there so the result broadcasts back against the original image when subtracted.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    mx = einops.reduce(img, 'h w c -> () w c', 'max')
    return img - mx

print(solve())
```

## Why this works

Reducing over height with 'max' while keeping () in the height slot produces a per-column, per-channel maximum of shape (1, w, c); subtracting it broadcasts across the height axis, leaving the image dimensions unchanged.
