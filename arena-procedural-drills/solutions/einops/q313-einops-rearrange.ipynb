## Problem

Space-to-depth with block size 2 on a BCHW batch: pack each non-overlapping 2x2 spatial block into the channel dim. H and W halve; channels grow 4x. Use einops.rearrange with named factors p1 and p2.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy')
b = 2

def solve():
    return einops.rearrange(arr, 'b c (h p1) (w p2) -> b (c p1 p2) h w', p1=2, p2=2)

print(solve())
```

## Why this works

Space-to-depth factors each spatial axis into a coarse part (h, w) and a 2-wide block part (p1, p2). Moving the block factors into the channel grouping (c p1 p2) shrinks H and W by 2 and grows the channel count 4x, repacking every 2x2 spatial block into the channel dimension.
