## Problem

Write a function solve(arr, p1, p2) that takes a tensor of shape (b, c*p1*p2, h, w) — channel groups ordered with c slowest — and performs DEPTH-TO-SPACE: return shape (b, c, h*p1, w*p2) via 'b (c p1 p2) h w -> b c (h p1) (w p2)', spreading each group of p1*p2 channels over a finer spatial grid.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the channels into (c, p1, p2) and merge the two block factors into height and width — the exact inverse of space-to-depth.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr, p1, p2):
    return einops.rearrange(arr, 'b (c p1 p2) h w -> b c (h p1) (w p2)', p1=p1, p2=p2)


print(solve(t.arange(4.0).reshape(1, 4, 1, 1), 2, 2))
```

## Why this works

'b (c p1 p2) h w -> b c (h p1) (w p2)' sends each channel-group member to its sub-pixel position, multiplying spatial resolution by p1 x p2 while dividing channels — pixel shuffle again, this time with the group order c-major.
