## Problem

For a batch of images with color channels grouped into blocks, rearrange the data by splitting the color channels into separate groups and expanding the height and width dimensions by a factor of two, resulting in a larger spatial resolution.


<details><summary>💡 Hint (click to reveal)</summary>

Split channels into the true channel count times two size-2 factors, then merge those factors into height and width.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.arange(2 * 12 * 3 * 4).reshape(2, 12, 3, 4)

def solve():
    return einops.rearrange(arr, 'b (c p1 p2) h w -> b c (h p1) (w p2)', p1=2, p2=2)

print(solve())
```

## Why this works

The channel axis of 12 is decomposed into c=3 real channels and two 2-wide patch factors p1, p2. Those factors are fused into height and width, doubling each spatial dimension, which is the pixel-shuffle / depth-to-space upsampling pattern.
