## Problem

Transform a batch of images by rearranging the color channels into spatial dimensions, effectively increasing the height and width of each image while reducing the number of color channels.


<details><summary>💡 Hint (click to reveal)</summary>

Split the channel axis into named height/width block factors plus a remaining channel, then fuse those block factors into the spatial axes.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.load('/delta_numbers.npy')
x = np.arange(2 * 8 * 4 * 4).reshape(2, 8, 4, 4)
b = 2

def solve():
    return einops.rearrange(x, 'b (h1 w1 c) h w -> b c (h h1) (w w1)', h1=4, w1=2)

print(solve())
```

## Why this works

The channel axis 8 is decomposed into (h1 w1 c) with h1=4, w1=2 (leaving c=1), and those spatial-block factors are merged into height and width. This depth-to-space upsampling enlarges the spatial dimensions while shrinking the channel count.
