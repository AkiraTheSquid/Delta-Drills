## Problem

You are given x, a 4D uint8 array of shape (2, 6, 150, 150) in (batch, channels, height, width) layout where the 6 channels are conceptually two stacked halves of 3 channels each. Split the channel axis into its first 3-channel half and its second 3-channel half, returning a 5D array of shape (2, 2, 3, 150, 150) in (split, batch, channels_per_half, height, width) layout whose two slices along the leading split axis are part1 and part2.


<details><summary>💡 Hint (click to reveal)</summary>

Split the channel axis into a leading factor of 2 and the remaining channels, then bring that factor to the front.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
x = np.concatenate([arr[:2], arr[:2]], axis=1)
b = 2

def solve():
    return einops.rearrange(x, 'b (split c) h w -> split b c h w', split=2)

print(solve())
```

## Why this works

Factoring the 6-channel axis as `(split c)` with split=2 separates it into two consecutive 3-channel halves, and moving `split` to the front yields a leading axis whose two slices are the first and second halves. The original answer unpacks these as part1, part2, but the full 5D array is the returned value.
