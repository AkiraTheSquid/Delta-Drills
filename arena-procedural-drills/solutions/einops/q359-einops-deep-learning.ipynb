## Problem

Write a function solve(x, split) that takes a tensor of shape (b, split*c, h, w) whose channel axis holds `split` consecutive groups (group index SLOWEST) and returns shape (split, b, c, h, w) — the groups separated and moved to the front, ready to unpack as part1, part2, ... = solve(x, 2).


<details><summary>💡 Hint (click to reveal)</summary>

Factor the channels with the split index OUTERMOST and name it first in the output — tuple unpacking does the rest.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x, split):
    return einops.rearrange(x, 'b (split c) h w -> split b c h w', split=split)


print(solve(np.arange(8.0).reshape(1, 4, 1, 2), 2))
```

## Why this works

'b (split c) h w -> split b c h w' separates consecutive channel halves (the first c channels form group 0) and leads with the group axis so `a, b = solve(x, 2)` destructures them — the gating-network split of GLU-style layers.
