## Problem

Write a function solve(x, k) that takes a batch (b, c, h, w) — h, w divisible by k — and does max-pooling with k x k windows FOLLOWED by a full flatten: return shape (b, c*(h//k)*(w//k)) in ONE einops.reduce: 'b c (h k1) (w k2) -> b (c h w)' with 'max'.


<details><summary>💡 Hint (click to reveal)</summary>

Pooling factors reduce away while the surviving axes merge into the flat feature vector — one pattern, two jobs.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(x, k):
    return einops.reduce(x, 'b c (h k1) (w k2) -> b (c h w)', 'max', k1=k, k2=k)


print(solve(t.arange(16.0).reshape(1, 1, 4, 4), 2))
```

## Why this works

The reduce consumes the within-window offsets (max) and the output group '(c h w)' simultaneously flattens what remains — pool-then-flatten fused, the tail end of many small convnets.
