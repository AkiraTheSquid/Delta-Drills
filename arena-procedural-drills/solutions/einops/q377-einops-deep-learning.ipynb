## Problem

You are given x, a 4D array of shape (batch, channels, height, width) = (2, 3, 150, 150) holding two channels-first images. First apply a 2x2 maximum pooling on the spatial axes, taking the maximum value within each non-overlapping 2x2 window of height and width so each spatial axis is halved. Then flatten the channel axis together with the two pooled spatial axes into a single trailing feature axis, returning a 2D array of shape (batch, channels * (height/2) * (width/2)) = (2, 16875).

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = _load_numbers().astype(float)
x = arr[:2]

def solve():
    return einops.reduce(x, 'b c (h two1) (w two2) -> b (c h w)', 'max', two1=2, two2=2)

print(solve())
```

## Why this works

reduce both pools and flattens in one pattern: factoring each spatial axis as (out 2) and taking 'max' performs 2x2 max pooling, and emitting (c h w) as a single merged output axis flattens channels with the pooled spatial grid into one feature vector per image.
