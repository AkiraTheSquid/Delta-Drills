## Problem

For a batch of images with channels and spatial dimensions, combine the channel, height, and width into a single dimension, resulting in a flattened representation for each image.


<details><summary>💡 Hint (click to reveal)</summary>

Use rearrange to combine the three non-batch axes into one composite dimension.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
x = arr[:2]
b = 2

def solve():
    return einops.rearrange(x, 'b c h w -> b (c h w)')

print(solve())
```

## Why this works

Merging channels, height, and width into one composite `(c h w)` flattens each image into a single feature vector while preserving the batch axis. This is the standard flatten step before a dense layer.
