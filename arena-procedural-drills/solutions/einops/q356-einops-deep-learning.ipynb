## Problem

Write a function solve(x) that takes a batch (b, c, h, w) and FLATTENS each image completely: return shape (b, c*h*w) via 'b c h w -> b (c h w)' — the standard conv-features-to-classifier-head flatten.


<details><summary>💡 Hint (click to reveal)</summary>

Merging already-adjacent trailing axes is a reshape — batch stays, everything else concatenates in order.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x):
    return einops.rearrange(x, 'b c h w -> b (c h w)')


print(solve(np.arange(8.0).reshape(2, 1, 2, 2)))
```

## Why this works

'b (c h w)' rolls channels, then rows, then columns into one feature vector per item — x.reshape(b, -1), the bridge from convolutional features to a dense classifier head.
