## Problem

Write solve(imgs) for a (B, H, W, C) batch: divide each image into two equal halves along HEIGHT and stack the halves into the batch axis — output (2·B, H/2, W, C) with all the TOP halves first, then all the bottom halves. Pattern: 'b (two h) w c -> (two b) h w c' with two=2.

## Reference solution


```python
import torch as t
import einops

arr = t.movedim(t.tensor(np.load('/delta_numbers.npy')), 1, -1)
imgs = arr

def solve(imgs):
    return einops.rearrange(imgs, 'b (two h) w c -> (two b) h w c', two=2)

print(solve(imgs))
```

## Why this works

Splitting height as (two h) names the top half two=0 and bottom half two=1; moving `two` to the FRONT of the output batch group means all tops come first, then all bottoms. '(b two)' instead would interleave per-image — order inside a group is the whole semantics.
