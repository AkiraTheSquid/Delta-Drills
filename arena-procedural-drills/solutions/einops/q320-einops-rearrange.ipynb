## Problem

For a batch of images, divide each image into two equal halves along the height dimension and then stack these halves vertically, resulting in a new batch that has twice the number of images.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the height axis into a leading half-selector of size 2, then merge that selector into the batch axis to produce twice as many shorter images.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    return einops.rearrange(arr, 'b (two h) w c -> (two b) h w c', two=2)

print(solve())
```

## Why this works

Splitting height into (two h) with two=2 separates each image's top and bottom halves, and merging that factor into the batch via (two b) stacks the halves as new batch elements. The result doubles the batch count while halving the height.
