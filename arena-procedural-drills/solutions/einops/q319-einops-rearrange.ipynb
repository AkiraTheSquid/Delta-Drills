## Problem

Write a function solve(arr) that takes a channels-FIRST batch of shape (b, c, h, w) and returns the channels-LAST layout (b, h, w, c) — same values, the channel axis moved to the end.


<details><summary>💡 Hint (click to reveal)</summary>

A pure axis permutation — every letter appears on both sides, just reordered.

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b c h w -> b h w c')


print(solve(np.arange(6.0).reshape(1, 3, 1, 2)))
```

## Why this works

rearrange(arr, 'b c h w -> b h w c') is np.transpose(arr, (0, 2, 3, 1)): the PyTorch-to-image-library layout conversion. No data moves conceptually — only the axis order changes.
