## Problem

You are given arr, a 4D uint8 array of shape (6, 3, 150, 150) representing a batch of 6 RGB digit images stored in channels-first (batch, channels, height, width) layout. Return a 4D array of shape (6, 150, 150, 3) that holds the same images in channels-last (batch, height, width, channels) layout by moving the channel axis to the trailing position.


<details><summary>💡 Hint (click to reveal)</summary>

Just relist the axes in the target order on the right-hand side; no factors or merges are needed for a layout swap.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b = 2

def solve():
    return einops.rearrange(arr, 'b c h w -> b h w c')

print(solve())
```

## Why this works

This is a pure axis permutation moving the channel axis from position 1 to the last position, converting BCHW to BHWC. einops.rearrange handles it by naming the axes and listing them in the desired output order.
