## Problem

You are given arr, a 4D array of shape (batch, channels, height, width) = (6, 3, 150, 150) holding channels-first digit images, and you use only the first 2 images arr[0:2]. Stack these 2 images vertically into a single tall image, and at the same time duplicate the result horizontally so the right half is an exact copy of the left half, producing a 3D channels-first array of shape (channels, 2 * height, 2 * width) = (3, 300, 300).


<details><summary>💡 Hint (click to reveal)</summary>

Use repeat: stack the batch into the height composite and fold a literal 2 into the width, putting channels first.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]
b = 2

def solve():
    return einops.repeat(arr[0:2], 'b c h w -> c (b h) (2 w)')

print(solve())
```

## Why this works

Merging the batch axis into height with `(b h)` stacks the two images vertically, and a literal 2 folded into width `(2 w)` duplicates the columns to copy the left half onto the right. The channel axis is moved to the front for channels-first output.
