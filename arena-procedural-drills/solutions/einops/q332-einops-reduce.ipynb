## Problem

For an image represented by its height, width, and color channels, compute the maximum value for each row across the color channels and then subtract this maximum from the original image, maintaining the same dimensions.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce over width with max while keeping a singleton width axis, then rely on broadcasting to subtract.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    mx = einops.reduce(img, 'h w c -> h () c', 'max')
    return img - mx

print(solve())
```

## Why this works

reduce with output pattern 'h () c' takes the max over width for each (row, channel) but keeps a size-1 width axis so the result broadcasts back against the original image. Subtracting it removes the per-row, per-channel maximum at every spatial location.
