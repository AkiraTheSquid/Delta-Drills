## Problem

You are given x, a 4D floating-point array of shape (batch, channels, height, width). For each (batch, channels) pair, compute the mean over the spatial axes (height, width) and subtract that scalar mean from every spatial location of the corresponding (batch, channel) feature map. Return a 4D array of shape (batch, channels, height, width) in which each per-image, per-channel slice now has zero spatial mean.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce the spatial axes to size-1 with mean, keeping batch and channel, then subtract via broadcasting.

</details>

## Reference solution


```python
import numpy as np
import einops

x = np.arange(2 * 12 * 8 * 8).reshape(2, 12, 8, 8).astype(np.float32)
b = 2

def solve():
    return x - einops.reduce(x, 'b c h w -> b c 1 1', 'mean')

print(solve())
```

## Why this works

reduce with output 'b c 1 1' averages over the spatial axes per (batch, channel) while retaining singleton h and w axes. Those size-1 axes broadcast against x so the subtraction zeros the spatial mean of every per-image, per-channel feature map.
