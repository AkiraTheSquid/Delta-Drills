## Problem

Write a function solve(x) that takes a 4-D tensor of shape (b, c, h, w) and CENTERS each (batch, channel) feature map: subtract from every pixel the mean over ITS OWN map's h and w. Return the same shape; the trick is reducing to 'b c 1 1' so the means broadcast back.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce to literal '1 1' instead of dropping the letters — keepdims in pattern form, so the subtraction broadcasts.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(x):
    return x - einops.reduce(x, 'b c h w -> b c 1 1', 'mean')


print(solve(np.arange(8.0).reshape(1, 2, 2, 2)))
```

## Why this works

reduce(x, 'b c h w -> b c 1 1', 'mean') keeps singleton spatial axes (einops' keepdims), so x minus it aligns pixel-for-pixel — x - x.mean(axis=(2,3), keepdims=True), the centering step of instance normalization.
