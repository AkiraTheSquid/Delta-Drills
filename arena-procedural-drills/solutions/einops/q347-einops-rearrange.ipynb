## Problem

Write a function solve(img) that takes a channels-first image (c, h, w) and lays its channels out HORIZONTALLY as one 2-D array: return shape (h, c*w) via 'c h w -> h (c w)' — channel 0's image, then channel 1's, side by side.


<details><summary>💡 Hint (click to reveal)</summary>

Merging '(c w)' with c outermost puts whole channel images next to each other — height stays shared.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import rearrange, reduce, repeat

def solve(img):
    return einops.rearrange(img, 'c h w -> h (c w)')


print(solve(np.arange(8.0).reshape(2, 2, 2)))
```

## Why this works

'c h w -> h (c w)' drops the channel axis into width, c-major, so the channels appear as separate grayscale panels in a row — the standard per-channel visualization strip.
