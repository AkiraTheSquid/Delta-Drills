## Problem

You are given img, a 3D array of shape (channels, height, width) representing a single channels-first RGB image. Return a 2D array of shape (height, channels * width) in which the per-channel 2D slices have been laid out side-by-side along the width axis, so the three color planes appear as three grayscale panels stacked horizontally in channel order.


<details><summary>💡 Hint (click to reveal)</summary>

Use rearrange and merge the channel dimension into the width by writing it as a composite output axis, keeping height first.

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
img = arr[0]

def solve():
    return einops.rearrange(arr[0], 'c h w -> h (c w)')

print(solve())
```

## Why this works

The channels-first slice (c, h, w) is rearranged so that the channel axis is merged into the width axis with `(c w)`, laying each color plane out as a horizontal panel. Keeping `h` as the leading output axis preserves vertical structure while the three planes sit side by side along the merged width.
