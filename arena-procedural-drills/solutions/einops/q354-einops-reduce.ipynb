## Problem

For a (b, h, w, c) tensor, subtract its per-image spatial minimum (one min value per (b, c) pair) using broadcasting.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce with 'min' over the two spatial axes but keep them as size-1 dimensions, then subtract.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.arange(2 * 3 * 4 * 5).reshape(2, 3, 4, 5)

def solve():
    mn = einops.reduce(arr, 'b h w c -> b 1 1 c', 'min')
    return arr - mn

print(solve())
```

## Why this works

Reducing over the spatial axes h and w with keep-dims `b 1 1 c` yields one min per (b, c) pair while preserving broadcastable shape. Subtracting this from the original tensor relies on NumPy broadcasting across the singleton spatial dims.
