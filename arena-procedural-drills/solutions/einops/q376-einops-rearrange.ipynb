## Problem

You are given img_a and img_b, two 3D uint8 arrays each of shape (150, 150, 3) in (height, width, channels) layout. Treat them as a batch of 2 images and concatenate them side by side along the width axis, returning a 3D array of shape (150, 2*150, 3) in (height, batch*width, channels) layout so img_b sits immediately to the right of img_a.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

# Standalone fixture mirroring the ARENA digits array (6, 3, 150, 150) uint8.
# In the grader this is replaced by np.load('/delta_numbers.npy').
np.random.seed(0)
_NUMBERS = np.random.randint(0, 256, size=(6, 3, 150, 150), dtype=np.uint8)
def _load_numbers():
    try:
        return np.load('/delta_numbers.npy')
    except FileNotFoundError:
        return _NUMBERS

arr = np.moveaxis(_load_numbers(), 1, -1)
img = arr[0]
b = 2
img_a = img
img_b = img

def solve():
    return einops.rearrange([img_a, img_b], 'b h w c -> h (b w) c')

print(solve())
```

## Why this works

Passing a Python list to rearrange stacks the elements into a new leading batch axis b=2. Folding (b w) then places img_b's columns immediately after img_a's along width, producing a (h, 2*w, c) side-by-side composite.
