## Problem

Write solve(imgs) for a float (B, H, W, C) batch with W even: halve the WIDTH by averaging each pair of adjacent columns. Pattern: einops.reduce 'b h (w w2) c -> b h w c' with 'mean', w2=2.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy').astype(np.float32), 1, -1)[::-1].copy()
imgs = arr

def solve(imgs):
    return einops.reduce(imgs, 'b h (w w2) c -> b h w c', 'mean', w2=2)

print(solve(imgs))
```

## Why this works

Grouping width as (w w2) with w2=2 pairs adjacent columns; reducing with 'mean' averages each pair, halving the width. 1-D average pooling along a single axis — the same window mechanics as 2-D pooling, applied to width only.
