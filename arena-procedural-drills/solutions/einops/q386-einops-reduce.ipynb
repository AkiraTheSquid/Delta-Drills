## Problem

For a batch of images with multiple color channels, reduce the width of each image by averaging pairs of adjacent columns, resulting in a new width that is half the original.


<details><summary>💡 Hint (click to reveal)</summary>

Factor the width into an outer index and an inner group of 2, then let reduce collapse the inner group with a mean.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy').astype(np.float32), 1, -1)
b = 2

def solve():
    return einops.reduce(arr, 'b h (w w2) c -> b h w c', 'mean', w2=2)

print(solve())
```

## Why this works

Splitting the width axis into a (w, w2) pair with w2=2 groups every adjacent column-pair together; reducing with 'mean' over the disappearing w2 axis averages each pair, halving the width while keeping batch, height, and channels intact.
