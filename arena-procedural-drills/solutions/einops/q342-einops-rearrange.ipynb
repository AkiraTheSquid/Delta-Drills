## Problem

For a batch of images with spatial dimensions and color channels, first combine the batch and height dimensions into a single dimension, and then restore the original structure to separate the batch and height back into distinct dimensions.


<details><summary>💡 Hint (click to reveal)</summary>

Merge batch with height, then invert by splitting the combined axis with both sizes provided.

## Reference solution


```python
import numpy as np
import einops

arr = np.moveaxis(np.load('/delta_numbers.npy'), 1, -1)
b, h = 6, 150

def solve():
    tmp = einops.rearrange(arr, 'b h w c -> (b h) w c')
    return einops.rearrange(tmp, '(b h) w c -> b h w c', b=b, h=h)

print(solve())
```

## Why this works

The first rearrange merges batch and height into one axis; the second splits '(b h)' back apart, supplying both b and h so the factorization is determined, restoring the original (6, 150, 150, 3) tensor.
