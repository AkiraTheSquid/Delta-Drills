## Reference solution


```python
import torch as t
import einops

arr = t.tensor(np.load('/delta_numbers.npy')).to(t.float32)
img = arr[0]

def solve(img):
    return einops.reduce(img, 'c (h h2) (w w2) -> c h w', 'mean', h2=3, w2=3)

print(solve(img))
```

## Why this works

Same pooling idea as 2×2 average pooling, but the window factors are 3 — each output pixel is the mean of a 3×3 block, shrinking H and W by 3× while channels pass through. Works for any window size divisible into the spatial dims.
