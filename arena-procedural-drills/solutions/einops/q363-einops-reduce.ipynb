## Problem

Apply average pooling with a 3x3 filter to a single image, reducing the height and width dimensions by a factor of three while keeping the number of channels the same.


<details><summary>💡 Hint (click to reveal)</summary>

Factor height and width into outer indices and size-3 inner blocks, then reduce those blocks with 'mean'.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy').astype(np.float32)
img = arr[0]

def solve():
    return einops.reduce(img, 'c (h h2) (w w2) -> c h w', 'mean', h2=3, w2=3)

print(solve())
```

## Why this works

Factoring each spatial axis into an outer index and a size-3 inner block, then reducing the inner blocks with 'mean', performs non-overlapping 3x3 average pooling. Channels are preserved and the spatial dimensions shrink by a factor of three.
