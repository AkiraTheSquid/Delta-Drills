## Problem

Write a function solve(arr) that takes a batch (b, c, h, w) and merges batch into width with the batch index INNERMOST: return shape (c, h, w*b) via 'b c h w -> c h (w b)' — the images' columns INTERLEAVE (column 0 of every image, then column 1, ...) instead of sitting side by side. (Contrast '(b w)', which concatenates whole images.)


<details><summary>💡 Hint (click to reveal)</summary>

'(w b)' with b INNER interleaves the batches column by column — the merge order is the semantics.

</details>

## Reference solution


```python
import torch as t
import einops
from einops import rearrange, reduce, repeat

def solve(arr):
    return einops.rearrange(arr, 'b c h w -> c h (w b)')


print(solve(t.arange(8.0).reshape(2, 1, 2, 2)))
```

## Why this works

In '(w b)', b varies fastest: for each original column position, every image contributes its column before moving on — a comb-interleave. '(b w)' would lay whole images side by side. The two patterns produce identical shapes and totally different pictures.
