## Problem

Given a 4D array arr of shape (batch, channels, height, width) holding a batch of channels-first color images, return a 3D array of shape (channels, height, batch * width) that lays every image side by side into one long horizontal strip: image 0 occupies the leftmost width columns, image 1 the next width columns, and so on, with channels and height preserved.


<details><summary>💡 Hint (click to reveal)</summary>

Keep channels and height up front, then merge batch into the width axis so images sit side by side.

</details>

## Reference solution


```python
import numpy as np
import einops

arr = np.load('/delta_numbers.npy')
b = 2

def solve():
    return einops.rearrange(arr, 'b c h w -> c h (b w)')

print(solve())
```

## Why this works

Channels and height move to the front and the batch axis is composed with width as (b w), so each image's columns are laid out consecutively along the width, forming one wide horizontal strip.
