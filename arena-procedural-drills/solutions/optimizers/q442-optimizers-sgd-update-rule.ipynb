## Problem

Minimize (x − c)² by hand-rolled gradient descent. Write solve(x0, c, lr, steps): start from the leaf `x = torch.tensor([x0], requires_grad=True)`. Each step: build the loss `(x - c) ** 2`, backprop it, update `x` IN PLACE inside `torch.no_grad()` (why is that context required for a leaf that requires grad?), and zero the gradient before the next iteration (what would accumulate if you didn't?). Return `round(float(x.detach()), 4)`.

Each step: fresh loss, `backward()`, `x -= lr * x.grad` under no_grad, then `x.grad.zero_()`.

## Reference solution


```python
import torch

def solve(x0, c, lr, steps):
    x = torch.tensor([x0], requires_grad=True)
    for _ in range(steps):
        loss = (x - c) ** 2
        loss.backward()
        with torch.no_grad():
            x -= lr * x.grad
        x.grad.zero_()
    return round(float(x.detach()), 4)

print(solve(2.0, 5.0, 0.3, 3))
```

## Why this works

Two traps: the in-place update must sit inside `torch.no_grad()` (mutating a requires-grad leaf mid-graph is an error), and `.grad` ACCUMULATES — skip `zero_()` and step 2 applies grad₁+grad₂. For (x−c)² each step multiplies the distance to c by (1−2lr), which is how the closed-form expected values fall out.
