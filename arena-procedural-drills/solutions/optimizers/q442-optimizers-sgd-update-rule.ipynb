## Problem

Use autograd plus the manual SGD update rule to minimize loss = (x - 5.0)**2 starting from x = 2.0 with learning rate lr = 0.3. Run EXACTLY 3 gradient-descent steps. In each step: compute the loss, call loss.backward(), then update x in-place under torch.no_grad() with x -= lr * x.grad, and finally zero x.grad. After the 3 steps, print round(float(x.detach()), 4).

## Reference solution


```python
import torch

x = torch.tensor([2.0], requires_grad=True)
lr = 0.3
for _ in range(3):
    loss = (x - 5.0) ** 2
    loss.backward()
    with torch.no_grad():
        x -= lr * x.grad
    x.grad.zero_()
print(round(float(x.detach()), 4))
```

## Why this works

Each step computes the gradient of (x-5)^2 (which is 2*(x-5)), then nudges x toward 5 by lr times that gradient under no_grad, zeroing the gradient before the next step. Three such steps drive x from 2.0 to about 4.808.
