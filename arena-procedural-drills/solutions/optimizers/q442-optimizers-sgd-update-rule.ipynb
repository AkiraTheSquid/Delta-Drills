## Problem

Minimize $(x-5)^2$ from $x=2$ with three hand-rolled gradient-descent steps (lr 0.3 — all in the starter). Each step: build the loss, backprop it, update `x` in place inside `torch.no_grad()` (why is that context required for a leaf that requires grad?), and zero the gradient before the next iteration (what would accumulate if you didn't?). Then print round(float(x.detach()), 4).

## Reference solution


```python
import torch

x = torch.tensor([2.0], requires_grad=True)
lr = 0.3
for _ in range(3):
    loss = (x - 5.0) ** 2
    loss.backward()
    with torch.no_grad():
        x -= lr * x.grad
    x.grad.zero_()
print(round(float(x.detach()), 4))
```

## Why this works

Each step computes the gradient of (x-5)^2 (which is 2*(x-5)), then nudges x toward 5 by lr times that gradient under no_grad, zeroing the gradient before the next step. Three such steps drive x from 2.0 to about 4.808.
