## Problem

One step of vanilla SGD by hand: parameters move a small step AGAINST the gradient. Write solve(theta, grad, lr): apply the SGD update rule from memory and return the updated parameter tensor (do not modify `theta` in place).


<details><summary>💡 Hint (click to reveal)</summary>

theta_new = `theta - lr * grad` — move AGAINST the gradient.

## Reference solution


```python
import torch

def solve(theta, grad, lr):
    return theta - lr * grad

theta = torch.tensor([1.0, 2.0, 3.0], dtype=torch.float64)
grad = torch.tensor([0.5, -1.0, 2.0], dtype=torch.float64)
print(solve(theta, grad, 0.1))
```

## Why this works

Vanilla SGD: θ ← θ − lr·∇L. The minus sign is the whole rule — the gradient points uphill, parameters step downhill. With lr = 0 nothing moves; with lr = 1 and grad = θ you land at zero.
