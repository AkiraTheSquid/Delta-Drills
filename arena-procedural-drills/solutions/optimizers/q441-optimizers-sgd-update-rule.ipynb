## Problem

Apply ONE step of vanilla SGD by hand (no optimizer object). Given parameters theta = [1.0, 2.0, 3.0], gradient grad = [0.5, -1.0, 2.0], and learning rate lr = 0.1, compute the updated parameters theta_new = theta - lr*grad as a float64 torch tensor. Print the result as a Python list with each element rounded to 4 decimals, i.e. print([round(v, 4) for v in theta_new.tolist()]).


<details><summary>💡 Hint (click to reveal)</summary>

Subtract lr times the gradient from the parameters — one elementwise update.

</details>

## Reference solution


```python
import torch

theta = torch.tensor([1.0, 2.0, 3.0], dtype=torch.float64)
grad = torch.tensor([0.5, -1.0, 2.0], dtype=torch.float64)
lr = 0.1
theta_new = theta - lr * grad
print([round(v, 4) for v in theta_new.tolist()])
```

## Why this works

Vanilla SGD moves parameters one learning-rate step opposite the gradient: theta_new = theta - lr*grad. Applied elementwise to the given vectors this yields [0.95, 2.1, 2.8].
