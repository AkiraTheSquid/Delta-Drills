## Problem

Iterators exhaust; lists don't. Write solve(make_params) where `make_params()` returns a ONE-SHOT iterator over parameter tensors. First sum `p.numel()` over the same iterator twice — `first_pass`, then `second_pass` (predict the second before running). Then materialize a fresh `make_params()` into a list and take both passes again (`right_first`, `right_second`). Return the list `[first_pass, second_pass, right_first, right_second]`. This is exactly why optimizers materialize `model.parameters()` up front.

Sum numel over the iterator twice — the second pass sees nothing. Lists don't exhaust.

## Reference solution


```python
import torch

def solve(make_params):
    it = make_params()
    first_pass = sum(p.numel() for p in it)
    second_pass = sum(p.numel() for p in it)
    params = list(make_params())
    right_first = sum(p.numel() for p in params)
    right_second = sum(p.numel() for p in params)
    return [first_pass, second_pass, right_first, right_second]

make_params = lambda: iter([torch.tensor([1.0, 2.0, 3.0]), torch.tensor([4.0, 5.0])])
print(solve(make_params))
```

## Why this works

An iterator yields each item once, ever: the second sum over the same iterator gets 0 items. `list(make_params())` materializes the tensors so every later pass sees all of them — exactly why `torch.optim` converts `model.parameters()` (a generator) to a list at construction.
