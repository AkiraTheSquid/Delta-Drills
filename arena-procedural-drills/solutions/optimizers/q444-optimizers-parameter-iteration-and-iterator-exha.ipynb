## Problem

An iterator over parameters exhausts after one pass. Define make_params() that returns iter([torch.tensor([1.,2.,3.]), torch.tensor([4.,5.])]). First, get one iterator `it = make_params()` and compute the total number of elements (sum of p.numel()) by iterating it TWICE: first_pass then second_pass. Then convert a fresh make_params() to a list and again compute the total twice: right_first then right_second. Print the list [first_pass, second_pass, right_first, right_second] (use plain print on the list of ints).

Iterating an iterator exhausts it; iterating a list does not. Watch what the second pass over each sees.

## Reference solution


```python
import torch

def make_params():
    return iter([torch.tensor([1.0, 2.0, 3.0]), torch.tensor([4.0, 5.0])])

it = make_params()
first_pass = sum(p.numel() for p in it)
second_pass = sum(p.numel() for p in it)

params = list(make_params())
right_first = sum(p.numel() for p in params)
right_second = sum(p.numel() for p in params)

print([first_pass, second_pass, right_first, right_second])
```

## Why this works

An iterator is consumed once: the first sum drains it (3+2=5) and the second sees nothing (0). Materializing into a list captures the elements so they can be traversed repeatedly, giving 5 on both passes. This is exactly why optimizers store parameters in a list rather than an iterator.
