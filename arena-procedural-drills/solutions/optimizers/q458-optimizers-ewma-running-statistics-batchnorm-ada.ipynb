## Problem

Adam's first-moment estimate is an EWMA of the gradients with decay `beta`: the moment keeps a `beta` share of itself and absorbs a `(1−beta)` share of each new gradient. Write solve(m, beta, grads): fold the list of gradient tensors into `m` in order with that update, and return the final biased (un-corrected) moment tensor.


<details><summary>💡 Hint (click to reveal)</summary>

Fold in order: `m = beta * m + (1 - beta) * g` per gradient.

</details>

## Reference solution


```python
import torch

def solve(m, beta, grads):
    for g in grads:
        m = beta * m + (1 - beta) * g
    return m

m = torch.zeros(2)
grads = [torch.tensor([1.0, -2.0]), torch.tensor([3.0, 0.0]), torch.tensor([-1.0, 4.0])]
print(solve(m, 0.9, grads))
```

## Why this works

Adam's first moment is an exponentially-weighted moving average: keep a β share of the running value, absorb (1−β) of the new gradient. From m₀ = 0 the estimate is biased toward zero early — that's what Adam's bias correction divides away; with β = 0 the EWMA just tracks the latest gradient.
