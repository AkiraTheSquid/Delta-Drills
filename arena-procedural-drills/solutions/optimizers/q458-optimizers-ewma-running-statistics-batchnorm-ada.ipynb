## Problem

Adam's first-moment estimate m is an EWMA of the gradients: m = beta * m + (1 - beta) * g, which is the same running-stat update with mu = (1 - beta). Using torch, start m = torch.zeros(2) and beta = 0.9. Apply the update once per gradient in this order: torch.tensor([1.0, -2.0]), torch.tensor([3.0, 0.0]), torch.tensor([-1.0, 4.0]). Print the final (biased, un-corrected) first moment as a plain python list of floats each rounded to 4 decimals: print([round(float(x), 4) for x in m]).

Use the running-stat update with mu = 1 - beta, applied component-wise to each gradient in order; skip any bias correction.

## Reference solution


```python
import torch
beta = 0.9
m = torch.zeros(2)
grads = [
    torch.tensor([1.0, -2.0]),
    torch.tensor([3.0, 0.0]),
    torch.tensor([-1.0, 4.0]),
]
for g in grads:
    m = beta * m + (1 - beta) * g
print([round(float(x), 4) for x in m])
```

## Why this works

Adam's first moment m is an EWMA of gradients with decay beta, equivalently the running-stat update with mu = 1-beta. Applying it element-wise per gradient gives the biased (un-corrected) first moment, which Adam would later divide by 1-beta^t for bias correction.
