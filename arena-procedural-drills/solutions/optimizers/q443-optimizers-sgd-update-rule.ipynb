## Problem

Write a function solve(X, y, theta, lr) performing ONE gradient-descent step on linear-regression MSE loss mean((X @ theta - y)^2): the gradient is (2/n) * X.T @ (X @ theta - y) where n = len(y); return theta - lr * gradient (a new tensor; do not modify theta).


<details><summary>💡 Hint (click to reveal)</summary>

Residual first (X theta - y), then push it back through X.T with the 2/n factor from differentiating the squared mean.

</details>

## Reference solution


```python
import torch as t

def solve(X, y, theta, lr):
    return theta - lr * (2.0 / len(y)) * (X.T @ (X @ theta - y))


print(solve(t.tensor([[1.0, 1.0], [1.0, 2.0]]), t.tensor([2.0, 3.0]), t.zeros(2), 0.1))
```

## Why this works

d/dtheta mean((X theta - y)^2) = (2/n) X.T (X theta - y) — residuals weighted by each feature. Stepping against it is one SGD (full-batch) update; at the exact solution (case 2) the residual is zero and theta doesn't move.
