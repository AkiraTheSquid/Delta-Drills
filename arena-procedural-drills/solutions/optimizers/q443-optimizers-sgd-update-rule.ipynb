## Problem

One SGD step on linear-regression MSE in numpy. The starter defines the design matrix `X`, targets `y`, initial `theta`, `lr`, and `n`. The MSE gradient is $\nabla_\theta = \frac{2}{n} X^\top (X\theta - y)$ — translate it to code, apply one descent step, and print [round(float(v), 4) for v in theta_new].


<details><summary>💡 Hint (click to reveal)</summary>

Compute the residual X@theta - y first, project it back through X.T, scale by 2/n, then take a single gradient-descent step.

## Reference solution


```python
import numpy as np
theta = np.array([0.0, 0.0])
X = np.array([[1.0, 1.0], [1.0, 2.0], [1.0, 3.0]])
y = np.array([1.0, 2.0, 3.0])
lr = 0.1
n = len(y)
grad = (2.0 / n) * X.T @ (X @ theta - y)
theta_new = theta - lr * grad
print([round(float(v), 4) for v in theta_new])
```

## Why this works

One SGD step is just theta minus the learning rate times the MSE gradient. The gradient (2/n)*X.T@(X@theta - y) is the analytic derivative of mean squared error w.r.t. the parameters; evaluating it at theta=0 gives the residuals' contribution, and subtracting lr*grad nudges theta downhill.
