## Problem

Write a function solve(values, mu) that folds a list of values into an EWMA running statistic starting from 0.0: for each v, running = (1 - mu) * running + mu * v. Return the final running value as a float.


<details><summary>💡 Hint (click to reveal)</summary>

Each step keeps a (1-mu) share of the old value and adds a mu share of the new one — a loop with one line of state.

</details>

## Reference solution


```python
import torch as t

def solve(values, mu):
    return (lambda r: r)(__import__('functools').reduce(lambda acc, v: (1 - mu) * acc + mu * v, values, 0.0))


print(solve([5.0, 7.0, 3.0, 9.0, 2.0], 0.1))
```

## Why this works

running = (1-mu)*running + mu*v is the exponentially weighted moving average: recent values dominate with rate mu. mu=1 tracks only the latest value, mu=0 never learns — the two degenerate cases bracketing BatchNorm's momentum and Adam's moment estimates.
