## Problem

An exponentially-weighted moving average (EWMA) updates a running statistic as running = (1 - mu) * running + mu * new_value. Start with running = 0.0 and mu = 0.1, then apply the update once for each value in the list [5.0, 7.0, 3.0, 9.0, 2.0] in order. Print the final running value rounded to 4 decimals via print(round(float(running), 4)).


<details><summary>💡 Hint (click to reveal)</summary>

Loop through the values in order, each time mixing (1-mu) of the old running value with mu of the new value.

</details>


## Reference solution


```python
running = 0.0
mu = 0.1
values = [5.0, 7.0, 3.0, 9.0, 2.0]
for v in values:
    running = (1 - mu) * running + mu * v
print(round(float(running), 4))
```

## Why this works

An EWMA blends the old running value with each new observation, weighting the new one by mu and decaying the past by (1-mu). Applying it sequentially over the list lets recent values dominate while older ones fade geometrically.
