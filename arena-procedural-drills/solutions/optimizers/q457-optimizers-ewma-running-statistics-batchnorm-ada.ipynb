## Problem

Write a function solve(batches, momentum) that maintains BatchNorm-style RUNNING statistics: starting from running_mean = zeros(d) and running_var = ones(d), fold in each batch (a 2-D tensor of shape (n, d)) with running = (1 - momentum) * running + momentum * batch_stat, using the batch's per-column mean and POPULATION variance. Return the tuple (running_mean, running_var).

## Reference solution


```python
import torch as t

def solve(batches, momentum):
    return (lambda rm_rv: rm_rv)(__import__('functools').reduce(lambda acc, b: ((1 - momentum) * acc[0] + momentum * b.mean(dim=0), (1 - momentum) * acc[1] + momentum * b.var(dim=0, correction=0)), batches, (t.zeros(batches[0].shape[1]), t.ones(batches[0].shape[1]))))


print(solve([t.tensor([[1.0, 2.0], [3.0, 4.0]])], 0.2))
```
