## Problem

Write a function solve(batches, momentum) that maintains BatchNorm-style RUNNING statistics: starting from running_mean = zeros(d) and running_var = ones(d), fold in each batch (a 2-D array of shape (n, d)) with running = (1 - momentum) * running + momentum * batch_stat, using the batch's per-column mean and POPULATION variance. Return the tuple (running_mean, running_var).


<details><summary>💡 Hint (click to reveal)</summary>

Two EWMAs side by side — one over batch means, one over batch (population) variances — starting from the (0, 1) prior.

</details>

## Reference solution


```python
import numpy as np

def solve(batches, momentum):
    return (lambda rm_rv: rm_rv)(__import__('functools').reduce(lambda acc, b: ((1 - momentum) * acc[0] + momentum * b.mean(axis=0), (1 - momentum) * acc[1] + momentum * b.var(axis=0)), batches, (np.zeros(batches[0].shape[1]), np.ones(batches[0].shape[1]))))


print(solve([np.array([[1.0, 2.0], [3.0, 4.0]])], 0.2))
```

## Why this works

BatchNorm's inference-time statistics are exactly these EWMAs: each training batch nudges running_mean toward its column means and running_var toward its column variances (ddof=0) at rate `momentum`. The zeros/ones initialization is why early estimates are biased toward a standard normal.
