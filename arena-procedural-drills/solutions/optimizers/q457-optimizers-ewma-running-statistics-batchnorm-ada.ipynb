## Problem

BatchNorm's running statistics are EWMAs with rate `momentum`. The starter gives `running_mean` (zeros), `running_var` (ones), momentum 0.2, and two 2×3 batches (rows are samples, columns are features). For each batch in order: compute the per-feature mean and per-feature POPULATION variance, then EWMA-update both running stats. Print running_mean.round(4).tolist() on line 1 and running_var.round(4).tolist() on line 2. (running_var comes out all ones — check the batch variances to see why.)

## Reference solution


```python
import numpy as np
running_mean = np.zeros(3)
running_var = np.ones(3)
momentum = 0.2
batches = [
    np.array([[1.0, 2.0, 3.0], [3.0, 4.0, 5.0]]),
    np.array([[2.0, 0.0, 4.0], [4.0, 2.0, 6.0]]),
]
for b in batches:
    bm = b.mean(axis=0)
    bv = b.var(axis=0)
    running_mean = (1 - momentum) * running_mean + momentum * bm
    running_var = (1 - momentum) * running_var + momentum * bv
print(running_mean.round(4).tolist())
print(running_var.round(4).tolist())
```

## Why this works

BatchNorm tracks running_mean and running_var with the same EWMA update, using the per-feature batch statistics computed along axis 0. Each feature column gets its own mean and population variance; here both batches happen to have variance 1 per feature, so running_var stays 1.
