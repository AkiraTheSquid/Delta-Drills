## Problem

BatchNorm keeps running_mean and running_var updated with EWMA: running = (1 - momentum) * running + momentum * batch_stat. Using numpy, start running_mean = zeros(3) and running_var = ones(3) with momentum = 0.2. For each of the two batches b1 = [[1,2,3],[3,4,5]] and b2 = [[2,0,4],[4,2,6]] (rows are samples, columns are features), compute the per-feature batch mean (b.mean(axis=0)) and per-feature batch variance (b.var(axis=0), population variance) and apply the EWMA update to running_mean and running_var respectively, processing b1 then b2. Print running_mean.round(4).tolist() on the first line and running_var.round(4).tolist() on the second line.

## Reference solution


```python
import numpy as np
running_mean = np.zeros(3)
running_var = np.ones(3)
momentum = 0.2
batches = [
    np.array([[1.0, 2.0, 3.0], [3.0, 4.0, 5.0]]),
    np.array([[2.0, 0.0, 4.0], [4.0, 2.0, 6.0]]),
]
for b in batches:
    bm = b.mean(axis=0)
    bv = b.var(axis=0)
    running_mean = (1 - momentum) * running_mean + momentum * bm
    running_var = (1 - momentum) * running_var + momentum * bv
print(running_mean.round(4).tolist())
print(running_var.round(4).tolist())
```

## Why this works

BatchNorm tracks running_mean and running_var with the same EWMA update, using the per-feature batch statistics computed along axis 0. Each feature column gets its own mean and population variance; here both batches happen to have variance 1 per feature, so running_var stays 1.
