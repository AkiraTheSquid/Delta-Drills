## Problem

A generator expression is a one-shot iterator. Create gen = (torch.tensor(float(i)) for i in [2.0, 3.0, 4.0]). Convert it once with params = list(gen). Compute count = len(params), then total = sum(p.item() for p in params), then total_again = sum(p.item() for p in params) (a second pass over the SAME list). Print [count, round(total,4), round(total_again,4)].


<details><summary>💡 Hint (click to reveal)</summary>

Convert the generator to a list immediately, then both the count and the two sum passes operate on stored elements.

</details>

## Reference solution


```python
import torch

gen = (torch.tensor(float(i)) for i in [2.0, 3.0, 4.0])
params = list(gen)
count = len(params)
total = sum(p.item() for p in params)
total_again = sum(p.item() for p in params)
print([count, round(total, 4), round(total_again, 4)])
```

## Why this works

A generator expression is a one-shot iterator, so converting it to a list once is what makes repeated traversal possible. After list(gen) the three tensors live in memory; len gives 3 and both sum passes over the list yield 2+3+4=9. A second pass over the generator itself would have given 0.
