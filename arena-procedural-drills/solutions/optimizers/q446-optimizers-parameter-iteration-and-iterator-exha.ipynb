## Problem

A generator expression is a one-shot iterator, but converting it to a list makes it reusable. Write solve(vals): build `gen = (torch.tensor(float(v)) for v in vals)`, materialize `params = list(gen)`, then compute the count of params, the sum of their scalar values (`.item()`), and that same sum AGAIN over the same list — does a list exhaust? Return `[count, round(total, 4), round(total_again, 4)]`.


<details><summary>💡 Hint (click to reveal)</summary>

`list(gen)` materializes; a list can be summed twice and both sums agree.

</details>

## Reference solution


```python
import torch

def solve(vals):
    gen = (torch.tensor(float(v)) for v in vals)
    params = list(gen)
    count = len(params)
    total = sum(p.item() for p in params)
    total_again = sum(p.item() for p in params)
    return [count, round(total, 4), round(total_again, 4)]

print(solve([2.0, 3.0, 4.0]))
```

## Why this works

Generator expressions are one-shot, but converting to a list stores the tensors: `count`, `total`, and `total_again` all work because iteration over a LIST restarts from the top. Contrast with the previous drill where the second pass over a raw iterator saw nothing.
