## Problem

Print the number of learnable parameters in nn.BatchNorm2d(num_features=32). Expected: 2 * num_features (weight + bias).


<details><summary>💡 Hint (click to reveal)</summary>

Each learnable BN tensor has num_features entries and there are two of them.

</details>

## Reference solution


```python
import torch.nn as nn

def solve():
    bn = nn.BatchNorm2d(32)
    return sum(p.numel() for p in bn.parameters())

print(solve())
```

## Why this works

BatchNorm2d learns an affine scale (weight) and shift (bias), each a vector of length num_features. With 32 features that is 32 + 32 = 64 learnable parameters; the running mean/var are buffers, not parameters.
