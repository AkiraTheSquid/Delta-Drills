## Problem

How many LEARNABLE parameters does `nn.BatchNorm2d(num_features)` carry? Think about which per-channel vectors are learnable (weight, bias) and which are running buffers (mean, var — buffers don't count). Write solve(num_features): build the module and return the parameter count (sum of `numel()`) as an int.


<details><summary>💡 Hint (click to reveal)</summary>

2 × num_features — γ and β; the running stats are buffers.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(num_features):
    bn = nn.BatchNorm2d(num_features)
    return sum(p.numel() for p in bn.parameters())

print(solve(32))
```

## Why this works

BatchNorm2d carries two LEARNABLE per-channel vectors (scale γ, shift β) → 2·num_features parameters. Running mean/var look like parameters but are registered as buffers — updated by the forward pass, not by gradients.
