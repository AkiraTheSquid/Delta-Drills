## Problem

How many LEARNABLE parameters does `nn.BatchNorm2d(num_features=32)` carry? Think about which per-channel vectors are learnable and which are running buffers (buffers don't count), then verify by counting parameters programmatically on the starter's `bn`. Print the total as an int.

## Reference solution


```python
import torch.nn as nn

def solve():
    bn = nn.BatchNorm2d(32)
    return sum(p.numel() for p in bn.parameters())

print(solve())
```

## Why this works

BatchNorm2d learns an affine scale (weight) and shift (bias), each a vector of length num_features. With 32 features that is 32 + 32 = 64 learnable parameters; the running mean/var are buffers, not parameters.
