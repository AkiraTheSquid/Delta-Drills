## Problem

A raw tensor stored as a Module attribute is INVISIBLE to `.parameters()`. Inside a Module subclass's `__init__`, wrap `torch.randn(3, 4)` in the class that registers it as learnable, then print exactly `param_count: N` where N counts the module's parameters (an f-string over the count).


<details><summary>💡 Hint (click to reveal)</summary>

nn.Parameter assigned as a Module attribute is auto-registered; count what .parameters() yields.

</details>

## Reference solution


```python
import torch
import torch.nn as nn


def solve():
    class M(nn.Module):
        def __init__(self):
            super().__init__()
            self.w = nn.Parameter(torch.randn(3, 4))
    model = M()
    return f'param_count: {len(list(model.parameters()))}'


print(solve())
```

## Why this works

Wrapping a tensor in nn.Parameter registers it on the Module, so it shows up in model.parameters(). A single registered parameter means len(list(model.parameters())) is 1.
