## Problem

A raw tensor stored as a Module attribute is INVISIBLE to `.parameters()`. Write solve(shape): define a Module subclass whose `__init__` (after `super().__init__()`) stores `torch.randn(*shape)` wrapped in the class that registers it as learnable, and return an instance. The grader counts its parameters and checks the parameter's shape.


<details><summary>💡 Hint (click to reveal)</summary>

`nn.Parameter(torch.randn(*shape))` — the wrapper IS the registration.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(shape):
    class M(nn.Module):
        def __init__(self, s):
            super().__init__()
            self.w = nn.Parameter(torch.randn(*s))
    return M(shape)

model = solve((3, 4))
print(model)
```

## Why this works

A raw tensor attribute is invisible to `.parameters()`; wrapping it in `nn.Parameter` marks it learnable and auto-registers it on assignment. That's the entire difference — nn.Parameter is a Tensor subclass whose only job is 'count me'.
