## Problem

Wrap a raw torch.randn(3, 4) tensor as a learnable parameter on a Module using nn.Parameter. Print whether it appears in .parameters() as `param_count: 1`.


<details><summary>💡 Hint (click to reveal)</summary>

nn.Parameter assigned as a Module attribute is auto-registered; count what .parameters() yields.

</details>

## Reference solution


```python
import torch
import torch.nn as nn


def solve():
    class M(nn.Module):
        def __init__(self):
            super().__init__()
            self.w = nn.Parameter(torch.randn(3, 4))
    model = M()
    return f'param_count: {len(list(model.parameters()))}'


print(solve())
```

## Why this works

Wrapping a tensor in nn.Parameter registers it on the Module, so it shows up in model.parameters(). A single registered parameter means len(list(model.parameters())) is 1.
