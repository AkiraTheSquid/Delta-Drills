## Problem

Why is bias=False used when a Conv2d is immediately followed by BatchNorm2d? Build the pair `nn.Sequential(nn.Conv2d(3,16,3,bias=False), nn.BatchNorm2d(16))` and print the number of learnable parameters using sum(p.numel() for p in seq.parameters()).


<details><summary>💡 Hint (click to reveal)</summary>

Sum p.numel() over the Sequential; remember the conv has no bias and BN contributes two vectors of size num_features.

## Reference solution


```python
import torch.nn as nn

def solve():
    seq = nn.Sequential(nn.Conv2d(3, 16, 3, bias=False), nn.BatchNorm2d(16))
    return sum(p.numel() for p in seq.parameters())

print(solve())
```

## Why this works

BatchNorm immediately re-centers and re-scales conv outputs, so any conv bias would be cancelled by BN's mean subtraction, making it redundant. The conv has 16*3*3*3 = 432 weights and no bias; BN adds weight+bias of 16 each, so 432 + 32 = 464 learnable params.
