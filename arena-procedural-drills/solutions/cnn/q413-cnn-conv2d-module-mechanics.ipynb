## Problem

Why is `bias=False` safe when a Conv2d feeds straight into BatchNorm2d? Because BatchNorm's own per-channel shift makes a conv bias redundant. Write solve(seq): count the learnable parameters of the given module — sum `p.numel()` over `seq.parameters()` — and return the total as an int.


<details><summary>💡 Hint (click to reveal)</summary>

Sum `p.numel()` over `.parameters()`; BN(16) adds 32 learnables (weight+bias).

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(seq):
    return sum(p.numel() for p in seq.parameters())

seq = nn.Sequential(nn.Conv2d(3, 16, 3, bias=False), nn.BatchNorm2d(16))
print(solve(seq))
```

## Why this works

The conv contributes out·in·kH·kW weights (no bias here); BatchNorm2d(16) adds two learnable length-16 vectors (weight γ and bias β) = 32. Running mean/var are buffers, not parameters, so they don't appear in the count.
