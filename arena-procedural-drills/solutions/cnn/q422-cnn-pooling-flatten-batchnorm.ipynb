## Problem

Modules record train/eval state on a boolean flag. Write solve(module, train): put the module in train mode when `train` is True, eval mode when False (the two methods `.train()` / `.eval()`), then return the value of the module attribute that records the mode.


<details><summary>💡 Hint (click to reveal)</summary>

`.train()` / `.eval()` flip the boolean `.training` attribute.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(module, train):
    if train:
        module.train()
    else:
        module.eval()
    return module.training

bn = nn.BatchNorm2d(8)
print(solve(bn, False))
```

## Why this works

`module.training` is the flag both methods flip: `.train()` sets True, `.eval()` sets False. BatchNorm reads it to pick batch vs running statistics; Dropout reads it to drop or pass through — same flag, different behaviors.
