## Problem

Modules record train/eval state on a boolean flag. The starter puts a BatchNorm2d into eval mode — in eval mode it normalizes with running statistics instead of batch statistics. Print the value of the module attribute that records this mode.


<details><summary>💡 Hint (click to reveal)</summary>

Switch the module to eval mode and read the boolean training attribute.

## Reference solution


```python
import torch.nn as nn

def solve():
    bn = nn.BatchNorm2d(8)
    bn.eval()
    return bn.training

print(solve())
```

## Why this works

Calling .eval() sets the module's training flag to False, which makes BatchNorm use its stored running_mean/running_var instead of batch statistics. So bn.training reads False after the toggle.
