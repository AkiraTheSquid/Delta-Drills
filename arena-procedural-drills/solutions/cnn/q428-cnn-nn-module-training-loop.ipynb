## Problem

Distinguish .train() vs .eval() modes on a Module. Build nn.Sequential(nn.Linear(4,2), nn.Dropout(0.5)), put in eval mode, then print model.training.


<details><summary>💡 Hint (click to reveal)</summary>

model.eval() sets the training flag — just read it back.

</details>

## Reference solution


```python
import torch.nn as nn


def solve():
    model = nn.Sequential(nn.Linear(4, 2), nn.Dropout(0.5))
    model.eval()
    return model.training


print(solve())
```

## Why this works

A Module's .training flag is True in train mode and False in eval mode; .eval() flips it (and submodules) to False. Reading model.training after .eval() yields False.
