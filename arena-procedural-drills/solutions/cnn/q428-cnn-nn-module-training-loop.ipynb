## Problem

Dropout behaves differently in train vs eval mode, so getting the mode right matters — and `.eval()` on a container recurses into every child. Write solve(model): call `model.eval()` and return the tuple `(model.training, child_flags)` where `child_flags` is the list of `.training` for each direct child (iterate the container).


<details><summary>💡 Hint (click to reveal)</summary>

`.eval()` recurses — check `model.training` AND each child's flag.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(model):
    model.eval()
    return (model.training, [m.training for m in model])

model = nn.Sequential(nn.Linear(4, 2), nn.Dropout(0.5))
print(solve(model))
```

## Why this works

Mode-setting recurses through the tree: `model.eval()` flips the container and every child, including the Dropout that actually changes behavior. Checking child flags is how you debug 'my model behaves differently at test time' surprises.
