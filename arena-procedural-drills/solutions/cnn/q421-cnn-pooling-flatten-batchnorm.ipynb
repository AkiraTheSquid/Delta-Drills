## Problem

BatchNorm tracks its running statistics as BUFFERS, not parameters — so they get no gradients and don't appear in `.parameters()`. Using the starter's `bn = nn.BatchNorm2d(8)`, collect the NAMES of all registered buffers (the module has an API for iterating buffers by name) and print them as a Python list of strings.

Enumerate named_buffers and collect their keys; expect the two running stats plus a counter.

## Reference solution


```python
import torch.nn as nn

def solve():
    bn = nn.BatchNorm2d(8)
    return list(dict(bn.named_buffers()).keys())

print(solve())
```

## Why this works

BatchNorm tracks inference-time statistics as registered buffers (not parameters): running_mean, running_var, and num_batches_tracked. Buffers move with the module and are saved in state_dict but receive no gradient.
