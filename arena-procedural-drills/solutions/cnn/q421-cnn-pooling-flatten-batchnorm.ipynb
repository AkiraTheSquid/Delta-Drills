## Problem

BatchNorm tracks its running statistics as BUFFERS, not parameters — so they get no gradients and don't appear in `.parameters()`. Write solve(module): collect the NAMES of all registered buffers (the module has an API for iterating buffers by name) and return them as a Python list of strings.


<details><summary>💡 Hint (click to reveal)</summary>

`module.named_buffers()` yields (name, tensor) pairs.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(module):
    return [name for name, _ in module.named_buffers()]

bn = nn.BatchNorm2d(8)
print(solve(bn))
```

## Why this works

`named_buffers()` iterates registered buffers by name: BatchNorm shows `running_mean`, `running_var`, `num_batches_tracked`. Inside a container the names get the child prefix (`0.running_mean`) — same dotted scheme as `named_parameters()`.
