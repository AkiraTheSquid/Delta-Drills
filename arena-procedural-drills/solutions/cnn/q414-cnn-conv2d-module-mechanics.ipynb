## Problem

Apply a filter with the FUNCTIONAL conv API (no module). Write solve(x, k): convolve input `x` (shape (B, C, H, W)) with kernel `k` (shape (out, C, kH, kW)) using `torch.nn.functional.conv2d` and return the output tensor. Predict the output's spatial size first.

`F.conv2d(x, k)` — the functional API takes input and kernel directly.

## Reference solution


```python
import torch
import torch.nn.functional as F

def solve(x, k):
    return F.conv2d(x, k)

x = torch.ones(1, 1, 4, 4)
k = torch.ones(1, 1, 3, 3) / 9
print(solve(x, k))
```

## Why this works

`torch.nn.functional.conv2d` is the module-free conv: you supply the kernel tensor yourself, shaped (out_channels, in_channels, kH, kW). A 3×3 kernel with no padding shrinks each spatial side by 2; an all-ones/9 kernel is a mean filter, so a ones input maps to ones.
