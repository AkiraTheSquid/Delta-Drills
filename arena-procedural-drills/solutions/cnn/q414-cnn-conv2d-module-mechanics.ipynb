## Problem

Apply an averaging filter with the FUNCTIONAL conv API (no module). The starter defines a (1,1,4,4) ones input `x` and a 3×3 averaging kernel `k` (every entry 1/9). Convolve `x` with `k` using `torch.nn.functional.conv2d` and print the output tensor directly. Predict the output's spatial size first.


<details><summary>💡 Hint (click to reveal)</summary>

Call F.conv2d(x, k) directly; with all-ones input the averaging kernel returns ones at every valid position.

</details>

## Reference solution


```python
import torch
import torch.nn.functional as F

def solve():
    x = torch.ones(1, 1, 4, 4)
    k = torch.ones(1, 1, 3, 3)/9
    return F.conv2d(x, k).detach().numpy()

print(solve())
```

## Why this works

F.conv2d with no padding slides the 3x3 kernel over the 4x4 input giving a 2x2 output. Each window is all ones averaged by the 1/9 kernel, so every output value is 1.0, yielding a (1,1,2,2) tensor of ones.
