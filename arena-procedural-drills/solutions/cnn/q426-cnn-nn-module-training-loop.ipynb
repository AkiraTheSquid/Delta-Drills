## Problem

Evaluation code shouldn't build autograd graphs. Write solve(x): inside a `torch.inference_mode()` context, compute `y` = the sum of `x`, and return the tuple `(y.requires_grad, round(float(y), 4))`. Even if `x` requires grad, no graph gets built inside the context.


<details><summary>💡 Hint (click to reveal)</summary>

`with torch.inference_mode():` — outputs inside don't track grad.

</details>

## Reference solution


```python
import torch

def solve(x):
    with torch.inference_mode():
        y = x.sum()
    return (y.requires_grad, round(float(y), 4))

x = torch.ones(3)
print(solve(x))
```

## Why this works

`inference_mode` is the stronger sibling of `no_grad`: ops inside build no graph and their outputs can never re-enter autograd. Even summing a requires-grad input yields `requires_grad=False` — evaluation code should live in one of these contexts.
