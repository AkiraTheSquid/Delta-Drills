## Problem

Use torch.inference_mode() to disable autograd for evaluation. Inside the context, print whether the result of torch.ones(3).sum() requires grad.


<details><summary>💡 Hint (click to reveal)</summary>

Tensors created inside the inference_mode context never track gradients.

</details>

## Reference solution


```python
import torch


def solve():
    with torch.inference_mode():
        y = torch.ones(3).sum()
    return y.requires_grad


print(solve())
```

## Why this works

Inside torch.inference_mode() autograd is disabled, so any tensor produced there has requires_grad=False. The summed tensor therefore reports False.
