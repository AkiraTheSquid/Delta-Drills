## Problem

Subclass `nn.Module`: in `__init__`, call `super().__init__()` and register a single `nn.Linear(4, 2)` as an attribute, and give the class a `forward` that applies it. Instantiate it, predict how many learnable scalars the linear contributes (weight AND bias), then print the parameter count as an int.


<details><summary>💡 Hint (click to reveal)</summary>

Register the Linear inside __init__ and sum p.numel() over model.parameters(); count weight plus bias.

</details>

## Reference solution


```python
import torch.nn as nn

class M(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(4, 2)
    def forward(self, x):
        return self.fc(x)

def solve():
    model = M()
    return sum(p.numel() for p in model.parameters())

print(solve())
```

## Why this works

A submodule assigned in __init__ is auto-registered, so its parameters appear in model.parameters(). nn.Linear(4,2) has a 2x4 weight (8) plus a length-2 bias (2), totaling 10 learnable parameters.
