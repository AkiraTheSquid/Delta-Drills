## Problem

Subclass `nn.Module`. Write solve(in_features, out_features): define a class whose `__init__` calls `super().__init__()` and registers a single `nn.Linear(in_features, out_features)` as an attribute, with a `forward` that applies it. Return an INSTANCE. Predict how many learnable scalars the linear contributes (weight AND bias) — the grader counts them.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(in_features, out_features):
    class M(nn.Module):
        def __init__(self, i, o):
            super().__init__()
            self.fc = nn.Linear(i, o)
        def forward(self, x):
            return self.fc(x)
    return M(in_features, out_features)

model = solve(4, 2)
print(model)
```

## Why this works

Assigning an nn.Module to an attribute auto-registers it, so `.parameters()` finds the linear's weight (out·in) and bias (out) — Linear(4, 2) contributes 10. The `super().__init__()` call must come first: it builds the registries the assignment hooks into.
