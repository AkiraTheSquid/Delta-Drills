## Problem

Global average pooling collapses each channel's spatial map to a single number. Write solve(x): reduce a (B, C, H, W) feature map to shape (B, C) by averaging over the spatial dimensions — pick the right `dim` argument yourself. Return the pooled tensor.


<details><summary>💡 Hint (click to reveal)</summary>

`x.mean(dim=(2, 3))` — average away exactly the two spatial dims.

</details>

## Reference solution


```python
import torch

def solve(x):
    return x.mean(dim=(2, 3))

x = torch.zeros(1, 64, 7, 7)
print(solve(x))
```

## Why this works

Global average pooling reduces each (H, W) map to its mean: `x.mean(dim=(2,3))` takes (B, C, H, W) to (B, C). It's the standard bridge from conv features to a classifier head — no learned weights, no spatial assumptions.
