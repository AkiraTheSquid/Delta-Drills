## Problem

Global average pooling collapses each channel's spatial map to a single number. Reduce the starter's (1, 64, 7, 7) feature map `x` to shape (1, 64) by averaging over the spatial dimensions — pick the right `dim` argument yourself. Print the resulting shape as a plain tuple.

Reduce the two spatial axes with mean(dim=(2,3)) and inspect the remaining shape.

## Reference solution


```python
import torch

def solve():
    x = torch.zeros(1, 64, 7, 7)
    return tuple(x.mean(dim=(2,3)).shape)

print(solve())
```

## Why this works

Averaging over dims 2 and 3 collapses the entire spatial grid into a single value per channel, producing (1, 64). This global average pooling is the parameter-free head used before a classifier in many CNNs.
