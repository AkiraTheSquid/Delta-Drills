## Problem

Zero-pad a 1-D tensor by hand. Write solve(x, pad): allocate a buffer of length `len(x) + 2*pad` filled with 0 using `x.new_full` (which preserves x's dtype and device — that's why it beats `torch.zeros` here), then copy `x` into the middle with slice assignment. Return the padded tensor.


<details><summary>💡 Hint (click to reveal)</summary>

`x.new_full((n + 2*pad,), 0)` allocates; `out[pad:-pad] = x` fills the middle.

</details>

## Reference solution


```python
import torch

def solve(x, pad):
    out = x.new_full((x.shape[0] + 2 * pad,), 0)
    out[pad:out.shape[0] - pad] = x
    return out

x = torch.tensor([1, 2, 3, 4])
print(solve(x, 1))
```

## Why this works

`x.new_full` creates a tensor with x's dtype and device — `torch.zeros` would silently give float32 even for an int input. Slice assignment copies x into the middle; the third case checks the dtype survived, which is the whole reason new_full exists.
