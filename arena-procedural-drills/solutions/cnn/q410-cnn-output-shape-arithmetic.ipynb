## Problem

Pad a 1-D tensor x=tensor([1,2,3,4]) with one zero on each side using new_full to allocate the padded buffer then copy in. Print the result tensor.


<details><summary>💡 Hint (click to reveal)</summary>

Allocate a zero buffer two longer than x, then copy x into the middle slice.

</details>

## Reference solution


```python
import torch

def solve():
    x = torch.tensor([1, 2, 3, 4])
    out = x.new_full((x.shape[0] + 2,), 0)
    out[1:-1] = x
    return out.numpy()

print(solve())
```

## Why this works

new_full allocates a buffer of length len(x)+2 filled with zeros and matching x's dtype/device, then the interior slice out[1:-1] is overwritten with x. This leaves a single zero on each end, the manual equivalent of one-element zero padding.
