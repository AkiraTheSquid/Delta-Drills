## Problem

Write a function solve(c, k) that returns, as a plain int, the number of multiplications in ONE output position of a convolution with c input channels and a k x k kernel — the dot-product length c * k * k.


<details><summary>💡 Hint (click to reveal)</summary>

One output value is a dot product across every input channel of every kernel cell.

</details>


## Reference solution


```python
import numpy as np

def solve(c, k):
    return int(c * k * k)


print(solve(64, 3))
```

## Why this works

Each output position multiplies the kernel against a c x k x k input patch — c*k*k products summed. This dot-product length is why channel count dominates conv compute.
