## Problem

How many input multiply-accumulates per output position for a Conv2d with in_channels=C, kernel=K? Print the formula's value for C=64, K=3 (use C*K*K).


<details><summary>💡 Hint (click to reveal)</summary>

Count taps per output position as channels times the kernel area.

</details>


## Reference solution


```python
def solve():
    C, K = 64, 3
    return C * K * K

print(solve())
```

## Why this works

Each output pixel of a conv is a dot product over the receptive field: C input channels times K*K spatial taps, so C*K*K multiply-accumulates. For C=64, K=3 that is 64*9 = 576 (bias adds one more but is not counted here).
