## Problem

Same-padding rule for odd kernel: given kernel=5 and stride=1, what padding P keeps output side equal to input side? Use P = (K-1)//2. Print as int.


<details><summary>💡 Hint (click to reveal)</summary>

The padding that preserves size for an odd kernel is half of K-1.

</details>


```python
%pip install -q numpy torch --index-url https://download.pytorch.org/whl/cpu
```

## Reference solution


```python
def solve():
    K = 5
    return (K - 1)//2

print(solve())
```

## Why this works

For an odd kernel at stride 1, padding (K-1)//2 on each side exactly compensates the K-1 pixels the kernel would otherwise shrink, keeping output side equal to input side. For K=5 that is 2.
