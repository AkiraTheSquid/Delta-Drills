## Problem

Write a function solve(k) that takes an ODD kernel size k and returns the 'same' padding as a plain int — the padding that keeps output size equal to input size at stride 1: (k - 1) // 2.


<details><summary>💡 Hint (click to reveal)</summary>

The kernel loses k-1 pixels of extent; splitting that evenly between the two sides restores them.

</details>

## Reference solution


```python
import torch as t

def solve(k):
    return int((k - 1) // 2)


print(solve(5))
```

## Why this works

With stride 1 the output shrinks by k-1 unless padding adds it back; p=(k-1)/2 on each side does exactly that for odd k. Substituting into (h+2p-k)+1 gives h — 'same' padding, only cleanly defined when k is odd.
