## Problem

For feature extraction you freeze a pretrained backbone. Write solve(model): turn off `requires_grad` on EVERY parameter (iterate `model.parameters()` and call `requires_grad_(False)` on each), then return the tuple `(n_still_on, flags)` — the count of parameters still requiring grad (an int; iterate and count, don't guess) and the per-parameter flag list.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(model):
    for p in model.parameters():
        p.requires_grad_(False)
    flags = [p.requires_grad for p in model.parameters()]
    return (sum(1 for f in flags if f), flags)

model = nn.Linear(4, 2)
print(solve(model))
```

## Why this works

Freezing a backbone = flipping `requires_grad` off on every parameter; gradient computation then skips them entirely. Counting the survivors (0) programmatically instead of guessing is the habit — partial freezes make wrong guesses expensive.
