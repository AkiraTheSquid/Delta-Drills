## Problem

Freeze all parameters of a model for feature-extraction fine-tuning. After `for p in model.parameters(): p.requires_grad_(False)`, print the count of parameters still requiring grad.


<details><summary>💡 Hint (click to reveal)</summary>

After freezing every parameter, count how many still have requires_grad True.

</details>

## Reference solution


```python
import torch.nn as nn


def solve():
    model = nn.Linear(4, 2)
    for p in model.parameters():
        p.requires_grad_(False)
    return sum(1 for p in model.parameters() if p.requires_grad)


print(solve())
```

## Why this works

Calling requires_grad_(False) on every parameter freezes the whole model, so none of them require grad afterward. The count of still-trainable parameters is therefore 0.
