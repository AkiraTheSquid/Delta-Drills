## Problem

For feature extraction you freeze a pretrained backbone: the starter turns off `requires_grad` on every parameter of a small model. Count how many parameters still require grad — iterate `model.parameters()` yourself and count, don't guess — and print the count as an int.


<details><summary>💡 Hint (click to reveal)</summary>

After freezing every parameter, count how many still have requires_grad True.

## Reference solution


```python
import torch.nn as nn


def solve():
    model = nn.Linear(4, 2)
    for p in model.parameters():
        p.requires_grad_(False)
    return sum(1 for p in model.parameters() if p.requires_grad)


print(solve())
```

## Why this works

Calling requires_grad_(False) on every parameter freezes the whole model, so none of them require grad afterward. The count of still-trainable parameters is therefore 0.
