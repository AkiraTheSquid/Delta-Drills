## Problem

Conv layers immediately followed by BatchNorm are usually built with `bias=False`. Write solve(conv): check the module's `bias` attribute and return the plain bool `conv.bias is None`.


<details><summary>💡 Hint (click to reveal)</summary>

`conv.bias is None` — with bias=False the attribute is literally None.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(conv):
    return conv.bias is None

conv = nn.Conv2d(3, 16, kernel_size=3, bias=False)
print(solve(conv))
```

## Why this works

With `bias=False` PyTorch stores None in the bias slot, not a zero tensor — so `conv.bias is None` is the definitive check. The pattern matters because conv→BatchNorm pairs drop the bias: BN's own shift makes it redundant.
