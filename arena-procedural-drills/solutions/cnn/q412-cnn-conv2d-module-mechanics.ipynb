## Problem

Create nn.Conv2d(3, 16, kernel_size=3, bias=False) and confirm bias is None. Print 'bias is None: True' or False.

## Reference solution


```python
import torch.nn as nn

def solve():
    conv = nn.Conv2d(3, 16, kernel_size=3, bias=False)
    return f'bias is None: {conv.bias is None}'

print(solve())
```

## Why this works

Passing bias=False tells Conv2d not to register a bias parameter, so conv.bias is literally None rather than a zero tensor. The f-string interpolates the boolean check, which is True.
