## Problem

Max-pooling with kernel 2 and stride 2 halves each spatial side. Write solve(x, pool): apply the pooling module to `x` and return the output shape as a plain tuple — note which dims changed and which didn't.


<details><summary>💡 Hint (click to reveal)</summary>

`tuple(pool(x).shape)` — batch and channels pass through, spatial dims halve.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(x, pool):
    return tuple(pool(x).shape)

x = torch.zeros(1, 4, 8, 8)
pool = nn.MaxPool2d(kernel_size=2, stride=2)
print(solve(x, pool))
```

## Why this works

MaxPool2d(2, 2) tiles each spatial map with non-overlapping 2×2 windows: H and W halve (floor division for odd sizes), while batch and channel dims are untouched — pooling is purely spatial.
