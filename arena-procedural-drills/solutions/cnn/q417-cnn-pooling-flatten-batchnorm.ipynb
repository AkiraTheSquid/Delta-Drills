## Problem

Max-pooling with kernel 2 and stride 2 halves each spatial side. The starter defines `x` of shape (1, 4, 8, 8) and the pool module. Apply it and print the output shape as a plain tuple — note which dims changed and which didn't.

Pass the tensor through the pool and read the shape; a stride-2 2x2 window halves H and W.

## Reference solution


```python
import torch
import torch.nn as nn

def solve():
    x = torch.zeros(1, 4, 8, 8)
    pool = nn.MaxPool2d(kernel_size=2, stride=2)
    return tuple(pool(x).shape)

print(solve())
```

## Why this works

A 2x2 max pool with stride 2 halves each spatial dimension while leaving batch and channel dims untouched, so (1,4,8,8) becomes (1,4,4,4). Pooling reduces resolution without learnable parameters.
