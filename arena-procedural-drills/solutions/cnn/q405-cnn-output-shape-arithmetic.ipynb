## Problem

A convolution slides a $K \times K$ kernel with stride $S$ over an input padded by $P$ on each side. For an input of side $H$, the output side is $\lfloor (H + 2P - K)/S \rfloor + 1$.

The starter defines `H, K, S, P` for a 32×32 input, kernel 5, stride 1, no padding. Compute the output side with torch-free integer arithmetic and print it as an int.


<details><summary>💡 Hint (click to reveal)</summary>

Plug the four values straight into floor((H + 2P - K)/S + 1) using integer division for the floor.

</details>

## Reference solution


```python
def solve():
    H, K, S, P = 32, 5, 1, 0
    return (H + 2*P - K)//S + 1

print(solve())
```

## Why this works

The Conv2d spatial output formula floor((H + 2P - K)/S + 1) counts how many valid kernel placements fit along a dimension after padding. With H=32, K=5, S=1, P=0 there are 28 positions. Integer floor-division // implements the floor for stride steps.
