## Problem

A convolution slides a K×K kernel with stride S over an input padded by P on each side. For an input of side H, the output side is floor((H + 2P − K)/S) + 1. Write solve(H, K, S, P): compute the output side with plain integer arithmetic (floor division) and return it as an int.


<details><summary>💡 Hint (click to reveal)</summary>

floor((H + 2P − K)/S) + 1, with `//` for the floor.

</details>

## Reference solution


```python
import torch as t

def solve(H, K, S, P):
    return (H + 2 * P - K) // S + 1

print(solve(32, 5, 1, 0))
```

## Why this works

The conv output-size formula: each output position needs a full K-window, the padded extent is H+2P, and stride S skips positions — floor((H+2P−K)/S)+1. It's pure integer arithmetic; `//` is Python's floor division.
