## Problem

Write a function solve(x, w) that computes a VALID 1-D convolution (cross-correlation, no flip) via unfold+einsum: x has shape (b, c_in, l), w has shape (c_out, c_in, k); return shape (b, c_out, l-k+1) by sliding windows over x and contracting with the einsum 'bilk,oik->bol' (i = input channel, l = output position, k = within-window offset).


<details><summary>💡 Hint (click to reveal)</summary>

Unfold the length axis into overlapping windows first — then the conv is one einsum contracting input channels and window offsets.

</details>

## Reference solution


```python
import numpy as np

def solve(x, w):
    return np.einsum('bilk,oik->bol', np.lib.stride_tricks.sliding_window_view(x, w.shape[2], axis=2), w)


print(solve(np.arange(8.0).reshape(1, 1, 8), np.ones((1, 1, 3))))
```

## Why this works

sliding_window_view exposes shape (b, i, l_out, k); einsum 'bilk,oik->bol' contracts i and k against each output filter — convolution as matrix contraction, exactly what im2col-based implementations do under the hood.
