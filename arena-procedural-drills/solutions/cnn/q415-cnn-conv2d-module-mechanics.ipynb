## Why this works

Unfolded patches xp are (B, C_in, L_out, K) and the weight is (C_out, C_in, K). A conv sums over the shared input-channel i and kernel-tap k, keeping batch b, output-length l, and output-channel o, which is exactly the einsum 'bilk,oik->bol'.
