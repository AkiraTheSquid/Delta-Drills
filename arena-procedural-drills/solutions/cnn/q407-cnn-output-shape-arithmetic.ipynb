## Problem

Shape arithmetic is easy to get subtly wrong, so verify it empirically. The starter builds `x` of shape (1, 3, 32, 32) and `conv = nn.Conv2d(3, 8, kernel_size=3, stride=2, padding=1)`. Predict the output shape, then run the input through the conv and print the ACTUAL output shape as a plain tuple (wrap the shape in `tuple(...)` for the expected format).

## Reference solution


```python
import torch
import torch.nn as nn

def solve():
    x = torch.zeros(1, 3, 32, 32)
    conv = nn.Conv2d(3, 8, kernel_size=3, stride=2, padding=1)
    return tuple(conv(x).shape)

print(solve())
```

## Why this works

Running a real Conv2d confirms the arithmetic: out_channels becomes 8, and each spatial side follows floor((32 + 2 - 3)/2 + 1) = 16. The batch dim is unchanged, giving (1, 8, 16, 16).
