## Problem

Verify the output shape of a Conv2d empirically. Input shape (1, 3, 32, 32), kernel=3, stride=2, padding=1, out_channels=8. Print the output shape as tuple.


<details><summary>💡 Hint (click to reveal)</summary>

Build the actual nn.Conv2d, push a zero tensor through it, and read tuple(out.shape).

## Reference solution


```python
import torch
import torch.nn as nn

def solve():
    x = torch.zeros(1, 3, 32, 32)
    conv = nn.Conv2d(3, 8, kernel_size=3, stride=2, padding=1)
    return tuple(conv(x).shape)

print(solve())
```

## Why this works

Running a real Conv2d confirms the arithmetic: out_channels becomes 8, and each spatial side follows floor((32 + 2 - 3)/2 + 1) = 16. The batch dim is unchanged, giving (1, 8, 16, 16).
