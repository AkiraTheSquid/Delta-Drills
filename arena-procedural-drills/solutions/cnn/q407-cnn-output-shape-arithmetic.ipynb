## Problem

Shape arithmetic is easy to get subtly wrong, so verify it empirically. Write solve(x, conv): predict the output shape of running `x` through the conv module, then actually run it and return the shape as a plain tuple (wrap in `tuple(...)`).

## Reference solution


```python
import torch
import torch.nn as nn

def solve(x, conv):
    return tuple(conv(x).shape)

x = torch.zeros(1, 3, 32, 32)
conv = nn.Conv2d(3, 8, kernel_size=3, stride=2, padding=1)
print(solve(x, conv))
```

## Why this works

Predict with the formula, then verify by running the module: (32 + 2·1 − 3)//2 + 1 = 16, channels become out_channels=8, batch passes through. Wrapping `.shape` in `tuple(...)` converts torch.Size to a plain tuple.
