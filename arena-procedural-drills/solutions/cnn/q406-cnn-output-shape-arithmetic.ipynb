## Problem

Write a function solve(h, k, s, p) that computes a convolution's OUTPUT SIDE LENGTH from input side h, kernel size k, stride s, and padding p using the standard formula floor((h + 2p - k) / s) + 1, returned as a plain int.


<details><summary>💡 Hint (click to reveal)</summary>

Padded extent minus kernel gives the sliding room; divide by stride (floor) and add the starting position.

</details>

## Reference solution


```python
import numpy as np

def solve(h, k, s, p):
    return int((h + 2 * p - k) // s + 1)


print(solve(224, 7, 2, 3))
```

## Why this works

(h + 2p - k)//s + 1 counts kernel placements: the padded input spans h+2p, the kernel needs k, and every s pixels starts a new placement. Kernel 7/stride 2/padding 3 exactly halves 224 to 112 — the classic ResNet stem.
