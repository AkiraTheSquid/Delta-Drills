## Problem

Conv2d output size with stride and padding: input H=W=224, kernel=7, stride=2, padding=3. Compute output side using floor((H+2P-K)/S+1). Print as int.


<details><summary>💡 Hint (click to reveal)</summary>

Apply floor((H + 2P - K)/S + 1); the stride of 2 is what roughly halves the side length.

</details>

## Reference solution


```python
def solve():
    H, K, S, P = 224, 7, 2, 3
    return (H + 2*P - K)//S + 1

print(solve())
```

## Why this works

Same Conv2d output formula, now with stride 2 and padding 3. (224 + 6 - 7)//2 + 1 = 223//2 + 1 = 111 + 1 = 112. The stride in the denominator halves the spatial extent, which is the classic ResNet stem downsampling.
