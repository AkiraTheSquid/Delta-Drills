## Problem

Write a function solve(c_in, c_out, k, bias) that returns the LEARNABLE PARAMETER COUNT of a 2-D conv layer as a plain int: c_out filters of size c_in*k*k, plus c_out bias terms when bias is True.


<details><summary>💡 Hint (click to reveal)</summary>

Weights are one c_in x k x k block per output channel; bias adds one scalar per output channel.

</details>

## Reference solution


```python
import torch as t

def solve(c_in, c_out, k, bias):
    return int(c_out * (c_in * k * k) + (c_out if bias else 0))


print(solve(64, 128, 3, True))
```

## Why this works

The weight tensor has shape (c_out, c_in, k, k) — c_out * c_in * k^2 numbers — and bias contributes c_out more. Note what's absent: the input's height and width, because the same filters slide everywhere (weight sharing).
