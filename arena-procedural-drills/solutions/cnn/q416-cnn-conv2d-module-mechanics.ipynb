## Problem

How many learnable parameters does `nn.Conv2d(64, 128, kernel_size=3, padding=1)` have with the default `bias=True`? Work it out from the shapes of the layer's weight and bias (the starter defines `C_in, C_out, K`). Print the count as an int.


<details><summary>💡 Hint (click to reveal)</summary>

Multiply out-channels by per-filter weights (in_ch*K*K) and add one bias per out-channel.

</details>


## Reference solution


```python
def solve():
    C_in, C_out, K = 64, 128, 3
    return C_out * (C_in * K * K) + C_out

print(solve())
```

## Why this works

A Conv2d has one K*K*C_in weight tensor per output channel plus one bias per output channel: C_out*(C_in*K*K) + C_out. Padding does not change parameter count. For 64->128, K=3 that is 128*576 + 128 = 73856.
