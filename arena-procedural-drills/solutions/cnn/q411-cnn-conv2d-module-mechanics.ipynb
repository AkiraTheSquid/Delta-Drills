## Problem

The starter builds `conv = nn.Conv2d(3, 16, kernel_size=3)`. In what ORDER does PyTorch lay out a Conv2d weight's four dimensions? Predict it, then check yourself: print the weight's shape as a plain tuple.


<details><summary>💡 Hint (click to reveal)</summary>

Read conv.weight.shape and recall the order is out, in, then the kernel height and width.

## Reference solution


```python
import torch.nn as nn

def solve():
    conv = nn.Conv2d(3, 16, kernel_size=3)
    return tuple(conv.weight.shape)

print(solve())
```

## Why this works

Conv2d stores its weight as (out_channels, in_channels, kH, kW), so a conv from 3 to 16 channels with a 3x3 kernel has weight shape (16, 3, 3, 3). This layout lets one big tensor hold every output filter stacked along dim 0.
