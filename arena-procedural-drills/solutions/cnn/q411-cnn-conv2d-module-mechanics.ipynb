## Problem

In what ORDER does PyTorch lay out a Conv2d weight's four dimensions? Write solve(conv): predict it — (out_channels, in_channels, kH, kW) or something else? — then return the weight's shape as a plain tuple.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(conv):
    return tuple(conv.weight.shape)

conv = nn.Conv2d(3, 16, kernel_size=3)
print(solve(conv))
```

## Why this works

Conv2d weights are laid out (out_channels, in_channels, kH, kW): one kernel stack per output channel, each spanning every input channel. Kernel_size can be a tuple — (3, 7) lands as kH=3, kW=7.
