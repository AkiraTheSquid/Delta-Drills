## Problem

Between a conv stack and a linear head, activations of shape (B, C, H, W) must become (B, C·H·W). Apply `nn.Flatten()` (default `start_dim=1`) to the starter's `x` of shape (8, 16, 4, 4) and print the flattened shape as a plain tuple. Why does the default leave dim 0 alone?


<details><summary>💡 Hint (click to reveal)</summary>

Flatten everything except the batch dimension; multiply the non-batch dims together.

## Reference solution


```python
import torch
import torch.nn as nn

def solve():
    x = torch.zeros(8, 16, 4, 4)
    return tuple(nn.Flatten()(x).shape)

print(solve())
```

## Why this works

nn.Flatten() defaults to start_dim=1, so it keeps the batch dim and merges channels and spatial dims: 16*4*4 = 256. The result (8, 256) is ready to feed a linear classification head.
