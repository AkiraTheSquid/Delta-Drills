## Problem

Between a conv stack and a linear head, activations of shape (B, C, H, W) must become (B, C·H·W). Write solve(x): apply `nn.Flatten()` (default `start_dim=1`) and return the flattened tensor. Why does the default leave dim 0 alone?


<details><summary>💡 Hint (click to reveal)</summary>

`nn.Flatten()(x)` — default start_dim=1 keeps the batch dim.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(x):
    return nn.Flatten()(x)

x = torch.zeros(8, 16, 4, 4)
out = solve(x)
print(out)
```

## Why this works

`nn.Flatten()` defaults to start_dim=1: it collapses everything AFTER the batch dim, (B, C, H, W) → (B, C·H·W). Flattening dim 0 too would merge samples — the default exists precisely so the batch survives into the linear head.
