## Problem

Write a function solve(w, x, y, lr) implementing ONE full training step for the scalar model pred = w * x with squared-error loss (pred - y)^2, entirely by hand: compute the loss's gradient dL/dw = 2 * (w*x - y) * x, take the gradient-descent step, and return the updated w as a float. This is zero_grad/forward/backward/step compressed to arithmetic.


<details><summary>💡 Hint (click to reveal)</summary>

Chain rule on (wx - y)^2: derivative 2(wx - y) times the inner derivative x; then step downhill.

</details>

## Reference solution


```python
import torch as t

def solve(w, x, y, lr):
    return float(w - lr * 2 * (w * x - y) * x)


print(solve(0.0, 2.0, 4.0, 0.1))
```

## Why this works

dL/dw = 2(wx - y)x by the chain rule, and w <- w - lr * grad is the whole optimizer. At the optimum (case 2) the gradient vanishes and w is unchanged — the fixed point every training loop is trying to reach.
