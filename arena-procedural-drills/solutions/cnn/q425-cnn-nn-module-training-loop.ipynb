## Problem

From memory, write the canonical training step as a single printed string: zero the grads, compute the loss from the forward pass, backpropagate, step the optimizer. Use exactly the four calls `opt.zero_grad()`, `loss = loss_fn(model(x), y)`, `loss.backward()`, `opt.step()`, joined by `; ` in the right ORDER — the order is the drill.


<details><summary>💡 Hint (click to reveal)</summary>

Order is zero_grad, forward+loss, backward, step — emit them as one semicolon-joined string.

</details>

## Reference solution


```python
def solve():
    return 'opt.zero_grad(); loss = loss_fn(model(x), y); loss.backward(); opt.step()'


print(solve())
```

## Why this works

The canonical PyTorch training step is exactly four ordered calls: clear stale gradients, compute the forward loss, backpropagate, then step the optimizer. The string reproduces that order with semicolon separators.
