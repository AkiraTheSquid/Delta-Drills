## Problem

Autograd has a GLOBAL on/off switch, and it beats per-tensor settings. Flip gradient tracking off globally (the subtopic names the function), then create the starter's `x` (requires_grad=True) and compute `y`, the sum of the elementwise square of `x`. No graph gets built despite `requires_grad=True`. Print two lines: print(y.requires_grad), then print(round(float(y), 4)).


<details><summary>💡 Hint (click to reveal)</summary>

The global flag wins over a tensor's requires_grad for graph construction, but it does not change the computed numbers.

</details>

## Reference solution


```python
import torch
torch.set_grad_enabled(False)
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
y = (x * x).sum()
print(y.requires_grad)
print(round(float(y), 4))
```

## Why this works

torch.set_grad_enabled(False) flips the global grad-mode flag, which overrides per-tensor requires_grad during forward computation: no autograd graph is built, so y.requires_grad is False. The numeric value is unaffected: 1+4+9 = 14.0.
