## Problem

Autograd has a GLOBAL on/off switch, and it beats per-tensor settings. Write solve(x): record `prev = torch.is_grad_enabled()`, flip tracking off with `torch.set_grad_enabled(False)`, compute `y` = the sum of the elementwise square of `x` (no graph gets built despite `requires_grad=True`), then RESTORE the flag to `prev` before returning — global state must not leak. Return `(y.requires_grad, round(float(y), 4))`.


<details><summary>💡 Hint (click to reveal)</summary>

`set_grad_enabled(False)` beats per-tensor flags; restore `prev` before returning.

## Reference solution


```python
import torch

def solve(x):
    prev = torch.is_grad_enabled()
    torch.set_grad_enabled(False)
    y = (x * x).sum()
    torch.set_grad_enabled(prev)
    return (y.requires_grad, round(float(y), 4))

x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
print(solve(x))
```

## Why this works

The global switch overrides `requires_grad=True` on the input — ops run while it's off produce untracked outputs. Restoring the recorded `prev` (rather than blindly setting True) is the discipline: library code must not leak global state, which is why `torch.no_grad()` is a context manager.
