## Problem

Write a function solve(x, grad_out) implementing the SIGMOID BACKWARD PASS by hand: forward is y = 1/(1+exp(-x)) elementwise; given the upstream gradient grad_out (dL/dy), return the downstream gradient dL/dx = grad_out * y * (1 - y).


<details><summary>💡 Hint (click to reveal)</summary>

Sigmoid's derivative is expressible from its OUTPUT alone — y(1-y) — then multiply by whatever gradient arrived from above.

</details>

## Reference solution


```python
import numpy as np

def solve(x, grad_out):
    return (lambda y: grad_out * y * (1 - y))(1.0 / (1.0 + np.exp(-x)))


print(solve(np.array([0.0]), np.array([1.0])))
```

## Why this works

dy/dx = y(1-y) for y = sigmoid(x), so the chain rule gives dL/dx = grad_out * y * (1-y). At x=0 the local slope peaks at 0.25 (times grad_out 4 gives exactly 1), and at saturated inputs it vanishes — the vanishing-gradient problem in one test case.
