## Problem

Implement a sigmoid backward pass by hand in numpy — no autograd. Forward: $y = \sigma(x) = 1/(1+e^{-x})$ for the starter's `x`. For the backward pass you need sigmoid's local derivative $dy/dx$ — express it in terms of $y$ itself (a classic identity worth memorizing), then apply the chain rule to the starter's incoming gradient `grad_out` to get `dx`. Print with print([round(float(v), 4) for v in dx]).


<details><summary>💡 Hint (click to reveal)</summary>

Use the sigmoid's own output to form its derivative y*(1-y), then multiply by the upstream gradient.

## Reference solution


```python
import numpy as np

x = np.array([0.0, 1.0, -1.0])
y = 1.0 / (1.0 + np.exp(-x))
grad_out = np.array([2.0, 0.5, 1.0])
dx = grad_out * (y * (1.0 - y))
print([round(float(v), 4) for v in dx])
```

## Why this works

The local derivative of the sigmoid is y*(1-y) where y is its output, so backprop multiplies the incoming gradient by that Jacobian elementwise. Applying grad_out * y*(1-y) yields the input-side gradient.
