## Problem

Using PyTorch autograd, let X = [[1,2,3],[4,5,6]] and Y = [[1,0],[0,1],[1,1]] both with requires_grad=True. Compute M = X @ Y, then backpropagate the upstream gradient dL/dM = [[1,2],[3,4]] by calling M.backward(dLdM). Print X.grad.tolist() on the first line and Y.grad.tolist() on the second line (each is a Python nested list of floats).

## Reference solution


```python
import torch
X = torch.tensor([[1., 2., 3.], [4., 5., 6.]], requires_grad=True)
Y = torch.tensor([[1., 0.], [0., 1.], [1., 1.]], requires_grad=True)
M = X @ Y
dLdM = torch.tensor([[1., 2.], [3., 4.]])
M.backward(dLdM)
print(X.grad.tolist())
print(Y.grad.tolist())
```

## Why this works

For M = X @ Y, autograd applies the matmul backward rules dL/dX = dL/dM @ Y^T and dL/dY = X^T @ dL/dM. Passing dLdM as the upstream gradient to M.backward seeds the vector-Jacobian product, and the .grad fields hold exactly those two products.
