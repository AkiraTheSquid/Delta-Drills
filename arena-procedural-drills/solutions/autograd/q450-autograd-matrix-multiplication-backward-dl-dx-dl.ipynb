## Problem

Autograd through a matmul. Write solve(X, Y, dLdM) where `X` (n×k) and `Y` (k×m) require grad and `dLdM` (n×m) is the upstream gradient: compute the product `M = X @ Y`, then backpropagate by passing the upstream gradient to `M.backward(...)`. Predict the SHAPES of `X.grad` and `Y.grad` before running. Return the tuple `(X.grad, Y.grad)`.


<details><summary>💡 Hint (click to reveal)</summary>

`M.backward(dLdM)`; X.grad has X's shape (dLdM @ Yᵀ), Y.grad has Y's (Xᵀ @ dLdM).

</details>

## Reference solution


```python
import torch

def solve(X, Y, dLdM):
    M = X @ Y
    M.backward(dLdM)
    return (X.grad, Y.grad)

X = torch.tensor([[1., 2., 3.], [4., 5., 6.]], requires_grad=True)
Y = torch.tensor([[1., 0.], [0., 1.], [1., 1.]], requires_grad=True)
dLdM = torch.tensor([[1., 2.], [3., 4.]])
print(solve(X, Y, dLdM))
```

## Why this works

Backprop through a matmul produces gradients with the SAME shapes as the inputs: X.grad = dLdM·Yᵀ is (n×k), Y.grad = Xᵀ·dLdM is (k×m). Passing dLdM to `M.backward(...)` supplies the upstream gradient the graph needs since M isn't a scalar.
