## Problem

Autograd through a matmul. The starter's `X` (2×3) and `Y` (3×2) require grad, and an upstream gradient `dLdM` (2×2) is given. Compute the product `M`, then backpropagate the upstream gradient through it by passing it to `M.backward(...)`. Predict the SHAPES of `X.grad` and `Y.grad` before running. Print X.grad.tolist() on the first line and Y.grad.tolist() on the second.

## Reference solution


```python
import torch
X = torch.tensor([[1., 2., 3.], [4., 5., 6.]], requires_grad=True)
Y = torch.tensor([[1., 0.], [0., 1.], [1., 1.]], requires_grad=True)
M = X @ Y
dLdM = torch.tensor([[1., 2.], [3., 4.]])
M.backward(dLdM)
print(X.grad.tolist())
print(Y.grad.tolist())
```

## Why this works

For M = X @ Y, autograd applies the matmul backward rules dL/dX = dL/dM @ Y^T and dL/dY = X^T @ dL/dM. Passing dLdM as the upstream gradient to M.backward seeds the vector-Jacobian product, and the .grad fields hold exactly those two products.
