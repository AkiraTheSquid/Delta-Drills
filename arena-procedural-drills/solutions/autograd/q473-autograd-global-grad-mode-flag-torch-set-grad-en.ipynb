## Problem

How do optimizers edit parameters that require grad without autograd objecting to in-place ops on a leaf? They work inside `torch.no_grad()`. Within the starter's `no_grad` block, take an in-place SGD-style step on `w`: subtract 0.1 times the all-ones length-3 vector, in place. Print [round(v, 4) for v in w.tolist()], then print(w.requires_grad) — the attribute survives the block.

Wrap the in-place update in the context manager that turns off tracking, then check that the leaf's requires_grad survives outside it.

## Reference solution


```python
import torch
w = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
with torch.no_grad():
    w -= 0.1 * torch.tensor([1.0, 1.0, 1.0])
print([round(v, 4) for v in w.tolist()])
print(w.requires_grad)
```

## Why this works

torch.no_grad() locally disables grad tracking, which is what permits an in-place edit of a leaf that requires grad without raising. The values shift down by 0.1 each, and after the block the leaf retains its requires_grad=True attribute since no_grad only suppressed tracking inside it.
