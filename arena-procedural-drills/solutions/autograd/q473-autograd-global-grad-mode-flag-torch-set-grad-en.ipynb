## Problem

How do optimizers edit parameters that require grad without autograd objecting to in-place ops on a leaf? They work inside `torch.no_grad()`. Write solve(w, lr, step): within a `no_grad` block, subtract `lr * step` from `w` IN PLACE. Return the tuple `(w, w.requires_grad)` — the grader checks the ORIGINAL tensor object was updated and still requires grad.


<details><summary>💡 Hint (click to reveal)</summary>

The in-place update needs `torch.no_grad()`; afterwards w still requires grad.

</details>

## Reference solution


```python
import torch

def solve(w, lr, step):
    with torch.no_grad():
        w -= lr * step
    return (w, w.requires_grad)

w = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
print(solve(w, 0.1, torch.ones(3)))
```

## Why this works

In-place edits on a requires-grad leaf inside the graph raise a RuntimeError — autograd couldn't rewind them. `torch.no_grad()` suspends recording so the optimizer step is legal; the tensor object is mutated in place (same identity) and keeps requires_grad=True for the next forward pass.
