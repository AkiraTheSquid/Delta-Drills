## Problem

Verify the matmul backward rules against autograd. Write solve(X, Y) where both require grad: compute `M = X @ Y`, build the scalar loss L = ½·Σ M² — chosen precisely so that dL/dM = M — and backprop it. Then compute the MANUAL gradients from the matmul backward rules (detach them) and compare. Return `(round(float((X.grad - gX).abs().max()), 4), round(float((Y.grad - gY).abs().max()), 4), X.grad)` — the two max-abs-differences should be 0.


<details><summary>💡 Hint (click to reveal)</summary>

L = ½ΣM² makes dL/dM = M; manual gX = M @ Yᵀ, gY = Xᵀ @ M (detached).

</details>

## Reference solution


```python
import torch

def solve(X, Y):
    M = X @ Y
    L = 0.5 * (M * M).sum()
    L.backward()
    gX = (M @ Y.t()).detach()
    gY = (X.t() @ M).detach()
    return (round(float((X.grad - gX).abs().max()), 4), round(float((Y.grad - gY).abs().max()), 4), X.grad)

X = torch.tensor([[1., 2.], [3., 4.], [5., 6.]], requires_grad=True)
Y = torch.tensor([[2., 1., 0., 1.], [1., 0., 2., 3.]], requires_grad=True)
print(solve(X, Y))
```

## Why this works

The loss ½ΣM² is chosen so dL/dM = M exactly — plugging M into the backward rules gives closed-form gradients you can compare with autograd's. Max-abs-difference 0.0 on both is the verification; detaching the manual computation keeps it out of the graph.
