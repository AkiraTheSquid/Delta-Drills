## Problem

Verify the matmul backward rules against autograd. Let X = [[1,2],[3,4],[5,6]] and Y = [[2,1,0,1],[1,0,2,3]] with requires_grad=True, and M = X @ Y. Define scalar loss L = 0.5 * sum(M*M), so dL/dM = M. Call L.backward(), then compute the manual gradients gX = M @ Y^T and gY = X^T @ M (detached). Print three lines: round(float(max abs difference between X.grad and gX), 4), then the same for Y.grad and gY, then X.grad.tolist().

The derivative of 0.5*sum(M*M) w.r.t. M is just M; plug that in as dL/dM and compare the manual products against autograd.

## Reference solution


```python
import torch
X = torch.tensor([[1., 2.], [3., 4.], [5., 6.]], requires_grad=True)
Y = torch.tensor([[2., 1., 0., 1.], [1., 0., 2., 3.]], requires_grad=True)
M = X @ Y
L = 0.5 * (M * M).sum()
L.backward()
gX = (M @ Y.t()).detach()
gY = (X.t() @ M).detach()
print(round(float((X.grad - gX).abs().max()), 4))
print(round(float((Y.grad - gY).abs().max()), 4))
print(X.grad.tolist())
```

## Why this works

With L = 0.5*sum(M*M) the upstream gradient is dL/dM = M itself, so the matmul rules give gX = M @ Y^T and gY = X^T @ M. These match autograd's X.grad and Y.grad exactly, so both max-abs differences are 0.
