## Problem

Verify the matmul backward rules against autograd. The starter has `X`, `Y` (requires grad) and `M = X @ Y` is the product. Build the scalar loss $L = \tfrac{1}{2}\sum M^2$ — chosen precisely so that $\partial L/\partial M = M$ — and backprop it. Then compute the MANUAL gradients from the matmul backward rules (detach them) and compare. Print three lines: the max abs difference between `X.grad` and manual (rounded to 4), the same for `Y`, then X.grad.tolist().


<details><summary>💡 Hint (click to reveal)</summary>

The derivative of 0.5*sum(M*M) w.r.t. M is just M; plug that in as dL/dM and compare the manual products against autograd.

## Reference solution


```python
import torch
X = torch.tensor([[1., 2.], [3., 4.], [5., 6.]], requires_grad=True)
Y = torch.tensor([[2., 1., 0., 1.], [1., 0., 2., 3.]], requires_grad=True)
M = X @ Y
L = 0.5 * (M * M).sum()
L.backward()
gX = (M @ Y.t()).detach()
gY = (X.t() @ M).detach()
print(round(float((X.grad - gX).abs().max()), 4))
print(round(float((Y.grad - gY).abs().max()), 4))
print(X.grad.tolist())
```

## Why this works

With L = 0.5*sum(M*M) the upstream gradient is dL/dM = M itself, so the matmul rules give gX = M @ Y^T and gY = X^T @ M. These match autograd's X.grad and Y.grad exactly, so both max-abs differences are 0.
