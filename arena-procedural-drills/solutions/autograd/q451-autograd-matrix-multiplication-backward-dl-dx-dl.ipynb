## Problem

Apply the matmul backward rules directly — no autograd. For M = XY: dL/dX = (dL/dM) Yᵀ and dL/dY = Xᵀ (dL/dM). Write solve(X, Y, dLdM): translate both rules to torch code and return the tuple `(dLdX, dLdY)`.


<details><summary>💡 Hint (click to reveal)</summary>

Straight translation: `dLdM @ Y.t()` and `X.t() @ dLdM`.

</details>

## Reference solution


```python
import torch

def solve(X, Y, dLdM):
    dLdX = dLdM @ Y.t()
    dLdY = X.t() @ dLdM
    return (dLdX, dLdY)

X = torch.tensor([[2., 0., 1.], [1., 3., 2.]])
Y = torch.tensor([[1., 2.], [0., 1.], [3., 0.]])
dLdM = torch.tensor([[1., 1.], [2., 0.]])
print(solve(X, Y, dLdM))
```

## Why this works

The two matmul backward rules ARE the implementation: dL/dX = (dL/dM)Yᵀ and dL/dY = Xᵀ(dL/dM). Sanity-check with shapes: (n×m)(m×k) → n×k matches X; (k×n)(n×m) → k×m matches Y.
