## Problem

Without using autograd, apply the matmul backward rules directly. Given X = [[2,0,1],[1,3,2]], Y = [[1,2],[0,1],[3,0]], and an upstream gradient dL/dM = [[1,1],[2,0]] for M = X @ Y, compute dL/dX = dL/dM @ Y^T and dL/dY = X^T @ dL/dM using torch tensors. Print dLdX.tolist() on the first line and dLdY.tolist() on the second line.


<details><summary>💡 Hint (click to reveal)</summary>

Post-multiply the upstream gradient by Y transposed for dL/dX; pre-multiply by X transposed for dL/dY.

</details>

## Reference solution


```python
import torch
X = torch.tensor([[2., 0., 1.], [1., 3., 2.]])
Y = torch.tensor([[1., 2.], [0., 1.], [3., 0.]])
dLdM = torch.tensor([[1., 1.], [2., 0.]])
dLdX = dLdM @ Y.t()
dLdY = X.t() @ dLdM
print(dLdX.tolist())
print(dLdY.tolist())
```

## Why this works

The matmul backward rules come straight from the chain rule on M = X @ Y: dL/dX = dL/dM @ Y^T and dL/dY = X^T @ dL/dM. Applying them with transposes gives the same result autograd would, without building a graph.
