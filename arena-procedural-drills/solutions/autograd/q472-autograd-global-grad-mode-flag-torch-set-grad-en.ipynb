## Problem

The global grad flag can be flipped back and forth mid-script. Write solve(x): record `prev = torch.is_grad_enabled()` before anything changes; flip the flag OFF and compute `a` — the sum of `x + 1`; flip it back ON and compute `b` the same way. Predict all three booleans. Return `[a.requires_grad, b.requires_grad, prev]`.


<details><summary>💡 Hint (click to reveal)</summary>

Record `is_grad_enabled()` first; a computed with the flag off never tracks.

## Reference solution


```python
import torch

def solve(x):
    prev = torch.is_grad_enabled()
    torch.set_grad_enabled(False)
    a = (x + 1).sum()
    torch.set_grad_enabled(True)
    b = (x + 1).sum()
    return [a.requires_grad, b.requires_grad, prev]

x = torch.tensor([2.0, 4.0], requires_grad=True)
print(solve(x))
```

## Why this works

Tracking is decided AT OP TIME by the global flag: `a` (computed while off) has requires_grad=False forever; `b` (computed after re-enabling) tracks again — for the zeros input without requires_grad, nothing ever tracks. `prev` records the default state: True.
