## Problem

Using torch, create x = torch.tensor([2.0, 4.0], requires_grad=True). Record prev = torch.is_grad_enabled() (its default value before you change anything). Then set the global grad flag to False and compute a = (x + 1).sum(). Then set the global grad flag back to True and compute b = (x + 1).sum(). Print exactly one list: print([a.requires_grad, b.requires_grad, prev]).

## Reference solution


```python
import torch
x = torch.tensor([2.0, 4.0], requires_grad=True)
prev = torch.is_grad_enabled()
torch.set_grad_enabled(False)
a = (x + 1).sum()
torch.set_grad_enabled(True)
b = (x + 1).sum()
print([a.requires_grad, b.requires_grad, prev])
```

## Why this works

Grad tracking defaults to enabled, so prev is True. Computing a while the flag is off yields requires_grad=False; turning it back on makes b track gradients again (True). The flag is global state queried via torch.is_grad_enabled().
