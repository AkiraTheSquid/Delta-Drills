## Problem

Watch autograd do the unbroadcast for you. The starter defines `x` of shape (1,3) and `y` of shape (2,1), both requiring grad. Add them — broadcasting gives (2,3) — reduce the sum to a scalar, and backprop it. Predict both gradients first: how many broadcast copies does each original element fan out to? Print both with print(x.grad.tolist(), y.grad.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Each operand's gradient equals the number of times broadcasting replicated it along its size-1 axis.

## Reference solution


```python
import torch
x = torch.tensor([[1.0, 2.0, 3.0]], requires_grad=True)
y = torch.tensor([[1.0],[2.0]], requires_grad=True)
z = (x + y).sum()
z.backward()
print(x.grad.tolist(), y.grad.tolist())
```

## Why this works

x (1,3) and y (2,1) broadcast to (2,3); summing gives an upstream gradient of all ones at shape (2,3). Autograd unbroadcasts each operand: x is summed over its size-1 row axis (2 copies) giving 2s, and y over its size-1 column axis (3 copies) giving 3s.
