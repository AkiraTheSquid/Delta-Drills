## Problem

Watch autograd do the unbroadcast for you. Write solve(x, y) where `x` has shape (1, n), `y` has shape (m, 1), both requiring grad: add them — broadcasting gives (m, n) — reduce the sum to a scalar, and backprop. Predict both gradients first: how many broadcast copies does each original element fan out to? Return the tuple `(x.grad, y.grad)`.


<details><summary>💡 Hint (click to reveal)</summary>

`(x + y).sum().backward()`; each grad = number of broadcast copies.

## Reference solution


```python
import torch

def solve(x, y):
    z = (x + y).sum()
    z.backward()
    return (x.grad, y.grad)

x = torch.tensor([[1.0, 2.0, 3.0]], requires_grad=True)
y = torch.tensor([[1.0], [2.0]], requires_grad=True)
print(solve(x, y))
```

## Why this works

Autograd applies unbroadcast automatically: broadcasting (1, n) + (m, 1) to (m, n) means each x element was copied m times and each y element n times, so after backward x.grad is all-m and y.grad all-n — with each grad matching its ORIGINAL shape.
