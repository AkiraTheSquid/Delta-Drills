## Problem

Let x = torch.tensor([[1.,2.,3.]], requires_grad=True) (shape (1,3)) and y = torch.tensor([[1.],[2.]], requires_grad=True) (shape (2,1)). Compute z = (x + y).sum(), which broadcasts x and y to shape (2,3), then call z.backward(). PyTorch's autograd internally unbroadcasts each gradient back to the original operand shapes. Print both accumulated gradients on one line with print(x.grad.tolist(), y.grad.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Each operand's gradient equals the number of times broadcasting replicated it along its size-1 axis.

</details>

## Reference solution


```python
import torch
x = torch.tensor([[1.0, 2.0, 3.0]], requires_grad=True)
y = torch.tensor([[1.0],[2.0]], requires_grad=True)
z = (x + y).sum()
z.backward()
print(x.grad.tolist(), y.grad.tolist())
```

## Why this works

x (1,3) and y (2,1) broadcast to (2,3); summing gives an upstream gradient of all ones at shape (2,3). Autograd unbroadcasts each operand: x is summed over its size-1 row axis (2 copies) giving 2s, and y over its size-1 column axis (3 copies) giving 3s.
