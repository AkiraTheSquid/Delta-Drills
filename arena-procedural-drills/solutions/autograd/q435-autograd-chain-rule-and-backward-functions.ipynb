## Problem

Backprop through a cube. The starter's `x` requires grad. Compute `out` as x cubed elementwise, then backprop the starter's all-ones upstream gradient through it by passing that gradient to `out.backward(...)`. Before printing, predict `x.grad`: what is the derivative of $x^3$, evaluated at 2, −3, and 4? Print with print(x.grad.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Differentiate x cubed elementwise (3x^2) and scale by the incoming gradient.

## Reference solution


```python
import torch

x = torch.tensor([2.0, -3.0, 4.0], requires_grad=True)
out = x * x * x
grad_out = torch.tensor([1.0, 1.0, 1.0])
out.backward(grad_out)
print(x.grad.tolist())
```

## Why this works

For out = x^3 the elementwise derivative is 3*x^2, and passing grad_out into backward applies the chain rule grad_out * 3*x^2. With unit upstream gradients this gives 3*[4,9,16] = [12,27,48].
