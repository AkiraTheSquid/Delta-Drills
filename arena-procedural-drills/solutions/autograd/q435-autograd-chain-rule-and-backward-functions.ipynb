## Problem

Backprop through a cube. Write solve(x, grad_out) where `x` requires grad: compute `out` as x cubed elementwise, then backprop the given upstream gradient through it by passing it to `out.backward(...)`. Before running, predict `x.grad`: the derivative of x³ is 3x², scaled elementwise by `grad_out`. Return `x.grad`.


<details><summary>💡 Hint (click to reveal)</summary>

out = `x * x * x`; pass grad_out into `out.backward(grad_out)`; x.grad = 3x²·grad_out.

</details>

## Reference solution


```python
import torch

def solve(x, grad_out):
    out = x * x * x
    out.backward(grad_out)
    return x.grad

x = torch.tensor([2.0, -3.0, 4.0], requires_grad=True)
grad_out = torch.tensor([1.0, 1.0, 1.0])
print(solve(x, grad_out))
```

## Why this works

`out.backward(gradient)` is how you backprop from a NON-scalar: the tensor you pass is the upstream dL/d(out). The chain rule gives x.grad = 3x² × grad_out elementwise — evaluate at each x to check your prediction.
