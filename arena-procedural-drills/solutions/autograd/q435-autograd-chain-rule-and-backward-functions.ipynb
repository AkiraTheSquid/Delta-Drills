## Problem

Using PyTorch autograd, create x = torch.tensor([2.0, -3.0, 4.0], requires_grad=True). Compute out = x*x*x (i.e. x cubed elementwise), then call out.backward with grad_out = torch.tensor([1.0, 1.0, 1.0]). The chain rule gives dL/dx = grad_out * d(out)/dx where d(out)/dx = 3*x^2. Print x.grad as a Python list using print(x.grad.tolist()).

## Reference solution


```python
import torch

x = torch.tensor([2.0, -3.0, 4.0], requires_grad=True)
out = x * x * x
grad_out = torch.tensor([1.0, 1.0, 1.0])
out.backward(grad_out)
print(x.grad.tolist())
```

## Why this works

For out = x^3 the elementwise derivative is 3*x^2, and passing grad_out into backward applies the chain rule grad_out * 3*x^2. With unit upstream gradients this gives 3*[4,9,16] = [12,27,48].
