## Problem

Manually backpropagate through a two-stage chain (linear then ReLU) using numpy. Let W = np.array([[1.0, -2.0], [0.5, 3.0]]) and x = np.array([2.0, -1.0]). Forward: z = W @ x, then out = relu(z). Given grad_out = np.array([1.0, 1.0]) flowing into out, apply the chain rule stage by stage: first multiply by ReLU's local Jacobian (1 where z>0 else 0) to get grad_z, then by the linear layer's local Jacobian (W transpose) to get grad_x = W.T @ grad_z. Print grad_x as a list of values each rounded to 4 decimals, using print([round(float(v), 4) for v in grad_x]).

Gate the upstream gradient by where z>0, then pull it back through the linear layer with W transposed.

## Reference solution


```python
import numpy as np

W = np.array([[1.0, -2.0], [0.5, 3.0]])
x = np.array([2.0, -1.0])
z = W @ x
relu_mask = (z > 0).astype(float)
grad_out = np.array([1.0, 1.0])
grad_z = grad_out * relu_mask
grad_x = W.T @ grad_z
print([round(float(v), 4) for v in grad_x])
```

## Why this works

ReLU's local Jacobian is a 0/1 gate on whether each pre-activation was positive, so grad_z masks grad_out where z<=0. The linear layer backprops by multiplying with W transpose, giving grad_x = W.T @ grad_z.
