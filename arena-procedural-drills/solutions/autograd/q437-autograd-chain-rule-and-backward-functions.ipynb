## Problem

Write a function solve(w, x, grad_out) backpropagating BY HAND through z = w @ x followed by a = relu(z): given upstream gradient grad_out (dL/da), return the tuple (grad_w, grad_x) where grad_z = grad_out * (z > 0), grad_w = outer(grad_z, x), and grad_x = w.T @ grad_z.


<details><summary>💡 Hint (click to reveal)</summary>

ReLU's gradient gates grad_out by which pre-activations were positive; then the linear layer's two gradients are an outer product (for W) and a transposed matvec (for x).

</details>

## Reference solution


```python
import torch as t

def solve(w, x, grad_out):
    return (lambda gz: (t.outer(gz, x), w.T @ gz))(grad_out * ((w @ x) > 0))


print(solve(t.eye(2), t.tensor([1.0, -1.0]), t.tensor([1.0, 1.0])))
```

## Why this works

grad_z = grad_out * (z > 0) kills gradient where ReLU was dead; dL/dW = grad_z outer x (each weight's gradient is its output-row gradient times its input), and dL/dx = W.T @ grad_z routes gradient back through the weights. The all-negative case shows a completely dead unit: both gradients zero.
