## Problem

Backprop by hand through a two-stage chain: linear, then ReLU. The starter gives `W`, `x`, and the upstream gradient `grad_out`. Forward: $z = Wx$, then out = relu($z$). Now walk the chain rule backwards one stage at a time: what is ReLU's local Jacobian (a mask determined by $z$)? What does the linear stage contribute (which transpose of `W`, applied to which gradient)? Compute `grad_x` and print with print([round(float(v), 4) for v in grad_x]).


<details><summary>💡 Hint (click to reveal)</summary>

Gate the upstream gradient by where z>0, then pull it back through the linear layer with W transposed.

## Reference solution


```python
import numpy as np

W = np.array([[1.0, -2.0], [0.5, 3.0]])
x = np.array([2.0, -1.0])
z = W @ x
relu_mask = (z > 0).astype(float)
grad_out = np.array([1.0, 1.0])
grad_z = grad_out * relu_mask
grad_x = W.T @ grad_z
print([round(float(v), 4) for v in grad_x])
```

## Why this works

ReLU's local Jacobian is a 0/1 gate on whether each pre-activation was positive, so grad_z masks grad_out where z<=0. The linear layer backprops by multiplying with W transpose, giving grad_x = W.T @ grad_z.
