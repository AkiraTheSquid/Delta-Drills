## Problem

An elementwise op broadcast a tensor of original shape (3,1) up to (3,2) during the forward pass. The upstream gradient flowing back has shape (3,2) and is all ones (torch.ones(3,2)). Implement the 'unbroadcast' reduction that reverses broadcasting: first sum over any prepended leading dims, then sum (with keepdim) over each axis whose original size was 1. Apply it to reduce the (3,2) ones gradient back to the original (3,1) shape, and print the result with print(result.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Reverse broadcasting by summing the upstream gradient over exactly the axes that were size 1 in the original shape, keeping the dimension.

## Reference solution


```python
import torch
def unbroadcast(grad, shape):
    while grad.dim() > len(shape):
        grad = grad.sum(0)
    for i, s in enumerate(shape):
        if s == 1:
            grad = grad.sum(i, keepdim=True)
    return grad
a = torch.tensor([[1.0],[2.0],[3.0]])
out_grad = torch.ones(3, 2)
print(unbroadcast(out_grad, a.shape).tolist())
```

## Why this works

Broadcasting copies a size-1 axis across the broadcast dimension, so its gradient is the sum of all copies. Unbroadcast reverses this by first dropping any prepended leading dims via summation, then summing (keepdim) over each axis whose original size was 1, restoring the operand's shape.
