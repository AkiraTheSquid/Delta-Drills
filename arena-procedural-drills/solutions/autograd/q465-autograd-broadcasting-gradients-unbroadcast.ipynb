## Problem

When the forward pass broadcasts a (3, 1) tensor up to (3, 2), the backward pass must reduce the (3, 2) gradient back to (3, 1) — summing over every position the forward copied to. Write solve(grad, shape): first sum away any PREPENDED leading dims (while grad has more dims than the target shape, sum dim 0), then sum WITH `keepdim=True` over each axis whose original size was 1. Return the reduced gradient.


<details><summary>💡 Hint (click to reveal)</summary>

Sum dim 0 while too many dims; then `sum(i, keepdim=True)` per size-1 axis.

</details>

## Reference solution


```python
import torch

def solve(grad, shape):
    while grad.dim() > len(shape):
        grad = grad.sum(0)
    for i, s in enumerate(shape):
        if s == 1:
            grad = grad.sum(i, keepdim=True)
    return grad

out_grad = torch.ones(3, 2)
print(solve(out_grad, (3, 1)))
```

## Why this works

Unbroadcast reverses the two broadcast mechanisms separately: prepended dims (grad has MORE dims than the original) get summed away, and stretched size-1 axes get summed WITH keepdim so the axis survives at size 1. Every forward copy contributes one gradient term — that's why it's a sum.
