## Problem

Write a function solve(z) that takes a 1-D boolean PyTorch tensor and returns the number of True entries as a plain Python int. The tensor may be all True, all False, or any mix, and may have any length >= 1.


<details><summary>💡 Hint (click to reveal)</summary>

True counts as nonzero (and as 1 in a sum) — just remember to convert the PyTorch result to a plain int.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return int(t.count_nonzero(z))


example = t.tensor([True, False, True, True])
print(solve(example))
```

## Why this works

t.count_nonzero counts True entries because True is the only nonzero boolean; z.sum() gives the same number since True adds 1. Both return PyTorch integer types, so int(...) is needed to satisfy the plain-Python-int requirement the tests check with isinstance.
