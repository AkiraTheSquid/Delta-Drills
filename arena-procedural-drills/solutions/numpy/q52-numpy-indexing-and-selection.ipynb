## Problem

You are given Z, a 1D boolean array of length 4 with a mix of True and False entries. Return the total number of True entries in Z as a single integer count. The answer must reflect how many positions hold True, ignoring False entries entirely. (★☆☆)
Print the result.

Booleans are 1/0 under the hood; count the nonzero entries.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.array([True, False, True, True])
    return np.count_nonzero(Z)


print(solve())
```

## Why this works

True counts as nonzero and False as zero, so np.count_nonzero on a boolean array returns the number of True entries as an integer.
