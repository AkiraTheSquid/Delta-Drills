## Problem

Write a function solve(n) that takes a positive integer n and returns an n x n numpy array of floats whose outer ring — the first and last row and the first and last column — is 1.0, and whose interior block is 0.0. Use slice assignment rather than Python loops. Note that for n = 1 or n = 2 every entry lies on the border, so the whole matrix should be 1.0.


<details><summary>💡 Hint (click to reveal)</summary>

Start from a uniform matrix and overwrite its interior block with one 2-D slice assignment — remember that negative slice bounds count from the end of an axis.

</details>

## Reference solution


```python
import numpy as np

def solve(n):
    z = np.ones((n, n))
    z[1:-1, 1:-1] = 0.0
    return z


print(solve(4))
```

## Why this works

np.ones((n, n)) builds the full float matrix, and the slice z[1:-1, 1:-1] selects every row and column except the first and last, so assigning 0.0 to it zeros exactly the interior block in one vectorized step. Because the slice 1:-1 is empty when n is 1 or 2, the assignment is a harmless no-op in those edge cases and the matrix stays all ones, which is correct since every entry is on the border.
