## Problem

You are given Z, a 10x10 float array whose outer ring (first row, last row, first column, last column) holds 1.0 and whose interior 8x8 block holds 0.0. Return Z as a framed pattern: a solid border of ones surrounding a hollow rectangle of zeros, with the original (10, 10) shape preserved. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Begin with all ones and set the interior to zero with a slice that drops the outermost rows and columns.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.ones((10, 10))
    Z[1:-1, 1:-1] = 0
    return Z

print(solve())
```

## Why this works

Start from an all-ones array, then zero out the interior using a slice that excludes the first and last index on each axis. The 1:-1 slice on both dimensions leaves a one-cell border of ones around a block of zeros.
