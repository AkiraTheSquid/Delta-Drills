## Problem

Build Z, a 1D structured array of length 10 whose dtype has two top-level fields: 'position', itself a nested structure with float fields 'x' and 'y'; and 'color', a nested structure with float fields 'r', 'g', and 'b'. All entries should be zero-initialized. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Express the dtype as a list of (name, sub-dtype) pairs where the sub-dtype is itself a list of named float fields.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros(10, [('position', [('x', float, 1),
                                    ('y', float, 1)]),
                      ('color', [('r', float, 1),
                                 ('g', float, 1),
                                 ('b', float, 1)])])
    return Z

print(solve())
```

## Why this works

A NumPy structured dtype can nest sub-dtypes: each top-level field ('position', 'color') is itself a list of named float subfields. Passing this nested dtype spec to np.zeros allocates a length-10 record array with every field zero-initialized.
