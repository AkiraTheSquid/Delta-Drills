## Problem

Write a function solve(n) that takes a positive integer n and returns a 1-D structured numpy array of length n, initialized to zeros, whose dtype has two nested fields: 'position' containing float subfields 'x' and 'y', and 'color' containing float subfields 'r', 'g', and 'b'.


<details><summary>💡 Hint (click to reveal)</summary>

Structured dtypes nest: a field's type can itself be a list of (name, type) pairs.

</details>

## Reference solution


```python
import numpy as np

def solve(n):
    return np.zeros(n, [('position', [('x', float), ('y', float)]),
                        ('color', [('r', float), ('g', float), ('b', float)])])


print(solve(2))
```

## Why this works

np.zeros(n, dtype) with a nested dtype list creates records whose 'position' and 'color' fields are themselves structures; access chains like z['position']['x'] then address the float leaves. Zero-initialization makes every leaf 0.0, which the sum check verifies.
