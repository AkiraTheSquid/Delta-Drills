## Problem

Write a function solve(z, fill) that takes a 2-D numpy array z and a scalar fill, and returns a new array with a one-element-wide border of fill surrounding z on all four sides. Shape (r, c) becomes (r+2, c+2), with the original values in the interior.


<details><summary>💡 Hint (click to reveal)</summary>

The padding function's constant mode takes the border value as an argument — it isn't limited to zeros.

</details>

## Reference solution


```python
import numpy as np

def solve(z, fill):
    return np.pad(z, pad_width=1, mode="constant", constant_values=fill)


example = np.ones((2, 2), dtype=int)
print(solve(example, -1))
```

## Why this works

np.pad with mode='constant' and constant_values=fill wraps the array in one layer of the chosen value on every side. Allocating np.full((r+2, c+2), fill) and assigning the interior slice out[1:-1, 1:-1] = z builds the same result.
