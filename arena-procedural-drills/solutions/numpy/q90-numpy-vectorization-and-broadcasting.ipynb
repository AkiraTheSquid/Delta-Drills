## Problem

Given a 2D array Z, create a new array that surrounds Z with a border of -1, where the border is one element wide. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a single NumPy function for adding a constant-valued border of a given width around an array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.ones((3, 3), dtype=int)
    return np.pad(Z, pad_width=1, mode='constant', constant_values=-1)

print(solve())
```

## Why this works

np.pad adds a frame of cells around the original array; pad_width=1 makes that frame exactly one element thick on every side. Using mode='constant' with constant_values=-1 fills the new border cells with -1 while leaving the interior untouched.
