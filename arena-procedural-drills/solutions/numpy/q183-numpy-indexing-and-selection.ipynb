## Problem

Write a function solve(z) that takes a 2-D float array with DISTINCT values within each row and returns an integer array of the same shape where each entry is replaced by its 0-based rank WITHIN ITS ROW (0 = the row's smallest). Do not use Python loops.

The double-argsort rank trick works along an axis too — apply both passes with axis=1.

## Reference solution


```python
import numpy as np

def solve(z):
    return z.argsort(axis=1).argsort(axis=1)


print(solve(np.array([[0.3, 0.1, 0.2]])))
```

## Why this works

argsort(axis=1) gives each row's sorting permutation; a second argsort(axis=1) inverts it row-wise, turning positions into ranks. This is the 1-D double-argsort idiom lifted to rows; the scatter alternative (writing arange into argsorted positions) computes the same table.
