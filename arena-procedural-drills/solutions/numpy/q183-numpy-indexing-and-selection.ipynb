## Problem

Given the 2D array Z of shape (rows, cols) = (4, 6) containing random floats in [0, 1), produce an integer array of the same shape where each entry is replaced by its rank within its row, with 1 assigned to the smallest value in that row and cols (=6) to the largest. Each row is ranked independently. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

argsort each row, then scatter 0..cols-1 into those sorted positions to invert the permutation into ranks.

</details>

## Reference solution


```python
import numpy as np

def solve():
    order = np.argsort(Z, axis=1)
    ranks = np.empty_like(order)
    ranks[np.arange(Z.shape[0])[:, None], order] = np.arange(Z.shape[1])
    return ranks + 1
```

## Why this works

argsort gives, for each row, the column positions in ascending order. Scattering the rank sequence 0..cols-1 back into those positions inverts the permutation so each element learns its own rank; adding 1 shifts to 1-based ranks with the smallest value getting 1.
