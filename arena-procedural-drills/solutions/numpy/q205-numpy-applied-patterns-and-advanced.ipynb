## Problem

Write a function solve(z) that takes a 2-D PyTorch integer tensor (rows may repeat) and returns the tensor of DISTINCT rows in order of FIRST APPEARANCE — not sorted. Each duplicated row appears once, at the position order its first copy occupied.

## Reference solution


```python
import torch as t
def solve(z):
    _, inverse = t.unique(z, dim=0, return_inverse=True)
    first = t.full((int(inverse.max()) + 1,), z.shape[0], dtype=t.int64)
    first.scatter_reduce_(0, inverse, t.arange(z.shape[0]), reduce="amin")
    return z[t.sort(first).values]


print(solve(t.tensor([[1, 0], [0, 1], [1, 0]])))
```
