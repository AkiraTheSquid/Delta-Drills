## Problem

Write a function solve(z) that takes a 2-D numpy integer array (rows may repeat) and returns the array of DISTINCT rows in order of FIRST APPEARANCE — not sorted. Each duplicated row appears once, at the position order its first copy occupied.


<details><summary>💡 Hint (click to reveal)</summary>

Row-wise unique can report each distinct row's first index — sort THOSE indices to restore scan order, then gather.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    _, idx = np.unique(z, axis=0, return_index=True)
    return z[np.sort(idx)]


print(solve(np.array([[1, 0], [0, 1], [1, 0]])))
```

## Why this works

np.unique(z, axis=0, return_index=True) deduplicates whole rows and yields each survivor's first-occurrence position, but ordered by row VALUE; sorting the indices recovers the original encounter order and indexing z with them returns the rows themselves — unlike plain unique(axis=0), which would return them lexicographically sorted.
