## Problem

You are given Z, a 2D integer array of shape (6, 3) of random 0/1 values where some rows may be duplicates of each other, and uZ, a 2D array containing only the distinct rows of Z with the original column count preserved. Return uZ so duplicate rows of Z appear exactly once in the output. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

np.unique can deduplicate whole rows directly when you tell it which axis to operate over.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 2, (6, 3))
    uZ = np.unique(Z, axis=0)
    return uZ

print(solve())
```

## Why this works

np.unique with axis=0 treats each row as a single unit, returning the distinct rows with the column count preserved and (as a side effect) in lexicographically sorted order. This is the modern one-liner replacement for the older void-view trick that achieved the same row deduplication.
