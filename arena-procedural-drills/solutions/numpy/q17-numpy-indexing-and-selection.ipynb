## Problem

Write a function solve(z) that takes a 2-D numpy array of floats and returns a new array with one extra row on the top and bottom and one extra column on each side, all filled with zeros — a one-cell-thick zero border surrounding the original values. An input of shape (r, c) therefore produces output of shape (r+2, c+2).


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a padding function whose width argument adds the same margin on every side — or allocate a bigger zero array and drop z into its interior slice.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.pad(z, pad_width=1, mode="constant", constant_values=0)


example = np.ones((2, 2))
print(solve(example))
```

## Why this works

np.pad(z, pad_width=1, mode='constant', constant_values=0) adds one zero-filled cell on every side, growing (r, c) to (r+2, c+2). Building a zeros((r+2, c+2)) array and assigning out[1:-1, 1:-1] = z is an equally valid route the tests accept.
