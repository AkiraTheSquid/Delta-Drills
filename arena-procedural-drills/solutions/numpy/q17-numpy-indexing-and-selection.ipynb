## Problem

Starting from a 5x5 array Z filled with ones, surround the existing values with a one-cell-thick border of zeros so the result is a 7x7 array whose outermost row and column on every side contain zeros and whose interior 5x5 region preserves the original ones. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

There is a padding function that adds a one-cell border of a constant value around an array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.ones((5, 5))
    Z = np.pad(Z, pad_width=1, mode='constant', constant_values=0)
    return Z

print(solve())
```

## Why this works

np.pad with pad_width=1 adds one cell on every side, and mode='constant' with constant_values=0 fills that new border with zeros, growing the 5x5 array to 7x7 while preserving the interior ones. This is the direct way to wrap an array in a zero frame.
