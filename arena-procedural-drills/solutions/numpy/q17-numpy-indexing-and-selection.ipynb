## Problem

Write a function solve(z) that takes a 2-D PyTorch tensor of floats and returns a new tensor with one extra row on the top and bottom and one extra column on each side, all filled with zeros — a one-cell-thick zero border surrounding the original values. An input of shape (r, c) therefore produces output of shape (r+2, c+2).

## Reference solution


```python
import torch as t
def solve(z):
    return t.nn.functional.pad(z, (1, 1, 1, 1), mode="constant", value=0)


example = t.ones((2, 2))
print(solve(example))
```
