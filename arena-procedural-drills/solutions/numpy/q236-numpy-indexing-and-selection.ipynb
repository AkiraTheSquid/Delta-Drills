## Problem

Write a function solve(x, threshold) that takes a 2-D numpy array x of floats and a scalar threshold. It should return a boolean numpy array with the same shape as x that is True exactly where the corresponding entry of x is strictly greater than threshold. Entries equal to the threshold must come out False. Do not use Python loops.

Comparing a numpy array to a scalar with a comparison operator broadcasts elementwise and yields a boolean mask of the same shape.

## Reference solution


```python
import numpy as np

def solve(x, threshold):
    return x > threshold


example = np.array([[1.0, 5.0, 2.0], [7.0, 0.5, 3.0]])
print(solve(example, 2.5))
```

## Why this works

A comparison between a numpy array and a scalar is applied elementwise: numpy broadcasts the scalar across every entry and returns an array of dtype bool with the same shape as the input. Using the strict greater-than comparison means entries exactly equal to the threshold produce False, which matches the required contract. No loops are needed because the vectorized operator handles any 2-D shape, including a single row or a single element.
