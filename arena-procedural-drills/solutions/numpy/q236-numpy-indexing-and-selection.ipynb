## Problem

You are given Z, a 2D float array of shape (3, 3) of random values in [0, 1). Return a 2D boolean array of the same (3, 3) shape where each entry is True if the corresponding entry of Z is strictly greater than 0.5 and False otherwise. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

A single elementwise comparison against the threshold produces the mask.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((3, 3))
    return Z > 0.5

print(solve())
```

## Why this works

The elementwise comparison Z > 0.5 broadcasts the threshold across every entry, returning a boolean array of the same (3, 3) shape that is True only where the value strictly exceeds 0.5.
