## Problem

Write a function solve(z, step, v) that takes a 1-D PyTorch tensor z, a positive integer step, and a value v. Return a new tensor equal to z except that every step-th element — indices 0, step, 2*step, ... — has been overwritten with v. The input tensor must not be modified.


<details><summary>💡 Hint (click to reveal)</summary>

A strided slice can be the TARGET of an assignment — do it on a copy so the caller's tensor survives.

</details>

## Reference solution


```python
import torch as t

def solve(z, step, v):
    out = z.clone()
    out[::step] = v
    return out


example = t.arange(12)
print(solve(example, 3, -1))
```

## Why this works

out[::step] = v assigns the scalar into every step-th slot starting at index 0 in one vectorized statement. Copying first keeps the caller's tensor intact, which the first test verifies by checking z after the call.
