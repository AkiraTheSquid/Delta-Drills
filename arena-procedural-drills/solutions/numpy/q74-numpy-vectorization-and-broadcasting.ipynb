## Problem

Given the 1D array Z = np.arange(12) (values 0..11), overwrite every third element starting at index 0 with the value -1, leaving all other entries untouched. The result is still a length-12 1D array of integers where the overwritten positions hold -1 and the remaining positions keep their original arange values. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Assign the scalar to a step-3 slice of the vector.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(12)
    Z[::3] = -1
    return Z

print(solve())
```

## Why this works

A strided slice Z[::3] addresses every third element starting at index 0, and assigning a scalar to it broadcasts -1 into those positions in place while leaving the others untouched. The result stays a length-12 integer vector.
