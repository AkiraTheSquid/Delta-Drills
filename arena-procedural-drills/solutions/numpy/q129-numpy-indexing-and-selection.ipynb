## Problem

Given a set of points in 2D, find the indices of the two closest points for each point in the set based on Euclidean distance. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Form the full squared-distance matrix by broadcasting, then use a partial sort to grab the smallest distances per row while skipping the point's own zero-distance entry.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(6, 2)
    X, Y = Z[:, None, :], Z[None, :, :]
    dist2 = ((X - Y)**2).sum(axis=2)
    k = 2
    idx = np.argpartition(dist2, kth=k+1, axis=1)[:, 1:k+1]
    return idx

print(solve())
```

## Why this works

Broadcasting Z against itself along new axes builds the (6,6) matrix of squared distances without an explicit loop. Each row's nearest points are the smallest entries, but the self-distance (zero) is always smallest, so argpartition is used to cheaply pull the smallest few indices and the first column (self) is dropped, leaving the two closest neighbors per point.
