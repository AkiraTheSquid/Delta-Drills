## Problem

Write a function solve(z) that takes an (n, 2) tensor of 2-D points with all pairwise distances distinct, and returns an (n, 2) integer tensor whose row i holds the indices of the TWO points closest to point i (excluding i itself). The two indices within each row are graded as a set — either order passes.


<details><summary>💡 Hint (click to reveal)</summary>

Build the squared-distance matrix by broadcasting, then argsort each row — position 0 is always the point itself, so the neighbors sit at positions 1 and 2.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    dist2 = ((z[:, None, :] - z[None, :, :]) ** 2).sum(dim=2)
    return t.argsort(dist2, dim=1)[:, 1:3]


example = t.tensor([[0.0, 0.0], [1.0, 0.0], [5.0, 0.0], [1.2, 0.2]])
print(solve(example))
```

## Why this works

Squared distances sort identically to distances, so the sqrt is unnecessary. Row-wise argsort puts each point's own index first (distance 0), making columns 1 and 2 the two nearest neighbors. argpartition(dist2, kth=2)[:, 1:3] finds the same pair faster; the grader sorts each row so either method's ordering passes.
