## Problem

Write a function solve(x) that takes a 1-D numpy float array with at least 3 entries and returns the moving average over every window of 3 consecutive entries: output entry i is the mean of x[i], x[i+1], x[i+2]. The result has length len(x) - 2.


<details><summary>💡 Hint (click to reveal)</summary>

A moving average is a convolution with a uniform kernel — and the 'valid' mode keeps only fully overlapping windows.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.convolve(x, np.ones(3) / 3.0, mode="valid")


print(solve(np.arange(10, dtype=float)))
```

## Why this works

Convolving with the kernel [1/3, 1/3, 1/3] in 'valid' mode averages each complete window of three neighbors, yielding len(x)-2 outputs. Summing shifted slices (x[:-2] + x[1:-1] + x[2:]) / 3 computes the identical result.
