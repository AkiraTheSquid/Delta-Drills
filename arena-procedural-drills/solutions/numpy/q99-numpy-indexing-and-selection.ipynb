## Problem

Given a 1D array x of length 10, calculate the average of each consecutive triplet of values, returning a new array that contains these averages. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Convolve the signal with a uniform length-3 kernel of weight 1/3 using mode='valid'.

## Reference solution


```python
import numpy as np

def solve():
    x = np.arange(10, dtype=float)
    kernel = np.ones(3) / 3.0
    return np.convolve(x, kernel, mode='valid')

print(solve())
```

## Why this works

Convolving with a length-3 kernel of equal weights 1/3 computes a sliding window average of every consecutive triplet. mode='valid' keeps only windows that fully overlap the input, yielding len(x)-2 = 8 averages.
