## Problem

Write a function solve(v, length) that takes a 1-D PyTorch float tensor v (not all zeros) and a positive target length, and returns v rescaled by a single positive constant so that its Euclidean norm equals exactly `length`, preserving direction. Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Dividing by the current norm gives a unit vector — then scale that up to whatever length you were asked for.

</details>

## Reference solution


```python
import torch as t

def solve(v, length):
    return v * (length / t.linalg.norm(v))


print(solve(t.tensor([3.0, 4.0]), 10.0))
```

## Why this works

v / ||v|| has norm 1, so multiplying by the target gives norm `length` exactly, and a single positive scale factor cannot change direction. The last case checks the achieved norm numerically rather than the entries, so any algebraically equivalent scaling passes.
