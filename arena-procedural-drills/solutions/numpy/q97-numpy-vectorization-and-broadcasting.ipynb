## Problem

You are given Z, a 1D vector of 10 random floats in [0, 1). Rescale Z by a single positive constant so that the resulting vector has Euclidean length equal to 1, i.e. the square root of the sum of its squared entries equals 1. The result has shape (10,) and points in the same direction as Z. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Divide the vector by its L2 norm (np.linalg.norm).

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(10)
    return Z / np.linalg.norm(Z)

print(solve())
```

## Why this works

Dividing a vector by its Euclidean (L2) norm rescales it by a single positive constant, leaving its direction unchanged while forcing the sum of squared entries to 1. np.linalg.norm computes that L2 length by default.
