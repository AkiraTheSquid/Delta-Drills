## Problem

Write a function solve(x) that takes a 2-D PyTorch float tensor (no constant columns) and returns a new tensor in which every COLUMN has been standardized: subtract that column's mean and divide by that column's population standard deviation, so each output column has mean 0 and std 1.

## Reference solution


```python
import torch as t

def solve(x):
    return (x - x.mean(dim=0)) / x.std(dim=0, correction=0)


example = t.tensor([[1.0, 10.0], [3.0, 30.0]])
print(solve(example))
```
