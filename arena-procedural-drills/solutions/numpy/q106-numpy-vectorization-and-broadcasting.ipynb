## Problem

Write a function solve(x) that takes a 2-D numpy float array (no constant columns) and returns a new array in which every COLUMN has been standardized: subtract that column's mean and divide by that column's population standard deviation, so each output column has mean 0 and std 1.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce along axis 0 for both statistics — the resulting per-column vectors broadcast back across the rows on their own.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return (x - x.mean(axis=0)) / x.std(axis=0)


example = np.array([[1.0, 10.0], [3.0, 30.0]])
print(solve(example))
```

## Why this works

x.mean(axis=0) and x.std(axis=0) are length-cols vectors of per-column statistics; subtracting and dividing broadcasts them down every row, standardizing columns independently. The last case verifies the defining property (each column: mean 0, std 1) directly.
