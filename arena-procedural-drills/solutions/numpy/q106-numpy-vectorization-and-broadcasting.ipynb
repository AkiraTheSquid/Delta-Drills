## Problem

Given a matrix X, transform each column so that its mean is zero and its standard deviation is one. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Subtract the column means and divide by the column standard deviations, both taken with axis=0.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.randn(5, 4)
    return (X - X.mean(axis=0)) / X.std(axis=0, ddof=0)

print(solve())
```

## Why this works

Per-column standardization subtracts each column's mean and divides by its standard deviation. Using axis=0 reduces over the rows so the statistics are computed independently per column, and ddof=0 gives the population std; broadcasting applies the (4,) stats across all rows.
