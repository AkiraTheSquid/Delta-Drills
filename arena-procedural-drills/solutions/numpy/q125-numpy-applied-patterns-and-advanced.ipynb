## Problem

Write a function solve(text) that takes a string of CSV data — rows of comma-separated integer fields where some fields are blank (whitespace only) — and returns a 2-D integer numpy array parsed from it, with every blank field filled in as 0.


<details><summary>💡 Hint (click to reveal)</summary>

np.genfromtxt (unlike loadtxt) tolerates missing fields — and takes an argument saying what to substitute for them.

</details>

## Reference solution


```python
import numpy as np
from io import StringIO

def solve(text):
    return np.genfromtxt(StringIO(text), delimiter=",", dtype=int, filling_values=0)


example = "1, 2, 3\n4,  , 6\n"
print(solve(example))
```

## Why this works

StringIO wraps the string as a file-like object; np.genfromtxt parses it with delimiter=',' and substitutes filling_values=0 wherever a field is blank. np.loadtxt would raise on the missing fields — handling gaps is exactly what genfromtxt adds. (numpy 2.x note: dtype=int, not the removed np.int alias.)
