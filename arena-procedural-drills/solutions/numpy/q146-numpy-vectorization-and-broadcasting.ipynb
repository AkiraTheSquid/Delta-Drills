## Problem

Write a function solve(a, b, c) that takes three 1-D numpy arrays of the same length n and returns a single array of length 3n interleaving them position by position: the result reads a[0], b[0], c[0], a[1], b[1], c[1], and so on. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Allocate the full-length result and fill it with three strided assignments — every third slot per source.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b, c):
    out = np.empty(a.size * 3, dtype=np.result_type(a, b, c))
    out[0::3] = a
    out[1::3] = b
    out[2::3] = c
    return out


print(solve(np.array([1, 2]), np.array([10, 20]), np.array([100, 200])))
```

## Why this works

Strided slices out[0::3], out[1::3], out[2::3] address the interleaved positions of each source, and slice assignment fills them without loops. np.column_stack((a, b, c)).ravel() is the same interleave via a reshape: each row (a[i], b[i], c[i]) flattens consecutively.
