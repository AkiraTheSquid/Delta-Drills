## Problem

Given two 1D arrays A and B of equal length n, produce a length-2n array C in which the entries of A and B alternate position by position, so C[2i] is A[i] and C[2i+1] is B[i] for every valid i. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Make an empty length-2n array and assign A and B into its even and odd strided slices.

</details>

## Reference solution


```python
import numpy as np

def solve():
    A = np.arange(5)
    B = 10 + np.arange(5)
    C = np.empty(A.size + B.size, dtype=A.dtype)
    C[0::2] = A
    C[1::2] = B
    return C

print(solve())
```

## Why this works

Allocate an output of length 2n, then use strided assignment: the even slice [0::2] receives A and the odd slice [1::2] receives B. This writes each source array into alternating positions in a single vectorized step, avoiding any Python-level interleave loop.
