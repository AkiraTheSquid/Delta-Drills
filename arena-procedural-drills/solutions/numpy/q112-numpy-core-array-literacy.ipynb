## Problem

Given a square matrix Z, retain only the elements that are within one position of the main diagonal, setting all other elements to zero. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Build a mask from the absolute difference of row and column indices and keep only the band within one step.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(25).reshape(5, 5)
    mask = np.abs(np.arange(5)[:, None] - np.arange(5)[None, :]) <= 1
    return Z * mask

print(solve())
```

## Why this works

The outer difference of row and column index vectors gives, at each position, its distance from the main diagonal. Keeping entries where that absolute distance is at most 1 produces a tridiagonal band mask, and multiplying zeroes out everything else.
