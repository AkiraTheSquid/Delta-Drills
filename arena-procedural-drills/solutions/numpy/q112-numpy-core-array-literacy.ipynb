## Problem

Write a function solve(z) that takes a square 2-D numpy array and returns a new array of the same shape in which entries within one position of the main diagonal (the diagonal itself plus the two adjacent diagonals) keep their values and every other entry is zero. Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

An entry [i, j] is near the diagonal exactly when |i - j| is small — build that condition as a broadcast boolean mask.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    n = z.shape[0]
    mask = np.abs(np.arange(n)[:, None] - np.arange(n)[None, :]) <= 1
    return z * mask


example = np.arange(16).reshape(4, 4)
print(solve(example))
```

## Why this works

Broadcasting a column of row indices against a row of column indices gives |i - j| for every cell; comparing to 1 marks the tridiagonal band, and multiplying by the boolean mask zeroes everything else (True acts as 1, False as 0). np.tril(np.triu(z, -1), 1) composes the same band.
