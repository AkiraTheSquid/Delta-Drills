## Problem

Given a 1D float array Z, return a new float array of the same shape where each entry has been replaced by the largest integer that is less than or equal to it (so positives lose their fractional part toward zero and negatives are pushed further negative). (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

There is a single ufunc that rounds each element down toward negative infinity.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.array([1.2, 3.7, -0.3])
    return np.floor(Z)


print(solve())
```

## Why this works

np.floor returns the largest integer not exceeding each element, as floats. It rounds toward negative infinity, so positives drop their fraction and negatives move further negative (-0.3 -> -1.0).
