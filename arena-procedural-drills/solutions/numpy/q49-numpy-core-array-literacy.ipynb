## Problem

Write a function solve(z) that takes a 1-D PyTorch tensor of floats and returns a new float tensor of the same shape in which each entry is replaced by the largest integer value less than or equal to it (the floor). Positive numbers lose their fractional part, while negative numbers with any fractional part move further from zero (for example -0.3 becomes -1.0).

PyTorch has an elementwise rounding family — pick the member that always rounds toward negative infinity, not toward zero.

## Reference solution


```python
import torch as t

def solve(z):
    return t.floor(z)


example = t.tensor([1.2, 3.7, -0.3])
print(solve(example))
```

## Why this works

t.floor rounds every entry toward negative infinity, which is what 'largest integer less than or equal' means. It differs from truncation on negatives: floor(-0.3) is -1.0 while int-style truncation would give 0 — the negative cases pin that down. Whole numbers are already their own floor.
