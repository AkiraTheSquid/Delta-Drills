## Problem

Write a function solve(a, mapping) that takes a 1-D non-negative integer array a and a Python dict of {old_value: new_value} integer pairs whose keys include every value appearing in a. Return a new array with each entry replaced by its mapped value, WITHOUT looping over the elements of a (looping over the small dict is fine).


<details><summary>💡 Hint (click to reveal)</summary>

Turn the dict into a dense lookup TABLE indexed by old value — then one fancy-index gathers every replacement at once.

</details>

## Reference solution


```python
import numpy as np

def solve(a, mapping):
    table = np.arange(max(mapping) + 1)
    for k, v in mapping.items():
        table[k] = v
    return table[a]


print(solve(np.array([1, 2, 3, 2, 1, 0]), {0: 9, 1: 8, 2: 7, 3: 6}))
```

## Why this works

Building table so table[old] == new converts the dict into an array the size of the largest key; table[a] then performs a vectorized gather, replacing every element in one operation regardless of a's length. The per-element work happens in numpy, with Python only touching the handful of dict entries.
