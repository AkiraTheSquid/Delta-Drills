## Problem

Given a 1D array `a` and a dict `mapping` of `{old: new}` integer pairs, return a new array with each value replaced by its mapped value (assume keys cover the integer range up to max key). (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Make a lookup array large enough to cover the maximum key, fill in the mappings, then index it with a.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.array([1, 2, 3, 2, 1, 0])
    mapping = {0: 9, 1: 8, 2: 7, 3: 6}
    table = np.arange(max(mapping) + 1)
    for k, v in mapping.items():
        table[k] = v
    return table[a]

print(solve())
```

## Why this works

Building a lookup table indexed by the old value lets the remapping happen as a single fancy-index operation table[a]. The table must span up to the largest key so every value in a has a valid slot; the original bug used a table sized only to the dict length.
