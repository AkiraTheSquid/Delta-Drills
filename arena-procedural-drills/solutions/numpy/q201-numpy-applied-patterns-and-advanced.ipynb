## Problem

Write a function solve(p, p0, p1) that takes an (n, 2) tensor of 2-D points p and two (m, 2) tensors p0 and p1 giving the endpoints of m infinite lines. Return the (n, m) tensor of perpendicular distances from every point to every LINE (the infinite line through the endpoints, not the segment). No two endpoints coincide.


<details><summary>💡 Hint (click to reveal)</summary>

The 2-D cross product of the line direction with the point-to-line-start vector measures twice the triangle area — divide by the direction's length to recover the height.

</details>

## Reference solution


```python
import torch as t

def solve(p, p0, p1):
    d = p1 - p0
    cross = t.abs(d[None, :, 0] * (p[:, None, 1] - p0[None, :, 1])
                   - d[None, :, 1] * (p[:, None, 0] - p0[None, :, 0]))
    return cross / t.sqrt((d ** 2).sum(-1))[None, :]


pts = t.tensor([[0.0, 1.0]])
print(solve(pts, t.tensor([[0.0, 0.0]]), t.tensor([[1.0, 0.0]])))
```

## Why this works

For direction d = p1 - p0 and offset w = point - p0, |d x w| equals |d| times the perpendicular distance, so dividing the absolute 2-D cross product by |d| gives the distance. Broadcasting the n points against the m lines fills the whole table without loops.
