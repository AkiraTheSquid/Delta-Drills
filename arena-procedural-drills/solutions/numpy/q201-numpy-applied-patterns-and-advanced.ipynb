## Problem

Given a set of 2D points P, plus two sets of 2D points P0 and P1 each describing the endpoints of a line, compute the perpendicular distance from every point in P to every line, returning a (len(P), len(P0)) matrix of distances. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Write each line in implicit ax+by+c form from its endpoints, then broadcast point coordinates against the per-line coefficients and divide by the line length.

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    P0 = np.random.uniform(-10, 10, (5, 2))
    P1 = np.random.uniform(-10, 10, (5, 2))
    P = np.random.uniform(-10, 10, (5, 2))
    dx = P1[:, 0] - P0[:, 0]
    dy = P1[:, 1] - P0[:, 1]
    cross = P1[:, 0] * P0[:, 1] - P1[:, 1] * P0[:, 0]
    num = np.abs(dy[None, :] * P[:, 0, None] - dx[None, :] * P[:, 1, None] + cross[None, :])
    den = np.sqrt(dx ** 2 + dy ** 2)
    return num / den[None, :]

print(solve())
```

## Why this works

Each line through P0->P1 has the implicit form dy*x - dx*y + (x1*y0 - y1*x0) = 0, and the perpendicular distance of a point is the absolute value of that expression divided by the line's length sqrt(dx^2+dy^2). Broadcasting point coordinates (shape (n,1)) against per-line coefficients (shape (1,m)) evaluates every point-line pair into the (len(P), len(P0)) distance matrix in one shot.
