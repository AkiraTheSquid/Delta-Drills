## Problem

You are given Z, a 1D float array of length 10 in which every entry is 1.0. Return Z so the printed result is a length-10 vector consisting of ten 1.0 values with the default floating-point dtype, with no zeros and no other values present anywhere along its single axis. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the ones constructor with the desired length; the default dtype is already floating point.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.ones(10)
    return Z

print(solve())
```

## Why this works

np.ones(10) creates a length-10 vector filled with 1.0 using the default float dtype, so every entry is the floating-point value one.
