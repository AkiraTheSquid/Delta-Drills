## Problem

Write a function solve(n) that takes a non-negative integer n and returns a 1-D PyTorch tensor of length n in which every entry is 1.0. The result must use torch's default floating-point dtype, not integers. Note that torch's default float is float32, where numpy's is float64 — the grader checks the dtype as well as the values.

## Reference solution


```python
import torch as t

def solve(n):
    return t.ones(n)


example = 4
print(solve(example))
```

## Why this works

t.ones(n) allocates a new tensor of shape (n,) and fills every slot with 1.0, using PyTorch's default float64 dtype since no dtype argument is given. Because the shape argument is just the integer n, the result is 1-D, and it works for any non-negative length including 0 (an empty float tensor) and 1.
