## Problem

Write a function solve(n) that takes a non-negative integer n and returns a 1-D numpy array of length n in which every entry is 1.0. The result must use numpy's default floating-point dtype, not integers. The grader checks the values, the shape, and the dtype for several different lengths.

## Reference solution


```python
import numpy as np

def solve(n):
    return np.ones(n)


example = 4
print(solve(example))
```

## Why this works

np.ones(n) allocates a new array of shape (n,) and fills every slot with 1.0, using numpy's default float64 dtype since no dtype argument is given. Because the shape argument is just the integer n, the result is 1-D, and it works for any non-negative length including 0 (an empty float array) and 1.
