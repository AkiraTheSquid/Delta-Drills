## Problem

You are given Z, a 1D float array of length 10 that started as all zeros and then had its entries at indices 3, 4, 5, 6, 7 (the slice 3:8) overwritten with 5.0. Return Z, a length-10 vector that is 0.0 outside that index range and 5.0 inside it. (★☆☆)
Print the result.

Start from a vector of zeros and assign a scalar into a half-open slice; the stop index is exclusive.

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros(10)
    Z[3:8] = 5
    return Z

print(solve())
```

## Why this works

Allocate a length-10 zero vector, then use slice assignment Z[3:8] = 5 to overwrite the five entries at indices 3 through 7 in place. Slice assignment broadcasts the scalar 5.0 across the targeted view, leaving every other entry at its initial 0.0.
