## Problem

Write a function solve(tensors) that takes a list of 2-D PyTorch tensors of arbitrary (possibly different) shapes and returns the BLOCK-DIAGONAL matrix containing them in order along the diagonal: total shape is (sum of heights, sum of widths), with each input occupying its own diagonal block and zeros everywhere else.


<details><summary>💡 Hint (click to reveal)</summary>

Total the heights and widths for the canvas, then walk the blocks, dropping each into its slice and advancing both offsets by its shape.

</details>

## Reference solution


```python
import torch as t

def solve(arrays):
    rows = sum(a.shape[0] for a in arrays)
    cols = sum(a.shape[1] for a in arrays)
    out = t.zeros((rows, cols))
    r = c = 0
    for a in arrays:
        out[r:r + a.shape[0], c:c + a.shape[1]] = a
        r += a.shape[0]
        c += a.shape[1]
    return out


print(solve([t.ones((1, 2)), 2 * t.ones((2, 1))]))
```

## Why this works

The canvas is zeros of the summed shape; each block lands at the running (row, col) offset via one slice assignment, and advancing both offsets by the block's dimensions keeps the blocks on the diagonal with zero overlap. scipy.linalg.block_diag does this in one call — this drill builds the mechanism by hand.
