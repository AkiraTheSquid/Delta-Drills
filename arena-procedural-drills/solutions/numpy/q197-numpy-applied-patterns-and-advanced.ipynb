## Problem

Given a list of 2D arrays, create a new 2D array that contains the input arrays along its diagonal, filling the rest with zeros. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Allocate a zero matrix sized to the total rows and columns, then walk a running (row, col) offset and paste each block into its slice.

</details>

## Reference solution


```python
import numpy as np

def solve():
    arrays = [np.ones((2, 3)), 2 * np.ones((1, 2)), 3 * np.ones((3, 1))]
    R = sum(a.shape[0] for a in arrays)
    C = sum(a.shape[1] for a in arrays)
    out = np.zeros((R, C))
    ri = ci = 0
    for a in arrays:
        r, c = a.shape
        out[ri:ri + r, ci:ci + c] = a
        ri += r
        ci += c
    return out

print(solve())
```

## Why this works

A block-diagonal layout places each input block so its rows and columns occupy a disjoint band, with everything else zero. Tracking running row and column offsets and writing each array into the matching slice of a zero matrix builds exactly that structure (equivalent to scipy.linalg.block_diag).
