## Problem

Given a one-dimensional array D and a corresponding index array S, compute the average of the elements in D for each unique index in S. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Use weighted bincount for the per-group sums and unweighted bincount for the per-group counts, then divide.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    D = np.random.uniform(0, 1, 100)
    S = np.random.randint(0, 10, 100)
    D_sums = np.bincount(S, weights=D)
    D_counts = np.bincount(S)
    D_means = D_sums / D_counts
    return D_means

print(solve())
```

## Why this works

Group means are the per-group sums divided by per-group counts. bincount(S, weights=D) scatter-adds each D value into the bin named by its index S, while plain bincount(S) counts the group sizes; dividing elementwise gives the mean of D within each unique index.
