## Problem

Write a function solve(x, k) that takes a 1-D numpy array x and a positive integer k, and returns a new 1-D array in which each element of x appears k times in a row before the next element starts. For example, with x = [1, 2, 3] and k = 3 the result is [1, 1, 1, 2, 2, 2, 3, 3, 3]. Note that this is different from laying whole copies of the array end to end. Your function must work for any element values and any k >= 1.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a built-in that duplicates each individual element a given number of times consecutively — it is not the one that stacks whole copies of the array end to end.

## Reference solution


```python
import numpy as np

def solve(x, k):
    return np.repeat(x, k)


example = np.array([1, 2, 3])
print(solve(example, 3))
```

## Why this works

np.repeat(x, k) walks through x element by element and emits k consecutive copies of each one, so the relative order of the original elements is preserved and the result has x.size * k entries. This is the element-wise counterpart of np.tile, which instead concatenates whole copies of the array one after another. Because k = 1 is a valid repeat count, the same call also handles the identity case without any special-casing.
