## Problem

Given the array [1, 2, 3], return a new array where each element appears three times in order. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use the element-wise duplication function, not the block-tiling one.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([1, 2, 3])
    return np.repeat(Z, 3)

print(solve())
```

## Why this works

np.repeat duplicates each element in place the given number of times, so repeating [1,2,3] three times yields 1 1 1 2 2 2 3 3 3. This contrasts with tiling, which repeats the whole block.
