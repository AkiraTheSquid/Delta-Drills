## Problem

Write a function solve(x, k) that takes a 1-D PyTorch tensor x and a positive integer k, and returns a new 1-D tensor in which each element of x appears k times in a row before the next element starts. For example, with x = [1, 2, 3] and k = 3 the result is [1, 1, 1, 2, 2, 2, 3, 3, 3]. Note that this is different from laying whole copies of the tensor end to end — torch has a separate operation for each. Your function must work for any element values and any k >= 1.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has a built-in that duplicates each individual element a given number of times consecutively — it is not the one that stacks whole copies of the tensor end to end.

</details>

## Reference solution


```python
import torch as t

def solve(x, k):
    return t.repeat_interleave(x, k)


example = t.tensor([1, 2, 3])
print(solve(example, 3))
```
