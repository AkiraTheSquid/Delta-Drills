## Problem

Build a 5x5 floating-point matrix where every entry on or above the main diagonal equals 1.0 and every entry strictly below the main diagonal equals 0.0. The result is the upper-triangular indicator of a 5x5 matrix. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

There is a dedicated NumPy function that extracts the upper triangle of a matrix.

</details>

## Reference solution


```python
import numpy as np

def solve():
    return np.triu(np.ones((5, 5)))

print(solve())
```

## Why this works

np.triu keeps the upper triangle (on and above the main diagonal) of its input and zeros everything below. Applied to an all-ones 5x5 array it yields the upper-triangular indicator matrix.
