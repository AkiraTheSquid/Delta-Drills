## Problem

Write a function solve(n) that takes a positive integer n and returns an n x n numpy array of floats. Every entry on or above the main diagonal must be 1.0, and every entry strictly below the diagonal must be 0.0. For n = 1 the result is simply [[1.0]]. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Start from a matrix of all ones, then use numpy's upper-triangle helper to zero out everything strictly below the main diagonal.

</details>

## Reference solution


```python
import numpy as np

def solve(n):
    return np.triu(np.ones((n, n)))


example = 4
print(solve(example))
```

## Why this works

np.ones((n, n)) builds an n x n matrix of ones, which defaults to float64, so the entries are already 1.0. np.triu keeps every entry on and above the main diagonal and replaces everything strictly below it with zero, producing exactly the upper-triangular indicator. The same single call works for any side length, including the n = 1 edge case where the result is just [[1.0]].
