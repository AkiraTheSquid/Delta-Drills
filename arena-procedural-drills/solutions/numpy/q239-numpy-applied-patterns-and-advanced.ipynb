## Problem

Write a function solve(a, b) that takes two 2-D PyTorch tensors a and b, where a has shape (m, k) and b has shape (k, n). It should return the matrix product of a and b as a 2-D tensor of shape (m, n). Do not use Python loops; the shapes vary between test cases, so your function must work for any compatible pair.


<details><summary>💡 Hint (click to reveal)</summary>

Elementwise multiplication is not what you want here — look for the operation that combines each row of the first tensor with each column of the second.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return a @ b


a = t.tensor([[1.0, 2.0], [3.0, 4.0]])
b = t.tensor([[5.0, 6.0], [7.0, 8.0]])
print(solve(a, b))
```
