## Problem

You are given a 2D array of shape (5, 3) filled with 1.0 and a 2D array of shape (3, 2) filled with 1.0. Return their matrix product as a 2D array of shape (5, 2), where each output entry is the sum of three 1*1 products. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the real matrix product (not elementwise) so the shared inner dimension is contracted.

</details>

## Reference solution


```python
import numpy as np

def solve():
    return np.ones((5, 3)) @ np.ones((3, 2))

print(solve())
```

## Why this works

Matrix multiplication of a (5,3) all-ones matrix with a (3,2) all-ones matrix contracts the shared inner dimension of length 3, so every output entry sums three 1*1 products and equals 3.0, giving a (5,2) result. The @ operator is equivalent to np.matmul here.
