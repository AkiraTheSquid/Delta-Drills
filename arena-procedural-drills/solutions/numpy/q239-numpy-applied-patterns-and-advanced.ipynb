## Problem

Write a function solve(a, b) that takes two 2-D numpy arrays a and b, where a has shape (m, k) and b has shape (k, n). It should return the matrix product of a and b as a 2-D array of shape (m, n). Do not use Python loops; the shapes vary between test cases, so your function must work for any compatible pair.


<details><summary>💡 Hint (click to reveal)</summary>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return a @ b


a = np.array([[1.0, 2.0], [3.0, 4.0]])
b = np.array([[5.0, 6.0], [7.0, 8.0]])
print(solve(a, b))
```

## Why this works

The @ operator (equivalently np.matmul) performs true matrix multiplication: entry (i, j) of the result is the dot product of row i of a with column j of b. Because a has shape (m, k) and b has shape (k, n), the inner dimensions match and the result is (m, n). This works for any compatible pair, including a single row times a single column, which yields a 1-by-1 array.
