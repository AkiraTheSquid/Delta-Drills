## Problem

Compute the Khatri-Rao product (column-wise Kronecker) of two matrices with the same number of columns (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcast A and B with inserted singleton row axes so the column axis aligns, multiply, then collapse the two row axes into one.

## Reference solution


```python
import numpy as np

def solve():
    A = np.array([[1, 2], [3, 4], [5, 6]])
    B = np.array([[7, 8], [9, 10]])
    KR = (A[:, None, :] * B[None, :, :]).reshape(A.shape[0] * B.shape[0], A.shape[1])
    return KR

print(solve())
```

## Why this works

The Khatri-Rao product is the column-wise Kronecker product: for each shared column, every entry of A's column is multiplied by every entry of B's column. Broadcasting A as (rows_A, 1, cols) against B as (1, rows_B, cols) forms all those products, and reshaping merges the two row axes into the stacked (rows_A*rows_B, cols) result.
