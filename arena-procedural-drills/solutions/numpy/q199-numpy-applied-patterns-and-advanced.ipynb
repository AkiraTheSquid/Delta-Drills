## Problem

Write a function solve(a, b) that takes two 2-D arrays with the SAME number of columns — a of shape (ra, c) and b of shape (rb, c) — and returns their Khatri-Rao product (column-wise Kronecker product): a (ra*rb, c) array whose column k is the outer product of a's column k with b's column k, flattened. Row index i*rb + j of the result holds a[i] * b[j] elementwise.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcast a's rows against b's rows into an (ra, rb, c) product cube, then collapse the two row axes into one.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return (a[:, None, :] * b[None, :, :]).reshape(-1, a.shape[1])


print(solve(np.array([[1, 2], [3, 4]]), np.array([[5, 6]])))
```

## Why this works

a[:, None, :] * b[None, :, :] pairs every row of a with every row of b column-by-column; reshaping (ra, rb, c) to (ra*rb, c) lays those pairings out with a's index varying slowest — the Khatri-Rao layout used in tensor decompositions.
