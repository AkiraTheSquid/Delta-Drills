## Problem

Write a function solve(z) that takes a 1-D PyTorch tensor and returns a tuple of two tensors: the distinct values of z in ascending order, and, aligned with them, the number of times each value occurs. Do not use Python loops.

## Reference solution


```python
import torch as t

def solve(z):
    return t.unique(z, return_counts=True)


example = t.tensor([1, 2, 2, 3, 3, 3])
print(solve(example))
```

## Why this works

t.unique(z, return_counts=True) returns the sorted distinct values and a parallel tensor of their occurrence counts in one call. The values come back in ascending order, which is exactly the required alignment for the counts.
