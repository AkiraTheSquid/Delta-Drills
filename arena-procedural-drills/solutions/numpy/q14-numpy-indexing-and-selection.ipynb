## Problem

Given an array Z, return the distinct elements along with the number of times each element appears in the array. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

One function gives both the sorted distinct values and, with the right flag, the matching occurrence counts.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([1, 2, 2, 3, 3, 3])
    vals, cnts = np.unique(Z, return_counts=True)
    return vals, cnts

print(solve())
```

## Why this works

np.unique returns the sorted distinct values, and with return_counts=True it additionally returns how many times each distinct value occurs. The two parallel arrays line up element by element: value and its frequency.
