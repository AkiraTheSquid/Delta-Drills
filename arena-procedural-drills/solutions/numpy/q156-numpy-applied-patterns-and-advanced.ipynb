## Problem

Given the 2D array Z of shape (6, 7) containing random integers in [0, 5), build the summed-area table (integral image) I of the same shape, where I[i, j] equals the sum of all entries Z[i', j'] with i' <= i and j' <= j (the total of the upper-left rectangle ending at row i and column j). (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Cumsum along one axis, then cumsum the result along the other.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 5, (6, 7))
    I = np.cumsum(np.cumsum(Z, axis=0), axis=1)
    return I

print(solve())
```

## Why this works

A summed-area table accumulates over both axes. Taking a cumulative sum down the rows (axis=0) gives column-wise prefix sums, and a second cumulative sum across the columns (axis=1) of that result yields the full upper-left rectangle total at every position.
