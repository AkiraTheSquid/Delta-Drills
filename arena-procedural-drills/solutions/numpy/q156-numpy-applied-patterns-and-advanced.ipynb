## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns its summed-area table (integral image): an array of the same shape where entry [i, j] is the sum of ALL entries z[i', j'] with i' <= i and j' <= j. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Accumulate down the rows, then accumulate that result across the columns — two cumulative sums compose into the 2-D prefix sum.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.cumsum(np.cumsum(z, axis=0), axis=1)


print(solve(np.array([[1, 2], [3, 4]])))
```

## Why this works

A cumsum along axis 0 makes each cell the sum of its column above it; a second cumsum along axis 1 then sums those column-sums leftward, yielding the full upper-left rectangle total at every cell. On an all-ones array the entry [i, j] is (i+1)*(j+1) — the rectangle's area.
