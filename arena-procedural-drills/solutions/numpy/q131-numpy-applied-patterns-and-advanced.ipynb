## Problem

Given a vector of ones, increment each element specified by a second vector of indices, ensuring that repeated indices are handled correctly. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Count occurrences with bincount and add them, then apply an unbuffered np.add.at pass so repeated indices are not dropped.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.ones(10)
    I = np.random.randint(0, len(Z), 20)
    Z += np.bincount(I, minlength=len(Z))
    np.add.at(Z, I, 1)
    return Z

print(solve())
```

## Why this works

Naive Z[I] += 1 silently drops repeats because fancy indexing buffers the read. The canonical answer applies two accumulation passes: bincount counts each index's occurrences and adds them, then np.add.at performs an unbuffered scatter-add of 1 per occurrence, so repeated indices accumulate correctly in both steps.
