## Problem

Write a function solve(z, idx) that takes a 1-D float array z and a 1-D integer array idx of positions into z (positions may REPEAT). Return a new array equal to z but where each position has been incremented by 1.0 once for EVERY time it appears in idx — a position appearing 3 times gains 3.0. Do not modify the input; do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Naive fancy-index += collapses repeats (each position updates once) — count the occurrences first, or use the ufunc method built for unbuffered updates.

</details>

## Reference solution


```python
import numpy as np

def solve(z, idx):
    return z + np.bincount(idx, minlength=len(z))


print(solve(np.ones(5), np.array([0, 1, 1, 3])))
```

## Why this works

out[idx] += 1 applies only ONE increment per distinct position because the reads are buffered — the classic repeated-index bug. np.bincount(idx, minlength=len(z)) counts every occurrence, and adding the counts applies them all; np.add.at(out, idx, 1) is the in-place unbuffered equivalent.
