## Problem

Write a function solve(z, idx) that takes a 1-D float tensor z and a 1-D integer tensor idx of positions into z (positions may REPEAT). Return a new tensor equal to z but where each position has been incremented by 1.0 once for EVERY time it appears in idx — a position appearing 3 times gains 3.0. Do not modify the input; do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Naive fancy-index += collapses repeats (each position updates once) — count the occurrences first, or use the ufunc method built for unbuffered updates.

</details>

## Reference solution


```python
import torch as t

def solve(z, idx):
    return z + t.bincount(idx, minlength=len(z))


print(solve(t.ones(5), t.tensor([0, 1, 1, 3])))
```
