## Problem

Write a function solve(a) that takes a 1-D PyTorch integer tensor and returns a new tensor with every run of consecutive equal entries collapsed to a single occurrence (keeping each run's first element), so no two neighboring entries of the result are equal. Values may reappear later — only ADJACENT duplicates collapse.


<details><summary>💡 Hint (click to reveal)</summary>

Compare the tensor to itself shifted by one — an entry starts a new run exactly when it differs from its left neighbor (and the first entry always counts).

</details>

## Reference solution


```python
import torch as t
def solve(a):
    mask = t.cat([t.tensor([True]), a[1:] != a[:-1]])
    return a[mask]


print(solve(t.tensor([1, 1, 2, 2, 2, 3, 1, 1, 4])))
```
