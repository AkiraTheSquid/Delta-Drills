## Problem

Write a function solve(a) that takes a 1-D numpy integer array and returns a new array with every run of consecutive equal entries collapsed to a single occurrence (keeping each run's first element), so no two neighboring entries of the result are equal. Values may reappear later — only ADJACENT duplicates collapse.


<details><summary>💡 Hint (click to reveal)</summary>

Compare the array to itself shifted by one — an entry starts a new run exactly when it differs from its left neighbor (and the first entry always counts).

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    mask = np.r_[True, a[1:] != a[:-1]]
    return a[mask]


print(solve(np.array([1, 1, 2, 2, 2, 3, 1, 1, 4])))
```

## Why this works

a[1:] != a[:-1] flags positions that differ from their predecessor; prepending True (np.r_ concatenates) keeps the first element, and boolean indexing extracts one representative per run. Unlike np.unique, this preserves order and keeps values that recur in separate runs.
