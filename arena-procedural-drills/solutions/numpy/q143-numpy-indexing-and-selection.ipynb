## Problem

Given a 1D integer array a, return a new 1D array obtained by collapsing every run of consecutive equal entries down to a single occurrence (keeping the first entry of each run), so that no two adjacent entries in the output are equal. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Compare each element to its predecessor to find run boundaries, keeping the first element unconditionally.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.array([1, 1, 2, 2, 2, 3, 1, 1, 4])
    mask = np.r_[True, a[1:] != a[:-1]]
    out = a[mask]
    return out

print(solve())
```

## Why this works

Comparing the array shifted by one (a[1:] != a[:-1]) marks every position that differs from its predecessor, i.e. the start of a new run. Prepending True with np.r_ always keeps the first element, and masking selects one representative per run of equal values.
