## Problem

Write a function solve(tensors) that takes a list of 1-D PyTorch integer tensors (possibly of different lengths) and returns their cartesian product as a 2-D tensor of shape (product of lengths, len(tensors)): every combination of one element per input tensor, ordered so the FIRST tensor varies slowest and the last varies fastest.


<details><summary>💡 Hint (click to reveal)</summary>

meshgrid with indexing='ij' expands every combination into aligned grids — ravel each grid and stack them as columns.

</details>

## Reference solution


```python
import torch as t
def solve(arrays):
    grids = t.meshgrid(*arrays, indexing="ij")
    return t.stack([g.ravel() for g in grids], dim=1)


print(solve([t.tensor([1, 2]), t.tensor([10, 20, 30])]))
```
