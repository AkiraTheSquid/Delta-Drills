## Problem

Given a list of 1D vectors of varying lengths, build the cartesian product — every combination of one element from each vector — and return it as a 2D array of shape (prod_lens, n_vectors). (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

np.meshgrid(*arrays, indexing='ij') gives the broadcasted coordinate grids; stack on the last axis and reshape to (-1, n).

</details>

## Reference solution


```python
import numpy as np

def solve():
    arrays = [np.array([1, 2, 3]), np.array([4, 5]), np.array([6, 7])]
    return np.stack(np.meshgrid(*arrays, indexing='ij'), axis=-1).reshape(-1, len(arrays))

print(solve())
```

## Why this works

np.meshgrid with indexing='ij' produces one grid per input array, each broadcast over the full cartesian shape; stacking them on a new last axis and reshaping to (-1, n_vectors) flattens the grid into one row per combination in lexicographic order.
