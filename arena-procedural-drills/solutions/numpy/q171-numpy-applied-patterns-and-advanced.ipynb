## Problem

Write a function solve(arrays) that takes a list of 1-D numpy integer arrays (possibly of different lengths) and returns their cartesian product as a 2-D array of shape (product of lengths, len(arrays)): every combination of one element per input array, ordered so the FIRST array varies slowest and the last varies fastest.


<details><summary>💡 Hint (click to reveal)</summary>

meshgrid with indexing='ij' expands every combination into aligned grids — ravel each grid and stack them as columns.

</details>

## Reference solution


```python
import numpy as np

def solve(arrays):
    grids = np.meshgrid(*arrays, indexing="ij")
    return np.stack([g.ravel() for g in grids], axis=1)


print(solve([np.array([1, 2]), np.array([10, 20, 30])]))
```

## Why this works

np.meshgrid(*arrays, indexing='ij') produces one grid per input whose value at a multi-index is that input's element in the combination; raveling in C order makes the first axis vary slowest, matching the required lexicographic order, and stacking the flat grids column-wise lists one combination per row.
