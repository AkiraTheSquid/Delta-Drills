## Problem

Write a function solve(z, k, rng) that takes a 2-D numpy array z with all-distinct rows, an integer k, and a numpy random Generator rng. Return a 2-D array of k rows sampled from z uniformly at random WITHOUT replacement using rng. The grader checks the shape, that every returned row exists in z, and that the k sampled rows are all different — any valid sample passes.

## Reference solution


```python
import numpy as np

def solve(z, k, rng):
    idx = rng.choice(z.shape[0], k, replace=False)
    return z[idx]


example = np.arange(12).reshape(4, 3)
print(solve(example, 2, np.random.default_rng(0)))
```

## Why this works

rng.choice(n_rows, k, replace=False) picks k distinct row positions; z[idx] gathers them. Because the sample is random, the tests verify the contract — right shape, rows drawn from z, no repeats (k = all rows forces the whole set) — rather than one specific sample.
