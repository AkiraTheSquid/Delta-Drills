## Problem

You are given Z, a 2D integer array of shape (10, 3), k=4, and idx, a 1D array of 4 distinct row indices chosen uniformly at random from 0..9 without replacement. Return Z[idx], a 2D array of shape (4, 3) consisting of the 4 selected rows of Z in the order given by idx. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Sample k distinct row indices without replacement, then index the array by that vector of row indices.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 10, (10, 3))
    k = 4
    idx = np.random.choice(Z.shape[0], k, replace=False)
    return Z[idx]

print(solve())
```

## Why this works

np.random.choice with replace=False draws k distinct row indices from 0..9. Fancy-indexing Z by that index array selects exactly those rows, in the sampled order, producing a (k, 3) submatrix.
