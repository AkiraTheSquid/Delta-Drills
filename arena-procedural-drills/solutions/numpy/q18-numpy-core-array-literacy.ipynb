## Problem

You are given Z, a 1D float array of length 3 that originally held a mix of finite floats and at least one NaN, but with every NaN entry already replaced by 0.0. Return Z so it can be inspected; the result must be a NaN-free copy of the original with the same shape, dtype, and ordering of the non-NaN values. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

A single conversion function swaps NaN entries for a chosen replacement value while preserving the rest.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([0.0, np.nan, 1.0])
    Z = np.nan_to_num(Z, nan=0.0)
    return Z

print(solve())
```

## Why this works

np.nan_to_num replaces NaN entries with the supplied value (0.0 here) while leaving finite values untouched, producing a NaN-free array of the same shape, dtype, and ordering. The non-NaN values keep their original positions.
