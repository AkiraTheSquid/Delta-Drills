## Problem

Write a function solve(z) that takes a 1-D numpy array of floats that may contain NaN entries (but no infinities) and returns a new array in which every NaN has been replaced by 0.0 while every other value is unchanged. Do not modify the input array.

## Reference solution


```python
import numpy as np

def solve(z):
    return np.nan_to_num(z, nan=0.0)


example = np.array([0.5, np.nan, 1.0])
print(solve(example))
```

## Why this works

np.nan_to_num(z, nan=0.0) returns a copy with NaNs replaced by 0.0. np.where(np.isnan(z), 0.0, z) does the same thing explicitly: NaN never equals anything (even itself), so np.isnan is the only reliable way to find those entries.
