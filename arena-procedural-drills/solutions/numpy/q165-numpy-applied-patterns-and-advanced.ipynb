## Problem

Write a function solve(n, edges) that takes a node count n and a 2-D integer array of shape (m, 2) whose rows [u, v] are undirected edges (no self-loops), and returns the n x n adjacency matrix: entry [u, v] AND entry [v, u] equal 1 for every edge, all other entries 0, dtype integer. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

The two edge-list columns form coordinate arrays — one paired-array assignment sets all (u, v) cells, and swapping the columns sets the mirrors.

</details>

## Reference solution


```python
import numpy as np

def solve(n, edges):
    a = np.zeros((n, n), dtype=int)
    a[edges[:, 0], edges[:, 1]] = 1
    a[edges[:, 1], edges[:, 0]] = 1
    return a


print(solve(3, np.array([[0, 1], [1, 2]])))
```

## Why this works

Indexing a[edges[:, 0], edges[:, 1]] addresses every edge's cell simultaneously (paired coordinate arrays), and repeating with the columns swapped writes the symmetric halves, making A equal to its transpose. Zero-initializing with dtype=int keeps the required integer kind.
