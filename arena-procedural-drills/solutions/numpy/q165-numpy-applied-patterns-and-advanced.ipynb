## Problem

You are given n=6 (the number of nodes), edges, a 2D integer array of shape (4, 2) where each row [u, v] is one undirected edge, and A, a 2D integer array of shape (6, 6) initialized to zero and then updated so that A[u, v] and A[v, u] are both set to 1 for every edge. Return A. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the edge endpoint columns as paired row/column indices to set 1s both ways.

</details>

## Reference solution


```python
import numpy as np

def solve():
    n = 6
    edges = np.array([[0,1],[0,2],[3,5],[2,5]])
    A = np.zeros((n,n), dtype=int)
    A[edges[:,0], edges[:,1]] = 1
    A[edges[:,1], edges[:,0]] = 1
    return A

print(solve())
```

## Why this works

The edge list's two columns are the source and target node indices. Using them as paired fancy indices sets A[u,v]=1 for all edges at once, and repeating with the columns swapped sets A[v,u]=1, producing a symmetric undirected adjacency matrix.
