## Problem

Write a function solve(n, edges) that takes a node count n and a 2-D integer tensor of shape (m, 2) whose rows [u, v] are undirected edges (no self-loops), and returns the n x n adjacency matrix: entry [u, v] AND entry [v, u] equal 1 for every edge, all other entries 0, dtype integer. Do not use Python loops.

## Reference solution


```python
import torch as t

def solve(n, edges):
    a = t.zeros((n, n), dtype=t.int64)
    a[edges[:, 0], edges[:, 1]] = 1
    a[edges[:, 1], edges[:, 0]] = 1
    return a


print(solve(3, t.tensor([[0, 1], [1, 2]])))
```
