## Problem

You are given z, a 1D floating-point array of shape (3,) holding three real-valued logits. Return softmax, a 1D array of the same shape whose entries are non-negative and sum to 1, where each output entry is proportional to the exponential of the corresponding input entry. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Exponentiate the logits (shifting by the max for stability) and divide by their sum.

</details>

## Reference solution


```python
import numpy as np

def solve():
    z = np.array([1.0, 2.0, 3.0])
    e = np.exp(z - z.max())
    return e / e.sum()

print(solve())
```

## Why this works

Softmax exponentiates each logit and normalizes by the total so the outputs are non-negative and sum to 1. Subtracting z.max() before exponentiating is a numerically stable shift that does not change the result, since the constant factor cancels in the normalization.
