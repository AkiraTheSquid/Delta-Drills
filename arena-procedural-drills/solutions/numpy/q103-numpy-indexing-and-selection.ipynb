## Problem

Write a function solve(z) that takes a 1-D numpy float array of logits and returns its softmax: a same-shape array of non-negative entries summing to 1, where entry i is exp(z[i]) divided by the sum of exponentials. Your implementation must be numerically stable — it will be tested with large logits like 1000 where naive exponentiation overflows to inf.


<details><summary>💡 Hint (click to reveal)</summary>

Softmax is unchanged by subtracting any constant from all logits — subtract the maximum so the largest exponent is exp(0).

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    e = np.exp(z - z.max())
    return e / e.sum()


print(solve(np.array([1.0, 2.0, 3.0])))
```

## Why this works

Shifting by z.max() makes every exponent <= 0, so np.exp never overflows, while the ratio is mathematically identical (the shift multiplies numerator and denominator by the same constant). The 1000-scale case makes the naive exp(z)/exp(z).sum() produce inf/inf = NaN and fail.
