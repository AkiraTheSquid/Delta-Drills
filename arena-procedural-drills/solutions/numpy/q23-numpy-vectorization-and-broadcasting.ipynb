## Problem

Given a 3x4 integer matrix Z whose entries are 0 through 11 in row-major order, return a 1D array of length 12 that lists Z's entries in column-major order (reading down each column from top to bottom, columns ordered left to right). (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Flatten the array but request column-major (Fortran) ordering so it reads down columns first.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(12).reshape(3, 4)
    return Z.flatten('F')

print(solve())
```

## Why this works

Flattening with order 'F' (Fortran / column-major) reads the matrix down each column before moving to the next column, which is the requested column-major linearization. The default 'C' order would instead read row by row.
