## Problem

Write a function solve(z) that takes a 2-D PyTorch tensor and returns a 1-D tensor listing all of z's entries in column-major order: read down the first column top to bottom, then the second column, and so on. Note that this is not the default row-major flattening, and unlike numpy, torch's flatten has no order argument — so change the axis order first.


<details><summary>💡 Hint (click to reveal)</summary>

Flattening accepts an order argument — the Fortran convention walks down columns first. Transposing before a default flatten works too.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return z.T.flatten()


example = t.arange(6).reshape(2, 3)
print(solve(example))
```

## Why this works

z.flatten('F') traverses the tensor in Fortran (column-major) order: down each column, then across. z.T.ravel() is the same thing — transposing swaps the roles of rows and columns so the default row-major walk visits columns of the original.
