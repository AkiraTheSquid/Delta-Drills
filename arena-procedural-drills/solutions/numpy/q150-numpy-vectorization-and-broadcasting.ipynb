## Problem

You are given y, a 1D integer label vector of length 5 whose values lie in {0, 1, ..., C-1} where C = y.max() + 1 = 3. Return a 2D integer array of shape (5, 3) whose i-th row is the C-dimensional one-hot indicator of y[i]: a 1 in column y[i] and 0s elsewhere. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Index the rows of an identity matrix of size C by the label vector.

</details>

## Reference solution


```python
import numpy as np

def solve():
    y = np.array([0,2,1,2,0])
    C = y.max() + 1
    I = np.eye(C, dtype=int)
    one_hot = I[y]
    return one_hot

print(solve())
```

## Why this works

The C-by-C identity matrix has exactly one 1 per row, in the position equal to the row index. Fancy-indexing that identity with the label vector y selects row y[i] for each sample, which is precisely the one-hot encoding of each label.
