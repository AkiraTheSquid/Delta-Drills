## Problem

Write a function solve(y) that takes a 1-D integer label vector whose values lie in {0, ..., C-1} where C = y.max() + 1 (every class up to the max may appear), and returns a 2-D INTEGER array of shape (len(y), C) whose row i is the one-hot encoding of y[i]. Unlike a float one-hot, the dtype here must be an integer kind, and C is inferred from y itself.


<details><summary>💡 Hint (click to reveal)</summary>

Infer the width from the largest label, build an integer identity of that size, and gather its rows by label.

</details>

## Reference solution


```python
import numpy as np

def solve(y):
    c = y.max() + 1
    return np.eye(c, dtype=int)[y]


print(solve(np.array([0, 2, 1, 2, 0])))
```

## Why this works

C = y.max() + 1 sizes the encoding to the labels actually present; np.eye(C, dtype=int)[y] gathers integer one-hot rows. The all-zeros case yields width 1, and a single label 3 forces width 4 with the mark in the last column — both pin the inference logic.
