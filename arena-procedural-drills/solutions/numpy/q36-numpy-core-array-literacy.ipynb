## Problem

Write a function solve(a, shape) that takes a 1-D PyTorch tensor a and a tuple shape of three positive integers whose product equals the length of a. Return a 3-D tensor with exactly that shape, containing a's values in the same row-major order (the last axis varies fastest). The values themselves must be unchanged — only how they are organized into dimensions changes.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch can reinterpret a tensor's existing data under a new set of dimensions in a single call, as long as the total element count stays the same.

</details>

## Reference solution


```python
import torch as t

def solve(a, shape):
    return a.reshape(shape)


example = t.arange(12.0)
print(solve(example, (2, 2, 3)))
```

## Why this works

a.reshape(shape) returns a view of the same data organized under the new dimensions, without copying or reordering values. Because PyTorch stores tensors in row-major (C) order by default, the 1-D sequence fills the 3-D tensor with the last axis varying fastest, which is exactly the required layout. It works for any compatible target, including shapes with size-1 axes like (4, 1, 1) and the single-element case (1, 1, 1).
