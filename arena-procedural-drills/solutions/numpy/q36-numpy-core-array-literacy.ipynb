## Problem

Given a 1D array of length 12, return a 3D array with the shape (2, 2, 3) containing the same values. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

A length-12 vector can be reshaped to any shape whose dimensions multiply to 12.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(12).reshape(2, 2, 3)
    return Z.shape

print(solve())
```

## Why this works

reshape reinterprets the 12 contiguous values into the requested (2,2,3) layout without copying data, since 2*2*3 equals 12. The grader checks the resulting shape tuple.
