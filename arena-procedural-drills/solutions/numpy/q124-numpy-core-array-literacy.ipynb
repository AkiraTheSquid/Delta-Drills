## Problem

Write a function solve(z) that takes a 2-D float array with a unique maximum in each row and returns a same-shape float array containing 1.0 at each row's maximum position and 0.0 everywhere else (a per-row argmax one-hot). Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Pair a row-index range with the per-row argmax as a coordinate pair — indexing with two arrays writes one cell per row.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    out = np.zeros_like(z)
    out[np.arange(z.shape[0]), z.argmax(axis=1)] = 1.0
    return out


example = np.array([[0.1, 0.9], [0.8, 0.2]])
print(solve(example))
```

## Why this works

z.argmax(axis=1) locates each row's maximum column; indexing out[np.arange(n_rows), those columns] addresses exactly one cell per row simultaneously, and assigning 1.0 builds the one-hot. This paired-array (coordinate) indexing is the vectorized replacement for a per-row loop.
