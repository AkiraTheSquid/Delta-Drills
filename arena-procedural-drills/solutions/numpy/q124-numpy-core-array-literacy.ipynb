## Problem

You are given Z, a 2D float array of shape (4, 5) of random values in [0, 1), idx, a 1D array of length 4 holding the column index of the largest value in each row of Z, and out, a 2D array of shape (4, 5) that is zero everywhere except for a single 1.0 in row i at column idx[i]. Return out. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Find each row's argmax column, then use paired row/column fancy indexing to drop a 1 at those spots.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(4, 5)
    idx = Z.argmax(axis=1)
    out = np.zeros_like(Z)
    out[np.arange(Z.shape[0]), idx] = 1
    return out

print(solve())
```

## Why this works

argmax(axis=1) gives the winning column per row. Pairing a row-index range with those column indices in a fancy-index assignment places a single 1.0 at each row's maximum, leaving the zero-initialized matrix untouched elsewhere.
