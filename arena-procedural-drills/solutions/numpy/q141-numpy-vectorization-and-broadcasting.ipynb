## Problem

You are given A and B, two 2D arrays of shape (5, 5) with random entries in [0, 1). Return a 1D array of length 5 whose i-th entry equals the i-th diagonal entry of the matrix product A @ B, i.e. the dot product of row i of A with column i of B. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Write the diagonal entry as sum_j A[i,j]*B[j,i] and translate it directly into an einsum subscript string.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

def solve():
    np.random.seed(0)
    A = np.random.uniform(0, 1, (5, 5))
    B = np.random.uniform(0, 1, (5, 5))
    return np.einsum("ij,ji->i", A, B)

print(solve())
```

## Why this works

The diagonal of A @ B is the dot product of row i of A with column i of B, i.e. sum_j A[i,j] * B[j,i]. The einsum subscript "ij,ji->i" expresses exactly that contraction, computing only the diagonal entries without forming the full matrix product.
