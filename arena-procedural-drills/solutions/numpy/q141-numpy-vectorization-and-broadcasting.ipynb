## Problem

Write a function solve(a, b) that takes two square float matrices of the same shape (n, n) and returns a 1-D tensor of length n holding the DIAGONAL of the matrix product a @ b — without computing the full product (entry i is the dot product of row i of a with column i of b). Any approach producing those values passes, but the einsum form is the one worth learning.


<details><summary>💡 Hint (click to reveal)</summary>

Diagonal entry i multiplies row i of the first matrix with COLUMN i of the second — an einsum can say that with a transposed index, summing over j only.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.einsum('ij,ji->i', a, b)


example_a = t.tensor([[1.0, 2.0], [3.0, 4.0]])
example_b = t.tensor([[5.0, 6.0], [7.0, 8.0]])
print(solve(example_a, example_b))
```

## Why this works

einsum('ij,ji->i', a, b) pairs a's row index with b's column index and sums over j, computing only the n diagonal entries — O(n^2) work instead of the O(n^3) full product that t.diag(a @ b) throws away. (a * b.T).sum(axis=1) is the same contraction written with broadcasting.
