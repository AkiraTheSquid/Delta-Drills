## Problem

Write a function solve(x) that takes a 2-D PyTorch tensor x of floats and returns a pair (min, max) containing the smallest and largest element in the whole tensor, in that order. Both returned values must be plain Python numbers, not 0-dimensional tensors. The grader passes the tensor in as a parameter, so do not build or regenerate any data inside solve.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch tensors have reduction methods that, when called with no axis argument, collapse the entire tensor down to one scalar.

## Reference solution


```python
import torch as t

def solve(x):
    return (x.min().item(), x.max().item())


example = t.tensor([[3.5, -2.0, 7.25], [0.0, 9.5, -8.75]])
print(solve(example))
```

## Why this works

Called without an axis, x.min() and x.max() reduce over every element of the tensor regardless of its shape, each returning a single scalar. Packing the two results into a tuple gives the required (min, max) pair. Because the reductions ignore shape, the same two calls handle a single row, a single column, a 1x1 tensor, and ties where every element is equal.
