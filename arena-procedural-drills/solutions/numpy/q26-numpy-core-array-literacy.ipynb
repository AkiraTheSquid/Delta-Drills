## Problem

Write a function solve(x) that takes a 2-D numpy array x of floats and returns a pair (min, max) containing the smallest and largest element in the whole array, in that order. Both returned values should be scalars, not arrays. The grader passes the array in as a parameter, so do not build or regenerate any data inside solve.

numpy arrays have reduction methods that, when called with no axis argument, collapse the entire array down to one scalar.

## Reference solution


```python
import numpy as np

def solve(x):
    return (x.min(), x.max())


example = np.array([[3.5, -2.0, 7.25], [0.0, 9.5, -8.75]])
print(solve(example))
```

## Why this works

Called without an axis, x.min() and x.max() reduce over every element of the array regardless of its shape, each returning a single scalar. Packing the two results into a tuple gives the required (min, max) pair. Because the reductions ignore shape, the same two calls handle a single row, a single column, a 1x1 array, and ties where every element is equal.
