## Problem

Generate a 10x10 array filled with random values and determine the smallest and largest values in that array. (★☆☆)

Two array reduction methods give the global smallest and largest values; return them together.

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((10, 10))
    return Z.min(), Z.max()

print(solve())
```

## Why this works

Array reductions .min() and .max() scan every element and return the global extremes regardless of dimensionality. Returning both as a tuple matches the expected (Zmin, Zmax) pair.
