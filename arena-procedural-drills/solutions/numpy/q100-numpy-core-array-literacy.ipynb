## Problem

Write a function solve(a, b) that takes two 1-D float tensors of equal length, where b may contain zeros, and returns the elementwise quotient a / b — except at positions where b is zero, the output must be exactly 0.0 instead of inf or NaN. No divide-by-zero warnings should be raised.

## Reference solution


```python
import torch as t
def solve(a, b):
    out = t.zeros_like(a)
    nz = b != 0
    out[nz] = a[nz] / b[nz]
    return out


print(solve(t.tensor([1.0, 2.0]), t.tensor([2.0, 0.0])))
```
