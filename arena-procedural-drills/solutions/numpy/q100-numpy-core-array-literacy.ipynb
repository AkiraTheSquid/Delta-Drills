## Problem

Write a function solve(a, b) that takes two 1-D float arrays of equal length, where b may contain zeros, and returns the elementwise quotient a / b — except at positions where b is zero, the output must be exactly 0.0 instead of inf or NaN. No divide-by-zero warnings should be raised.


<details><summary>💡 Hint (click to reveal)</summary>

ufuncs accept a where= mask plus an out= array of defaults — division simply never happens at the masked-off positions.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.divide(a, b, out=np.zeros_like(a), where=b != 0)


print(solve(np.array([1.0, 2.0]), np.array([2.0, 0.0])))
```

## Why this works

np.divide(..., out=np.zeros_like(a), where=b != 0) divides only where b is nonzero and leaves the pre-filled 0.0 elsewhere, so no inf, NaN, or warning is ever produced. Dividing first and patching afterwards would raise the warning the drill forbids.
