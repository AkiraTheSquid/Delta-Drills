## Problem

Given two 1D float arrays a and b of the same length, where b may contain zero entries, compute the elementwise quotient a / b, but wherever the corresponding entry of b is zero, set that output position to 0 instead of producing inf or nan. The result has the same shape and dtype as a. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the out= and where= arguments of np.divide so divisions only happen where the denominator is nonzero, with zeros pre-filled elsewhere.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.array([1.0, 2.0, 0.0])
    b = np.array([2.0, 0.0, 3.0])
    return np.divide(a, b, out=np.zeros_like(a), where=b != 0)

print(solve())
```

## Why this works

np.divide's where= argument restricts the division to positions where b is nonzero, so no inf/nan is ever produced. The out= array is pre-filled with zeros, so every skipped (divide-by-zero) position keeps its 0 value, giving a clean same-shape, same-dtype result.
