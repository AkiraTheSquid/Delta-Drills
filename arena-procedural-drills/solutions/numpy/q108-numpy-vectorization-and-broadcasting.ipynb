## Problem

Given a 2D array X of shape (100, 3) representing 100 observations across 3 columns, return a 3x3 array whose rows give the 25th, 50th, and 75th percentile values for each of the 3 columns (row 0 = 25th percentiles per column, row 1 = medians per column, row 2 = 75th percentiles per column). (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Pass a list of three probabilities to a single quantile call and reduce along the observation axis.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.randn(100, 3)
    return np.quantile(X, [0.25, 0.5, 0.75], axis=0)

print(solve())
```

## Why this works

np.quantile with a list of probabilities computes several quantiles at once, stacking them along a new leading axis. Passing axis=0 reduces over the 100 observations so each column is treated independently, yielding a (3, 3) array whose rows are the 25th, 50th, and 75th percentiles per column.
