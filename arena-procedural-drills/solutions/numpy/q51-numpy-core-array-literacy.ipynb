## Problem

Write a function solve(z) that takes any numpy array and returns the size of its data buffer as the string "<n> bytes", where <n> is the total number of bytes the array's elements occupy (element count times bytes per element). For example, a 10x10 float64 array gives "800 bytes".


<details><summary>💡 Hint (click to reveal)</summary>

Arrays carry their buffer size as an attribute — or multiply the element count by the per-element item size.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return f"{z.nbytes} bytes"


example = np.zeros((10, 10))
print(solve(example))
```

## Why this works

z.nbytes is exactly z.size * z.itemsize: the number of elements times bytes per element. Formatting it with an f-string produces the required '<n> bytes' text; the int32 and uint8 cases verify the count follows the dtype, not just the shape.
