## Problem

Write a function solve(z) that takes any PyTorch tensor and returns the size of its data buffer as the string "<n> bytes", where <n> is the total number of bytes the tensor's elements occupy (element count times bytes per element). For example, a 10x10 float32 tensor gives "400 bytes". Build the number from the element count and the per-element size rather than reading it off a single attribute.

## Reference solution


```python
import torch as t

def solve(z):
    return f"{z.numel() * z.element_size()} bytes"


example = t.zeros((10, 10))
print(solve(example))
```
