## Problem

Given a 10x10 array Z of zeros (float64), report how much memory the array's data buffer occupies in bytes and print that count, formatted as the string "<n> bytes" where <n> is the byte count. Count only the element storage, not Python or array-object overhead. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

The array has an attribute that reports total bytes of its data buffer; format it with the required string.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.zeros((10, 10))
    return "%d bytes" % Z.nbytes


print(solve())
```

## Why this works

An array's element storage equals size * itemsize, which numpy exposes directly as the nbytes attribute. A 10x10 float64 array is 100 elements * 8 bytes = 800 bytes.
