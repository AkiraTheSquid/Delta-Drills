## Problem

Write a function solve(n, d0, d1) that takes a size n and two scalars, and returns an n x n float matrix with d0 at every position of the main diagonal and d1 on both diagonals directly above and below it, all other entries zero — the classic tridiagonal stencil matrix.


<details><summary>💡 Hint (click to reveal)</summary>

fill_diagonal writes the MAIN diagonal of whatever view you hand it — offset views like m[:-1, 1:] turn that into the off-diagonals.

</details>

## Reference solution


```python
import numpy as np

def solve(n, d0, d1):
    m = np.zeros((n, n))
    np.fill_diagonal(m, d0)
    np.fill_diagonal(m[:-1, 1:], d1)
    np.fill_diagonal(m[1:, :-1], d1)
    return m


print(solve(4, 5.0, -1.0))
```

## Why this works

np.fill_diagonal(m, d0) sets the main diagonal in place; the shifted views m[:-1, 1:] and m[1:, :-1] have the super- and sub-diagonal as THEIR main diagonals, so filling those places d1 correctly. Summing scaled np.eye(n, k=...) matrices builds the same stencil.
