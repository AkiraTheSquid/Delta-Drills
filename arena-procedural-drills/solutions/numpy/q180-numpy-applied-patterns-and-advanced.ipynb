## Problem

Construct a square matrix of size n with specified values along the main diagonal and the diagonals directly above and below it, using efficient assignment methods. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

np.fill_diagonal on shifted slices M[:-1, 1:] and M[1:, :-1] targets the upper and lower diagonals.

</details>

## Reference solution


```python
import numpy as np

def solve():
    n = 6
    d0 = 5 * np.ones(n)
    d1 = -1 * np.ones(n - 1)
    M = np.zeros((n, n))
    np.fill_diagonal(M, d0)
    np.fill_diagonal(M[:-1, 1:], d1)
    np.fill_diagonal(M[1:, :-1], d1)
    return M

print(solve())
```

## Why this works

np.fill_diagonal writes the main diagonal of any 2D view in place. Slicing M[:-1, 1:] selects a view whose main diagonal is the superdiagonal of M, and M[1:, :-1] the subdiagonal, so three fill_diagonal calls build the tridiagonal band efficiently.
