## Problem

What is the equivalent of `enumerate` for numpy arrays? Iterate a 3x3 array and return a list of `(index_tuple, value)` pairs. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a dedicated ndenumerate iterator that gives you the multi-index alongside each value.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    return list(np.ndenumerate(Z))

print(solve())
```

## Why this works

np.ndenumerate is the N-dimensional analogue of Python's enumerate: it yields (index_tuple, value) pairs walking the array in C order. Wrapping it in list() materialises all nine pairs for a 3x3 array.
