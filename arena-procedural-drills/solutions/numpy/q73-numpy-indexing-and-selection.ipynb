## Problem

Write a function solve(z) that takes a 2-D numpy array and returns a list of (index, value) pairs covering every element in row-major order, where index is the (row, column) tuple of the element's position — numpy's equivalent of Python's enumerate.


<details><summary>💡 Hint (click to reveal)</summary>

numpy ships an N-dimensional enumerate that yields a position tuple alongside each value.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return [(index, value) for index, value in np.ndenumerate(z)]


example = np.arange(4).reshape(2, 2)
print(solve(example))
```

## Why this works

np.ndenumerate(z) iterates the array in row-major order yielding ((row, col), value) pairs — the N-dimensional analogue of enumerate. Iterating np.ndindex(z.shape) and reading z[idx] builds the identical list.
