## Problem

Given an RGB image represented as a 3D array, convert it to a 2D array representing the grayscale version by applying a specific weighted sum to the color channels. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

A dot product of the channel axis against the luminance weight vector collapses RGB to one grayscale value per pixel.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    I = np.random.randint(0, 256, (4, 4, 3)).astype(np.uint8)
    return np.dot(I[..., :3], [0.299, 0.587, 0.114])

print(solve())
```

## Why this works

Grayscale conversion is a weighted sum over the 3 color channels with the standard luminance coefficients. np.dot contracts the trailing length-3 channel axis of the (H, W, 3) image against the length-3 weight vector, collapsing it to a single (H, W) intensity per pixel.
