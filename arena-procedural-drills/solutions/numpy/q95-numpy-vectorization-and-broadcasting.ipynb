## Problem

Write a function solve(img) that takes an RGB image as a 3-D numpy array of shape (h, w, 3) and returns the 2-D grayscale version of shape (h, w), where each output pixel is the weighted sum 0.299*R + 0.587*G + 0.114*B of that pixel's three channels. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

A dot product between the trailing channel axis and a length-3 weight vector collapses (h, w, 3) to (h, w) in one operation.

</details>

## Reference solution


```python
import numpy as np

def solve(img):
    return img @ np.array([0.299, 0.587, 0.114])


example = np.zeros((2, 2, 3))
example[0, 0] = [255, 255, 255]
print(solve(example))
```

## Why this works

Matrix-multiplying an (h, w, 3) array by a length-3 vector contracts the last axis, computing the weighted channel sum at every pixel simultaneously. Writing it as 0.299*img[...,0] + 0.587*img[...,1] + 0.114*img[...,2] is the same computation spelled out per channel; the equal-channels case (128 everywhere) confirms the weights sum to 1.
