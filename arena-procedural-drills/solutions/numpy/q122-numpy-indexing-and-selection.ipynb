## Problem

Given a 2D array X of shape (100, 4) representing 100 observations across 4 variables (one per column), return the 4x4 symmetric matrix whose entry [i, j] is the Pearson correlation coefficient between column i and column j of X, with 1.0 along the diagonal. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the built-in correlation routine and tell it that each variable is a column, not a row.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.randn(100, 4)
    return np.corrcoef(X, rowvar=False)

print(solve())
```

## Why this works

np.corrcoef computes the Pearson correlation matrix; rowvar=False tells it that variables sit in columns rather than rows. The result is a symmetric 4x4 matrix with ones on the diagonal.
