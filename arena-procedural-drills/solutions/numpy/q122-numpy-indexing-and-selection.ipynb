## Problem

Write a function solve(x) that takes a 2-D float array of shape (n_obs, n_var) — observations in rows, variables in columns — and returns the n_var x n_var matrix of Pearson correlation coefficients between the columns. The diagonal should be (numerically) 1.


<details><summary>💡 Hint (click to reveal)</summary>

numpy's correlation function assumes VARIABLES ARE ROWS by default — one flag flips it to columns.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.corrcoef(x, rowvar=False)


example = np.array([[1.0, 2.0], [2.0, 4.0], [3.0, 6.5]])
print(solve(example))
```

## Why this works

np.corrcoef(x, rowvar=False) treats each column as a variable and each row as an observation, returning the symmetric matrix of pairwise Pearson coefficients with 1s on the diagonal. Perfectly proportional columns give +1, perfectly opposing trends give -1 — the two hand-built cases.
