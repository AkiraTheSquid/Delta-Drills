## Problem

Write a function solve(x) that takes a 2-D float tensor of shape (n_obs, n_var) — observations in rows, variables in columns — and returns the n_var x n_var matrix of Pearson correlation coefficients between the columns. The diagonal should be (numerically) 1.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch's correlation function assumes VARIABLES ARE ROWS by default — one flag flips it to columns.

## Reference solution


```python
import torch as t
def solve(x):
    return t.corrcoef(x.T)


example = t.tensor([[1.0, 2.0], [2.0, 4.0], [3.0, 6.5]])
print(solve(example))
```
