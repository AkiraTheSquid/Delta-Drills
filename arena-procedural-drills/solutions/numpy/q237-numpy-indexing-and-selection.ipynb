## Problem

You are given Z, a 2D floating-point array of shape (4, 4) with entries drawn from the uniform distribution on [0, 1). Return a single scalar equal to the sum of the four main-diagonal entries Z[i, i] for i in 0 through 3. (★☆☆)

There is a dedicated NumPy function for the sum of the main diagonal.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((4, 4))
    return np.trace(Z)

print(solve())
```

## Why this works

np.trace sums the main-diagonal entries Z[i, i], returning the scalar trace of the matrix. This avoids manually indexing each diagonal element.
