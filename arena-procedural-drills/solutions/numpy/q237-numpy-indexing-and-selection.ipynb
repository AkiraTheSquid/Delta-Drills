## Problem

Write a function solve(z) that takes a square 2-D numpy array z of floats and returns the trace: the sum of its main-diagonal entries z[i, i]. The result must be a single scalar number, not an array. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

The entries z[i, i] form the main diagonal — numpy can extract that diagonal (or sum it directly) in one vectorized step.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.trace(z)


example = np.array([[1.0, 2.0], [3.0, 4.0]])
print(solve(example))
```

## Why this works

np.trace(z) sums the main-diagonal entries z[i, i] in a single call, returning a scalar. Equivalently, z.diagonal().sum() first selects the diagonal as a 1-D array and then reduces it. Both work for any square size, including a 1x1 array, and for a matrix whose diagonal is all zeros (trace 0.0).
