## Problem

Convert a Python nested list of two inner lists, each holding two integers, into a 2D NumPy array Z of shape (2, 2) preserving row order, so each inner list becomes one row of Z in the order it appeared. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Pass the nested list straight to the array constructor; the nesting structure becomes the rows.

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([[1, 2], [3, 4]])
    return Z

print(solve())
```

## Why this works

np.array interprets a nested Python list as a 2D array: each inner list becomes a row in order, producing the (2, 2) array directly.
