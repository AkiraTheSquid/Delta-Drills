## Problem

Write a function solve(rows) that takes a Python nested list — a list of equal-length inner lists of integers — and returns a 2-D NumPy array in which each inner list becomes one row, in the same order. The input is a plain Python list, not an array, and your function must return an actual NumPy array, not a list.

## Reference solution


```python
import numpy as np

def solve(rows):
    return np.array(rows)


example = [[1, 2, 3], [4, 5, 6]]
print(solve(example))
```

## Why this works

np.array(rows) copies the nested list into a new ndarray, treating each inner list as one row and inferring the shape (number_of_rows, number_of_columns) plus an integer dtype from the values. Because every inner list has the same length, the result is a proper 2-D array with the rows in their original order. The tests check isinstance and ndim as well as the values, so returning the untouched Python list would fail.
