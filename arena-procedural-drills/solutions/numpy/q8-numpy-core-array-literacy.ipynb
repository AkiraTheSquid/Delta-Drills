## Problem

You are given rng, a NumPy random Generator seeded with a fixed integer so its output stream is deterministic. Produce a 1D array of length 3 containing the next three uniform floats in [0, 1) drawn from this generator. Because the seed is fixed, running the code repeatedly must yield the exact same three numbers. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Construct a Generator with a fixed seed, then ask it for three uniform samples; the fixed seed guarantees reproducibility.

</details>

## Reference solution


```python
import numpy as np

def solve():
    rng = np.random.default_rng(42)
    return rng.random(3)

print(solve())
```

## Why this works

A Generator created with a fixed seed via np.random.default_rng produces a deterministic stream, so rng.random(3) returns the same three uniform floats in [0,1) on every run. The seed fully determines the output sequence.
