## Problem

Write a function solve(rng, n) that takes a torch.Generator rng and a positive integer n, and returns a 1-D tensor of the next n uniform floats in [0, 1) drawn from that generator. Use the generator you are given — do not create a new one or reseed, since the grader checks the exact deterministic stream.

## Reference solution


```python
import torch as t

def solve(rng, n):
    return t.rand(n, generator=rng)


example_rng = t.Generator().manual_seed(42)
print(solve(example_rng, 3))
```
