## Problem

Write a function solve(rng, n) that takes a numpy random Generator rng and a positive integer n, and returns a 1-D array of the next n uniform floats in [0, 1) drawn from that generator. Use the generator you are given — do not create a new one or reseed, since the grader checks the exact deterministic stream.

## Reference solution


```python
import numpy as np

def solve(rng, n):
    return rng.random(n)


example_rng = np.random.default_rng(42)
print(solve(example_rng, 3))
```

## Why this works

rng.random(n) advances the generator's deterministic stream and returns the next n uniform floats as a 1-D array. Because the grader rebuilds a generator with the same seed to compute the expected values, only drawing from the PASSED-IN generator (no reseeding) matches the stream.
