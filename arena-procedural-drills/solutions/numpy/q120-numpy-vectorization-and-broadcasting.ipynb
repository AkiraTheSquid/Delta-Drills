## Problem

Given a 2D array X with some NaN values, replace each NaN in the array with the average of the non-NaN values in its respective column. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Compute NaN-ignoring column means, find the NaN coordinates, and scatter the matching column mean into each hole.

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.randn(5, 4)
    X[::2, 1] = np.nan
    col_means = np.nanmean(X, axis=0)
    idx = np.where(np.isnan(X))
    X[idx] = np.take(col_means, idx[1])
    return X

print(solve())
```

## Why this works

np.nanmean computes each column's mean while ignoring NaNs. np.where locates the NaN positions, and indexing col_means by the column component of those positions fills each NaN with its own column's mean via fancy assignment.
