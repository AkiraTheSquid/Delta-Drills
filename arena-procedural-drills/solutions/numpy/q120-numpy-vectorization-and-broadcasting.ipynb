## Problem

Write a function solve(x) that takes a 2-D numpy float array in which some entries are NaN (but no column is entirely NaN) and returns a new array where every NaN has been replaced by the mean of the NON-NaN values in its own column. Finite entries are unchanged. Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

nanmean ignores the NaNs while averaging; np.where hands you the (row, col) positions of the holes, and the col half of that picks which mean to drop in.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    out = x.copy()
    col_means = np.nanmean(out, axis=0)
    idx = np.where(np.isnan(out))
    out[idx] = np.take(col_means, idx[1])
    return out


example = np.array([[1.0, np.nan], [3.0, 4.0]])
print(solve(example))
```

## Why this works

np.nanmean(x, axis=0) computes per-column means over finite entries only. np.where(np.isnan(x)) lists NaN coordinates; indexing the means by the column coordinates (idx[1]) aligns one fill value per hole, and fancy-index assignment patches them all at once on a copy.
