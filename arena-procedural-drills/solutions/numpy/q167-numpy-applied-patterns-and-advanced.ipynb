## Problem

Given a random 10x10 matrix Z, return all possible 3x3 submatrices that can be formed from it. (★★★)

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve():
    return sliding_window_view(Z, window_shape=(3, 3))
```

## Why this works

sliding_window_view with a 2D window shape produces an array of shape (8, 8, 3, 3): for every valid top-left anchor it yields the corresponding 3x3 submatrix as a strided view, so all overlapping submatrices are gathered without any explicit loops.
