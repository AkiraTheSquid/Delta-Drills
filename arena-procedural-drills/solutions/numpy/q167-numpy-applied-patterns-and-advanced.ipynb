## Problem

Write a function solve(z, k) that takes a 2-D PyTorch tensor z of shape (r, c) and a window size k, and returns ALL k x k contiguous submatrices of z as a 4-D tensor of shape (r - k + 1, c - k + 1, k, k), where entry [i, j] is the block whose top-left corner sits at z[i, j]. Do not use Python loops.

## Reference solution


```python
import torch as t
def solve(z, k):
    return z.unfold(0, k, 1).unfold(1, k, 1)


print(solve(t.arange(9).reshape(3, 3), 2))
```
