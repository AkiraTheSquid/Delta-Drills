## Problem

Write a function solve(z, k) that takes a 2-D numpy array z of shape (r, c) and a window size k, and returns ALL k x k contiguous submatrices of z as a 4-D array of shape (r - k + 1, c - k + 1, k, k), where entry [i, j] is the block whose top-left corner sits at z[i, j]. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

The sliding-window view accepts a 2-D window shape — it slides over both axes at once and never copies.

</details>

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve(z, k):
    return sliding_window_view(z, (k, k))


print(solve(np.arange(9).reshape(3, 3), 2))
```

## Why this works

sliding_window_view(z, (k, k)) reinterprets z's strides so position [i, j] of the result IS the k x k block anchored at z[i, j], producing the (r-k+1, c-k+1, k, k) family of overlapping submatrices with zero copying — the basis of every im2col-style convolution trick.
