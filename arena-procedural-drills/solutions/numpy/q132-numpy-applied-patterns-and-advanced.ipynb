## Problem

Write a function solve(x, i) that takes a 1-D float array x of values and an equal-length integer array i of target positions, and returns a 1-D accumulation array F of length max(i)+1 where F[k] is the SUM of all values x[j] whose target i[j] equals k. Positions receiving no values hold 0.0.


<details><summary>💡 Hint (click to reveal)</summary>

bincount doesn't just count — hand it weights and each occurrence contributes its value instead of 1.

</details>

## Reference solution


```python
import numpy as np

def solve(x, i):
    return np.bincount(i, weights=x)


print(solve(np.array([1.0, 2.0, 3.0]), np.array([0, 2, 0])))
```

## Why this works

np.bincount(i, weights=x) sums x[j] into slot i[j] for every j, returning an array long enough for the largest index; slots never targeted stay 0.0. It's the vectorized scatter-add, the same accumulation np.add.at(F, i, x) performs on a preallocated array.
