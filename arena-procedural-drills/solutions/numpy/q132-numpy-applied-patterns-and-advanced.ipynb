## Problem

How to accumulate elements of a vector (X) to an array (F) based on an index list (I)? (★★★)
Print the result.

bincount takes a weights argument that accumulates each value into the bin named by the matching index.

## Reference solution


```python
import numpy as np

def solve():
    X = [1, 2, 3, 4, 5, 6]
    I = [1, 3, 9, 3, 4, 1]
    F = np.bincount(I, X)
    return F

print(solve())
```

## Why this works

np.bincount with a weights argument adds each weight X[k] into the bin given by I[k], summing contributions that share an index. This is exactly scatter-add: F[i] becomes the total of all X values whose index equals i, with the output length set by the largest index.
