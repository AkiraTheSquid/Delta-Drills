## Problem

Write a function solve(x, i) that takes a 1-D float tensor x of values and an equal-length integer tensor i of target positions, and returns a 1-D accumulation tensor F of length max(i)+1 where F[k] is the SUM of all values x[j] whose target i[j] equals k. Positions receiving no values hold 0.0.


<details><summary>💡 Hint (click to reveal)</summary>

bincount doesn't just count — hand it weights and each occurrence contributes its value instead of 1.

</details>

## Reference solution


```python
import torch as t

def solve(x, i):
    return t.bincount(i, weights=x)


print(solve(t.tensor([1.0, 2.0, 3.0]), t.tensor([0, 2, 0])))
```
