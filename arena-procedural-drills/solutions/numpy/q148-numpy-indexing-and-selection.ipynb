## Problem

Given a 1D array of random integers, return the indices of the first appearance of each unique value in the order they were originally found. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

np.unique can return first-occurrence indices; then sort the indices to restore original discovery order.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    a = np.random.randint(0, 10, 20)
    _, idx = np.unique(a, return_index=True)
    idx_sorted = np.sort(idx)
    return idx_sorted

print(solve())
```

## Why this works

np.unique with return_index=True gives the index of the first occurrence of each unique value, but ordered by value. Sorting those indices ascending recovers the order in which the distinct values were first encountered in the original array.
