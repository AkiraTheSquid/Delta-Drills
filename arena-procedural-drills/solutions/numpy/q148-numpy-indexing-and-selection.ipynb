## Problem

Write a function solve(a) that takes a 1-D numpy integer array and returns the indices of the FIRST occurrence of each distinct value, in ascending index order — the positions where a previously unseen value appears as you scan left to right.


<details><summary>💡 Hint (click to reveal)</summary>

np.unique can report WHERE each distinct value first occurs — but it returns them in value order, not scan order.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    _, idx = np.unique(a, return_index=True)
    return np.sort(idx)


print(solve(np.array([4, 2, 4, 1, 2, 5])))
```

## Why this works

return_index=True makes np.unique hand back the first-occurrence index of each distinct value; those indices come sorted by VALUE, so np.sort reorders them into left-to-right scan order. Each index in the result marks the debut of a new value.
