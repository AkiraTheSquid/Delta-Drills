## Problem

Given two arrays a and b, return a new array containing the larger value from each pair of corresponding elements. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use the element-wise pairwise comparison function, not the reduction that returns a single max.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.array([1, 5, 3])
    b = np.array([2, 4, 3])
    return np.maximum(a, b)

print(solve())
```

## Why this works

np.maximum compares the two arrays element by element and returns the larger value at each position, giving [2,5,3] here. It is the element-wise pairwise max, distinct from the reduction np.max.
