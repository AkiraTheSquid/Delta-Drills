## Problem

Write a function solve(a, b) that takes two 1-D numpy arrays of equal length and returns a new array of the same length, where each element is the larger of the two corresponding elements from a and b. If the two values at a position are equal, keep that shared value. Do not use Python loops or Python's built-in max.

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.maximum(a, b)


a = np.array([1.0, 5.0, 3.0, 2.0])
b = np.array([4.0, 2.0, 3.0, 7.0])
print(solve(a, b))
```

## Why this works

np.maximum(a, b) is a binary elementwise ufunc: it compares the two arrays position by position and keeps the larger value at each slot, returning a new array of the same shape. This is different from np.max, which reduces one array to a single number. Because the comparison is per-position, ties simply keep the shared value, and it works for any length, including length-1 arrays.
