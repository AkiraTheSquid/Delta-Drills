## Problem

Write a function solve(names) that takes a list of field names (for example ['r', 'g', 'b', 'a']) and returns a numpy structured dtype in which each named field, in the given order, is a single unsigned byte (0-255). Calling it with ['r', 'g', 'b', 'a'] should produce a dtype describing one RGBA pixel.


<details><summary>💡 Hint (click to reveal)</summary>

np.dtype accepts a list of (name, type) pairs; numpy's one-byte unsigned integer type goes in the second slot.

</details>

## Reference solution


```python
import numpy as np

def solve(names):
    return np.dtype([(n, np.ubyte) for n in names])


print(solve(['r', 'g', 'b', 'a']))
```

## Why this works

np.dtype([(n, np.ubyte) for n in names]) builds a structured dtype with one unsigned-byte field per name, in order. np.ubyte is an alias of np.uint8, so each field occupies exactly one byte and the whole item size equals the number of fields.
