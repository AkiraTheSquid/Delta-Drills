## Problem

Define a custom NumPy structured dtype named color that describes an RGBA pixel as four unsigned byte fields, in the order r, g, b, a (each one byte, range 0-255). The result should be a dtype object usable to construct or interpret structured arrays where every element holds the four named color channels. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

np.dtype accepts a list of (field_name, field_type) tuples; pick the unsigned single-byte integer type for each channel.

</details>

## Reference solution


```python
import numpy as np

def solve():
    color = np.dtype([('r', np.ubyte), ('g', np.ubyte), ('b', np.ubyte), ('a', np.ubyte)])
    return color

print(solve())
```

## Why this works

A structured dtype is built from a list of (name, type) pairs; np.ubyte is the unsigned 8-bit integer giving the 0-255 range. The resulting dtype object names four contiguous one-byte fields r, g, b, a that can describe each pixel of a structured array.
