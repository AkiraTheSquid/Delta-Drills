## Problem

Write a function solve(x) that takes a 1-D numpy array of floats containing both positive and negative values, and returns a same-shape array in which every entry has been rounded away from zero to the nearest integer. That means positive entries move up to the next whole number, negative entries move down to the next whole number, and entries that are already whole (including zero) stay exactly as they are. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Think of each value as a magnitude and a sign: the magnitude always rounds in the same direction, and the sign can be reattached elementwise afterwards.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.sign(x) * np.ceil(np.abs(x))


example = np.array([1.2, -3.7, 0.5, -2.0])
print(solve(example))
```

## Why this works

Rounding away from zero is a ceiling operation on the magnitude with the original sign restored: np.ceil(np.abs(x)) pushes every magnitude up to the next whole number, and multiplying by np.sign(x) reattaches the sign, with zero mapping to zero automatically. Whole-number entries are unchanged because the ceiling of an integer magnitude is itself. An equivalent vectorized route is np.where(x >= 0, np.ceil(x), np.floor(x)); both preserve the input shape and avoid Python loops.
