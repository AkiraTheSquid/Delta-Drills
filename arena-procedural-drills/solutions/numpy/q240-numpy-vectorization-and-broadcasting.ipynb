## Problem

Given a 1D float array Z of length 10 with both positive and negative entries, return a new length-10 float array where each entry has been replaced by the nearest integer that is at least as far from zero as the original (so positive entries move up to the next integer and negative entries move down to the next more-negative integer, preserving sign). (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Round positives up and negatives down; a sign-conditioned select (or a copysign on the ceiled magnitude) does it elementwise.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.uniform(-10, +10, 10)
    return np.where(Z > 0, np.ceil(Z), np.floor(Z))

print(solve())
```

## Why this works

Rounding away from zero means ceiling positive values and flooring negative ones, which np.where selects per element based on the sign of Z. An equivalent trick is np.copysign(np.ceil(np.abs(Z)), Z), which rounds the magnitude up and restores the original sign.
