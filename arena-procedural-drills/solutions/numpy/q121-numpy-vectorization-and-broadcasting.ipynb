## Problem

Write a function solve(a, b) that takes two 2-D float tensors of identical shape (n, m), treats corresponding rows as vector pairs, and returns a 1-D tensor of length n whose i-th entry is the dot product of row i of a with row i of b. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Multiply elementwise and sum across each row — or say exactly that with an einsum whose output keeps only the row index.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.einsum('ij,ij->i', a, b)


print(solve(t.tensor([[1.0, 2.0]]), t.tensor([[3.0, 4.0]])))
```

## Why this works

einsum('ij,ij->i', a, b) multiplies matching entries and sums over j, leaving one dot product per row — equivalent to (a * b).sum(axis=1) but without materializing the product tensor. Computing t.diag(a @ b.T) gets the same numbers while wastefully forming the whole n x n product.
