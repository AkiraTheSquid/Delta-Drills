## Problem

Given two matrices A and B of shape (5, 4), compute the dot product for each pair of corresponding rows and return the result as a 1D array of length 5. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Write an einsum that shares both indices on the inputs but keeps only the row index in the output.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    A = np.random.rand(5, 4)
    B = np.random.rand(5, 4)
    return np.einsum('ij,ij->i', A, B)

print(solve())
```

## Why this works

The einsum subscript 'ij,ij->i' multiplies the row-aligned matrices elementwise and sums over the column index j, leaving the row index i free. That is exactly a per-row dot product, giving a length-5 vector.
