## Problem

Write a function solve(v) that takes a 1-D numpy array v of floats and returns the index of its smallest element as a plain Python int. If the smallest value appears more than once, return the index of its first occurrence. You may assume v always contains at least one element.

## Reference solution


```python
import numpy as np

def solve(v):
    return int(np.argmin(v))


example = np.array([4.0, 2.0, 7.0, 2.5, 9.0])
print(solve(example))
```

## Why this works

np.argmin(v) scans the vector once and returns the index of the smallest element, and numpy guarantees that on ties the first occurrence wins — exactly the contract asked for here. Wrapping the result in int() converts numpy's integer scalar into a plain Python int. This works for any length, including a single-element vector where the answer is 0.
