## Problem

Write a function solve(v) that takes a 1-D PyTorch tensor v of floats and returns the index of its smallest element as a plain Python int. If the smallest value appears more than once, return the index of its first occurrence. You may assume v always contains at least one element.

## Reference solution


```python
import torch as t

def solve(v):
    return int(t.argmin(v))


example = t.tensor([4.0, 2.0, 7.0, 2.5, 9.0])
print(solve(example))
```

## Why this works

t.argmin(v) scans the vector once and returns the index of the smallest element, and PyTorch guarantees that on ties the first occurrence wins — exactly the contract asked for here. Wrapping the result in int() converts PyTorch's integer scalar into a plain Python int. This works for any length, including a single-element vector where the answer is 0.
