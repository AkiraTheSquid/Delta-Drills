## Problem

Write a function solve(shape) that takes a tuple of three integers, e.g. (3, 3, 3), and returns a new numpy array of exactly that shape filled with uniform random floats drawn from the interval [0, 1). Each element must be an independent draw, so a multi-element result should not be constant. The grader checks the returned array's shape, floating dtype, value range, and that the values actually vary.


<details><summary>💡 Hint (click to reveal)</summary>

numpy's random module has functions that take a shape (or its unpacked dimensions) and return that many independent draws from the unit interval.

## Reference solution


```python
import numpy as np

def solve(shape):
    return np.random.random(shape)


example = (3, 3, 3)
print(solve(example))
```

## Why this works

np.random.random(shape) allocates an array of the requested shape and fills every slot with an independent uniform draw from [0, 1), returning float64 by default. Because the shape tuple is passed straight through, the same one-liner works for any 3-D shape, including the size-1 (1, 1, 1) edge case. Since the output is random, the grader inspects shape, dtype, value range, and non-constancy instead of comparing exact values, so any correct uniform generator (np.random.rand, np.random.random, or a default_rng equivalent) passes.
