## Problem

Build Z, a 3D array of shape (3, 3, 3) filled with random values drawn from the uniform distribution on [0, 1). The result is a single array of that shape and a floating-point dtype. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Pass the desired shape tuple directly to the uniform random generator.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((3, 3, 3))
    return Z

print(solve())
```

## Why this works

np.random.random accepts a shape tuple and fills an array of that shape with uniform draws on [0,1). Seeding to 0 reproduces the exact (3,3,3) array the grader expects.
