## Problem

You are given a, a 3D array of shape (2, 3, 4) with arbitrary (uninitialized) contents, and Z, a new array built to match a's shape and dtype but filled entirely with zeros. Return Z.shape so the student can confirm the new array carries the same (2, 3, 4) shape as the source array. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

There is a *_like constructor that mirrors an array's shape and dtype while filling with zeros.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.empty((2, 3, 4))
    Z = np.zeros_like(a)
    return Z.shape

print(solve())
```

## Why this works

np.zeros_like builds a new zero-filled array matching the source's shape and dtype, so Z inherits the (2,3,4) shape of a. The grader confirms the shape tuple is preserved.
