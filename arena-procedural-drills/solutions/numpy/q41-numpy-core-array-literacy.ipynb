## Problem

Write a function solve(x) that takes a numpy array x of any shape and returns a brand-new array filled entirely with zeros, matching both the shape and the dtype of x. The input may have any number of dimensions and any numeric or boolean dtype, and it must not be modified. The returned array's dtype must be identical to x's dtype, not just a default float array of the right shape.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a one-call constructor that builds a zero-filled array by borrowing another array's shape and dtype — you never have to read .shape or .dtype yourself.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.zeros_like(x)


example = np.array([[3, -1, 4], [1, 5, -9]], dtype=np.int32)
print(solve(example))
```

## Why this works

np.zeros_like(x) allocates a fresh array with exactly the same shape and dtype as x and fills it with zeros, leaving x untouched. Because it copies the dtype from the input, an int32 input yields int32 zeros and a bool input yields False values, which a hand-rolled np.zeros(x.shape) would get wrong (it defaults to float64). It works for any dimensionality, including a single-element 3-D array.
