## Problem

Write a function solve(z) that takes a 2-D float array and returns the array with each ROW rescaled to unit Euclidean length — except rows that are entirely zero, which must remain all zeros (no NaN or inf anywhere, and no warnings raised).


<details><summary>💡 Hint (click to reveal)</summary>

Per-row norms with keepdims broadcast back over the row — and the where=/out= division pattern skips the zero norms instead of dividing by them.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    norms = np.linalg.norm(z, axis=1, keepdims=True)
    return np.divide(z, norms, out=np.zeros_like(z), where=norms != 0)


print(solve(np.array([[3.0, 4.0], [0.0, 0.0]])))
```

## Why this works

np.linalg.norm(z, axis=1, keepdims=True) gives an (n, 1) column of row lengths that broadcasts across each row. Routing the division through where=norms != 0 with a zero-filled out array means zero rows are never divided — they inherit the 0.0 defaults — so no warning, NaN, or inf ever appears.
