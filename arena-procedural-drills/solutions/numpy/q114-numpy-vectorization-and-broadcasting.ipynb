## Problem

Write a function solve(z) that takes a 1-D numpy array with DISTINCT values and returns an integer array of the same length holding each element's 0-based rank: the smallest value gets rank 0, the next gets 1, and so on, with each rank reported at the position of its element. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

argsort answers 'which element goes in each sorted slot' — apply it twice and it flips into 'which slot does each element go to', which is the rank.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return z.argsort().argsort()


print(solve(np.array([10, 40, 20, 30])))
```

## Why this works

The first argsort gives the permutation that sorts z; the second argsort inverts that permutation, yielding for each original position the index it would occupy in sorted order — its 0-based rank. With distinct values this double-argsort trick is exact.
