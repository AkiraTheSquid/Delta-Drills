## Problem

Given a vector Z, return the ranks of its elements in a 0-based format by sorting the elements twice. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Apply argsort twice; the second call inverts the sorting permutation into ranks.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([10, 40, 20, 30])
    return Z.argsort().argsort()

print(solve())
```

## Why this works

The first argsort gives the indices that would sort the array; applying argsort again inverts that permutation, which is exactly each element's 0-based rank. Double argsort is the classic O(n log n) ranking trick.
