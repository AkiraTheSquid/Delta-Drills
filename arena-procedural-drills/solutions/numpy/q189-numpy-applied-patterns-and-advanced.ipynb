## Problem

Given a 2D array Z, return the array rotated by 90 degrees a specified number of times, where the rotation direction is counter-clockwise. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

One CCW quarter-turn equals transposing the array and then flipping it along the first axis.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(12).reshape(3, 4)
    return Z.T[::-1]

print(solve())
```

## Why this works

A single counter-clockwise 90-degree rotation is the transpose followed by reversing the row order: transposing swaps axes, and flipping rows top-to-bottom completes the rotation. This is exactly what np.rot90 does, expressed with basic transpose plus a reversed slice.
