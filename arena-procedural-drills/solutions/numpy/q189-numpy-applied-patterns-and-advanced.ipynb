## Problem

Write a function solve(z, k) that takes a 2-D numpy array and an integer k (possibly 0 or larger than 3), and returns z rotated counterclockwise by k quarter-turns (k=1 is 90 degrees CCW; k is taken modulo 4). Do not modify the input.

## Reference solution


```python
import numpy as np

def solve(z, k):
    return np.rot90(z, k)


print(solve(np.arange(6).reshape(2, 3), 1))
```

## Why this works

np.rot90(z, k) rotates counterclockwise k times, reducing k modulo 4 internally — k=4 is the identity and k=7 equals k=3. Each turn is equivalent to transposing and reversing rows, so the composition never copies more than needed.
