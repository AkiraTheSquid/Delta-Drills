## Problem

You are given Z, a 2D integer array of shape (10, 4) holding the values 0..39 in row-major order, m=3, and idx, a 1D array of 3 distinct row indices drawn uniformly without replacement from 0..9. Return Z[idx], a 2D array of shape (3, 4) containing the chosen rows in the order they appear in idx. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Draw distinct row indices with np.random.choice(..., replace=False) and gather them with fancy indexing.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.arange(40).reshape(10, 4)
    m = 3
    idx = np.random.choice(Z.shape[0], size=m, replace=False)
    return Z[idx]

print(solve())
```

## Why this works

np.random.choice with replace=False draws m distinct row indices uniformly without repeats. Fancy indexing Z[idx] then gathers exactly those rows in the order the indices were drawn, yielding the (3,4) subarray.
