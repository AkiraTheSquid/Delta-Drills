## Problem

Write a function solve(x, idx) that takes a 2-D PyTorch tensor x and a 1-D tensor (or list) of row indices idx, and returns the 2-D tensor made of exactly those rows of x, in the order the indices appear in idx. Indices may repeat, and negative indices count from the end as usual. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

An integer tensor used as an index picks out whole rows at once, in exactly the order the integers are listed.

## Reference solution


```python
import torch as t

def solve(x, idx):
    return x[idx]


example = t.tensor([[10.0, 20.0], [30.0, 40.0], [50.0, 60.0]])
example_idx = t.tensor([2, 0])
print(solve(example, example_idx))
```

## Why this works

Indexing a 2-D tensor with an integer tensor (fancy indexing) gathers whole rows along axis 0: the result has one row per entry of idx, in that order. Repeated entries simply copy the same row again, and negative entries count from the end, just like scalar indexing. The result is a new tensor, so the original x is untouched.
