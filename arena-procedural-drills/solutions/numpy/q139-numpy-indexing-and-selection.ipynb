## Problem

Write a function solve(x, idx) that takes a 2-D numpy array x and a 1-D array (or list) of row indices idx, and returns the 2-D array made of exactly those rows of x, in the order the indices appear in idx. Indices may repeat, and negative indices count from the end as usual. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

An integer array used as an index picks out whole rows at once, in exactly the order the integers are listed.

</details>

## Reference solution


```python
import numpy as np

def solve(x, idx):
    return x[idx]


example = np.array([[10.0, 20.0], [30.0, 40.0], [50.0, 60.0]])
example_idx = np.array([2, 0])
print(solve(example, example_idx))
```

## Why this works

Indexing a 2-D array with an integer array (fancy indexing) gathers whole rows along axis 0: the result has one row per entry of idx, in that order. Repeated entries simply copy the same row again, and negative entries count from the end, just like scalar indexing. The result is a new array, so the original x is untouched.
