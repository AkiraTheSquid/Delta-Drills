## Problem

Write a function solve(b) that takes a 2-D binary (0/1 integer) array of shape (n, u) whose rows encode subsets of a u-element universe, with NO empty rows, and returns the n x n matrix of pairwise JACCARD DISTANCES: entry [i, j] is 1 - |intersection| / |union| of the sets encoded by rows i and j. The diagonal is 0. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcast the row sets against each other — bitwise AND counts the intersection, bitwise OR the union, and the distance is one minus their ratio.

</details>

## Reference solution


```python
import numpy as np

def solve(b):
    inter = (b[:, None, :] & b[None, :, :]).sum(-1)
    union = (b[:, None, :] | b[None, :, :]).sum(-1)
    return 1 - inter / union


print(solve(np.array([[1, 1, 0], [0, 1, 1]])))
```

## Why this works

b[:, None, :] against b[None, :, :] pairs every two rows; & marks shared elements and | marks combined ones, so summing the last axis yields |intersection| and |union| for all pairs at once. Identical sets give distance 0 (the diagonal) and disjoint sets give 1 — the two hand cases.
