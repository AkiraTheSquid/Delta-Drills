## Problem

You are given B, a 2D binary array of shape (6, 8) whose 6 rows are interpreted as subsets of an 8-element universe (a 1 at column k means element k is in the set). Return the 6x6 symmetric Jaccard-distance matrix J where J[i, j] = 1 - |row_i intersect row_j| / |row_i union row_j|; when the union of two rows is empty, substitute the denominator with 1 so the entry stays 1 instead of becoming NaN. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcast the binary rows pairwise and use bitwise AND/OR summed over columns for intersection and union sizes; guard a zero union before dividing.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    B = (np.random.rand(6, 8) > 0.6).astype(int)
    inter = (B[:, None, :] & B[None, :, :]).sum(-1)
    union = (B[:, None, :] | B[None, :, :]).sum(-1)
    J = 1 - inter / np.where(union == 0, 1, union)
    return J

print(solve())
```

## Why this works

Treating rows as bit-sets, broadcasting B against itself and AND-ing/OR-ing along the feature axis gives pairwise intersection and union sizes for all row pairs at once. Jaccard distance is 1 minus intersection over union; np.where guards the empty-union case by substituting 1 in the denominator so the entry stays 1 instead of NaN.
