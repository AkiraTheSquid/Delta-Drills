## Problem

Write a function solve(a) that takes a 1-D numpy integer array and returns its RUN-LENGTH ENCODING as a tuple of two arrays (values, counts): values holds each run's value in order of appearance, counts how long that run is. No Python loops over the elements.


<details><summary>💡 Hint (click to reveal)</summary>

Find where runs START (positions differing from their predecessor, plus position 0) — the values sit there, and the gaps between consecutive starts are the lengths.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    starts = np.r_[0, 1 + np.flatnonzero(a[1:] != a[:-1])]
    values = a[starts]
    counts = np.diff(np.r_[starts, a.size])
    return values, counts


print(solve(np.array([1, 1, 2, 3, 3, 3])))
```

## Why this works

a[1:] != a[:-1] marks run boundaries; flatnonzero + 1 converts them into start positions, with 0 prepended for the first run. Indexing a at the starts recovers each run's value, and differencing the starts (with the array length appended) measures each run's extent — the exact inverse of np.repeat(values, counts).
