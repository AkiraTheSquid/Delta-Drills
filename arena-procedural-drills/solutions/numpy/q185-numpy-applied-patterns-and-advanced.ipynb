## Problem

Write a function solve(a) that takes a 1-D PyTorch integer tensor and returns its RUN-LENGTH ENCODING as a tuple of two tensors (values, counts): values holds each run's value in order of appearance, counts how long that run is. No Python loops over the elements.

## Reference solution


```python
import torch as t
def solve(a):
    changes = t.nonzero(a[1:] != a[:-1], as_tuple=True)[0] + 1
    starts = t.cat([t.zeros(1, dtype=t.int64), changes])
    values = a[starts]
    counts = t.diff(t.cat([starts, t.tensor([a.numel()])]))
    return values, counts


print(solve(t.tensor([1, 1, 2, 3, 3, 3])))
```
