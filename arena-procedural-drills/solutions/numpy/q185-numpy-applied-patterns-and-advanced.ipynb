## Problem

Run-length encode a 1D integer array (values and counts) without Python loops over elements (★★★)
Print the result.

## Reference solution


```python
import numpy as np

def solve():
    starts = np.r_[0, 1 + np.flatnonzero(a[1:] != a[:-1])]
    values = a[starts]
    counts = np.diff(np.r_[starts, a.size])
    return values, counts
```

## Why this works

Run boundaries occur wherever consecutive elements differ, found vectorized by comparing a[1:] to a[:-1]; prepending 0 marks the first run start. The values at those starts are the run values, and differencing the start positions (with the array length appended) gives the run lengths.
