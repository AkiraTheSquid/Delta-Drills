## Problem

Given a 3x3 matrix Z, exchange the positions of the first and last columns and return the modified matrix. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Index columns with a list and assign a reordered column selection back to swap them in one step.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    Z[:, [0, 2]] = Z[:, [2, 0]]
    return Z

print(solve())
```

## Why this works

Column fancy indexing with [0,2] selects the first and last columns; assigning the reversed selection [2,0] swaps them across all rows at once. The slice : keeps every row.
