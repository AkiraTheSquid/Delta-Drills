## Problem

Write a function solve(x) that takes a 2-D PyTorch integer tensor x and returns a new tensor of the same shape in which the first and last columns have exchanged places. Every other column, and the order of the rows, must be unchanged. Do not modify x itself — the grader checks that the original tensor is left intact. The matrix can be any shape, including a single row or a single column.


<details><summary>💡 Hint (click to reveal)</summary>

Indexing with a list of column positions lets you read or write several columns at once — just do it on a copy so the caller's tensor survives.

</details>

## Reference solution


```python
import torch as t

def solve(x):
    out = x.clone()
    out[:, [0, -1]] = out[:, [-1, 0]]
    return out


example = t.arange(9).reshape(3, 3)
print(solve(example))
```

## Why this works

The canonical solution first copies x so the caller's tensor is untouched, then uses fancy indexing: assigning out[:, [-1, 0]] into out[:, [0, -1]] exchanges the two columns in one vectorized step. The right-hand side is evaluated before the assignment happens, so no temporary variable is needed. When the matrix has only one column, positions 0 and -1 name the same column, so the swap is a harmless no-op and the tensor comes back unchanged.
