## Problem

You are given Z, a 2D float array of shape (5, 6) of random values in [0, 1), k=2, idx, a 2D integer array of shape (5, 2) holding the column indices of the two largest values in each row of Z, and mask, a 2D integer array of shape (5, 6) that is zero everywhere except for a 1 at each (row, idx[row]) position. Return mask. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

argpartition gives you the top-k column indices per row cheaply; scatter 1s into a zeros_like array using broadcasted row indices against those columns.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(5, 6)
    k = 2
    idx = np.argpartition(Z, -k, axis=1)[:, -k:]
    mask = np.zeros_like(Z, dtype=int)
    mask[np.arange(Z.shape[0])[:, None], idx] = 1
    return mask

print(solve())
```

## Why this works

np.argpartition with -k along axis 1 puts the k largest values of each row into the last k positions, so slicing [:, -k:] yields the top-k column indices per row (order within the group unspecified, which is fine for a mask). Advanced indexing with a column-vector of row indices broadcast against those column indices sets exactly one 1 per selected cell in a zero matrix.
