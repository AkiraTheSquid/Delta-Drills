## Problem

Write a function solve(z, k) that takes a 2-D float array with distinct values in each row and returns an INTEGER mask of the same shape containing 1 at the positions of each row's k largest values and 0 elsewhere.


<details><summary>💡 Hint (click to reveal)</summary>

argpartition along rows finds the top-k columns cheaply; pair them with a broadcast column of row numbers to scatter the 1s.

</details>

## Reference solution


```python
import numpy as np

def solve(z, k):
    idx = np.argpartition(z, -k, axis=1)[:, -k:]
    mask = np.zeros_like(z, dtype=int)
    mask[np.arange(z.shape[0])[:, None], idx] = 1
    return mask


print(solve(np.array([[0.1, 0.9, 0.5]]), 2))
```

## Why this works

np.argpartition(z, -k, axis=1)[:, -k:] yields each row's top-k column indices (order irrelevant for a mask). Indexing with the (n, 1) row-number column broadcast against the (n, k) column indices addresses all marked cells at once — the row-wise generalization of the single-argmax one-hot.
