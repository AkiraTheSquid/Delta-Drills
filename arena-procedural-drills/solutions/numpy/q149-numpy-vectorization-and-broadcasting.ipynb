## Problem

Write a function solve(a, k) that takes a 1-D numpy integer array and a positive integer k, and returns the k-th order discrete difference of a: applying the adjacent-difference operation k times in succession. The result has length len(a) - k.


<details><summary>💡 Hint (click to reveal)</summary>

The difference function has an order argument — n applications in one call, shrinking the array by one each time.

</details>

## Reference solution


```python
import numpy as np

def solve(a, k):
    return np.diff(a, n=k)


print(solve(np.array([1, 4, 9, 16, 25]), 2))
```

## Why this works

np.diff(a, n=k) iterates the adjacent-difference k times: each pass turns consecutive pairs into their difference and shortens the array by 1. On the squares [1,4,9,16,25] the second difference is the constant 2 — the discrete analogue of a second derivative.
