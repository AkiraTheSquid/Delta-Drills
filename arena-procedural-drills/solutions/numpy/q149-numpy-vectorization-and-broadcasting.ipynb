## Problem

Write a function solve(a, k) that takes a 1-D PyTorch integer tensor and a positive integer k, and returns the k-th order discrete difference of a: applying the adjacent-difference operation k times in succession. The result has length len(a) - k.


<details><summary>💡 Hint (click to reveal)</summary>

The difference function has an order argument — n applications in one call, shrinking the tensor by one each time.

</details>

## Reference solution


```python
import torch as t

def solve(a, k):
    return t.diff(a, n=k)


print(solve(t.tensor([1, 4, 9, 16, 25]), 2))
```

## Why this works

t.diff(a, n=k) iterates the adjacent-difference k times: each pass turns consecutive pairs into their difference and shortens the tensor by 1. On the squares [1,4,9,16,25] the second difference is the constant 2 — the discrete analogue of a second derivative.
