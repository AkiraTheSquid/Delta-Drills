## Problem

You are given a, a 1D integer array of shape (10,) containing the values 0 through 9 in order, and an integer k (here 3). Return d, a 1D array of length 10 - k whose entries are the result of repeatedly taking pairwise consecutive differences of a a total of k times. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

np.diff takes an order argument n that repeats the differencing.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.arange(10)
    k = 3
    d = np.diff(a, n=k)
    return d

print(solve())
```

## Why this works

np.diff with the n argument applies the discrete first-difference operator k times in succession, each pass shortening the array by one. For a linear ramp the first difference is constant, so higher-order differences collapse to zeros.
