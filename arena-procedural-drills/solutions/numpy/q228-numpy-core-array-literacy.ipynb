## Problem

Build a 2D floating-point identity matrix of shape (3, 3): the three entries on the main diagonal equal 1.0 and the six off-diagonal entries equal 0.0. The result is the 3x3 multiplicative identity for matrix multiplication. (★☆☆)
Print the result.

## Reference solution


```python
import numpy as np

def solve():
    Z = np.eye(3)
    return Z

print(solve())
```

## Why this works

np.eye(3) builds the 3x3 identity matrix: ones on the main diagonal and zeros elsewhere, with the default float dtype, which is the multiplicative identity for matrix multiplication.
