## Problem

Write a function solve(n) that takes a positive integer n and returns the n-by-n identity matrix as a 2-D PyTorch tensor of floats: 1.0 on the main diagonal and 0.0 everywhere else. The grader checks both the values and that the result has a floating-point dtype. Your function must work for any n >= 1, including the smallest case n = 1.

## Reference solution


```python
import torch as t

def solve(n):
    return t.eye(n)


example = 3
print(solve(example))
```
