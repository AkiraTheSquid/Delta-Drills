## Problem

Write a function solve(n) that takes a positive integer n and returns the n-by-n identity matrix as a 2-D numpy array of floats: 1.0 on the main diagonal and 0.0 everywhere else. The grader checks both the values and that the result has a floating-point dtype. Your function must work for any n >= 1, including the smallest case n = 1.

## Reference solution


```python
import numpy as np

def solve(n):
    return np.eye(n)


example = 3
print(solve(example))
```

## Why this works

np.eye(n) builds an n-by-n array with 1.0 on the main diagonal and 0.0 everywhere else, returning float64 by default, so it satisfies both the value and dtype checks for every n >= 1. np.identity(n) is an equivalent one-liner, and building np.zeros((n, n)) then setting the diagonal to 1.0 also passes. The last test case additionally inspects dtype.kind to confirm the result is a float array rather than an integer one.
