## Problem

Given an array Z, determine if there are any elements that are not a number (NaN) and print the result as a boolean value. (★☆☆)

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([0.0, np.nan, 1.0])
    return np.isnan(Z).any()

print(solve())
```

## Why this works

np.isnan returns a boolean mask flagging NaN positions, and .any() collapses that mask to True if at least one NaN is present. Here the array contains one NaN, so the result is True.
