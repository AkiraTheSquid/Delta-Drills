## Problem

Write a function solve(x) that takes a 1-D numpy array of floats and returns a single Python boolean: True if any element of x is NaN, and False otherwise. The array may contain negative numbers or infinities, which do not count as NaN. Your function should work for arrays of any length, including a single element.


<details><summary>💡 Hint (click to reveal)</summary>

NaN never compares equal to anything — even itself — so use numpy's elementwise NaN test and then collapse the resulting boolean array to a single yes/no answer.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return bool(np.isnan(x).any())


example = np.array([1.0, np.nan, 3.0])
print(solve(example))
```

## Why this works

np.isnan(x) produces a boolean array marking exactly which entries are NaN; infinities and ordinary floats map to False. Calling .any() on that boolean array collapses it to a single truth value that is True when at least one entry was NaN. Wrapping the result in bool() converts numpy's scalar np.bool_ into a plain Python boolean, matching the return contract.
