## Problem

Write a function solve(x) that takes a 1-D PyTorch tensor of floats and returns a single Python boolean: True if any element of x is NaN, and False otherwise. The tensor may contain negative numbers or infinities, which do not count as NaN. Your function should work for tensors of any length, including a single element.


<details><summary>💡 Hint (click to reveal)</summary>

NaN never compares equal to anything — even itself — so use PyTorch's elementwise NaN test and then collapse the resulting boolean tensor to a single yes/no answer.

</details>

## Reference solution


```python
import torch as t

def solve(x):
    return bool(t.isnan(x).any())


example = t.tensor([1.0, t.nan, 3.0])
print(solve(example))
```
