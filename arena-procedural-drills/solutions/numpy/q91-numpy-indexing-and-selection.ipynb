## Problem

Write a function solve(n, p, rng) that takes a matrix size n, a count p (1 <= p <= n*n), and a numpy random Generator rng. Return an n x n floating-point matrix that is all zeros except for exactly p entries equal to 1.0, at p distinct positions chosen at random using rng. The grader checks the shape, the total count of ones, and that values are only 0.0 and 1.0 — any valid random placement passes.


<details><summary>💡 Hint (click to reveal)</summary>

Choose p distinct FLAT indices out of n*n without replacement, then scatter 1s into them — np.put writes into an array as if it were flat.

</details>

## Reference solution


```python
import numpy as np

def solve(n, p, rng):
    z = np.zeros((n, n))
    flat_positions = rng.choice(n * n, p, replace=False)
    np.put(z, flat_positions, 1.0)
    return z


print(solve(4, 3, np.random.default_rng(0)))
```

## Why this works

rng.choice(n * n, p, replace=False) picks p distinct flat positions; np.put assigns 1.0 at those flat indices in the 2-D array. Because the placement is random, the tests verify the invariants (shape, exactly p ones, only 0/1 values) rather than exact positions — the p = n*n case forces every entry to 1.
