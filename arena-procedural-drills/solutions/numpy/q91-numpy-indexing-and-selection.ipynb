## Problem

Given n = 10 and p = 3, build a 10x10 floating-point matrix that starts as all zeros and then has exactly p of its 100 entries set to 1, with the p positions chosen uniformly at random without repetition from the full grid. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Sample distinct flat indices over range(n*n) without replacement, then write into the array's flattened view.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    n = 10
    p = 3
    Z = np.zeros((n, n))
    np.put(Z, np.random.choice(range(n * n), p, replace=False), 1)
    return Z

print(solve())
```

## Why this works

Treating the n*n grid as a flat sequence of indices lets us pick p distinct positions with np.random.choice(..., replace=False), guaranteeing no repeats. np.put writes 1 into the flattened view of Z at exactly those linear indices, scattering p ones across the 2D grid.
