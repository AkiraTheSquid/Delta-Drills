## Problem

Write a function solve(n, p, rng) that takes a matrix size n, a count p (1 <= p <= n*n), and a PyTorch random Generator rng. Return an n x n floating-point matrix that is all zeros except for exactly p entries equal to 1.0, at p distinct positions chosen at random using rng. The grader checks the shape, the total count of ones, and that values are only 0.0 and 1.0 — any valid random placement passes.

## Reference solution


```python
import torch as t
def solve(n, p, rng):
    z = t.zeros(n * n)
    z[t.randperm(n * n, generator=rng)[:p]] = 1.0
    return z.reshape(n, n)


print(solve(4, 3, t.Generator().manual_seed(0)))
```
