## Problem

Write a function solve(z) that takes a 2-D PyTorch integer tensor and returns, for each row, the column index of the row's largest value — but breaking ties by the LAST occurrence instead of argmax's usual first-occurrence rule.


<details><summary>💡 Hint (click to reveal)</summary>

argmax always reports the FIRST maximum — run it on the reversed rows and convert the index back.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    rev = z.flip(1)
    return z.shape[1] - 1 - rev.argmax(dim=1)


print(solve(t.tensor([[1, 5, 5], [7, 2, 7]])))
```

## Why this works

Reversing each row makes the last occurrence the first one argmax sees; position i in the reversed row corresponds to column c-1-i of the original, hence the index arithmetic. The all-equal row (every entry tied) correctly maps to the final column.
