## Problem

Given a 2D array Z of random integers, return the column index of the largest value in each row, breaking ties by selecting the last occurrence. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

argmax picks the first max; reverse each row so the last max becomes first, then convert the index back.

</details>

## Reference solution


```python
import numpy as np

def solve():
    rev = Z[:, ::-1]
    idx_rev = rev.argmax(axis=1)
    return Z.shape[1] - 1 - idx_rev
```

## Why this works

argmax returns the index of the first maximum, but we want the last occurrence on ties. Reversing each row makes the last occurrence become the first, so argmax on the reversed view finds it; mapping the reversed index back with cols-1-idx gives the original column.
