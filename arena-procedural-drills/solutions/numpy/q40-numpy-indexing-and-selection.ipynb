## Problem

Given a 1D integer sequence with a mix of zero and nonzero entries, return the positions (indices) of every entry that is not zero, in their natural left-to-right order. The result should follow NumPy's convention for reporting nonzero positions of an array. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a function that reports nonzero positions and returns them as a tuple of index arrays.

</details>

## Reference solution


```python
import numpy as np

def solve():
    return np.nonzero([1, 2, 0, 0, 4, 0])

print(solve())
```

## Why this works

np.nonzero returns a tuple of index arrays, one per dimension; for a 1D input that tuple holds a single array of the positions where entries are nonzero, in left-to-right order.
