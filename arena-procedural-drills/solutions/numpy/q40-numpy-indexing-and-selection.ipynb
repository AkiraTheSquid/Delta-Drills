## Problem

Write a function solve(x) that takes a Python list of integers mixing zero and nonzero values. Return the positions of all nonzero entries in left-to-right order. The result must be a tuple containing exactly one element: a 1-D numpy integer array of those indices — the same tuple-of-index-arrays structure numpy uses when reporting where entries are nonzero. Note that x arrives as a plain Python list, not a numpy array.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a single function that reports where entries are nonzero, packaging the result as one index array per dimension.

## Reference solution


```python
import numpy as np

def solve(x):
    return np.nonzero(np.asarray(x))


example = [0, 3, 0, 0, 7, 0, -2]
print(solve(example))
```

## Why this works

np.nonzero converts its input to an array if necessary and returns a tuple with one index array per dimension, so for a 1-D list it yields a tuple holding a single array of the nonzero positions, already in left-to-right order. Negative values are nonzero, so they are included, and an all-zero input produces a tuple containing an empty index array.
