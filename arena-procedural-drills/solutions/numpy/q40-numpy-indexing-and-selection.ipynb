## Problem

Write a function solve(x) that takes a Python list of integers mixing zero and nonzero values. Return the positions of all nonzero entries in left-to-right order. The result must be a tuple containing exactly one element: a 1-D PyTorch integer tensor of those indices — the same tuple-of-index-tensors structure PyTorch uses when reporting where entries are nonzero. Note that x arrives as a plain Python list, not a PyTorch tensor.

## Reference solution


```python
import torch as t
def solve(x):
    return t.nonzero(t.as_tensor(x), as_tuple=True)


example = [0, 3, 0, 0, 7, 0, -2]
print(solve(example))
```
