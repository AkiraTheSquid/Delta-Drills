## Problem

You are given Z, a 1D array of length 10 holding the integers 0 through 9 stored with dtype int32 (32-bit signed integers). Return Z so the printed result shows the sequence [0, 1, ..., 9] with its 32-bit integer dtype intact. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

arange takes a dtype keyword; specify the 32-bit signed integer type explicitly.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(10, dtype=np.int32)
    return Z

print(solve())
```

## Why this works

np.arange(10) produces 0..9, and passing dtype=np.int32 forces the elements to be stored as 32-bit signed integers rather than the platform default integer type.
