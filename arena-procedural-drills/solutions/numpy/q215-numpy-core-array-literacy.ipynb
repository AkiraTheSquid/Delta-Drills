## Problem

Write a function solve(n, dtype_str) that takes a length n and a torch dtype NAME as a string (like 'int32' or 'float64'), and returns a 1-D tensor of the integers 0 through n-1 stored with exactly that dtype. The grader checks both the values and the dtype name. Look the dtype up from the string rather than branching on it.

## Reference solution


```python
import torch as t

def solve(n, dtype_str):
    return t.arange(n, dtype=getattr(t, dtype_str))


print(solve(5, "int32"))
```
