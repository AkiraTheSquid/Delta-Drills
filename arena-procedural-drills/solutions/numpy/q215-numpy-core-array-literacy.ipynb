## Problem

Write a function solve(n, dtype_str) that takes a length n and a numpy dtype NAME as a string (like 'int32' or 'float64'), and returns a 1-D array of the integers 0 through n-1 stored with exactly that dtype. The grader checks both values and dtype name.


<details><summary>💡 Hint (click to reveal)</summary>

np.dtype turns the string name into a dtype object, and arange accepts it directly.

</details>

## Reference solution


```python
import numpy as np

def solve(n, dtype_str):
    return np.arange(n, dtype=np.dtype(dtype_str))


print(solve(5, "int32"))
```

## Why this works

np.dtype('int32') resolves the textual name into the dtype object, and np.arange(n, dtype=...) allocates with it from the start — no post-hoc astype copy. The dtype comparison in the tests is by name string, so exactness matters.
