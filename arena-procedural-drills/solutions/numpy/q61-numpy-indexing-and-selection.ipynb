## Problem

Write a function solve(z, v) that takes a 1-D PyTorch numeric tensor z and a scalar v, and returns the single element of z that is closest to v (minimizing the absolute difference). If two elements are equally close, return the one at the lower index. Return a scalar value, not an tensor.


<details><summary>💡 Hint (click to reveal)</summary>

Turn 'closest' into 'smallest absolute difference', then ask PyTorch for the POSITION of that smallest value.

</details>

## Reference solution


```python
import torch as t

def solve(z, v):
    return z[t.abs(z - v).argmin()]


example = t.tensor([0, 10, 20, 30])
print(solve(example, 12.4))
```

## Why this works

t.abs(z - v) measures each element's distance to v; argmin() finds the index of the smallest distance, and indexing z with it recovers the element itself. argmin returns the first index on ties, matching the lower-index rule.
