## Problem

Write a function solve(z, v) that takes a 1-D numpy numeric array z and a scalar v, and returns the single element of z that is closest to v (minimizing the absolute difference). If two elements are equally close, return the one at the lower index. Return a scalar value, not an array.

Turn 'closest' into 'smallest absolute difference', then ask numpy for the POSITION of that smallest value.

## Reference solution


```python
import numpy as np

def solve(z, v):
    return z[np.abs(z - v).argmin()]


example = np.array([0, 10, 20, 30])
print(solve(example, 12.4))
```

## Why this works

np.abs(z - v) measures each element's distance to v; argmin() finds the index of the smallest distance, and indexing z with it recovers the element itself. argmin returns the first index on ties, matching the lower-index rule.
