## Problem

You are given Z, a 1D integer array of the values 0 through 99, a scalar v drawn uniformly from [0, 100), and index, the position in Z whose value is nearest to v in absolute distance. Return Z[index] so that the printed result is the single element of Z that minimizes |Z - v|. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Compute absolute distances to v, then take argmin to find the closest element.

</details>

## Reference solution


```python
import numpy as np

np.random.seed(0)


def solve():
    np.random.seed(0)
    Z = np.arange(100)
    v = np.random.uniform(0, 100)
    index = np.abs(Z - v).argmin()
    return Z[index]


print(solve())
```

## Why this works

The absolute differences |Z - v| measure distance from each element to v; argmin gives the position of the smallest distance, and indexing returns the nearest value. Seed 0 fixes v so the result is reproducible.
