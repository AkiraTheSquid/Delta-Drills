## Problem

Write a function solve(x, kern) that takes a 2-D float tensor x and a smaller 2-D kernel kern, and returns the VALID cross-correlation of x with kern: output entry [i, j] is the sum of the elementwise product of kern with the window of x anchored at [i, j] (no kernel flipping, windows fully inside x). Output shape is (xr - kr + 1, xc - kc + 1).


```python
%pip install -q numpy torch --index-url https://download.pytorch.org/whl/cpu
```

## Reference solution


```python
import torch as t
def solve(x, kern):
    kh, kw = kern.shape
    windows = x.unfold(0, kh, 1).unfold(1, kw, 1)
    return (windows * kern).sum(dim=(2, 3))


print(solve(t.arange(9, dtype=t.float32).reshape(3, 3), t.ones((2, 2))))
```
