## Problem

Given a 2D array X and a kernel K, compute the result of applying the kernel to the array using a sliding window approach, ensuring that the kernel only overlaps with valid regions of the array. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Use sliding_window_view to expose every valid window, then contract each window against the kernel with tensordot over both kernel axes.

</details>

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve():
    X = np.arange(7 * 7).reshape(7, 7)
    K = np.array([[1, 0, -1], [1, 0, -1], [1, 0, -1]])
    W = sliding_window_view(X, K.shape)
    Y = np.tensordot(W, K, axes=K.ndim)
    return Y

print(solve())
```

## Why this works

sliding_window_view materializes every valid K-sized window of X as the trailing two axes of a 4D view, so valid (no-padding) convolution becomes a contraction of each window against the kernel. np.tensordot with axes=K.ndim sums the element-wise products over both kernel axes, producing the filtered output.
