## Problem

Write a function solve(x, kern) that takes a 2-D float array x and a smaller 2-D kernel kern, and returns the VALID cross-correlation of x with kern: output entry [i, j] is the sum of the elementwise product of kern with the window of x anchored at [i, j] (no kernel flipping, windows fully inside x). Output shape is (xr - kr + 1, xc - kc + 1).


<details><summary>💡 Hint (click to reveal)</summary>

Expose every valid window as trailing axes with a sliding view — the kernel broadcasts against all of them at once, and one double-axis sum finishes each response.

</details>

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve(x, kern):
    windows = sliding_window_view(x, kern.shape)
    return (windows * kern).sum(axis=(2, 3))


print(solve(np.arange(9, dtype=float).reshape(3, 3), np.ones((2, 2))))
```

## Why this works

sliding_window_view(x, kern.shape) has shape (out_h, out_w, kh, kw); multiplying by kern broadcasts the kernel over every window simultaneously, and summing the two window axes produces each output response. This is convolution's sliding-window definition minus the kernel flip — i.e. cross-correlation, the operation deep-learning 'conv' layers actually compute.
