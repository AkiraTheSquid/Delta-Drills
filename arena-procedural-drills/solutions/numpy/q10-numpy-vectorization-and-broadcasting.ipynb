## Problem

You are given Z, a 2D array of shape (5, 5) filled with random values drawn from the uniform distribution on [0, 1). Rescale Z so that its overall mean becomes 0 and its overall standard deviation becomes 1, computing those statistics across all 25 entries (not per-row or per-column). The result should be a (5, 5) array of the same dtype. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the whole-array (no-axis) mean and standard deviation as scalars to center then scale every entry.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((5, 5))
    Z = (Z - np.mean(Z)) / np.std(Z)
    return Z

print(solve())
```

## Why this works

Standardizing across all 25 entries means subtracting the global scalar mean and dividing by the global scalar standard deviation, which np.mean and np.std return when called with no axis. The result has overall mean 0 and standard deviation 1 while preserving shape and dtype.
