## Problem

Write a function solve(z) that takes a numpy array of floats (any shape, not all entries equal) and returns a new array of the same shape rescaled so that its overall mean is 0 and its overall standard deviation is 1. Use the population standard deviation (numpy's default). Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Center first by subtracting the overall mean, then divide by the overall standard deviation.

## Reference solution


```python
import numpy as np

def solve(z):
    return (z - z.mean()) / z.std()


example = np.array([[1.0, 2.0], [3.0, 4.0]])
print(solve(example))
```

## Why this works

Subtracting z.mean() shifts the distribution to mean 0; dividing by z.std() (population std, numpy's default ddof=0) scales the spread to 1. Both reductions are over the WHOLE array — no axis argument — so any shape standardizes the same way.
