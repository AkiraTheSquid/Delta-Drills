## Problem

Write a function solve(start, end) that returns a 1-D PyTorch tensor containing every consecutive integer from start to end, including both endpoints, in increasing order. Both start and end are Python integers, and you may assume start <= end. Note that t.arange's endpoint is exclusive. The result must be an integer tensor, and it must still work when start and end are equal or negative.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch's range-building function treats its stop argument as exclusive, so pick a stop that pushes one past the last value you want.

</details>

## Reference solution


```python
import torch as t

def solve(start, end):
    return t.arange(start, end + 1)


example_start, example_end = 3, 8
print(solve(example_start, example_end))
```

## Why this works

t.arange(start, stop) produces consecutive integers from start up to but not including stop, so passing end + 1 as the stop makes end itself the last element. Because both arguments are Python integers, the result is an integer tensor without any extra casting. The same call handles negative ranges and the start == end case, where it returns a length-1 tensor containing just that value.
