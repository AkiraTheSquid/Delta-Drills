## Problem

Build a 1D integer vector Z whose entries are the consecutive integers starting at 10 and ending at 49 inclusive, listed in strictly increasing order. The result is a length-40 vector. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

arange takes start and stop; remember the stop is exclusive when choosing the upper bound.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(10, 50)
    return Z

print(solve())
```

## Why this works

np.arange(10, 50) produces consecutive integers from the start (inclusive) up to but not including the stop, giving 10..49, a length-40 increasing vector.
