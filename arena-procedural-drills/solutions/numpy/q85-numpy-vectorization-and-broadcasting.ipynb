## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns a 2-D array containing only the rows of z that have at least one nonzero entry, in their original order. If every row is all zeros, return an empty array with zero rows.


<details><summary>💡 Hint (click to reveal)</summary>

Build a boolean per-row test — 'is this whole row zero?' — then invert it and use it as a row mask.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return z[~(z == 0).all(axis=1)]


example = np.array([[0, 0], [3, 0], [0, 0]])
print(solve(example))
```

## Why this works

(z == 0).all(axis=1) marks rows that are entirely zero; negating with ~ selects the keepers, and boolean row indexing preserves their order. z[z.any(axis=1)] is the same test phrased positively.
