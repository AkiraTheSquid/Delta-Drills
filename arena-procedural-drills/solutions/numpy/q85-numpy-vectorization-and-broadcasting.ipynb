## Problem

You are given Z, a 2D integer array with 4 rows and 3 columns where some rows are entirely zero. Return a 2D array containing only the rows of Z that have at least one nonzero entry, preserving the original column count and the original ordering of the surviving rows. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Build a per-row boolean mask of all-zero rows, negate it, and use it to index the rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([[0, 0, 0], [1, 0, 2], [0, 0, 0], [3, 4, 5]])
    return Z[~(Z == 0).all(axis=1)]

print(solve())
```

## Why this works

(Z == 0).all(axis=1) is True for rows that are entirely zero; negating it with ~ gives a boolean mask of rows with at least one nonzero entry, and boolean row-indexing keeps exactly those rows in their original order.
