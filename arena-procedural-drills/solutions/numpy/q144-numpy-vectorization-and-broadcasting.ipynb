## Problem

Given a 2D array Z with shape (6, 5) and a vector v of length 5, calculate the result of multiplying each row of Z by the vector v, returning a 1D array of length 6. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

A single matrix-vector product (Z @ v) already gives the dot of every row with v.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(6, 5)
    v = np.random.rand(5)
    dots = Z @ v
    return dots

print(solve())
```

## Why this works

Matrix-vector multiplication Z @ v computes, for each row of Z, the dot product with v, producing one scalar per row. With Z shape (6,5) and v shape (5,), the result is the length-6 vector of row-wise dot products, no loop required.
