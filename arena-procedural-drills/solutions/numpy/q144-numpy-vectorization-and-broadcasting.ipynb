## Problem

Write a function solve(z, v) that takes a 2-D float array z of shape (n, m) and a vector v of length m, and returns a 1-D array of length n whose i-th entry is the dot product of row i of z with v. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Matrix-vector multiplication IS the per-row dot product — one operator does the whole job.

</details>

## Reference solution


```python
import numpy as np

def solve(z, v):
    return z @ v


print(solve(np.array([[1.0, 2.0], [3.0, 4.0]]), np.array([10.0, 1.0])))
```

## Why this works

z @ v contracts z's columns against v, producing exactly one dot product per row. The broadcast form (z * v).sum(axis=1) computes the same thing while materializing the intermediate product array that @ avoids.
