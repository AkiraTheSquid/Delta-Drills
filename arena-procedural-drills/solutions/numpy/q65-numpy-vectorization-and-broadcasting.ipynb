## Problem

Given Z, a 1D float vector of length 10 whose entries are all 0.0, mark Z so that any subsequent attempt to assign into it (for example writing Z[0] = 1) raises a ValueError reporting that the assignment destination is read-only. The shape and dtype of Z must stay unchanged. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Arrays expose a writeable flag in .flags that can be turned off.

</details>


```python
%pip install -q numpy
```

## Reference solution


```python
import numpy as np


def solve():
    Z = np.zeros(10)
    Z.flags.writeable = False
    return Z


print(solve())
```

## Why this works

Setting the writeable flag to False marks the array read-only without changing its shape or dtype, so any later assignment like Z[0]=1 raises a ValueError. The array still holds ten zeros, matching the expected value.
