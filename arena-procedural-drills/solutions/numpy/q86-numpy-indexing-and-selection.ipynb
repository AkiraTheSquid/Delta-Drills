## Problem

Write a function solve(z) that takes a 2-D numpy integer array used as a mask (zeros and nonzeros) and returns a 2-D array of shape (k, 2) listing the (row, column) coordinates of every nonzero entry, in row-major scan order. If there are no nonzero entries, the result has zero rows.


<details><summary>💡 Hint (click to reveal)</summary>

One numpy function returns nonzero positions already grouped as one (row, col) pair per hit — a cousin of np.nonzero with a different packaging.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.argwhere(z)


example = np.array([[0, 1], [1, 0]])
print(solve(example))
```

## Why this works

np.argwhere(z) returns one row per nonzero element, each row holding that element's index tuple — exactly the (k, 2) coordinate list. np.nonzero gives the same information as a tuple of separate axis arrays; stacking them column-wise reproduces argwhere.
