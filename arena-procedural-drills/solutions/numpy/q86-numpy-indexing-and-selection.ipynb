## Problem

You are given Z, a 2D integer array of shape (4, 5) that acts as a binary mask: most entries are 0 and a few are 1. Return coords, a 2D array of shape (k, 2) where k is the number of nonzero entries in Z; each row of coords gives the (row, column) index of one nonzero entry. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a single NumPy function that returns the coordinates of nonzero entries as rows of an array.

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros((4, 5), dtype=int)
    Z[1, 2] = 1
    Z[3, 4] = 1
    return np.argwhere(Z)

print(solve())
```

## Why this works

np.argwhere returns the indices of all nonzero entries as a (k, 2) array, one (row, column) pair per nonzero element, in row-major scan order. This is the standard way to recover coordinates from a binary mask.
