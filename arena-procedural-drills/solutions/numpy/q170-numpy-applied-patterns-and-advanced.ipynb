## Problem

Write a function solve(z) that takes a 2-D float array containing NaN entries (no column entirely NaN) and returns a new array with every NaN replaced by the mean of the NON-NaN values in its own COLUMN, computed with keepdims so the statistics broadcast cleanly. Finite entries stay unchanged; the input must not be modified.


<details><summary>💡 Hint (click to reveal)</summary>

nanmean with keepdims gives a (1, cols) row of fill values — one np.where selects between it and the original everywhere at once.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    means = np.nanmean(z, axis=0, keepdims=True)
    return np.where(np.isnan(z), means, z)


print(solve(np.array([[1.0, np.nan], [3.0, 6.0]])))
```

## Why this works

np.nanmean(z, axis=0, keepdims=True) shapes the column means as (1, cols) so they broadcast over every row; np.where(np.isnan(z), means, z) then swaps each NaN for its column's mean in a single vectorized select, returning a new array and leaving z intact.
