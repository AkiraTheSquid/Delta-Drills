## Problem

Given a 2D array Z with some NaN values, replace each NaN in the array with the mean of its respective column. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Use np.nanmean per column, locate the NaNs with np.where(np.isnan(...)), then scatter the matching column means back in.

</details>

## Reference solution


```python
import numpy as np

def solve():
    out = Z.copy()
    col_means = np.nanmean(out, axis=0, keepdims=True)
    inds = np.where(np.isnan(out))
    out[inds] = np.take(col_means, inds[1])
    return out
```

## Why this works

np.nanmean computes each column's mean while ignoring NaNs. np.where(np.isnan(Z)) gives the row/column indices of the missing entries, and indexing col_means by those column indices selects the correct replacement value for each NaN position.
