## Problem

Given a 3x3 matrix Z, calculate the square root of the sum of the absolute squares of its elements. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a single linear-algebra function whose default behavior on a matrix gives the square-root-of-sum-of-squares you need.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((3, 3))
    return np.linalg.norm(Z)

print(solve())
```

## Why this works

The square root of the sum of squared absolute values of all entries is exactly the Frobenius norm of the matrix. np.linalg.norm with no ord argument on a 2D array computes that Frobenius norm directly.
