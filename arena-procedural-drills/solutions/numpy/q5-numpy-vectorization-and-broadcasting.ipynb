## Problem

Write a function solve(z) that takes a 2-D PyTorch tensor of floats and returns its Frobenius norm: the square root of the sum of the squares of all its entries, as a single scalar. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch's linear-algebra module has a one-call norm, or build it yourself from squaring, summing, and a square root.

## Reference solution


```python
import torch as t

def solve(z):
    return t.linalg.norm(z)


example = t.tensor([[3.0, 4.0], [0.0, 0.0]])
print(solve(example))
```

## Why this works

t.linalg.norm on a 2-D tensor defaults to the Frobenius norm — sqrt of the sum of squared entries. Squaring elementwise, summing everything, and taking t.sqrt computes the identical value, and the tests accept either route.
