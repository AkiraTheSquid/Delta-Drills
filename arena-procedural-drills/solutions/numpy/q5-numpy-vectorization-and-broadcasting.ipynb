## Problem

Write a function solve(z) that takes a 2-D numpy array of floats and returns its Frobenius norm: the square root of the sum of the squares of all its entries, as a single scalar. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.linalg.norm(z)


example = np.array([[3.0, 4.0], [0.0, 0.0]])
print(solve(example))
```

## Why this works

np.linalg.norm on a 2-D array defaults to the Frobenius norm — sqrt of the sum of squared entries. Squaring elementwise, summing everything, and taking np.sqrt computes the identical value, and the tests accept either route.
