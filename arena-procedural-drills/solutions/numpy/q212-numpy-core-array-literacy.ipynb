## Problem

Write a function solve(rows, cols) that takes two positive integers and returns a rows x cols PyTorch tensor of BOOLEAN dtype in which every entry is True. The dtype must be torch.bool — an integer or float tensor of ones does not count.

## Reference solution


```python
import torch as t

def solve(rows, cols):
    return t.ones((rows, cols), dtype=t.bool)


print(solve(2, 3))
```
