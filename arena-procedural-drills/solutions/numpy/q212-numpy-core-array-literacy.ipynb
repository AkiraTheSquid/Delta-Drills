## Problem

Write a function solve(rows, cols) that takes two positive integers and returns a rows x cols numpy array of BOOLEAN dtype in which every entry is True. The dtype must be bool — an integer or float array of ones does not count.


<details><summary>💡 Hint (click to reveal)</summary>

The ones constructor takes a dtype — and for booleans, one IS True. np.full works too.

</details>

## Reference solution


```python
import numpy as np

def solve(rows, cols):
    return np.ones((rows, cols), dtype=bool)


print(solve(2, 3))
```

## Why this works

np.ones((rows, cols), dtype=bool) builds the array directly with True entries since 1 casts to True; np.full((rows, cols), True) states the same intent explicitly. The dtype check rejects numeric lookalikes.
