## Problem

Build a 2D array of shape (3, 3) with boolean dtype in which every one of the 9 entries holds the value True. The result must use the bool dtype (not integer or float) and contain no False entries anywhere in the matrix. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

The ones constructor accepts a dtype argument; pick the boolean dtype so the fill value is True.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.ones((3, 3), dtype=bool)
    return Z

print(solve())
```

## Why this works

np.ones with dtype=bool fills the array with the boolean truth value (True), giving a 3x3 matrix where every entry is True and the dtype is genuinely bool rather than int or float.
