## Problem

Write a function solve(x) that takes a 1-D PyTorch tensor x of non-negative integers and returns the value that occurs most frequently in x. If several values are tied for the highest count, return the smallest of those values. Return the result as a plain Python int. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Build a tensor of occurrence counts indexed by value, then find the position of the largest count — first-match behavior on ties is exactly what you need.

## Reference solution


```python
import torch as t

def solve(x):
    return int(t.bincount(x).argmax())


example = t.tensor([3, 1, 3, 2, 3, 1])
print(solve(example))
```

## Why this works

t.bincount(x) returns a tensor whose i-th entry counts how many times the value i appears in x, so its indices are the values themselves. Calling argmax() on those counts returns the index of the largest count, and because argmax returns the first maximum it encounters, ties are automatically resolved in favor of the smallest value. Wrapping the result in int() turns the PyTorch scalar into a plain Python integer.
