## Problem

Given a 1D integer array Z of length 50 whose entries are all non-negative, return the single integer value that occurs most frequently in Z. If two or more values share the highest frequency, return the smallest such value.
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Tally counts per value with bincount, then take the argmax — its tie-breaking already favors the smallest value.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 10, 50)
    return np.bincount(Z).argmax()

print(solve())
```

## Why this works

bincount tallies the occurrences of each non-negative integer value, producing a frequency array indexed by value. argmax returns the index of the largest count, and because argmax returns the first maximum it naturally breaks ties toward the smallest value.
