## Problem

Write a function solve(faces) that takes an (m, 3) integer array of triangles, each row the three vertex ids of one triangle, and returns the set of UNIQUE UNDIRECTED EDGES used by the triangles: a (k, 2) array where each row is an edge with its two vertex ids in ascending order, rows sorted lexicographically, no duplicates.


<details><summary>💡 Hint (click to reveal)</summary>

Each triangle contributes the vertex pairs (0,1), (1,2), (2,0) — list them all, canonicalize each pair by sorting its two ids, then deduplicate rows.

</details>

## Reference solution


```python
import numpy as np

def solve(faces):
    edges = faces[:, [0, 1, 1, 2, 2, 0]].reshape(-1, 2)
    edges = np.sort(edges, axis=1)
    return np.unique(edges, axis=0)


print(solve(np.array([[0, 1, 2]])))
```

## Why this works

Fancy-indexing the columns [0,1,1,2,2,0] and reshaping to pairs enumerates all 3m directed edges; sorting within each row makes (2,1) and (1,2) the same canonical edge, and np.unique(axis=0) removes duplicates while sorting lexicographically — edges shared between adjacent triangles collapse to one.
