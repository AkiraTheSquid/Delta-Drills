## Problem

Given a set of 10 triplets representing triangles defined by shared vertices, determine the unique pairs of vertices that form the edges of these triangles. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Duplicate each vertex and roll to form consecutive (a,b) edge pairs, sort each pair so direction doesn't matter, then take unique rows via a structured-dtype view.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    faces = np.random.randint(0, 100, (10, 3))
    F = np.roll(faces.repeat(2, axis=1), -1, axis=1)
    F = F.reshape(len(F) * 3, 2)
    F = np.sort(F, axis=1)
    G = F.view(dtype=[('p0', F.dtype), ('p1', F.dtype)])
    G = np.unique(G)
    return G

print(solve())
```

## Why this works

Each triangle's three vertices yield three edges; repeating each vertex and rolling by one pairs consecutive vertices so reshaping to (-1, 2) lists every directed edge. Sorting each pair makes an edge orientation-independent, and viewing the two columns as a structured dtype lets np.unique collapse duplicate undirected edges into the unique set of line segments.
