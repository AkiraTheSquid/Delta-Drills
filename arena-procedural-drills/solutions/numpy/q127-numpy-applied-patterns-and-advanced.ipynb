## Problem

Consider a random vector with shape (10,2) representing coordinates, find point by point distances (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Turn each coordinate column into a row vector and subtract it from its own transpose so broadcasting builds every pairwise difference at once.

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((10, 2))
    X, Y = np.atleast_2d(Z[:, 0], Z[:, 1])
    D = np.sqrt((X - X.T)**2 + (Y - Y.T)**2)
    return D

Z = np.random.random((10, 2))
X, Y = np.atleast_2d(Z[:, 0], Z[:, 1])
D = np.sqrt((X - X.T)**2 + (Y - Y.T)**2)
print(D)
```

## Why this works

Pairwise distances come from broadcasting each coordinate against its own transpose. Promoting the x and y columns to row vectors with atleast_2d lets X - X.T form the full matrix of x-differences (and likewise for y); summing the squares and taking the square root yields the symmetric (10,10) Euclidean distance matrix with a zero diagonal.
