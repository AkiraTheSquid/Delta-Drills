## Problem

Write a function solve(z) that takes a 2-D array of shape (n, 2) whose rows are 2-D points, and returns the n x n matrix of pairwise Euclidean distances: entry [i, j] is the distance between point i and point j. The diagonal is 0. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Insert one new axis on each operand so the points broadcast against each other — the difference array then has shape (n, n, 2).

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    diff = z[:, None, :] - z[None, :, :]
    return np.sqrt((diff ** 2).sum(axis=2))


example = np.array([[0.0, 0.0], [3.0, 4.0]])
print(solve(example))
```

## Why this works

z[:, None, :] against z[None, :, :] broadcasts to all (i, j) point pairs at once; squaring, summing the coordinate axis, and square-rooting gives every distance simultaneously. The diagonal is zero because each point's difference with itself vanishes.
