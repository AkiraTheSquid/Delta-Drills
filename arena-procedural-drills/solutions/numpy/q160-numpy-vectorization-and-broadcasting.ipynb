## Problem

Write a function solve(v) that takes a 1-D PyTorch tensor v of length n and returns an n x n lower-triangular matrix whose row i contains the first i+1 entries of v followed by zeros: L[i, j] = v[j] when j <= i and 0 otherwise. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Stack n copies of v as rows, then let the lower-triangle helper zero out everything past each row's prefix.

</details>

## Reference solution


```python
import torch as t

def solve(v):
    n = v.numel()
    return t.tril(t.tile(v, (n, 1)))


print(solve(t.tensor([1, 2, 3])))
```

## Why this works

t.tile(v, (n, 1)) repeats v as every row; t.tril zeroes entries above the main diagonal, leaving row i with exactly its first i+1 values. Note row content is v's PREFIX (v[0]..v[i]), not v shifted — the fourth case with mixed values pins that.
