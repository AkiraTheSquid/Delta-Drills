## Problem

Construct a lower-triangular matrix whose i-th row contains the first i+1 elements of v (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Tile the vector into n rows, then keep the lower triangle.

</details>

## Reference solution


```python
import numpy as np

def solve():
    v = np.arange(1, 6)
    n = v.size
    L = np.tril(np.tile(v, (n,1)))
    return L

print(solve())
```

## Why this works

np.tile stacks v into n identical rows, giving a matrix where every row is the full vector. np.tril then zeros everything strictly above the main diagonal, leaving row i holding only its first i+1 entries.
