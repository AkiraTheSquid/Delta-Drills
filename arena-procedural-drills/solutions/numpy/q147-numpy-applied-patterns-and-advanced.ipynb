## Problem

You are given Z, a 1D float array of 1000 samples drawn from a standard normal distribution, the scalars lo and hi equal to the 5th and 95th percentiles of Z respectively, and Z_wins, a 1D array of length 1000 obtained by replacing every entry of Z below lo with lo and every entry above hi with hi. Return the tuple (Z_wins.min(), Z_wins.max()). (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Get the two percentile cut points, then clamp the array between them.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randn(1000)
    lo, hi = np.percentile(Z, [5, 95])
    Z_wins = np.clip(Z, lo, hi)
    return Z_wins.min(), Z_wins.max()

print(solve())
```

## Why this works

Winsorizing means pulling extreme values in to fixed cut points. Compute the 5th and 95th percentiles with np.percentile, then np.clip bounds every entry into [lo, hi]. After clipping the min becomes lo and the max becomes hi.
