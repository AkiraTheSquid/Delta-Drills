## Problem

Write a function solve(z, lo_pct, hi_pct) that takes a 1-D float array and two percentile levels (e.g. 5 and 95), and returns the WINSORIZED array: values below the lo_pct percentile are raised to it, values above the hi_pct percentile are lowered to it, everything else is unchanged. Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Two percentiles give the fences; one clip call pushes the outliers onto them.

</details>

## Reference solution


```python
import numpy as np

def solve(z, lo_pct, hi_pct):
    lo, hi = np.percentile(z, [lo_pct, hi_pct])
    return np.clip(z, lo, hi)


print(solve(np.array([0.0, 1.0, 2.0, 3.0, 100.0]), 10, 90))
```

## Why this works

np.percentile(z, [lo_pct, hi_pct]) computes both cutoffs in one call, and np.clip(z, lo, hi) maps everything outside the fences onto them while leaving interior values alone — the standard winsorization used to tame outliers. Percentiles 0 and 100 are the min and max, so that case is the identity.
