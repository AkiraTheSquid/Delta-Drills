## Problem

Write a function solve(a, b) that takes two float arrays of the same shape and computes ((a + b) * (-a / 2)) WITHOUT allocating any new arrays: use the out= argument of numpy's ufuncs (np.add, np.divide, np.negative, np.multiply) so every intermediate lands in a or b. The final result must end up stored in a, and the function returns a. b may be used as scratch space.


<details><summary>💡 Hint (click to reveal)</summary>

Every numpy ufunc accepts out= to write its result into an existing array — sequence the four operations so nothing you still need gets overwritten.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    np.add(a, b, out=b)
    np.divide(a, 2, out=a)
    np.negative(a, out=a)
    np.multiply(a, b, out=a)
    return a


example_a = np.ones(3)
example_b = np.ones(3) * 2
print(solve(example_a, example_b))
```

## Why this works

The ordering matters: a+b is stored into b first (a is still needed), then a is halved and negated in place, and the final multiply writes into a. The grader checks both the value and that the returned object IS the original a, so a fresh-array solution fails the identity check.
