## Problem

Write a function solve(a, b) that takes two float tensors of the same shape and computes ((a + b) * (-a / 2)) WITHOUT allocating any new tensors: use PyTorch's in-place methods (add_, div_, neg_, mul_) so every intermediate lands in a or b. The final result must end up stored in a, and the function returns a. b may be used as scratch space.

## Reference solution


```python
import torch as t
def solve(a, b):
    b.add_(a)
    a.div_(2)
    a.neg_()
    a.mul_(b)
    return a


example_a = t.ones(3)
example_b = t.ones(3) * 2
print(solve(example_a, example_b))
```
