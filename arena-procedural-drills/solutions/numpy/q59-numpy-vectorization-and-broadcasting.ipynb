## Problem

Given two length-3 vectors A (all ones) and B (all twos), compute the value of ((A + B) * (-A / 2)) and store the final result in A, performing each intermediate step so that no new arrays are allocated and only A and B hold the working values. The final A should be a length-3 vector. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Every numpy ufunc accepts an out= target; chain add, divide, negative, multiply writing into A and B.

</details>

## Reference solution


```python
import numpy as np


def solve():
    A = np.ones(3) * 1
    B = np.ones(3) * 2
    np.add(A, B, out=B)
    np.divide(A, 2, out=A)
    np.negative(A, out=A)
    np.multiply(A, B, out=A)
    return A


print(solve())
```

## Why this works

Using the out= argument of each ufunc writes results back into existing buffers instead of allocating new arrays. B becomes A+B, A becomes -A/2, then A *= B gives (A+B)*(-A/2) entirely in place.
