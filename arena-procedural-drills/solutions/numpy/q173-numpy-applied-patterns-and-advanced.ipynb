## Problem

Write a function solve(a, b) that takes an invertible n x n matrix a and an n x m matrix b (m right-hand sides, one per column), and returns the n x m solution matrix x satisfying a @ x = b — all systems solved simultaneously.


<details><summary>💡 Hint (click to reveal)</summary>

The linear solver accepts a MATRIX right-hand side directly — one factorization covers every column.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.linalg.solve(a, b)


example_a = t.tensor([[2.0, 0.0], [0.0, 4.0]])
example_b = t.tensor([[2.0, 4.0], [8.0, 12.0]])
print(solve(example_a, example_b))
```

## Why this works

t.linalg.solve(a, b) with a 2-D b treats each column as an independent right-hand side and returns the matching columns of solutions, factorizing a only once. Looping columns or computing inv(a) @ b gives the same values with more work.
