## Problem

Given a square matrix A and a matrix b with multiple columns, find the solution vector x for the equation Ax = b for each column of b simultaneously. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

np.linalg.solve accepts a multi-column b and returns one solution column per right-hand side.

</details>

## Reference solution


```python
import numpy as np

def solve():
    return np.linalg.solve(A, b)
```

## Why this works

np.linalg.solve handles a right-hand side with multiple columns directly: it solves Ax = b for each column of b in a single factorization, returning an x whose columns are the corresponding solution vectors. This is more accurate and faster than inverting A.
