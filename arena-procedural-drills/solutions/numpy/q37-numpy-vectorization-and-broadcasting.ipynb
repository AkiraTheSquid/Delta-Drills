## Problem

Write a function solve(a, b) that takes two 1-D PyTorch tensors of floats with the same length and returns their dot product: the single number obtained by combining corresponding elements of the two vectors. The grader calls solve with several different vector pairs, so do not hard-code any values. Return a scalar (a plain float or a 0-d PyTorch value is fine), not an tensor with more than one element.

Think of the dot product as an elementwise product followed by a sum — PyTorch offers a single built-in reduction that does both at once for a pair of vectors.

## Reference solution


```python
import torch as t

def solve(a, b):
    return float(t.dot(a, b))


a_example = t.tensor([1.0, 2.0, 3.0])
b_example = t.tensor([4.0, -5.0, 6.0])
print(solve(a_example, b_example))
```

## Why this works

t.dot(a, b) (equivalently a @ b) multiplies corresponding elements of the two vectors and sums the products in one vectorized call, returning a scalar. Because the whole computation happens inside PyTorch, it needs no Python loop and works for any equal-length pair, including length-1 vectors and vectors whose dot product is zero. The same value can also be computed as (a * b).sum(), which makes the multiply-then-reduce structure explicit.
