## Problem

Given two random vectors of length 3, return the scalar value that represents the sum of the products of their corresponding elements. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

The sum of element-wise products of two vectors is exactly their dot product.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    a = np.random.random(3)
    b = np.random.random(3)
    return np.dot(a, b)

print(solve())
```

## Why this works

The dot product multiplies corresponding elements of the two vectors and sums the products into one scalar. Seeding to 0 reproduces the same two vectors the grader draws.
