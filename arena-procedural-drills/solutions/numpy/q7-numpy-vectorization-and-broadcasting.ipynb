## Problem

Given a 5x5 matrix Z of random values, return a new matrix where each element in a row has had the mean of that row subtracted from it. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Take the mean over the column axis but keep the reduced dimension so it broadcasts back against each row.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((5, 5))
    Z -= Z.mean(axis=1, keepdims=True)
    return Z

print(solve())
```

## Why this works

Computing the mean along axis=1 with keepdims=True yields a (5,1) column of per-row means that broadcasts cleanly against the (5,5) matrix, so each element has its own row's mean removed. Without keepdims the shape would not align for the row-wise subtraction.
