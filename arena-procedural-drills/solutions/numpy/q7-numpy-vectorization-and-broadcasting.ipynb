## Problem

Write a function solve(z) that takes a 2-D numpy array of floats and returns a new array of the same shape in which every entry has had its own row's mean subtracted. Each row of the result therefore averages to zero. Do not use Python loops, and do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Compute the mean along the row axis and keep its dimensions so broadcasting lines each mean up with its own row.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return z - z.mean(axis=1, keepdims=True)


example = np.array([[1.0, 2.0, 3.0], [10.0, 20.0, 30.0]])
print(solve(example))
```

## Why this works

z.mean(axis=1, keepdims=True) has shape (rows, 1), so subtracting it broadcasts one mean across each row. Without keepdims the shape (rows,) would broadcast against columns instead — the single-column test case catches that classic mistake.
