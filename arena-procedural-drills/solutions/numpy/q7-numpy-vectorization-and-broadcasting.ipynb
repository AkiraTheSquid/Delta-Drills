## Problem

Write a function solve(z) that takes a 2-D PyTorch tensor of floats and returns a new tensor of the same shape in which every entry has had its own row's mean subtracted. Each row of the result therefore averages to zero. Do not use Python loops, and do not modify the input.

## Reference solution


```python
import torch as t

def solve(z):
    return z - z.mean(dim=1, keepdim=True)


example = t.tensor([[1.0, 2.0, 3.0], [10.0, 20.0, 30.0]])
print(solve(example))
```
