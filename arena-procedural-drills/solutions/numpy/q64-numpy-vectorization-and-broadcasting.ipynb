## Problem

Write a function solve(a, b) that takes two PyTorch tensors of the same shape and returns a tuple of two plain Python booleans: first, whether a and b are equal elementwise WITHIN floating-point tolerance; second, whether they are EXACTLY equal (same shape and identical values). The two answers differ when values are only approximately equal.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has one comparison that forgives tiny float error and another that demands bit-for-bit agreement — this drill wants both.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return bool(t.allclose(a, b)), bool(t.equal(a, b))


print(solve(t.tensor([1.0, 2.0]), t.tensor([1.0, 2.0])))
```
