## Problem

Write a function solve(a, b) that takes two numpy arrays of the same shape and returns a tuple of two plain Python booleans: first, whether a and b are equal elementwise WITHIN floating-point tolerance; second, whether they are EXACTLY equal (same shape and identical values). The two answers differ when values are only approximately equal.

numpy has one comparison that forgives tiny float error and another that demands bit-for-bit agreement — this drill wants both.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return bool(np.allclose(a, b)), bool(np.array_equal(a, b))


print(solve(np.array([1.0, 2.0]), np.array([1.0, 2.0])))
```

## Why this works

np.allclose(a, b) tests equality within relative/absolute tolerance, so values differing by 1e-12 still count as equal; np.array_equal(a, b) requires identical shape and exactly identical values. Wrapping both in bool() converts numpy booleans to the plain Python bools the grader type-checks.
