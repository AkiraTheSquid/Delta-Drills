## Problem

Given the 1D array Z = np.linspace(-3, 3, 7), apply a ReLU-style clip so every negative entry becomes 0 while non-negative entries are left unchanged. The output keeps the original shape (7,) and dtype. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Clip the array at a lower bound of 0.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.linspace(-3, 3, 7)
    return Z.clip(min=0)


print(solve())
```

## Why this works

clip(min=0) raises any value below 0 up to 0 and leaves the rest unchanged, implementing ReLU while preserving shape and dtype.
