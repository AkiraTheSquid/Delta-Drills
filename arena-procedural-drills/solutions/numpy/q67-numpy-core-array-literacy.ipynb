## Problem

Write a function solve(z) that takes a 1-D PyTorch float tensor and returns a new tensor of the same shape where every negative entry has been replaced by 0.0 and every non-negative entry is unchanged (a ReLU-style clamp). Do not modify the input tensor.

## Reference solution


```python
import torch as t

def solve(z):
    return z.clamp(min=0.0)


example = t.tensor([-2.0, 0.5, 3.0])
print(solve(example))
```
