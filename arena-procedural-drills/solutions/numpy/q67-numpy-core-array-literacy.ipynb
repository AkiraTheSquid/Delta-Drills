## Problem

Write a function solve(z) that takes a 1-D numpy float array and returns a new array of the same shape where every negative entry has been replaced by 0.0 and every non-negative entry is unchanged (a ReLU-style clip). Do not modify the input array.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a clamping method that takes a lower bound — or take the elementwise maximum against zero.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return z.clip(min=0)


example = np.array([-2.0, 0.5, 3.0])
print(solve(example))
```

## Why this works

z.clip(min=0) (equivalently np.maximum(z, 0)) replaces everything below 0 with 0 while leaving other values as they are, returning a new array. This is exactly the ReLU activation used throughout deep learning.
