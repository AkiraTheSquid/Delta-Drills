## Problem

Write a function solve(z) that takes a 1-D PyTorch tensor of floats and returns a NEW tensor containing the same values reordered from smallest to largest. The input tensor must be left unmodified — the grader checks it afterwards. Note that torch's sort returns both values and indices.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has two sorts: the top-level FUNCTION returns a sorted copy, while the tensor METHOD sorts in place.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return t.sort(z).values


example = t.tensor([0.4, 0.1, 0.9])
print(solve(example))
```

## Why this works

t.sort(z) returns a new ascending tensor and leaves z untouched, unlike z.sort(), which mutates z and would fail the input-unmodified check. Duplicates keep their multiplicity and negatives order before positives.
