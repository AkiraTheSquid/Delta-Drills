## Problem

Draw a 1D vector of 10 random floats in [0, 1), then return the same values reordered from smallest to largest. The result has shape (10,) with entries in non-decreasing order. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Draw the vector, then sort it in place.

## Reference solution


```python
import numpy as np

np.random.seed(0)


def solve():
    np.random.seed(0)
    Z = np.random.random(10)
    Z.sort()
    return Z


print(solve())
```

## Why this works

np.random.random draws 10 floats in [0,1). Sorting in place reorders them non-decreasingly. The grader reseeds to 0 before calling, so the same draws are produced and sorted identically.
