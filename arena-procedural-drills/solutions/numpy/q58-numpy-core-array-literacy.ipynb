## Problem

Write a function solve(z) that takes a 1-D numpy array of floats and returns a NEW array containing the same values reordered from smallest to largest. The input array must be left unmodified — the grader checks it afterwards.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has two sorts: the top-level FUNCTION returns a sorted copy, while the array METHOD sorts in place.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.sort(z)


example = np.array([0.4, 0.1, 0.9])
print(solve(example))
```

## Why this works

np.sort(z) returns a new ascending array and leaves z untouched, unlike z.sort(), which mutates z and would fail the input-unmodified check. Duplicates keep their multiplicity and negatives order before positives.
