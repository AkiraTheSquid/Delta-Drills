## Problem

Given an integer n and a 2D array X, select from X the rows which can be interpreted as draws from a multinomial distribution with n degrees, i.e., the rows which only contain integers and which sum to n. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Test integrality with the fractional part (modulo 1) and require it across the whole row, then intersect with a row-sum-equals-n condition. Use the resulting boolean mask to index the rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    X = np.asarray([[1.0, 0.0, 3.0, 8.0],
                    [2.0, 0.0, 1.0, 1.0],
                    [1.5, 2.5, 1.0, 0.0]])
    n = 4
    M = np.logical_and.reduce(np.mod(X, 1) == 0, axis=-1)
    M &= (X.sum(axis=-1) == n)
    return X[M]

print(solve())
```

## Why this works

A row qualifies if every entry is an integer (so its fractional part is zero) and the row sums to n. np.mod(X, 1) == 0 flags integer entries; logical_and.reduce along the last axis requires all entries in a row to pass. Combining that mask with the row-sum equality and boolean-indexing X selects the multinomial rows.
