## Problem

Write a function solve(x, n) that takes a 2-D float array x and an integer n, and returns the subarray of rows that could be draws from a multinomial with n trials: rows whose entries are all NON-NEGATIVE INTEGERS (integer-valued floats count) summing exactly to n. Preserve row order.


<details><summary>💡 Hint (click to reveal)</summary>

Three independent row conditions — integer-valued (check the remainder mod 1), non-negative, and the right total — AND them into one mask.

</details>

## Reference solution


```python
import numpy as np

def solve(x, n):
    integral = (np.mod(x, 1) == 0).all(axis=1)
    non_negative = (x >= 0).all(axis=1)
    sums_to_n = x.sum(axis=1) == n
    return x[integral & non_negative & sums_to_n]


print(solve(np.array([[2.0, 2.0], [1.5, 2.5]]), 4))
```

## Why this works

np.mod(x, 1) == 0 detects integer-valued floats exactly (2.0000001 fails, as the last case demands); combining it with non-negativity and the row-sum condition via & builds one boolean row mask, and boolean indexing keeps qualifying rows in order.
