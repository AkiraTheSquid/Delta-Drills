## Problem

Write a function solve(z, k) that takes a 2-D PyTorch tensor z and a column index k, and returns a new tensor with the ROWS of z reordered so that column k reads in ascending order from top to bottom. Rows move as whole units. You may assume the values in column k are distinct.


<details><summary>💡 Hint (click to reveal)</summary>

Get the ORDER that would sort just the key column, then apply that order to the whole tensor's rows with fancy indexing.

</details>

## Reference solution


```python
import torch as t

def solve(z, k):
    return z[z[:, k].argsort()]


example = t.tensor([[5, 2], [1, 9]])
print(solve(example, 0))
```

## Why this works

z[:, k].argsort() computes the row order that sorts column k; indexing z with it moves entire rows into that order. This argsort-then-gather idiom is the PyTorch version of sorting records by a key field.
