## Problem

Given the 2D array Z of shape (3, 3) containing random integers in [0, 10), reorder its rows so that the values in column index 1 are in ascending order from top to bottom. Rows move as whole units; values within a row stay together. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Get the ordering of the target column with argsort, then use that index array to reorder the rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 10, (3, 3))
    return Z[Z[:, 1].argsort()]

print(solve())
```

## Why this works

argsort on column index 1 returns the row order that would put that column in ascending order. Indexing Z by that order vector reorders whole rows as units, so each row's values stay together while the rows are sorted by their column-1 key.
