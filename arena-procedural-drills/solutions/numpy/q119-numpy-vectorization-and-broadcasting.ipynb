## Problem

Given a vector Z of 10 random values, return the indices of the three largest elements in descending order. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Partition out the top-k indices, then sort only that small subset by value in descending order.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(10)
    k = 3
    idx = np.argpartition(Z, -k)[-k:]
    return idx[np.argsort(Z[idx])[::-1]]

print(solve())
```

## Why this works

np.argpartition cheaply isolates the indices of the k largest values (unordered) in the last k slots. Sorting just those k by their values and reversing yields the top-k indices in descending order.
