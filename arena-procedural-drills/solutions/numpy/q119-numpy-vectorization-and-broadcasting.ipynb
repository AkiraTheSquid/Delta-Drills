## Problem

Write a function solve(z, k) that takes a 1-D numpy array with DISTINCT values and an integer k, and returns the INDICES of the k largest elements, ordered so the index of the largest comes first, then the second largest, and so on.


<details><summary>💡 Hint (click to reveal)</summary>

argpartition finds the top-k cheaply but unordered — a small argsort of just those k finishes the ranking.

</details>

## Reference solution


```python
import numpy as np

def solve(z, k):
    idx = np.argpartition(z, -k)[-k:]
    return idx[np.argsort(z[idx])[::-1]]


print(solve(np.array([5.0, 9.0, 1.0, 7.0]), 2))
```

## Why this works

np.argpartition(z, -k)[-k:] isolates the k largest indices in O(n) without fully sorting; sorting only those k values (descending) puts the largest first. A plain np.argsort(z)[::-1][:k] is the simpler O(n log n) route and grades identically.
