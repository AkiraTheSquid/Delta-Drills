## Problem

Write a function solve(z, k) that takes a 1-D PyTorch tensor with DISTINCT values and an integer k, and returns the INDICES of the k largest elements, ordered so the index of the largest comes first, then the second largest, and so on.

## Reference solution


```python
import torch as t
def solve(z, k):
    return t.topk(z, k).indices


print(solve(t.tensor([5.0, 9.0, 1.0, 7.0]), 2))
```
