## Problem

Given a 3x3 integer matrix Z whose entries are the values 0 through 8 laid out in row-major order, return a 1D integer array of length 3 whose i-th entry is the total of the three values in the i-th column of Z (one column total per output position). (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Reducing with sum over axis 0 aggregates down the rows, producing one total per column.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    return Z.sum(axis=0)

print(solve())
```

## Why this works

Summing along axis 0 collapses the rows, adding the three values in each column and yielding a length-3 vector of column totals. For arange(9).reshape(3,3) the columns are (0,3,6),(1,4,7),(2,5,8), giving [9,12,15].
