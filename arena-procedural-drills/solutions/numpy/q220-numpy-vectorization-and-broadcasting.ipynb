## Problem

Write a function solve(x) that takes a 2-D PyTorch integer matrix x and returns a 1-D PyTorch tensor containing the sum of each column, so the result has one entry per column. The function must work for any matrix shape, including a single row or a single column. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch reductions take an axis argument — pick the axis that runs down the rows so it collapses, leaving one value per column.

</details>

## Reference solution


```python
import torch as t

def solve(x):
    return x.sum(dim=0)


example = t.tensor([[1, 2, 3], [10, 20, 30]])
print(solve(example))
```

## Why this works

Summing with axis=0 collapses axis 0 (the rows), so each output entry is the total of one column. PyTorch performs the whole reduction in vectorized C code, so no Python loop is needed and it works for any shape, including a single-row matrix (where the result equals that row) and a single-column matrix (where the result has length 1).
