## Problem

Write a function solve(z, k) that takes a square 2-D PyTorch tensor z and an integer offset k, and returns the 1-D tensor of entries along the k-th diagonal of z: k = 0 is the main diagonal, positive k counts diagonals above it, negative k below it.


<details><summary>💡 Hint (click to reveal)</summary>

The same PyTorch function that BUILDS a matrix from a diagonal also EXTRACTS diagonals when handed a matrix — and it takes an offset.

</details>

## Reference solution


```python
import torch as t

def solve(z, k):
    return t.diag(z, diagonal=k)


example = t.arange(25).reshape(5, 5)
print(solve(example, 1))
```

## Why this works

t.diag(z, k) applied to a 2-D tensor extracts the diagonal offset by k: entries z[i, i+k]. Positive offsets move above the main diagonal and shorten the result; the corner case k = n-1 leaves a single element.
