## Problem

You are given Z, a 2D square array of shape (5, 5) containing the integers 0 through 24 in row-major order. Return d, a 1D array of shape (4,) containing the entries that sit one position above the main diagonal of Z, that is the entries Z[i, i+1] for i in 0 through 3, in that order. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

np.diag takes an offset argument k that selects diagonals above or below the main one.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(25).reshape(5, 5)
    return np.diag(Z, k=1)

print(solve())
```

## Why this works

np.diag with offset k=1 extracts the first superdiagonal, i.e. the entries Z[i, i+1]. For a 5x5 matrix that produces a length-4 vector of the values sitting just above the main diagonal.
