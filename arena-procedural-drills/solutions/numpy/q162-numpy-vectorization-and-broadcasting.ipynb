## Problem

Write a function solve(z) that takes a 2-D float array with non-negative entries and returns the array with each ROW rescaled to sum to 1 — a probability distribution per row. Rows that sum to zero must remain all zeros, with no warnings, NaN, or inf produced.


<details><summary>💡 Hint (click to reveal)</summary>

Row sums with keepdims broadcast the divisor across each row; the where=/out= division pattern sidesteps the zero-sum rows.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    sums = z.sum(axis=1, keepdims=True)
    return np.divide(z, sums, out=np.zeros_like(z), where=sums != 0)


print(solve(np.array([[1.0, 3.0], [0.0, 0.0]])))
```

## Why this works

z.sum(axis=1, keepdims=True) is the (n, 1) column of row totals; dividing through where=sums != 0 with zero-filled defaults normalizes real rows and leaves empty ones untouched. This is L1 (sum) normalization, the sibling of the L2 (length) version in the unit-row drill.
