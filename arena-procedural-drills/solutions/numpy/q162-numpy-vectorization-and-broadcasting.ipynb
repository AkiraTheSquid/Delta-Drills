## Problem

You are given Z, a 2D array of shape (5, 3) with non-negative entries. Return a 2D array of the same shape (5, 3) where each row of Z is rescaled by a single positive constant so that its entries sum to 1; for any row whose sum is 0, leave that row as all zeros to avoid division by zero. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Divide by per-row sums (keepdims), using the where/out guard to skip zero-sum rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(5, 3)
    row_sums = Z.sum(axis=1, keepdims=True)
    Z_norm = np.divide(Z, row_sums, out=np.zeros_like(Z), where=row_sums!=0)
    return Z_norm

print(solve())
```

## Why this works

Each row is rescaled to sum to 1 by dividing by its own total. Computing row sums with keepdims lets them broadcast over the columns, and np.divide with out and where performs the division only where the sum is nonzero, leaving zero-sum rows as zeros.
