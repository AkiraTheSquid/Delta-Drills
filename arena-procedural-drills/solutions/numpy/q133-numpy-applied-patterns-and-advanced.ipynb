## Problem

Given an image represented as a 3D array of shape (height, width, 3) with values in the range of 0 to 255, determine the total number of distinct color combinations present in the image. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Reshape the pixels into an (N,3) list of colors and let np.unique with axis=0 deduplicate entire rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    w, h = 256, 256
    I = np.random.randint(0, 4, (h, w, 3)).astype(np.ubyte)
    colors = np.unique(I.reshape(-1, 3), axis=0)
    n = len(colors)
    return n

print(solve())
```

## Why this works

Flattening the image to a list of RGB triples with reshape(-1, 3) turns the problem into counting unique rows. np.unique with axis=0 deduplicates whole rows (colors), and the length of that result is the number of distinct colors present.
