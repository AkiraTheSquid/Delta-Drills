## Problem

Write a function solve(img) that takes an image as a 3-D PyTorch tensor of shape (height, width, 3) with integer channel values and returns the number of DISTINCT colors it contains — a color being a full (R, G, B) triple — as a plain Python int.

## Reference solution


```python
import torch as t

def solve(img):
    colors = t.unique(img.reshape(-1, 3), dim=0)
    return int(len(colors))


example = t.zeros((2, 2, 3), dtype=t.int64)
example[0, 0] = t.tensor([255, 0, 0])
print(solve(example))
```
