## Problem

You are given Z, a 1D integer array of shape (10,) containing the values 0 through 9 in order. Produce a single scalar equal to the total of all entries of Z. The output is one integer value, not an array. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce the array with the add ufunc rather than calling np.sum.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.arange(10)
    return np.add.reduce(Z)


print(solve())
```

## Why this works

np.add.reduce applies the add ufunc cumulatively across the array, returning the scalar sum. For small arrays this avoids np.sum's overhead while giving the same total (45).
