## Problem

Write a function solve(z) that takes a 1-D PyTorch integer tensor and returns the sum of all its entries as a plain Python int — not a 0-dimensional tensor.

## Reference solution


```python
import torch as t

def solve(z):
    return int(z.sum())


example = t.arange(10)
print(solve(example))
```
