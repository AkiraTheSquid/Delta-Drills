## Problem

Write a function solve(z) that takes a 1-D numpy integer array and returns the sum of all its entries as a plain Python int (not a numpy scalar and not an array).


<details><summary>💡 Hint (click to reveal)</summary>

Any of numpy's summing routes works — the catch is converting the numpy scalar into Python's own integer type.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return int(z.sum())


example = np.arange(10)
print(solve(example))
```

## Why this works

z.sum() (or np.add.reduce(z)) collapses the array to a numpy integer scalar; wrapping it in int() produces the plain Python int the grader type-checks with isinstance.
