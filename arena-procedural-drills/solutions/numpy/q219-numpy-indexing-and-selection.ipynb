## Problem

Write a function solve(x) that takes a 1-D numpy array of floats and returns an array of the same shape in which the largest entry has been replaced with 0. Every other entry must keep its original value. If the maximum value appears more than once, replace only its first occurrence.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a reduction that gives you the position (index) of the largest value rather than the value itself — use that index to assign into a copy.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    result = x.copy()
    result[np.argmax(x)] = 0.0
    return result


example = np.array([3.0, -1.0, 7.5, 2.0])
print(solve(example))
```

## Why this works

np.argmax(x) returns the index of the first occurrence of the maximum, which is exactly the tie-breaking rule the task asks for. Assigning 0.0 at that single index changes one entry while leaving the rest untouched. Copying x before assigning keeps the caller's array unmodified and preserves the original shape.
