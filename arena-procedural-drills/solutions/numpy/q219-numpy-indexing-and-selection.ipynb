## Problem

You are given Z, a 1D vector of 10 random floats in [0, 1). Return the same vector but with the single entry that holds the largest value replaced by 0; all other entries keep their original values. The result has shape (10,). (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Locate the position of the maximum with argmax, then assign zero to just that index.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random(10)
    Z[Z.argmax()] = 0
    return Z

print(solve())
```

## Why this works

Z.argmax() gives the index of the largest entry; assigning 0 to that single position replaces the maximum while leaving every other value untouched, preserving the (10,) shape.
