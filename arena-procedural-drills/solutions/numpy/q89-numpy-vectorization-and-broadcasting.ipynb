## Problem

You are given a and b, two 1D arrays of equal length 3. Return a single 1D array of length 6 whose entries alternate between a and b starting with a[0], so the result is [a[0], b[0], a[1], b[1], a[2], b[2]]. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Pair the two arrays up as columns of a 2-column matrix, then flatten it in row-major order.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.array([1, 3, 5])
    b = np.array([2, 4, 6])
    return np.ravel(np.column_stack((a, b)))

print(solve())
```

## Why this works

Stacking the two vectors as columns produces a (3, 2) matrix whose rows are the (a[i], b[i]) pairs. Reading that matrix in C (row-major) order with np.ravel walks each row left-to-right, which is exactly the interleaved sequence a[0], b[0], a[1], b[1], .... This avoids any manual index juggling.
