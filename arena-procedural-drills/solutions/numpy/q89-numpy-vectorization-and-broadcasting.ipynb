## Problem

Write a function solve(a, b) that takes two 1-D numpy arrays of the same length n and returns a single 1-D array of length 2n whose entries alternate between a and b starting with a[0]: [a[0], b[0], a[1], b[1], ...]. Do not use Python loops.

Pair the arrays up as columns of one matrix, then read that matrix in row-major order.

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.ravel(np.column_stack((a, b)))


print(solve(np.array([1, 3, 5]), np.array([2, 4, 6])))
```

## Why this works

np.column_stack((a, b)) builds an (n, 2) array whose rows are (a[i], b[i]); raveling it row-major visits a[0], b[0], a[1], b[1], ... Preallocating out[0::2] = a; out[1::2] = b is an equally valid strided-assignment route.
