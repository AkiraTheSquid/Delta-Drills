## Problem

Write a function solve(x) that takes a 1-D PyTorch tensor of integers and returns an tensor of the same shape whose i-th entry is the total of elements 0 through i inclusive. For example, the running totals of [2, 3, 5] are [2, 5, 10]. Do not use Python loops.

## Reference solution


```python
import torch as t
def solve(x):
    return t.cumsum(x, dim=0)


example = t.tensor([2, 3, 5])
print(solve(example))
```
