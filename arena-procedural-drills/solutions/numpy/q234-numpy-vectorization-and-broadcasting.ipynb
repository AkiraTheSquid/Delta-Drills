## Problem

Write a function solve(x) that takes a 1-D numpy array of integers and returns an array of the same shape whose i-th entry is the total of elements 0 through i inclusive. For example, the running totals of [2, 3, 5] are [2, 5, 10]. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a built-in that computes a running (cumulative) total along an array in one call.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.cumsum(x)


example = np.array([2, 3, 5])
print(solve(example))
```

## Why this works

np.cumsum(x) walks along the array once and emits, at each position i, the sum of x[0] through x[i], so the output has the same shape as the input. It handles negatives and a single-element array the same way, and it runs in compiled code with no Python loop.
