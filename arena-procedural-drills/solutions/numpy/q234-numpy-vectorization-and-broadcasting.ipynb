## Problem

Given a 1D integer vector Z of length 5, return a 1D vector of the same shape whose i-th entry is the running total of Z's entries from position 0 through position i inclusive (so the first output entry equals Z[0] and the last output entry equals the total of every entry in Z). (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

There is a single NumPy reduction that yields prefix sums.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(5)
    return np.cumsum(Z)

print(solve())
```

## Why this works

np.cumsum returns the running total: each output position holds the sum of all input entries up to and including that index. The first entry equals Z[0] and the last equals the grand total, exactly the prefix-sum definition.
