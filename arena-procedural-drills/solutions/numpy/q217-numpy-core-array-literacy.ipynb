## Problem

Build Z, a 2D integer array of shape (8, 8) whose entries form a checkerboard: Z[i, j] is 0 when i + j is even and 1 when i + j is odd, with the top-left entry Z[0, 0] equal to 0. Adjacent cells (horizontally or vertically) must hold opposite values, producing the classic alternating 0/1 grid. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Find the smallest 2x2 block that already alternates correctly, then replicate it across the grid with np.tile.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.tile(np.array([[0, 1], [1, 0]]), (4, 4))
    return Z

print(solve())
```

## Why this works

A 2x2 tile [[0,1],[1,0]] is the smallest checkerboard unit with 0 at its top-left. Tiling it 4x4 produces an 8x8 grid where Z[i,j] = (i+j) % 2, giving the alternating pattern with adjacent cells holding opposite values.
