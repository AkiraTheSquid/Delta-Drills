## Problem

Write a function solve(block, reps_r, reps_c) that takes a 2-D numpy array block and two positive repetition counts, and returns the array formed by TILING the block reps_r times vertically and reps_c times horizontally — shape (block_rows * reps_r, block_cols * reps_c).


<details><summary>💡 Hint (click to reveal)</summary>

np.tile's repetition argument is a tuple — one count per axis.

</details>

## Reference solution


```python
import numpy as np

def solve(block, reps_r, reps_c):
    return np.tile(block, (reps_r, reps_c))


print(solve(np.array([[0, 1], [1, 0]]), 2, 2))
```

## Why this works

np.tile(block, (reps_r, reps_c)) stamps the block in a reps_r x reps_c grid, repeating along both axes at once. Tiling the 2x2 checker block is the classic loop-free way to build a full checkerboard.
