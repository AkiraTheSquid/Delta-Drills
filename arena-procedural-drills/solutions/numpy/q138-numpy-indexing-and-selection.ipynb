## Problem

Write a function solve(z, p) that takes a 2-D PyTorch tensor z and a permutation p of its row indices, and reorders the rows of z IN PLACE according to p (row i of the result is the old row p[i]). Return z itself — the grader verifies that the returned object IS the caller's tensor, mutated.


<details><summary>💡 Hint (click to reveal)</summary>

z[p] builds the reordered copy; assigning it through z[:] pours those values back into the ORIGINAL buffer instead of rebinding the name.

</details>

## Reference solution


```python
import torch as t

def solve(z, p):
    z[:] = z[p]
    return z


example = t.arange(6).reshape(3, 2)
print(solve(example, t.tensor([2, 0, 1])))
```

## Why this works

z[p] is a new tensor (fancy indexing copies — which is also why the right-hand side is safe to read while writing). The slice-assignment z[:] = ... overwrites the existing buffer in place, so the caller's object mutates; plain z = z[p] would rebind a local name and leave the caller's tensor untouched, failing the identity check.
