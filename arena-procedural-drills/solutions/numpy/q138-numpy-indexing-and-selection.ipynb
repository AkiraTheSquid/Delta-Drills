## Problem

Given the 2D array Z of shape (5, 4) holding the values 0..19 in row-major order, and a length-5 random permutation p of the row indices, reorder the rows of Z according to p, writing the result back into Z so that the array is modified in place. The shape stays (5, 4). (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Index the rows with the permutation, then write the result back through a full slice (Z[:] = ...) to mutate in place.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.arange(20).reshape(5, 4)
    p = np.random.permutation(Z.shape[0])
    Z[:] = Z[p]
    return Z

print(solve())
```

## Why this works

Z[p] gathers the rows in permuted order into a new array; assigning it back with the slice Z[:] writes those rows into the original buffer in place rather than rebinding the name. The shape is preserved because a permutation reorders rows without changing their count.
