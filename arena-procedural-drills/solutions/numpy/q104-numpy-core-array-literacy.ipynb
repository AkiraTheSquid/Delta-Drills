## Problem

For n = 5, build an n x n floating-point permutation matrix: each row and each column contains exactly one 1.0 and the remaining entries are 0.0, with the positions of the 1.0s determined by a uniformly random permutation of {0, 1, ..., n-1}. The result has shape (5, 5). (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Permute the rows of an n×n identity matrix using a random permutation of 0..n-1.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    n = 5
    return np.eye(n)[np.random.permutation(n)]

print(solve())
```

## Why this works

Each row of the identity matrix has a single 1.0; reordering those rows by a random permutation yields a matrix where every row and every column still contains exactly one 1.0, i.e. a permutation matrix. np.random.permutation(n) supplies the random row order.
