## Problem

Write a function solve(n, rng) that takes a size n and a numpy random Generator rng, and returns an n x n floating-point permutation matrix: every row and every column contains exactly one 1.0 with all other entries 0.0, the arrangement chosen at random using rng. The grader checks the permutation-matrix properties, so any valid random permutation passes.


<details><summary>💡 Hint (click to reveal)</summary>

Shuffling the ROWS of the identity matrix keeps exactly one 1 in every row and column.

</details>

## Reference solution


```python
import numpy as np

def solve(n, rng):
    return np.eye(n)[rng.permutation(n)]


print(solve(3, np.random.default_rng(0)))
```

## Why this works

np.eye(n) is a permutation matrix already (the identity); indexing its rows with rng.permutation(n) shuffles them, preserving the one-1-per-row-and-column property. The tests assert those invariants (row sums, column sums, values) rather than a specific arrangement.
