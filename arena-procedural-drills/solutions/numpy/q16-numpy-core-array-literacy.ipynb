## Problem

Construct a 5x5 floating-point matrix Z whose entries strictly below the main diagonal are 1.0 and every other entry (the main diagonal itself and everything above it) is 0.0. The result is a 2D array of shape (5, 5) with ones forming a triangle that excludes the diagonal. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Take the lower triangle of an all-ones matrix but shift the cutoff so the main diagonal itself is excluded.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.tril(np.ones((5, 5)), k=-1)
    return Z

print(solve())
```

## Why this works

np.tril keeps the lower-triangular part of an array and zeros the rest; the offset k=-1 excludes the main diagonal, leaving ones strictly below it. Applied to an all-ones 5x5 matrix this yields the strict lower-triangular pattern of ones.
