## Problem

Write a function solve(a, b) that takes two 2-D numpy arrays a and b with the same shape. It should return a tuple of two arrays: the first is the vertical combination, with the rows of b stacked below the rows of a; the second is the horizontal combination, with b placed side by side to the right of a. Do not modify the input arrays.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a pair of stacking helpers that join a sequence of arrays along the row axis or along the column axis.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.vstack([a, b]), np.hstack([a, b])


a = np.array([[1.0, 2.0], [3.0, 4.0]])
b = np.array([[5.0, 6.0], [7.0, 8.0]])
print(solve(a, b))
```

## Why this works

np.vstack([a, b]) joins the arrays along axis 0, so b's rows appear below a's, while np.hstack([a, b]) joins along axis 1, placing b's columns to the right of a's. Because a and b share a shape, both stacks are always valid, including for single-row, single-column, and 1x1 inputs. Both helpers allocate new arrays, so the inputs are left untouched.
