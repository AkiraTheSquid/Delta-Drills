## Problem

Given two 2x2 arrays, combine them into a single array by placing one on top of the other and then by placing them side by side. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

There are two complementary stacking helpers: one concatenates rows, the other concatenates columns.

</details>

## Reference solution


```python
import numpy as np

def solve():
    a = np.ones((2, 2))
    b = np.zeros((2, 2))
    print(np.vstack([a, b]))
    return np.hstack([a, b])

print(solve())
```

## Why this works

np.vstack stacks the two 2x2 blocks along axis 0 (one on top of the other, giving a 4x2 array), while np.hstack joins them along axis 1 (side by side, giving a 2x4 array). The grader checks the horizontal stack returned by solve.
