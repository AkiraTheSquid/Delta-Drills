## Problem

Write a function solve(n) that takes a non-negative integer n and returns a 1-D PyTorch tensor of length n in which every entry is 0.0. The result must use torch's default floating-point dtype, not integers. Your function must also work when n is 0, returning an empty tensor.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has a one-call constructor that allocates an all-zero tensor of a given size, and its default dtype is already floating point.

</details>

## Reference solution


```python
import torch as t

def solve(n):
    return t.zeros(n)


example = 4
print(solve(example))
```

## Why this works

t.zeros(n) allocates a 1-D tensor with n entries, each initialized to 0.0, and its default dtype is float64 — no loop or explicit fill is needed. Because the size is just a parameter, the same call handles every length, including n = 0, which produces an empty float tensor.
