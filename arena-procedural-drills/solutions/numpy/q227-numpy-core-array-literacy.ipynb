## Problem

Build Z, a 1D floating-point array of shape (10,) in which every entry is 0.0. The result is a single null vector of length 10 with a floating-point dtype, where each of the 10 positions holds the value zero. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

The zeros constructor with a single length argument gives a float null vector.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros(10)
    return Z

print(solve())
```

## Why this works

np.zeros(10) allocates a length-10 vector of the default float dtype with every entry initialized to 0.0.
