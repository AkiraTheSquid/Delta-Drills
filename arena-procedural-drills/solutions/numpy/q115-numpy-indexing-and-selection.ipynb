## Problem

Write a function solve(z) that takes a 2-D numpy float array that may contain NaN entries and returns a 1-D boolean array with one entry per COLUMN: True where that column consists entirely of NaN values, False otherwise. The result's dtype must be bool.


<details><summary>💡 Hint (click to reveal)</summary>

Turn the array into a NaN mask first, then ask along the column axis whether EVERY entry is flagged.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.isnan(z).all(axis=0)


example = np.array([[0.0, np.nan], [1.0, np.nan]])
print(solve(example))
```

## Why this works

np.isnan(z) marks NaN cells; .all(axis=0) collapses down each column, staying True only when every row in that column was NaN. A column with even one finite value (last case) comes out False.
