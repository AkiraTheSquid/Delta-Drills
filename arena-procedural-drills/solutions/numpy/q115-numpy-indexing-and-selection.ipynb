## Problem

You are given Z, a 2D float array of shape (3, 3) whose third column is entirely NaN while the other columns are finite. Return a 1D boolean array of length 3 where entry j is True if column j of Z is filled with NaN values from top to bottom and False otherwise. (★★☆)
Print the result.

Build a NaN mask and ask whether it holds for every row within each column.

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([
        [0, 1, np.nan],
        [1, 2, np.nan],
        [4, 5, np.nan],
    ])
    return np.isnan(Z).all(axis=0)

print(solve())
```

## Why this works

np.isnan produces a boolean mask of NaN positions, and reducing it with .all(axis=0) checks whether every entry down a column is NaN. The result is a length-3 boolean array flagging fully-NaN columns.
