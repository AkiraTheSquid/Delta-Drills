## Problem

Write a function solve(z) that takes a 2-D PyTorch float tensor that may contain NaN entries and returns a 1-D boolean tensor with one entry per COLUMN: True where that column consists entirely of NaN values, False otherwise. The result's dtype must be bool.


<details><summary>💡 Hint (click to reveal)</summary>

Turn the tensor into a NaN mask first, then ask along the column axis whether EVERY entry is flagged.

## Reference solution


```python
import torch as t

def solve(z):
    return t.isnan(z).all(dim=0)


example = t.tensor([[0.0, t.nan], [1.0, t.nan]])
print(solve(example))
```

## Why this works

t.isnan(z) marks NaN cells; .all(axis=0) collapses down each column, staying True only when every row in that column was NaN. A column with even one finite value (last case) comes out False.
