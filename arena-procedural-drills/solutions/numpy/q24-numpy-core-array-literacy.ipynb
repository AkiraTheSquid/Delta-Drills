## Problem

Write a function solve(z) that takes a 1-D PyTorch tensor of floats and returns a new tensor in which the largest value has been replaced by 0.0, with everything else unchanged. If the maximum occurs more than once, replace only its first occurrence. Do not modify the input tensor.


<details><summary>💡 Hint (click to reveal)</summary>

Find the POSITION of the maximum rather than its value, then assign into a copy at that single index.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    out = z.clone()
    out[out.argmax()] = 0.0
    return out


example = t.tensor([2.0, 9.0, 4.0])
print(solve(example))
```

## Why this works

z.argmax() gives the index of the first occurrence of the maximum; assigning 0.0 at that index on a copy leaves the original untouched and changes exactly one entry. Using z == z.max() as a mask would wrongly zero EVERY tied maximum — the tie case catches that.
