## Problem

Given a random vector Z of length 10, set the largest value in the vector to 0 and return the modified vector. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a single-call way to locate the position of the largest entry, then write to that position.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random(10)
    Z[Z.argmax()] = 0
    return Z

print(solve())
```

## Why this works

argmax returns the position of the largest element; assigning 0 to Z at that index overwrites the maximum in place. Returning Z gives the vector with its peak zeroed out.
