## Problem

Write a function solve(z) that takes a 2-D PyTorch integer tensor and returns an tensor of the same shape where entry [i, j] holds the RUNNING MAXIMUM of row i from column 0 through column j — the largest value seen so far scanning that row left to right. Do not use Python loops.

## Reference solution


```python
import torch as t
def solve(z):
    return t.cummax(z, dim=1).values


print(solve(t.tensor([[3, 1, 4, 1, 5]])))
```
