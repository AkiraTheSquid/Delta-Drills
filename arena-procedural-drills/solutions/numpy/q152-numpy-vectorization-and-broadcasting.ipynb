## Problem

Given a 2D array Z, return an array where each element at position (i, j) contains the maximum value from the start of the row up to column j for each row. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

The maximum ufunc has an accumulate method; run it along the row axis.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 100, (4, 7))
    cummax = np.maximum.accumulate(Z, axis=1)
    return cummax

print(solve())
```

## Why this works

np.maximum is a binary ufunc, and its .accumulate method performs a running fold along an axis. Applied with axis=1 it produces, at each column, the maximum of all entries from the start of the row up to that column, i.e. a row-wise cumulative maximum.
