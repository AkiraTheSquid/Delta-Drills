## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns an array of the same shape where entry [i, j] holds the RUNNING MAXIMUM of row i from column 0 through column j — the largest value seen so far scanning that row left to right. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

cumsum is just add.accumulate — the maximum ufunc has the same accumulate method, along whichever axis you choose.

## Reference solution


```python
import numpy as np

def solve(z):
    return np.maximum.accumulate(z, axis=1)


print(solve(np.array([[3, 1, 4, 1, 5]])))
```

## Why this works

Every binary ufunc exposes .accumulate; np.maximum.accumulate(z, axis=1) carries the largest-so-far along each row exactly the way np.add.accumulate carries a running total. A strictly decreasing row stays pinned at its first value — the last test.
