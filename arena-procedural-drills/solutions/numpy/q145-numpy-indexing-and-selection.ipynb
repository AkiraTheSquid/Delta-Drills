## Problem

Given a 1D array, identify the indices of elements that are greater than their immediate neighbors, excluding the first and last elements. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Compare the interior slice against its left and right shifts, then offset the resulting indices by one to account for the dropped first element.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    a = np.random.randn(20)
    peaks = np.where((a[1:-1] > a[:-2]) & (a[1:-1] > a[2:]))[0] + 1
    return peaks

print(solve())
```

## Why this works

An interior element is a strict local maximum when it exceeds both neighbors. Comparing the interior slice a[1:-1] against the left-shifted a[:-2] and right-shifted a[2:] vectorizes both checks; np.where gives positions within the interior, and adding 1 maps them back to original indices.
