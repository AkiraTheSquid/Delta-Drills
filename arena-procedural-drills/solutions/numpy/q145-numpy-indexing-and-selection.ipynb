## Problem

Write a function solve(a) that takes a 1-D float array and returns the indices of its LOCAL PEAKS: positions i (with 1 <= i <= len(a)-2) where a[i] is strictly greater than both immediate neighbors. The first and last positions can never qualify. Return an integer array (possibly empty), indices ascending.


<details><summary>💡 Hint (click to reveal)</summary>

Three shifted slices line every interior element up with its two neighbors — compare, AND the masks, and remember the slice offset.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    return np.where((a[1:-1] > a[:-2]) & (a[1:-1] > a[2:]))[0] + 1


print(solve(np.array([1.0, 3.0, 2.0, 5.0, 4.0])))
```

## Why this works

a[1:-1], a[:-2], and a[2:] align each interior element with its left and right neighbors; the conjunction of both strict comparisons marks peaks, and np.where gives positions WITHIN the interior slice — hence the +1 shift back to original indices. The plateau case ([2.0, 2.0]) yields nothing because the comparison is strict.
