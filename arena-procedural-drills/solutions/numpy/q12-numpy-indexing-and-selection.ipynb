## Problem

Given a 1D array Z, modify the array such that all elements with values greater than 3 and less than 8 are negated. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Build a boolean mask from the two comparisons combined with &, then multiply the masked elements by -1 in place.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(11)
    Z[(3 < Z) & (Z < 8)] *= -1
    return Z

print(solve())
```

## Why this works

A boolean mask combining the two element-wise comparisons with & selects the entries strictly between 3 and 8, and in-place multiplication by -1 negates exactly those positions. Boolean indexing edits the selected elements while leaving the rest unchanged.
