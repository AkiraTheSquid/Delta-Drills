## Problem

Write a function solve(z) that takes a 1-D PyTorch integer tensor and returns a new tensor in which every entry strictly greater than 3 and strictly less than 8 has been negated, while all other entries are unchanged. The input tensor itself must not be modified. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Combine two comparisons with the elementwise AND operator to build a boolean mask, then assign through it on a copy.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    out = z.clone()
    out[(out > 3) & (out < 8)] *= -1
    return out


example = t.arange(11)
print(solve(example))
```

## Why this works

(z > 3) & (z < 8) is a boolean mask of the entries inside the open interval; indexing a copy with it and multiplying by -1 negates exactly those positions. The & operator (not Python's `and`) is required for elementwise logic, and both bounds are strict so 3 and 8 stay positive.
