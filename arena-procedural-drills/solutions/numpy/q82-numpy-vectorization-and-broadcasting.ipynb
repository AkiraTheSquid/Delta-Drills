## Problem

Write a function solve(z) that takes a 2-D PyTorch integer tensor and returns an tensor of the same shape in which entry [i, j] equals the running total of row i from column 0 through column j inclusive (a cumulative sum along each row).


<details><summary>💡 Hint (click to reveal)</summary>

The cumulative-sum function takes an axis — choose the one that walks across each row.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return z.cumsum(dim=1)


example = t.arange(6).reshape(2, 3)
print(solve(example))
```

## Why this works

z.cumsum(axis=1) accumulates along axis 1, so each row's totals build left to right independently of the other rows. With axis=0 the totals would instead run down the columns — the single-column case distinguishes the two.
