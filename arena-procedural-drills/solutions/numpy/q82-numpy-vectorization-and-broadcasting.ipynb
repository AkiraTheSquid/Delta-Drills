## Problem

Given the 2D array Z of shape (rows, cols) = (3, 4) holding the values 0..11 in row-major order, produce R of the same shape whose entry R[i, j] equals the running total along that row from column 0 up to and including column j. Each row is summed independently. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

A cumulative sum along the column axis gives per-row running totals.

</details>


```python
%pip install -q numpy
```

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(12).reshape(3, 4)
    return Z.cumsum(axis=1)

print(solve())
```

## Why this works

cumsum along axis=1 accumulates left-to-right within each row independently, so R[i, j] is the running total from column 0 through column j. Each row is treated separately because the reduction is along the column axis.
