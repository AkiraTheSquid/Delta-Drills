## Problem

Write a function solve(z, perm) that takes a 2-D numpy array z and a 1-D integer array perm holding a permutation of the row indices of z. Return a new array whose i-th row is row perm[i] of z — the rows of z reordered according to perm.


<details><summary>💡 Hint (click to reveal)</summary>

Indexing a 2-D array with an integer array along the first axis selects whole rows in exactly the order you list them.

</details>

## Reference solution


```python
import numpy as np

def solve(z, perm):
    return z[perm]


example = np.arange(6).reshape(3, 2)
print(solve(example, np.array([2, 0, 1])))
```

## Why this works

Fancy indexing z[perm] gathers row perm[i] into output position i, returning a new array — precisely 'reorder the rows by perm'. It works for any permutation, including the identity and a full reversal.
