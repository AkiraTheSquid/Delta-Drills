## Problem

You are given Z, a 2D array of shape (4, 3) containing the integers 0 through 11 in row-major order, along with perm, a 1D array of shape (4,) holding a random permutation of the row indices 0 through 3. Return a (4, 3) array whose rows are the rows of Z reordered according to perm, so each output row is one row of Z taken in the order specified by perm. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Fancy-index the array's first axis with the permutation array to pick rows in a new order.

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.arange(12).reshape(4, 3)
    perm = np.random.permutation(Z.shape[0])
    return Z[perm]

print(solve())
```

## Why this works

Indexing a 2D array with a 1D integer array selects whole rows in the order given, so Z[perm] reorders the rows of Z according to the permutation. Seeding the RNG with 0 makes the permutation reproducible to match the grader.
