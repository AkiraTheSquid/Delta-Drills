## Problem

Write a function solve(z, perm) that takes a 2-D PyTorch tensor z and a 1-D integer tensor perm holding a permutation of the row indices of z. Return a new tensor whose i-th row is row perm[i] of z — the rows of z reordered according to perm.


<details><summary>💡 Hint (click to reveal)</summary>

Indexing a 2-D tensor with an integer tensor along the first axis selects whole rows in exactly the order you list them.

</details>

## Reference solution


```python
import torch as t

def solve(z, perm):
    return z[perm]


example = t.arange(6).reshape(3, 2)
print(solve(example, t.tensor([2, 0, 1])))
```

## Why this works

Fancy indexing z[perm] gathers row perm[i] into output position i, returning a new tensor — precisely 'reorder the rows by perm'. It works for any permutation, including the identity and a full reversal.
