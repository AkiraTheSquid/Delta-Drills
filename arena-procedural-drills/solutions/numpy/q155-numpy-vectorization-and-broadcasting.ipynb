## Problem

Write a function solve(values, counts) that takes a 1-D array of run values and an equal-length integer array of run lengths, and returns the DECODED run-length-encoded sequence: values[0] repeated counts[0] times, then values[1] repeated counts[1] times, and so on. A count of 0 contributes nothing.


<details><summary>💡 Hint (click to reveal)</summary>

Element-wise repetition with a per-element count array IS run-length decoding.

</details>

## Reference solution


```python
import numpy as np

def solve(values, counts):
    return np.repeat(values, counts)


print(solve(np.array([1, 3, 2]), np.array([2, 1, 3])))
```

## Why this works

np.repeat(values, counts) emits each value its own number of times back to back — exactly the inverse of run-length encoding. Zero counts silently drop their value, so [7,8,9]/[1,0,2] decodes without an 8.
