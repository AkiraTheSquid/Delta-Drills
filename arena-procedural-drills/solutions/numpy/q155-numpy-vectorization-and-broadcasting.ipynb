## Problem

Given a 1D array values and a 1D integer array counts of the same length, return a 1D array where each entry values[i] is reproduced counts[i] times in order, reconstructing the original run-length-encoded sequence. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

np.repeat can take a per-element count array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    values = np.array([1, 3, 2, 2])
    counts = np.array([2, 1, 3, 2])
    decoded = np.repeat(values, counts)
    return decoded

print(solve())
```

## Why this works

np.repeat accepts an array of repeat counts, one per element, so values[i] is emitted counts[i] times in order. This is exactly run-length decoding, done in a single vectorized call with no explicit loop.
