## Problem

You are given Z, a 2D floating-point array of shape (10, 10) with entries drawn from the uniform distribution on [0, 1). Return rank, a single integer scalar giving the rank of Z as a matrix (the number of linearly independent rows or columns). (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

The rank equals the number of non-negligible singular values; numpy has a one-call helper that does the SVD and thresholding for you.

</details>

## Reference solution


```python
import numpy as np
np.random.seed(0)

def solve():
    Z = np.random.uniform(0, 1, (10, 10))
    return np.linalg.matrix_rank(Z)

print(solve())
```

## Why this works

A random 10x10 matrix is full rank with probability 1, so the rank is 10. np.linalg.matrix_rank computes this via the SVD, counting singular values above a numerical tolerance, which is the robust way to measure linear independence of rows/columns.
