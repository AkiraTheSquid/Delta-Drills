## Problem

Write a function solve(z) that takes a 2-D float tensor and returns its matrix RANK — the number of linearly independent rows/columns — as a plain Python int, judged numerically (tiny singular values do not count).


<details><summary>💡 Hint (click to reveal)</summary>

Rank equals the number of singular values above a tolerance — PyTorch wraps the whole SVD recipe in one linalg call.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return int(t.linalg.matrix_rank(z))


print(solve(t.tensor([[1.0, 2.0], [2.0, 4.0]])))
```

## Why this works

t.linalg.matrix_rank runs an SVD and counts singular values above a scale-aware threshold, so an outer product (all rows proportional) is rank 1 even with float noise, while a random 10x10 matrix is almost surely full rank. int(...) converts the PyTorch integer for the type check.
