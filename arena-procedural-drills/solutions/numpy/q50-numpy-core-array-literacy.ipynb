## Problem

Write a function solve(n, v) that takes an integer n >= 1 and a number v, and returns an n x n floating-point matrix with v at every position on the main diagonal and 0.0 everywhere else.


<details><summary>💡 Hint (click to reveal)</summary>

Scale the identity matrix, or hand a constant diagonal vector to the diagonal-matrix builder.

</details>

## Reference solution


```python
import numpy as np

def solve(n, v):
    return np.eye(n) * v


print(solve(4, 2.0))
```

## Why this works

np.eye(n) is the n x n identity; multiplying by v scales exactly the diagonal entries since everything else is zero. np.diag(np.full(n, float(v))) constructs the same matrix directly from a constant diagonal vector.
