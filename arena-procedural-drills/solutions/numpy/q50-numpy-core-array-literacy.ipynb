## Problem

Construct a 4x4 square matrix where all the elements on the main diagonal are 2, and all other elements are 0. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Start from an identity matrix and scale it.

## Reference solution


```python
import numpy as np


def solve():
    return np.eye(4) * 2


print(solve())
```

## Why this works

np.eye(4) is the 4x4 identity (1s on the diagonal, 0s elsewhere). Multiplying by 2 scales the diagonal ones to 2 while zeros stay zero.
