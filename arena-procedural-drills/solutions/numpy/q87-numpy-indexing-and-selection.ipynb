## Problem

Write a function solve(z, bins) that takes a 1-D numpy array of floats in [0, 1) and a positive integer bins, and returns a 1-D integer array of length bins counting how many values fall into each of the bins equal-width intervals dividing [0, 1]. Values sitting exactly on an interior edge count toward the higher bin (numpy's convention).


<details><summary>💡 Hint (click to reveal)</summary>

numpy's histogram function does the binning and counting in one call — give it the bin count and the (0, 1) range explicitly.

</details>

## Reference solution


```python
import numpy as np

def solve(z, bins):
    counts, _ = np.histogram(z, bins=bins, range=(0.0, 1.0))
    return counts


example = np.array([0.1, 0.5, 0.55, 0.9])
print(solve(example, 5))
```

## Why this works

np.histogram(z, bins=bins, range=(0.0, 1.0)) splits [0, 1] into equal-width intervals and returns (counts, edges); the function keeps just the counts. Passing range explicitly keeps the bins anchored to [0, 1] even if z doesn't reach the ends.
