## Problem

Given a symmetric 2D array Z and indices `i`, `j` plus a value `v`, return Z with both Z[i, j] and Z[j, i] set to v so the matrix stays symmetric. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Mirror the write: set both Z[i, j] and Z[j, i] to v.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(25).reshape(5, 5)
    Z = Z + Z.T - np.diag(np.diag(Z))
    i, j, v = 2, 3, 42
    Z = Z.copy()
    Z[i, j] = v
    Z[j, i] = v
    return Z

print(solve())
```

## Why this works

To keep a symmetric matrix symmetric, any assignment to Z[i, j] must be mirrored to Z[j, i]. Setting both entries to v in a copy preserves symmetry without the overhead of a custom ndarray subclass.
