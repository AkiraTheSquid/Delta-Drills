## Problem

Write a function solve(z, i, j, v) that takes a SYMMETRIC 2-D PyTorch tensor z, two indices, and a value, and returns a new tensor equal to z except that BOTH z[i, j] and z[j, i] are set to v — preserving symmetry. The input must not be modified. i and j may be equal (the diagonal).


<details><summary>💡 Hint (click to reveal)</summary>

Symmetry is an invariant YOU maintain — write the mirrored cell in the same operation.

</details>

## Reference solution


```python
import torch as t

def solve(z, i, j, v):
    out = z.clone()
    out[i, j] = v
    out[j, i] = v
    return out


print(solve(t.zeros((3, 3)), 0, 2, 5.0))
```

## Why this works

Setting both out[i, j] and out[j, i] keeps the matrix equal to its transpose; when i == j the two writes coincide on the diagonal harmlessly. Doing it on a copy protects the caller's tensor — the first case checks z afterwards.
