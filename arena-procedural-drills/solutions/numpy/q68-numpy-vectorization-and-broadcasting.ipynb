## Problem

Write a function solve(z) that takes a 1-D numpy numeric array and returns a plain Python bool: True when z is non-decreasing (every element is >= the one before it), False otherwise. An array with a single element counts as non-decreasing.


<details><summary>💡 Hint (click to reveal)</summary>

Consecutive differences tell the whole story: the sequence is non-decreasing exactly when none of them is negative.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return bool(np.all(np.diff(z) >= 0))


example = np.array([1, 2, 2, 4, 5])
print(solve(example))
```

## Why this works

np.diff(z) computes each step z[i+1] - z[i]; the array is non-decreasing when all steps are >= 0, which np.all reduces to one boolean. For a single-element array diff is empty and np.all of an empty array is True — matching the required convention.
