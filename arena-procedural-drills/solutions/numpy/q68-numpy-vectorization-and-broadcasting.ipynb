## Problem

You are given Z, a 1D integer array, and is_mono, a boolean flag set to True when every consecutive pair in Z satisfies Z[i+1] >= Z[i] (non-decreasing) and False otherwise. Return is_mono as a single boolean value summarizing monotonicity of the whole sequence. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Look at consecutive differences and check they are all non-negative.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.array([1, 2, 2, 4, 5])
    return np.all(np.diff(Z) >= 0)


print(solve())
```

## Why this works

np.diff gives consecutive differences Z[i+1]-Z[i]; the sequence is non-decreasing exactly when all those differences are >= 0, which np.all reduces to a single boolean.
