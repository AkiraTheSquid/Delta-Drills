## Problem

Write a function solve(rows, cols) that takes two positive integers and returns a rows x cols matrix in which every row contains the values 0 through cols-1 in ascending order — every row is identical, and each entry equals its own column index.


<details><summary>💡 Hint (click to reveal)</summary>

One row is just the 0..cols-1 range — then either tile it down the rows or let broadcasting add it to a zero matrix.

</details>

## Reference solution


```python
import torch as t

def solve(rows, cols):
    return t.tile(t.arange(cols), (rows, 1))


print(solve(5, 5))
```

## Why this works

t.tile(t.arange(cols), (rows, 1)) stacks rows copies of the range vertically. The broadcasting route — t.zeros((rows, cols)) + t.arange(cols) — produces the same values because the length-cols vector broadcasts across every row (as floats there; the tests compare values, and both routes pass).
