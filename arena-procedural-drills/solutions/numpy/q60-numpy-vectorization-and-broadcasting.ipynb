## Problem

Build a 5x5 matrix Z (shape (rows, cols)) in which every row contains the integers 0 through cols-1 in ascending order, so each row is identical and the column index equals the value stored. The five rows are stacked vertically to form the final 5x5 matrix. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcast a 1D arange across the rows of a zeros matrix by adding it.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.zeros((5, 5))
    Z += np.arange(5)
    return Z


print(solve())
```

## Why this works

A (5,5) zero matrix plus a length-5 row vector broadcasts the vector across every row, so each row becomes [0,1,2,3,4] and the column index equals the stored value.
