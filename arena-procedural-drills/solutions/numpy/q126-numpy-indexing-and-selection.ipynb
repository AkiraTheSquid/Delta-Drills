## Problem

Write a function solve(z, bh, bw) that takes a 2-D tensor z whose shape (r, c) is exactly divisible by the block size (bh, bw), and returns a 2-D tensor of shape (r//bh, c//bw) in which each entry is the SUM of the corresponding non-overlapping bh x bw block of z. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Reshape each axis into (blocks, within-block) pairs — then one sum over both within-block axes collapses every block at once.

</details>

## Reference solution


```python
import torch as t

def solve(z, bh, bw):
    r, c = z.shape
    return z.reshape(r // bh, bh, c // bw, bw).sum(dim=(1, 3))


example = t.arange(16).reshape(4, 4)
print(solve(example, 2, 2))
```

## Why this works

Reshaping (r, c) to (r//bh, bh, c//bw, bw) splits each axis into block index and within-block offset without moving data; summing over axes 1 and 3 (the two offsets) leaves one total per block. This reshape-reduce idiom is the standard loop-free pooling pattern.
