## Problem

Given a 2D array Z with shape (4, 6), rearrange it into 2x2 blocks and compute the sum of each block. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Reshape so each spatial axis splits into a block index and an in-block index, then sum the in-block axes.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(4 * 6).reshape(4, 6)
    Z2 = Z.reshape(2, 2, 3, 2)
    return Z2.sum(axis=(1, 3))

print(solve())
```

## Why this works

Reshaping (4, 6) into (2, 2, 3, 2) splits each axis into (block, within-block) pairs: rows become 2 row-blocks of 2, columns become 3 col-blocks of 2. Summing over the two within-block axes (1 and 3) collapses each 2x2 tile to its sum.
