## Problem

Write a function solve(x) that takes a 2-D numpy float array and returns a new array of the same shape in which every entry has had its own COLUMN's mean subtracted, so each column of the result averages to zero. (Contrast: an earlier drill centers each row.) Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Averaging along axis 0 collapses the rows, leaving one mean per column — and that vector broadcasts back over the rows without any reshaping.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return x - x.mean(axis=0)


example = np.array([[1.0, 10.0], [3.0, 20.0]])
print(solve(example))
```

## Why this works

x.mean(axis=0) has shape (cols,), which broadcasts against (rows, cols) by aligning trailing axes — each column's mean is subtracted down that column. Note the asymmetry with row-centering, which needs keepdims (or a reshape) because a (rows,) vector would NOT align correctly; the single-row and single-column cases expose any axis mix-up.
