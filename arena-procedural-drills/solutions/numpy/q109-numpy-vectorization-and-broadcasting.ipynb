## Problem

Given a matrix X, return a new matrix where each element in a row has had the average of that row subtracted from it. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Take the per-row mean while keeping dimensions so it broadcasts back against the original matrix.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.rand(5, 10)
    return X - X.mean(axis=1, keepdims=True)

print(solve())
```

## Why this works

Each row's mean is computed with axis=1, and keepdims=True keeps a length-1 second axis so the (5, 1) means broadcast cleanly against the (5, 10) matrix during subtraction. This centers every row at zero without any explicit loop.
