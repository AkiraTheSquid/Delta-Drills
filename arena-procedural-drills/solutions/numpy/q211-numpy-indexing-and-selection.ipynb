## Problem

Consider two arrays A and B of shape (8,3) and (2,2). How to find rows of A that contain elements of each row of B regardless of the order of the elements in B? (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Add two trailing length-1 axes to A so it broadcasts against B, producing a 4D match tensor. Reduce with any over the right axes to ask 'does this B element appear in the A row', then all to require every B row be covered.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    A = np.random.randint(0, 5, (8, 3))
    B = np.random.randint(0, 5, (2, 2))
    # Broadcast each A element against every B element.
    C = (A[..., np.newaxis, np.newaxis] == B)
    # For each (A-row, B-row): each B element must appear somewhere in the A row.
    rows = np.where(C.any((3, 1)).all(1))[0]
    return rows

print(solve())
```

## Why this works

Reshaping A to (8, 3, 1, 1) and comparing against B (2, 2) broadcasts every element of each A row against every element of each B row, giving a (8, 3, 2, 2) boolean tensor C. C.any((3,1)) asks, for each A row and each B-row position, whether any A-row element matches any element in that B row; .all(1) then requires this to hold for both B rows. np.where extracts the indices of qualifying A rows.
