## Problem

Write a function solve(a, b) that takes a 2-D integer tensor a of shape (ra, ca) and a 2-D integer tensor b of shape (rb, cb), and returns the sorted indices of the rows of a that contain at least one element from EVERY row of b (order within rows irrelevant). Return a 1-D integer tensor of row indices.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcast a 4-D equality cube (a-row, a-col, b-row, b-col), then reduce: ANY match within a b-row, but ALL b-rows must be hit.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    c = a[..., None, None] == b
    return t.where(c.any(dim=(3, 1)).all(dim=1))[0]


a = t.tensor([[0, 1, 2], [3, 4, 5]])
b = t.tensor([[0, 9], [2, 9]])
print(solve(a, b))
```

## Why this works

a[..., None, None] == b compares every element of a with every element of b, shaped (ra, ca, rb, cb). Reducing with any over the b-column and a-column axes asks 'does this a-row touch this b-row at all?', and all over the b-row axis demands every b-row is touched — t.where converts the surviving row mask to indices.
