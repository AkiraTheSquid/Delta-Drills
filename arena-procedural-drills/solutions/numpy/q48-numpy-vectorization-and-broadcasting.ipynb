## Problem

Write a function solve(v, fill) that takes a 1-D numpy integer array v and an integer fill value. It should return a numpy array of the same length in which every element at an odd index (indices 1, 3, 5, ...) has been replaced by fill, while every element at an even index keeps its original value. Do not use Python loops, and leave the caller's array v unmodified.


<details><summary>💡 Hint (click to reveal)</summary>

A strided slice can appear on the left-hand side of an assignment, and a scalar broadcasts into every position that slice selects.

</details>

## Reference solution


```python
import numpy as np

def solve(v, fill):
    out = v.copy()
    out[1::2] = fill
    return out


example = np.array([3, 8, 2, 7, 5, 1])
print(solve(example, -1))
```

## Why this works

Copying v first keeps the caller's array intact; the slice out[1::2] selects every second element starting at index 1, which is exactly the odd indices. Assigning the scalar fill to that slice broadcasts it into all selected slots in one vectorized step, so no loop is needed. For a length-1 array the slice is empty, the assignment does nothing, and the copy comes back unchanged.
