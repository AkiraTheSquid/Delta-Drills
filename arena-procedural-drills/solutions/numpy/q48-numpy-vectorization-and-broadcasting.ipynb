## Problem

Given a vector containing the integers from 0 to 9, modify the elements at odd positions to be -1. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use strided slice assignment starting at index 1 with step 2 to target the odd positions.

</details>

## Reference solution


```python
import numpy as np


def solve():
    Z = np.arange(10)
    Z[1::2] = -1
    return Z


print(solve())
```

## Why this works

np.arange(10) gives 0..9. The slice 1::2 selects every odd index (1,3,5,7,9); assigning -1 broadcasts to all of them in place, leaving even positions untouched.
