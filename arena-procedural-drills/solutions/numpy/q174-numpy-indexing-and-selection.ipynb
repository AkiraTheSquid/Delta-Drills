## Problem

Write a function solve(z) that takes a 2-D float array with an ODD number of columns and returns a 1-D array holding each row's MEDIAN — the middle value of that row when sorted.


<details><summary>💡 Hint (click to reveal)</summary>

With an odd row length the median is one specific order statistic — a partial sort (partition) places exactly that element without sorting the whole row.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    m = z.shape[1] // 2
    part = np.partition(z, m, axis=1)
    return part[:, m]


print(solve(np.array([[3.0, 1.0, 2.0]])))
```

## Why this works

np.partition(z, m, axis=1) rearranges each row so the element at position m is the one that WOULD be there in full sorted order — the middle element, since columns are odd. Reading column m gives every row's median in O(n) per row; np.median(z, axis=1) is the direct call and grades identically.
