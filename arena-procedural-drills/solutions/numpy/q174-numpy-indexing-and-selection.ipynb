## Problem

Given a 2D array Z with an odd number of columns, return the median value from each row as a 1D array. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

The median of an odd-length row is the middle order statistic; np.partition places exactly that element at index cols//2.

</details>

## Reference solution


```python
import numpy as np

def solve():
    m = Z.shape[1] // 2
    part = np.partition(Z, m, axis=1)
    return part[:, m]
```

## Why this works

With an odd number of columns the median is the middle element after sorting. np.partition is cheaper than a full sort: it guarantees the element at position m is the one that would be there if sorted, so part[:, m] is the row-wise median.
