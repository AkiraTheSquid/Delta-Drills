## Problem

Build Z, a 1D floating-point array of shape (5,) containing 5 evenly spaced values that start at 0 and end at 1 inclusive, with equal spacing between consecutive entries. The first entry must be exactly 0.0, the last entry must be exactly 1.0, and entries in between fill in at uniform steps. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

One numpy call gives evenly spaced values with both endpoints included.

</details>

## Reference solution


```python
import numpy as np


def solve():
    return np.linspace(0, 1, 5)


print(solve())
```

## Why this works

np.linspace(start, stop, num) produces num evenly spaced samples including both endpoints, so it gives exactly 0.0 and 1.0 at the ends with uniform 0.25 steps between.
