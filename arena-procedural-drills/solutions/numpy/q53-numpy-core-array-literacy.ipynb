## Problem

Write a function solve(n) that takes an integer n >= 2 and returns a 1-D floating-point array of n evenly spaced values that starts at exactly 0.0 and ends at exactly 1.0, with equal spacing between consecutive entries.


<details><summary>💡 Hint (click to reveal)</summary>

## Reference solution


```python
import numpy as np

def solve(n):
    return np.linspace(0.0, 1.0, n)


print(solve(5))
```

## Why this works

np.linspace(0.0, 1.0, n) places n points with equal gaps, including both endpoints exactly. Building it as np.arange(n) / (n - 1) is equivalent to within rounding; the tests round to 10 decimals so either construction passes.
