## Problem

Given a 2D integer array Z of shape (10, 3), return the subarray containing only those rows whose 3 entries are not all equal to each other (i.e., drop any row in which all three columns hold the same value). The result is 2D with shape (k, 3) where k is the number of surviving rows. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

A row is all-equal iff its row max equals its row min; keep the rows where they differ via a boolean mask.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 5, (10, 3))
    U = Z[Z.max(axis=1) != Z.min(axis=1), :]
    return U

print(solve())
```

## Why this works

A row has all three entries equal exactly when its maximum equals its minimum, so the rows to keep are those where max and min along axis 1 differ. Boolean-indexing Z with that per-row mask drops the constant rows and returns the surviving (k, 3) subarray.
