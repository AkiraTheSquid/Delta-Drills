## Problem

Write a function solve(z) that takes a 2-D PyTorch integer tensor and returns the subarray of rows that are NOT constant — rows containing at least two different values. Rows where every entry is equal are dropped; the rest keep their order.


<details><summary>💡 Hint (click to reveal)</summary>

A row is constant exactly when every entry equals its FIRST entry — compare against the first column and reduce.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    mask = (z != z[:, :1]).any(dim=1)
    return z[mask]


print(solve(t.tensor([[1, 1, 1], [1, 2, 3]])))
```

## Why this works

z[:, :1] keeps the first column as an (n, 1) slice that broadcasts across each row; (z != it).any(axis=1) flags rows where anything deviates from the row's first value. Comparing max against min per row is an equivalent constancy test for numeric dtypes.
