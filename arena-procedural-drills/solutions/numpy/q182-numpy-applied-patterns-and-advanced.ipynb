## Problem

Given a 2D array pts of shape (5, 2) holding 5 points in the plane (each row is an (x, y) pair) and a 3x3 homogeneous transform matrix T, return a (5, 2) array of the transformed points, where each output row is the (x, y) image of the corresponding input point under T applied in homogeneous coordinates. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Augment the points with a ones column for homogeneous coordinates, multiply by T.T, then keep the first two columns.

</details>

## Reference solution


```python
import numpy as np

def solve():
    H = np.c_[pts, np.ones(len(pts))]
    return (H @ T.T)[:, :2]
```

## Why this works

Appending a column of ones turns each 2D point into homogeneous coordinates (x, y, 1). Multiplying the (5, 3) matrix by T.T applies the transform to every row at once, and slicing the first two columns drops the homogeneous coordinate to recover the transformed (x, y) points.
