## Problem

Write a function solve(pts, m) that takes an (n, 2) tensor of 2-D points and a 3x3 homogeneous transform matrix m, and returns the (n, 2) tensor of transformed points: append a 1 to each point, multiply by m, then divide the first two coordinates by the resulting third (homogeneous divide).

## Reference solution


```python
import torch as t
def solve(pts, m):
    h = t.cat([pts, t.ones(len(pts), 1)], dim=1)
    out = h @ m.T
    return out[:, :2] / out[:, 2:3]


shift = t.tensor([[1.0, 0.0, 2.0], [0.0, 1.0, -1.0], [0.0, 0.0, 1.0]])
print(solve(t.tensor([[0.0, 0.0]]), shift))
```
