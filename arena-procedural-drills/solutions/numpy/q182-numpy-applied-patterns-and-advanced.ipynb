## Problem

Write a function solve(pts, t) that takes an (n, 2) array of 2-D points and a 3x3 homogeneous transform matrix t, and returns the (n, 2) array of transformed points: append a 1 to each point, multiply by t, then divide the first two coordinates by the resulting third (homogeneous divide).


<details><summary>💡 Hint (click to reveal)</summary>

Row-vector points need t TRANSPOSED in the product — and keep the divisor 2-D (a column) so it broadcasts over both coordinates.

</details>

## Reference solution


```python
import numpy as np

def solve(pts, t):
    h = np.c_[pts, np.ones(len(pts))]
    out = h @ t.T
    return out[:, :2] / out[:, 2:3]


shift = np.array([[1.0, 0.0, 2.0], [0.0, 1.0, -1.0], [0.0, 0.0, 1.0]])
print(solve(np.array([[0.0, 0.0]]), shift))
```

## Why this works

np.c_ appends the homogeneous 1s; because points sit in rows, the transform applies as h @ t.T. The perspective divide out[:, :2] / out[:, 2:3] uses a (n, 1) slice so it broadcasts across x and y — the last case's w = 2 row shows the divide actually happening.
