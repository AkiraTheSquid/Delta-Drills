## Problem

Write a function solve(x, n_samples, rng) that takes a 1-D float tensor x, a bootstrap sample count, and a PyTorch random Generator. Draw n_samples bootstrap resamples using EXACTLY rng.integers(0, x.size, (n_samples, x.size)) as the index matrix, compute each resample's mean, and return the (2.5th, 97.5th) percentiles of those means as a tuple — the 95% bootstrap confidence interval for the mean.

## Reference solution


```python
import torch as t
def solve(x, n_samples, rng):
    idx = t.randint(0, x.numel(), (n_samples, x.numel()), generator=rng)
    means = x[idx].mean(dim=1)
    qs = t.tensor([0.025, 0.975], dtype=means.dtype)
    lo, hi = t.quantile(means, qs)
    return lo, hi


print(solve(t.arange(10, dtype=t.float32), 100, t.Generator().manual_seed(0)))
```
