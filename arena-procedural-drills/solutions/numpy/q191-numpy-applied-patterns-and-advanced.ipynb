## Problem

Write a function solve(x, n_samples, rng) that takes a 1-D float array x, a bootstrap sample count, and a numpy random Generator. Draw n_samples bootstrap resamples using EXACTLY rng.integers(0, x.size, (n_samples, x.size)) as the index matrix, compute each resample's mean, and return the (2.5th, 97.5th) percentiles of those means as a tuple — the 95% bootstrap confidence interval for the mean.


<details><summary>💡 Hint (click to reveal)</summary>

One (n_samples, n) integer index matrix fancy-indexes ALL the resamples at once — then a row-mean and two percentiles finish the interval.

</details>

## Reference solution


```python
import numpy as np

def solve(x, n_samples, rng):
    idx = rng.integers(0, x.size, (n_samples, x.size))
    means = x[idx].mean(axis=1)
    lo, hi = np.percentile(means, [2.5, 97.5])
    return lo, hi


print(solve(np.arange(10, dtype=float), 100, np.random.default_rng(0)))
```

## Why this works

x[idx] with a 2-D index matrix materializes every bootstrap resample as a row; mean(axis=1) gives the resampled means and their 2.5/97.5 percentiles bracket the central 95%. The exact rng call is prescribed so the deterministic stream matches the grader; a constant array collapses the interval to a point.
