## Problem

Given a 1D array X, perform resampling with replacement to create multiple samples, calculate the mean for each sample, and then determine the 95% confidence interval for these means using percentiles. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Generate an (N, len(X)) matrix of random integer indices to fancy-index X all at once, take row means, then read off the 2.5 and 97.5 percentiles.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.randn(100)
    N = 1000
    idx = np.random.randint(0, X.size, (N, X.size))
    means = X[idx].mean(axis=1)
    confint = np.percentile(means, [2.5, 97.5])
    return confint

print(solve())
```

## Why this works

Bootstrapping resamples the data with replacement N times; here an (N, len(X)) integer index array drawn uniformly with np.random.randint selects N resamples in one vectorized fancy-index, and the row means form the bootstrap distribution of the sample mean. The 2.5th and 97.5th percentiles of those means give a 95% confidence interval.
