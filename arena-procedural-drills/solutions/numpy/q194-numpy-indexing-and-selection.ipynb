## Problem

Given a 2D array Z, return the indices of the k largest values in each row, where k is specified. The output should be sorted in descending order based on the values. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

argpartition gets the top-k indices cheaply; then argsort the corresponding values per row and reverse to reorder those indices descending.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(5, 10)
    k = 3
    idx = np.argpartition(Z, -k, axis=1)[:, -k:]
    rows = np.arange(Z.shape[0])[:, None]
    idx_sorted = idx[rows, np.argsort(Z[rows, idx], axis=1)[:, ::-1]]
    return idx_sorted

print(solve())
```

## Why this works

argpartition isolates the top-k column indices per row in unspecified order, so a second pass sorts them by their actual values. Gathering Z[rows, idx] gives the top-k values, argsort orders them ascending, and reversing produces descending order; reindexing idx with that permutation yields the indices sorted by value.
