## Problem

Write a function solve(z, k) that takes a 2-D float array with distinct values within each row and returns an (rows, k) integer array where row i holds the column indices of that row's k largest values, ordered by DESCENDING value (largest first).


<details><summary>💡 Hint (click to reveal)</summary>

argpartition isolates each row's top-k cheaply but unordered — gather those k values and argsort just them, descending, to finish.

</details>

## Reference solution


```python
import numpy as np

def solve(z, k):
    idx = np.argpartition(z, -k, axis=1)[:, -k:]
    rows = np.arange(z.shape[0])[:, None]
    order = np.argsort(z[rows, idx], axis=1)[:, ::-1]
    return idx[rows, order]


print(solve(np.array([[0.4, 0.9, 0.1]]), 2))
```

## Why this works

The partition step finds WHICH k columns matter per row in O(cols); the small per-row argsort over only those k values (reversed for descending) supplies the order. z[rows, idx] uses a broadcast row column to gather per-row selections — the same paired-indexing pattern as the top-k mask drill. Full np.argsort(z)[:, ::-1][:, :k] is the simpler equivalent.
