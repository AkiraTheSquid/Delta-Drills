## Problem

Write a function solve(z, n) that takes a 1-D numeric array with distinct values and returns the n LARGEST values (the values themselves, not indices) in ASCENDING order, as efficiently as a partial sort allows.


<details><summary>💡 Hint (click to reveal)</summary>

Partition against the n-th-from-last position — the top n land in the tail unordered, and a tiny sort of just those n finishes.

</details>

## Reference solution


```python
import numpy as np

def solve(z, n):
    return np.sort(np.partition(z, -n)[-n:])


print(solve(np.array([5, 1, 9, 3, 7]), 2))
```

## Why this works

np.partition(z, -n) rearranges so the last n slots hold the n largest values (order arbitrary) in O(len) time; sorting only that n-element tail is cheap. Full np.sort(z)[-n:] gives the same answer at O(n log n) for the whole array — the partition route is the scalable idiom.
