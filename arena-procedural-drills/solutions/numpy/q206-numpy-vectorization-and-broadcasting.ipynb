## Problem

Write a function solve(z, n) that takes a 1-D numeric tensor with distinct values and returns the n LARGEST values (the values themselves, not indices) in ASCENDING order, as efficiently as a partial sort allows.

## Reference solution


```python
import torch as t
def solve(z, n):
    return t.topk(z, n).values.flip(0)


print(solve(t.tensor([5, 1, 9, 3, 7]), 2))
```
