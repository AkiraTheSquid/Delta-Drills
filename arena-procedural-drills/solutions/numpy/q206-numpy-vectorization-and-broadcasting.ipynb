## Problem

You are given Z, a 1D integer array of shape (10000,) containing a random permutation of the values 0 through 9999, and an integer n (here 5). Return a 1D array of length n containing the n largest values of Z. The output ordering need only contain those n values; their internal order is not constrained. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Partition on the negated array so the largest values' indices land first, then take the first n and index back into the array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.arange(10000)
    np.random.shuffle(Z)
    n = 5
    return print(Z[np.argpartition(-Z, n)[:n]])

solve()
```

## Why this works

argpartition on -Z places the n indices of the largest values into the first n slots (in unspecified order), so slicing [:n] and gathering from Z returns the n largest values without a full sort. This is the O(N) approach versus the O(N log N) argsort-based alternative.
