## Problem

Given the 1D integer array Z = np.arange(5), produce a new array with the same shape (5,) and the same numerical values along its single axis, but with element dtype float32 instead of integer (so the new dtype is a 32-bit floating-point type and the values themselves are preserved as floats). (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Build the integer range, then convert dtype with the astype method specifying the 32-bit float type.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(5)
    return Z.astype(np.float32)

print(solve())
```

## Why this works

arange(5) yields integers 0..4; calling .astype(np.float32) returns a new array with the same shape and values converted to 32-bit floating point, leaving the original untouched.
