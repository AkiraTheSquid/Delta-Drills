## Problem

Write a function solve(x) that takes a 1-D numpy array of integers and returns a new array with the same shape and the same numerical values, but whose element dtype is float32. The original array must not be modified. The grader checks both the values and the dtype of the array you return.

numpy arrays have a method that returns a copy of the array converted to whatever dtype you ask for.

## Reference solution


```python
import numpy as np

def solve(x):
    return x.astype(np.float32)


example = np.array([1, 2, 3])
print(solve(example))
```

## Why this works

x.astype(np.float32) allocates a fresh array with the requested dtype and copies the values in, so the original array is left untouched. The integer values here are small enough to be represented exactly in float32, so only the dtype changes while shape and values stay identical. Simply returning x would fail every case, because the grader compares str(result.dtype) and an integer array reports int64 or int32, not float32.
