## Problem

Write a function solve(x) that takes a 1-D PyTorch tensor of integers and returns a new tensor with the same shape and the same numerical values, but whose element dtype is float32. The original tensor must not be modified. The grader checks both the values and the dtype of the tensor you return.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch tensors have a method that returns a copy of the tensor converted to whatever dtype you ask for.

## Reference solution


```python
import torch as t

def solve(x):
    return x.to(t.float32)


example = t.tensor([1, 2, 3])
print(solve(example))
```

## Why this works

x.astype(t.float32) allocates a fresh tensor with the requested dtype and copies the values in, so the original tensor is left untouched. The integer values here are small enough to be represented exactly in float32, so only the dtype changes while shape and values stay identical. Simply returning x would fail every case, because the grader compares str(result.dtype) and an integer tensor reports int64 or int32, not float32.
