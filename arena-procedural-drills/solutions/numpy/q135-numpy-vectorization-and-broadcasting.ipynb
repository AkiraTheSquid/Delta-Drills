## Problem

Write a function solve(x) that takes a 4-D numpy integer array x of shape (a, b, c, d) and returns a 2-D array of shape (a, b). Each entry [i, j] of the result must equal the total of all c*d elements in the slice x[i, j]. Your function must work for any 4-D shape, including dimensions of size 1, and must not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

numpy reductions accept a tuple of axes, so you can collapse several dimensions in one call.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return x.sum(axis=(-2, -1))


example = np.arange(24).reshape(2, 3, 2, 2)
print(solve(example))
```

## Why this works

Passing a tuple of axes to a reduction sums over all of those axes at once: x.sum(axis=(-2, -1)) collapses the last two dimensions and leaves the first two intact, giving an (a, b) result. Using negative axis indices makes the code work for any 4-D shape without naming the leading dimensions. This also handles size-1 axes cleanly, where each output entry is just the single element in its slice.
