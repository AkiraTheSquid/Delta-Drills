## Problem

Considering a four dimensions array, how to get sum over the last two axis at once? (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

sum accepts a tuple of axes, so you can reduce the last two dimensions in a single call.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    A = np.random.randint(0, 10, (3, 4, 3, 4))
    s = A.sum(axis=(-2, -1))
    return s

print(solve())
```

## Why this works

Passing a tuple of axes to sum reduces over multiple dimensions at once, here the last two, collapsing the (3,4,3,4) array to (3,4). This is equivalent to flattening the trailing two axes and summing the resulting last axis, but the tuple form is more direct.
