## Problem

Write a function solve(z, bh, bw) that takes a 2-D integer tensor and a window shape (bh, bw), and returns a 2-D tensor of shape (rows - bh + 1, cols - bw + 1) whose entry [i, j] is the SUM of the OVERLAPPING bh x bw window anchored at z[i, j]. Do not use Python loops.

## Reference solution


```python
import torch as t
def solve(z, bh, bw):
    return z.unfold(0, bh, 1).unfold(1, bw, 1).sum(dim=(2, 3))


print(solve(t.arange(9).reshape(3, 3), 2, 2))
```
