## Problem

Write a function solve(z, bh, bw) that takes a 2-D integer array and a window shape (bh, bw), and returns a 2-D array of shape (rows - bh + 1, cols - bw + 1) whose entry [i, j] is the SUM of the OVERLAPPING bh x bw window anchored at z[i, j]. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

A sliding-window view exposes every window as trailing axes — one sum over those two axes collapses each window.

</details>

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve(z, bh, bw):
    return sliding_window_view(z, (bh, bw)).sum(axis=(2, 3))


print(solve(np.arange(9).reshape(3, 3), 2, 2))
```

## Why this works

sliding_window_view(z, (bh, bw)) has shape (out_h, out_w, bh, bw): the last two axes ARE the window contents. Summing over axes (2, 3) reduces each window to its total. Unlike the non-overlapping reshape trick, overlapping windows need this view (or convolution) because each input cell belongs to many windows.
