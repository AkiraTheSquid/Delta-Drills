## Problem

Given a 2D array Z, extract all overlapping subarrays of shape (2, 4) and return a new array containing the sum of each subarray. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Build the 2x4 sliding windows, then reduce over the two trailing window axes with .sum(axis=(-2, -1)).

</details>

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve():
    Z = np.arange(8 * 10).reshape(8, 10)
    W = sliding_window_view(Z, (2, 4))
    return W.sum(axis=(-2, -1))

print(solve())
```

## Why this works

sliding_window_view with window (2, 4) yields a (7, 7, 2, 4) array of all overlapping 2x4 blocks. Summing over the last two axes collapses each block to its scalar sum, giving the (7, 7) array of window sums in one vectorized pass.
