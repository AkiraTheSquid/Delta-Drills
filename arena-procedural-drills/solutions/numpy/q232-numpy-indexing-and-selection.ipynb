## Problem

Write a function solve(x, d) that takes a 1-D PyTorch tensor of integers x and a positive integer d, and returns a boolean PyTorch tensor of the same shape. Entry i of the result must be True exactly when x[i] is divisible by d, and False otherwise. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Elementwise arithmetic on a PyTorch tensor gives each element's remainder at once, and comparing that whole result to a scalar yields a boolean tensor.

</details>

## Reference solution


```python
import torch as t

def solve(x, d):
    return x % d == 0


example = t.tensor([3, 5, 9, 12, 14])
print(solve(example, 3))
```

## Why this works

x % d computes the remainder of every element in one vectorized step, producing an integer tensor the same shape as x. Comparing that tensor to 0 with == is also elementwise, so it yields a boolean tensor that is True wherever the remainder is zero, i.e. wherever the element is divisible by d. PyTorch's modulo returns a value in [0, d) for positive d, so negative multiples like -5 with d = 5 are still correctly marked True.
