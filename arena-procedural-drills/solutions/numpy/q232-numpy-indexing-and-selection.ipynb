## Problem

You are given Z, a 1D integer array of shape (30,) containing the values 0 through 29 in order. Return a 1D boolean array of the same shape whose i-th entry is True exactly when Z[i] is divisible by 3. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Take the remainder of each element under division by 3 and compare it elementwise to zero.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(30)
    return Z % 3 == 0

print(solve())
```

## Why this works

np.arange(30) gives 0..29, and the elementwise modulo Z % 3 returns each remainder; comparing to 0 yields a boolean array that is True exactly where the value is a multiple of 3. The whole operation is vectorized, so no Python loop is needed.
