## Problem

Write a function solve(x, d) that takes a 1-D numpy array of integers x and a positive integer d, and returns a boolean numpy array of the same shape. Entry i of the result must be True exactly when x[i] is divisible by d, and False otherwise. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Elementwise arithmetic on a numpy array gives each element's remainder at once, and comparing that whole result to a scalar yields a boolean array.

</details>

## Reference solution


```python
import numpy as np

def solve(x, d):
    return x % d == 0


example = np.array([3, 5, 9, 12, 14])
print(solve(example, 3))
```

## Why this works

x % d computes the remainder of every element in one vectorized step, producing an integer array the same shape as x. Comparing that array to 0 with == is also elementwise, so it yields a boolean array that is True wherever the remainder is zero, i.e. wherever the element is divisible by d. numpy's modulo returns a value in [0, d) for positive d, so negative multiples like -5 with d = 5 are still correctly marked True.
