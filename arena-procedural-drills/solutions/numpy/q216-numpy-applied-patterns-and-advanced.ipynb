## Problem

Write a function solve(c) that takes a 1-D integer tensor c interpreted as COUNTS — c[k] is how many times the integer k should appear — and returns the sorted tensor containing each k repeated c[k] times: the inverse of t.bincount, so that t.bincount(solve(c), minlength=len(c)) equals c.


<details><summary>💡 Hint (click to reveal)</summary>

The values are just 0..len(c)-1 — repeat each one by its own count.

</details>

## Reference solution


```python
import torch as t
def solve(c):
    return t.repeat_interleave(t.arange(len(c)), c)


print(solve(t.tensor([0, 2, 1, 0, 2])))
```
