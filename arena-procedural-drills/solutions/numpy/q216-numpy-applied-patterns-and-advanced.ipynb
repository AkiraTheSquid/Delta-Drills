## Problem

Write a function solve(c) that takes a 1-D integer array c interpreted as COUNTS — c[k] is how many times the integer k should appear — and returns the sorted array containing each k repeated c[k] times: the inverse of np.bincount, so that np.bincount(solve(c), minlength=len(c)) equals c.


<details><summary>💡 Hint (click to reveal)</summary>

The values are just 0..len(c)-1 — repeat each one by its own count.

</details>

## Reference solution


```python
import numpy as np

def solve(c):
    return np.repeat(np.arange(len(c)), c)


print(solve(np.array([0, 2, 1, 0, 2])))
```

## Why this works

np.repeat(np.arange(len(c)), c) emits integer k exactly c[k] times, in ascending order — precisely inverting np.bincount. Zero counts skip their value, as the round-trip test confirms.
