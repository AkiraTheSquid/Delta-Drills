## Problem

Given a sorted array C that represents the counts of each integer, return an array A that contains each integer repeated according to its count in C. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

The values to emit are just the indices 0..len(C)-1; repeat each index by its corresponding count using np.repeat.

</details>

## Reference solution


```python
import numpy as np

def solve():
    C = np.bincount([1, 1, 2, 3, 4, 4, 6])
    A = np.repeat(np.arange(len(C)), C)
    return A

print(solve())
```

## Why this works

C is a bincount, so C[k] is how many times integer k should appear. np.repeat takes the value array arange(len(C)) (the integers 0..len(C)-1) and repeats each one C[k] times, reconstructing the original multiset such that np.bincount(A) == C.
