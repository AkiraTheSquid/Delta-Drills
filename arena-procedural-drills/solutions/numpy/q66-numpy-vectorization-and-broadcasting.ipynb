## Problem

Write a function solve(z) that takes a 2-D tensor of shape (n, 2) whose rows are cartesian (x, y) coordinates and returns a tuple of two 1-D tensors of length n: the polar radii r = sqrt(x^2 + y^2) and the polar angles theta = atan2(y, x) in radians.


<details><summary>💡 Hint (click to reveal)</summary>

Slice the two coordinate columns apart, then use the two-argument arctangent — it gets the quadrant right where a plain arctan of y/x cannot.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    x, y = z[:, 0], z[:, 1]
    r = t.sqrt(x ** 2 + y ** 2)
    theta = t.arctan2(y, x)
    return r, theta


example = t.tensor([[1.0, 0.0], [0.0, 2.0]])
print(solve(example))
```
