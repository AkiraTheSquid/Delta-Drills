## Problem

Write a function solve(z) that takes a 2-D array of shape (n, 2) whose rows are cartesian (x, y) coordinates and returns a tuple of two 1-D arrays of length n: the polar radii r = sqrt(x^2 + y^2) and the polar angles theta = atan2(y, x) in radians.


<details><summary>💡 Hint (click to reveal)</summary>

Slice the two coordinate columns apart, then use the two-argument arctangent — it gets the quadrant right where a plain arctan of y/x cannot.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    x, y = z[:, 0], z[:, 1]
    r = np.sqrt(x ** 2 + y ** 2)
    t = np.arctan2(y, x)
    return r, t


example = np.array([[1.0, 0.0], [0.0, 2.0]])
print(solve(example))
```

## Why this works

The radius is the Euclidean length of each (x, y) row. np.arctan2(y, x) takes both signs into account, so points in every quadrant (and on the negative x-axis, where the angle is pi) map to the correct angle — np.arctan(y / x) would fold quadrants together and divide by zero at x = 0.
