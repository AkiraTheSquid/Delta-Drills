Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)
Print the result.

## Reference solution


```python
import numpy as np

np.random.seed(0)


def solve():
    np.random.seed(0)
    Z = np.random.random((10, 2))
    X, Y = Z[:, 0], Z[:, 1]
    T = np.arctan2(Y, X)
    return T


print(solve())
```

## Why this works

Cartesian (x,y) maps to polar via R = sqrt(x^2+y^2) and the angle T = arctan2(y, x), which picks the correct quadrant. The grader's expected value is the angle array T. Seed 0 reproduces the coordinates.
