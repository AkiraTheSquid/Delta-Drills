## Problem

Given a 5x5 array A, exchange the positions of the first two rows and return the modified array. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Assign A[[1,0]] into A[[0,1]] — fancy indexing copies the source rows first, so the swap is atomic.

</details>

## Reference solution


```python
import numpy as np

def solve():
    A = np.arange(25).reshape(5, 5)
    A[[0, 1]] = A[[1, 0]]
    return A

print(solve())
```

## Why this works

Fancy indexing on both sides swaps rows in one statement: the right-hand side A[[1,0]] materializes a copy of rows 1 and 0, which is then assigned back into positions 0 and 1. Because the source is copied before assignment, the swap is safe and needs no temporary.
