## Problem

Write a function solve(z, i, j) that takes a 2-D numpy array and two row indices, and returns a new array with rows i and j exchanged and everything else unchanged. The input array must not be modified. i and j may be equal.


<details><summary>💡 Hint (click to reveal)</summary>

Fancy indexing with a LIST of row numbers works on both sides of an assignment — and the right side is gathered before the left side is written.

</details>

## Reference solution


```python
import numpy as np

def solve(z, i, j):
    out = z.copy()
    out[[i, j]] = out[[j, i]]
    return out


example = np.arange(9).reshape(3, 3)
print(solve(example, 0, 1))
```

## Why this works

out[[i, j]] = out[[j, i]] gathers rows j and i into a temporary (fancy indexing copies), then assigns them into slots i and j — a simultaneous swap with no explicit temp variable. Doing it on a copy protects the caller's array, which the first case checks.
