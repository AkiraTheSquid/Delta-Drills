## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns a 1-D array with one entry per row: the column index of the row's FIRST nonzero element, or -1 if the row contains only zeros.


<details><summary>💡 Hint (click to reveal)</summary>

On a boolean row, argmax stops at the first True — but an all-False row ALSO returns 0, so you need a second reduction to tell those apart.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    mask = z != 0
    return np.where(mask.any(axis=1), mask.argmax(axis=1), -1)


example = np.array([[0, 0, 3], [0, 2, 0], [0, 0, 0]])
print(solve(example))
```

## Why this works

mask.argmax(axis=1) returns the first True position per row because True > False, but it yields 0 for all-zero rows too. mask.any(axis=1) distinguishes the two cases, and np.where merges them: the real index where a nonzero exists, -1 otherwise. The all-zero rows are exactly the trap this drill teaches.
