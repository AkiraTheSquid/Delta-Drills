## Problem

Given a 2D array Z, return the index of the first nonzero element in each row, or -1 for rows that contain only zeros. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Build a boolean mask of nonzeros; argmax on it finds the first True per row, and any() tells you which rows should fall back to -1.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([[0, 0, 3], [0, 2, 0], [0, 0, 0]])
    mask = Z != 0
    idx = np.where(mask.any(axis=1), mask.argmax(axis=1), -1)
    return idx

print(solve())
```

## Why this works

A boolean mask of nonzero entries lets argmax along each row return the position of the first True (since argmax returns the first maximal index). Rows that are entirely zero have no True, so any(axis=1) flags them and np.where overrides their result with -1.
