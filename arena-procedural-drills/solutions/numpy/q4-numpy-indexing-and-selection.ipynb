## Problem

Construct a 5x5 matrix where the values 1, 2, 3, and 4 are placed just below the main diagonal, with all other elements being zero. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

np.diag can build a matrix from a vector placed on an offset diagonal; a negative offset moves it below the main diagonal.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.diag(1 + np.arange(4), k=-1)
    return Z

print(solve())
```

## Why this works

np.diag(v, k) places the vector v on the k-th diagonal of an otherwise-zero square matrix; k=-1 is the first sub-diagonal (just below the main diagonal). The vector 1,2,3,4 of length 4 therefore lands below the diagonal of a 5x5 matrix.
