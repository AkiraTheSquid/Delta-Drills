## Problem

Write a function solve(n) that takes an integer n >= 1 and returns an n x n integer matrix that is all zeros except immediately below the main diagonal, where the values 1, 2, ..., n-1 appear in order. For n = 1 there is no room below the diagonal, so the result is just [[0]].

The function that builds a matrix from a diagonal vector takes an offset argument that shifts which diagonal the vector lands on.

## Reference solution


```python
import numpy as np

def solve(n):
    return np.diag(1 + np.arange(n - 1), k=-1)


print(solve(5))
```

## Why this works

np.diag(v, k=-1) places vector v on the diagonal one step below the main one, zero-filling everything else. With v = 1 + np.arange(n - 1) the sub-diagonal reads 1..n-1; for n = 1 the vector is empty, giving the 1x1 zero matrix.
