## Problem

Build a 4x4 2D array where every entry equals its own row index, so each row holds a single repeated integer value and that value equals the row's position from top to bottom. The result must have integer dtype and shape (4, 4). (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

np.indices produces coordinate grids; the first of the returned grids holds the row index at every position.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.indices((4, 4))[0]
    return Z

print(solve())
```

## Why this works

np.indices((4,4)) returns a stack of two grids; element [0] is the array of row indices, where every entry equals its own row position, giving each row a single repeated integer. The dtype is integer and the shape is (4,4) as required.
