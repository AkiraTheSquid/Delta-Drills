## Problem

Write a function solve(rows, cols) that takes two positive integers and returns a rows x cols integer matrix in which every entry equals its own row index: the top row is all 0s, the next row all 1s, and so on. Do not use Python loops.

## Reference solution


```python
import numpy as np

def solve(rows, cols):
    return np.indices((rows, cols))[0]


print(solve(4, 4))
```

## Why this works

np.indices((rows, cols))[0] is the row-index grid: entry [i, j] equals i. Broadcasting np.arange(rows)[:, None] against a (rows, cols) shape (e.g. with np.broadcast_to or multiplication by ones) builds the same integer matrix.
