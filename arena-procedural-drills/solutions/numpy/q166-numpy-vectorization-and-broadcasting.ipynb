## Problem

You are given Z, a 1D integer array of shape (20,) containing the values 0 through 19 in order, and a window size n (here 3). Return a 1D array of length 20 - n + 1 whose i-th entry is the arithmetic mean of the n consecutive entries of Z starting at index i. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

numpy.lib.stride_tricks.sliding_window_view turns the 1D array into a 2D array of overlapping windows; reduce along the window axis.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(20)
    return np.lib.stride_tricks.sliding_window_view(Z, window_shape=3).mean(axis=-1)

print(solve())
```

## Why this works

A sliding window view exposes every length-3 contiguous slice of Z as a (18, 3) array without copying data, then taking the mean along the last axis gives the moving average in one vectorized step. This is equivalent to the cumulative-sum trick but reads more directly.
