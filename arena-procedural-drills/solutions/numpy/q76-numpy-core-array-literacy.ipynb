## Problem

Write a function solve(z) that takes a 2-D numpy array and returns a tuple of two new arrays: first, z mirrored left-right (each row reversed); second, z mirrored top-bottom (the order of the rows reversed). The input must not be modified.


<details><summary>💡 Hint (click to reveal)</summary>

Reversing a slice along one axis mirrors the array across the other — pick the axis carefully for each of the two mirrors.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return z[:, ::-1], z[::-1, :]


example = np.arange(6).reshape(2, 3)
print(solve(example))
```

## Why this works

z[:, ::-1] reverses within each row (a left-right mirror), while z[::-1, :] reverses the row order (a top-bottom mirror). np.fliplr and np.flipud are the named equivalents; slicing produces views without touching the original data.
