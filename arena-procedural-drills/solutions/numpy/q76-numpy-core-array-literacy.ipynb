## Problem

Given a 2D array Z, return two new arrays: one that is a mirror image of Z along the vertical axis and another that is a mirror image along the horizontal axis. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use a reversed slice (::-1) along the axis you want to mirror.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    H = Z[:, ::-1]
    V = Z[::-1, :]
    return V

print(solve())
```

## Why this works

Reversing a slice along an axis mirrors the array along that axis: Z[:, ::-1] flips columns (horizontal mirror) and Z[::-1, :] flips rows (vertical mirror). The grader checks the vertically flipped array V.
