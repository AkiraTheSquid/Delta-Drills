## Problem

Write a function solve(z, target) that takes a 1-D numpy float array and a scalar, and returns the INDEX (a plain Python int) of the entry closest to target — the position minimizing the absolute difference. On a tie, return the smallest such index. (Contrast: an earlier drill asks for the closest VALUE; this one wants where it lives.)


<details><summary>💡 Hint (click to reveal)</summary>

Distance-to-target is an array too — you want the position of its minimum, converted to a plain int.

</details>

## Reference solution


```python
import numpy as np

def solve(z, target):
    return int(np.abs(z - target).argmin())


print(solve(np.array([0.1, 0.4, 0.8]), 0.5))
```

## Why this works

np.abs(z - target) turns the question into 'where is the smallest value', which argmin answers with the first index on ties. int(...) converts the numpy integer to the plain Python int the grader type-checks.
