## Problem

Given a 1D vector Z of 10 real numbers and a scalar target z = 0.5, return the single entry of Z whose value is closest to z (the entry whose absolute difference from z is the smallest across the whole vector). (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Take the absolute differences from the target, find the argmin, and index back into the array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.uniform(0, 1, 10)
    z = 0.5
    return Z.flat[np.abs(Z - z).argmin()]

print(solve())
```

## Why this works

The entry closest to z is the one with the smallest absolute difference |Z - z|. argmin gives the index of that minimum distance, and indexing Z (via its flat view) at that index returns the nearest value itself rather than the distance.
