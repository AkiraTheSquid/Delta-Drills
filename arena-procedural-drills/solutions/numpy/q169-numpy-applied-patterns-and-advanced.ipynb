## Problem

You are given X, a 2D array of shape (6, 5) whose 6 rows are treated as 5-dimensional vectors. Return the 6x6 cosine-similarity matrix S where S[i, j] equals the dot product of rows i and j of X divided by the product of their Euclidean lengths. The diagonal of S is 1 and S is symmetric. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Normalize each row to unit length first, then a single matrix product with the transpose gives all pairwise cosines.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Xn = X / np.linalg.norm(X, axis=1, keepdims=True)
    return Xn @ Xn.T
```

## Why this works

Cosine similarity is the dot product of unit-normalized vectors. Dividing each row by its Euclidean norm yields unit rows, so the matrix product Xn @ Xn.T computes every pairwise cosine at once, giving a symmetric matrix with ones on the diagonal.
