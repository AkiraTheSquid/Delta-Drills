## Problem

Write a function solve(x, y) that takes two 2-D float arrays x of shape (n, d) and y of shape (m, d), with no zero rows, and returns the (n, m) CROSS cosine-similarity matrix: entry [i, j] is the dot product of row i of x with row j of y divided by the product of their Euclidean norms. (An earlier drill compares a matrix's rows with themselves; here the two sides differ.)


<details><summary>💡 Hint (click to reveal)</summary>

Normalize each side's rows to unit length first — then the whole similarity table is a single matrix product.

</details>

## Reference solution


```python
import numpy as np

def solve(x, y):
    xn = x / np.linalg.norm(x, axis=1, keepdims=True)
    yn = y / np.linalg.norm(y, axis=1, keepdims=True)
    return xn @ yn.T


print(solve(np.array([[1.0, 0.0]]), np.array([[0.0, 2.0], [3.0, 0.0]])))
```

## Why this works

Once both row sets are unit-length, dot products ARE cosines, so xn @ yn.T fills the entire (n, m) table in one multiplication. Normalizing first (instead of dividing the Gram matrix by an outer product of norms) is the cleaner factorization and generalizes to two different matrices.
