## Problem

Write a function solve(x, y) that takes two 2-D float tensors x of shape (n, d) and y of shape (m, d), with no zero rows, and returns the (n, m) CROSS cosine-similarity matrix: entry [i, j] is the dot product of row i of x with row j of y divided by the product of their Euclidean norms. (An earlier drill compares a matrix's rows with themselves; here the two sides differ.)

## Reference solution


```python
import torch as t

def solve(x, y):
    xn = x / t.linalg.norm(x, dim=1, keepdim=True)
    yn = y / t.linalg.norm(y, dim=1, keepdim=True)
    return xn @ yn.T


print(solve(t.tensor([[1.0, 0.0]]), t.tensor([[0.0, 2.0], [3.0, 0.0]])))
```
