## Problem

Write a function solve(stop, step) that takes a positive float stop (an exact multiple of step) and a positive float step, and returns a 1-D PyTorch tensor running 0, step, 2*step, ... up to AND INCLUDING stop. Note t.arange's endpoint is exclusive — the drill is getting the inclusive end right, and doing it without accumulating floating-point error.


<details><summary>💡 Hint (click to reveal)</summary>

Count how many points there are (stop/step plus one) and scale an integer range — safer than nudging a float endpoint.

</details>

## Reference solution


```python
import torch as t

def solve(stop, step):
    n = int(round(stop / step)) + 1
    return t.arange(n) * step


print(solve(10.0, 0.5))
```
