## Problem

Build a 1D array Z that starts at 0 and steps by 0.5 up to but not exceeding 10, so the values are 0.0, 0.5, 1.0, ..., 10.0. The endpoint 10.0 must be included. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

arange's stop is exclusive; push the stop slightly beyond your desired last value so the endpoint is included.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(0, 10.5, 0.5)
    return Z

print(solve())
```

## Why this works

np.arange is half-open (excludes the stop value), so to include 10.0 with a step of 0.5 the stop must be set just past it (10.5). This yields 0.0, 0.5, ..., 10.0.
