## Problem

Write a function solve(stop, step) that takes a positive float stop (an exact multiple of step) and a positive float step, and returns a 1-D array running 0, step, 2*step, ... up to AND INCLUDING stop. Note np.arange's endpoint is exclusive — the drill is getting the inclusive end right.


<details><summary>💡 Hint (click to reveal)</summary>

Count how many points there are (stop/step plus one) and scale an integer range — safer than nudging a float endpoint.

</details>

## Reference solution


```python
import numpy as np

def solve(stop, step):
    n = int(round(stop / step)) + 1
    return np.arange(n) * step


print(solve(10.0, 0.5))
```

## Why this works

Multiplying an integer arange by step avoids the float-accumulation and exclusive-endpoint pitfalls of np.arange(0, stop + step/2, step), guaranteeing exactly stop/step + 1 points ending precisely at stop. np.linspace(0, stop, n) with the same count is equally robust.
