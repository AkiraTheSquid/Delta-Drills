## Problem

Write a function solve(v) that takes a 1-D numpy array of floats with nonzero length (and at least one nonzero entry) and returns the vector rescaled to unit length: the result points in the same direction but its Euclidean length equals 1. Do not modify the input array.


<details><summary>💡 Hint (click to reveal)</summary>

Divide the vector by its own Euclidean length — and return a new array rather than scaling in place.

</details>

## Reference solution


```python
import numpy as np

def solve(v):
    return v / np.linalg.norm(v)


example = np.array([3.0, 4.0])
print(solve(example))
```

## Why this works

Dividing v by np.linalg.norm(v) rescales every component by the same factor, preserving direction while making the length exactly 1. The division creates a new array, so the caller's vector is untouched — one test checks that.
