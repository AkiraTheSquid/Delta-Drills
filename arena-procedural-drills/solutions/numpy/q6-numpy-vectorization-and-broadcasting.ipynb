## Problem

Given a random vector of length 5, adjust its values so that the total length of the vector equals 1, maintaining the direction of the vector. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Divide the vector by its own Euclidean length so the magnitude becomes 1 while the direction is unchanged.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    v = np.random.random(5)
    v /= np.linalg.norm(v)
    return v

print(solve())
```

## Why this works

Dividing a vector by its L2 norm keeps its direction but rescales its length to exactly 1, the definition of a unit vector. np.linalg.norm on a 1D array returns the L2 (Euclidean) length used as the divisor.
