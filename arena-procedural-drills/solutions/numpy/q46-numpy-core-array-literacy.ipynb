## Problem

Build a 2D integer matrix of shape (3, 3) whose entries, read row by row from top-left to bottom-right, are the consecutive integers 0 through 8 in row-major order, so the smallest value sits in the top-left cell and the largest value sits in the bottom-right cell. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Generate the consecutive integers, then reshape them into the square grid in row-major order.

</details>

## Reference solution


```python
import numpy as np

def solve():
    return np.arange(9).reshape(3, 3)

print(solve())
```

## Why this works

np.arange(9) produces 0..8 and reshape lays them out row-major into a 3x3 grid, so the smallest sits top-left and the largest bottom-right.
