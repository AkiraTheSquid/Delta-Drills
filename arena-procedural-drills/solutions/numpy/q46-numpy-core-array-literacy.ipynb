## Problem

Write a function solve(n) that takes a positive integer n and returns an n x n numpy array of integers containing every value from 0 up to n*n - 1 exactly once. The values must increase left to right within each row and continue onto the next row, so 0 sits in the top-left corner and n*n - 1 sits in the bottom-right corner. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

First build the full run of consecutive integers as a flat array, then change its shape into the square grid you need.

</details>

## Reference solution


```python
import numpy as np

def solve(n):
    return np.arange(n * n).reshape(n, n)


example = 3
print(solve(example))
```

## Why this works

np.arange(n * n) produces a 1-D array of the integers 0 through n*n-1 already in increasing order. Calling .reshape(n, n) reinterprets that flat run as an n x n grid filled row by row (row-major order), which is exactly the reading order the task demands. This works for any positive n, including the 1 x 1 case, without any Python loops.
