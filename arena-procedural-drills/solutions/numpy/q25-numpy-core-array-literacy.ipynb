## Problem

Given a 3x3 integer matrix Z, swap the top row with the bottom row so the row that was on top now sits at the bottom and the row that was at the bottom now sits on top. The middle row stays in place and the contents of each swapped row are kept intact. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

You can index multiple rows with a list of row numbers and assign a reordered view back into the same rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    Z[[0, 2]] = Z[[2, 0]]
    return Z

print(solve())
```

## Why this works

Fancy indexing with a row-index list lets you read and assign whole rows at once. Assigning Z[[2,0]] (bottom then top) into Z[[0,2]] swaps the first and last rows while the middle row is untouched.
