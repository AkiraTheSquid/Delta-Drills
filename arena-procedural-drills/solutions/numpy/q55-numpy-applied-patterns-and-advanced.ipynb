## Problem

Given a date range from July 1, 2016, to July 31, 2016, return all the dates within that month as an array. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

np.arange works on datetime64 day units; bound it from the month start to the next month start.

</details>


```python
%pip install -q numpy
```

## Reference solution


```python
import numpy as np


def solve():
    return np.arange('2016-07', '2016-08', dtype='datetime64[D]')


print(solve())
```

## Why this works

np.arange with a datetime64[D] dtype steps one day at a time. Bounding it by the month start (inclusive) and the next month start (exclusive) yields every calendar day of July 2016.
