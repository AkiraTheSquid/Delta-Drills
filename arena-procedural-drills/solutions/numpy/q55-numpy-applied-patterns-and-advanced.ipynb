## Problem

Write a function solve(year, month) that takes a year and a month (1-12) as integers and returns a 1-D numpy datetime64 array containing every calendar date in that month, in order — from the 1st through the last day. The function must get the month length right, including February in leap years.


<details><summary>💡 Hint (click to reveal)</summary>

A month-precision datetime64 plus 1 is the NEXT month — and np.arange can walk between two months in steps of days.

</details>

## Reference solution


```python
import numpy as np

def solve(year, month):
    start = np.datetime64(f"{year}-{month:02d}")
    return np.arange(start, start + 1, dtype="datetime64[D]")


print(solve(2016, 7))
```

## Why this works

np.datetime64(f'{year}-{month:02d}') is a month-precision timestamp; adding 1 advances one month, handling year rollover and leap years automatically. np.arange between the two with dtype datetime64[D] enumerates every day of the month, excluding the exclusive upper bound.
