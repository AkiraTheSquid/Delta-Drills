## Problem

Write a function solve(z) that takes a 1-D numpy float array whose entries are not all equal, and returns a new array linearly rescaled so its smallest entry is exactly 0.0 and its largest is exactly 1.0, with every other entry keeping its relative position in between. Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Shift so the minimum sits at zero, then divide by the full spread between the extremes.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return (z - z.min()) / (z.max() - z.min())


example = np.array([2.0, 4.0, 6.0])
print(solve(example))
```

## Why this works

Subtracting z.min() makes the smallest entry 0; dividing by the range z.max() - z.min() scales the largest to 1, and the transformation is affine so relative spacing is preserved. (numpy 2.x removed the old Z.ptp() method for the range — compute the spread explicitly or use np.ptp.)
