## Problem

Write a function solve(z) that takes a 1-D PyTorch float tensor whose entries are not all equal, and returns a new tensor linearly rescaled so its smallest entry is exactly 0.0 and its largest is exactly 1.0, with every other entry keeping its relative position in between. Do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Shift so the minimum sits at zero, then divide by the full spread between the extremes.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return (z - z.min()) / (z.max() - z.min())


example = t.tensor([2.0, 4.0, 6.0])
print(solve(example))
```
