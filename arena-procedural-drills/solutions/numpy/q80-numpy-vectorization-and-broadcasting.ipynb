## Problem

Given a 1D vector Z of length 10 with arbitrary real entries, return a new length-10 vector whose smallest entry is exactly 0.0, whose largest entry is exactly 1.0, and where every other entry is linearly placed between them according to its original value's position in Z's value range (min-max normalisation to the interval [0, 1]). (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Shift by the minimum, then scale by the range (max minus min).

</details>

## Reference solution


```python
import numpy as np
np.random.seed(0)

Z = np.random.rand(10)

def solve():
    return (Z - Z.min()) / (Z.max() - Z.min())

print(solve())
```

## Why this works

Min-max normalisation subtracts the minimum so the smallest value maps to 0, then divides by the value range (max minus min) so the largest maps to 1, placing every other value linearly in between. Seeding the RNG keeps the input reproducible.
