## Problem

Given a 2D array Z, transform each column so that it has a mean of zero and a standard deviation of one, handling any columns with zero standard deviation appropriately. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Compute column means and stds along axis 0 (keepdims for broadcasting), subtract and divide, but replace any zero std with 1 before dividing to handle constant columns safely.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randn(5, 4)
    mu = Z.mean(axis=0, keepdims=True)
    sd = Z.std(axis=0, ddof=0, keepdims=True)
    # Avoid divide-by-zero: a constant column gets divided by 1, leaving it centered at 0.
    Zz = (Z - mu) / np.where(sd == 0, 1, sd)
    return Zz.mean(0), Zz.std(0)

print(solve())
```

## Why this works

Per-column standardization subtracts each column's mean and divides by its standard deviation, both computed along axis 0 with keepdims so they broadcast back over the rows. Guarding the divisor with np.where(sd==0, 1, sd) prevents division by zero for constant columns (which stay zero after centering). The result has near-zero column means and unit column standard deviations.
