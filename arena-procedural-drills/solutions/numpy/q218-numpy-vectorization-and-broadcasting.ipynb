## Problem

Write a function solve(z) that takes a 2-D float array and standardizes each COLUMN (subtract its mean, divide by its population std) — but columns whose std is ZERO (constant columns) must come out as all zeros (just centered), with no division warnings or NaN. (An earlier drill assumes no constant columns; handling them is this one's point.)


<details><summary>💡 Hint (click to reveal)</summary>

Swap the zero stds for a harmless 1 BEFORE dividing — a centered constant column is already all zeros, so dividing by 1 leaves it correct.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    mu = z.mean(axis=0, keepdims=True)
    sd = z.std(axis=0, keepdims=True)
    return (z - mu) / np.where(sd == 0, 1, sd)


print(solve(np.array([[1.0, 5.0], [3.0, 5.0]])))
```

## Why this works

A constant column centers to exactly zero, so substituting its zero std with 1 via np.where makes the division a no-op instead of 0/0. Real columns are untouched by the substitution. This guard-the-denominator-first pattern avoids both the RuntimeWarning and the NaNs of dividing then patching.
