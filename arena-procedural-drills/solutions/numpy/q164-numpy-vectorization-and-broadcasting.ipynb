## Problem

Write a function solve(c) that takes a 1-D numpy array c of length n and returns the n x n symmetric Toeplitz matrix T with T[i, j] = c[|i - j|]: c[0] runs down the main diagonal, c[1] along both first off-diagonals, and so on. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Build the |i - j| index grid by broadcasting a column of indices against a row — then use that whole grid to index c at once.

</details>

## Reference solution


```python
import numpy as np

def solve(c):
    n = c.size
    i = np.arange(n)[:, None]
    j = np.arange(n)[None, :]
    return c[np.abs(i - j)]


print(solve(np.array([10, 20, 30])))
```

## Why this works

Broadcasting the row/column index vectors gives the (n, n) grid of |i - j| offsets; fancy-indexing c with an INTEGER ARRAY of any shape returns an array of that same shape, so c[np.abs(i - j)] materializes the entire Toeplitz matrix in one gather.
