## Problem

Given a 1D integer vector c of length n=6, build the n-by-n symmetric Toeplitz matrix T whose entry T[i, j] equals c[|i - j|]. The main diagonal therefore holds c[0], the first off-diagonals hold c[1], and so on out to c[n-1] in the corners. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Build the |i-j| index matrix via broadcasting, then use it to gather from c.

</details>

## Reference solution


```python
import numpy as np

def solve():
    c = np.arange(6)
    i = np.arange(c.size)[:, None]
    j = np.arange(c.size)[None, :]
    T = c[np.abs(i - j)]
    return T

print(solve())
```

## Why this works

Broadcasting a column index vector against a row index vector forms the matrix of i-j, and taking its absolute value gives |i-j| at every cell. Indexing c with that integer matrix produces the symmetric Toeplitz matrix T[i,j] = c[|i-j|].
