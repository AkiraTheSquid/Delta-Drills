## Problem

Write a function solve(z) that takes a 2-D numpy array and returns a 2-D array containing each distinct ROW of z exactly once, ordered lexicographically (compare first column, then second, and so on). Duplicate rows are collapsed to a single occurrence.


<details><summary>💡 Hint (click to reveal)</summary>

The deduplication function takes an axis argument — along axis 0 it treats whole rows as the things to compare.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.unique(z, axis=0)


example = np.array([[1, 2], [1, 2], [0, 9]])
print(solve(example))
```

## Why this works

np.unique(z, axis=0) deduplicates along the first axis, so entire rows are the units of comparison, and it returns them lexicographically sorted — first by column 0, then column 1, as the [[2,1],[2,0]] case confirms.
