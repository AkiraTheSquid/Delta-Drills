## Problem

Write a function solve(z) that takes a 2-D PyTorch tensor and returns a 2-D tensor containing each distinct ROW of z exactly once, ordered lexicographically (compare first column, then second, and so on). Duplicate rows are collapsed to a single occurrence.


<details><summary>💡 Hint (click to reveal)</summary>

The deduplication function takes an axis argument — along axis 0 it treats whole rows as the things to compare.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return t.unique(z, dim=0)


example = t.tensor([[1, 2], [1, 2], [0, 9]])
print(solve(example))
```

## Why this works

t.unique(z, axis=0) deduplicates along the first axis, so entire rows are the units of comparison, and it returns them lexicographically sorted — first by column 0, then column 1, as the [[2,1],[2,0]] case confirms.
