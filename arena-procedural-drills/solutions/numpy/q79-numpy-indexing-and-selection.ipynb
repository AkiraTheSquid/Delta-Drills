## Problem

You are given Z, a 2D integer array of shape (3, 3) whose first and second rows happen to be identical. Return a 2D array containing each unique row of Z exactly once, with rows ordered lexicographically and the original column count preserved. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

np.unique can operate over whole rows when you tell it which axis to deduplicate along.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([[1, 2, 3], [1, 2, 3], [2, 3, 4]])
    return np.unique(Z, axis=0)

print(solve())
```

## Why this works

np.unique with axis=0 treats each row as a single item, removing duplicate rows and returning the survivors sorted lexicographically while preserving the column count. The two identical first rows collapse to one.
