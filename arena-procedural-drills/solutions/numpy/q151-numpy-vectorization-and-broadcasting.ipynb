## Problem

Write a function solve(a, b) that takes a 3-D float tensor a of shape (h, w, c) and a 2-D tensor b of shape (h, w), and returns the elementwise product in which every channel slice a[:, :, k] is multiplied by b. The shapes differ, so direct multiplication needs an explicit new axis on b.


<details><summary>💡 Hint (click to reveal)</summary>

Broadcasting aligns TRAILING axes — (h, w) against (h, w, c) fails, but (h, w, 1) stretches over the channels.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return a * b[:, :, None]


print(solve(t.ones((2, 2, 3)), t.tensor([[1.0, 2.0], [3.0, 4.0]])))
```

## Why this works

b[:, :, None] appends a length-1 channel axis, making (h, w, 1) broadcast-compatible with (h, w, c): the single channel value at each pixel stretches over all c channels. Without the new axis, PyTorch would try to align w with c and fail — the drill's core broadcasting rule.
