## Problem

Consider an array of dimension (5,5,3), how to multiply it by an array with dimensions (5,5)? (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Add a trailing singleton axis to the (5,5) array so it broadcasts over the last dimension.

</details>

## Reference solution


```python
import numpy as np

def solve():
    A = np.ones((5,5,3))
    B = 2*np.ones((5,5))
    return A * B[:,:,None]

print(solve())
```

## Why this works

Broadcasting aligns shapes from the trailing axis. To multiply a (5,5,3) array by a (5,5) array you add a trailing length-1 axis to B with B[:,:,None], making it (5,5,1); it then broadcasts across the size-3 channel dimension.
