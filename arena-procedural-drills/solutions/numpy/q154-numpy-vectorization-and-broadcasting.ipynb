## Problem

Write a function solve(z, limit) that takes a 2-D float array (no constant columns) and a positive float limit. First standardize each COLUMN (subtract its mean, divide by its population std), then clip the standardized values into [-limit, +limit], and return the result.


<details><summary>💡 Hint (click to reveal)</summary>

Two independent steps: axis-0 statistics broadcast down the columns, then a symmetric clip caps the z-scores.

</details>

## Reference solution


```python
import numpy as np

def solve(z, limit):
    standardized = (z - z.mean(axis=0)) / z.std(axis=0)
    return np.clip(standardized, -limit, limit)


example = np.array([[1.0, 10.0], [2.0, 20.0], [30.0, 30.0]])
print(solve(example, 1.0))
```

## Why this works

Column standardization uses axis=0 reductions that broadcast back over the rows; np.clip then maps any |z-score| beyond the limit onto ±limit. With a generous limit nothing clips (last case), while extreme outliers pin exactly to the boundary.
