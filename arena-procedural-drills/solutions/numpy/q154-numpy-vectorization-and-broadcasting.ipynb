## Problem

Standardize each column of a matrix by subtracting the mean and dividing by the standard deviation, then limit the resulting values to be within the range of -3 to 3. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Subtract column means, divide by column stds (guard against zero), then clip to the range.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randn(100, 5)
    mu = Z.mean(axis=0, keepdims=True)
    sd = Z.std(axis=0, ddof=0, keepdims=True)
    Zz = (Z - mu) / np.where(sd==0, 1, sd)
    Zc = np.clip(Zz, -3, 3)
    return Zc.min(), Zc.max()

print(solve())
```

## Why this works

Standardize per column by subtracting the column mean and dividing by the column standard deviation, both computed with keepdims so they broadcast. Guarding the denominator with np.where avoids dividing by a zero std, and np.clip then bounds the z-scores into [-3, 3].
