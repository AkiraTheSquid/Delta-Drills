## Problem

Write a function solve(a, b) that takes two PyTorch float tensors of identical shape and returns their elementwise average: stack the two tensors along a new axis and average across it, or combine them arithmetically — entry [i, j] of the result is (a[i, j] + b[i, j]) / 2.


<details><summary>💡 Hint (click to reveal)</summary>

Stacking along a NEW leading axis turns the pair into one tensor whose first axis you can average away.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.stack([a, b], dim=0).mean(dim=0)


example_a = t.tensor([[1.0, 2.0]])
example_b = t.tensor([[3.0, 6.0]])
print(solve(example_a, example_b))
```

## Why this works

t.stack([a, b], axis=0) creates a (2, ...) tensor holding both inputs; mean(axis=0) collapses that new axis, averaging corresponding entries. The direct (a + b) / 2 computes the same values and the tests accept it — the stack form generalizes to averaging any number of tensors.
