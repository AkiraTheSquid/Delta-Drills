## Problem

Given two arrays A and B of shape (3, 4), combine them into a new array that adds an extra dimension, then calculate the average along that new dimension. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Stack the two arrays along a new leading axis, then take the mean over that axis.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    A = np.random.rand(3, 4)
    B = np.random.rand(3, 4)
    return np.stack([A, B], axis=0).mean(axis=0)

print(solve())
```

## Why this works

np.stack adds a brand-new axis (here axis 0) producing a (2, 3, 4) array, and averaging along that axis collapses it back to (3, 4) by taking the elementwise mean of A and B. Seeding the RNG makes A and B reproducible for the grader.
