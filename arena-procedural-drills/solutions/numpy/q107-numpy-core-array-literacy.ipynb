## Problem

Write a function solve(a, b) that takes an invertible n x n float PyTorch matrix a and a length-n vector b, and returns the vector x satisfying the linear system a @ x = b. Use a linear solver rather than computing an explicit inverse.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch's linear-algebra module has a direct system solver — faster and more accurate than inverting and multiplying.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.linalg.solve(a, b)


example_a = t.tensor([[2.0, 0.0], [0.0, 4.0]])
example_b = t.tensor([6.0, 8.0])
print(solve(example_a, example_b))
```

## Why this works

t.linalg.solve(a, b) factorizes a and solves a @ x = b directly, which is numerically better than t.linalg.inv(a) @ b (the tests round to 8 decimals, so the inverse route also passes on these well-conditioned systems — but solve is the idiom to learn). The random case adds 3*I to keep the matrix comfortably invertible.
