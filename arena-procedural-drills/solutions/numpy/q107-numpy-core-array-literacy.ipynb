## Problem

Given a random 3x3 matrix A and a vector b of length 3, find the vector x that satisfies the equation Ax = b. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use the dedicated linear-system solver rather than inverting the matrix.

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    A = np.random.rand(3, 3)
    b = np.random.rand(3)
    return np.linalg.solve(A, b)

print(solve())
```

## Why this works

np.linalg.solve directly solves the linear system Ax = b for x using a stable LU factorization, which is preferable to forming the explicit inverse. For a nonsingular 3x3 A this returns the unique solution vector.
