## Problem

Given a 1D vector created with `np.arange(5)`, replicate it to form a 2D matrix with 4 rows. Then, calculate the average of each column in this matrix. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Tile the vector into rows, then reduce with mean along the row axis (axis=0).

</details>

## Reference solution


```python
import numpy as np

def solve():
    v = np.arange(5)
    M = np.tile(v, (4, 1))
    return M.mean(axis=0)

print(solve())
```

## Why this works

np.tile with reps (4, 1) stacks the length-5 vector into a 4x5 matrix of identical rows, and taking the mean along axis 0 averages down each column. Because every row is identical, the column means just equal the original vector as floats.
