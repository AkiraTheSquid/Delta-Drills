## Problem

Write a function solve(v, reps) that takes a 1-D PyTorch tensor v and a positive integer reps. Build the 2-D matrix whose reps rows are each a copy of v, then return a tuple of (that matrix, the per-column means of the matrix). Since every row is identical, the column means should reproduce v's values as floats.


<details><summary>💡 Hint (click to reveal)</summary>

Tile the vector with a (reps, 1) repetition pattern, then average down the rows (axis 0).

</details>

## Reference solution


```python
import torch as t
def solve(v, reps):
    m = t.tile(v, (reps, 1))
    return m, m.to(t.float32).mean(dim=0)


example = t.arange(5)
print(solve(example, 4))
```

## Why this works

t.tile(v, (reps, 1)) repeats v as reps stacked rows. Averaging along axis 0 collapses the rows into one mean per column; identical rows make those means equal to v itself, converted to floats by the division inside mean.
