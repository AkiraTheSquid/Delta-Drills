## Problem

Write a function solve(n) that takes a positive integer n and returns a 1-D numpy array of floats containing the n evenly spaced breakpoints strictly inside the open interval (0, 1). In other words, splitting [0, 1] into n+1 equal sub-intervals produces exactly n interior breakpoints; return those, excluding both 0.0 and 1.0. For example, n = 3 should give the three quarter-points between 0 and 1.


<details><summary>💡 Hint (click to reveal)</summary>

If you generate the evenly spaced points including both endpoints, you need two extra points — then drop the first and last.

</details>

## Reference solution


```python
import numpy as np

def solve(n):
    return np.linspace(0.0, 1.0, n + 2)[1:-1]


example = 3
print(solve(example))
```

## Why this works

np.linspace(0.0, 1.0, n + 2) produces n + 2 evenly spaced points from 0 to 1 inclusive, which are exactly the boundaries of the n + 1 equal sub-intervals. Slicing off the first and last entries with [1:-1] removes the endpoints 0.0 and 1.0, leaving the n interior breakpoints. This works for any positive n, including n = 1, where the only interior breakpoint is 0.5.
