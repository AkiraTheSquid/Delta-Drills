## Problem

Write a function solve(n) that takes a positive integer n and returns a 1-D PyTorch tensor of floats containing the n evenly spaced breakpoints strictly inside the open interval (0, 1). In other words, splitting [0, 1] into n+1 equal sub-intervals produces exactly n interior breakpoints; return those, excluding both 0.0 and 1.0. For example, n = 3 should give the three quarter-points between 0 and 1.


<details><summary>💡 Hint (click to reveal)</summary>

If you generate the evenly spaced points including both endpoints, you need two extra points — then drop the first and last.

</details>

## Reference solution


```python
import torch as t

def solve(n):
    return t.linspace(0.0, 1.0, n + 2)[1:-1]


example = 3
print(solve(example))
```
