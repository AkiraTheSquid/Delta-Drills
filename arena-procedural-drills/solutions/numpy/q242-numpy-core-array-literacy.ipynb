## Problem

You are given Z, a 1D float array of length 10 holding 10 evenly spaced values strictly between 0 and 1 (both endpoints excluded), so the spacing matches what you get from splitting [0, 1] into 11 equal sub-intervals and taking the 10 interior breakpoints. Return Z. (★★☆)
Print the result.

Generate evenly spaced points with the right step, then slice off the zero endpoint.

</details>

## Reference solution


```python
import numpy as np

def solve():
    return np.linspace(0, 1, 11, endpoint=False)[1:]

print(solve())
```

## Why this works

np.linspace(0, 1, 11, endpoint=False) splits [0, 1) into 11 equal steps starting at 0; dropping the first element with [1:] leaves the 10 interior breakpoints, both endpoints excluded. The spacing is 1/11.
