## Problem

Given a 16x16 array, compute the sum of non-overlapping 4x4 blocks and return the resulting array. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Reshape the matrix into (blocks, within, blocks, within) and sum over the two within-block axes.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.ones((16, 16))
    return Z.reshape(4, 4, 4, 4).sum((1, 3))

print(solve())
```

## Why this works

Reshaping a 16x16 array to (4, 4, 4, 4) splits each axis into a block index and an in-block index, so axes 1 and 3 enumerate the positions inside each 4x4 block. Summing over those two in-block axes collapses every non-overlapping block to its total, giving the 4x4 block-sum array.
