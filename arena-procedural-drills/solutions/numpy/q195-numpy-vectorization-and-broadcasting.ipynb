## Problem

Write a function solve(z, k) that takes a 2-D integer array and a block size k, and returns the sums of NON-OVERLAPPING k x k blocks tiling z from the top-left — where z's dimensions need NOT be multiples of k: trailing blocks are simply smaller (whatever remains). Output shape is (ceil(rows/k), ceil(cols/k)). (The evenly-divisible reshape trick fails here — reduceat handles the ragged edges.)


<details><summary>💡 Hint (click to reveal)</summary>

reduceat sums between the START positions you give it — starts every k rows, then every k columns, and the ragged tail takes care of itself.

</details>

## Reference solution


```python
import numpy as np

def solve(z, k):
    r = np.add.reduceat(z, np.arange(0, z.shape[0], k), axis=0)
    return np.add.reduceat(r, np.arange(0, z.shape[1], k), axis=1)


print(solve(np.ones((5, 5), dtype=int), 2))
```

## Why this works

np.add.reduceat(z, starts, axis) sums each segment from one start to the next (the final segment runs to the end, however short), so tiling starts at np.arange(0, n, k) handle non-divisible shapes naturally. Applying it along rows then columns composes into 2-D tile sums — where the reshape trick would simply raise on a 5x5 with k=2.
