## Problem

Write a function solve(z, k) that takes a 2-D integer tensor and a block size k, and returns the sums of NON-OVERLAPPING k x k blocks tiling z from the top-left — where z's dimensions need NOT be multiples of k: trailing blocks are simply smaller (whatever remains). Output shape is (ceil(rows/k), ceil(cols/k)). (The evenly-divisible reshape trick fails here — reduceat handles the ragged edges.)

## Reference solution


```python
import torch as t
def solve(z, k):
    rows = t.stack([chunk.sum(dim=0) for chunk in z.split(k, dim=0)])
    return t.stack([chunk.sum(dim=1) for chunk in rows.split(k, dim=1)], dim=1)


print(solve(t.ones((5, 5), dtype=t.int64), 2))
```
