## Problem

Write a function solve(z, w, step) that takes a 1-D numpy array, a window length w, and a positive stride step. Return the 2-D array whose rows are the length-w windows of z starting at positions 0, step, 2*step, ... (only windows that fit completely). Row count is (len(z) - w) // step + 1. (An earlier drill slides by 1; the stride parameter is what's new.)


<details><summary>💡 Hint (click to reveal)</summary>

Take ALL stride-1 windows first, then keep every step-th row — slicing a view is still a view.

</details>

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve(z, w, step):
    return sliding_window_view(z, w)[::step]


print(solve(np.arange(8), 3, 2))
```

## Why this works

sliding_window_view(z, w) lists every stride-1 window; the row slice [::step] keeps those starting at multiples of step. Both operations are views, so no data is copied. The row-count formula (len-w)//step + 1 falls out of which starts still fit.
