## Problem

Consider a one-dimensional array Z, build a two-dimensional array whose first row is (Z[0],Z[1],Z[2]) and each subsequent row is shifted by 1 (last row should be (Z[-3],Z[-2],Z[-1]).


<details><summary>💡 Hint (click to reveal)</summary>

A length-3 sliding window over the 1D array yields each shifted triple as a row.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(10)
    return np.lib.stride_tricks.sliding_window_view(Z, window_shape=3)

print(solve())
```

## Why this works

sliding_window_view produces every contiguous length-3 window of Z as rows of a (8, 3) array, which is exactly the requested matrix whose rows shift by one each step from (Z[0:3]) down to (Z[-3:]).
