## Problem

Write a function solve(z) that takes a PyTorch tensor of floats (any shape) and returns a new tensor of the same shape in which the mean of ALL entries has been subtracted from every entry, so the result's overall mean is zero. Do not modify the input tensor.


<details><summary>💡 Hint (click to reveal)</summary>

A no-axis mean collapses the whole tensor to one number, and subtracting a scalar broadcasts to every entry.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    return z - z.mean()


example = t.tensor([[1.0, 2.0], [3.0, 6.0]])
print(solve(example))
```

## Why this works

z.mean() with no axis argument averages every entry regardless of shape, and z - scalar broadcasts the subtraction across the whole tensor, returning a new centered tensor and leaving z untouched.
