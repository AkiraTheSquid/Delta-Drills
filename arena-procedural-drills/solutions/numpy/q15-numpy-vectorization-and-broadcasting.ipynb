## Problem

Given a 3x3 matrix Z of random values, return the matrix after adjusting each element by removing the average value of all elements. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Subtract the single scalar mean of all entries from the matrix so the overall average becomes zero.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random((3, 3))
    Z = Z - Z.mean()
    return Z

print(solve())
```

## Why this works

Z.mean() with no axis is the scalar average of all nine entries; subtracting it from every element centers the matrix so its overall mean becomes zero. Broadcasting applies the single scalar to the whole array.
