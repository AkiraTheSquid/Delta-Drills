Write a function solve(x) that takes a 1-D PyTorch float tensor and returns its elementwise CUBE — each entry raised to the third power. Any torch route works (pow, repeated multiply, or einsum).

## Reference solution


```python
import torch as t

def solve(x):
    return t.pow(x, 3)


print(solve(t.tensor([1.0, 2.0, -3.0])))
```
