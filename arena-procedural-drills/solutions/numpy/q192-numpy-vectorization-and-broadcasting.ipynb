## Reference solution


```python
import numpy as np

def solve(x):
    return np.power(x, 3)


print(solve(np.array([1.0, 2.0, -3.0])))
```

## Why this works

np.power(x, 3), x ** 3, and x * x * x all compute the same elementwise cube (the repeated multiply is often fastest since it avoids the general pow routine; einsum('i,i,i->i', x, x, x) is the exotic spelling). Negative bases keep their sign since the exponent is odd.
