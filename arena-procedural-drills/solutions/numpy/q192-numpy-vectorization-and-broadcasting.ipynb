## Problem

Compute the element-wise cube of a vector x using `np.power(x, 3)`. Return the resulting array. (★★★)

## Reference solution


```python
import numpy as np

def solve():
    x = np.arange(10, dtype=float)
    return np.power(x, 3)

print(solve())
```

## Why this works

The element-wise cube is just np.power(x, 3), which raises each entry to the third power. The original gallery snippet only benchmarked equivalent ways to cube an array (power, repeated multiply, einsum); the actual returned quantity is the cubed array itself.
