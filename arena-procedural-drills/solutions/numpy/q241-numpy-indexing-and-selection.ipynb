## Problem

Given two 1D integer arrays Z1 and Z2 of length 10, each containing random values in [0, 10), produce the sorted 1D array of values that appear in both Z1 and Z2, with each common value listed only once. (★☆☆)
Print the result.

There is a single set-style NumPy function that returns sorted unique common values of two arrays.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z1 = np.random.randint(0, 10, 10)
    Z2 = np.random.randint(0, 10, 10)
    return np.intersect1d(Z1, Z2)

print(solve())
```

## Why this works

np.intersect1d returns the sorted, de-duplicated set of values present in both input arrays, which is exactly the common-values requirement. It handles the uniqueness and ordering for you.
