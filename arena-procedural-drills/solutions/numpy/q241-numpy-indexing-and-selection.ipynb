## Problem

Write a function solve(a, b) that takes two 1-D numpy integer arrays and returns a 1-D array of the values that appear in both a and b. Each shared value must appear exactly once in the result, and the result must be sorted in ascending order. If the two arrays have no values in common, return an empty 1-D array.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has set-style operations that work directly on arrays — look for the one that keeps only the values common to both inputs.

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.intersect1d(a, b)


a = np.array([4, 1, 3, 2, 3])
b = np.array([3, 4, 4, 8, 0])
print(solve(a, b))
```

## Why this works

np.intersect1d(a, b) treats both arrays as sets: it deduplicates each input, keeps only the values that occur in both, and returns them sorted in ascending order — exactly the required contract, including an empty array when nothing overlaps. An equivalent manual route is np.unique(a[np.isin(a, b)]), since np.unique both deduplicates and sorts.
