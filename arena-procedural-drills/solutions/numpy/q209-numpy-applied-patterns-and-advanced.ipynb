## Problem

Implement Conway's Game of Life on a 50x50 binary board with zero-padded boundaries. Given the initial board Z and an integer `iterations`, return the board after that many steps. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Count neighbours with eight offset slices of the board summed together, then derive birth and survival masks from that count and the current cell state. Write into a fresh zero board each iteration so reads aren't corrupted by writes.

</details>

## Reference solution


```python
import numpy as np

np.random.seed(0)
Z = np.random.randint(0, 2, (50, 50))
iterations = 10

def solve():
    def iterate(z):
        N = (z[0:-2, 0:-2] + z[0:-2, 1:-1] + z[0:-2, 2:] +
             z[1:-1, 0:-2]                 + z[1:-1, 2:] +
             z[2:  , 0:-2] + z[2:  , 1:-1] + z[2:  , 2:])
        birth = (N == 3) & (z[1:-1, 1:-1] == 0)
        survive = ((N == 2) | (N == 3)) & (z[1:-1, 1:-1] == 1)
        out = np.zeros_like(z)
        out[1:-1, 1:-1][birth | survive] = 1
        return out
    z = Z.copy()
    for _ in range(iterations):
        z = iterate(z)
    return z

print(solve())
```

## Why this works

Each Game of Life step counts the eight neighbours of every interior cell via eight shifted slices of the padded board, then applies the rules: a dead cell with exactly 3 live neighbours is born, a live cell with 2 or 3 survives. Boundary cells stay 0 because the slicing only writes the interior [1:-1, 1:-1]. Building a fresh zeroed output each step (rather than mutating in place) keeps the neighbour counts consistent within a step.
