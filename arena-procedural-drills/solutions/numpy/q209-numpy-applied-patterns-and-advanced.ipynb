## Problem

Write a function solve(z, iterations) that takes a 2-D 0/1 integer array (a Game of Life board with DEAD cells beyond every border) and returns the board after the given number of Conway steps: a live cell survives with 2 or 3 live neighbors, a dead cell becomes alive with exactly 3, everything else dies. Return a new array; do not modify the input.


<details><summary>💡 Hint (click to reveal)</summary>

Pad the board with a dead border, sum the eight shifted slices to count neighbors everywhere at once, then apply both birth/survival rules as one boolean expression.

</details>

## Reference solution


```python
import numpy as np

def solve(z, iterations):
    board = z.copy()
    for _ in range(iterations):
        padded = np.pad(board, 1)
        n = (padded[:-2, :-2] + padded[:-2, 1:-1] + padded[:-2, 2:]
             + padded[1:-1, :-2] + padded[1:-1, 2:]
             + padded[2:, :-2] + padded[2:, 1:-1] + padded[2:, 2:])
        board = ((n == 3) | ((board == 1) & (n == 2))).astype(int)
    return board


blinker = np.zeros((5, 5), dtype=int)
blinker[2, 1:4] = 1
print(solve(blinker, 1))
```

## Why this works

Zero-padding makes every border cell's out-of-bounds neighbors dead; the eight shifted slices of the padded board align each cell with its neighbors so one sum counts them all. The next generation is (n == 3) | (alive & (n == 2)) — birth on exactly 3, survival on 2 or 3. The blinker (period 2), block (still life), and lone cell (dies) test all three behaviors.
