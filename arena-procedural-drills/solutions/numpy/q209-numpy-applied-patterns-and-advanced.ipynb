## Problem

Write a function solve(z, iterations) that takes a 2-D 0/1 integer tensor (a Game of Life board with DEAD cells beyond every border) and returns the board after the given number of Conway steps: a live cell survives with 2 or 3 live neighbors, a dead cell becomes alive with exactly 3, everything else dies. Return a new tensor; do not modify the input.

## Reference solution


```python
import torch as t
def solve(z, iterations):
    board = z.clone()
    for _ in range(iterations):
        padded = t.nn.functional.pad(board, (1, 1, 1, 1))
        n = (padded[:-2, :-2] + padded[:-2, 1:-1] + padded[:-2, 2:]
             + padded[1:-1, :-2] + padded[1:-1, 2:]
             + padded[2:, :-2] + padded[2:, 1:-1] + padded[2:, 2:])
        board = ((n == 3) | ((board == 1) & (n == 2))).to(t.int64)
    return board


blinker = t.zeros((5, 5), dtype=t.int64)
blinker[2, 1:4] = 1
print(solve(blinker, 1))
```
