## Problem

Write a function solve(a, b) that takes a column vector a of shape (m, 1) and a row vector b of shape (1, n), and returns their broadcast sum of shape (m, n), where entry [i, j] equals a[i, 0] + b[0, j]. (Broadcasting is the whole answer: shapes (m, 1) and (1, n) line up to (m, n) on their own, so adding the two tensors directly grades the same as any explicit loop.)

## Reference solution


```python
import torch as t
def solve(a, b):
    return a + b


print(solve(t.arange(3).reshape(3, 1), t.arange(3).reshape(1, 3)))
```
