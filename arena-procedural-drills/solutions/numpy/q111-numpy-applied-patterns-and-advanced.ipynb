## Problem

Considering two arrays with shape (1,3) and (3,1), how to compute their sum using an iterator? (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Let nditer allocate the output by passing None, then write the sum into the third operand inside the loop.

</details>

## Reference solution


```python
import numpy as np

def solve():
    A = np.arange(3).reshape(3, 1)
    B = np.arange(3).reshape(1, 3)
    it = np.nditer([A, B, None])
    for x, y, z in it:
        z[...] = x + y
    return it.operands[2]

print(solve())
```

## Why this works

np.nditer with a None operand allocates an output broadcast to the combined (3, 3) shape of the (3, 1) and (1, 3) inputs. Iterating writes each broadcast sum into the output buffer, which is then read back from it.operands[2].
