## Problem

Write a function solve(a, b) that takes a column vector a of shape (m, 1) and a row vector b of shape (1, n), and returns their broadcast sum of shape (m, n), where entry [i, j] equals a[i, 0] + b[0, j]. (The classic exercise does this with np.nditer([a, b, None]) — try the iterator, though any broadcasting approach grades the same.)


<details><summary>💡 Hint (click to reveal)</summary>

np.nditer with a None operand allocates the broadcast output for you and iterates all three in lockstep — assigning through z[...] fills it.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    it = np.nditer([a, b, None])
    for x, y, z in it:
        z[...] = x + y
    return it.operands[2]


print(solve(np.arange(3).reshape(3, 1), np.arange(3).reshape(1, 3)))
```

## Why this works

np.nditer([a, b, None]) broadcasts the operands to (m, n) and allocates the output; each iteration exposes one aligned element triple, and z[...] = x + y writes the sum. Plain a + b relies on the very same broadcasting rules — the iterator makes the machinery explicit, which is the point of the exercise.
