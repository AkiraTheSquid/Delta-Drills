## Problem

Write a function solve(n) that takes an integer n >= 2 and returns a structured numpy array of shape (n, n) with two float fields named 'x' and 'y', where entry [i, j] holds the coordinates of a grid point covering the unit square: 'x' varies across the columns from 0 to 1, and 'y' varies down the rows from 0 to 1, both in n evenly spaced values.

Allocate a zeroed structured array with two float fields, then fill both fields at once from a coordinate-grid pair.

## Reference solution


```python
import numpy as np

def solve(n):
    z = np.zeros((n, n), [('x', float), ('y', float)])
    z['x'], z['y'] = np.meshgrid(np.linspace(0, 1, n), np.linspace(0, 1, n))
    return z


print(solve(2))
```

## Why this works

np.zeros((n, n), [('x', float), ('y', float)]) reserves the two named fields; np.meshgrid over two linspaces produces the X grid (varying across columns) and Y grid (varying down rows), which slot into the fields by name. Field access z['x'] is how structured arrays expose their components.
