## Problem

Create a structured array with `x` and `y` coordinates covering the [0,1]x[0,1] area (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Define a structured dtype with x and y fields, then fill them from a meshgrid of two linspaces.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros((5, 5), [('x', float), ('y', float)])
    Z['x'], Z['y'] = np.meshgrid(np.linspace(0, 1, 5), np.linspace(0, 1, 5))
    return Z

print(solve())
```

## Why this works

A structured dtype with named float fields 'x' and 'y' lets one array hold paired coordinates per cell. np.meshgrid over two linspaces on [0,1] fills the x and y fields with the grid coordinate values, giving a 5x5 grid covering the unit square.
