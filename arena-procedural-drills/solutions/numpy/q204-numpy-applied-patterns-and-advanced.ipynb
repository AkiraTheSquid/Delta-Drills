## Problem

Given a 1D integer vector I of length n=9 whose entries fit in 8 unsigned bits, build the n-by-8 matrix B of 0s and 1s in which row i holds the 8-bit binary representation of I[i] with the most significant bit in column 0 and the least significant bit in column 7. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Cast the vector to uint8, give it a trailing axis, and let np.unpackbits along that axis produce the MSB-first 8-bit rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    I = np.array([0, 1, 2, 3, 15, 16, 32, 64, 128], dtype=np.uint8)
    return np.unpackbits(I[:, np.newaxis], axis=1)

print(solve())
```

## Why this works

np.unpackbits expands each uint8 into its 8 constituent bits, most-significant-bit first, which matches the requested column order (MSB in column 0, LSB in column 7). Adding a trailing axis with [:, np.newaxis] makes each integer its own row so unpacking along axis 1 yields the n-by-8 binary matrix.
