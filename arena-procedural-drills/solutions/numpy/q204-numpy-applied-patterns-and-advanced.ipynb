## Problem

Write a function solve(v) that takes a 1-D tensor of unsigned integers below 256 and returns an (len(v), 8) integer matrix of 0s and 1s where row i is the 8-bit binary representation of v[i], MOST significant bit first (so 1 encodes as 00000001, 128 as 10000000).

## Reference solution


```python
import torch as t

def solve(v):
    bits = ((v.reshape(-1, 1) & (2 ** t.arange(8))) != 0).to(t.int64)
    return bits.flip(1)


print(solve(t.tensor([1, 128])))
```
