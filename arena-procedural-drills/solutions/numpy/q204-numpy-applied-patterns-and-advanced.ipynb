## Problem

Write a function solve(v) that takes a 1-D array of unsigned integers below 256 and returns an (len(v), 8) integer matrix of 0s and 1s where row i is the 8-bit binary representation of v[i], MOST significant bit first (so 1 encodes as 00000001, 128 as 10000000).


<details><summary>💡 Hint (click to reveal)</summary>

Bitwise-AND each value against the eight powers of two at once by broadcasting — then flip the column order for MSB-first.

</details>

## Reference solution


```python
import numpy as np

def solve(v):
    bits = ((v.reshape(-1, 1) & (2 ** np.arange(8))) != 0).astype(int)
    return bits[:, ::-1]


print(solve(np.array([1, 128])))
```

## Why this works

v.reshape(-1, 1) & 2**np.arange(8) tests every bit of every value simultaneously (nonzero where the bit is set); casting the boolean to int gives LSB-first columns, and [:, ::-1] reverses to the required MSB-first layout. np.unpackbits(v.astype(np.uint8)[:, None], axis=1) is the built-in shortcut.
