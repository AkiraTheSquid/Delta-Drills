## Problem

Write a function solve(z) that takes a 1-D float32 numpy array with non-negative values and converts it IN PLACE into an int32 view of the SAME memory buffer: return an int32 array that shares z's memory and holds the truncated integer value of each original float. No new data buffer may be allocated (the grader checks np.shares_memory).


<details><summary>💡 Hint (click to reveal)</summary>

view() re-types the SAME 4 bytes per element; assigning the float values through the int view is what performs the conversion without allocating.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    y = z.view(np.int32)
    y[:] = z
    return y


example = np.array([1.5, 99.75, 0.25], dtype=np.float32)
print(solve(example))
```

## Why this works

z.view(np.int32) reinterprets the float32 buffer as int32 (same size, same memory) — at that instant it holds garbage bit patterns. The assignment y[:] = z reads the floats and stores their truncated integer values back into the shared buffer, so the result both shares memory and holds the converted numbers. z.astype(np.int32) would allocate a new buffer and fail the shares_memory check.
