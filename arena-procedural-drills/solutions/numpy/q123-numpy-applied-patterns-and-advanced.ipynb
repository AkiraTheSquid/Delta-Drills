## Problem

Given a length-10 float32 array Z whose values are uniform random numbers in [0, 100), convert it to an int32 array Y that occupies the same underlying memory buffer (no new allocation for the data) and holds the truncated integer value of each original float. Y has shape (10,) and dtype int32. 
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Reinterpret the buffer with a same-width int view, then assign the floats back into it to trigger the cast.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = (np.random.rand(10) * 100).astype(np.float32)
    Y = Z.view(np.int32)
    Y[:] = Z
    return Y

print(solve())
```

## Why this works

Z.view(np.int32) reinterprets the same float32 buffer as int32 without allocating new data. Assigning Y[:] = Z then casts the float values (truncating toward zero) into that shared buffer, performing the conversion in place.
