## Problem

Write a function solve(z) that takes a 1-D float32 PyTorch tensor with non-negative values and converts it IN PLACE into an int32 view of the SAME memory buffer: return an int32 tensor that shares z's memory and holds the truncated integer value of each original float. No new data buffer may be allocated (the grader checks that the result and z report the same untyped_storage().data_ptr()).


<details><summary>💡 Hint (click to reveal)</summary>

view() re-types the SAME 4 bytes per element; assigning the float values through the int view is what performs the conversion without allocating.

</details>

## Reference solution


```python
import torch as t

def solve(z):
    y = z.view(t.int32)
    y[:] = z
    return y


example = t.tensor([1.5, 99.75, 0.25], dtype=t.float32)
print(solve(example))
```
