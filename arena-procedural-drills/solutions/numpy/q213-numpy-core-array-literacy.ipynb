## Problem

Write a function solve(n, idx) that takes a length n and a valid index idx, and returns a 1-D float PyTorch tensor of n zeros except for a 1.0 at position idx — a standard basis (one-hot) vector.

## Reference solution


```python
import torch as t

def solve(n, idx):
    z = t.zeros(n)
    z[idx] = 1.0
    return z


print(solve(10, 4))
```

## Why this works

t.zeros(n) allocates the float vector and a single indexed assignment plants the 1.0. This is the standard basis vector e_idx, the building block of one-hot encodings.
