## Problem

Build a 1D float vector Z of length 10 whose entries are all 0.0 except the entry at index position 4 (the fifth entry, counting from position 0), which holds the value 1.0. The result is a length-10 vector that is zero everywhere except at that one position. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Make a zero vector of the right length, then assign into the single index you need (remember positions are counted from 0).

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros(10)
    Z[4] = 1
    return Z

print(solve())
```

## Why this works

Start from a length-10 zero vector and assign 1.0 to index 4 (the fifth position, zero-based), leaving every other entry at 0.0.
