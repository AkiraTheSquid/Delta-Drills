## Problem

Compute the pairwise Euclidean distance matrix between rows of X (n,d) without explicit Python loops (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Insert a new axis so the two copies broadcast into an (n, n, d) difference tensor, then reduce over the feature axis.

## Reference solution


```python
import numpy as np

def solve():
    diff = X[:, None, :] - X[None, :, :]
    return np.sqrt(np.sum(diff * diff, axis=-1))
```

## Why this works

Broadcasting X[:, None, :] against X[None, :, :] forms an (n, n, d) tensor of coordinate differences for every row pair. Squaring, summing over the feature axis, and taking the square root yields the full pairwise Euclidean distance matrix with no Python loops.
