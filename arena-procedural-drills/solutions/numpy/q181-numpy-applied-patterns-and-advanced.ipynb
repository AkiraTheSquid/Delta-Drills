## Problem

Write a function solve(x, y) that takes two 2-D float tensors x of shape (n, d) and y of shape (m, d), and returns the (n, m) matrix of pairwise Euclidean distances: entry [i, j] is the distance between row i of x and row j of y. (An earlier drill measures one point set against itself; here the two sets differ.) Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Give x a new middle axis and y a new leading axis — the difference then broadcasts to (n, m, d) and one reduction finishes it.

</details>

## Reference solution


```python
import torch as t

def solve(x, y):
    diff = x[:, None, :] - y[None, :, :]
    return t.sqrt((diff ** 2).sum(dim=-1))


print(solve(t.tensor([[0.0, 0.0]]), t.tensor([[3.0, 4.0], [6.0, 8.0]])))
```

## Why this works

x[:, None, :] is (n, 1, d) and y[None, :, :] is (1, m, d); their difference broadcasts to every cross pair. Squaring, summing the coordinate axis, and rooting gives the rectangular distance table — the two-set generalization of the self-distance drill.
