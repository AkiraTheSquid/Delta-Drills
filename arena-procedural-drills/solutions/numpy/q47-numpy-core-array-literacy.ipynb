## Problem

Write a function solve(vals) that takes a 1-D sequence of k numbers (a numpy array or a plain Python list) and returns the k x k matrix that has those values along its main diagonal and zeros everywhere else. The result must be a 2-D numpy array. Do not use Python loops.

## Reference solution


```python
import numpy as np

def solve(vals):
    return np.diag(np.asarray(vals))


example = np.array([1.0, 2.0, 3.0])
print(solve(example))
```

## Why this works

np.diag(v), when given a 1-D array v of length k, constructs a k x k array with v along the main diagonal and zeros everywhere else. Wrapping the input in np.asarray first lets the same code accept plain Python lists as well as arrays. Note that np.diag is its own inverse of sorts: applied to a 2-D array it instead extracts the diagonal, so the dimensionality of the input decides which behavior you get.
