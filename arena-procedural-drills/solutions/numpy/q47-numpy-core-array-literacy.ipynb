## Problem

Construct a 3x3 square matrix where the values along the main diagonal are 1, 2, and 3, while all other elements are zero. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

np.diag turns a 1D vector into a diagonal matrix. Build the vector [1,2,3] from a range first.

</details>

## Reference solution


```python
import numpy as np


def solve():
    return np.diag(1 + np.arange(3))


print(solve())
```

## Why this works

np.arange(3) gives [0,1,2]; adding 1 yields [1,2,3]. np.diag of a 1D array builds a square matrix with that vector on the main diagonal and zeros elsewhere, exactly the requested 3x3.
