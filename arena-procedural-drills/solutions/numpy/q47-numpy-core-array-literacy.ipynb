## Problem

Write a function solve(vals) that takes a 1-D sequence of k numbers (a PyTorch tensor or a plain Python list) and returns the k x k matrix that has those values along its main diagonal and zeros everywhere else. The result must be a 2-D PyTorch tensor. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has a single function that, given a 1-D vector, builds the square matrix with that vector on the main diagonal.

</details>

## Reference solution


```python
import torch as t

def solve(vals):
    return t.diag(t.as_tensor(vals))


example = t.tensor([1.0, 2.0, 3.0])
print(solve(example))
```
