## Problem

Given an array containing the elements [1, 2, 3], create a new array that repeats these elements three times in sequence. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use the function that repeats the entire array as a block, not the one that repeats individual elements.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([1, 2, 3])
    return np.tile(Z, 3)

print(solve())
```

## Why this works

np.tile concatenates whole copies of the array end to end, so tiling [1,2,3] three times produces the block-repeated sequence 1 2 3 1 2 3 1 2 3.
