## Problem

Write a function solve(x, k) that takes a 1-D numpy array x and a positive integer k, and returns a new 1-D array containing the entire sequence of x repeated k times end-to-end. For example, repeating [1, 2, 3] twice gives [1, 2, 3, 1, 2, 3] — the whole block repeats in order, not each element individually. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a built-in that lays whole copies of an array end-to-end a given number of times — a different function from the one that repeats each element in place.

## Reference solution


```python
import numpy as np

def solve(x, k):
    return np.tile(x, k)


example = np.array([1.0, 2.0, 3.0])
print(solve(example, 2))
```

## Why this works

np.tile(x, k) concatenates k full copies of x end-to-end, producing a new array of length k * len(x). Unlike np.repeat, which duplicates each element where it stands ([1, 1, 2, 2, 3, 3]), tile preserves the whole block's ordering on every pass ([1, 2, 3, 1, 2, 3]). It returns a fresh array and handles the edge cases naturally: k = 1 gives a copy of the input, and a single-element array simply grows to length k.
