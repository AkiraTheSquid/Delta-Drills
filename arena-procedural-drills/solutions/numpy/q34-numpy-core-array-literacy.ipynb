## Problem

Write a function solve(x, k) that takes a 1-D PyTorch tensor x and a positive integer k, and returns a new 1-D tensor containing the entire sequence of x repeated k times end-to-end. For example, repeating [1, 2, 3] twice gives [1, 2, 3, 1, 2, 3] — the whole block repeats in order, not each element individually. torch has two similarly named operations here and only one does this. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has a built-in that lays whole copies of a tensor end-to-end a given number of times — a different function from the one that repeats each element in place.

</details>

## Reference solution


```python
import torch as t

def solve(x, k):
    return x.repeat(k)


example = t.tensor([1.0, 2.0, 3.0])
print(solve(example, 2))
```
