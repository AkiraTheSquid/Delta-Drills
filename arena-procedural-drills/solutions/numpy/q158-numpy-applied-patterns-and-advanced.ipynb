## Problem

You are given Z, a 2D integer array of shape (5, 8) whose entries are random integers from 0 to 5 inclusive. Return counts, a 1D array of shape (5,) where counts[i] is the number of distinct values that appear in row i of Z. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Apply a per-row function that returns the size of the unique values along the row axis.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 6, (5, 8))
    counts = np.apply_along_axis(lambda r: np.unique(r).size, 1, Z)
    return counts

print(solve())
```

## Why this works

For each row the number of distinct values is the size of np.unique on that row. np.apply_along_axis runs that small function over axis 1, returning one count per row in a length-5 array.
