## Problem

Write a function solve(z) that takes a 2-D PyTorch integer tensor and returns a 1-D tensor with one entry per row: the number of DISTINCT values appearing in that row.


```python
%pip install -q numpy torch --index-url https://download.pytorch.org/whl/cpu
```

## Reference solution


```python
import torch as t
def solve(z):
    return t.tensor([t.unique(row).numel() for row in z])


print(solve(t.tensor([[1, 1, 2], [3, 4, 5]])))
```
