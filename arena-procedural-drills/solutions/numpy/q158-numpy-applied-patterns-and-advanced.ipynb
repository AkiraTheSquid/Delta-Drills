## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns a 1-D array with one entry per row: the number of DISTINCT values appearing in that row.


<details><summary>💡 Hint (click to reveal)</summary>

When a reduction has no vectorized form, apply_along_axis runs your one-row function over every row — here, the size of each row's unique set.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.apply_along_axis(lambda r: np.unique(r).size, 1, z)


print(solve(np.array([[1, 1, 2], [3, 4, 5]])))
```

## Why this works

np.unique(r).size counts a single row's distinct values; np.apply_along_axis maps that over axis 1, one row at a time. A fully vectorized alternative sorts each row and counts positions that differ from their neighbor: (np.diff(np.sort(z, axis=1), axis=1) != 0).sum(axis=1) + 1.
