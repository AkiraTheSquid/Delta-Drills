## Problem

Write a function solve(z, y, threshold) that takes two float arrays of the same shape and a scalar threshold. Return a new array equal to z except that every position where y is strictly greater than threshold is replaced by -1.0. Do not modify the inputs.


<details><summary>💡 Hint (click to reveal)</summary>

A boolean mask built from ONE array can index a DIFFERENT array of the same shape.

</details>

## Reference solution


```python
import numpy as np

def solve(z, y, threshold):
    out = z.copy()
    out[y > threshold] = -1.0
    return out


example_z = np.array([1.0, 2.0, 3.0])
example_y = np.array([0.9, 0.1, 0.6])
print(solve(example_z, example_y, 0.5))
```

## Why this works

y > threshold is a boolean mask; using it to index a copy of z assigns -1.0 exactly where the condition holds in y. The strictly-greater comparison leaves y == threshold positions untouched, which the last case pins down.
