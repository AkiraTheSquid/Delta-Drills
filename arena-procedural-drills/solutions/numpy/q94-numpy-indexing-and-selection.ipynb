## Problem

Given two 4x4 arrays Z and Y of random values, replace the elements in Z with -1 where the corresponding elements in Y are greater than 0.5. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Build a boolean mask from the comparison on the second array and assign into the first array through that mask.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.rand(4, 4)
    Y = np.random.rand(4, 4)
    Z[Y > 0.5] = -1
    return Z

print(solve())
```

## Why this works

Y > 0.5 builds a boolean mask aligned elementwise with Z. Assigning -1 through that boolean mask updates only the positions where the condition holds, leaving every other entry of Z unchanged.
