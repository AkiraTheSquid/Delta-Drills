## Problem

Write a function solve(x, k) that takes an (n, d) float array of points (all pairwise distances distinct) and an integer k, and returns an (n, k) integer array whose row i lists the indices of the k points CLOSEST to point i, nearest first, always excluding the point itself.


<details><summary>💡 Hint (click to reveal)</summary>

Row-sorted squared distances put each point's own index first (distance zero) — the k neighbors follow immediately after.

</details>

## Reference solution


```python
import numpy as np

def solve(x, k):
    d2 = ((x[:, None, :] - x[None, :, :]) ** 2).sum(axis=-1)
    return np.argsort(d2, axis=1)[:, 1:k + 1]


pts = np.array([[0.0, 0.0], [1.0, 0.0], [4.0, 0.0]])
print(solve(pts, 1))
```

## Why this works

Broadcasting builds the full squared-distance matrix; sorting each row by distance yields the point itself at column 0 and its neighbors in nearest-first order at columns 1..k. Filling the diagonal with inf before an argpartition is the faster large-n variant, giving the same set.
