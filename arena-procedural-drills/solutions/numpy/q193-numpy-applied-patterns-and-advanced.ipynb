## Problem

Given an array X of shape (n, d) representing points in a d-dimensional space, return the indices of the k closest points to each point, excluding the point itself. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Build a pairwise squared-distance matrix by broadcasting, mask the diagonal with infinity, then argpartition each row to grab the k nearest indices.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.rand(8, 3)
    diff = X[:, None, :] - X[None, :, :]
    D2 = np.sum(diff * diff, axis=-1)
    np.fill_diagonal(D2, np.inf)
    k = 2
    idx = np.argpartition(D2, k, axis=1)[:, :k]
    return idx

print(solve())
```

## Why this works

Broadcasting X against itself gives all pairwise difference vectors, and summing their squares along the last axis yields a squared-distance matrix. Setting the diagonal to infinity excludes each point from being its own neighbor, then argpartition along each row pulls the k smallest-distance column indices to the front.
