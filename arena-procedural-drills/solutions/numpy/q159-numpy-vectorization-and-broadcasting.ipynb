## Problem

Consider the vector [1, 2, 3, 4, 5], how to build a new vector with 3 consecutive zeros interleaved between each value? (★★★)
Print the result.

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([1,2,3,4,5])
    nz = 3
    Z0 = np.zeros(len(Z) + (len(Z)-1)*nz)
    Z0[::nz+1] = Z
    return Z0

print(solve())
```

## Why this works

With nz zeros between each value, the output length is the count of originals plus nz zeros for each of the n-1 gaps. Allocating that zero array and writing the originals at every (nz+1)-th position via a strided slice leaves the zero runs in between.
