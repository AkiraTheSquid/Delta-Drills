## Problem

Write a function solve(z, nz) that takes a 1-D PyTorch tensor z and a non-negative integer nz, and returns a new tensor in which nz zeros are interleaved between every pair of consecutive entries of z: the result has length len(z) + (len(z) - 1) * nz, with z's values at positions 0, nz+1, 2*(nz+1), ...


<details><summary>💡 Hint (click to reveal)</summary>

Allocate the zero-filled result at its final size, then drop z's values in with ONE strided assignment — the stride is nz + 1.

</details>

## Reference solution


```python
import torch as t

def solve(z, nz):
    out = t.zeros(len(z) + (len(z) - 1) * nz)
    out[:: nz + 1] = z
    return out


print(solve(t.tensor([1, 2, 3, 4, 5]), 3))
```

## Why this works

The output length is len(z) plus nz gaps for each of the len(z)-1 intervals. Since consecutive values land nz+1 apart, out[::nz+1] = z places them all at once into the zero canvas. nz = 0 degenerates to a copy and a single element needs no gaps — the two edge cases.
