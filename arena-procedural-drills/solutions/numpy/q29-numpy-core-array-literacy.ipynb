## Problem

Write a function solve(x, y) that takes two 1-D PyTorch tensors: x with n elements and y with m elements. It must return a tuple (X, Y) of two 2-D tensors, each of shape (m, n). Every row of X is a copy of x, and every column of Y is a copy of y, so that (X[i, j], Y[i, j]) gives the coordinates of one point on the grid spanned by x and y. torch's meshgrid requires you to state the indexing convention explicitly. The two input lengths can differ, and either tensor may have just one element.


<details><summary>💡 Hint (click to reveal)</summary>

PyTorch has one function that turns a pair of coordinate vectors into exactly this pair of grids — the same one plotting code calls right before evaluating f(X, Y) over a surface.

</details>

## Reference solution


```python
import torch as t

def solve(x, y):
    X, Y = t.meshgrid(x, y, indexing='xy')
    return X, Y


example_x = t.tensor([1.0, 2.0, 3.0])
example_y = t.tensor([10.0, 20.0])
print(solve(example_x, example_y))
```
