## Problem

Write a function solve(x, y) that takes two 1-D numpy arrays: x with n elements and y with m elements. It must return a tuple (X, Y) of two 2-D arrays, each of shape (m, n). Every row of X is a copy of x, and every column of Y is a copy of y, so that (X[i, j], Y[i, j]) gives the coordinates of one point on the grid spanned by x and y. The two input lengths can differ, and either array may have just one element.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has one function that turns a pair of coordinate vectors into exactly this pair of grids — the same one plotting code calls right before evaluating f(X, Y) over a surface.

</details>

## Reference solution


```python
import numpy as np

def solve(x, y):
    X, Y = np.meshgrid(x, y)
    return X, Y


example_x = np.array([1.0, 2.0, 3.0])
example_y = np.array([10.0, 20.0])
print(solve(example_x, example_y))
```

## Why this works

np.meshgrid(x, y) with its default indexing returns exactly this pair: X of shape (len(y), len(x)) whose every row equals x, and Y of the same shape whose every column equals y. These coordinate matrices let you evaluate a function of two variables at every grid point in one vectorized expression. The same grids can also be built by hand with broadcasting or np.tile, which is what meshgrid does internally.
