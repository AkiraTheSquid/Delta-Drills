## Problem

Given two 1D arrays x and y with values [0, 1, 2] and [0, 1] respectively, create two 2D arrays that represent the coordinate matrices from these input arrays. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

A single NumPy function turns two 1D coordinate vectors into a pair of 2D coordinate grids.

</details>

## Reference solution


```python
import numpy as np

def solve():
    x, y = np.meshgrid(np.arange(3), np.arange(2))
    return y

print(solve())
```

## Why this works

np.meshgrid broadcasts the two 1D coordinate arrays into 2D coordinate matrices: x repeats the row values down each row, y repeats the column values across each row. The grader checks the y coordinate matrix, which has shape (2,3) with each row constant.
