## Problem

Write a function solve(x) that takes a 2-D float array of shape (n, m) with no zero rows, treats the rows as vectors, and returns the n x n cosine-similarity matrix: entry [i, j] is the dot product of rows i and j divided by the product of their Euclidean norms. The diagonal is (numerically) 1.


<details><summary>💡 Hint (click to reveal)</summary>

x @ x.T computes every pairwise dot product in one shot; an outer product of the row norms gives the matching denominators.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    norms = np.linalg.norm(x, axis=1, keepdims=True)
    return (x @ x.T) / (norms @ norms.T)


example = np.array([[1.0, 0.0], [0.0, 2.0]])
print(solve(example))
```

## Why this works

The Gram matrix x @ x.T holds all dot products; dividing by norms @ norms.T (the outer product of row lengths, kept 2-D via keepdims) normalizes each entry into a cosine. Parallel rows give 1, opposite rows -1, orthogonal rows 0 — the three hand-built cases.
