## Problem

You are given X, a 2D array of shape (5, 4) whose 5 rows are treated as 4-dimensional vectors. Return the 5x5 cosine-similarity matrix S where S[i, j] equals the dot product of rows i and j of X divided by the product of their Euclidean lengths. The diagonal of S is 1 and S is symmetric. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Compute all pairwise dot products with X @ X.T, then divide by the outer product of the per-row Euclidean norms.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    X = np.random.rand(5, 4)
    norms = np.linalg.norm(X, axis=1, keepdims=True)
    S = (X @ X.T) / (norms @ norms.T)
    return S

print(solve())
```

## Why this works

Cosine similarity is the matrix of pairwise dot products normalized by the outer product of the row lengths. X @ X.T gives every dot product at once; norms is the column of row magnitudes, so norms @ norms.T is the matrix whose (i,j) entry is |row_i||row_j|. Dividing elementwise produces the symmetric similarity matrix with ones on the diagonal.
