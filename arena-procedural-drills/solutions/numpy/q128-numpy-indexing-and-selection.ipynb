## Problem

Write a function solve(x) that takes a 2-D float tensor of shape (n, m) with no zero rows, treats the rows as vectors, and returns the n x n cosine-similarity matrix: entry [i, j] is the dot product of rows i and j divided by the product of their Euclidean norms. The diagonal is (numerically) 1.

## Reference solution


```python
import torch as t

def solve(x):
    norms = t.linalg.norm(x, dim=1, keepdim=True)
    return (x @ x.T) / (norms @ norms.T)


example = t.tensor([[1.0, 0.0], [0.0, 2.0]])
print(solve(example))
```
