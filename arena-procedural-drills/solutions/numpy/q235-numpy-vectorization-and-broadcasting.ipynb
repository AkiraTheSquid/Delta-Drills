## Problem

Given the 1D array Z of shape (10,) containing random floats in [0, 1), reorder its entries so they are in ascending order, modifying Z in place rather than producing a separate sorted copy. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use the in-place sort method on the array rather than a function that returns a new array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.random(10)
    Z.sort()
    return Z

print(solve())
```

## Why this works

ndarray.sort() reorders the array in place (unlike np.sort, which returns a copy), so Z itself ends up ascending. Returning Z hands back the same buffer now in sorted order.
