## Problem

Write a function solve(x) that takes a 1-D numpy array of floats and rearranges its entries into ascending order in place, meaning the array object passed in must itself be modified. The grader inspects the contents of x after calling solve(x), not just the return value, so building and returning a sorted copy while leaving x untouched will not pass. You may return x for convenience, but the mutation is what is checked.


<details><summary>💡 Hint (click to reveal)</summary>

An ndarray has a method that reorders its own contents, unlike the top-level numpy function that hands back a sorted copy.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    x.sort()
    return x


example = np.array([3.0, 1.0, 2.0])
solve(example)
print(example)
```

## Why this works

The ndarray method x.sort() sorts the array's own memory buffer into ascending order and returns None, so the caller's array is changed — which is exactly what an in-place contract requires. The top-level np.sort(x) instead allocates a brand-new sorted array and leaves x untouched, so it fails here unless you write the result back with slice assignment (x[:] = np.sort(x)). The grader checks the array it handed you after the call, so only genuinely mutating implementations pass.
