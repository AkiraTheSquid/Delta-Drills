## Problem

Write a function solve(x) that takes a 1-D PyTorch tensor of floats and rearranges its entries into ascending order in place, meaning the tensor object passed in must itself be modified. The grader inspects the contents of x after calling solve(x), not just the return value, so building and returning a sorted copy while leaving x untouched will not pass. You may return x for convenience, but the mutation is what is checked.


<details><summary>💡 Hint (click to reveal)</summary>

A tensor has a method that reorders its own contents, unlike the top-level PyTorch function that hands back a sorted copy.

</details>

## Reference solution


```python
import torch as t
def solve(x):
    x.copy_(x.sort().values)
    return x


example = t.tensor([3.0, 1.0, 2.0])
solve(example)
print(example)
```

## Why this works

The tensor method x.sort() sorts the tensor's own memory buffer into ascending order and returns None, so the caller's tensor is changed — which is exactly what an in-place contract requires. The top-level t.sort(x) instead allocates a brand-new sorted tensor and leaves x untouched, so it fails here unless you write the result back with slice assignment (x[:] = t.sort(x)). The grader checks the tensor it handed you after the call, so only genuinely mutating implementations pass.
