## Problem

Given an integer n, create a 2D boolean array where the upper triangular part, including the diagonal, is marked as True and the rest as False. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Start from an all-True matrix and apply the upper-triangle function.

## Reference solution


```python
import numpy as np

def solve():
    n = 5
    return np.triu(np.ones((n, n), dtype=bool))

print(solve())
```

## Why this works

np.triu zeroes out everything below the main diagonal, so applying it to an all-True boolean matrix leaves True on and above the diagonal and False below. This is the upper-triangular mask including the diagonal.
