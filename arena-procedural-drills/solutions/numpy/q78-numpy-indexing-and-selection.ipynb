## Problem

Write a function solve(n) that takes an integer n >= 1 and returns an n x n BOOLEAN numpy array in which the upper-triangular part including the main diagonal is True and everything strictly below the diagonal is False. The dtype must be bool, not integers.

## Reference solution


```python
import numpy as np

def solve(n):
    return np.triu(np.ones((n, n), dtype=bool))


print(solve(3))
```

## Why this works

np.triu keeps entries on and above the main diagonal and zeroes (False) the rest, so applying it to np.ones((n, n), dtype=bool) yields the boolean mask directly. The dtype check rejects the integer-0/1 version of the same shape.
