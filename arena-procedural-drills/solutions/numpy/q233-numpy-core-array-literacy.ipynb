## Problem

You are given Z, a 1D integer vector of length 50 containing the values 0, 1, 2, ..., 49 in order. Return a 1D vector of the same length whose entries appear in the reverse order, so the result starts at 49 and ends at 0. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

A negative slice step reverses traversal order.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(50)
    return Z[::-1]

print(solve())
```

## Why this works

A step of -1 in the slice Z[::-1] walks the array from the last element to the first, producing a reversed view. Returning it gives 49 down to 0 without any explicit indexing math.
