## Problem

Write a function solve(x) that takes a 1-D PyTorch tensor x and returns a new 1-D tensor of the same length with the elements in reverse order, so the last element comes first and the first element comes last. Do not use Python loops, and note that PyTorch — unlike numpy — does not support negative-step slicing, so x[::-1] will raise. The input tensor itself should not be modified.

## Reference solution


```python
import torch as t

def solve(x):
    return t.flip(x, [0])


example = t.tensor([1.0, 2.0, 3.0, 4.0])
print(solve(example))
```
