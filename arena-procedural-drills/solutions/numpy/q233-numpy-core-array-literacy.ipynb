## Problem

Write a function solve(x) that takes a 1-D numpy array x and returns a new 1-D array of the same length with the elements in reverse order, so the last element comes first and the first element comes last. Do not use Python loops. The input array itself should not be modified.


<details><summary>💡 Hint (click to reveal)</summary>

Slicing accepts a negative step, which walks the array from the end back to the start.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return x[::-1]


example = np.array([1.0, 2.0, 3.0, 4.0])
print(solve(example))
```

## Why this works

A slice with a negative step traverses the array from the last element to the first, so x[::-1] yields the whole array in reverse order. It returns a view rather than modifying x, and it works for any length, including a single-element array where the reversed result equals the input.
