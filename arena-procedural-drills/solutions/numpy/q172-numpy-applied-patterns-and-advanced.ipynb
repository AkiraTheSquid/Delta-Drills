## Problem

Given a small matrix Z of integers, determine the most frequently occurring value in each row, resolving ties by selecting the smallest value. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Count value occurrences per row with np.bincount; argmax over the counts naturally breaks ties toward the smaller value.

</details>

## Reference solution


```python
import numpy as np

def solve():
    m = Z.max() + 1
    return np.apply_along_axis(lambda r: np.bincount(r, minlength=m).argmax(), 1, Z)
```

## Why this works

For each row, np.bincount counts occurrences of every value 0..max, and argmax returns the index (the value) with the highest count, breaking ties toward the smallest value since argmax picks the first maximum. apply_along_axis runs this per row.
