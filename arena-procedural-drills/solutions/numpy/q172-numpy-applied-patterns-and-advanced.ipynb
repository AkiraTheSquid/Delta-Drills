## Problem

Write a function solve(z) that takes a 2-D array of small NON-NEGATIVE integers and returns a 1-D array with each row's MODE — its most frequently occurring value — breaking ties by choosing the SMALLEST tied value.


<details><summary>💡 Hint (click to reveal)</summary>

Per row: count occurrences by value with bincount, then take argmax — its first-index tie rule lands on the smallest value automatically.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    m = int(z.max()) + 1
    return np.apply_along_axis(lambda r: np.bincount(r, minlength=m).argmax(), 1, z)


print(solve(np.array([[1, 2, 2], [3, 3, 0]])))
```

## Why this works

np.bincount(row) indexes counts BY VALUE, so argmax over the counts returns the most frequent value, and because argmax reports the first maximum, tied counts resolve to the smallest value — exactly the required tie-break. apply_along_axis maps the recipe over rows; the all-distinct row ([2,1] etc.) degenerates to the smallest entry.
