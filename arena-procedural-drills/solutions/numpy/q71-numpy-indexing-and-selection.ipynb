## Problem

Write a function solve(labels, values) that takes a 1-D integer tensor labels holding class indices in the range 0..K-1 and a 1-D lookup table values of length K. Return a 1-D tensor the same length as labels whose i-th entry is values[labels[i]] — each label replaced by its looked-up value. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

Index the TABLE with the label tensor — fancy indexing gathers one table entry per label in a single operation.

</details>

## Reference solution


```python
import torch as t

def solve(labels, values):
    return values[labels]


example_labels = t.tensor([0, 2, 1, 2, 0])
example_values = t.tensor([10, 20, 30])
print(solve(example_labels, example_values))
```

## Why this works

values[labels] is a gather: for every position i it reads values[labels[i]], producing an output aligned with labels. Repeated labels simply read the same table slot repeatedly — the vectorized equivalent of a per-element dictionary lookup.
