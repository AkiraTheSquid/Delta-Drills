## Problem

You are given labels, a 1D integer array of length 5 holding class indices in the range 0..K-1, and values, a 1D lookup table of length K indexed by class. Produce a 1D array of length 5 whose i-th entry is the lookup-table value at position labels[i]. (★★☆)
Print the result.

Index the lookup table directly with the integer label array.

## Reference solution


```python
import numpy as np

def solve():
    labels = np.array([0, 2, 1, 2, 0])
    values = np.array([10, 20, 30])
    return values[labels]

print(solve())
```

## Why this works

Using an integer array to index another array performs a gather: values[labels] produces an array the same shape as labels, where each position holds values[labels[i]]. This is the standard lookup-table / one-to-many mapping pattern in NumPy.
