## Problem

Given p matrices M with shape (p, n, n) and p vectors V with shape (p, n, 1), compute the sum over p of M[i] @ V[i] in a single call. Return the (n, 1) result. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

A matrix-vector product is a contraction over the matrix columns and the vector rows; the batch sum is a contraction over the leading axis of each. np.tensordot can take both pairs of axes in one call.

</details>

## Reference solution


```python
import numpy as np

def solve():
    p, n = 10, 20
    M = np.ones((p, n, n))
    V = np.ones((p, n, 1))
    # Sum over p (axis 0 of both) and contract M's axis 2 with V's axis 1.
    S = np.tensordot(M, V, axes=[[0, 2], [0, 1]])
    return S

print(solve())
```

## Why this works

Summing M[i] @ V[i] over i is a single tensor contraction: pair M's batch axis (0) with V's batch axis (0) to sum over p, and pair M's last axis (2, the column index of each matrix) with V's middle axis (1, the vector's rows) to perform the matrix-vector product. np.tensordot with axes=[[0,2],[0,1]] does both contractions at once, leaving M's axis 1 and V's axis 2 to form the (n, 1) result.
