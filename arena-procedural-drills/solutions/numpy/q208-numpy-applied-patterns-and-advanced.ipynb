## Problem

Write a function solve(m, v) that takes a stack of p matrices m with shape (p, n, n) and a stack of p column vectors v with shape (p, n, 1), and returns the (n, 1) SUM over the batch of the products m[i] @ v[i] — all in vectorized numpy (tensordot or einsum), no Python loop over p.


<details><summary>💡 Hint (click to reveal)</summary>

Write the contraction as index notation: the batch index p and the inner index j both disappear, leaving ik.

</details>

## Reference solution


```python
import numpy as np

def solve(m, v):
    return np.einsum('pij,pjk->ik', m, v)


print(solve(np.ones((2, 2, 2)), np.ones((2, 2, 1))))
```

## Why this works

einsum('pij,pjk->ik', m, v) multiplies each m[p] with v[p] (contracting j) and, because p is absent from the output, sums the batch in the same pass. np.tensordot(m, v, axes=[[0, 2], [0, 1]]) expresses the identical double contraction.
