## Problem

Write a function solve(m, v) that takes a stack of p matrices m with shape (p, n, n) and a stack of p column vectors v with shape (p, n, 1), and returns the (n, 1) SUM over the batch of the products m[i] @ v[i] — all in vectorized PyTorch (tensordot or einsum), no Python loop over p.


<details><summary>💡 Hint (click to reveal)</summary>

Write the contraction as index notation: the batch index p and the inner index j both disappear, leaving ik.

</details>

## Reference solution


```python
import torch as t

def solve(m, v):
    return t.einsum('pij,pjk->ik', m, v)


print(solve(t.ones((2, 2, 2)), t.ones((2, 2, 1))))
```
