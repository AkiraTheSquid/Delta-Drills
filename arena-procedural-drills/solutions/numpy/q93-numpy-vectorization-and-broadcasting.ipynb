## Problem

Write a function solve(labels, k) that takes a 1-D integer array of class IDs in the range 0..k-1 and the class count k, and returns a 2-D float array of shape (len(labels), k) in which row i is the one-hot encoding of labels[i]: 1.0 in column labels[i], 0.0 elsewhere. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

The rows of the k x k identity matrix ARE the k one-hot vectors — gather the right ones by fancy indexing with the label array.

</details>

## Reference solution


```python
import numpy as np

def solve(labels, k):
    return np.eye(k)[labels]


print(solve(np.array([0, 2, 1, 2]), 3))
```

## Why this works

np.eye(k) holds every one-hot vector as its rows; indexing it with the labels array gathers row labels[i] into output row i. Passing k explicitly (rather than inferring labels.max()+1) keeps trailing unused classes — the k=5 case checks that.
