## Problem

You are given labels, a 1D integer array of length 4 with class IDs in the range 0..K-1, K, the number of distinct classes (here 3), and onehot, a 2D array of shape (4, K) where row i contains a 1.0 in column labels[i] and 0.0 elsewhere. Return onehot so each input label becomes its corresponding one-hot row vector. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Rows of an identity matrix are exactly the one-hot vectors; index it by the label array.

</details>

## Reference solution


```python
import numpy as np

def solve():
    labels = np.array([0, 2, 1, 2])
    K = labels.max() + 1
    return np.eye(K)[labels]

print(solve())
```

## Why this works

np.eye(K) is the K×K identity matrix whose row j is the one-hot vector for class j. Fancy-indexing that identity by the labels array gathers, for each label, its corresponding identity row, producing the (len(labels), K) one-hot matrix in a single step.
