## Problem

Write a function solve(labels, k) that takes a 1-D integer tensor of class IDs in the range 0..k-1 and the class count k, and returns a 2-D float tensor of shape (len(labels), k) in which row i is the one-hot encoding of labels[i]: 1.0 in column labels[i], 0.0 elsewhere. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

The rows of the k x k identity matrix ARE the k one-hot vectors — gather the right ones by fancy indexing with the label tensor.

</details>

## Reference solution


```python
import torch as t

def solve(labels, k):
    return t.eye(k)[labels]


print(solve(t.tensor([0, 2, 1, 2]), 3))
```

## Why this works

t.eye(k) holds every one-hot vector as its rows; indexing it with the labels tensor gathers row labels[i] into output row i. Passing k explicitly (rather than inferring labels.max()+1) keeps trailing unused classes — the k=5 case checks that.
