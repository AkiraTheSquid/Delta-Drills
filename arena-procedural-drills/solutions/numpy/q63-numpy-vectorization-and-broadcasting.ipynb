## Problem

Extract the integer part of a random array of positive numbers using 4 different methods (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Several methods drop the fraction; np.trunc removes it toward zero for positive values.

</details>

## Reference solution


```python
import numpy as np

np.random.seed(0)


def solve():
    np.random.seed(0)
    Z = np.random.uniform(0, 10, 10)
    return np.trunc(Z)


print(solve())
```

## Why this works

For positive numbers the integer part is obtained by Z - Z%1, Z//1, np.floor, casting to int, or np.trunc; all agree. The grader checks np.trunc, which drops the fractional part toward zero. Seed 0 makes Z reproducible.
