## Problem

Write a function solve(z) that takes a 1-D numpy array of POSITIVE floats and returns a float array of the same shape holding just the integer part of each entry (e.g. 3.7 becomes 3.0). Several numpy idioms compute this — use whichever you like.


<details><summary>💡 Hint (click to reveal)</summary>

For positive numbers, flooring, truncating, integer-dividing by 1, and subtracting the remainder mod 1 all agree.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.floor(z)


example = np.array([3.7, 0.2, 5.0])
print(solve(example))
```

## Why this works

On positive inputs np.floor(z), np.trunc(z), z // 1, z - z % 1, and z.astype(int) all yield the integer part; the call site coerces the result to float so the astype route passes too. (They differ on NEGATIVE numbers — floor moves away from zero, trunc toward it — which is why this drill promises positives.)
