## Problem

Write a function solve(n) that takes an odd positive integer n and returns an n x n integer tensor where entry [i, j] is the Manhattan (L1) distance from the center cell: |i - n//2| + |j - n//2|. The center itself is 0.

## Reference solution


```python
import torch as t
def solve(n):
    y = t.arange(n)[:, None]
    x = t.arange(n)[None, :]
    c = n // 2
    return (y - c).abs() + (x - c).abs()


print(solve(3))
```
