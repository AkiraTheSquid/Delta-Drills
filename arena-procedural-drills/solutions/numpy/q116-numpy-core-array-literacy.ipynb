## Problem

Given a 5x5 grid, compute a 2D array where each element represents the sum of the absolute differences from the center coordinates. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use open grid coordinates and add the absolute row and column distances from the center.

</details>

## Reference solution


```python
import numpy as np

def solve():
    n = 5
    Y, X = np.ogrid[:n, :n]
    center = (n // 2, n // 2)
    return np.abs(Y - center[0]) + np.abs(X - center[1])

print(solve())
```

## Why this works

np.ogrid yields open (broadcastable) coordinate arrays of shape (n, 1) and (1, n). Summing the absolute row-offset and column-offset from the center broadcasts them into a full (5, 5) Manhattan-distance map.
