## Problem

Write a function solve(n) that takes an odd positive integer n and returns an n x n integer array where entry [i, j] is the Manhattan (L1) distance from the center cell: |i - n//2| + |j - n//2|. The center itself is 0.


<details><summary>💡 Hint (click to reveal)</summary>

np.ogrid hands you a column of row indices and a row of column indices — broadcasting their absolute offsets adds up into the full grid.

</details>

## Reference solution


```python
import numpy as np

def solve(n):
    y, x = np.ogrid[:n, :n]
    c = n // 2
    return np.abs(y - c) + np.abs(x - c)


print(solve(3))
```

## Why this works

np.ogrid[:n, :n] yields an (n,1) column and a (1,n) row of indices; subtracting the center and taking absolute values gives each axis's distance, and the + broadcasts them into the (n, n) Manhattan-distance grid, 0 at the center and n-1+n-1... growing toward the corners.
