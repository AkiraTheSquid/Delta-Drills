## Problem

You are given Z, a 2D array of shape (3, 3) containing the integers 0 through 8 in row-major order, and R, the same array rotated 90 degrees counterclockwise about its center. Return R, a (3, 3) array where the original top row becomes the new left column and the original left column becomes the new bottom row. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

There is a single NumPy function for a 90-degree rotation.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    return np.rot90(Z)

print(solve())
```

## Why this works

np.rot90 rotates an array 90 degrees counterclockwise in the plane of its first two axes, so the original top row becomes the left column and the original left column becomes the bottom row. It is a pure view-rearrangement, no arithmetic involved.
