## Problem

Write a function solve(z) that takes a 2-D numpy array and returns it rotated 90 degrees counterclockwise: the last column of the input becomes the first row of the output. An input of shape (r, c) produces output of shape (c, r).

## Reference solution


```python
import numpy as np

def solve(z):
    return np.rot90(z)


example = np.arange(6).reshape(2, 3)
print(solve(example))
```

## Why this works

np.rot90(z) rotates counterclockwise by default: column c-1 becomes row 0. The composition z.T[::-1] is identical; the non-square and single-row cases pin the direction down, since a CLOCKWISE rotation produces a different answer there.
