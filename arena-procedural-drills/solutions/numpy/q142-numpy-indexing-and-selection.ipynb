## Problem

Identify the indices of rows in a 2D array Z that contain at least one non-finite value, such as NaN or Inf. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

Use isfinite plus an all-reduction per row, negate it, and read off the row indices with np.where.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randn(6, 4)
    Z[1, 2] = np.nan
    Z[4, 0] = np.inf
    mask = ~np.isfinite(Z).all(axis=1)
    return np.where(mask)[0]

print(solve())
```

## Why this works

np.isfinite flags every finite entry; reducing with all(axis=1) gives rows that are entirely finite. Negating that marks rows containing any NaN or Inf, and np.where returns those row indices.
