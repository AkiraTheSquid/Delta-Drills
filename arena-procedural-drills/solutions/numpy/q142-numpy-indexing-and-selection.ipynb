## Problem

Write a function solve(z) that takes a 2-D numpy array of floats which may contain NaN, Inf, or -Inf entries. Return a 1-D array of the row indices, in increasing order, of every row that contains at least one non-finite value. If every entry is finite, return an empty 1-D array. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

An elementwise finiteness test gives a boolean array the same shape as z — reduce it along the right axis, then turn the surviving booleans into index positions.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.where(~np.isfinite(z).all(axis=1))[0]


example = np.array([[1.0, 2.0, 3.0],
                    [4.0, np.nan, 6.0],
                    [7.0, 8.0, 9.0],
                    [np.inf, 10.0, 11.0]])
print(solve(example))
```

## Why this works

np.isfinite(z) marks every finite cell True, and .all(axis=1) collapses each row to a single boolean that is True only when the whole row is finite. Negating that vector flags exactly the rows containing at least one NaN or Inf, and np.where(...)[0] converts the boolean vector into the sorted array of matching row indices. Because np.where returns an empty integer array when nothing matches, the fully-finite case needs no special handling.
