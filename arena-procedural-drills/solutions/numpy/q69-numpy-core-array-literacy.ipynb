## Problem

Write a function solve(x, repeats) that takes a 1-D numpy array x and an integer array repeats of the same length, and returns a 1-D array in which each element x[i] appears exactly repeats[i] times consecutively, in order. A count of 0 drops that element entirely.

The element-repeating function accepts not just one count but a whole array of per-element counts.

## Reference solution


```python
import numpy as np

def solve(x, repeats):
    return np.repeat(x, repeats)


example_x = np.array([1, 2, 3])
example_r = np.array([1, 3, 2])
print(solve(example_x, example_r))
```

## Why this works

np.repeat(x, repeats) with an ARRAY second argument repeats each element by its own count — including zero, which removes the element. This per-element form is what distinguishes it from np.tile, which copies the whole array.
