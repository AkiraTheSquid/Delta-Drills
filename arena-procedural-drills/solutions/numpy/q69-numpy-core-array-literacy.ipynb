## Problem

Given a vector x and an array of integers indicating how many times to repeat each corresponding element, return a new vector that contains each element of x repeated according to the specified counts. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

np.repeat accepts an array of per-element counts, not just a single scalar.

</details>

## Reference solution


```python
import numpy as np

def solve():
    x = np.array([1, 2, 3])
    repeats = np.array([1, 3, 2])
    return np.repeat(x, repeats)

print(solve())
```

## Why this works

np.repeat takes an element-wise repeat count when its second argument is an array of the same length as the input, expanding each element x[i] exactly repeats[i] times and concatenating the results into one flat vector. This is the vectorised alternative to building the output with a Python loop.
