## Problem

Write a function solve(p, rng) that takes a 2-D tensor p of shape (rows, k) whose rows are probability distributions (non-negative, each summing to 1) and a PyTorch random Generator. Sample ONE category index per row via inverse-CDF: draw u = rng.random((rows, 1)), then return for each row the first category whose cumulative probability exceeds u. Return a 1-D integer tensor of length rows. Use exactly that recipe — the grader replays the same generator stream.

## Reference solution


```python
import torch as t
def solve(p, rng):
    u = t.rand((p.shape[0], 1), generator=rng)
    cs = t.cumsum(p, dim=1)
    return (u < cs).to(t.int64).argmax(dim=1)


probs = t.tensor([[0.2, 0.8], [1.0, 0.0]])
print(solve(probs, t.Generator().manual_seed(0)))
```
