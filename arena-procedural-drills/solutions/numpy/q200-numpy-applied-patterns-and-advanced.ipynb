## Problem

Sample categories according to row-wise probability matrix P (rows sum to 1) (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Take the per-row cumulative sum, draw one uniform per row, and find the first cumulative bin that exceeds it with argmax.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    P = np.random.dirichlet(np.ones(5), size=10)
    u = np.random.rand(P.shape[0], 1)
    cs = np.cumsum(P, axis=1)
    samples = (u < cs).argmax(axis=1)
    return samples

print(solve())
```

## Why this works

Inverse-CDF sampling per row: the cumulative sum of each probability row gives category thresholds, and a single uniform draw u per row falls into exactly one bin. (u < cs).argmax returns the first column where the cumulative probability exceeds u, which is the sampled category index.
