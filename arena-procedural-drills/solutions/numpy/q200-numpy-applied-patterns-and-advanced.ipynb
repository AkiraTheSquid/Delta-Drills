## Problem

Write a function solve(p, rng) that takes a 2-D array p of shape (rows, k) whose rows are probability distributions (non-negative, each summing to 1) and a numpy random Generator. Sample ONE category index per row via inverse-CDF: draw u = rng.random((rows, 1)), then return for each row the first category whose cumulative probability exceeds u. Return a 1-D integer array of length rows. Use exactly that recipe — the grader replays the same generator stream.


<details><summary>💡 Hint (click to reveal)</summary>

A row's cumulative sums split [0, 1) into segments sized by the probabilities — the first cumulative value exceeding the uniform draw names the sampled category.

</details>

## Reference solution


```python
import numpy as np

def solve(p, rng):
    u = rng.random((p.shape[0], 1))
    cs = np.cumsum(p, axis=1)
    return (u < cs).argmax(axis=1)


probs = np.array([[0.2, 0.8], [1.0, 0.0]])
print(solve(probs, np.random.default_rng(0)))
```

## Why this works

np.cumsum(p, axis=1) turns each row into its CDF; comparing a per-row uniform draw against it gives False until the draw's segment, and argmax finds that first True. Degenerate rows (probability 1 on one category) always return that category regardless of the draw.
