## Problem

Write a function solve(x, y) that takes two 1-D float arrays of the same length n with x[i] != y[j] for all pairs, builds the n x n Cauchy matrix C with C[i, j] = 1 / (x[i] - y[j]), and returns its determinant as a single scalar.


<details><summary>💡 Hint (click to reveal)</summary>

Every ufunc has an .outer form that applies it to all pairs — subtraction's outer gives the x[i] - y[j] table in one call.

</details>

## Reference solution


```python
import numpy as np

def solve(x, y):
    c = 1.0 / np.subtract.outer(x, y)
    return np.linalg.det(c)


example_x = np.arange(4).astype(float)
print(solve(example_x, example_x + 0.5))
```

## Why this works

np.subtract.outer(x, y) builds the full pairwise-difference matrix; taking reciprocals yields the Cauchy matrix, and np.linalg.det finishes the job. Broadcasting x[:, None] - y[None, :] is the same table written by hand.
