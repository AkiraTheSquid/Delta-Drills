## Problem

You are given X = [0, 1, 2, ..., 7] and Y = X + 0.5, two 1D arrays of length 8. Build the 8x8 Cauchy matrix C with C[i, j] = 1 / (X[i] - Y[j]), then return the determinant of C as a single scalar. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Form the pairwise-difference matrix with an outer subtraction, invert elementwise, then take the determinant.

</details>

## Reference solution


```python
import numpy as np

def solve():
    X = np.arange(8)
    Y = X + 0.5
    C = 1.0 / np.subtract.outer(X, Y)
    return np.linalg.det(C)

print(solve())
```

## Why this works

np.subtract.outer forms the full (8, 8) matrix of pairwise differences X[i] - Y[j], and taking the reciprocal builds the Cauchy matrix in one vectorized step. np.linalg.det then returns its determinant as a scalar.
