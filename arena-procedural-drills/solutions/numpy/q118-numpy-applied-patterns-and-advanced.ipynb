## Problem

Write a function solve(x, y) that takes two 1-D float tensors of the same length n with x[i] != y[j] for all pairs, builds the n x n Cauchy matrix C with C[i, j] = 1 / (x[i] - y[j]), and returns its determinant as a single scalar.

## Reference solution


```python
import torch as t
def solve(x, y):
    c = 1.0 / (x[:, None] - y[None, :])
    return t.linalg.det(c)


example_x = t.arange(4).to(t.float32)
print(solve(example_x, example_x + 0.5))
```
