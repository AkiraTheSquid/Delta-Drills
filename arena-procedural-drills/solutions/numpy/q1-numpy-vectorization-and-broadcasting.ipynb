## Problem

Given a 5x5 matrix Z of random integers, return the index of the largest value in each row as a 1D array. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

There is a single array method that returns the index of the maximum; tell it which axis to reduce over so it works row by row.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 10, (5, 5))
    return Z.argmax(axis=1)

print(solve())
```

## Why this works

argmax over axis=1 scans each row independently and returns the column index of that row's maximum, giving one index per row in a 1D array of length 5. Passing axis collapses only that dimension, which is what 'per row' requires.
