## Problem

Write a function solve(z) that takes a 2-D PyTorch tensor z and returns a 1-D integer tensor containing, for each row, the index of that row's largest value. If a row's maximum appears more than once, report the first (leftmost) occurrence. The matrix can have any shape, including a single row or a single column.


<details><summary>💡 Hint (click to reveal)</summary>

Reductions that report a position (not a value) also take an axis argument — pick the axis that runs across each row.

## Reference solution


```python
import torch as t

def solve(z):
    return z.argmax(dim=1)


example = t.tensor([[1, 9, 3], [7, 2, 5]])
print(solve(example))
```

## Why this works

z.argmax(axis=1) scans along axis 1 (the columns within each row) and returns one index per row. PyTorch's argmax reports the first occurrence on ties, which is exactly the required tie-break.
