## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns the successive differences between adjacent columns within each row: entry [i, j] of the result equals z[i, j+1] - z[i, j]. For input of shape (r, c) the result has shape (r, c-1). You may assume z has at least two columns. Do not use Python loops.


<details><summary>💡 Hint (click to reveal)</summary>

numpy has a discrete-difference function that takes an axis argument — or subtract two shifted column slices.

## Reference solution


```python
import numpy as np

def solve(z):
    return np.diff(z, axis=1)


example = np.arange(9).reshape(3, 3)
print(solve(example))
```

## Why this works

np.diff(z, axis=1) subtracts each column from the next within every row, dropping one column. The slicing identity z[:, 1:] - z[:, :-1] computes exactly the same result and both pass.
