## Problem

You are given Z, a 3x3 integer matrix containing the values 0 through 8 arranged row by row. For each row, compute the successive differences between adjacent column entries, producing a 2D array of shape (3, 2) where entry (i, j) equals Z[i, j+1] - Z[i, j]. (★☆☆)
Print the result.

There is a function for successive differences; point it at the column axis to difference adjacent entries within each row.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.arange(9).reshape(3, 3)
    return np.diff(Z, axis=1)

print(solve())
```

## Why this works

np.diff along axis=1 subtracts each column entry from the next within every row, producing successive differences and reducing that axis length by one, hence shape (3,2). This is exactly Z[i,j+1]-Z[i,j].
