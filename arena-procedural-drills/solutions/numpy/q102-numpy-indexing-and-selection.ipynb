## Problem

Write a function solve(z) that takes a 1-D numpy integer array and returns a tuple (values, counts): the distinct values of z and their occurrence counts, both ordered by DESCENDING count (most frequent value first). You may assume all counts are distinct, so the order is unambiguous.


<details><summary>💡 Hint (click to reveal)</summary>

Get values with their counts first, then compute the ordering that sorts the COUNTS and apply it to both arrays — reversed.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    vals, cnts = np.unique(z, return_counts=True)
    order = np.argsort(cnts)[::-1]
    return vals[order], cnts[order]


print(solve(np.array([4, 4, 4, 7, 7, 1])))
```

## Why this works

np.unique(..., return_counts=True) yields values ascending with aligned counts; argsort on the counts gives ascending count order, and [::-1] flips it to descending. Indexing both arrays with the same order keeps values and counts aligned.
