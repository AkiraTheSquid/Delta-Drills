## Problem

Given an array Z of random integers, return the distinct values along with their occurrences, arranged in descending order of the counts. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Get unique values with their counts, then argsort the counts and reverse to order both arrays by frequency descending.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 5, 20)
    vals, cnts = np.unique(Z, return_counts=True)
    order = np.argsort(cnts)[::-1]
    print(vals[order])
    return cnts[order]

print(solve())
```

## Why this works

np.unique with return_counts=True gives the distinct values and how often each occurs. argsort on the counts then reversing ([::-1]) produces the descending-count ordering, which is applied to both the values and the counts so they stay aligned.
