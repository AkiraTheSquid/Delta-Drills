## Problem

Write a function solve(x, w) that takes a 1-D PyTorch tensor x and a window length w (1 <= w <= len(x)), and returns a 2-D tensor of shape (len(x) - w + 1, w) whose row i is the contiguous window x[i : i + w]. Do not use Python loops.

## Reference solution


```python
import torch as t
def solve(x, w):
    return x.unfold(0, w, 1)


print(solve(t.arange(7), 3))
```
