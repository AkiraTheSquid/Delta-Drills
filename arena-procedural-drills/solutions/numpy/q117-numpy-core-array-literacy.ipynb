## Problem

You are given x, a 1D integer array of shape (7,) containing the values 0 through 6 in order. Return W, a 2D array of shape (5, 3) where each row is a contiguous length-3 window of x starting at successive positions, so W[i] equals x[i:i+3]. (★★☆)
Print the result.

Reach for the stride-trick that produces overlapping windows of a fixed size.

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve():
    x = np.arange(7)
    return sliding_window_view(x, window_shape=3)

print(solve())
```

## Why this works

sliding_window_view returns a strided view where each row is a length-3 contiguous window advanced by one position. For a length-7 input this gives 7 - 3 + 1 = 5 windows without copying data.
