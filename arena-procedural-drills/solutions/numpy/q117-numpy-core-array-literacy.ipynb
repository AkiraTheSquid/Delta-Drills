## Problem

Write a function solve(x, w) that takes a 1-D numpy array x and a window length w (1 <= w <= len(x)), and returns a 2-D array of shape (len(x) - w + 1, w) whose row i is the contiguous window x[i : i + w]. Do not use Python loops.

numpy's stride-tricks module has a purpose-built sliding-window view — no copying, no loops.

## Reference solution


```python
import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

def solve(x, w):
    return sliding_window_view(x, window_shape=w)


print(solve(np.arange(7), 3))
```

## Why this works

sliding_window_view(x, w) reinterprets x's memory as overlapping length-w windows using strides, producing (len(x)-w+1, w) rows without copying data. The w=1 and w=len(x) extremes confirm the shape arithmetic at both ends.
