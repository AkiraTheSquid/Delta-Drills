## Problem

Using the provided grid generated from X and Y coordinates, create a 2D array that represents a Gaussian distribution based on the specified parameters for standard deviation and mean. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Compute the radial distance from the grid, then apply the Gaussian exp(-(D-mu)^2 / (2 sigma^2)).

</details>

## Reference solution


```python
import numpy as np

def solve():
    X, Y = np.meshgrid(np.linspace(-1, 1, 10), np.linspace(-1, 1, 10))
    D = np.sqrt(X * X + Y * Y)
    sigma, mu = 1.0, 0.0
    return np.exp(-((D - mu) ** 2 / (2.0 * sigma ** 2)))

print(solve())
```

## Why this works

meshgrid builds 2D coordinate arrays; D is the radial distance of each grid point from the origin. Plugging that distance into the Gaussian formula exp(-(D-mu)^2 / (2 sigma^2)) gives a radially symmetric bump peaking at the center where D is smallest.
