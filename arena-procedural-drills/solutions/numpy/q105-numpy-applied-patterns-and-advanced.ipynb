## Problem

Write a function solve(n, sigma) that takes a grid size n and a standard deviation sigma, and returns an n x n tensor sampling a Gaussian bump over the unit square: build X and Y coordinates each as n evenly spaced values from -1 to 1 (X varying across columns, Y down rows), let D = sqrt(X^2 + Y^2), and return exp(-(D^2) / (2 * sigma^2)).


<details><summary>💡 Hint (click to reveal)</summary>

meshgrid over two linspaces gives every grid point's coordinates at once; the rest is elementwise arithmetic on those grids.

</details>

## Reference solution


```python
import torch as t
def solve(n, sigma):
    x, y = t.meshgrid(t.linspace(-1, 1, n), t.linspace(-1, 1, n), indexing="xy")
    d2 = x ** 2 + y ** 2
    return t.exp(-d2 / (2.0 * sigma ** 2))


print(solve(3, 1.0))
```
