## Problem

Write a function solve(n, sigma) that takes a grid size n and a standard deviation sigma, and returns an n x n array sampling a Gaussian bump over the unit square: build X and Y coordinates each as n evenly spaced values from -1 to 1 (X varying across columns, Y down rows), let D = sqrt(X^2 + Y^2), and return exp(-(D^2) / (2 * sigma^2)).


<details><summary>💡 Hint (click to reveal)</summary>

meshgrid over two linspaces gives every grid point's coordinates at once; the rest is elementwise arithmetic on those grids.

</details>

## Reference solution


```python
import numpy as np

def solve(n, sigma):
    x, y = np.meshgrid(np.linspace(-1, 1, n), np.linspace(-1, 1, n))
    d2 = x ** 2 + y ** 2
    return np.exp(-d2 / (2.0 * sigma ** 2))


print(solve(3, 1.0))
```

## Why this works

np.meshgrid(np.linspace(-1, 1, n), np.linspace(-1, 1, n)) produces the X and Y coordinate grids; squaring, summing, and exponentiating evaluates the Gaussian at all n^2 points simultaneously. For odd n the center point sits at (0, 0), so the bump's peak there is exactly 1.0.
