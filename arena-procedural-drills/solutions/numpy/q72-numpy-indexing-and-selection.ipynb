## Problem

Build an 8x8 integer matrix filled with a checkerboard pattern of 0s and 1s, where the top-left entry is 0, entries alternate between 0 and 1 along each row and column, and no two horizontally or vertically adjacent entries share the same value. (★★☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Use step-2 slices on rows and columns to assign 1 to the alternating cells of a zero matrix.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros((8, 8), dtype=int)
    Z[1::2, ::2] = 1
    Z[::2, 1::2] = 1
    return Z

print(solve())
```

## Why this works

Starting from a zero matrix, strided slicing sets the two diagonals of the checkerboard to 1: odd rows / even columns and even rows / odd columns. Together those two assignments cover exactly the cells where row+column is odd, giving a checkerboard whose top-left cell stays 0.
