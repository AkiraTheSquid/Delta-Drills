## Problem

Write a function solve(rows, cols, top_left) that takes two positive integers and a corner value top_left (0 or 1), and returns a rows x cols integer checkerboard: the entry at [0, 0] equals top_left, and values alternate between 0 and 1 along every row and every column so no two adjacent entries match.


<details><summary>💡 Hint (click to reveal)</summary>

A checkerboard cell's value is decided by the parity of row+column — the corner value just flips which parity gets the 1s.

</details>

## Reference solution


```python
import numpy as np

def solve(rows, cols, top_left):
    return (np.indices((rows, cols)).sum(axis=0) + top_left) % 2


print(solve(3, 3, 1))
```

## Why this works

(i + j + top_left) % 2 gives top_left at the corner (i = j = 0) and flips on every horizontal or vertical step, which is the definition of a checkerboard. np.indices supplies the i and j grids so the whole board is one vectorized expression.
