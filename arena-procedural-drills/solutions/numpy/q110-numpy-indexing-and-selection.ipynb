## Problem

Write a function solve(z) that takes a 2-D numpy array containing at least one nonzero entry and returns a tuple of four plain ints (row_min, row_max, col_min, col_max): the tightest bounding box of the nonzero entries — the first and last row indices that contain any nonzero value, and likewise for columns (both bounds inclusive).

## Reference solution


```python
import numpy as np

def solve(z):
    rows = np.any(z, axis=1)
    cols = np.any(z, axis=0)
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]
    return int(rmin), int(rmax), int(cmin), int(cmax)


example = np.zeros((4, 4), dtype=int)
example[1:3, 2] = 1
print(solve(example))
```

## Why this works

np.any(z, axis=1) flags rows holding any nonzero; np.where turns those flags into indices, whose first and last elements are the row bounds. The same along axis 0 gives the column bounds — together the tightest inclusive box, even for scattered corners.
