## Problem

Given a 2D array Z, identify the smallest rectangle that contains all the nonzero entries and return its row and column indices as a tuple of the form (row_min, row_max, col_min, col_max). (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Collapse nonzero presence onto each axis, then take the first and last indices that are True.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros((6, 7), dtype=int)
    Z[2:5, 3:6] = 1
    rows = np.any(Z, axis=1)
    cols = np.any(Z, axis=0)
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]
    return (rmin, rmax, cmin, cmax)

print(solve())
```

## Why this works

Reducing with np.any along each axis tells you which rows and columns contain any nonzero value. The first and last True positions (via np.where indexed with [0, -1]) give the tight bounding box of the nonzero region.
