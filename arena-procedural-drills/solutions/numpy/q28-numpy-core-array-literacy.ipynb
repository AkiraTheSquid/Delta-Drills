## Problem

Write a function solve(x) that takes a 1-D numpy array of floats and returns its arithmetic mean as a single float scalar. The grader calls solve() with several different vectors — including negatives and a length-1 vector — so do not hard-code any values or regenerate the data inside the function.

numpy arrays have a built-in reduction that collapses the whole vector to one number — the average of its entries.

## Reference solution


```python
import numpy as np

def solve(x):
    return float(x.mean())


example = np.array([1.0, 2.0, 3.0, 4.0])
print(solve(example))
```

## Why this works

x.mean() sums every entry of the vector and divides by its length, collapsing the whole array to one value. Wrapping it in float() makes the return type an honest Python scalar rather than a 0-d numpy value. Because the data arrives as the parameter x, the same one-liner works for any vector the grader passes in, including a single-element array.
