## Problem

You are given A, a 2D array of shape (5, 5) containing the integers 1 through 25 in row-major order, and an integer k (here -2) which may be positive, zero, or negative. Return diag_k, a 1D array containing the entries A[i+|k|, i] when k is negative or A[i, i+k] when k is non-negative, taken along the k-th diagonal of A in order of increasing index. (★★★)


<details><summary>💡 Hint (click to reveal)</summary>

np.diag takes a k offset — negative values pull diagonals below the main diagonal.

</details>

## Reference solution


```python
import numpy as np

def solve():
    A = np.arange(1, 26).reshape(5, 5)
    k = -2
    diag_k = np.diag(A, k=k)
    return diag_k

print(solve())
```

## Why this works

np.diag with the k argument extracts an off-diagonal: negative k selects diagonals below the main one (rows shifted down), positive k selects above. For k=-2 it returns A[i+2, i] for valid i, giving the third sub-diagonal in increasing-index order.
