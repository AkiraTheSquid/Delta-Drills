## Problem

Write a function solve(z, k) that takes a 2-D square integer array and an integer offset k (positive, zero, or negative), and returns the SUM of the entries on the k-th diagonal — the main diagonal for k = 0, above it for positive k, below it for negative k — as a plain Python int. (An earlier drill extracts the diagonal; this one reduces it.)


<details><summary>💡 Hint (click to reveal)</summary>

Extract the offset diagonal and sum it — or use the trace function, which takes the very same offset directly.

## Reference solution


```python
import numpy as np

def solve(z, k):
    return int(np.diag(z, k=k).sum())


example = np.arange(1, 10).reshape(3, 3)
print(solve(example, -1))
```

## Why this works

np.diag(z, k).sum() and np.trace(z, offset=k) both add up the entries z[i, i+k]; trace skips materializing the diagonal. Either way, int(...) converts the numpy scalar into the plain int the grader type-checks.
