## Problem

You are given Z, a 1D complex-valued array of shape (2,) whose two entries each carry a real part and an imaginary part. Produce two separate real-valued arrays of shape (2,): the first contains the real components of each entry, and the second contains the imaginary components. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Two companion functions pull out the real part and the imaginary part of a complex array separately.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([1 + 2j, 3 + 4j])
    return np.real(Z), np.imag(Z)

print(solve())
```

## Why this works

np.real and np.imag extract the real and imaginary components of a complex array element-wise, each returning a real-valued array of the same shape. Together they split the (2,) complex array into two (2,) real arrays.
