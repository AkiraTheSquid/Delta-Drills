## Problem

Write a function solve(z) that takes a 1-D complex-valued numpy array and returns a tuple of two real-valued arrays of the same length: the first holding the real part of each entry, the second holding the imaginary part.


<details><summary>💡 Hint (click to reveal)</summary>

Complex arrays expose their two components either as array attributes or through a matching pair of numpy functions.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    return np.real(z), np.imag(z)


example = np.array([1 + 2j, 3 + 4j])
print(solve(example))
```

## Why this works

np.real(z) and np.imag(z) (equivalently z.real and z.imag) extract the two real-valued components of a complex array as separate arrays of the same shape, which the function returns as a tuple.
