## Problem

Write a function solve(rows, cols) that takes two positive integers and returns a rows x cols integer matrix holding a checkerboard pattern of 0s and 1s. The top-left entry is 0, and values alternate along every row and every column, so no two horizontally or vertically adjacent entries are equal.

## Reference solution


```python
import numpy as np

def solve(rows, cols):
    z = np.zeros((rows, cols), dtype=int)
    z[1::2, ::2] = 1
    z[::2, 1::2] = 1
    return z


print(solve(3, 4))
```

## Why this works

Strided slice assignment paints the two interleaved sublattices: odd rows get 1s in even columns, even rows get 1s in odd columns. Equivalently, (i + j) % 2 gives each cell's value directly — both produce the same integer checkerboard with 0 in the top-left corner.
