## Problem

Build an 8x8 integer matrix filled with a checkerboard pattern of 0s and 1s, where the top-left entry is 0, entries alternate between 0 and 1 along each row and column, and no two horizontally or vertically adjacent entries share the same value. (★☆☆)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Start from a zero matrix and use strided slicing (start::2) on rows and columns to set the two interleaved groups of cells to 1.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.zeros((8, 8), dtype=int)
    Z[1::2, ::2] = 1
    Z[::2, 1::2] = 1
    return Z

print(solve())
```

## Why this works

A checkerboard with a 0 in the top-left has 1s exactly where the row and column indices have different parity. Two strided slice assignments set the odd-row/even-col cells and the even-row/odd-col cells to 1, leaving the rest at the zero fill.
