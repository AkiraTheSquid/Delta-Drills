## Problem

Write a function solve(rows, cols) that takes two positive integers and returns a rows x cols integer matrix holding a checkerboard pattern of 0s and 1s. The top-left entry is 0, and values alternate along every row and every column, so no two horizontally or vertically adjacent entries are equal.

## Reference solution


```python
import torch as t

def solve(rows, cols):
    z = t.zeros((rows, cols), dtype=t.int64)
    z[1::2, ::2] = 1
    z[::2, 1::2] = 1
    return z


print(solve(3, 4))
```
