## Problem

Write a function solve(z, names) that takes a 2-D string array (each row holding string representations of the fields of one record) and a comma-separated field-name string like 'col1, col2, col3'. Return a numpy RECORD array of length z.shape[0] with those named fields, where each field k holds the k-th column of z. Field values stay strings; access must work as r['col1'] etc.


<details><summary>💡 Hint (click to reveal)</summary>

Records are built column-by-column: transpose the row-oriented array so each COLUMN becomes one field array, then hand the list to the record constructor.

</details>

## Reference solution


```python
import numpy as np

def solve(z, names):
    return np.rec.fromarrays(z.T, names=names)


example = np.array([('Hello', '2.5', '3'), ('World', '3.6', '2')])
print(solve(example, 'col1, col2, col3'))
```

## Why this works

np.rec.fromarrays expects one array per FIELD, so z.T supplies the columns of the row-oriented input; names labels them in order. The result supports r['col1']-style access per field. (In numpy 2.x use np.rec.fromarrays — the old np.core.records path is deprecated.)
