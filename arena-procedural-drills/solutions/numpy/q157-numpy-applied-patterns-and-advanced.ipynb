## Problem

You are given Z, a 2D string array of shape (2, 3) where each row holds a name, a numeric string, and an integer string, and R, a structured record array of length 2 with named fields col1 (8-byte string), col2 (float64), and col3 (int64) populated from the columns of Z. Return R so each row is accessible by field name. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Transpose so each column is one field, then feed the columns to a record-array constructor with names and formats.

</details>

## Reference solution


```python
import numpy as np

def solve():
    Z = np.array([("Hello", 2.5, 3),
                  ("World", 3.6, 2)])
    R = np.rec.fromarrays(Z.T,
                          names='col1, col2, col3',
                          formats='S8, f8, i8')
    return R

print(solve())
```

## Why this works

Z is stored as strings, so transposing it groups the data by column. np.rec.fromarrays builds a structured record array from those per-column sequences, casting each to its declared format (S8 string, f8 float, i8 int) so rows become accessible by field name.
