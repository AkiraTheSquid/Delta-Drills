## Problem

Given a 2D integer array Z of shape (8, 5), return the subarray containing exactly those rows whose 5 entries are strictly increasing from left to right (each entry strictly greater than the one before it). The result is 2D with shape (k, 5) where k is the number of qualifying rows. (★★★)
Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Check that every row-wise consecutive difference is positive, then mask-select those rows.

</details>

## Reference solution


```python
import numpy as np

def solve():
    np.random.seed(0)
    Z = np.random.randint(0, 10, (8, 5))
    mask = np.all(np.diff(Z, axis=1) > 0, axis=1)
    return Z[mask]

print(solve())
```

## Why this works

A row is strictly increasing exactly when all of its consecutive differences are positive. np.diff along axis 1 gives those differences, and np.all over each row builds a boolean mask that selects the qualifying rows via fancy indexing.
