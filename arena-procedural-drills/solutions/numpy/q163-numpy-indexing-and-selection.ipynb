## Problem

Write a function solve(z) that takes a 2-D numpy integer array and returns the subarray of rows that are STRICTLY INCREASING left to right (each entry greater than the one before it), preserving their original order. If no row qualifies, the result has zero rows.


<details><summary>💡 Hint (click to reveal)</summary>

Row-wise differences must ALL be positive for a strictly increasing row — reduce the comparison along the row axis into a row mask.

</details>

## Reference solution


```python
import numpy as np

def solve(z):
    mask = np.all(np.diff(z, axis=1) > 0, axis=1)
    return z[mask]


print(solve(np.array([[1, 2, 3], [3, 2, 1]])))
```

## Why this works

np.diff(z, axis=1) computes the within-row steps; requiring every step > 0 (np.all over axis 1) yields one boolean per row, and boolean indexing keeps the qualifying rows in order. Equal neighbors fail because the comparison is strict — the [5, 5, 6] row.
