## Problem

Using torch, you have a 1D signal x = [[-3.0, -1.0, -7.0, -2.0]] (shape [1,4]) whose values are all negative. Pad one cell on each side, then take a sliding max over windows of size 3 with stride 1 (use F.pad followed by .unfold(1,3,1).max(dim=2).values). Do this TWICE: once padding with value=float('-inf') (correct) and once padding with value=0.0 (wrong, since 0 beats every real negative at the edges). Print exactly print([correct[0].tolist(), wrong[0].tolist()]).

## Reference solution


```python
import torch
import torch.nn.functional as F
x = torch.tensor([[-3.0, -1.0, -7.0, -2.0]])
correct = F.pad(x, (1, 1), value=float('-inf')).unfold(1, 3, 1).max(dim=2).values
wrong = F.pad(x, (1, 1), value=0.0).unfold(1, 3, 1).max(dim=2).values
print([correct[0].tolist(), wrong[0].tolist()])
```

## Why this works

Max pooling must pad with the identity element of max, which is -inf, so padded cells can never win. Padding with 0 corrupts the edges of an all-negative signal because 0 beats every real value, producing spurious 0.0 maxima at the boundaries.
