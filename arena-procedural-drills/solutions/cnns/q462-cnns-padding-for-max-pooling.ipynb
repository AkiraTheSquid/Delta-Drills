## Problem

Padding for max-pooling must never beat real values. Write solve(x, pad_value) for a (1, N) signal: pad one cell on each side with `F.pad(..., value=pad_value)`, carve windows of size 3, stride 1 with `.unfold`, and take each window's max. Return the resulting (1, N) tensor. On an all-negative signal, padding with float('-inf') is correct; padding with 0.0 fabricates winners at the edges — the grader checks both.


<details><summary>💡 Hint (click to reveal)</summary>

`F.pad(x, (1, 1), value=...)` then `.unfold(1, 3, 1).max(dim=2).values`.

</details>

## Reference solution


```python
import torch
import torch.nn.functional as F

def solve(x, pad_value):
    padded = F.pad(x, (1, 1), value=pad_value)
    return padded.unfold(1, 3, 1).max(dim=2).values

x = torch.tensor([[-3.0, -1.0, -7.0, -2.0]])
print(solve(x, float('-inf')))
```

## Why this works

Padding participates in the max, so its value must never win: −∞ is the identity for max. Padding with 0.0 on an all-negative signal fabricates 0 winners at the edges — compare the two outputs and the bug is visible in the first and last positions.
