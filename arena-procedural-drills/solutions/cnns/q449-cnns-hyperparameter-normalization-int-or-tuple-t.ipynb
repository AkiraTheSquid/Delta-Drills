## Problem

Write a helper to_pair(x) that maps an int x to (x, x) and a length-2 sequence x to (int(x[0]), int(x[1])). Normalise each value in the list [5, (3, 1), 7, (2, 2)] and print the list of resulting (h, w) tuples, i.e. print([to_pair(v) for v in [5, (3, 1), 7, (2, 2)]]).


<details><summary>💡 Hint (click to reveal)</summary>

Apply the int-or-sequence normalizer to each list element with a comprehension.

</details>


## Reference solution


```python
def to_pair(x):
    if isinstance(x, int):
        return (x, x)
    return (int(x[0]), int(x[1]))

print([to_pair(v) for v in [5, (3, 1), 7, (2, 2)]])
```

## Why this works

to_pair normalizes mixed int-or-tuple hyperparameters into uniform (h, w) tuples: ints are duplicated and length-2 sequences are coerced to int tuples. Mapping it over the list yields one (h, w) tuple per entry.
