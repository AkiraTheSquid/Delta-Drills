## Problem

Write a function solve(kernel, stride, padding) that normalizes ALL THREE conv hyperparameters at once — each may be an int or a length-2 sequence — returning a tuple of three (h, w) int pairs: (kernel_pair, stride_pair, padding_pair).


<details><summary>💡 Hint (click to reveal)</summary>

Write to_pair once and map it over the three arguments — mixed int/list/tuple inputs all canonicalize the same way.

</details>

## Reference solution


```python
import torch as t

def solve(kernel, stride, padding):
    return (lambda P: (P(kernel), P(stride), P(padding)))(lambda v: (int(v), int(v)) if isinstance(v, int) else (int(v[0]), int(v[1])))


print(solve(3, (1, 2), [0, 1]))
```

## Why this works

Factoring the normalizer out and applying it uniformly is exactly how conv layer __init__ methods sanitize their arguments — after this, downstream arithmetic can assume pairs of plain ints everywhere.
