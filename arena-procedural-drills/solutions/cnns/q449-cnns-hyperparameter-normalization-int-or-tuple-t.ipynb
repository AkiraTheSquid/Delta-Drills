## Problem

Normalization helpers must cope with mixed inputs. Implement `to_pair` (int → symmetric pair; length-2 sequence → tuple of ints) and map it over the mixed list [5, (3, 1), 7, (2, 2)]. The starter's print line emits the list of resulting pairs.


<details><summary>💡 Hint (click to reveal)</summary>

Apply the int-or-sequence normalizer to each list element with a comprehension.

</details>


## Reference solution


```python
def to_pair(x):
    if isinstance(x, int):
        return (x, x)
    return (int(x[0]), int(x[1]))

print([to_pair(v) for v in [5, (3, 1), 7, (2, 2)]])
```

## Why this works

to_pair normalizes mixed int-or-tuple hyperparameters into uniform (h, w) tuples: ints are duplicated and length-2 sequences are coerced to int tuples. Mapping it over the list yields one (h, w) tuple per entry.
