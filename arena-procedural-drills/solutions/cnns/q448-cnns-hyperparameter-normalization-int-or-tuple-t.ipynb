## Problem

Write a function solve(hw, kernel, stride, padding) that combines the int-or-pair normalizer with per-axis conv arithmetic: each of kernel/stride/padding may be an int or an (h, w) pair; hw is the input (h, w) pair. Return the output (h', w') as a tuple of ints using floor((size + 2p - k)/s) + 1 per axis.


<details><summary>💡 Hint (click to reveal)</summary>

Normalize every hyperparameter to a pair FIRST, then run the size formula independently per axis.

</details>

## Reference solution


```python
import torch as t

def solve(hw, kernel, stride, padding):
    return (lambda P: (int((hw[0] + 2 * P(padding)[0] - P(kernel)[0]) // P(stride)[0] + 1), int((hw[1] + 2 * P(padding)[1] - P(kernel)[1]) // P(stride)[1] + 1)))(lambda v: (v, v) if isinstance(v, int) else (int(v[0]), int(v[1])))


print(solve((32, 32), 3, 1, 1))
```

## Why this works

Rectangular kernels and anisotropic strides make the two axes genuinely independent — normalize each argument with to_pair, then apply (size + 2p - k)//s + 1 twice. The mixed case (5,3)/(1,2) is where int-assumptions break.
