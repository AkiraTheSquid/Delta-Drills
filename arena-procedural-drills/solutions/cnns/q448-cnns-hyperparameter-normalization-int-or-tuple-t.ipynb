## Problem

Combine the `to_pair` normalizer with per-axis conv arithmetic. Implement `to_pair` (int → symmetric pair; length-2 sequence → tuple of ints), normalize kernel 3, stride (2, 1), padding 1, then compute the conv output size for H=32, W=28 — apply the output-size formula independently per axis, indexing the normalized pairs. Print exactly print((out_h, out_w)).


<details><summary>💡 Hint (click to reveal)</summary>

Normalize each hyperparameter to a pair, then apply the conv output formula separately for the height and width axes.

</details>


## Reference solution


```python
def to_pair(x):
    return (x, x) if isinstance(x, int) else (int(x[0]), int(x[1]))

H, W = 32, 28
k = to_pair(3)
s = to_pair((2, 1))
p = to_pair(1)
out_h = (H + 2 * p[0] - k[0]) // s[0] + 1
out_w = (W + 2 * p[1] - k[1]) // s[1] + 1
print((out_h, out_w))
```

## Why this works

After normalizing kernel, stride, and padding into (h, w) pairs, the standard conv output-size formula (size + 2*pad - kernel)//stride + 1 is applied independently to each spatial axis. Height and width use different stride components, so they shrink by different amounts.
