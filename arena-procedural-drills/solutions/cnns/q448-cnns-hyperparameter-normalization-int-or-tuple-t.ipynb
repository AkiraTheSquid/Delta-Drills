## Problem

Write a helper to_pair(x) that returns (x, x) when x is an int and (int(x[0]), int(x[1])) when x is a length-2 sequence. Use it to normalise kernel_size=3, stride=(2, 1), and padding=1, then compute the output spatial size of a 2D convolution on an input of height H=32, width W=28 using out = (size + 2*pad - kernel) // stride + 1 independently for the height and width axes. Print the result as a tuple, i.e. print((out_h, out_w)).


<details><summary>💡 Hint (click to reveal)</summary>

Normalize each hyperparameter to a pair, then apply the conv output formula separately for the height and width axes.

</details>


## Reference solution


```python
def to_pair(x):
    return (x, x) if isinstance(x, int) else (int(x[0]), int(x[1]))

H, W = 32, 28
k = to_pair(3)
s = to_pair((2, 1))
p = to_pair(1)
out_h = (H + 2 * p[0] - k[0]) // s[0] + 1
out_w = (W + 2 * p[1] - k[1]) // s[1] + 1
print((out_h, out_w))
```

## Why this works

After normalizing kernel, stride, and padding into (h, w) pairs, the standard conv output-size formula (size + 2*pad - kernel)//stride + 1 is applied independently to each spatial axis. Height and width use different stride components, so they shrink by different amounts.
