## Problem

A length-C vector broadcasts against a (B, C, H, W) tensor only if its channel axis lines up with dim 1. Write solve(weight): reshape the length-C `weight` so the batch, height, and width axes are singletons and C sits on the channel axis — shape (1, C, 1, 1). Return the reshaped tensor.


<details><summary>💡 Hint (click to reveal)</summary>

`weight.reshape(1, weight.shape[0], 1, 1)` — C on dim 1, singletons elsewhere.

</details>

## Reference solution


```python
import torch

def solve(weight):
    return weight.reshape(1, weight.shape[0], 1, 1)

weight = torch.tensor([1.0, 2.0, 3.0, 4.0])
print(solve(weight))
```

## Why this works

Against (B, C, H, W), a raw length-C vector would align with the trailing W axis. Reshaping to (1, C, 1, 1) puts C on the channel dim and lets broadcasting expand the three singleton axes — the standard shape for every per-channel BatchNorm vector.
