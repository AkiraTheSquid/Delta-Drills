## Problem

A BatchNorm layer with C=4 channels stores its weight as a length-C vector (shape (4,)). Before it can broadcast against a (B, C, H, W) activation tensor, it must be reshaped so the channel axis lines up with dim 1 and the batch/height/width axes are singleton. Take weight = torch.tensor([1.0, 2.0, 3.0, 4.0]), reshape it to (1, C, 1, 1), and print the resulting shape as a list of ints via print(list(shaped.shape)).


<details><summary>💡 Hint (click to reveal)</summary>

Put the channel count on dim 1 and make the batch and spatial dims singletons.

</details>

## Reference solution


```python
import torch

C = 4
weight = torch.tensor([1.0, 2.0, 3.0, 4.0])
shaped = weight.reshape(1, C, 1, 1)
print(list(shaped.shape))
```

## Why this works

Reshaping a length-C weight to (1, C, 1, 1) places the channel count on dim 1 with singleton batch and spatial dims, which is what lets it broadcast against a (B, C, H, W) activation. The resulting shape is simply [1, 4, 1, 1].
