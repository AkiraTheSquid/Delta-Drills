## Problem

Write a function solve(x) that normalizes a conv hyperparameter: an int becomes the symmetric pair (x, x); a length-2 sequence (list or tuple) becomes a tuple of its two values as plain ints. Always return a tuple of two ints.


<details><summary>💡 Hint (click to reveal)</summary>

Branch on isinstance(int): scalars duplicate, sequences convert element-wise — and always emit a plain tuple of ints.

</details>


## Reference solution


```python
import numpy as np

def solve(x):
    return (int(x), int(x)) if isinstance(x, int) else (int(x[0]), int(x[1]))


print(solve(3), solve((1, 2)))
```

## Why this works

This is the normalizer behind every kernel_size/stride/padding argument in conv APIs: ints mean 'same in both axes'. Converting sequence elements through int() canonicalizes lists and numpy scalars alike.
