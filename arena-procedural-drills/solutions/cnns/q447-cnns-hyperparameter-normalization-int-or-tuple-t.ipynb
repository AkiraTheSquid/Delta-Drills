## Problem

Conv hyperparameters like `kernel_size` accept either an int or an (h, w) pair. Write the normalizer `to_pair(x)`: an int becomes the symmetric pair, and a length-2 sequence becomes a tuple of its two values as ints. The starter's print line applies it to 3 and (2, 5).


<details><summary>💡 Hint (click to reveal)</summary>

Branch on whether the input is an int; duplicate it if so, otherwise unpack its two elements as ints.

</details>


## Reference solution


```python
def to_pair(x):
    if isinstance(x, int):
        return (x, x)
    return (int(x[0]), int(x[1]))

print([to_pair(3), to_pair((2, 5))])
```

## Why this works

Conv hyperparameters accept either a single int (same value for both spatial axes) or an explicit (h, w) pair. to_pair normalizes both forms: an int becomes (x, x) and a length-2 sequence is coerced to a tuple of ints.
