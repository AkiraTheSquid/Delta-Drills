## Problem

Using only numpy, take x = [-4.0, -1.0, -3.0, -2.0] (all negative). Build two padded versions by prepending and appending a single value: one with -np.inf and one with 0.0. For each, form the 4 sliding windows of size 3 (windows[i] = padded[i:i+3]) and take the max along each window. Print exactly print([win_inf.max(axis=1).tolist(), win_zero.max(axis=1).tolist()]) to show how the 0-pad corrupts the edge maxima while -inf does not.

## Reference solution


```python
import numpy as np
x = np.array([-4.0, -1.0, -3.0, -2.0])
pad_inf = np.concatenate([[-np.inf], x, [-np.inf]])
pad_zero = np.concatenate([[0.0], x, [0.0]])
win_inf = np.stack([pad_inf[i:i+3] for i in range(len(x))])
win_zero = np.stack([pad_zero[i:i+3] for i in range(len(x))])
print([win_inf.max(axis=1).tolist(), win_zero.max(axis=1).tolist()])
```

## Why this works

Padding with -inf is the correct identity for a max reduction, so the padded entries are inert and the windowed maxima reflect only real data. Zero-padding an all-negative signal lets the 0 win at the first and last windows, corrupting the edge maxima.
