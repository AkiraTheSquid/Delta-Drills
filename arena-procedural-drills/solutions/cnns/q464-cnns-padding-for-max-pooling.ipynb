## Problem

Write a function solve(x) demonstrating why max-pooling pads with -inf, not 0: x is a 1-D float array (possibly all negative). Pad one element on EACH side twice — once with -inf, once with 0.0 — then max-pool both padded arrays with window 2, stride 2. Return the tuple (pooled_inf_pad, pooled_zero_pad); they differ whenever x is negative at the borders.


<details><summary>💡 Hint (click to reveal)</summary>

Pad both ways, pool both ways, and compare — the zero pad wins the max at negative borders and corrupts the result.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return (lambda pi, pz: (np.maximum(pi[0::2], pi[1::2]), np.maximum(pz[0::2], pz[1::2])))(np.concatenate([[-np.inf], x, [-np.inf]]), np.concatenate([[0.0], x, [0.0]]))


print(solve(np.array([-4.0, -1.0, -3.0, -2.0])))
```

## Why this works

Max-pooling's identity element is -inf: it never beats a real value. Padding with 0 injects a fake maximum wherever the data is negative — the all-negative cases show pooled values snapping to 0. That's why frameworks pad pooling with -inf (and convolution, which SUMS, pads with its identity 0).
