## Problem

PyTorch's MaxPool pads with −∞, not 0 — so padding can never fabricate a winning value. Write solve(x): build an `nn.MaxPool1d` with kernel 2, stride 2, padding 1, apply it to the (1, 1, N) input, and return the output tensor. On an all-negative input, look at the edge outputs: real negative values, not 0.


<details><summary>💡 Hint (click to reveal)</summary>

`nn.MaxPool1d(2, 2, padding=1)`; edge outputs are real values, never 0.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

def solve(x):
    pool = nn.MaxPool1d(kernel_size=2, stride=2, padding=1)
    return pool(x)

x = torch.tensor([[[-5.0, -2.0, -9.0, -1.0, -6.0, -8.0]]])
print(solve(x))
```

## Why this works

MaxPool's built-in padding uses −∞ internally — the padded cells can never beat a real value, even on an all-negative input. That's why pooling padding is safe by construction, unlike hand-rolled zero-padding before a max.
