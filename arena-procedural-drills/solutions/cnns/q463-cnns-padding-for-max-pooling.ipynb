## Problem

PyTorch's MaxPool pads with $-\infty$, not 0 — so padding can never fabricate a winning value. Build an `nn.MaxPool1d` with kernel 2, stride 2, padding 1, apply it to the starter's all-negative (1,1,6) input, and print exactly print(y[0].tolist()). Look at the edge outputs: real negative values, not 0.

## Reference solution


```python
import torch
import torch.nn as nn
x = torch.tensor([[[-5.0, -2.0, -9.0, -1.0, -6.0, -8.0]]])
pool = nn.MaxPool1d(kernel_size=2, stride=2, padding=1)
y = pool(x)
print(y[0].tolist())
```

## Why this works

nn.MaxPool1d pads its input with -inf rather than 0, so the synthetic boundary cells never beat the real (negative) edge values. With padding=1 the length-6 signal becomes length-8, and stride-2 kernel-2 pooling yields 4 outputs that faithfully include the true negative edges.
