## Problem

Using torch.nn, build a MaxPool1d with kernel_size=2, stride=2, padding=1 and apply it to the all-negative input x = [[[-5.0, -2.0, -9.0, -1.0, -6.0, -8.0]]] (shape [1,1,6]). PyTorch's MaxPool pads with -inf (not 0), so the padded boundary cells never win the max and the output edges reflect the true negative edge values. Print exactly print(y[0].tolist()) where y is the pooled output.

## Reference solution


```python
import torch
import torch.nn as nn
x = torch.tensor([[[-5.0, -2.0, -9.0, -1.0, -6.0, -8.0]]])
pool = nn.MaxPool1d(kernel_size=2, stride=2, padding=1)
y = pool(x)
print(y[0].tolist())
```

## Why this works

nn.MaxPool1d pads its input with -inf rather than 0, so the synthetic boundary cells never beat the real (negative) edge values. With padding=1 the length-6 signal becomes length-8, and stride-2 kernel-2 pooling yields 4 outputs that faithfully include the true negative edges.
