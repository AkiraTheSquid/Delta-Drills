## Problem

Write a function solve(v, m) that takes a BATCH of vectors v of shape (b, n) and a single matrix m of shape (n, n), and returns the length-b vector of QUADRATIC FORMS: entry k is v[k] @ m @ v[k]. (A companion drill does one vector; the batch letter is what's new.)


<details><summary>💡 Hint (click to reveal)</summary>

Both copies of v share the batch letter so each row pairs with itself; i and j contract through the matrix.

</details>

## Reference solution


```python
import torch as t

def solve(v, m):
    return t.einsum('bi,ij,bj->b', v, m, v)


print(solve(t.tensor([[1.0, 2.0]]), t.eye(2)))
```

## Why this works

einsum('bi,ij,bj->b', v, m, v) evaluates every row's quadratic form in one pass — the shared b stops rows from cross-multiplying while the matrix contracts both feature indices. One Mahalanobis-style score per vector.
