## Problem

Given a vector v of shape (2) and a matrix M of shape (2, 2), compute the result of the quadratic form involving v and M, which produces a single scalar output. Print the result using the provided inputs. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Feed all three operands into one einsum, chaining the index letters so both vectors meet the matrix, and reduce to a scalar.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

v = np.arange(1, 3)
M = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(v, M, v, 'i, i j, j ->')

print(solve())
```

## Why this works

einsum accepts more than two operands. Labeling v as i, M as i j, and v as j with an empty output contracts all indices simultaneously, computing the quadratic form v^T M v in a single scalar reduction.
