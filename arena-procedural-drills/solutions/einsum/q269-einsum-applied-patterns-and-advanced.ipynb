## Problem

Write a function solve(img, w) that takes a channels-first image of shape (c, h, w) and a per-channel weight vector of length c, and returns the (h, w) weighted combination of the channels — each channel map scaled by its weight and summed.

Contract the channel index between the image and the weight vector — the spatial axes pass through untouched.

## Reference solution


```python
import torch as t

def solve(img, w):
    return t.einsum('chw,c->hw', img, w)


print(solve(t.ones((3, 2, 2)), t.tensor([0.5, 0.25, 0.25])))
```

## Why this works

einsum('chw,c->hw', img, w) scales each channel map by its weight and sums across channels — grayscale conversion when the weights are the luminance coefficients. A one-hot weight simply selects a channel.
