## Problem

Given an image represented as a 3D array with shape (2, 2, 2) and a 1D array of weights, combine the channels of the image into a single 2D spatial map using the provided weights. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Name the channel axis in both the image and the weight vector, then leave it out of the output pattern so it gets contracted away.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

img = np.arange(1, 9).reshape((2, 2, 2))
w = np.arange(1, 3)

def solve():
    return einsum(img, w, 'c h w, c -> h w')

print(solve())
```

## Why this works

Each output spatial cell is a weighted sum over the channel axis c. einsum names c in both operands but drops it from the output, so it is summed; h and w pass through unchanged, yielding a single 2D map blended by the channel weights.
