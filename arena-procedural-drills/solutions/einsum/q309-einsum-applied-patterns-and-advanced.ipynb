## Problem

Given a multi-head tensor x with shape (b, t, h, d) and a head-importance weight w with shape (h), compute a new tensor of shape (b, t, d) that represents the weighted sum of the heads. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Name the head axis in both operands and simply omit it from the output pattern so einsum contracts (weights then sums) over it.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(1, 17).reshape((2, 2, 2, 2))
w = np.arange(1, 3)

def solve():
    return einsum(x, w, 'b t h d, h -> b t d')

print(solve())
```

## Why this works

The head axis h appears in both inputs but is absent from the output, so einsum multiplies each head's slice by its weight and sums over h. Batch, time, and feature axes (b, t, d) are carried through unchanged, yielding the weighted combination of the heads.
