## Problem

Multiply two vectors element-wise and sum the results into a single scalar.
Use numpy (imported as np). Use the exact inputs below and print the result.
v1 = np.arange(1, 3)
v2 = np.arange(1, 3)


<details><summary>💡 Hint (click to reveal)</summary>

Repeat the index on both operands and leave the output side empty to contract it to a scalar.

</details>

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    v1 = np.arange(1, 3)
    v2 = np.arange(1, 3)
    return einsum(v1, v2, 'i, i ->')

print(solve())
```

## Why this works

Using the same index i on both vectors and an empty output pattern multiplies matching elements and sums over i, producing the scalar dot product (1*1 + 2*2 = 5).
