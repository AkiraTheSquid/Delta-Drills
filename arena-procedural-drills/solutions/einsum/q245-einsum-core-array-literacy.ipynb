## Problem

Write a function solve(v1, v2) that takes two 1-D float arrays of the same length and returns their INNER (dot) product as a single scalar: multiply elementwise and sum everything. Express the contraction in einsum or any equivalent.


<details><summary>💡 Hint (click to reveal)</summary>

A repeated index that vanishes from the output means multiply-and-sum over it — with nothing after the arrow, the result is a scalar.

</details>

## Reference solution


```python
import numpy as np

def solve(v1, v2):
    return np.einsum('i,i->', v1, v2)


print(solve(np.array([1.0, 2.0]), np.array([3.0, 4.0])))
```

## Why this works

In 'i,i->' the shared index i is contracted: matching elements multiply, and its absence from the (empty) output sums them all — the definition of the dot product. np.dot(v1, v2) and (v1 * v2).sum() are the same contraction.
