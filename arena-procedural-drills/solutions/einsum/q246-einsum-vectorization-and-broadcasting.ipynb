## Problem

Given two 2x2 matrices A and B, calculate their product and return the elements that are located on the main diagonal. (★☆☆)

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    A = np.arange(1, 5).reshape((2, 2))
    B = np.arange(1, 5).reshape((2, 2))
    return einsum(A, B, 'i k, k i -> i')

print(solve())
```

## Why this works

The pattern 'i k, k i -> i' contracts the shared index k while keeping i, computing sum_k A[i,k]*B[k,i] which is exactly the i-th diagonal entry of the product AB. This skips forming the full product matrix.
