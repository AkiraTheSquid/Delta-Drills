## Problem

Write a function solve(a, b) that takes two square matrices of the same shape (n, n) and returns the DIAGONAL of their matrix product a @ b as a 1-D array of length n — computing only those n entries, not the full product. Entry i is the dot product of row i of a with COLUMN i of b.


<details><summary>💡 Hint (click to reveal)</summary>

Give b's indices in transposed order so its column pairs with a's row, contract over the shared k, and keep i.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.einsum('ik,ki->i', a, b)


print(solve(np.array([[1.0, 2.0], [3.0, 4.0]]), np.array([[5.0, 6.0], [7.0, 8.0]])))
```

## Why this works

'ik,ki->i' contracts k between a's rows and b's columns while the surviving i ties output entry i to row i and column i — the product's diagonal, computed in O(n^2) instead of the O(n^3) full multiply. Identity times any matrix returns that matrix's diagonal, as the third case shows.
