## Problem

Given a vector x and a 2D matrix W, compute a weighted sum of the matrix's rows based on the vector, and then take the dot product of that result with the vector again. Print the final result. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use all three operands in one einsum, sharing i with the matrix rows and j with the matrix columns, and leave the output side empty.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(1, 3)
W = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(x, W, x, 'i, i j, j ->')

print(solve())
```

## Why this works

This is the quadratic form x^T W x. einsum accepts more than two operands: index i ties x to W's rows and j ties W's columns to the second x. With no free indices in the output, every axis is summed, yielding a scalar.
