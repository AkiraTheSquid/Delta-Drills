## Problem

Write a function solve(x, w) that takes a 1-D vector x of length n and an (n, n) matrix w, and returns the QUADRATIC FORM x^T W x as a single scalar: weight each pair of components x[i] * x[j] by w[i, j] and sum everything.


<details><summary>💡 Hint (click to reveal)</summary>

Three operands in one einsum: x's index pairs with W's rows, x's second copy with W's columns, and everything contracts to a scalar.

</details>

## Reference solution


```python
import numpy as np

def solve(x, w):
    return np.einsum('i,ij,j->', x, w, x)


print(solve(np.array([1.0, 2.0]), np.eye(2)))
```

## Why this works

einsum('i,ij,j->', x, w, x) computes sum_ij x[i] * w[i,j] * x[j] — the quadratic form x @ w @ x. With the identity weight it reduces to the squared norm (first case).
