## Problem

Given a 2D array X of shape (2, 2) and its mean along the first axis, compute the variance and covariance by contracting the batch dimension of the mean-centered data. Use the provided inputs and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Mean-center first, then contract the sample axis n of the centered data with itself and divide by n-1.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

X = np.arange(1, 5).reshape((2, 2))
mu = X.mean(axis=0)
n = 2

def solve():
    return einsum(X-mu, X-mu, 'n d, n e -> d e') / (n-1)

print(solve())
```

## Why this works

Centering by the column means and then contracting the sample axis n of (X-mu) with itself gives the unnormalized scatter matrix of features d against e. Dividing by (n-1) produces the unbiased sample covariance matrix.
