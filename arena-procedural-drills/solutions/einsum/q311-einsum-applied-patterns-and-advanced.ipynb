## Problem

Multiply a 2D matrix A by a vector representing a diagonal matrix, followed by another 2D matrix B, and return the resulting matrix. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Treat the diagonal as a 1D operand sharing the contracted index j with both matrices; the output keeps only the outer indices i and k.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))
v = np.arange(1, 3)
B = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, v, B, 'i j, j, j k -> i k')

print(solve())
```

## Why this works

A diagonal matrix multiply scales column/row j by v[j], which einsum expresses by sharing the single index j across A's second axis, the vector v, and B's first axis. Contracting over j computes A @ diag(v) @ B in one pass without materializing the diagonal matrix.
