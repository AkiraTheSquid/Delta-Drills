## Problem

Given a batch of 2x2 matrices A and a batch of 2x2 vectors x, compute the result of applying each matrix to its corresponding vector, producing a new array that maintains the batch structure. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Match the batch axis b on both inputs, contract the matrix column axis j with the vector, and keep b and i.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 9).reshape((2, 2, 2))
x = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, x, 'b i j, b j -> b i')

print(solve())
```

## Why this works

Per-batch matrix-vector product: the batch axis b is matched on both operands so each matrix pairs with its own vector, the column axis j is contracted, and the row axis i remains as the output vector per batch item.
