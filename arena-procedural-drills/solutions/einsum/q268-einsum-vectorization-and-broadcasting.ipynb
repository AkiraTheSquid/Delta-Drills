## Problem

Write a function solve(a, x) that takes a batch of matrices a of shape (b, i, j) and a batch of vectors x of shape (b, j), and returns the (b, i) batch of matrix-vector products: item n is a[n] @ x[n].


<details><summary>💡 Hint (click to reveal)</summary>

Pair each matrix with ITS OWN vector by sharing the batch index; contract j as usual.

</details>

## Reference solution


```python
import numpy as np

def solve(a, x):
    return np.einsum('bij,bj->bi', a, x)


print(solve(np.stack([np.eye(2)]), np.array([[3.0, 4.0]])))
```

## Why this works

einsum('bij,bj->bi', a, x) applies matrix n to vector n — the shared b prevents cross-batch mixing while j contracts. The matmul equivalent needs an explicit (a @ x[..., None])[..., 0] to add and strip the column axis.
