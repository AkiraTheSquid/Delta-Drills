## Problem

Given Query and Key tensors Q and K with shapes (2, 2, 2, 2), compute the attention scores by comparing these tensors across the batch dimension and returning a new tensor with the appropriate shape. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Keep b and h as batch axes, contract the head dim d, and let the distinct query index q and key index k form the output matrix.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

Q = np.arange(1, 17).reshape((2, 2, 2, 2))
K = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(Q, K, 'b h q d, b h k d -> b h q k')

print(solve())
```

## Why this works

Raw attention scores: for each batch b and head h, every query vector (indexed q) is dotted with every key vector (indexed k) over the head dimension d. d is shared and summed; b and h are batch axes; q and k become the score-matrix axes.
