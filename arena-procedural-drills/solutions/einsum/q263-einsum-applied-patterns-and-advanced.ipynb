## Problem

Write a function solve(q, k) that takes Query and Key tensors of shapes (b, h, q, d) and (b, h, k, d), and returns the (b, h, q, k) ATTENTION SCORE tensor: for each batch and head, the dot product of every query vector with every key vector (no scaling).

Contract only the shared feature depth d, keeping the separate query and key position indices — that cross pairing is the score table.

## Reference solution


```python
import numpy as np

def solve(q, k):
    return np.einsum('bhqd,bhkd->bhqk', q, k)


print(solve(np.ones((1, 1, 1, 2)), np.ones((1, 1, 2, 2))))
```

## Why this works

einsum('bhqd,bhkd->bhqk', q, k) dots each query against each key within every (batch, head), producing the raw attention logits that softmax then normalizes in a transformer. An orthonormal key basis reads off the query's coordinates (third case).
