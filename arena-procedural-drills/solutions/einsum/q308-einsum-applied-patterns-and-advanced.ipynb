## Problem

Given a sequence x of shape (t, d), calculate the sum of the products of each token and the token that follows it in the sequence. Use the exact inputs below and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Slice the sequence into all-but-last and all-but-first, then contract the feature axis while keeping the position axis.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(1, 7).reshape((3, 2))

def solve():
    return einsum(x[:-1], x[1:], 't d, t d -> t')

print(solve())
```

## Why this works

Slicing x[:-1] and x[1:] aligns each token with its successor. Sharing axis t and contracting only d gives a per-position dot product between consecutive tokens, leaving one value per adjacent pair.
