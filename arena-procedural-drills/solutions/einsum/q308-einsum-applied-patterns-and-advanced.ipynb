## Problem

Write a function solve(x) that takes a sequence x of shape (t, d) with t >= 2 and returns the length-(t-1) array whose entry i is the DOT PRODUCT of token i with the NEXT token i+1 — adjacent-pair similarities along the sequence.


<details><summary>💡 Hint (click to reveal)</summary>

Slice the sequence against its shifted self first — the einsum then pairs row i with row i+1 automatically.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.einsum('td,td->t', x[:-1], x[1:])


print(solve(np.array([[1.0, 0.0], [2.0, 3.0], [0.0, 1.0]])))
```

## Why this works

x[:-1] and x[1:] align each token with its successor; einsum('td,td->t') dots the aligned pairs, measuring how similar consecutive tokens are. The slicing does the temporal shift; the contraction does the geometry.
