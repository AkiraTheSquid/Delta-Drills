## Problem

Write a function solve(t) that takes a 4-D tensor of shape (b, h, s, d) and returns it with the h and s axes SWAPPED — shape (b, s, h, d), values untouched (pure permutation).


<details><summary>💡 Hint (click to reveal)</summary>

Same letters both sides, two of them exchanged after the arrow — no contraction, just reordering.

</details>

## Reference solution


```python
import numpy as np

def solve(t):
    return np.einsum('bhsd->bshd', t)


print(solve(np.ones((1, 2, 3, 1))))
```

## Why this works

einsum('bhsd->bshd', t) exchanges the head and sequence axes — np.swapaxes(t, 1, 2) — the reshuffle used when regrouping attention heads back into token order.
