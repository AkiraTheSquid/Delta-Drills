## Problem

Given a tensor T with shape (batch, head, seq, dim), rearrange the axes so that the head and seq dimensions are swapped. Print the result using the provided input. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

T = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(T, 'b h s d -> b s h d')

print(solve())
```

## Why this works

A single operand with all axes retained is a permutation. Swapping the positions of h and s in the output pattern transposes those two axes while leaving b and d fixed.
