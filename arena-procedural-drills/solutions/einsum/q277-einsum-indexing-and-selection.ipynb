## Problem

Write a function solve(a) that takes a square 2-D array and returns its TRACE — the scalar sum of the main-diagonal entries — via the repeated-index contraction.

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('ii->', a)


print(solve(np.array([[1.0, 2.0], [3.0, 4.0]])))
```

## Why this works

einsum('ii->', a) contracts the repeated index over the diagonal — the trace. Keeping the index ('ii->i') would return the diagonal ENTRIES instead; the empty output is what turns extraction into summation.
