## Problem

Given a square 2D matrix A of shape (2, 2) with integer entries, return a single scalar equal to the sum of the entries lying on A's main diagonal (those positions where the row index equals the column index). The result is a scalar, not an array. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use one repeated label for both axes to pick the diagonal, then leave the output empty to sum it to a scalar.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

A = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, 'i i ->')

print(solve())
```

## Why this works

Repeating i on both axes selects the diagonal, and the empty output pattern sums those diagonal entries into a scalar. That sum is the matrix trace.
