## Problem

Given a 2x2 matrix X, compute the result of multiplying the matrix by its transpose to analyze the relationships between its columns. Print the resulting matrix. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

X = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(X, X, 'n d, n e -> d e')

print(solve())
```

## Why this works

Sharing the row axis n between the two copies of X and keeping the two column axes d and e computes X^T X: entry (d,e) is the dot product of column d with column e over the rows. The shared n is summed; d and e stay free.
