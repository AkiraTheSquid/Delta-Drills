## Problem

Write a function solve(x) that takes a 2-D array of shape (n, d) and returns the (d, d) matrix X^T X — the Gram matrix of x's COLUMNS, where entry [p, q] is the dot product of column p with column q.


<details><summary>💡 Hint (click to reveal)</summary>

Share the ROW index between the two copies of x and keep two different column indices — the rows contract away.

## Reference solution


```python
import numpy as np

def solve(x):
    return np.einsum('nd,ne->de', x, x)


print(solve(np.array([[1.0, 2.0], [3.0, 4.0]])))
```

## Why this works

einsum('nd,ne->de', x, x) contracts the sample axis n between two views of the same matrix, leaving all column-pair dot products — X^T X, the symmetric matrix at the heart of least squares and covariance computations.
