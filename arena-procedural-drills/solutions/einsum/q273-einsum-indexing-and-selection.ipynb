## Problem

Write a function solve(a) that takes a batch of square matrices of shape (b, n, n) and returns the (b, n) tensor whose row i is the DIAGONAL of matrix i — the batched diagonal extraction.

## Reference solution


```python
import torch as t

def solve(a):
    return t.einsum('bii->bi', a)


print(solve(t.stack([t.diag(t.tensor([1.0, 2.0])), t.diag(t.tensor([3.0, 4.0]))])))
```
