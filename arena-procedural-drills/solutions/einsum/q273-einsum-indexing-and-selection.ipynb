## Problem

For a batch of matrices A, return a 2D array where each row contains the diagonal elements of the corresponding matrix. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Reuse the same letter for both inner matrix axes to select the diagonal, and keep the batch axis on the output.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

A = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(A, 'b i i -> b i')

print(solve())
```

## Why this works

Repeating the index i on both matrix axes forces einsum to read only positions where row equals column, i.e. the diagonal. Keeping b and i in the output produces one diagonal vector per batch element.
