## Problem

Write a function solve(a) that takes a batch of square matrices of shape (b, n, n) and returns the (b, n) array whose row i is the DIAGONAL of matrix i — the batched diagonal extraction.


<details><summary>💡 Hint (click to reveal)</summary>

Repeat the matrix index to follow each diagonal, but KEEP it in the output so the entries survive per position.

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('bii->bi', a)


print(solve(np.stack([np.diag([1.0, 2.0]), np.diag([3.0, 4.0])])))
```

## Why this works

einsum('bii->bi', a) walks each item's diagonal and, unlike the trace ('bii->b'), retains the i index so the diagonal VALUES come back per batch item — np.diagonal(a, axis1=1, axis2=2).
