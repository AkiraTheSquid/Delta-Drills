## Problem

Write a function solve(b_stack, m) that takes a batch of matrices of shape (b, i, j) and ONE constant matrix m of shape (j, k), and returns the (b, i, k) batch multiplying every item by the same m.


<details><summary>💡 Hint (click to reveal)</summary>

m carries no batch letter, so the same matrix hits every item — only j contracts.

</details>

## Reference solution


```python
import numpy as np

def solve(b_stack, m):
    return np.einsum('bij,jk->bik', b_stack, m)


print(solve(np.ones((2, 1, 2)), np.ones((2, 3))))
```

## Why this works

einsum('bij,jk->bik', b_stack, m) applies one shared linear map across the whole batch (broadcast matmul b_stack @ m). Contrast the batched-pair version 'bij,bjk->bik', where each item gets its own matrix.
