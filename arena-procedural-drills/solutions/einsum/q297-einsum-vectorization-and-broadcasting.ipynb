## Problem

Given a batch of matrices B with shape (b, i, j) and a constant matrix M with shape (j, k), compute the result of multiplying each matrix in the batch by the constant matrix. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Share the inner axis j between B and M, keep the batch axis, and let the output expose i and k.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

B = np.arange(1, 9).reshape((2, 2, 2))
M = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(B, M, 'b i j, j k -> b i k')

print(solve())
```

## Why this works

The contraction axis j is shared between the batch of matrices and the constant matrix and is dropped from the output, performing a matrix multiply per batch item; the batch axis b is carried through untouched.
