## Problem

In a multi-layer perceptron, compute the product of the hidden states and the output weights, resulting in an array that combines the last dimension of the hidden states with the second dimension of the weights. Use the provided inputs and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Share the hidden axis between h and W, keep batch and time, and let the weight's second axis become the new last dimension.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

h = np.arange(1, 9).reshape((2, 2, 2))
W = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(h, W, 'b t h, h d -> b t d')

print(solve())
```

## Why this works

The shared hidden axis h is contracted between the hidden states and the output weights, performing a batched matmul that replaces the hidden dimension with the weight's output dimension d, keeping b and t.
