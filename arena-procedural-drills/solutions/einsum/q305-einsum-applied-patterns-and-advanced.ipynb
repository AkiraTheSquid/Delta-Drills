## Problem

Write a function solve(h, w) that takes hidden states h of shape (b, t, hd) and an output weight matrix w of shape (hd, d), and returns the (b, t, d) tensor projecting every position's hidden vector through w — an MLP output layer over a sequence.

## Reference solution


```python
import torch as t

def solve(h, w):
    return t.einsum('bth,hd->btd', h, w)


print(solve(t.ones((1, 1, 2)), t.eye(2)))
```

## Why this works

einsum('bth,hd->btd', h, w) applies the output projection independently at every sequence position — h @ w by trailing-axis broadcast, the last linear layer of a per-token MLP.
