## Problem

Given two 3D arrays A and B, each with shape (2, 2, 2), compute the product of the last two dimensions for each item in the first dimension, resulting in a new array with shape (2, 2, 2). Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Keep the batch label on both inputs and the output, and share the inner contraction axis between the two operands.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

A = np.arange(1, 9).reshape((2, 2, 2))
B = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(A, B, 'b i k, b k j -> b i j')

print(solve())
```

## Why this works

The shared inner axis k is contracted while the shared batch axis b is broadcast (kept), giving an independent i-by-k times k-by-j matrix product for each batch element. This is batched matrix multiplication.
