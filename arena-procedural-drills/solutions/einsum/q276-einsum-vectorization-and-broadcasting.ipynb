## Problem

Write a function solve(a, b) that takes two 3-D tensors a of shape (b, i, k) and b of shape (b, k, j), and returns the (b, i, j) batched matrix product — contracting the shared inner axis independently for each leading batch index.


<details><summary>💡 Hint (click to reveal)</summary>

Same as plain matrix multiplication with a batch letter carried through on everything.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.einsum('bik,bkj->bij', a, b)


print(solve(t.ones((2, 1, 2)), t.ones((2, 2, 1))))
```

## Why this works

einsum('bik,bkj->bij', a, b) contracts k within each batch item while b passes through — the 3-D stacked matmul that the @ operator also performs on 3-D tensors.
