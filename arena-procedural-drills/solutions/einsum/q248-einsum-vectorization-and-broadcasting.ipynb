## Problem

Given two matrices X and Y, compute the result of multiplying each row of X with each row of Y, producing a new matrix where each entry corresponds to the product of the respective rows. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    X = np.arange(1, 5).reshape((2, 2))
    Y = np.arange(1, 5).reshape((2, 2))
    return einsum(X, Y, 'n d, m d -> n m')

print(solve())
```

## Why this works

Sharing the feature index d but giving X and Y distinct row indices n and m contracts d to form the dot product of every X-row with every Y-row, producing an (n, m) matrix of pairwise row dot products.
