## Problem

Write a function solve(x, y) that takes two 2-D float arrays x of shape (n, d) and y of shape (m, d), and returns the (n, m) matrix of ALL pairwise row dot products: entry [i, j] is the dot product of row i of x with row j of y. This is the Gram-style product x @ y.T, written as one contraction.


<details><summary>💡 Hint (click to reveal)</summary>

Share the feature index d between the operands and keep both row indices — the contraction over d does the dots.

</details>

## Reference solution


```python
import numpy as np

def solve(x, y):
    return np.einsum('nd,md->nm', x, y)


print(solve(np.array([[1.0, 0.0]]), np.array([[2.0, 3.0], [0.0, 5.0]])))
```

## Why this works

'nd,md->nm' contracts the shared feature axis d while preserving both row axes, filling the whole cross table of dot products — identical to x @ y.T, where the transpose exists purely to line d up for matmul.
