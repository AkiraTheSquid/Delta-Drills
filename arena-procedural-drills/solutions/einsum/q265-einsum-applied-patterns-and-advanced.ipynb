## Problem

Given two batches of 2x2 matrices T and U, compute the product of the inner matrices for each batch item and return the resulting batch of matrices. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Carry the batch axis a unchanged on both inputs and contract only the inner axis j.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

T = np.arange(1, 9).reshape((2, 2, 2))
U = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(T, U, 'a i j, a j k -> a i k')

print(solve())
```

## Why this works

Batched matrix multiplication: the batch axis a rides along on both operands, while the shared inner axis j is contracted, leaving i and k as the per-item product's row and column axes.
