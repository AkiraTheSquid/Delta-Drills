## Problem

Write a function solve(a, u) that takes two batches of matrices — a of shape (a, i, j) and u of shape (a, j, k) — and returns the (a, i, k) batch where each item is the matrix product a[n] @ u[n].


<details><summary>💡 Hint (click to reveal)</summary>

The batch index rides along on both operands and the output; only the inner j contracts.

</details>

## Reference solution


```python
import torch as t

def solve(a, u):
    return t.einsum('aij,ajk->aik', a, u)


print(solve(t.ones((1, 2, 2)), t.ones((1, 2, 2))))
```

## Why this works

einsum('aij,ajk->aik', t, u) performs an independent matrix multiply per batch item — the batched matmul that t @ u also computes for 3-D stacks, spelled out in indices.
