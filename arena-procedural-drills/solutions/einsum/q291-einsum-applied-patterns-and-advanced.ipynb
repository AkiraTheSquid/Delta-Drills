## Problem

Write a function solve(img, s) that takes a channels-first image of shape (c, h, w) and a per-channel scale vector of length c, and returns the image with each channel MULTIPLIED by its scale — same shape (c, h, w), no summation at all.


<details><summary>💡 Hint (click to reveal)</summary>

Keep the channel index in the OUTPUT — a repeated index that survives multiplies without summing.

</details>

## Reference solution


```python
import numpy as np

def solve(img, s):
    return np.einsum('chw,c->chw', img, s)


print(solve(np.ones((2, 1, 2)), np.array([2.0, 0.5])))
```

## Why this works

einsum('chw,c->chw', img, s) scales each channel by its factor: c appears in both inputs AND the output, so it aligns them without contracting — broadcasting img * s[:, None, None] in index form. Contrast 'chw,c->hw', which would also sum the channels.
