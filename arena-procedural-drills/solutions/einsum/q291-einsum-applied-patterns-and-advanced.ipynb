## Problem

Given an image represented as a 3D array with shape (c, h, w) and a 1D array of scale factors with shape (c), return a new array where each channel of the image is multiplied by its corresponding scale factor. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Pair the per-channel scalar's single axis with the image's channel axis and keep that axis in the output instead of reducing it.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

img = np.arange(1, 9).reshape((2, 2, 2))
s = np.arange(1, 3)

def solve():
    return einsum(img, s, 'c h w, c -> c h w')

print(solve())
```

## Why this works

When an axis label appears in two inputs and also in the output, einsum multiplies aligned elements without summing it away. The channel axis 'c' is shared and kept, so each channel plane is scaled by its matching factor.
