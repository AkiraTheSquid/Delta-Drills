## Problem

Given input x with shape (b, t, d) and a weight tensor W with shape (3, d, d), project x into three different spaces simultaneously, resulting in an output with shape (3, b, t, d). Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Contract the shared feature axis and let the leading weight axis n broadcast, placing it first on the output.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

x = np.arange(1, 9).reshape((2, 2, 2))
W = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(x, W, 'b t d, n d e -> n b t e')

print(solve())
```

## Why this works

The feature axis d is contracted while the new projection-bank axis n is broadcast across the batch, applying each of the n weight matrices to x independently. The output stacks the n projections, giving shape (n, b, t, e).
