## Problem

Write a function solve(x, w) that takes input x of shape (b, t, d) and a stacked weight tensor w of shape (n, d, e) holding n projection matrices, and returns the (n, b, t, e) tensor applying ALL n projections to x simultaneously — projection k in output slot k.


<details><summary>💡 Hint (click to reveal)</summary>

The projection-stack index n never contracts — it labels which output slice used which matrix, while d contracts as usual.

</details>

## Reference solution


```python
import numpy as np

def solve(x, w):
    return np.einsum('btd,nde->nbte', x, w)


print(solve(np.ones((1, 1, 2)), np.stack([np.eye(2), 2 * np.eye(2)])))
```

## Why this works

einsum('btd,nde->nbte', x, w) contracts the feature axis against each of the n stacked matrices, emitting all projections in one call — exactly how Q, K, and V are computed together from one input with a (3, d, e) weight stack.
