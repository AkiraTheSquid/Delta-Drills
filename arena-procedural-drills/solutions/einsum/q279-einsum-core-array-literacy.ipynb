## Problem

Given a batch of vectors x with shape (2, 2) and a transformation matrix W with shape (2, 2), compute the result of applying the transformation matrix to each vector in the batch. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

x = np.arange(1, 5).reshape((2, 2))
W = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(x, W, 'b d, d e -> b e')

print(solve())
```

## Why this works

The feature axis d is shared and contracted, transforming each batched vector by W into the new space e while the batch axis b passes through. This is exactly x @ W.
