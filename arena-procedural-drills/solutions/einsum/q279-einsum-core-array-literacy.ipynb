## Problem

Write a function solve(x, w) that takes a batch of vectors x of shape (b, d) and a transformation matrix w of shape (d, e), and returns the (b, e) batch of transformed vectors — each row of x multiplied through w (a linear layer without bias).

Contract the shared feature axis d; the batch axis b and output axis e both survive.

## Reference solution


```python
import torch as t

def solve(x, w):
    return t.einsum('bd,de->be', x, w)


print(solve(t.tensor([[1.0, 2.0]]), t.eye(2)))
```

## Why this works

einsum('bd,de->be', x, w) pushes every row of x through the matrix — x @ w, the forward pass of a dense layer. Index notation makes explicit which axis is consumed (d) and which is created (e).
