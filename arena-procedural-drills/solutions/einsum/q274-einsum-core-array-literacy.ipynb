## Problem

Write a function solve(a) that takes a 2-D tensor and returns a 1-D tensor of its ROW SUMS: entry i is the total of row i (the column axis is collapsed). (A companion drill sums columns — mind which index survives.)

## Reference solution


```python
import torch as t

def solve(a):
    return t.einsum('ij->i', a)


print(solve(t.tensor([[1, 2], [3, 4]])))
```

## Why this works

einsum('ij->i', a) sums over j (absent from the output), collapsing each row to its total — a.sum(axis=1). The single-column case distinguishes it from the column-sum sibling 'ij->j'.
