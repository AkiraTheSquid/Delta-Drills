## Problem

Given a 2D array A, return a 1D array where each element is the sum of the elements in each column of A. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Keep only one axis label on the output side; the omitted axis is the one summed over.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

A = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, 'i j -> i')

print(solve())
```

## Why this works

Dropping axis j from the output contracts it, summing across that axis for each fixed i. The result is a 1D vector of per-row sums.
