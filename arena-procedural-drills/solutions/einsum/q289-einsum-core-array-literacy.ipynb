## Problem

Given a 3D tensor X of shape (b, t, d), compute the total for each combination of t and d by aggregating values from all batches. Use the exact inputs below and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Name all three input axes but omit the batch axis from the output side; einsum reduces over any axis you don't keep.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

X = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(X, 'b t d -> t d')

print(solve())
```

## Why this works

Dropping an axis name from the right side of an einsum pattern sums (reduces) over it. Here 'b' is absent from the output, so values are accumulated across the batch dimension, leaving a (t, d) grid of totals.
