## Problem

Write a function solve(x) that takes a 3-D tensor of shape (b, t, d) and returns the (t, d) tensor summing over the BATCH axis: each (t, d) cell aggregates that position's values across all batch items.

Drop the leading letter — the batch is the axis that vanishes.

## Reference solution


```python
import torch as t

def solve(x):
    return t.einsum('btd->td', x)


print(solve(t.ones((3, 1, 2))))
```

## Why this works

einsum('btd->td', x) sums the batch axis away, leaving position-wise totals — x.sum(axis=0), e.g. accumulating gradients across a batch.
