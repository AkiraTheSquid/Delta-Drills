## Problem

Given two vectors u and v of shapes (m) and (n) respectively, compute their outer product and add the scalar A (broadcasts across the resulting matrix). Use the inputs provided below and print the final result.
u = np.arange(1, 3)
v = np.arange(1, 3)
A = 2

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

u = np.arange(1, 3)
v = np.arange(1, 3)
A = 2

def solve():
    return einsum(u, v, 'i, j -> i j') + A

print(solve())
```

## Why this works

Giving each vector a distinct output axis with no shared/reduced index produces the outer product u_i * v_j as an (m, n) matrix. The scalar A then broadcasts uniformly across that matrix.
