## Problem

Write a function solve(u, v, a) that takes vectors u (length m) and v (length n) and a scalar a, and returns the (m, n) matrix whose entry [i, j] is u[i] * v[j] + a — the outer product shifted by the scalar.


<details><summary>💡 Hint (click to reveal)</summary>

The outer product is the einsum part; the scalar rides in afterwards by ordinary broadcasting.

</details>

## Reference solution


```python
import numpy as np

def solve(u, v, a):
    return np.einsum('i,j->ij', u, v) + a


print(solve(np.array([1.0, 2.0]), np.array([10.0]), 5.0))
```

## Why this works

einsum('i,j->ij', u, v) builds the rank-1 matrix, and + a broadcasts the scalar over every entry — einsum handles multiplicative structure, addition stays outside it.
