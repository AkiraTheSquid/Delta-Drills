## Problem

Write a function solve(x, w1, w2) that takes input x of shape (b, i) and weight matrices w1 of shape (i, h) and w2 of shape (h, o), and returns the (b, o) result of pushing x through BOTH linear maps in one contraction (no intermediate activation).


<details><summary>💡 Hint (click to reveal)</summary>

Chain the indices b-i-h-o with i and h each shared by two operands — both contract in one call.

</details>

## Reference solution


```python
import numpy as np

def solve(x, w1, w2):
    return np.einsum('bi,ih,ho->bo', x, w1, w2)


print(solve(np.array([[1.0, 0.0]]), np.eye(2), np.eye(2)))
```

## Why this works

einsum('bi,ih,ho->bo', x, w1, w2) collapses the two-layer linear stack into one contraction equal to x @ w1 @ w2 — valid precisely because there's no nonlinearity between the layers, which is why depth without activations adds nothing.
