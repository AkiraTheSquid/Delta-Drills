## Problem

Given input x with shape (b, i), weight matrices W1 with shape (i, h) and W2 with shape (h, o), compute the output of applying both weight matrices to the input in a single operation. (★★☆)


<details><summary>💡 Hint (click to reveal)</summary>

Pass all three tensors and chain the contraction labels so each weight's output axis feeds the next weight's input axis.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

x = np.arange(1, 5).reshape((2, 2))
W1 = np.arange(1, 5).reshape((2, 2))
W2 = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(x, W1, W2, 'b i, i h, h o -> b o')

print(solve())
```

## Why this works

einsum accepts more than two operands; the chained shared axes i and h are both contracted in one expression, computing (x @ W1) @ W2 directly. Only the batch b and final output o survive.
