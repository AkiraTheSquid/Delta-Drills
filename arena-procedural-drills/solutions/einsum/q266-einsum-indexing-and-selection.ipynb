## Problem

Write a function solve(a) that takes a batch of square matrices of shape (b, n, n) and returns a 1-D array of length b holding each matrix's TRACE (the sum of its diagonal entries).


<details><summary>💡 Hint (click to reveal)</summary>

Repeat the matrix index within one operand to walk each diagonal, and drop it from the output to sum — the batch index stays.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('bii->b', a)


print(solve(np.stack([np.eye(2), 3 * np.eye(2)])))
```

## Why this works

einsum('bii->b', a) reads each item's diagonal (repeated i) and sums it (i absent from output), leaving one trace per batch element — np.trace with the matrix axes specified.
