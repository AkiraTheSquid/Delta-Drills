## Problem

For a batch of square matrices A, return a 1D array containing the sum of the diagonal elements for each matrix. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(A, 'b i i -> b')

print(solve())
```

## Why this works

Repeating index i on the last two axes selects each matrix's diagonal; dropping i from the output sums those diagonal entries, giving the trace per batch item b.
