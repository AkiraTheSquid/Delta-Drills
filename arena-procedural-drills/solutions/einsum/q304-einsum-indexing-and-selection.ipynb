## Problem

Given a 3D array x with shape (b, t, d) and a 1D weight vector w with shape (d), compute a new array with shape (b, t) by applying the weights along the last dimension of x. Use the exact inputs below and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Match the weight vector's single axis to x's last axis and omit it from the output so it gets contracted.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(1, 9).reshape((2, 2, 2))
w = np.arange(1, 3)

def solve():
    return einsum(x, w, 'b t d, d -> b t')

print(solve())
```

## Why this works

The feature axis d is shared between the tensor and the weight vector and is dropped from the output, so each (b, t) entry becomes a weighted sum over the last dimension of x.
