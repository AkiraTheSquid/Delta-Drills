## Problem

Write a function solve(x, w) that takes a tensor x of shape (b, t, d) and a weight vector w of length t — one weight per TIME position — and returns the (b, d) weighted sum over the time axis: each token's features scaled by its position's weight, then summed across t. (A companion drill weights the FEATURE axis; here the contraction letter is t.)


<details><summary>💡 Hint (click to reveal)</summary>

Match the weight vector's single letter to the TIME axis, not the feature axis — t contracts, d survives.

</details>

## Reference solution


```python
import numpy as np

def solve(x, w):
    return np.einsum('btd,t->bd', x, w)


print(solve(np.ones((1, 2, 2)), np.array([1.0, 3.0])))
```

## Why this works

einsum('btd,t->bd', x, w) pools the sequence with per-position weights — attention pooling with fixed weights. Compare 'btd,d->bt', where the weights act on features instead; which letter the vector shares decides everything.
