## Problem

Given three 2x2 matrices A, B, and C, perform a series of matrix multiplications on them in a single operation and return the resulting matrix. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use all three operands and let consecutive matrices share an inner index (k then m); keep only the first and last free indices.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))
B = np.arange(1, 5).reshape((2, 2))
C = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, B, C, 'i k, k m, m p -> i p')

print(solve())
```

## Why this works

einsum chains all three matrices in one call by sharing intermediate axes: k links A to B and m links B to C. Both shared axes are summed, leaving the outermost i and p as the (A@B@C) result.
