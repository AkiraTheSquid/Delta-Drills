## Problem

Write a function solve(a, b, c) that takes three matrices with chain-compatible shapes (i, k), (k, m), (m, p) and returns the triple MATRIX PRODUCT a @ b @ c as one einsum contraction.


<details><summary>💡 Hint (click to reveal)</summary>

Chain the letters so each inner index appears in exactly two neighboring operands — k and m both contract, i and p survive.

</details>

## Reference solution


```python
import torch as t

def solve(a, b, c):
    return t.einsum('ik,km,mp->ip', a, b, c)


print(solve(t.eye(2), t.tensor([[1.0, 2.0], [3.0, 4.0]]), t.eye(2)))
```

## Why this works

einsum('ik,km,mp->ip', a, b, c) expresses the double contraction of a matrix chain in one call, letting einsum choose the multiplication order. The result equals a @ b @ c.
