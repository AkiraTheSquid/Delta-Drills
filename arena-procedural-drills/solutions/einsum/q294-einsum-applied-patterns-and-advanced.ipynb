## Problem

Write a function solve(w, x) that takes a weight tensor w of shape (o, i, h, wd) and an input tensor x of shape (i, h, wd), and returns the length-o vector contracting x against each output filter — every axis of x pairs with the matching axis of w.


<details><summary>💡 Hint (click to reveal)</summary>

Every axis of x is shared with w and contracts; only the output-filter axis o survives.

</details>

## Reference solution


```python
import torch as t

def solve(w, x):
    return t.einsum('oihw,ihw->o', w, x)


print(solve(t.ones((2, 1, 2, 2)), t.ones((1, 2, 2))))
```

## Why this works

einsum('oihw,ihw->o', w, x) computes one full inner product between x and each of the o filters — a convolution layer's computation at a single spatial position, one response per output channel.
