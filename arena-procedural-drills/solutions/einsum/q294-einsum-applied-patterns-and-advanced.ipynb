## Problem

Given a weight tensor W of shape (2, 2, 2, 2) and an input tensor X of shape (2, 2, 2), compute a resulting vector by combining these tensors such that the output has shape (2). (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

W = np.arange(1, 17).reshape((2, 2, 2, 2))
X = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(W, X, 'o i h w, i h w -> o')

print(solve())
```

## Why this works

The three shared axes i, h, w appear in both operands but not in the output, so they are contracted (multiplied and summed) together, while the output axis 'o' is preserved, yielding a length-o vector.
