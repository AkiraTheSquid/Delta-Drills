## Problem

Write a function solve(a) that takes a 2-D numpy array and returns a 1-D array of its COLUMN SUMS: entry j is the total of column j (the row axis is collapsed).


<details><summary>💡 Hint (click to reveal)</summary>

Dropping an index from einsum's output side sums over it — drop i to collapse the rows.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('ij->j', a)


print(solve(np.array([[1, 2], [3, 4]])))
```

## Why this works

'ij->j' keeps the column index and omits the row index, so einsum sums over i — precisely a.sum(axis=0). The single-column case pins which axis collapsed.
