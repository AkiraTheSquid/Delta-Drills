## Problem

You are given A, a 2D array of shape (2, 2) with integer entries arranged row by row. Return a 1D array of length 2 whose j-th entry equals the sum of the j-th column of A, i.e. collapse the first axis by addition while keeping the second axis intact. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Omit the row index in the output pattern so it gets summed away, leaving only the column axis.

</details>

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    A = np.arange(1, 5).reshape((2, 2))
    return einsum(A, 'i j -> j')

print(solve())
```

## Why this works

Dropping the row index i from the output sums over that axis, collapsing rows while preserving the column axis j. The result's j-th entry is the sum of column j.
