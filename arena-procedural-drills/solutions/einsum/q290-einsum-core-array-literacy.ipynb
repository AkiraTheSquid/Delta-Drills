## Problem

Given a tensor X of any dimensionality, return a single scalar that represents the sum of all its elements. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Name each axis on the input and leave the output empty so every axis is summed away into one number.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

X = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(X, 'i j ->')

print(solve())
```

## Why this works

Listing every axis on the input and leaving the output side empty reduces all of them, summing the whole array into a single scalar. (einops.einsum cannot collapse an ellipsis directly to a scalar, so each axis is named explicitly.)
