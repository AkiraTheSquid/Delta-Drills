## Problem

Write a function solve(x) that takes a numpy array of ANY rank and returns its MEAN over all elements as a scalar — using a rank-generic einsum contraction (sum every axis), then dividing by the element count.


<details><summary>💡 Hint (click to reveal)</summary>

Sum all axes with a rank-generic contraction (the interleaved einsum form), then one division makes it a mean.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    total = np.einsum(x, list(range(x.ndim)), [])
    return total / x.size


print(solve(np.arange(6.0).reshape(2, 3)))
```

## Why this works

np.einsum(x, list(range(x.ndim)), []) uses the interleaved form to contract every axis whatever the rank, giving x.sum(); dividing by x.size makes it the mean. (The string form needs the ellipsis on both sides — '...->...' keeps axes, so the interleaved integer form is the clean way to sum them all.)
