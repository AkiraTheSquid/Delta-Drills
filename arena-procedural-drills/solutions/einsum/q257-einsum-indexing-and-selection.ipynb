## Problem

Given a 2x2 array A, return a 1D array containing the elements where the row and column indices match. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, 'i i -> i')

print(solve())
```

## Why this works

Repeating the same index i on both axes of A selects only the entries where row equals column, i.e. the diagonal. Keeping i in the output (rather than dropping it) returns those diagonal elements as a vector instead of summing them.
