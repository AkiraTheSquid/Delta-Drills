## Problem

Write a function solve(a) that takes a square 2-D tensor and returns its main DIAGONAL as a 1-D tensor — the entries a[i, i] — extracted via index notation (a repeated input index kept in the output).


<details><summary>💡 Hint (click to reveal)</summary>

Using the SAME letter for both axes walks the diagonal; keeping it in the output returns the entries instead of summing them.

</details>

## Reference solution


```python
import torch as t

def solve(a):
    return t.einsum('ii->i', a)


print(solve(t.tensor([[1, 2], [3, 4]])))
```

## Why this works

einsum('ii->i', a) reads positions where row equals column — the diagonal — and, because i remains in the output, returns those entries rather than reducing them (contrast 'ii->', the trace). Same as t.diag(a) on a matrix.
