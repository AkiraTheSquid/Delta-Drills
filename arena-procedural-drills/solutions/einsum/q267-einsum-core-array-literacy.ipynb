## Problem

Given two 2x2 matrices A and B, flatten both matrices and compute the sum of the products of their corresponding elements. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Match both axes between A and B and output nothing so all elementwise products are summed.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))
B = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, B, 'i j, i j ->')

print(solve())
```

## Why this works

Naming both axes identically (i, j) on the two operands pairs them elementwise, and an empty output sums every product, i.e. the Frobenius inner product of the flattened matrices.
