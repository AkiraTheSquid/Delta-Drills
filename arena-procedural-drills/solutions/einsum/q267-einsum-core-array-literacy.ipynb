## Problem

Write a function solve(a, b) that takes two tensors of identical 2-D shape and returns the SCALAR sum of the products of corresponding elements — the Frobenius inner product (flatten both, dot them).

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.einsum('ij,ij->', a, b)


print(solve(t.tensor([[1.0, 2.0], [3.0, 4.0]]), t.ones((2, 2))))
```

## Why this works

einsum('ij,ij->', a, b) multiplies matching entries and contracts both axes — the Frobenius inner product, equal to (a * b).sum() and to the trace of a.T @ b (identity against a matrix picks out its trace, the last case).
