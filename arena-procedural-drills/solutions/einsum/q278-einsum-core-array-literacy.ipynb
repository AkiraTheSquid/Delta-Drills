## Problem

Write a function solve(v) that takes a single 1-D vector of length n and returns the (n, n) symmetric matrix of ALL pairwise products of its entries: entry [i, j] is v[i] * v[j] — the vector's outer product with itself.

## Reference solution


```python
import numpy as np

def solve(v):
    return np.einsum('i,j->ij', v, v)


print(solve(np.array([1.0, 2.0, 3.0])))
```

## Why this works

einsum('i,j->ij', v, v) forms v's self outer product: rank-1, symmetric, with the squares on the diagonal. It's the building block of covariance updates and Hessian approximations — np.outer(v, v) in one index expression.
