## Problem

Given a vector v of size (2), construct a square matrix where each element at position (i, j) is the product of the i-th and j-th elements of v. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Pass the vector twice with different index labels and keep both labels in the output to build the outer product.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

v = np.arange(1, 3)

def solve():
    return einsum(v, v, 'i, j -> i j')

print(solve())
```

## Why this works

Giving the two copies of v distinct labels i and j and keeping both on the output forms an outer product, so entry (i, j) is v[i] * v[j].
