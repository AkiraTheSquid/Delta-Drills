## Problem

Write a function solve(x) that takes a batch of vectors of shape (b, d) and returns the (b, d, d) batch of SELF outer products: item n is the matrix x[n, i] * x[n, j] of all pairwise feature interactions within vector n.


<details><summary>💡 Hint (click to reveal)</summary>

Feed the batch in twice with two different feature letters — the shared b keeps each vector paired with itself.

</details>

## Reference solution


```python
import torch as t

def solve(x):
    return t.einsum('bd,be->bde', x, x)


print(solve(t.tensor([[1.0, 2.0]])))
```

## Why this works

einsum('bd,be->bde', x, x) forms each row's outer product with itself: symmetric (b, d, d) interaction matrices with the squared features on the diagonal — the per-sample building block of covariance and Fisher-information estimates.
