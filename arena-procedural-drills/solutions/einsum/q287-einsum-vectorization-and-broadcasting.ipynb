## Problem

Given a batch of vectors x with shape (b, d), return a batch of matrices of shape (b, d, d) that represent the interaction between each pair of features for every vector. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Pass x twice with different feature labels but the same batch label, and keep all three on the output.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

x = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(x, x, 'b d, b e -> b d e')

print(solve())
```

## Why this works

Reusing x with two distinct feature labels d and e while keeping the shared batch axis b forms a per-sample outer product. Entry (b, d, e) is x[b,d] * x[b,e], a batch of feature-interaction matrices.
