## Problem

Given two tensors P and Q, both with shape (b, h, n, d), calculate the similarity between the last two dimensions for each combination of the first two dimensions, resulting in a matrix of shape (n, n) for each batch and head. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Keep b and h aligned, give the two sequence axes distinct labels (n and m), and contract the shared last dimension d.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

P = np.arange(1, 17).reshape((2, 2, 2, 2))
Q = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(P, Q, 'b h n d, b h m d -> b h n m')

print(solve())
```

## Why this works

With batch and head axes (b, h) shared and carried through, and the feature axis d contracted, einsum computes an (n, m) similarity matrix independently for each batch-head pair, exactly like multi-head attention scores.
