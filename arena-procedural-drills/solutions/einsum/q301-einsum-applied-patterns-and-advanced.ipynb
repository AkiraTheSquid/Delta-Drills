## Problem

Write a function solve(p, q) that takes two tensors of shape (b, h, n, d) and (b, h, m, d), and returns the (b, h, n, m) similarity tensor: within each batch item and head, the dot product of every vector in p with every vector in q.

## Reference solution


```python
import torch as t

def solve(p, q):
    return t.einsum('bhnd,bhmd->bhnm', p, q)


print(solve(t.ones((1, 1, 1, 2)), t.ones((1, 1, 3, 2))))
```

## Why this works

einsum('bhnd,bhmd->bhnm', p, q) computes each (batch, head)'s full n x m table of vector dot products — the general similarity tensor behind attention scores and contrastive matching.
