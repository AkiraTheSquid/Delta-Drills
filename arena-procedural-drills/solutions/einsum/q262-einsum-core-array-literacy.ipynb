## Problem

Given a 2x2 matrix A and a 1D vector b, compute the resulting vector by performing a multiplication that combines the rows of the matrix with the elements of the vector. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Share the matrix's column axis j with the vector and keep only the row axis i.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))
b = np.arange(1, 3)

def solve():
    return einsum(A, b, 'i j, j -> i')

print(solve())
```

## Why this works

Matrix-vector product: the column axis j of A is shared with the vector's only axis and summed, leaving the row axis i as the output vector.
