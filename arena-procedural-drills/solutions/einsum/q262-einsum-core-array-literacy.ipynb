## Problem

Write a function solve(a, b) that takes an (i, j) matrix and a length-j vector, and returns the length-i MATRIX-VECTOR product: entry i is the dot product of row i of a with b.


<details><summary>💡 Hint (click to reveal)</summary>

The vector's single index matches the matrix's column index and contracts away, leaving one value per row.

</details>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.einsum('ij,j->i', a, b)


print(solve(t.tensor([[1.0, 2.0], [3.0, 4.0]]), t.tensor([10.0, 1.0])))
```

## Why this works

einsum('ij,j->i', a, b) contracts j — each row of a dots with b — which is precisely the matrix-vector product a @ b. Identity leaves the vector unchanged.
