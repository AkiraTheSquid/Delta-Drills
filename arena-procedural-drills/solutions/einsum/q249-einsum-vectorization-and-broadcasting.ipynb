## Problem

Write a function solve(x) that takes a 4-D float array of shape (b, c, h, w) — a batch of multi-channel images — and returns a 1-D array of length b where entry n is the SUM OF SQUARES of every value in image n (across all channels and pixels).


<details><summary>💡 Hint (click to reveal)</summary>

Feed x in twice with identical indices and keep only the batch index — every other axis multiplies with itself and sums away.

</details>

## Reference solution


```python
import numpy as np

def solve(x):
    return np.einsum('bchw,bchw->b', x, x)


print(solve(np.ones((2, 1, 2, 2))))
```

## Why this works

'bchw,bchw->b' pairs x with itself elementwise (squaring) and contracts c, h, and w, leaving one sum of squares per batch entry. (x ** 2).sum(axis=(1, 2, 3)) is the same reduction; negatives square positive, as the last case confirms.
