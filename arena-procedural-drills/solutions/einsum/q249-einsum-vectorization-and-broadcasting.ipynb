## Problem

For a batch of images represented by a 4D array X, compute the sum of the squares of all pixel values for each image. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Contract X against itself over the channel and spatial axes while keeping only the batch axis in the output.

</details>

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    X = np.arange(1, 17).reshape((2, 2, 2, 2))
    return einsum(X, X, 'b c h w, b c h w -> b')

print(solve())
```

## Why this works

Multiplying X by itself with identical index labels squares every pixel; keeping only the batch index b in the output sums those squares over channel, height, and width per image. The result is one scalar per batch element.
