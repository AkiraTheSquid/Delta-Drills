## Problem

Given a query vector q of shape (d) and a memory matrix M of shape (n, d), compute the similarity between q and each row of M, resulting in a 1D array of length n. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Share the query's only axis with M's feature axis and keep the row axis on the output.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

q = np.arange(1, 3)
M = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(q, M, 'd, n d -> n')

print(solve())
```

## Why this works

The feature axis d is shared and contracted, so each row n of M is dotted with q. Keeping only n in the output yields a similarity score per row.
