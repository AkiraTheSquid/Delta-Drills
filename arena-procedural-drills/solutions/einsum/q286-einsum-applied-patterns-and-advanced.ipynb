## Problem

Write a function solve(q, m) that takes a single query vector q of length d and a memory matrix m of shape (n, d), and returns the length-n similarity vector: entry i is the dot product of q with row i of m.


<details><summary>💡 Hint (click to reveal)</summary>

Contract the shared feature index d; only the memory-row index survives.

</details>

## Reference solution


```python
import numpy as np

def solve(q, m):
    return np.einsum('d,nd->n', q, m)


print(solve(np.array([1.0, 0.0]), np.array([[2.0, 5.0], [0.0, 3.0]])))
```

## Why this works

einsum('d,nd->n', q, m) dots the query against every memory row at once — m @ q, the retrieval scoring step of attention and nearest-neighbor lookups.
