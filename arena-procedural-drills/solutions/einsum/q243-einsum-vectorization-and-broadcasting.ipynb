## Problem

Given two batches of vectors, create a batch of matrices where each element of the first batch is multiplied by every element of the second batch. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Share the batch index on both inputs and give the two vector indices distinct names in the output so neither is contracted.

</details>

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    u = np.arange(1, 5).reshape((2, 2))
    v = np.arange(1, 5).reshape((2, 2))
    return einsum(u, v, 'b i, b j -> b i j')

print(solve())
```

## Why this works

Keeping the batch axis b shared and letting the vector axes i and j appear separately in the output forms an outer product per batch element. Because no index is summed away, each entry is u[b,i]*v[b,j], yielding a batch of matrices.
