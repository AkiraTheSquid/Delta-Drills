## Problem

Create a matrix where each element at position (i, j) is the result of multiplying the i-th element of vector v1 by the j-th element of vector v2. Use the provided inputs and print the resulting matrix.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

v1 = np.arange(1, 3)
v2 = np.arange(1, 3)

def solve():
    return einsum(v1, v2, 'i, j -> i j')

print(solve())
```

## Why this works

An outer product has no shared/contracted index: i comes only from v1 and j only from v2, and both appear in the output. Each entry (i,j) is simply v1[i]*v2[j].
