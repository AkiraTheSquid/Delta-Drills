## Problem

Write a function solve(v1, v2) that takes two 1-D vectors of lengths i and j and returns their (i, j) OUTER PRODUCT matrix: entry [p, q] is v1[p] * v2[q].


<details><summary>💡 Hint (click to reveal)</summary>

Two DIFFERENT indices with nothing contracted — every pairing survives into the output.

</details>

## Reference solution


```python
import numpy as np

def solve(v1, v2):
    return np.einsum('i,j->ij', v1, v2)


print(solve(np.array([1.0, 2.0]), np.array([10.0, 20.0, 30.0])))
```

## Why this works

einsum('i,j->ij', v1, v2) has no repeated index, so nothing sums: each output cell is one product v1[i] * v2[j] — the outer product np.outer computes, and the transpose of einsum('i,j->ji', ...).
