## Problem

Write a function solve(a, b) that takes two 3-D tensors of identical shape and returns the SCALAR total of the products of corresponding elements — the full elementwise contraction across all three axes.


<details><summary>💡 Hint (click to reveal)</summary>

## Reference solution


```python
import torch as t

def solve(a, b):
    return t.einsum('abc,abc->', a, b)


print(solve(t.ones((2, 2, 2)), t.ones((2, 2, 2))))
```

## Why this works

einsum('abc,abc->', a, b) is the rank-3 inner product: elementwise multiply and sum every axis, equal to (a * b).sum(). It generalizes the vector dot product to any matching shapes.
