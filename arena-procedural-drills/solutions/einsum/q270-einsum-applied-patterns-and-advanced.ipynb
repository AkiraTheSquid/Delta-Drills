## Problem

Given two 3D tensors A and B, compute the total sum of the products of their corresponding elements. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

A = np.arange(1, 9).reshape((2, 2, 2))
B = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(A, B, 'a b c, a b c ->')

print(solve())
```

## Why this works

Sharing all three index names a b c between the two tensors means corresponding elements are multiplied; an empty output pattern contracts every axis, summing the products into a single scalar (the full Frobenius inner product).
