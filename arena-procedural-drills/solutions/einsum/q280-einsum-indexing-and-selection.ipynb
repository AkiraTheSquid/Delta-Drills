## Problem

Write a function solve(a) that takes a 4-D tensor of shape (b, h, s, d) and returns the (b, s, d) tensor obtained by SUMMING across the second (h) dimension only.


<details><summary>💡 Hint (click to reveal)</summary>

Drop exactly one letter from the output — the h axis is the one summed away.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('bhsd->bsd', a)


print(solve(np.ones((1, 2, 1, 2))))
```

## Why this works

einsum('bhsd->bsd', a) omits h from the output, so it contracts — a.sum(axis=1) with the remaining axes untouched. Which letter you drop chooses which axis collapses.
