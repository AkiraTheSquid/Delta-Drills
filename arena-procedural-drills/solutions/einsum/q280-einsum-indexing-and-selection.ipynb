## Problem

Given a 4D tensor of shape (b, h, s, d), compute the sum across the second dimension to obtain a tensor of shape (b, s, d). Print the result using the provided input. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

activations = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(activations, 'b h s d -> b s d')

print(solve())
```

## Why this works

Omitting the head axis h from the output contracts it, summing across all heads for each (b, s, d) position. The remaining axes are preserved in place.
