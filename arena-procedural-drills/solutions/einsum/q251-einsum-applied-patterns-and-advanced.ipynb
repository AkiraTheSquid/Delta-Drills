## Problem

Given a 3D array X of shape (2, 2, 2) and a 1D array v of length 2, apply the vector to each token in the batch by performing a reduction that combines the last dimension of X with the vector, resulting in a 2D array of shape (2, 2). Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Match the hidden axis d on both X and v, and keep only b and t in the output.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

X = np.arange(1, 9).reshape((2, 2, 2))
v = np.arange(1, 3)

def solve():
    return einsum(X, v, 'b t d, d -> b t')

print(solve())
```

## Why this works

Each token's hidden vector (axis d) is dotted with the single shared weight vector v. Labeling d in both operands and omitting it from the output contracts it, leaving the batch b and token t axes, so one weight vector is applied to every token.
