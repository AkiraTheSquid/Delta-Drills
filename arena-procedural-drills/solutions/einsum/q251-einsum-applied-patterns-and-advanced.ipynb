## Problem

Write a function solve(x, v) that takes a 3-D array x of shape (b, t, d) — a batch of t tokens with d features — and a 1-D vector v of length d, and returns the (b, t) array where each token's feature vector has been reduced to its dot product with v.


<details><summary>💡 Hint (click to reveal)</summary>

The vector shares only the feature index with the batch — contract d, keep b and t.

</details>

## Reference solution


```python
import numpy as np

def solve(x, v):
    return np.einsum('btd,d->bt', x, v)


print(solve(np.ones((1, 2, 3)), np.array([1.0, 2.0, 3.0])))
```

## Why this works

einsum('btd,d->bt', x, v) dots every token's d-dimensional feature vector with v in one pass, collapsing the feature axis and leaving one scalar per token — the same as x @ v.
