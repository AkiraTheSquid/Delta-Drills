## Problem

Write a function solve(t) that takes a 4-D channels-second tensor of shape (b, c, h, w) and returns it in channels-LAST layout (b, h, w, c) — a pure axis permutation, no values changed.


<details><summary>💡 Hint (click to reveal)</summary>

List the same letters on both sides in the new order — einsum permutes without summing when nothing repeats within an operand.

</details>

## Reference solution


```python
import numpy as np

def solve(t):
    return np.einsum('bchw->bhwc', t)


print(solve(np.arange(8.0).reshape(1, 2, 2, 2)))
```

## Why this works

einsum('bchw->bhwc', t) relabels the axes so the channel axis lands last — identical to np.transpose(t, (0, 2, 3, 1)) or np.moveaxis(t, 1, -1), the standard conversion between PyTorch-style and image-style layouts.
