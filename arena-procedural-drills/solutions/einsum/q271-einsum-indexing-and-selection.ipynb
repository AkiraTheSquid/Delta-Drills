## Problem

Given a 4D tensor T with shape (2, 2, 2, 2), rearrange the dimensions so that the color channel, currently in the second position, is moved to the last position. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

A single-tensor pattern with the same axes on both sides is just a transpose; reorder the output labels to place the channel last.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

T = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(T, 'b c h w -> b h w c')

print(solve())
```

## Why this works

With a single operand and all four axes preserved, einsum performs a pure permutation. Listing the channel axis c last in the output moves it from the second position to the final position without altering any values.
