## Problem

Write a function solve(a) that takes a batch of matrices of shape (b, n, n) and returns the 1-D length-b tensor of each matrix's TOP-LEFT element a[k, 0, 0].


<details><summary>💡 Hint (click to reveal)</summary>

Fix the last two indices at 0 and let the batch axis run free — plain integer indexing, no contraction needed.

</details>

## Reference solution


```python
import torch as t

def solve(a):
    return a[:, 0, 0]


print(solve(t.arange(8.0).reshape(2, 2, 2)))
```

## Why this works

a[:, 0, 0] slices out one scalar per batch item. Not everything needs einsum — recognizing when simple indexing suffices is part of the vocabulary; an einsum can only reduce or permute, not select a fixed position.
