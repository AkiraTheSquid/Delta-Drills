## Problem

From a 3D tensor A with shape (b, n, n), return a 1D array containing the first element from each 2D matrix in the batch. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(A, 'b i j -> b i j')[:, 0, 0]

print(solve())
```

## Why this works

An identity einsum pattern simply returns the tensor unchanged; the real work is the slice [:, 0, 0], which picks the top-left (0,0) element of every matrix in the batch into a 1D array.
