## Problem

Given a 1D array x containing the values from 1 to 2, return a single value that represents the total of all elements in the array. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Single operand, name the one axis, and leave the output side empty to sum everything.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(1, 3)

def solve():
    return einsum(x, 'i ->')

print(solve())
```

## Why this works

With a single operand and an empty output, every axis is summed. Naming the only axis i and outputting nothing collapses the vector to the scalar total of its elements.
