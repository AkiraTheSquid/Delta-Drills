## Problem

Write a function solve(x) that takes a PyTorch tensor of ANY shape and returns the total of all its elements as a scalar, expressed as a full contraction (every index summed away).


<details><summary>💡 Hint (click to reveal)</summary>

Sum over EVERY index by leaving the output side empty — einsum's interleaved form handles any rank generically.

</details>

## Reference solution


```python
import torch as t

def solve(x):
    return t.einsum(x, list(range(x.ndim)), [])


print(solve(t.arange(6.0).reshape(2, 3)))
```

## Why this works

With an empty output-index list every axis contracts, producing the grand total — x.sum(). The interleaved form t.einsum(x, [0, 1, ...], []) generalizes '...->' notation to arbitrary rank without writing letters.
