## Problem

Write a function solve(a) that takes a 2-D numpy array and returns its TRANSPOSE — the array with row and column axes swapped, so output[j, i] equals a[i, j]. Express it as an axis-relabeling einsum (or any equivalent).


<details><summary>💡 Hint (click to reveal)</summary>

In einsum, writing the output indices in a different order IS the permutation — nothing is summed.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('ij->ji', a)


print(solve(np.array([[1, 2], [3, 4]])))
```

## Why this works

einsum('ij->ji', a) relabels the axes without contracting anything: each output position (j, i) reads input position (i, j). It's exactly a.T — the einsum form matters as the simplest example of pure axis permutation in index notation, the non-square case pinning the orientation.
