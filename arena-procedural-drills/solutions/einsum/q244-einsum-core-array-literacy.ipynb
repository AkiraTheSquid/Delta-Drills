## Problem

You are given A, a 2D integer array of shape (2, 2) containing the values 1 through 4 in row-major order. Return a 2D array of shape (2, 2) in which the row and column axes have been swapped, so the output entry at position (j, i) equals A[i, j]. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Reorder the two axis names on the output side to swap them.

</details>

## Reference solution


```python
import numpy as np
from einops import einsum

def solve():
    A = np.arange(1, 5).reshape((2, 2))
    return einsum(A, 'i j -> j i')

print(solve())
```

## Why this works

Listing the output axes as 'j i' instead of 'i j' permutes the axes, which is exactly a transpose. No index is summed, so values are merely relocated to position (j, i).
