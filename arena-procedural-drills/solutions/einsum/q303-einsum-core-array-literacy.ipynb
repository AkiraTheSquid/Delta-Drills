## Problem

Given a square matrix A, compute the difference between the matrix and its transpose to obtain the anti-symmetric part. Use the exact inputs below and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Get the transpose by reading the input axes in swapped order into the same output layout, then subtract from the original.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, 'i j -> i j') - einsum(A, 'j i -> i j')

print(solve())
```

## Why this works

The first einsum returns A unchanged; the second swaps the input axis order ('j i' read into output 'i j') to produce the transpose. Subtracting gives A - A.T, the anti-symmetric part.
