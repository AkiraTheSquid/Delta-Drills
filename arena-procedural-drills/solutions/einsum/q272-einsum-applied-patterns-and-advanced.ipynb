## Problem

Given a 3D array X representing a sequence of features and a 2D projection matrix M, apply the projection to transform the feature dimension of the sequence and return the resulting array. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Match the input feature axis to the matrix's first axis and let the matrix's second axis become the new feature dimension.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

X = np.arange(1, 9).reshape((2, 2, 2))
M = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(X, M, 'b t c, c f -> b t f')

print(solve())
```

## Why this works

The feature axis c is shared and contracted, mapping the input feature dimension into the new dimension f. Batch b and time t are carried through untouched, so this is a batched matrix projection of every (b, t) feature vector by M.
