## Problem

Write a function solve(x, m) that takes a sequence tensor x of shape (b, t, c) and a projection matrix m of shape (c, f), and returns the (b, t, f) tensor with the feature axis projected through m — every token's feature vector multiplied by the matrix.


<details><summary>💡 Hint (click to reveal)</summary>

Only the channel index is shared with the matrix — batch and time pass straight through while c contracts into f.

</details>

## Reference solution


```python
import torch as t

def solve(x, m):
    return t.einsum('btc,cf->btf', x, m)


print(solve(t.ones((1, 2, 2)), t.eye(2)))
```

## Why this works

einsum('btc,cf->btf', x, m) applies one linear layer to every token: c contracts against the matrix's input axis, f is the new feature axis. Broadcasting matmul x @ m does the same because the last axis of x aligns with m's first.
