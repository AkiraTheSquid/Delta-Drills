## Problem

Write a function solve(a, b) that takes two 2-D arrays of shapes (p, q) and (r, s) and returns their (p, r, q, s) TENSOR (outer) product with the axes interleaved as p, r, q, s: entry [i, k, j, l] equals a[i, j] * b[k, l].


<details><summary>💡 Hint (click to reveal)</summary>

Four distinct letters, none contracted — the output order after the arrow chooses the interleaving.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.einsum('pq,rs->prqs', a, b)


print(solve(np.array([[1.0, 2.0]]), np.array([[3.0], [4.0]])))
```

## Why this works

With no repeated index, every element of a multiplies every element of b; the output spec 'prqs' interleaves the row axes before the column axes, distinguishing it from np.tensordot's default 'pqrs' layout — axis ORDER is the drill.
