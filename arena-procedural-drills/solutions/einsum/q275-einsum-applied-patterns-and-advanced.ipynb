## Problem

Given two 2D arrays A and B, create a new tensor where each element of A is multiplied by every element of B, resulting in a 4D array. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use four completely different labels so no axis is summed, then order them on the output to interleave A's and B's dimensions.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

A = np.arange(1, 5).reshape((2, 2))
B = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, B, 'p q, r s -> p r q s')

print(solve())
```

## Why this works

No index is shared between A and B, so nothing is contracted and einsum forms the full outer product of every element of A with every element of B. The four distinct labels appear in the output, giving a 4D tensor.
