## Problem

Given two arrays, queries of shape (b, q, d) and keys of shape (b, k, d), calculate a score matrix that combines these inputs to produce an output of shape (b, q, k). (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Align the batch axis, contract the shared feature axis d, and let the distinct query and key position axes form the output matrix.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

Q = np.arange(1, 9).reshape((2, 2, 2))
K = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(Q, K, 'b q d, b k d -> b q k')

print(solve())
```

## Why this works

This is batched attention scoring: the feature axis d is shared and contracted while the batch axis b is kept aligned, producing a (b, q, k) matrix of query-key dot products per batch.
