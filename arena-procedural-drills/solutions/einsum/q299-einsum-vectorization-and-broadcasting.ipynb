## Problem

Write a function solve(q, k) that takes queries q of shape (b, q, d) and keys k of shape (b, k, d) — single-head attention, no head axis — and returns the (b, q, k) score tensor of all query-key dot products within each batch item.

Contract d, keep both position indices, share the batch — the multi-head version just carries one more letter.

## Reference solution


```python
import numpy as np

def solve(q, k):
    return np.einsum('bqd,bkd->bqk', q, k)


print(solve(np.ones((1, 1, 2)), np.ones((1, 2, 2))))
```

## Why this works

einsum('bqd,bkd->bqk', q, k) is the score computation of single-head attention: every query dots every key within its batch item. Equivalent to q @ k.transpose(0, 2, 1).
