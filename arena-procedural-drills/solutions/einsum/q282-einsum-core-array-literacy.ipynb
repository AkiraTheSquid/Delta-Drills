## Problem

Given a 3D tensor `data` of shape (b, t, d), compute the average of the values along the second dimension (t) by summing the values and then dividing by the number of elements in that dimension. Use the inputs provided and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

First contract the t axis with einsum to get sums, then divide by the size of that dimension.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

data = np.arange(1, 9).reshape((2, 2, 2))
t = 2

def solve():
    return einsum(data, 'b t d -> b d') / t

print(solve())
```

## Why this works

einsum sums out the t axis to give the per-(b, d) totals, then dividing by the known length t turns the sum into a mean. The result is float because of the division.
