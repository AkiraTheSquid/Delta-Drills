## Problem

Write a function solve(data) that takes a 3-D tensor of shape (b, t, d) and returns the (b, d) MEAN over the middle (t) axis: sum across t and divide by t's length.


<details><summary>💡 Hint (click to reveal)</summary>

einsum can only sum — the divide-by-count that turns a sum into a mean happens outside.

</details>

## Reference solution


```python
import numpy as np

def solve(data):
    return np.einsum('btd->bd', data) / data.shape[1]


print(solve(np.ones((1, 4, 2))))
```

## Why this works

einsum('btd->bd', data) sums over the dropped t axis; dividing by data.shape[1] converts it to the mean, matching data.mean(axis=1). Averaging over tokens like this is how sequence embeddings get pooled.
