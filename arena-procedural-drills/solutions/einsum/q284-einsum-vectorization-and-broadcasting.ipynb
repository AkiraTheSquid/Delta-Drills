## Problem

For a batch of vectors v with shape (b, d), compute the sum of the squares of each vector's elements. Use the provided input below and print the result. v = np.arange(1, 5).reshape((2, 2))


<details><summary>💡 Hint (click to reveal)</summary>

Multiply the batch of vectors by itself, contract the feature axis, and keep the batch axis to get per-vector squared norms.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum

v = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(v, v, 'b d, b d -> b')

print(solve())
```

## Why this works

Using v twice with the feature axis d shared and contracted multiplies each element by itself and sums over d, giving each vector's squared length. The batch axis b is kept, so there is one scalar per row.
