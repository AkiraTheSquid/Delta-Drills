## Problem

Write a function solve(v) that takes a batch of vectors of shape (b, d) and returns a length-b array where entry n is the SUM OF SQUARES of vector n's elements (its squared Euclidean norm).


<details><summary>💡 Hint (click to reveal)</summary>

Pair the batch with itself: matching d entries multiply (squares) and d contracts, keeping b.

</details>

## Reference solution


```python
import numpy as np

def solve(v):
    return np.einsum('bd,bd->b', v, v)


print(solve(np.array([[3.0, 4.0]])))
```

## Why this works

einsum('bd,bd->b', v, v) squares each component (elementwise self-product) and sums over d per batch row — the squared norms, (v ** 2).sum(axis=1).
