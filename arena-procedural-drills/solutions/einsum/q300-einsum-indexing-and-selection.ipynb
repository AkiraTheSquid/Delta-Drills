## Problem

Given a 3D tensor img with shape (c, h, w), return a 2D array that combines the height and width dimensions into a single dimension while preserving the channel dimension. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Use rearrange and group the two spatial axes in parentheses on the right to merge them into one.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

img = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einops.rearrange(img, 'c h w -> c (h w)')

print(solve())
```

## Why this works

rearrange merges axes when they are grouped with parentheses on the output side. '(h w)' flattens height and width into one contiguous dimension while the channel axis stays separate, giving shape (c, h*w).
