## Problem

Write a function solve(img) that takes a channels-first image of shape (c, h, w) and returns the (c, h*w) array with the two spatial axes FLATTENED into one, channels preserved — each channel becomes one long row.


<details><summary>💡 Hint (click to reveal)</summary>

A reshape with -1 merges the trailing axes in row-major order — h varies slower than w, exactly the '(h w)' grouping.

</details>

## Reference solution


```python
import numpy as np

def solve(img):
    return img.reshape(img.shape[0], -1)


print(solve(np.arange(8.0).reshape(2, 2, 2)))
```

## Why this works

img.reshape(c, -1) concatenates each channel's rows into one vector, matching einops' rearrange(img, 'c h w -> c (h w)'). Flattening spatial dims like this feeds conv features into dense layers.
