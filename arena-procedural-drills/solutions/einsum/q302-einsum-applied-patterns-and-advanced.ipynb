## Problem

For a 5D tensor v representing video data with dimensions for batch, time, channels, height, and width, compute the average across the time, height, and width dimensions while keeping the batch and channel dimensions intact. Use the inputs provided and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce to (b, c) with einsum to get the sum, then divide by the number of averaged positions to turn it into a mean.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

v = np.arange(1, 33).reshape((2, 2, 2, 2, 2))
h = 2
t = 2
w = 2

def solve():
    return einsum(v, 'b t c h w -> b c') / (t*h*w)

print(solve())
```

## Why this works

Keeping only b and c on the output sums over t, h, and w. Dividing by the count of summed elements (t*h*w) converts that sum into a mean over those three dimensions, preserving batch and channel.
