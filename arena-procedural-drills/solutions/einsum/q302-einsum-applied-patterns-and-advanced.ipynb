## Problem

Write a function solve(v) that takes a 5-D video tensor of shape (b, t, c, h, w) and returns the (b, c) array holding the MEAN over time and both spatial axes — sum out t, h, w and divide by their combined count.


<details><summary>💡 Hint (click to reveal)</summary>

Sum out three letters at once, then divide by the product of those axes' lengths for the mean.

</details>

## Reference solution


```python
import numpy as np

def solve(v):
    return np.einsum('btchw->bc', v) / (v.shape[1] * v.shape[3] * v.shape[4])


print(solve(np.ones((1, 2, 3, 2, 2))))
```

## Why this works

einsum('btchw->bc', v) collapses time and space in one contraction; dividing by t*h*w converts to the mean — v.mean(axis=(1, 3, 4)), a per-channel average over an entire clip.
