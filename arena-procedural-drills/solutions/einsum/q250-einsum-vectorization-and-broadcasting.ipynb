## Problem

Write a function solve(a) that takes a 3-D array of shape (b, i, d) — a batch of i feature vectors of dimension d — and returns the (b, i, i) batch of similarity matrices: entry [n, p, q] is the dot product of vectors p and q within batch item n.


<details><summary>💡 Hint (click to reveal)</summary>

Feed the batch in twice with DIFFERENT vector indices (i and j) but the same feature index d — contracting d gives every within-batch dot product.

</details>

## Reference solution


```python
import numpy as np

def solve(a):
    return np.einsum('bid,bjd->bij', a, a)


print(solve(np.arange(8.0).reshape(2, 2, 2)))
```

## Why this works

einsum('bid,bjd->bij', a, a) shares b and d between the two copies while keeping i and j separate, producing each batch item's Gram matrix of feature vectors. Orthonormal vectors give the identity — the second case.
