## Problem

For a 3D array A with shape (2, 2, 2), calculate the similarity between the internal feature vectors for each item in the batch, resulting in a 3D array that captures this relationship. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Both operands are the same array; share the hidden axis d and keep separate row indices i and j so every pair within a batch item gets dotted.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(A, A, 'b i d, b j d -> b i j')

print(solve())
```

## Why this works

For each batch item b we want the Gram matrix of its own feature vectors: entry (i,j) is the dot product of vector i with vector j over the hidden dimension d. Naming the shared contracted axis d in both operands and dropping it from the output sums it away, while keeping b, i, j free preserves the batch and the i x j pairing.
