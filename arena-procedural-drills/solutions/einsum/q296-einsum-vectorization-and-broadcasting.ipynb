## Problem

Given a 2D array x of shape (t, d), create a 3D array of shape (t, t, d) where each entry at position (i, j) contains the product of the ith and jth vectors from x. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

x = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(x, x, 'i d, j d -> i j d')

print(solve())
```

## Why this works

Using x twice with separate row indices i and j but a shared, retained feature axis d produces the element-wise product of vector i and vector j without summing over d, giving a (t, t, d) tensor.
