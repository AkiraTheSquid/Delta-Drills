## Problem

Write a function solve(x) that takes a 4-D tensor of shape (b, c, h, w) and returns the length-b vector whose entry n is the SUM of every channel and pixel value of batch item n.


<details><summary>💡 Hint (click to reveal)</summary>

Keep only the batch letter — the other three all vanish and sum together.

</details>

## Reference solution


```python
import torch as t

def solve(x):
    return t.einsum('bchw->b', x)


print(solve(t.ones((2, 1, 2, 2))))
```

## Why this works

einsum('bchw->b', x) contracts channels and both spatial axes in one go — x.sum(axis=(1, 2, 3)), one grand total per image.
