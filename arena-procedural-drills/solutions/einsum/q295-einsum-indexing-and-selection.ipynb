## Problem

Given a 4D tensor X with shape (b, c, h, w), compute a single value for each batch item by aggregating all pixel and channel values. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Name all four axes but keep just the batch axis in the output so the rest are summed.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

X = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(X, 'b c h w -> b')

print(solve())
```

## Why this works

Keeping only the batch axis on the output side reduces the channel, height, and width axes, summing every channel and pixel value into one number per batch item.
