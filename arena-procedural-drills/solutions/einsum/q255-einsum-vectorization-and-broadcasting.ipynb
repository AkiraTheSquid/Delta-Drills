## Problem

Write a function solve(a, w) that takes a 2-D array a of shape (n, d) and a 1-D weight vector w of length n, and returns the WEIGHTED AVERAGE of a's rows as a length-d vector: the w-weighted sum of the rows divided by the total weight (assume w.sum() is nonzero).

## Reference solution


```python
import numpy as np

def solve(a, w):
    return np.einsum('n,nd->d', w, a) / w.sum()


print(solve(np.array([[0.0, 0.0], [10.0, 20.0]]), np.array([1.0, 3.0])))
```

## Why this works

einsum('n,nd->d', w, a) computes the weighted sum of rows (w @ a); dividing by w.sum() turns it into a weighted mean. Equal weights recover the ordinary column average — the last case.
