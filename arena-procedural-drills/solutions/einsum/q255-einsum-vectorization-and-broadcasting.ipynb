## Problem

Given a 2D array A and a 1D weight vector w, compute the weighted sum of the rows of A, then normalize the result by the sum of the weights. (★☆☆)


<details><summary>💡 Hint (click to reveal)</summary>

Contract the row axis n between w and A to get a weighted row-sum, then divide by the total weight.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

w = np.arange(1, 3)
A = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(w, A, 'n, n d -> d') / w.sum()

print(solve())
```

## Why this works

The einsum weights each row of A by the matching scalar in w (shared axis n) and sums over the rows, leaving the feature axis d. Dividing by w.sum() turns that weighted sum into a weighted average.
