## Problem

Given a permutation matrix P of shape (n, n) and a vector v of shape (n), reorder the elements of v according to the structure of P. Use the inputs provided and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

This is a standard matrix-times-vector contraction: share the column index j between P and v and keep only the row index i.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

P = np.arange(1, 5).reshape((2, 2))
v = np.arange(1, 3)

def solve():
    return einsum(P, v, 'i j, j -> i')

print(solve())
```

## Why this works

Reordering v by a (permutation) matrix is a matrix-vector product P @ v: the shared index j is contracted while output index i is kept. Each output entry i is the sum over j of P[i,j]*v[j], which selects/permutes v's elements according to P's rows.
