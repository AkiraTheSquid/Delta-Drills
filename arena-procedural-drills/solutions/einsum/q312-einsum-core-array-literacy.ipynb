## Problem

Write a function solve(p, v) that takes an (n, n) PERMUTATION matrix p (exactly one 1 per row and column, zeros elsewhere) and a length-n vector v, and returns the reordered vector p @ v — entry i of the result is the v entry selected by row i's 1.


<details><summary>💡 Hint (click to reveal)</summary>

It's just a matrix-vector product — the permutation structure of p does the reordering by itself.

</details>

## Reference solution


```python
import numpy as np

def solve(p, v):
    return np.einsum('ij,j->i', p, v)


print(solve(np.array([[0.0, 1.0], [1.0, 0.0]]), np.array([10.0, 20.0])))
```

## Why this works

einsum('ij,j->i', p, v) is p @ v; because each row of a permutation matrix has a single 1, every output entry copies exactly one input entry — matrix multiplication as data movement. p = eye(n)[perm] reorders v into v[perm].
