## Problem

Write a function solve(s, v) that takes attention weights s of shape (b, h, q, k) and values v of shape (b, h, k, d), and returns the (b, h, q, d) result of applying the weights: for each batch and head, the weighted sum over the k (key) axis of the value vectors.


<details><summary>💡 Hint (click to reveal)</summary>

Only the key index k is shared and absent from the output — batch, head, query, and depth all ride along.

</details>

## Reference solution


```python
import numpy as np

def solve(s, v):
    return np.einsum('bhqk,bhkd->bhqd', s, v)


print(solve(np.ones((1, 1, 1, 2)) / 2, np.arange(4.0).reshape(1, 1, 2, 2)))
```

## Why this works

einsum('bhqk,bhkd->bhqd', s, v) contracts the key axis: each query position's output vector is the s-weighted mix of the value vectors. This is exactly the attention-weights-times-values step of a transformer; a one-hot weight row selects a single value vector (third case).
