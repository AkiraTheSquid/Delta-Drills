## Problem

Given two 4D arrays S and V with shapes (2, 2, 2, 2), combine them by summing over the dimension representing keys, resulting in a new array with the shape (2, 2, 2, d). Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Treat b and h as batch axes, contract the shared keys axis k, and leave q and d as the resulting matrix shape.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

S = np.arange(1, 17).reshape((2, 2, 2, 2))
V = np.arange(1, 17).reshape((2, 2, 2, 2))

def solve():
    return einsum(S, V, 'b h q k, b h k d -> b h q d')

print(solve())
```

## Why this works

This is batched attention-weighted aggregation: for each batch b and head h, the score matrix (q x k) multiplies the value matrix (k x d). The keys axis k is shared and summed away, while b and h ride along as batch dimensions and q, d remain as the output matrix axes.
