## Problem

Given two matrices A and B of shape (n, n), calculate the sum of the elements along the main diagonal of their product without explicitly forming the product itself. Use the exact inputs below and print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Cross the two index letters between A and B and reduce everything; the trace of a product is sum_ij A_ij B_ji.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

A = np.arange(1, 5).reshape((2, 2))
B = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(A, B, 'i j, j i ->')

print(solve())
```

## Why this works

trace(A @ B) equals the sum over i,j of A[i,j]*B[j,i]. By swapping the index order on B and reducing both axes (empty output), einsum computes that trace directly without ever materializing the product matrix.
