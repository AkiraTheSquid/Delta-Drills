## Problem

Write a function solve(a, b) that takes two (n, n) matrices and returns the TRACE OF THEIR PRODUCT tr(a @ b) as a scalar — computed directly by contraction, WITHOUT forming the product matrix.


<details><summary>💡 Hint (click to reveal)</summary>

Transpose one operand's letters and contract BOTH indices — the double contraction is the trace of the product.

</details>

## Reference solution


```python
import numpy as np

def solve(a, b):
    return np.einsum('ij,ji->', a, b)


print(solve(np.array([[1.0, 2.0], [3.0, 4.0]]), np.eye(2)))
```

## Why this works

einsum('ij,ji->', a, b) sums a[i,j] * b[j,i] over all i, j — exactly tr(a @ b), in O(n^2) instead of the O(n^3) product-then-trace. It equals the Frobenius inner product of a with b.T.
