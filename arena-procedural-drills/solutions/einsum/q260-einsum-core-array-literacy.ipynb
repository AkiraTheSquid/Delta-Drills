## Problem

Given two 2x2 matrices C and D, compute the resulting matrix by summing the products of corresponding elements over the shared inner dimension. Print the result.

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

C = np.arange(1, 5).reshape((2, 2))
D = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(C, D, 'i k, k j -> i j')

print(solve())
```

## Why this works

Standard matrix multiplication: the shared inner axis k is contracted (summed), while the outer row index i of C and column index j of D remain as the output shape.
