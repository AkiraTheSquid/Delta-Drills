## Problem

Write a function solve(c, d) that takes two matrices with compatible shapes (i, k) and (k, j) and returns their MATRIX PRODUCT: entry [i, j] sums the products c[i, k] * d[k, j] over the shared inner dimension k.


<details><summary>💡 Hint (click to reveal)</summary>

The inner index appears in both operands and not in the output — that repetition-then-omission IS matrix multiplication.

</details>

## Reference solution


```python
import torch as t

def solve(c, d):
    return t.einsum('ik,kj->ij', c, d)


print(solve(t.tensor([[1.0, 2.0]]), t.tensor([[3.0], [4.0]])))
```

## Why this works

einsum('ik,kj->ij', c, d) is the index-notation definition of the matrix product: k is contracted, i and j survive. Identity as either factor is a no-op, diagonal factors scale rows — the two structured cases.
