## Problem

Given a batch of vectors v with shape (b, d), compute the outer product of each vector with itself and then sum all resulting matrices together. (★☆☆)

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

v = np.arange(1, 5).reshape((2, 2))

def solve():
    return einsum(v, v, 'b i, b j -> i j')

print(solve())
```

## Why this works

Using v twice with index labels i and j builds the per-item outer product v_i v_j, and because the batch axis b is repeated in both operands but dropped from the output it is summed over. The result is sum_b v[b] outer v[b], i.e. v.T @ v.
