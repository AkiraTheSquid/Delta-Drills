## Problem

Write a function solve(v) that takes a batch of vectors of shape (b, d) and returns the SINGLE (d, d) matrix summing every vector's self outer product: sum over the batch of v[n] (x) v[n]. (Contrast the batched version that keeps one matrix per item.)

## Reference solution


```python
import torch as t

def solve(v):
    return t.einsum('bi,bj->ij', v, v)


print(solve(t.tensor([[1.0, 0.0], [0.0, 2.0]])))
```

## Why this works

einsum('bi,bj->ij', v, v) forms every row's outer product and, since b is absent from the output, sums them — v.T @ v, the scatter matrix accumulated across a batch. Keeping b ('bi,bj->bij') would return the individual matrices instead.
