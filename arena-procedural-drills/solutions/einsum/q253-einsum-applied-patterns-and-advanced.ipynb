## Problem

Given two 3D arrays F and G with shape (2, 2, 2), compute the sum of products of corresponding elements across the channel dimension for each spatial location, resulting in a 2D array. Print the result.


<details><summary>💡 Hint (click to reveal)</summary>

Contract only the channel axis c; keep h and w as matched free axes in both inputs and the output.

</details>

## Reference solution


```python
import numpy as np
import einops
from einops import einsum, rearrange, reduce, repeat

F = np.arange(1, 9).reshape((2, 2, 2))
G = np.arange(1, 9).reshape((2, 2, 2))

def solve():
    return einsum(F, G, 'c h w, c h w -> h w')

print(solve())
```

## Why this works

At each spatial location (h, w) we dot the channel vectors of F and G. The channel axis c is named on both operands but dropped from the output so it is summed, while h and w stay free and are matched elementwise.
