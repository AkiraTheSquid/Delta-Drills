## Problem

Write a function solve(f, g) that takes two 3-D tensors of identical shape (c, h, w) — c channels of h x w maps — and returns the (h, w) tensor formed by multiplying corresponding elements and summing across the CHANNEL dimension only.


<details><summary>💡 Hint (click to reveal)</summary>

Repeat all three indices on both operands but keep only h and w in the output — c is the one that multiplies-and-sums.

</details>

## Reference solution


```python
import torch as t

def solve(f, g):
    return t.einsum('chw,chw->hw', f, g)


print(solve(t.ones((2, 2, 2)), t.ones((2, 2, 2))))
```

## Why this works

einsum('chw,chw->hw', f, g) multiplies matching cells and contracts only the channel axis, leaving a per-pixel channel-dot-product — how feature maps get compared position by position; equivalent to (f * g).sum(axis=0).
