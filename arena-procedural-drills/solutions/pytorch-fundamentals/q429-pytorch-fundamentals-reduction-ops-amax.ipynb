## Problem

Using torch, build the fixed 2x3 float tensor x = [[1.0, 5.0, 3.0], [4.0, 2.0, 6.0]]. Use torch.amax to reduce over dim=1 (the columns), so you get the max of each row. Print the result with print(result.tolist()).

## Reference solution


```python
import torch

x = torch.tensor([[1.0, 5.0, 3.0], [4.0, 2.0, 6.0]])
result = x.amax(dim=1)
print(result.tolist())
```

## Why this works

amax over dim=1 reduces across each row's columns, returning the per-row maximum. For the two rows that gives 5.0 and 6.0.
