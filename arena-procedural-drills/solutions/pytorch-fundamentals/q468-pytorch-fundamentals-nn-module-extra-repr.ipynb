## Problem

Define an nn.Module subclass named Linear that stores in_features, out_features, and bias as attributes in __init__, and override extra_repr to return the string f"in_features={self.in_features}, out_features={self.out_features}, bias={self.bias}". Construct Linear(8, 4, bias=False) and print the result of calling its extra_repr() method directly.

## Reference solution


```python
import torch.nn as nn

class Linear(nn.Module):
    def __init__(self, in_features, out_features, bias=True):
        super().__init__()
        self.in_features = in_features
        self.out_features = out_features
        self.bias = bias
    def extra_repr(self):
        return f"in_features={self.in_features}, out_features={self.out_features}, bias={self.bias}"

m = Linear(8, 4, bias=False)
print(m.extra_repr())
```

## Why this works

extra_repr is the hook nn.Module exposes for adding custom descriptive text to a module's printed representation. Calling it directly just returns that formatted string built from the stored attributes.
