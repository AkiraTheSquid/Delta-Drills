## Problem

`extra_repr` is how a custom Module earns a readable one-line description. Write solve(in_features, out_features, bias): define a class `Linear(torch.nn.Module)` whose `__init__` stores the three arguments as plain attributes (calling `super().__init__()` first), and whose `extra_repr` returns the string `in_features=..., out_features=..., bias=...` filled with the stored values (an f-string). Return an INSTANCE constructed with the given arguments.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(in_features, out_features, bias):
    class Linear(nn.Module):
        def __init__(self, i, o, b):
            super().__init__()
            self.in_features = i
            self.out_features = o
            self.bias = b
        def extra_repr(self):
            return f"in_features={self.in_features}, out_features={self.out_features}, bias={self.bias}"
    return Linear(in_features, out_features, bias)

m = solve(8, 4, False)
print(m)
```

## Why this works

`extra_repr` is the hook `nn.Module.__repr__` calls to fill the parentheses of your module's printed line. Store constructor args as attributes and format them in one f-string — that's how built-in layers like `nn.Linear(8, 4, bias=False)` print themselves.
