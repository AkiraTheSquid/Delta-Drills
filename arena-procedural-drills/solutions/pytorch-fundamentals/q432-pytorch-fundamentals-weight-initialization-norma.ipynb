## Problem

The in-place initializers in `torch.nn.init` end with an underscore and mutate the tensor you hand them. Fill the starter's empty 2×3 tensor `w` with the constant 0.25 using the appropriate init function. Print with print(w.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

An in-place constant fill overwrites whatever torch.empty allocated.

</details>


```python
%pip install -q numpy torch --index-url https://download.pytorch.org/whl/cpu
```

## Reference solution


```python
import torch

w = torch.empty(2, 3)
torch.nn.init.constant_(w, 0.25)
print(w.tolist())
```

## Why this works

init.constant_ fills the tensor in place with the given scalar, so every entry of the 2x3 tensor becomes 0.25. The uninitialized values from torch.empty are fully overwritten.
