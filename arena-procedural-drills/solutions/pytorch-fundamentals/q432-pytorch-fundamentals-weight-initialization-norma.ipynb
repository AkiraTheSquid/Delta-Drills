## Problem

The in-place initializers in `torch.nn.init` end with an underscore and mutate the tensor you hand them. Write solve(w, value): fill the tensor `w` IN PLACE with the constant `value` using the appropriate init function, and return `w`. The grader checks the original tensor object afterward, so a filled copy does not count.


<details><summary>💡 Hint (click to reveal)</summary>

`torch.nn.init.constant_(w, value)` — the trailing underscore = in place.

</details>

## Reference solution


```python
import torch

def solve(w, value):
    torch.nn.init.constant_(w, value)
    return w

w = torch.empty(2, 3)
print(solve(w, 0.25))
```

## Why this works

`torch.nn.init.constant_(w, value)` writes into the tensor you pass (the trailing underscore marks in-place mutation) and returns it. The grader checks the ORIGINAL object, so building a new filled tensor doesn't count — init functions exist to fill already-allocated parameters.
