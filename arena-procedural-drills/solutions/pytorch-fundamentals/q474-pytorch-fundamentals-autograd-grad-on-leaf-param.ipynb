## Problem

Using torch, create a leaf tensor w = [1.0, 2.0, 3.0] with requires_grad=True and a constant x = [4.0, 5.0, 6.0] (no grad). Compute loss = (w * x).sum(), call loss.backward(), then print w.grad as a list via print(w.grad.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Differentiate a sum of products with respect to one factor and the other factor is what remains.

</details>

## Reference solution


```python
import torch
w = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
x = torch.tensor([4.0, 5.0, 6.0])
loss = (w * x).sum()
loss.backward()
print(w.grad.tolist())
```

## Why this works

For loss = sum(w_i * x_i), the partial derivative with respect to each w_i is simply x_i. So after backward the gradient stored on the leaf w equals x exactly: [4, 5, 6].
