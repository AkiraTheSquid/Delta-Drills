## Problem

The most basic autograd round trip. Write solve(w, x) where `w` already requires grad and `x` is a constant: build the scalar loss — the sum of the elementwise product of `w` and `x` — and backprop it with `.backward()`. Predict `w.grad` before running: for this loss you can read it directly off one of the inputs. Return the tuple `(round(float(loss), 4), w.grad)`.


<details><summary>💡 Hint (click to reveal)</summary>

loss = `(w * x).sum()`, then `loss.backward()`; d(loss)/dw is just x.

</details>

## Reference solution


```python
import torch

def solve(w, x):
    loss = (w * x).sum()
    loss.backward()
    return (round(float(loss), 4), w.grad)

w = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
x = torch.tensor([4.0, 5.0, 6.0])
print(solve(w, x))
```

## Why this works

For loss = Σ wᵢxᵢ the gradient w.grad equals x exactly — each wᵢ contributes xᵢ linearly. `.backward()` on the scalar populates `.grad` on every leaf that requires grad; reading the answer off the input before running is the point of the drill.
