## Problem

The most basic autograd round trip: the starter's `w` requires grad, `x` is a constant. Build the scalar loss — the sum of the elementwise product of `w` and `x` — and backprop it. Predict `w.grad` before running: for this loss you can read it directly off one of the inputs. Print with print(w.grad.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Differentiate a sum of products with respect to one factor and the other factor is what remains.

## Reference solution


```python
import torch
w = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
x = torch.tensor([4.0, 5.0, 6.0])
loss = (w * x).sum()
loss.backward()
print(w.grad.tolist())
```

## Why this works

For loss = sum(w_i * x_i), the partial derivative with respect to each w_i is simply x_i. So after backward the gradient stored on the leaf w equals x exactly: [4, 5, 6].
