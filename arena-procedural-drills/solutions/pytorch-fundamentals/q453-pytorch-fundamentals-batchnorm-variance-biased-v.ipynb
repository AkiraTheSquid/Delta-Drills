## Problem

`.var()` hides a convention choice: BIASED divides by $N$, UNBIASED by $N-1$ — and BatchNorm normalizes with the biased one while torch's default is unbiased. On the starter's 2×4 tensor, compute the variance over ALL elements both ways (the `unbiased=` flag) and print [biased, unbiased] with each value round(float(v), 4).


<details><summary>💡 Hint (click to reveal)</summary>

Call .var twice, toggling the unbiased flag; the only difference is dividing by N versus N-1.

</details>

## Reference solution


```python
import torch
x = torch.tensor([[1., 2., 3., 4.], [5., 6., 7., 8.]])
var_biased = x.var(unbiased=False)
var_unbiased = x.var(unbiased=True)
print([round(float(var_biased), 4), round(float(var_unbiased), 4)])
```

## Why this works

Biased variance divides the summed squared deviations by N, unbiased by N-1 (Bessel's correction). For these 8 values the biased value is 5.25 and the unbiased is 6.0; BatchNorm uses the biased estimator while torch's .var() defaults to unbiased.
