## Problem

`.var()` hides a convention choice: BIASED divides by N, UNBIASED by N−1 — and BatchNorm normalizes with the biased one while torch's default is unbiased. Write solve(x): compute the variance of ALL elements of `x` both ways (the `unbiased=` flag) and return `(round(float(var_biased), 4), round(float(var_unbiased), 4))`.


<details><summary>💡 Hint (click to reveal)</summary>

`x.var(unbiased=False)` divides by N; `unbiased=True` (the default) by N−1.

</details>

## Reference solution


```python
import torch

def solve(x):
    vb = x.var(unbiased=False)
    vu = x.var(unbiased=True)
    return (round(float(vb), 4), round(float(vu), 4))

x = torch.tensor([[1., 2., 3., 4.], [5., 6., 7., 8.]])
print(solve(x))
```

## Why this works

Biased variance divides by N, unbiased by N−1 (Bessel's correction). Torch's `.var()` DEFAULT is unbiased — but BatchNorm normalizes with the biased one, so mixing them up produces small, maddening numerical mismatches.
