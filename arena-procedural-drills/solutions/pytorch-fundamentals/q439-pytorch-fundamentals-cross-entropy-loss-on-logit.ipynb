## Problem

Show the identity: single-example cross-entropy equals the NEGATIVE log-softmax at the target index. Write solve(logits, target) for a 1-D logits tensor and a 0-D integer target: compute `manual` from `F.log_softmax` (reduce over dim 0, negate, index at the target), and `builtin` from `F.cross_entropy` — which expects a batch dimension, so unsqueeze both arguments. Return `(round(float(manual), 4), round(float(builtin), 4))` — the two should match.


<details><summary>💡 Hint (click to reveal)</summary>

`-F.log_softmax(logits, dim=0)[target]` vs `F.cross_entropy(logits.unsqueeze(0), target.unsqueeze(0))`.

</details>

## Reference solution


```python
import torch
import torch.nn.functional as F

def solve(logits, target):
    manual = -F.log_softmax(logits, dim=0)[target]
    builtin = F.cross_entropy(logits.unsqueeze(0), target.unsqueeze(0))
    return (round(float(manual), 4), round(float(builtin), 4))

logits = torch.tensor([3.0, 1.0, 0.2, -2.0])
target = torch.tensor(0)
print(solve(logits, target))
```

## Why this works

Cross-entropy for one example IS the negative log-softmax at the target index — the identity this drill has you verify. `F.cross_entropy` insists on a batch dim, hence the unsqueezes; both routes land on the same rounded number.
