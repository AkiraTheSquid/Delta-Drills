## Problem

Show the identity: single-example cross-entropy equals the NEGATIVE log-softmax at the target index. The starter gives 4 logits and target class 0. Compute `manual` from `F.log_softmax` (reduce over dim 0, negate, index at the target), and `builtin` from `F.cross_entropy` — which expects a batch dimension, so unsqueeze both arguments. Print [round(float(manual), 4), round(float(builtin), 4)] — the two should match.

## Reference solution


```python
import torch
import torch.nn.functional as F

logits = torch.tensor([3.0, 1.0, 0.2, -2.0])
target = torch.tensor(0)
manual = -F.log_softmax(logits, dim=0)[target]
builtin = F.cross_entropy(logits.unsqueeze(0), target.unsqueeze(0))
print([round(float(manual), 4), round(float(builtin), 4)])
```

## Why this works

Cross-entropy for one example is just the negative log-softmax probability of the true class, which is what F.cross_entropy computes after batching. Both the manual log-softmax indexing and the built-in call therefore agree.
