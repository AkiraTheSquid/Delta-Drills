## Problem

Show that cross-entropy on a single example equals the negative log-softmax of the logit at the target index, and that this matches F.cross_entropy. Use the logits [3.0, 1.0, 0.2, -2.0] with target class 0. Compute (a) the manual value -log_softmax(logits)[target] using F.log_softmax over dim 0, and (b) F.cross_entropy(logits.unsqueeze(0), target.unsqueeze(0)). Print [round(float(a), 4), round(float(b), 4)] as a list (they should be equal).

## Reference solution


```python
import torch
import torch.nn.functional as F

logits = torch.tensor([3.0, 1.0, 0.2, -2.0])
target = torch.tensor(0)
manual = -F.log_softmax(logits, dim=0)[target]
builtin = F.cross_entropy(logits.unsqueeze(0), target.unsqueeze(0))
print([round(float(manual), 4), round(float(builtin), 4)])
```

## Why this works

Cross-entropy for one example is just the negative log-softmax probability of the true class, which is what F.cross_entropy computes after batching. Both the manual log-softmax indexing and the built-in call therefore agree.
