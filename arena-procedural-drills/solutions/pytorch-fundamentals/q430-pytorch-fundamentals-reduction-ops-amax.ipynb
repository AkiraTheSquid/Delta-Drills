## Problem

Unlike `.max(dim=...)`, which returns a (values, indices) named tuple, `.amax(dim=...)` returns just the values tensor. On the starter's 2×3 tensor `x`, compute `vals` — the max down each COLUMN — with `.amax`, choosing the dim yourself. The starter already prints `vals.tolist()` and then the type NAME of what `x.max(dim=0)` returns.

## Reference solution


```python
import torch

x = torch.tensor([[1.0, 5.0, 3.0], [4.0, 2.0, 6.0]])
vals = x.amax(dim=0)
print(vals.tolist())
print(type(x.max(dim=0)).__name__)
```

## Why this works

amax(dim=0) reduces down the columns and returns only the values tensor, while max(dim=0) returns a named tuple (a torch.return_types.max object) carrying both values and indices. Printing the type name shows they are different return shapes.
