## Problem

Using torch, build the fixed 2x3 float tensor x = [[1.0, 5.0, 3.0], [4.0, 2.0, 6.0]]. Unlike .max(dim=...), which returns a (values, indices) tuple, .amax returns only the values. Compute vals = x.amax(dim=0) (max down each column). On the first line print vals.tolist(); on the second line print the type NAME of the object returned by x.max(dim=0) using print(type(x.max(dim=0)).__name__).

## Reference solution


```python
import torch

x = torch.tensor([[1.0, 5.0, 3.0], [4.0, 2.0, 6.0]])
vals = x.amax(dim=0)
print(vals.tolist())
print(type(x.max(dim=0)).__name__)
```

## Why this works

amax(dim=0) reduces down the columns and returns only the values tensor, while max(dim=0) returns a named tuple (a torch.return_types.max object) carrying both values and indices. Printing the type name shows they are different return shapes.
