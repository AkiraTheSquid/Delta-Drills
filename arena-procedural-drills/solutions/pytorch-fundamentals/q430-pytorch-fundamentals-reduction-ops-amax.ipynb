## Problem

Unlike `.max(dim=...)`, which returns a (values, indices) named tuple, `.amax(dim=...)` returns just the values tensor. Write solve(x) that takes a 2-D float tensor and returns the tuple `(vals, max_is_tensor, amax_is_tensor)`: `vals` = the max down each COLUMN via `.amax` (choose the dim yourself), `max_is_tensor` = `isinstance(x.max(dim=0), torch.Tensor)`, and `amax_is_tensor` = the same check on your `vals` — one of the two is not a plain tensor.

## Reference solution


```python
import torch

def solve(x):
    vals = x.amax(dim=0)
    return (vals, isinstance(x.max(dim=0), torch.Tensor), isinstance(vals, torch.Tensor))

x = torch.tensor([[1.0, 5.0, 3.0], [4.0, 2.0, 6.0]])
print(solve(x))
```

## Why this works

`x.amax(dim=0)` collapses the rows, giving the max of each column as a plain tensor (isinstance Tensor → True). `x.max(dim=0)` returns a `return_types.max` named tuple of (values, indices) — NOT a Tensor — which is why code that treats them interchangeably breaks.
