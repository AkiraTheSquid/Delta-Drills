## Problem

Logits are pre-softmax real-valued scores, and cross-entropy is computed directly from logits (not from probabilities) for numerical stability. Using PyTorch, build the logits tensor [[2.0, 0.5, -1.0], [0.1, 1.2, 3.0]] (a batch of 2 examples over 3 classes) and the integer class targets [0, 2]. Compute the mean cross-entropy loss with torch.nn.functional.cross_entropy (which expects raw logits, not softmax probabilities). Print round(float(loss), 4).

## Reference solution


```python
import torch
import torch.nn.functional as F

logits = torch.tensor([[2.0, 0.5, -1.0],
                       [0.1, 1.2, 3.0]])
targets = torch.tensor([0, 2])
loss = F.cross_entropy(logits, targets)
print(round(float(loss), 4))
```

## Why this works

F.cross_entropy consumes raw logits, internally applying log-softmax then negative-log-likelihood, and averages over the batch. Feeding the two examples with their target classes returns the mean loss.
