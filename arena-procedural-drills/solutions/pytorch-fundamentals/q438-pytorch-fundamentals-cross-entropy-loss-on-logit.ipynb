## Problem

Cross-entropy in PyTorch is computed straight from LOGITS (pre-softmax scores) — you never softmax first. The starter gives a 2×3 logits batch and integer class targets. Compute the mean cross-entropy with the right `torch.nn.functional` call and print round(float(loss), 4).


<details><summary>💡 Hint (click to reveal)</summary>

Pass raw logits and integer targets straight into F.cross_entropy — no manual softmax.

## Reference solution


```python
import torch
import torch.nn.functional as F

logits = torch.tensor([[2.0, 0.5, -1.0],
                       [0.1, 1.2, 3.0]])
targets = torch.tensor([0, 2])
loss = F.cross_entropy(logits, targets)
print(round(float(loss), 4))
```

## Why this works

F.cross_entropy consumes raw logits, internally applying log-softmax then negative-log-likelihood, and averages over the batch. Feeding the two examples with their target classes returns the mean loss.
