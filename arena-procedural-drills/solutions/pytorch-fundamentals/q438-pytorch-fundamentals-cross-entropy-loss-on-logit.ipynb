## Problem

Cross-entropy in PyTorch is computed straight from LOGITS (pre-softmax scores) — you never softmax first. Write solve(logits, targets) for a (N, C) logits batch and integer class targets: return the mean cross-entropy as a plain float rounded to 4 decimals, i.e. `round(float(loss), 4)`, using the right `torch.nn.functional` call.


<details><summary>💡 Hint (click to reveal)</summary>

`F.cross_entropy(logits, targets)` — it applies log-softmax internally.

</details>

## Reference solution


```python
import torch
import torch.nn.functional as F

def solve(logits, targets):
    loss = F.cross_entropy(logits, targets)
    return round(float(loss), 4)

logits = torch.tensor([[2.0, 0.5, -1.0], [0.1, 1.2, 3.0]])
targets = torch.tensor([0, 2])
print(solve(logits, targets))
```

## Why this works

`F.cross_entropy` fuses log-softmax and negative log-likelihood: you hand it raw logits and integer class targets. Softmaxing first is the classic bug — you'd be double-normalizing, and numerically much less stable than the fused version.
