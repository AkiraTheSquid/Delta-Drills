## Problem

Using torch, create a leaf parameter b = [2.0, -1.0] with requires_grad=True and target = [0.0, 0.0] (constant). Compute loss = ((b - target) ** 2).mean(), call loss.backward(). On the first line print the two grad components rounded to 4 decimals as: print(round(float(b.grad[0]), 4), round(float(b.grad[1]), 4)). Then, simulating one SGD step with lr = 0.1, compute updated = b - lr * b.grad under torch.no_grad() and on the second line print [round(float(v), 4) for v in updated].

## Reference solution


```python
import torch
b = torch.tensor([2.0, -1.0], requires_grad=True)
target = torch.tensor([0.0, 0.0])
loss = ((b - target) ** 2).mean()
loss.backward()
lr = 0.1
with torch.no_grad():
    updated = b - lr * b.grad
print(round(float(b.grad[0]), 4), round(float(b.grad[1]), 4))
print([round(float(v), 4) for v in updated])
```

## Why this works

For mean of (b-target)^2 over 2 elements, the gradient per component is (2/n)*(b_i - target_i) = b_i, so grad = [2.0, -1.0]. One SGD step b - 0.1*grad nudges each parameter toward the target: [1.8, -0.9].
