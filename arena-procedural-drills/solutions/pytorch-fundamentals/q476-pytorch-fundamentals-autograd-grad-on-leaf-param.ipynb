## Problem

The gradient of a MEAN divides by the element count — a classic off-by-factor trap. Write solve(b, target, lr) where leaf `b` requires grad: build loss = the mean squared error between `b` and `target` (elementwise (b−target)² then `.mean()`), backprop, then compute one SGD step `updated = b - lr * b.grad` inside `torch.no_grad()`. Return `(b.grad, updated)`.

## Reference solution


```python
import torch

def solve(b, target, lr):
    loss = ((b - target) ** 2).mean()
    loss.backward()
    with torch.no_grad():
        updated = b - lr * b.grad
    return (b.grad, updated)

b = torch.tensor([2.0, -1.0], requires_grad=True)
target = torch.tensor([0.0, 0.0])
print(solve(b, target, 0.1))
```

## Why this works

The mean divides by N, so b.grad = 2(b−target)/N — forget the /N and your gradient is N× too big (the off-by-factor trap). The SGD step must run under `torch.no_grad()` because in-place-updating a leaf that requires grad inside the graph is an autograd error.
