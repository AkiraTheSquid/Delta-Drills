## Problem

The gradient of a MEAN divides by the element count — a classic off-by-factor trap. The starter gives leaf `b` and constant `target`. Build loss = the mean squared error between them, backprop, and print the two gradient components with print(round(float(b.grad[0]), 4), round(float(b.grad[1]), 4)). Then take one SGD step with lr 0.1 inside the starter's `no_grad` block to compute `updated`, and print [round(float(v), 4) for v in updated].

## Reference solution


```python
import torch
b = torch.tensor([2.0, -1.0], requires_grad=True)
target = torch.tensor([0.0, 0.0])
loss = ((b - target) ** 2).mean()
loss.backward()
lr = 0.1
with torch.no_grad():
    updated = b - lr * b.grad
print(round(float(b.grad[0]), 4), round(float(b.grad[1]), 4))
print([round(float(v), 4) for v in updated])
```

## Why this works

For mean of (b-target)^2 over 2 elements, the gradient per component is (2/n)*(b_i - target_i) = b_i, so grad = [2.0, -1.0]. One SGD step b - 0.1*grad nudges each parameter toward the target: [1.8, -0.9].
