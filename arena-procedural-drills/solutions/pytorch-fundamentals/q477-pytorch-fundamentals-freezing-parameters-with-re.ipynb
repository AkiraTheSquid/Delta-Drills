## Problem

Build model = nn.Sequential(nn.Linear(3, 4), nn.ReLU(), nn.Linear(4, 2)). Call model.requires_grad_(False), which recursively turns off gradients on every parameter. Then, on one line, print the list [p.requires_grad for p in model.parameters()]. On a second line, print the number of trainable parameters: sum(p.numel() for p in model.parameters() if p.requires_grad).


<details><summary>💡 Hint (click to reveal)</summary>

The call freezes every parameter recursively; count only those whose requires_grad survives.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(3, 4),
    nn.ReLU(),
    nn.Linear(4, 2),
)
model.requires_grad_(False)
flags = [p.requires_grad for p in model.parameters()]
n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(flags)
print(n_trainable)
```

## Why this works

model.requires_grad_(False) recurses over every parameter in the module tree and clears its requires_grad flag. The Sequential has 4 parameter tensors (weight+bias per Linear; ReLU has none), all now frozen, so the trainable-parameter count is 0.
