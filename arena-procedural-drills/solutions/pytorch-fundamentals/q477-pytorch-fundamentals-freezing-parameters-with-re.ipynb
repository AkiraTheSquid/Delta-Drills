## Problem

Calling `requires_grad_(False)` on a MODULE recurses into every parameter. Write solve(model): freeze the given module with that single call, then return the tuple `(flags, n_trainable)` — `flags` = the list of `requires_grad` across `model.parameters()`, and `n_trainable` = the number of trainable parameter ELEMENTS remaining (weight the count by `numel()`, not per-tensor).

## Reference solution


```python
import torch
import torch.nn as nn

def solve(model):
    model.requires_grad_(False)
    flags = [p.requires_grad for p in model.parameters()]
    n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    return (flags, n_trainable)

model = nn.Sequential(nn.Linear(3, 4), nn.ReLU(), nn.Linear(4, 2))
print(solve(model))
```

## Why this works

`requires_grad_` on a MODULE walks every parameter recursively — no loop needed. After freezing, `sum(p.numel() for p in model.parameters() if p.requires_grad)` is 0: the count-by-elements form is what fine-tuning code logs as 'trainable params'.
