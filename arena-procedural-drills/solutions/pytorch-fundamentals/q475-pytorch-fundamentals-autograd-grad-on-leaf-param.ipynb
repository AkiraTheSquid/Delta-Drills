## Problem

A gradient always has the same shape as its parameter. The starter gives a matrix `W` (requires grad) and a constant vector `x`. Build the scalar loss: the sum of squares of the matvec $Wx$. Backprop, then print two lines: print(W.grad.shape == W.shape), then print(W.grad.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

The gradient of each weight is 2 times the corresponding pre-activation output times the input feature that multiplied it.

</details>

## Reference solution


```python
import torch
W = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True)
x = torch.tensor([1.0, 1.0])
loss = (W @ x).pow(2).sum()
loss.backward()
print(W.grad.shape == W.shape)
print(W.grad.tolist())
```

## Why this works

Gradients always match the shape of the leaf, so W.grad is (2,2) like W. With y = W@x, loss = sum(y^2), the gradient is 2*y_i*x_j per entry; here y = [3, 7] and x = [1, 1], giving rows [6,6] and [14,14].
