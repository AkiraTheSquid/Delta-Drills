## Problem

Using torch, create a leaf parameter matrix W = [[1.0, 2.0], [3.0, 4.0]] with requires_grad=True and a constant vector x = [1.0, 1.0]. Compute loss = (W @ x).pow(2).sum(), call loss.backward(). First print whether W.grad has the same shape as W via print(W.grad.shape == W.shape), then on the next line print W.grad.tolist().


<details><summary>💡 Hint (click to reveal)</summary>

The gradient of each weight is 2 times the corresponding pre-activation output times the input feature that multiplied it.

</details>

## Reference solution


```python
import torch
W = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True)
x = torch.tensor([1.0, 1.0])
loss = (W @ x).pow(2).sum()
loss.backward()
print(W.grad.shape == W.shape)
print(W.grad.tolist())
```

## Why this works

Gradients always match the shape of the leaf, so W.grad is (2,2) like W. With y = W@x, loss = sum(y^2), the gradient is 2*y_i*x_j per entry; here y = [3, 7] and x = [1, 1], giving rows [6,6] and [14,14].
