## Problem

A gradient always has the same shape as its parameter. Write solve(W, x) where matrix `W` requires grad and `x` is a constant vector: build the scalar loss — the sum of squares of the matvec W@x — backprop it, and return the tuple `(W.grad.shape == W.shape, W.grad)`.


<details><summary>💡 Hint (click to reveal)</summary>

loss = `(W @ x).pow(2).sum()`; a gradient always has its parameter's shape.

</details>

## Reference solution


```python
import torch

def solve(W, x):
    loss = (W @ x).pow(2).sum()
    loss.backward()
    return (W.grad.shape == W.shape, W.grad)

W = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True)
x = torch.tensor([1.0, 1.0])
print(solve(W, x))
```

## Why this works

Whatever the loss, `W.grad.shape == W.shape` — autograd accumulates one partial per parameter ELEMENT. Here d/dW of Σ(Wx)² = 2(Wx)xᵀ, an outer product with exactly W's shape.
