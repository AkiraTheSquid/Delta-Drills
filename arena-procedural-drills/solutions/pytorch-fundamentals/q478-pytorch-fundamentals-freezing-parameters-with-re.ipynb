## Problem

A frozen module's output is detached from the autograd graph. Write solve(model, x): freeze `model` with one recursive `requires_grad_(False)` call, run `x` through it and `.sum()` to get `y`. Return `(y.requires_grad, y.grad_fn is None)` — both tell the same story from different angles.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(model, x):
    model.requires_grad_(False)
    y = model(x).sum()
    return (y.requires_grad, y.grad_fn is None)

torch.manual_seed(0)
model = nn.Linear(4, 1)
x = torch.tensor([[1.0, 2.0, 3.0, 4.0]])
print(solve(model, x))
```

## Why this works

When no input to a computation requires grad, autograd doesn't record it: the output has `requires_grad=False` AND `grad_fn=None`. Two views of one fact — the tensor isn't connected to any graph, so `backward()` from it can't reach the weights.
