## Problem

A frozen module's output is detached from the autograd graph. Freeze the starter's `nn.Linear(4, 1)` with one recursive call, run `x` through it and sum to get `y`. Print two lines: print(y.requires_grad), then print(y.grad_fn is None) — both tell the same story from different angles.


<details><summary>💡 Hint (click to reveal)</summary>

If no input to an operation requires grad, the result carries neither a requires_grad flag nor a grad_fn.

</details>

## Reference solution


```python
import torch
import torch.nn as nn

torch.manual_seed(0)
model = nn.Linear(4, 1)
model.requires_grad_(False)
x = torch.tensor([[1.0, 2.0, 3.0, 4.0]])
y = model(x).sum()
print(y.requires_grad)
print(y.grad_fn is None)
```

## Why this works

Freezing all parameters means none of the forward op's inputs require grad, so the output is not attached to an autograd graph: y.requires_grad is False and it has no grad_fn. The seed only fixes the weight values and does not affect this.
