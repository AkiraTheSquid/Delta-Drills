## Problem

Using torch, implement a 2x2 max-pool with stride 2 over a single 4x4 feature map using amax over the (kh, kw) window dims. Build the fixed tensor of shape (1,1,4,4): x = [[[[1.,7.,2.,0.],[3.,4.,6.,5.],[8.,1.,9.,2.],[0.,4.,3.,7.]]]]. Extract 2x2 windows with x.unfold(2,2,2).unfold(3,2,2) (giving shape (1,1,2,2,2,2)), then take amax over dim=(4,5). Print the pooled result with print(result.tolist()).

## Reference solution


```python
import torch

x = torch.tensor([[[[1., 7., 2., 0.],
                    [3., 4., 6., 5.],
                    [8., 1., 9., 2.],
                    [0., 4., 3., 7.]]]])
x = x.unfold(2, 2, 2).unfold(3, 2, 2)
result = x.amax(dim=(4, 5))
print(result.tolist())
```

## Why this works

Two unfold calls tile the 4x4 map into non-overlapping 2x2 windows along the height and width axes, producing shape (1,1,2,2,2,2). Taking amax over the last two window dims is exactly 2x2 max-pooling with stride 2.
