## Problem

Implement a 2×2, stride-2 max-pool by hand on the starter's (1,1,4,4) tensor. Two steps: carve the two spatial dims into 2×2 windows with `.unfold` (called once per spatial dim — mind its size and step arguments), then reduce BOTH window dims away with a single `.amax` over the two trailing dims. Print the pooled result with print(result.tolist()).

## Reference solution


```python
import torch

x = torch.tensor([[[[1., 7., 2., 0.],
                    [3., 4., 6., 5.],
                    [8., 1., 9., 2.],
                    [0., 4., 3., 7.]]]])
x = x.unfold(2, 2, 2).unfold(3, 2, 2)
result = x.amax(dim=(4, 5))
print(result.tolist())
```

## Why this works

Two unfold calls tile the 4x4 map into non-overlapping 2x2 windows along the height and width axes, producing shape (1,1,2,2,2,2). Taking amax over the last two window dims is exactly 2x2 max-pooling with stride 2.
