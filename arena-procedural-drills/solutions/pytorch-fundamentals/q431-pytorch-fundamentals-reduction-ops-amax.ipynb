## Problem

Implement a 2×2, stride-2 max-pool by hand. Write solve(x) for a (1, 1, H, W) float tensor (H, W even): carve the two spatial dims into 2×2 windows with `.unfold` (called once per spatial dim — mind its size and step arguments), then reduce BOTH window dims away with a single `.amax` over the two trailing dims. Return the pooled (1, 1, H//2, W//2) tensor.


<details><summary>💡 Hint (click to reveal)</summary>

`.unfold(2, 2, 2).unfold(3, 2, 2)` carves 2×2 windows; then one `.amax(dim=(4, 5))`.

</details>

## Reference solution


```python
import torch

def solve(x):
    w = x.unfold(2, 2, 2).unfold(3, 2, 2)
    return w.amax(dim=(4, 5))

x = torch.tensor([[[[1., 7., 2., 0.],
                    [3., 4., 6., 5.],
                    [8., 1., 9., 2.],
                    [0., 4., 3., 7.]]]])
print(solve(x))
```

## Why this works

`.unfold(dim, size, step)` adds a trailing window axis per call: after unfolding both spatial dims the tensor is (1, 1, H/2, W/2, 2, 2). A single `.amax` over the two trailing window dims (4, 5) picks each window's max — exactly a 2×2/stride-2 max-pool.
