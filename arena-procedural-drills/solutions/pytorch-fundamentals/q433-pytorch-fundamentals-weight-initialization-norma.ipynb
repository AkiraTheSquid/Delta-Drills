## Problem

Using PyTorch, set torch.manual_seed(0), create a length-5 tensor with torch.empty, then apply torch.nn.init.normal_ in place with mean=2.0 and std=0.0. (A zero standard deviation makes every sampled value exactly equal to the mean.) Print the result with print(w.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Zero standard deviation makes every normal sample equal the mean.

</details>

## Reference solution


```python
import torch

torch.manual_seed(0)
w = torch.empty(5)
torch.nn.init.normal_(w, mean=2.0, std=0.0)
print(w.tolist())
```

## Why this works

Sampling from a normal distribution with std=0.0 collapses to a point mass at the mean, so every element equals 2.0 regardless of the RNG. The seed is irrelevant when the variance is zero.
