## Problem

Sampling initializers take distribution parameters. Fill the starter's empty length-5 tensor in place from a normal distribution with mean 2.0 and std 0.0 — a zero standard deviation makes every draw exactly the mean, which is why this output is deterministic. Print with print(w.tolist()).


<details><summary>💡 Hint (click to reveal)</summary>

Zero standard deviation makes every normal sample equal the mean.

</details>

## Reference solution


```python
import torch

torch.manual_seed(0)
w = torch.empty(5)
torch.nn.init.normal_(w, mean=2.0, std=0.0)
print(w.tolist())
```

## Why this works

Sampling from a normal distribution with std=0.0 collapses to a point mass at the mean, so every element equals 2.0 regardless of the RNG. The seed is irrelevant when the variance is zero.
