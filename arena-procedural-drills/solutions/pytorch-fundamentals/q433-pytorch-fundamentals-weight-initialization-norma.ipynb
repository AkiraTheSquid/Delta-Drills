## Problem

Sampling initializers take distribution parameters. Write solve(w, mean): fill `w` IN PLACE from a normal distribution with the given mean and std 0.0 (a zero standard deviation makes every draw exactly the mean, which is why this is deterministic), using `torch.nn.init.normal_`. Return `w` — the grader checks the original tensor object.


<details><summary>💡 Hint (click to reveal)</summary>

`torch.nn.init.normal_(w, mean=mean, std=0.0)` — std 0 makes every draw the mean.

</details>

## Reference solution


```python
import torch

def solve(w, mean):
    torch.nn.init.normal_(w, mean=mean, std=0.0)
    return w

w = torch.empty(5)
print(solve(w, 2.0))
```

## Why this works

`normal_` samples in place from N(mean, std). With std = 0.0 the distribution collapses to a point, so every element becomes exactly `mean` — deterministic, which is what makes this gradeable without seeding.
