## Problem

Partial unfreezing for fine-tuning. Write solve(model, idx) for a `nn.Sequential`: freeze the whole model recursively, then re-enable JUST the layer at position `idx` (index into the Sequential and call the same method with True). Return `(flags, n_on)` — the per-parameter `requires_grad` list and the count of parameter TENSORS still requiring grad.


<details><summary>💡 Hint (click to reveal)</summary>

Freeze all, then `model[idx].requires_grad_(True)` re-enables just that layer.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(model, idx):
    model.requires_grad_(False)
    model[idx].requires_grad_(True)
    flags = [p.requires_grad for p in model.parameters()]
    n_on = sum(int(p.requires_grad) for p in model.parameters())
    return (flags, n_on)

model = nn.Sequential(nn.Linear(2, 3), nn.Linear(3, 2))
print(solve(model, 1))
```

## Why this works

Freeze-then-selectively-unfreeze is the standard fine-tuning move: `model.requires_grad_(False)` flips everything off, indexing the Sequential gives the child module, and the same method with True flips just its parameters back on.
