## Problem

Partial unfreezing for fine-tuning: freeze the starter's two-layer Sequential recursively, then re-enable JUST the second layer (index into the Sequential and call the same method with True). Print two lines: the per-parameter `requires_grad` list, then the count of parameter TENSORS still requiring grad.

Freeze everything, then selectively unfreeze one submodule; the parameter order follows the module order.

## Reference solution


```python
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(2, 3),
    nn.Linear(3, 2),
)
model.requires_grad_(False)
model[1].requires_grad_(True)
flags = [p.requires_grad for p in model.parameters()]
n_on = sum(int(p.requires_grad) for p in model.parameters())
print(flags)
print(n_on)
```

## Why this works

Freezing the whole Sequential then re-enabling only model[1] leaves the first layer's two parameters off and the second layer's two parameters on. parameters() yields them in order, so flags are [False, False, True, True] and the count of trainable tensors is 2.
