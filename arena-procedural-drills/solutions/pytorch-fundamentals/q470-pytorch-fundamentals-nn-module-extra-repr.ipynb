## Problem

Define an nn.Module subclass named MyBlock whose __init__ stores channels and p as attributes and also creates a submodule self.lin = nn.Linear(channels, channels). Override extra_repr to return f"channels={self.channels}, p={self.p}". Construct MyBlock(3, 0.5) and print(repr(m)). Note the printed representation interleaves the extra_repr line with the child submodule line; print exactly that multi-line repr.


<details><summary>💡 Hint (click to reveal)</summary>

With a submodule present, the repr expands to multiple indented lines: your extra_repr first, then each child module.

## Reference solution


```python
import torch.nn as nn

class MyBlock(nn.Module):
    def __init__(self, channels, p):
        super().__init__()
        self.channels = channels
        self.p = p
        self.lin = nn.Linear(channels, channels)
    def extra_repr(self):
        return f"channels={self.channels}, p={self.p}"

m = MyBlock(3, 0.5)
print(repr(m))
```

## Why this works

When a module has child submodules, nn.Module's __repr__ switches to multi-line form: the extra_repr text is indented on its own line, followed by each named child's repr line, all wrapped between 'MyBlock(' and ')'. The registered nn.Linear child appears as '(lin): Linear(...)'.
