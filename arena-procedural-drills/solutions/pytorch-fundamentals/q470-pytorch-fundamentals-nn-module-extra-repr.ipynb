## Problem

When a module has BOTH an `extra_repr` and child submodules, the printed repr interleaves them. Write solve(channels, p): define `MyBlock(torch.nn.Module)` whose `__init__` stores `channels` and `p` as attributes and creates the child `self.lin = torch.nn.Linear(channels, channels)`; `extra_repr` returns `channels=..., p=...`. Return an instance — the grader checks the repr structure (your extra_repr line lands ABOVE the `(lin): ...` child line) and the child's weight shape.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(channels, p):
    class MyBlock(nn.Module):
        def __init__(self, c, prob):
            super().__init__()
            self.channels = c
            self.p = prob
            self.lin = nn.Linear(c, c)
        def extra_repr(self):
            return f"channels={self.channels}, p={self.p}"
    return MyBlock(channels, p)

m = solve(3, 0.5)
print(repr(m))
```

## Why this works

With both an `extra_repr` and children, the repr interleaves them: class name, then your extra_repr line, then one indented `(name): ...` line per child. Registering `self.lin = nn.Linear(c, c)` is what makes it a child — plain attribute assignment of a Module auto-registers it.
