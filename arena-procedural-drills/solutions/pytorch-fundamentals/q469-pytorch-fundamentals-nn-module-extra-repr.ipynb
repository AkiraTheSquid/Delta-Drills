## Problem

Where does `extra_repr` actually surface? Inside the module's repr. Write solve(dim, factor): define `Scale(torch.nn.Module)` storing `dim` and `factor` as attributes (super().__init__() first), with `extra_repr` returning `dim=..., factor=...` (f-string over the stored values). Return an instance built with the given arguments — the grader reads `repr()` of it, where your string shows up inside `Scale(...)`.


<details><summary>💡 Hint (click to reveal)</summary>

repr(m) becomes `Scale(dim=16, factor=2.5)` — your extra_repr fills the parens.

## Reference solution


```python
import torch
import torch.nn as nn

def solve(dim, factor):
    class Scale(nn.Module):
        def __init__(self, d, f):
            super().__init__()
            self.dim = d
            self.factor = f
        def extra_repr(self):
            return f"dim={self.dim}, factor={self.factor}"
    return Scale(dim, factor)

m = solve(16, 2.5)
print(repr(m))
```

## Why this works

For a module with no children, `repr(m)` is exactly `ClassName(<extra_repr>)`. That's the whole mechanism: the base class builds the skeleton, your `extra_repr` supplies the argument text.
