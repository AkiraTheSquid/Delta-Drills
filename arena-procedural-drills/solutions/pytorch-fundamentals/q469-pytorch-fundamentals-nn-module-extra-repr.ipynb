## Problem

Where does `extra_repr` actually surface? Inside the module's repr. Define `Scale(nn.Module)` storing `dim` and `factor` as attributes, with `extra_repr` returning `dim=..., factor=...` (f-string over the stored values). Construct `Scale(16, 2.5)` and print(repr(m)) — your string shows up inside `Scale(...)`.


<details><summary>💡 Hint (click to reveal)</summary>

With no submodules, the repr collapses to the class name wrapping your extra_repr string in parentheses.

</details>

## Reference solution


```python
import torch.nn as nn

class Scale(nn.Module):
    def __init__(self, dim, factor):
        super().__init__()
        self.dim = dim
        self.factor = factor
    def extra_repr(self):
        return f"dim={self.dim}, factor={self.factor}"

m = Scale(16, 2.5)
print(repr(m))
```

## Why this works

When a module has no child submodules, nn.Module's __repr__ places the extra_repr string inline inside the class-name parentheses. So repr(m) becomes 'Scale(<extra_repr>)' on a single line.
