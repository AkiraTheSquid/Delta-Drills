## Problem

BatchNorm2d computes PER-CHANNEL statistics over the batch and spatial dims. Write solve(x) for a (B, C, H, W) float tensor: return the tuple `(mean, var)` where `mean` is the per-channel mean and `var` the per-channel BIASED variance — reduce over exactly the dims that are not the channel dim. Both are length-C tensors.


<details><summary>💡 Hint (click to reveal)</summary>

Reduce over dims (0, 2, 3) — everything except the channel dim; var with `unbiased=False`.

</details>

## Reference solution


```python
import torch

def solve(x):
    mean = x.mean(dim=(0, 2, 3))
    var = x.var(dim=(0, 2, 3), unbiased=False)
    return (mean, var)

x = torch.arange(24, dtype=torch.float32).reshape(2, 3, 2, 2)
print(solve(x))
```

## Why this works

Per-channel statistics reduce over batch and both spatial dims, leaving one value per channel: `x.mean(dim=(0,2,3))` and `x.var(dim=(0,2,3), unbiased=False)`. Choosing the reduction dims by 'everything but channels' is the transferable habit.
