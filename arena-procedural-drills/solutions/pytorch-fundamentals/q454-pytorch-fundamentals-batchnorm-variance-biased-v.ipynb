## Problem

BatchNorm2d computes PER-CHANNEL statistics over the batch and spatial dims. On the starter's (2, 3, 2, 2) tensor, compute the per-channel mean and the BIASED variance — reduce over exactly the dims that are not the channel dim. Print mean.tolist() on the first line, then [round(v,4) for v in var.tolist()] on the second.

## Reference solution


```python
import torch
x = torch.arange(24, dtype=torch.float32).reshape(2, 3, 2, 2)
mean = x.mean(dim=(0, 2, 3))
var = x.var(dim=(0, 2, 3), unbiased=False)
print(mean.tolist())
print([round(v, 4) for v in var.tolist()])
```

## Why this works

BatchNorm2d computes per-channel statistics by reducing over the batch and both spatial dims, leaving one number per channel. Reducing over dims (0,2,3) collapses N, H, W and keeps C; using unbiased=False matches BatchNorm's biased variance.
