## Problem

BatchNorm2d normalizes per-channel over the batch and spatial dims. Build x = torch.arange(24, dtype=torch.float32).reshape(2,3,2,2) (N=2, C=3, H=2, W=2). For each of the 3 channels compute the per-channel mean and the BIASED variance (unbiased=False) over dims (0,2,3) — exactly what BatchNorm uses. First print mean.tolist(), then on a second line print the biased variance as [round(v,4) for v in var.tolist()].

## Reference solution


```python
import torch
x = torch.arange(24, dtype=torch.float32).reshape(2, 3, 2, 2)
mean = x.mean(dim=(0, 2, 3))
var = x.var(dim=(0, 2, 3), unbiased=False)
print(mean.tolist())
print([round(v, 4) for v in var.tolist()])
```

## Why this works

BatchNorm2d computes per-channel statistics by reducing over the batch and both spatial dims, leaving one number per channel. Reducing over dims (0,2,3) collapses N, H, W and keeps C; using unbiased=False matches BatchNorm's biased variance.
