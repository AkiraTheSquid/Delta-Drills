## Problem

Using PyTorch, set torch.manual_seed(0). Create a length-10000 weight tensor with torch.empty and initialize it in place with torch.nn.init.normal_(mean=0.0, std=1.0). Create a length-4 bias tensor with torch.empty and initialize it in place with torch.nn.init.constant_(value=0.0). Print, space-separated on one line: round(float(w.mean()),2), then round(float(w.std()),2), then b.tolist(). Use a single print statement: print(round(float(w.mean()), 2), round(float(w.std()), 2), b.tolist()).

## Reference solution


```python
import torch

torch.manual_seed(0)
w = torch.empty(10000)
torch.nn.init.normal_(w, mean=0.0, std=1.0)
b = torch.empty(4)
torch.nn.init.constant_(b, 0.0)
print(round(float(w.mean()), 2), round(float(w.std()), 2), b.tolist())
```

## Why this works

A large standard-normal sample has empirical mean near 0 and std near 1, and the constant init zeros the bias. Printing rounded statistics plus the bias list demonstrates both initializers at once.
