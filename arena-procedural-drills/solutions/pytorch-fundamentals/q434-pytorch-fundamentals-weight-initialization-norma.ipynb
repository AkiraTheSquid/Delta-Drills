## Problem

Write solve(n, c): call `torch.manual_seed(0)` FIRST, then create `w = torch.empty(n)` and fill it in place from a standard normal (mean 0, std 1) with `torch.nn.init.normal_`; then create `b = torch.empty(4)` and fill it in place with the constant `c`. With thousands of samples the empirical stats land close to the parameters. Return the tuple `(round(float(w.mean()), 2), round(float(w.std()), 2), b.tolist())`. Follow that exact order — seed, then w, then b — so the random draws are reproducible.


<details><summary>💡 Hint (click to reveal)</summary>

Seed first (`torch.manual_seed(0)`), then `normal_(w, 0.0, 1.0)`, then `constant_(b, c)`.

## Reference solution


```python
import torch

def solve(n, c):
    torch.manual_seed(0)
    w = torch.empty(n)
    torch.nn.init.normal_(w, mean=0.0, std=1.0)
    b = torch.empty(4)
    torch.nn.init.constant_(b, c)
    return (round(float(w.mean()), 2), round(float(w.std()), 2), b.tolist())

print(solve(10000, 0.0))
```

## Why this works

The exact order matters because `normal_` consumes random numbers: seed → fill w → fill b reproduces the same draws every run (constant_ consumes none). With 10k samples the empirical mean/std land within a couple of decimals of the distribution parameters — that's the law of large numbers in one print statement.
