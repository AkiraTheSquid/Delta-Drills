## Problem

Initialize two tensors in place with `torch.nn.init`: `w` (length 10000) from a standard normal — mean 0, std 1 — and `b` (length 4) to the constant 0. With 10k samples the empirical mean and std should land close to the distribution parameters; the starter's print line reports both (rounded to 2 decimals) plus `b.tolist()`.

## Reference solution


```python
import torch

torch.manual_seed(0)
w = torch.empty(10000)
torch.nn.init.normal_(w, mean=0.0, std=1.0)
b = torch.empty(4)
torch.nn.init.constant_(b, 0.0)
print(round(float(w.mean()), 2), round(float(w.std()), 2), b.tolist())
```

## Why this works

A large standard-normal sample has empirical mean near 0 and std near 1, and the constant init zeros the bias. Printing rounded statistics plus the bias list demonstrates both initializers at once.
