## Problem

Why is cross-entropy computed on logits directly? Because $\mathrm{CE} = \mathrm{logsumexp}(z) - z_{\mathrm{target}}$, and log-sum-exp can be evaluated stably even when exponentiating the logits would overflow — the starter's logits sit near 1000, where a naive softmax is all inf. Implement log-sum-exp the stable way (shift everything by the max before exponentiating), compute the loss, and print round(float(loss), 4).


<details><summary>💡 Hint (click to reveal)</summary>

Subtract the max logit before exponentiating, then take loss = logsumexp - target logit.

## Reference solution


```python
import numpy as np

logits = np.array([1000.0, 999.0, 998.0])
target = 0
m = np.max(logits)
logsumexp = m + np.log(np.sum(np.exp(logits - m)))
loss = logsumexp - logits[target]
print(round(float(loss), 4))
```

## Why this works

Cross-entropy equals logsumexp(logits) minus the target logit; computing logsumexp as m + log(sum(exp(logits - m))) subtracts the max first so the exponentials stay near 1 and never overflow. This stable form handles logits as large as 1000.
