## Problem

Write a function solve(logits, target) computing NUMERICALLY STABLE cross-entropy directly on logits: CE = logsumexp(logits) - logits[target], where logsumexp subtracts the max before exponentiating. Works even for logits around 1000 where naive softmax-then-log overflows. Return a float scalar.


<details><summary>💡 Hint (click to reveal)</summary>

Pull the max out of the logsumexp — the shifted exponentials stay tame and the max adds back linearly.

</details>

## Reference solution


```python
import numpy as np

def solve(logits, target):
    return (lambda m: float(m + np.log(np.sum(np.exp(logits - m))) - logits[target]))(np.max(logits))


print(solve(np.array([1.0, 2.0, 3.0]), 2))
```

## Why this works

logsumexp(z) = max + log(sum(exp(z - max))) is exact and overflow-free; CE = logsumexp(z) - z_target then never materializes softmax probabilities. The 1000-scale case would be inf - inf = NaN via naive log(softmax); done stably it's ~0.4076.
