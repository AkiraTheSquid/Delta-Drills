## Concept

After computing attention output per head, the standard multi-head attention implementation concatenates all heads by merging the `h` (heads) and `d` (per-head dimension) axes: `'b h t d -> b t (h d)'`. This produces the `(B, T, H*d)` tensor that feeds into the output projection, exactly as described in the Attention is All You Need architecture.

## Faded exercise 3

Complete `merge_heads`. Use `rearrange` to merge the heads and per-head dimension axes into a single model-dimension axis, moving the time axis back to position 2.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from einops import rearrange
import torch as t

def merge_heads(x):
    """x: (B, H, T, d_head) -> (B, T, H*d_head)"""
    return rearrange(x, 'b h t d -> b t (h d)')


def _test():
    import torch as t
    t.manual_seed(11)
    B, H, T, d = 2, 8, 10, 16
    x = t.randn(B, H, T, d)
    out = merge_heads(x)
    assert out.shape == (B, T, H * d)
    # Verify element content matches manual reshape
    # Transpose h and t first, then flatten h*d
    expected = x.permute(0, 2, 1, 3).reshape(B, T, H * d)
    assert t.allclose(out, expected)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
