## Concept

The `'b c h w -> b (c h w)'` pattern merges the three spatial-channel axes into one flat feature vector per batch element. This is the standard way to prepare a CNN feature map for a fully-connected head. All three inner axes are collapsed in row-major order: `c` varies slowest, `w` varies fastest.

## Faded exercise 1

Complete `flatten_chw`. Use `rearrange` to collapse the channel and spatial axes into one feature dimension.

**Fill in:** The `rearrange(x, 'b c h w -> b (c h w)')` call that flattens channels and spatial dimensions.

````
from einops import rearrange
import torch as t

def flatten_chw(x):
    return rearrange(x, 'b c h w -> b (c h w)')


def _test():
    import torch as t
    from einops import rearrange
    t.manual_seed(0)
    B, C, H, W = 4, 8, 6, 6
    x = t.randn(B, C, H, W)
    out = flatten_chw(x)
    assert out.shape == (B, C * H * W)
    # Verify element order matches nn.Flatten default
    assert t.allclose(out, x.flatten(start_dim=1))


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
