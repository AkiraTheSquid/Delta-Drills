## Setup

```
import numpy as np
import torch as t
from torch import Tensor
import einops
from einops import rearrange, reduce, repeat

t.manual_seed(0)
np.random.seed(0)
```

## Concept

Unbinding `dim=0` of a stacked `(T, H, W)` tensor returns a tuple of `T` separate `(H, W)` frames — the inverse of `t.stack`. Each returned slice is a view that shares storage with the source, so the split is free.

## Worked solution

We split a stack of frames along the time axis.

1. `stack.unbind(dim=0)` produces a tuple of `T` tensors, each of shape `(H, W)` — one frame per leading-axis position.
2. We confirm there are exactly `T` frames and that frame 0 matches `stack[0]` element-for-element.
3. Because each frame is a view, indexing into the tuple is cheap; we print the first frame to show the contents.

The printed frame count and the equality check confirm the leading-axis destructure.

```
import torch as t

t.manual_seed(0)
stack = t.randn(4, 2, 3)

def to_frames(stack: t.Tensor):
    return stack.unbind(dim=0)

frames = to_frames(stack)
print('num frames:', len(frames))
print('frame0 shape:', tuple(frames[0].shape))
print('frame0 matches:', t.equal(frames[0], stack[0]))
```
