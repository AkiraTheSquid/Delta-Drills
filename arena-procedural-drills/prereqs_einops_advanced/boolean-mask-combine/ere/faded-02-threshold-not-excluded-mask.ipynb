## Concept

To select entries that exceed a threshold but are not on an exclusion list, you AND a positive comparison with the negation of the exclusion flag. `~` flips the exclude-mask into a keep-mask; `&` intersects it with the threshold predicate. Parenthesise the comparison so `&` does not bind into it.
