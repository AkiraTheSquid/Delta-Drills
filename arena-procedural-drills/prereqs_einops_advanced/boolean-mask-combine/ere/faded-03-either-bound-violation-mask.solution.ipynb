## Concept

An out-of-bounds detector flags entries that violate EITHER bound: below the low limit OR above the high limit. This is a union of two comparison predicates, combined with `|`. Each comparison is parenthesised because `|` binds tighter than `<`/`>`.

## Faded exercise 3

### Faded — out-of-bounds via OR

Implement `out_of_bounds(x, lo, hi)`.

Given a `(N,)` tensor `x` and scalar limits `lo` and `hi` (with `lo < hi`), return a `(N,)` bool mask that is `True` where `x` violates EITHER bound — i.e. `x < lo` OR `x > hi`.

The two individual bound predicates are described in the task. **You must fill in the single line that combines a below-low comparison and an above-high comparison with OR** into the final mask.

**Fill in:** The bool mask that ORs the below-low comparison (x < lo) with the above-high comparison (x > hi).

````
import numpy as np
import torch as t
from torch import Tensor
import einops
from einops import rearrange, reduce, repeat
np.random.seed(0)
t.manual_seed(0)

def out_of_bounds(x: Tensor, lo: float, hi: float) -> Tensor:
    bad = (x < lo) | (x > hi)
    return bad

x = t.tensor([-1.0, 0.5, 2.0, 5.0])
print(out_of_bounds(x, 0.0, 3.0))


def _test():
    x = t.tensor([-1.0, 0.5, 2.0, 5.0, 0.0, 3.0])
    lo, hi = 0.0, 3.0
    got = out_of_bounds(x, lo, hi)
    expected = (x < lo) | (x > hi)
    assert got.dtype == t.bool, f'dtype {got.dtype}'
    assert got.shape == (x.shape[0],), f'shape {got.shape}'
    assert t.equal(got, expected), f'{got} vs {expected}'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
