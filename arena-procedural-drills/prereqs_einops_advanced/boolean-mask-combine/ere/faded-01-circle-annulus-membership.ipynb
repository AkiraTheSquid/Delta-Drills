## Concept

A point lies in an annulus (ring) when its squared distance from the center is at least the inner radius squared AND at most the outer radius squared. Building both bound predicates and AND-ing them gives the membership mask. Comparisons must be parenthesised because `&` binds tighter than `>=`/`<=`.
