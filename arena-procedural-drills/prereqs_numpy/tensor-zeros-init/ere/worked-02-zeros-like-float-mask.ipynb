## Setup

```
import numpy as np
import torch as t
from torch import Tensor

t.manual_seed(0)
np.random.seed(0)
```

## Concept

`t.zeros_like(x)` allocates a zero tensor that copies `x`'s shape, dtype, and device in one call — handy when the buffer must line up element-for-element with an existing tensor. Critically the result is a fresh allocation, not a view, so writing into it never mutates `x`.

## Worked solution

We are given a reference tensor `x` and need a matching zero buffer to write results into.

1. `t.zeros_like(x)` reads `x.shape`, `x.dtype`, and `x.device`, then allocates a new zero tensor with all three matched. We never have to restate the shape by hand.
2. We prove independence by writing a value into the buffer's first row and showing `x` is untouched — the buffer owns separate storage.
3. We print the buffer's shape and dtype next to `x`'s to confirm they agree.

This is the safe way to scaffold an accumulator that must broadcast against the input later.

```
import torch as t

t.manual_seed(0)
x = t.randn(2, 5)

def make_mirror(x: t.Tensor) -> t.Tensor:
    return t.zeros_like(x)

buf = make_mirror(x)
buf[0] = 9.0
print('buf shape == x shape:', buf.shape == x.shape)
print('buf dtype == x dtype:', buf.dtype == x.dtype)
print('x unchanged max:', x.abs().max().item() != 9.0)
print('buf row0 sum:', buf[0].sum().item())
```
