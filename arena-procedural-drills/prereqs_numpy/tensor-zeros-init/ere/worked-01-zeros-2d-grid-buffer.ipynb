## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

`t.zeros(shape, dtype=...)` allocates a fresh tensor filled with zeros. You pass the axis sizes either as positional arguments (`t.zeros(h, w)`) or as a single tuple (`t.zeros((h, w))`). The default dtype is `torch.float32`, which is exactly what you want for an empty accumulator you will later write floating-point values into.

## Worked solution

We want a 2-D float buffer of shape `(rows, cols)` that starts at all zeros.

1. We call `t.zeros(rows, cols)`. Passing the two sizes positionally is the most common idiom; PyTorch reads them as the shape.
2. Because we did not pass a `dtype`, the buffer is `torch.float32` — the right default for a grid that will hold real-valued data.
3. The result is a brand-new tensor that owns its own storage, so writing into it later cannot corrupt any other tensor.

The printed `shape` and `dtype` confirm we got a `(3, 4)` float buffer, and the sum confirms every entry really is zero.

```
import torch as t

def allocate_grid(rows: int, cols: int) -> t.Tensor:
    return t.zeros(rows, cols)

buf = allocate_grid(3, 4)
print('shape:', tuple(buf.shape))
print('dtype:', buf.dtype)
print('sum:', buf.sum().item())
```
