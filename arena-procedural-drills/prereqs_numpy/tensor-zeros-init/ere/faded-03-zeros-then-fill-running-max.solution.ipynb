## Faded exercise 3

### Fill a buffer with per-column maxima

Implement `column_maxima(x)` where `x` is `(rows, cols)`:
1. Allocate a `(cols,)` zero buffer with the same dtype as `x`.
2. Fill it with the maximum of each column of `x`.

Complete the reduction that produces the per-column maxima written into the buffer.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def column_maxima(x: t.Tensor) -> t.Tensor:
    out = t.zeros(x.shape[1], dtype=x.dtype)
    out[:] = x.max(dim=0).values
    return out

t.manual_seed(1)
x = t.randn(4, 3)
print(column_maxima(x))


def _test():
    x = t.tensor([
        [1.0, -2.0, 3.0],
        [4.0,  0.0, 1.0],
        [2.0,  5.0, 0.0],
    ])
    out = column_maxima(x)
    assert out.shape == (3,), out.shape
    assert out.dtype == x.dtype
    # independent ground truth: max per column computed by hand
    expected = t.tensor([4.0, 5.0, 3.0])
    assert t.allclose(out, expected), out


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
