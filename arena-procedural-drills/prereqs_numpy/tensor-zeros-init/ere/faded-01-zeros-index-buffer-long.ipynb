## Concept

`t.zeros(n, dtype=t.long)` allocates an integer zero buffer suitable for use as indices. PyTorch's `gather` / `index_select` / advanced indexing reject float indices, so any buffer that will hold positions must be an integer dtype (`t.long` = int64).
