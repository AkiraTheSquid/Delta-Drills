## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Unlike a sliding window (which overlaps with stride 1 along the row axis), a *block* partition reads each chunk once. For length-`b` non-overlapping blocks of a 1-D contiguous tensor, advancing one block means jumping `b` source elements, while advancing within a block jumps 1. So the two output strides are `(b * s, s)` where `s = x.stride(0)`.

## Faded exercise 1

### Faded — non-overlapping blocks via as_strided

Implement `blocks_1d(x, b)` to reshape a 1-D contiguous tensor of length `N` (with `N` divisible by `b`) into a zero-copy `(N // b, b)` view where row `i` is the block `x[i*b : i*b + b]`.

The shape and the call are written for you. Complete the **stride** passed to `as_strided`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def blocks_1d(x: Tensor, b: int) -> Tensor:
    n = x.shape[0]
    s = x.stride(0)
    stride = (b * s, s)
    return t.as_strided(x, size=(n // b, b), stride=stride)


def _test():
    x = t.arange(12, dtype=t.float32)
    b = 3
    out = blocks_1d(x, b)
    assert tuple(out.shape) == (4, 3), out.shape
    ref = x.reshape(4, 3)
    assert t.equal(out, ref), (out, ref)
    # zero-copy: must alias source storage
    assert out.data_ptr() == x.data_ptr()
    # spot-check block boundaries
    assert out[0].tolist() == [0.0, 1.0, 2.0]
    assert out[3].tolist() == [9.0, 10.0, 11.0]


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
