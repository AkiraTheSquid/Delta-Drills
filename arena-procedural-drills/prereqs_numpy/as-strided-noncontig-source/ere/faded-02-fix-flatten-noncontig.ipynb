`.view()` only works on contiguous storage; calling it on a non-contiguous view (like a transpose) raises `RuntimeError`. The fix is to materialize a fresh row-major copy with `.contiguous()` first, *then* `.view(-1)`. `.reshape(-1)` does this copy automatically when needed.

def flatten_transpose(x: Tensor) -> Tensor:
    contig = x.T.contiguous()
    return contig.view(-1)
