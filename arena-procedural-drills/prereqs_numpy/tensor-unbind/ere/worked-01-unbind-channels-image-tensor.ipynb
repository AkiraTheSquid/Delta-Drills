## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

`torch.unbind(x, dim=d)` peels a tensor apart along axis `d`, returning a Python tuple of views — one per index along that axis. For an image tensor of shape `(H, W, C)`, `unbind(dim=-1)` or `unbind(dim=2)` separates it into C tensors each of shape `(H, W)`. This is cleaner than slicing `x[:, :, 0]`, `x[:, :, 1]`, `x[:, :, 2]` manually.

## Worked solution

**Step 1 — Understand the input shape.**
An `(H, W, 3)` tensor has height, width, and three colour channels as its three axes. Axis 2 (the last) is the channel axis.

**Step 2 — Unbind along the channel axis.**
`torch.unbind(img, dim=-1)` returns a tuple `(R, G, B)` where each is an `(H, W)` tensor.

**Step 3 — Named destructuring.**
`r, g, b = img.unbind(dim=-1)` assigns each channel to a meaningful name. This makes subsequent computation readable: `gray = 0.299 * r + 0.587 * g + 0.114 * b`.

**Step 4 — Verify result shapes.**
Each of `r.shape`, `g.shape`, `b.shape` should be `(H, W)` — the channel axis is removed by unbind.

```
import torch as t

t.manual_seed(0)
H, W = 4, 6
img = t.rand(H, W, 3)      # (H, W, 3) RGB image, values in [0, 1]

# Unbind along the last axis (channels)
r, g, b = img.unbind(dim=-1)
print('img shape:', img.shape)   # (4, 6, 3)
print('r shape:', r.shape)       # (4, 6)
print('g shape:', g.shape)       # (4, 6)
print('b shape:', b.shape)       # (4, 6)

# Compute grayscale luma
gray = 0.299 * r + 0.587 * g + 0.114 * b
print('gray shape:', gray.shape)  # (4, 6)
print('gray[0, :3]:', gray[0, :3])

# Verify: same as manual indexing
assert t.allclose(r, img[:, :, 0])
assert t.allclose(g, img[:, :, 1])
assert t.allclose(b, img[:, :, 2])
print('All channel splits match manual indexing: True')
```
