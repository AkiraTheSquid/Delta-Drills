## Concept

The Y-axis rotation matrix R_y(θ) keeps the Y axis fixed while spinning points in the X-Z plane. The right-hand convention means +X rotates toward -Z for positive θ. The matrix entries are [[cos θ, 0, sin θ], [0, 1, 0], [-sin θ, 0, cos θ]].

## Worked solution

**Step 1 — Compute cos and sin.** We extract the scalar values with `.item()` so we can pass them to `t.tensor(...)` without autograd-graph baggage.

**Step 2 — Assemble the matrix.** The first and third rows carry the X-Z rotation; the second row is [0, 1, 0] because the Y axis itself is the rotation axis and is unaffected.

**Step 3 — Apply to the three standard basis vectors.** At θ = π/2: +X = (1,0,0) should map to (0,0,-1) (rotated into -Z), +Y = (0,1,0) stays (0,1,0), and +Z = (0,0,1) should map to (1,0,0) (rotated into +X). Checking these numerically confirms the formula is correct.

**Step 4 — Print the results.** We print both the matrix and the three transformed vectors so we can visually verify the rotation.

```
import torch as t
import math

t.manual_seed(42)

def ry_unit_axes(theta_deg: float):
    theta = t.tensor(math.radians(theta_deg))
    c = t.cos(theta).item()
    s = t.sin(theta).item()
    R = t.tensor([
        [ c,  0.0, s],
        [0.0, 1.0, 0.0],
        [-s,  0.0, c],
    ])
    x_hat = t.tensor([1.0, 0.0, 0.0])
    y_hat = t.tensor([0.0, 1.0, 0.0])
    z_hat = t.tensor([0.0, 0.0, 1.0])
    rx = R @ x_hat
    ry = R @ y_hat
    rz = R @ z_hat
    return R, rx, ry, rz

R, rx, ry, rz = ry_unit_axes(90.0)
print('R_y(90°):', R)
print('+X -> ', rx.round())    # expect (0, 0, -1)
print('+Y -> ', ry.round())    # expect (0, 1,  0)
print('+Z -> ', rz.round())    # expect (1, 0,  0)
```
