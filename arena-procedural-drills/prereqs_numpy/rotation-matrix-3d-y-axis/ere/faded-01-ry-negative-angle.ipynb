## Concept

A rotation by -θ is the inverse of a rotation by +θ. For R_y, this means R_y(-θ) = R_y(θ)^T = R_y(θ)^{-1}. You can verify this numerically: composing R_y(θ) with R_y(-θ) should give the identity matrix.
