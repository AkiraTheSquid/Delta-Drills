To create an animation or a multi-view dataset, you often need to rotate the same point cloud at many different angles. Building each rotation matrix separately, applying it, and stacking the results with `t.stack` is the standard pattern. The output shape is (M, N, 3) where M is the number of angles and N is the number of points.

## Faded exercise 3

Given a single 3D point `p = (2, 0, 1)` and angles `[0°, 45°, 90°, 135°, 180°]`, return a (5,) tensor of the X-coordinates of the rotated point at each angle.

1. Loop over each angle, build R_y, apply to `p`.
2. Collect the X coordinate of each rotated version.
3. The blank step is stacking the X coordinates into one tensor.

The blank step is combining all per-angle X-coordinate scalars into a single 1-D tensor.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import math

t.manual_seed(5)

def make_ry(deg: float) -> t.Tensor:
    theta = math.radians(deg)
    c, s = math.cos(theta), math.sin(theta)
    return t.tensor([[c, 0.0, s], [0.0, 1.0, 0.0], [-s, 0.0, c]], dtype=t.float32)

p = t.tensor([2.0, 0.0, 1.0])
angles_deg = [0.0, 45.0, 90.0, 135.0, 180.0]

x_coords_list = []
for deg in angles_deg:
    R = make_ry(deg)
    rotated_p = R @ p
    x_coords_list.append(rotated_p[0])

x_coords = t.stack(x_coords_list)

print('X coords at each angle:', x_coords.tolist())
print('At 0°: x should equal original x=2.0, got:', x_coords[0].item())
print('At 90°: x should be ~ +z=1.0, got:', x_coords[2].item())


def _test():
    import torch as t
    import math

    def make_ry(deg):
        theta = math.radians(deg)
        c, s = math.cos(theta), math.sin(theta)
        return t.tensor([[c, 0.0, s], [0.0, 1.0, 0.0], [-s, 0.0, c]], dtype=t.float32)

    p = t.tensor([2.0, 0.0, 1.0])
    angles_deg = [0.0, 45.0, 90.0, 135.0, 180.0]
    expected = t.stack([make_ry(d) @ p for d in angles_deg])[:, 0]
    assert x_coords.shape == (5,), f'shape: {x_coords.shape}'
    assert t.allclose(x_coords, expected, atol=1e-5), f'values: {x_coords} vs {expected}'
    # Specific checks: at 0 deg x=2, at 90 deg x=sin(90)*z_orig=1*1=1
    assert abs(x_coords[0].item() - 2.0) < 1e-4
    assert abs(x_coords[2].item() - 1.0) < 1e-4


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
