## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

`.item()` maps the tensor's dtype to a Python type: float tensors (`float32`, `float64`) return a Python `float`; integer tensors (`int32`, `int64`) return a Python `int`; boolean tensors return a Python `bool`. This matters when the returned value is used in context that checks type exactly, such as JSON serialisation or asserting strict equality.

## Worked solution

**Step 1 — Float tensors.**
`torch.tensor(3.14, dtype=t.float32).item()` returns `3.140000104904175` as a Python `float`. The slight precision change is due to float32's limited mantissa.

**Step 2 — Integer tensors.**
`torch.tensor(7, dtype=t.int64).item()` returns `7` as a Python `int`. No fractional part, no rounding.

**Step 3 — Boolean tensors.**
`torch.tensor(True).item()` returns `True` as a Python `bool`. Useful for control flow without triggering PyTorch's multi-element tensor bool warning.

**Step 4 — Check with `type().__name__`.**
Calling `type(x.item()).__name__` returns `'float'`, `'int'`, or `'bool'` respectively — handy for assertions and logging.

```
import torch as t

def describe_scalar(x: t.Tensor):
    """Show dtype -> Python type mapping for .item()."""
    val = x.item()
    python_type = type(val).__name__
    print(f'dtype={x.dtype}, .item() type={python_type}, value={val}')
    return val, python_type

# float32 -> float
describe_scalar(t.tensor(2.718, dtype=t.float32))
# float64 -> float
describe_scalar(t.tensor(2.718, dtype=t.float64))
# int32 -> int
describe_scalar(t.tensor(42, dtype=t.int32))
# int64 -> int
describe_scalar(t.tensor(100, dtype=t.int64))
# bool -> bool
describe_scalar(t.tensor(False))
```
