Inside a training loop, appending the tensor `loss` directly to a list would keep the entire computation graph alive for every step. To avoid memory leaks, call `.item()` to extract a plain Python float before appending. The graph is then freed after `.backward()` and the optimizer step, keeping memory constant across steps.

import torch as t
import torch.nn as nn

def run_loop(model, X, y, n_steps=8):
    opt = t.optim.SGD(model.parameters(), lr=0.01)
    log = []
    for _ in range(n_steps):
        opt.zero_grad()

        log.append(loss.item())
