## Faded exercise 2

### Faded — coordinate grid via broadcast

Implement `coord_sum_grid(rows, cols)`: given 1-D tensors `rows` of shape `(N,)` and `cols` of shape `(M,)`, return a grid `G` of shape `(N, M)` where `G[i, j] == rows[i] + cols[j]`.

Use broadcasting only — no loops, no `meshgrid`. The scaffold returns the sum of a reshaped `rows` and a reshaped `cols`. Fill in the reshape of `rows` into a column so it broadcasts against the row vector.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def coord_sum_grid(rows, cols):
    col = rows[:, None]            # (N, 1)
    row = cols[None, :]            # (1, M)
    return col + row               # (N, M)


def _test():
    t.manual_seed(0)
    rows = t.randn(4)
    cols = t.randn(6)
    G = coord_sum_grid(rows, cols)
    assert G.shape == (4, 6), G.shape
    ref = rows[:, None] + cols[None, :]
    assert t.allclose(G, ref, atol=1e-6)
    # Spot-check the defining property at a couple of entries.
    assert t.allclose(G[2, 5], rows[2] + cols[5], atol=1e-6)
    assert t.allclose(G[0, 3], rows[0] + cols[3], atol=1e-6)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
