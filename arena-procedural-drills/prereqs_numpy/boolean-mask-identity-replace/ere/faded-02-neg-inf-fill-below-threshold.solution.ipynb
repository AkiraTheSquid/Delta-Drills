`masked_fill` returns a copy with flagged positions overwritten by a scalar, broadcasting the mask over any leading batch axes. Filling masked-out logits with `float('-inf')` makes them vanish under a subsequent softmax — the standard way to forbid positions in attention or top-k filtering.

## Faded exercise 2

### Forbid low-score positions with -inf

Implement `mask_low_scores(scores, thresh)` for a `(B, T)` tensor `scores`: return a copy where every entry **strictly below** `thresh` is set to `float('-inf')`, and entries `>= thresh` are unchanged. Use `masked_fill` (do not mutate `scores`).

The function signature and return are given. Complete the line that produces the masked tensor.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def mask_low_scores(scores: Tensor, thresh: float) -> Tensor:
    out = scores.masked_fill(scores < thresh, float('-inf'))
    return out


def _test():
    t.manual_seed(0)
    scores = t.randn(2, 6)
    thresh = 0.0
    out = mask_low_scores(scores, thresh)
    below = scores < thresh
    assert t.isinf(out[below]).all() and (out[below] < 0).all(), 'below-thresh not -inf'
    assert t.equal(out[~below], scores[~below]), 'above-thresh entries changed'
    assert t.equal(scores, scores), 'input mutated'
    assert out.shape == scores.shape
    assert below.any() and (~below).any(), 'fixture should straddle the threshold'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
