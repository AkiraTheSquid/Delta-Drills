`EnableGrad` is the companion of `NoGrad`: it saves the previous flag value and SETS it to `True`, so it can temporarily cancel an outer `NoGrad`. Restoration on exit returns to the outer state.

## Faded exercise 3

Complete `EnableGrad.__enter__` so it sets the global flag to True (after saving the previous value). Fill in the enable line.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
grad_tracking_enabled = True

class EnableGrad:
    def __enter__(self):
        global grad_tracking_enabled
        self._prev = grad_tracking_enabled
        grad_tracking_enabled = True
        return self
    def __exit__(self, exc_type, exc_val, exc_tb):
        global grad_tracking_enabled
        grad_tracking_enabled = self._prev

grad_tracking_enabled = False
with EnableGrad():
    print(grad_tracking_enabled)
print(grad_tracking_enabled)


def _test():
    global grad_tracking_enabled
    # cancels an outer disabled state, then restores it
    grad_tracking_enabled = False
    with EnableGrad():
        assert grad_tracking_enabled is True
    assert grad_tracking_enabled is False
    # works on its own from True
    grad_tracking_enabled = True
    with EnableGrad():
        assert grad_tracking_enabled is True
    assert grad_tracking_enabled is True


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
