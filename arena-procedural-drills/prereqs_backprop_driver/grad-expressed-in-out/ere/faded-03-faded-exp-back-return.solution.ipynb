## Concept

For `out = exp(x)`, the derivative is `d/dx exp(x) = exp(x) = out`. This is the simplest possible 'grad in out' case: the local Jacobian is just the output itself. By the chain rule, `dL/dx = grad_out * out`. No extra computation — the forward result is the backward multiplier.

## Faded exercise 3

Complete `exp_back` by filling in the return expression. The function should use only `grad_out` and `out`.

**Fill in:** The return expression for the exponential backward: grad_out multiplied by the local Jacobian, which equals out.

````
import torch as t

t.manual_seed(0)

def exp_back(grad_out: t.Tensor, out: t.Tensor, x: t.Tensor) -> t.Tensor:
    return grad_out * out

# Exercise it
t.manual_seed(0)
x = t.randn(4)
out = t.exp(x)
grad_out = t.ones_like(x)
result = exp_back(grad_out, out, x)
print(result)  # should equal exp(x)


def _test():
    import torch as t
    t.manual_seed(77)
    x = t.randn(6)
    out = t.exp(x)
    grad_out = t.randn(6)

    result = exp_back(grad_out, out, x)

    expected = grad_out * t.exp(x)
    assert t.allclose(result, expected, atol=1e-6), f"Expected {expected}, got {result}"

    # x-independence
    wrong_x = t.randn_like(x) * 100
    assert t.allclose(exp_back(grad_out, out, wrong_x), result, atol=1e-6), \
        "exp_back must use only out, ignoring x"


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
