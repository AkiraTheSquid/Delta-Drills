## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

The hyperbolic tangent has a derivative that factors neatly: `d/dx tanh(x) = 1 - tanh(x)^2 = 1 - out^2`. By the chain rule, `dL/dx = grad_out * (1 - out^2)`. Just as with sigmoid, this form avoids recomputing the activation — `out` is already the cached forward result, so the backward pass requires no expensive calls.

## Faded exercise 2

Complete `tanh_back` by filling in the return expression. The function must use only `grad_out` and `out`, not `x` or any call to `t.tanh`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

t.manual_seed(0)

def tanh_back(grad_out: t.Tensor, out: t.Tensor, x: t.Tensor) -> t.Tensor:
    return grad_out * (1 - out ** 2)

# Exercise it
t.manual_seed(0)
x = t.randn(4)
out = t.tanh(x)
grad_out = t.ones_like(x)
result = tanh_back(grad_out, out, x)
print(result)


def _test():
    import torch as t
    t.manual_seed(55)
    x = t.randn(5)
    out = t.tanh(x)
    grad_out = t.randn(5)

    result = tanh_back(grad_out, out, x)

    # Ground truth
    expected = grad_out * (1.0 - t.tanh(x) ** 2)
    assert t.allclose(result, expected, atol=1e-6), f"Expected {expected}, got {result}"

    # Must not use x: wrong x, correct out -> same answer
    wrong_x = t.zeros_like(x)
    assert t.allclose(tanh_back(grad_out, out, wrong_x), result, atol=1e-6), \
        "tanh_back must use only out, not x"


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
