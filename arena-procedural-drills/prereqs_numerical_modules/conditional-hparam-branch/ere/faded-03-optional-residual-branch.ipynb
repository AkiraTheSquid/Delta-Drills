## Concept

A residual connection adds the block's input back to its output: `out = fc(x) + x`. Whether to add it is a constructor flag stored as `self.use_residual`. The forward pass gates the add behind `if self.use_residual`, so disabling residuals makes the block a plain linear layer.

class GatedResidual(t.nn.Module):
    def __init__(self, d, use_residual):
        super().__init__()
        self.fc = t.nn.Linear(d, d)
        self.use_residual = use_residual

    def forward(self, x):
        out = self.fc(x)
        if self.use_residual:
            out = out + x
        return out
