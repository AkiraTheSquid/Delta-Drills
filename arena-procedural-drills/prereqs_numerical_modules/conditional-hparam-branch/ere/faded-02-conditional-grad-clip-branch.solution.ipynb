## Concept

Gradient clipping rescales the gradient so its norm never exceeds `max_norm`. A `max_norm` of 0 (or non-positive) is the convention for 'no clipping', so the entire clip computation sits behind `if max_norm > 0`. Inside, we only rescale when the actual norm exceeds the threshold.

## Faded exercise 2

Implement `maybe_clip(grad, max_norm)`:

1. **If `max_norm > 0`:** compute the gradient norm; if it exceeds `max_norm`, rescale `grad` by `max_norm / norm` so the clipped norm equals `max_norm`.
2. **Else:** return `grad` unchanged.
3. Return the (possibly rescaled) gradient.

Complete the blanked rescaling step inside the `norm > max_norm` case.

**Fill in:** the rescaled gradient equal to grad times max_norm divided by its current norm

````
def maybe_clip(grad, max_norm):
    if max_norm > 0:
        norm = grad.norm()
        if norm > max_norm:
            grad = grad * (max_norm / norm)
    return grad


def _test():
    t.manual_seed(0)
    grad = t.tensor([3.0, 4.0])  # norm == 5
    g_off = maybe_clip(grad.clone(), 0.0)
    assert t.allclose(g_off, grad), 'max_norm=0 must be a no-op'
    g_clip = maybe_clip(grad.clone(), 1.0)
    assert abs(g_clip.norm().item() - 1.0) < 1e-5, 'clipped norm must equal max_norm'
    assert t.allclose(g_clip, grad * (1.0 / 5.0)), 'rescale factor wrong'
    g_big = maybe_clip(grad.clone(), 100.0)
    assert t.allclose(g_big, grad), 'norm below threshold must be unchanged'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
