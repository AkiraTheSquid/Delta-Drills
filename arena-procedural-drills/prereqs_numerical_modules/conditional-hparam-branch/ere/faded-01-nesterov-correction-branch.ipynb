## Concept

Nesterov momentum looks ahead: after updating the velocity buffer it forms the step as `grad + momentum * velocity` instead of just `velocity`. When `momentum == 0` there is no look-ahead and the update must collapse to the raw gradient. The look-ahead correction therefore lives behind an `if momentum != 0` branch.

## Faded exercise 1

Implement `momentum_update(velocity, grad, momentum)`. It must:

1. Update the buffer: `velocity = momentum * velocity + grad`.
2. **If `momentum != 0`:** form the Nesterov update `update = grad + momentum * velocity`.
3. **Else (`momentum == 0`):** the update is just `grad`.
4. Return `(update, velocity)`.

Complete the blanked `update` computation in the nonzero-momentum branch.

**Fill in:** the Nesterov-corrected update equal to grad plus momentum times the freshly updated velocity buffer
