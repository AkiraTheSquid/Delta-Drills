A weight vector built with `t.linspace(...)` defaults to CPU `float32`. To dot it against a tensor on another device/dtype it must be constructed with `device=x.device, dtype=x.dtype`. Otherwise the elementwise multiply raises a device error or upcasts the operand.

## Faded exercise 2

### Faded — linspace weights

Implement `weighted_sum(x)` for `x` of shape `(n,)`. It builds weights `linspace(0, 1, n)` and returns `(weights * x).sum()`. The weights MUST live on `x`'s device with `x`'s dtype. Complete the blanked construction of the weight vector.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def weighted_sum(x):
    n = x.shape[0]
    weights = t.linspace(0, 1, n, device=x.device, dtype=x.dtype)
    return (weights * x).sum()


def _test():
    for dt in (t.float32, t.float64):
        t.manual_seed(0)
        x = t.randn(6).to(dt)
        out = weighted_sum(x)
        assert out.dtype == dt, (out.dtype, dt)
        assert out.device == x.device
        w = t.linspace(0, 1, 6, device=x.device, dtype=dt)
        expected = (w * x).sum()
        assert t.allclose(out, expected)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
