Without an input tensor in hand, a Module can read its current device/dtype from `next(self.parameters())`. Allocating scratch with `device=ref.device, dtype=ref.dtype` makes it track the module after a `.to(...)` move, instead of hardcoding CPU/float32.

## Faded exercise 3

### Faded — scratch from a parameter

Implement a Module with a `nn.Linear(4, 2)` and a `make_scratch(self, shape)` method that returns a zero tensor matching the module's current device and dtype (read from its first parameter). Complete the blanked line that reads the reference parameter's device and dtype.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch.nn as nn

class _Scratch(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(4, 2)

    def make_scratch(self, shape):
        ref = next(self.parameters())
        device, dtype = ref.device, ref.dtype
        return t.zeros(shape, device=device, dtype=dtype)

def make_scratch_module():
    return _Scratch()


def _test():
    mod = make_scratch_module()
    s = mod.make_scratch((3, 5))
    ref = next(mod.parameters())
    assert s.shape == (3, 5)
    assert s.device == ref.device
    assert s.dtype == ref.dtype
    assert t.count_nonzero(s).item() == 0
    mod2 = make_scratch_module().to(t.float64)
    s2 = mod2.make_scratch((2,))
    assert s2.dtype == t.float64, s2.dtype


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
