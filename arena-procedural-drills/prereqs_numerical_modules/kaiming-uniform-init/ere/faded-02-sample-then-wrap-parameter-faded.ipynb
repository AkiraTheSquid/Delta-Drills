## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Calling in-place methods like `.uniform_()` directly on an `nn.Parameter` that has `requires_grad=True` raises a `RuntimeError`. The safe pattern is to sample into a plain tensor first (which has `requires_grad=False`) and then pass it to `nn.Parameter(...)`. The resulting Parameter carries the sampled values and enables gradients.

```python
import torch as t
import torch.nn as nn
import math

def init_weight_kaiming(in_features: int, out_features: int) -> nn.Parameter:
    bound = 1.0 / math.sqrt(in_features)
    w_init = t.empty(out_features, in_features).uniform_(-bound, bound)
    param = nn.Parameter(w_init)
    return param
```
