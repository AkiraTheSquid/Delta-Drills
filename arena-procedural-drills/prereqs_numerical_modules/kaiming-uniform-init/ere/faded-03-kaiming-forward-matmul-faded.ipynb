## Concept

The forward computation of a `Linear` layer is an affine map: multiply the input by the transposed weight matrix, then add the bias. The transpose is necessary because the weight is stored as `(out_features, in_features)`, so `weight.T` has shape `(in_features, out_features)`, aligning with the `in_features` dimension of the input.
