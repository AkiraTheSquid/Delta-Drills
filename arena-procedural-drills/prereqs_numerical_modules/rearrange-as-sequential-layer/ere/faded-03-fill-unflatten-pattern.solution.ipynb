To reshape a flat latent into a spatial map inside `nn.Sequential`, use `Rearrange('b (c h w) -> b c h w', c=..., h=..., w=...)`. The named axis sizes tell einops how to split the flat dimension.

## Faded exercise 3

Complete the decoder head's unflatten layer so it reshapes the flat vector into (B, c, h, w). Fill in the Rearrange layer with its axis sizes.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
from einops.layers.torch import Rearrange

t.manual_seed(8)

def decoder_head(latent_dim, channels, height, width):
    unflatten = Rearrange('b (c h w) -> b c h w', c=channels, h=height, w=width)
    return t.nn.Sequential(
        t.nn.Linear(latent_dim, channels * height * width),
        unflatten,
        t.nn.ReLU(),
    )

print(decoder_head(16, 4, 7, 7)(t.randn(3, 16)).shape)


def _test():
    t.manual_seed(88)
    model = decoder_head(16, 4, 7, 7)
    out = model(t.randn(3, 16))
    assert tuple(out.shape) == (3, 4, 7, 7)
    assert type(model) is t.nn.Sequential
    # ReLU at the end -> all outputs non-negative
    assert (out >= 0).all().item()


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
