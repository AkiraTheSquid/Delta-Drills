Flattening a conv feature map inside `nn.Sequential` uses `Rearrange('b c h w -> b (c h w)')`, which collapses channels and both spatial axes into one feature dimension while keeping the batch axis intact.

## Faded exercise 1

Complete the pipeline by inserting the Rearrange flatten layer between ReLU and Linear. Fill in the Rearrange layer.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
from einops.layers.torch import Rearrange

t.manual_seed(3)

def conv_classifier(in_channels, height, width, num_classes):
    flatten_layer = Rearrange('b c h w -> b (c h w)')
    return t.nn.Sequential(
        t.nn.Conv2d(in_channels, 8, kernel_size=3, padding=1),
        t.nn.ReLU(),
        flatten_layer,
        t.nn.Linear(8 * height * width, num_classes),
    )

print(conv_classifier(3, 8, 8, 5)(t.randn(4, 3, 8, 8)).shape)


def _test():
    t.manual_seed(33)
    model = conv_classifier(3, 8, 8, 5)
    out = model(t.randn(4, 3, 8, 8))
    assert tuple(out.shape) == (4, 5)
    # it must be a plain Sequential, not a custom Module subclass
    assert type(model) is t.nn.Sequential


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
