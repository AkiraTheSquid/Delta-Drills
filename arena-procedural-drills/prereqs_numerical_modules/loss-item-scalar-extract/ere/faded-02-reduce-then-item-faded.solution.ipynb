## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

When a loss is computed with `reduction='none'`, it returns a 1-D tensor of per-sample losses rather than a scalar. You cannot call `.item()` on a multi-element tensor — it raises a `RuntimeError`. The fix is to reduce first with `.mean()` or `.sum()` to get a 0-D tensor, then call `.item()` on the result.

The blanked step must first reduce the tensor, then call `.item()`.

````
import torch as t

def batch_mean_scalar(per_sample_loss):
    """
    Given a 1-D loss tensor, return the batch mean as a Python float.
    """
    mean_loss = per_sample_loss.mean().item()
    return mean_loss

# Exercise it
t.manual_seed(55)
losses = t.rand(6) + 0.1  # shape (6,), all positive
result = batch_mean_scalar(losses)
print(f"Batch mean loss: {result}  type={type(result).__name__}")


import torch as t

def _test():
    t.manual_seed(55)
    losses = t.rand(6) + 0.1

    result = batch_mean_scalar(losses)

    assert isinstance(result, float), "Result must be a Python float"

    # Verify numeric correctness against independent computation
    expected = losses.mean().item()
    assert abs(result - expected) < 1e-6

    # Try with a single-element tensor — should also work
    single = t.tensor([2.5])
    r = batch_mean_scalar(single)
    assert isinstance(r, float)
    assert abs(r - 2.5) < 1e-6


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
