## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

`expand` produces a zero-stride view sharing the source storage (no allocation), while `repeat` physically copies the data into fresh contiguous memory. Both yield the same values and shape, but only `repeat` allocates new storage, so their `data_ptr()` values differ.

## Worked solution

We broadcast a `(2,)` vector to `(3, 2)` two ways. `expand(3, 2)` gives stride `(0, 1)` and shares storage with `x`. `repeat(3, 1)` gives a contiguous `(3, 2)` with stride `(2, 1)` and a NEW data pointer. We print both strides and both storage-sharing booleans: expand shares, repeat does not. The values are identical, but the memory cost differs — expand is free, repeat copies `3*2` floats.

```
x = t.tensor([5.0, 7.0])
expanded = x.expand(3, 2)
repeated = x.repeat(3, 1)
print('expand stride:', tuple(expanded.stride()), 'shares:', expanded.data_ptr() == x.data_ptr())
print('repeat stride:', tuple(repeated.stride()), 'shares:', repeated.data_ptr() == x.data_ptr())
print('same values:', t.equal(expanded, repeated))
```
