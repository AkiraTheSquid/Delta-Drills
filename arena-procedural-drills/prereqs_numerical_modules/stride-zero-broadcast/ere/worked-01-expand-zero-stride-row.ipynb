## Setup

## Concept

`x.expand(R, N)` broadcasts a `(N,)` vector to `(R, N)` with stride `(0, 1)`: the zero stride on the new axis means every row reads the SAME underlying memory, so no data is copied. The result shares storage with `x`, confirmed by an equal `data_ptr()`.

## Worked solution

We take a `(3,)` vector and `expand` it to `(4, 3)`. The stride of the result is `(0, 1)`: walking down the rows advances the pointer by 0, so all four rows alias the original three floats. We confirm `expanded.data_ptr() == x.data_ptr()` (same storage) and print the stride. This is the no-copy broadcast: a `(4, 3)`-shaped view backed by only 3 floats of real memory.

```
x = t.tensor([10.0, 20.0, 30.0])
expanded = x.expand(4, 3)
print('shape:', tuple(expanded.shape))
print('stride:', tuple(expanded.stride()))
print('shares storage:', expanded.data_ptr() == x.data_ptr())
print('row 0 == row 3:', t.equal(expanded[0], expanded[3]))
```
