## Concept

Given a zero-mean data matrix `X` of shape `(N, D)`, the empirical covariance matrix is `C = X.T @ X / (N - 1)`, which has shape `(D, D)`. The contraction `(D, N) @ (N, D) → (D, D)` sums over the sample dimension `N`. The resulting matrix is symmetric and positive semi-definite.
