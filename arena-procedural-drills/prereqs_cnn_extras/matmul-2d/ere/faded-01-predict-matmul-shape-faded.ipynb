The shape rule for 2-D matrix multiplication is `(M, K) @ (K, N) → (M, N)`. The inner dimensions (both equal to `K`) are contracted away and disappear from the output. If the inner dimensions do not match, the operation raises an error. Predicting the output shape before running the operation is a critical debugging skill.

import torch as t
