## Concept

In the scaled dot-product attention mechanism, the raw attention scores are computed as `scores = Q @ K.T`, where `Q` and `K` both have shape `(T, d_head)`. The contraction produces shape `(T, T)`: entry `[i, j]` is the dot product between query `i` and key `j`. The shape rule is `(T, d_head) @ (d_head, T) → (T, T)`.

## Faded exercise 3

Complete the function below. Given query and key matrices of the same shape `(T, d_head)`, compute the raw score matrix. Fill in the single blanked line.

The key must be transposed before the matmul so the inner dimensions align.

**Fill in:** Compute Q @ K.T to produce the (T, T) attention score matrix.

````
import torch as t

def raw_attention_scores(Q, K):
    scores = Q @ K.T
    return scores

# Exercise it
t.manual_seed(88)
T, d_head = 6, 8
Q = t.randn(T, d_head)
K = t.randn(T, d_head)
scores = raw_attention_scores(Q, K)
print(f"Q shape:      {Q.shape}")
print(f"K.T shape:    {K.T.shape}")
print(f"scores shape: {scores.shape}")
print(f"scores[0,0] == Q[0]·K[0]: {t.isclose(scores[0,0], (Q[0]*K[0]).sum())}")


import torch as t

def _test():
    t.manual_seed(88)
    T, d_head = 6, 8
    Q = t.randn(T, d_head)
    K = t.randn(T, d_head)

    scores = raw_attention_scores(Q, K)

    assert scores.shape == (T, T), f"Expected ({T},{T}), got {scores.shape}"

    # Verify entry [i,j] is Q[i] . K[j]
    for i in range(T):
        for j in range(T):
            expected = (Q[i] * K[j]).sum()
            assert t.isclose(scores[i, j], expected, atol=1e-5), \
                f"scores[{i},{j}] mismatch"


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
