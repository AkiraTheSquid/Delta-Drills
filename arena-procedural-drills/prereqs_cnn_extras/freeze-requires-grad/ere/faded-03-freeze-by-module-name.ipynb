When fine-tuning large models, it is common to freeze parameters by module name prefix (e.g. freeze everything under `'encoder.'`) rather than by module reference. The `model.named_parameters()` iterator yields `(name, param)` pairs, making it easy to check whether a parameter belongs to a frozen section using `name.startswith(prefix)`.

```python
import torch as t
import torch.nn as nn

def freeze_by_prefix(model: nn.Module, prefix: str):
    frozen_names = []
    for name, param in model.named_parameters():
        if name.startswith(prefix):
            param.requires_grad = False
            frozen_names.append(name)
    return frozen_names
```
