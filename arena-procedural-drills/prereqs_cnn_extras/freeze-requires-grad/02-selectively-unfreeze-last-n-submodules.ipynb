## Selective unfreeze of the last N submodules — quick refresher

Pure 'freeze backbone, train head' is the simplest transfer-learning pattern. A finer-grained alternative: freeze EVERYTHING then **re-enable** training on only the last `n_last` submodules of the encoder. This trains a bit more of the network than head-only, but much less than full fine-tune.

```
for p in model.parameters():
    p.requires_grad = False                              # freeze all
children = list(model.encoder.children())                # ordered list
for layer in children[-n_last:]:                         # last n_last submodules
    for p in layer.parameters():
        p.requires_grad = True                           # unfreeze
```

**Exemplar.** For `encoder = Sequential(L0, L1, L2, L3)` and `n_last = 2`, layers `L2` and `L3` regain `requires_grad=True`; `L0` and `L1` stay frozen.

Implement `ex2_partial_unfreeze(model, n_last)` that performs **partial fine-tuning** in this order:

1. **Freeze everything**: set `p.requires_grad = False` on every param of `model`.
2. **Re-enable the last `n_last` encoder children**: take `list(model.encoder.children())[-n_last:]` and for each submodule set `p.requires_grad = True` on every param inside.
3. Leave the head `model.fc` FROZEN — that's the difference from the ex1 'replace head' pattern.
4. Return the list of trainable params (`[p for p in model.parameters() if p.requires_grad]`).

**Indexing detail.** `list(model.encoder.children())[-n_last:]` grabs the LAST `n_last` submodules regardless of which are param-bearing. ReLU children have no params, so they contribute nothing to the trainable list — but they still count toward the slice. This matches the standard 'last 2 blocks' partial-tune pattern.

**Boundary.** Assume `1 <= n_last <= len(list(model.encoder.children()))`.

````python
import torch.nn as nn

class ToyBackbone(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Linear(10, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 16),
        )
        self.fc = nn.Linear(16, 5)
    def forward(self, x):
        return self.fc(self.encoder(x))


def ex2_partial_unfreeze(model, n_last: int):
    """Freeze all, then re-enable requires_grad on the last n_last encoder children; head stays frozen."""
    raise NotImplementedError()


def _test_ex2():
    import torch.nn as nn
    from torch.nn import functional as F

    # n_last=1 — only the LAST encoder Linear (children[4]) is trainable.
    m = ToyBackbone()
    trainable = ex2_partial_unfreeze(m, n_last=1)
    # encoder children: 0 Linear, 1 ReLU, 2 Linear, 3 ReLU, 4 Linear.
    enc_children = list(m.encoder.children())
    # children[0] and children[2] (the first two Linears) must stay frozen.
    for p in enc_children[0].parameters():
        assert p.requires_grad is False, 'encoder[0] must be frozen at n_last=1'
    for p in enc_children[2].parameters():
        assert p.requires_grad is False, 'encoder[2] must be frozen at n_last=1'
    # children[4] must be trainable.
    for p in enc_children[4].parameters():
        assert p.requires_grad is True, 'encoder[4] must be trainable at n_last=1'
    # Head must stay frozen — unlike ex1.
    for p in m.fc.parameters():
        assert p.requires_grad is False, 'fc head must remain FROZEN (different from ex1)'
    # trainable list = exactly fc-free encoder[4] (weight + bias).
    assert len(trainable) == 2, f'n_last=1 → 2 trainable tensors (Linear weight+bias), got {len(trainable)}'

    # n_last=3 — slice [-3:] is (children[2] Linear, children[3] ReLU, children[4] Linear).
    # Both Linears trainable; ReLU contributes 0 params; fc still frozen.
    m3 = ToyBackbone()
    trainable3 = ex2_partial_unfreeze(m3, n_last=3)
    enc3 = list(m3.encoder.children())
    for p in enc3[0].parameters():
        assert p.requires_grad is False, 'encoder[0] still frozen at n_last=3'
    for p in enc3[2].parameters():
        assert p.requires_grad is True, 'encoder[2] trainable at n_last=3'
    for p in enc3[4].parameters():
        assert p.requires_grad is True, 'encoder[4] trainable at n_last=3'
    for p in m3.fc.parameters():
        assert p.requires_grad is False, 'fc still frozen at n_last=3'
    assert len(trainable3) == 4, f'n_last=3 → 4 trainable tensors (two Linears, w+b each), got {len(trainable3)}'

    # n_last=5 — every encoder Linear trainable; fc still frozen.
    m5 = ToyBackbone()
    trainable5 = ex2_partial_unfreeze(m5, n_last=5)
    enc5 = list(m5.encoder.children())
    for i in [0, 2, 4]:
        for p in enc5[i].parameters():
            assert p.requires_grad is True, f'encoder[{i}] trainable at n_last=5'
    for p in m5.fc.parameters():
        assert p.requires_grad is False, 'fc still frozen at n_last=5'
    assert len(trainable5) == 6, f'n_last=5 → 6 trainable tensors, got {len(trainable5)}'

    # Optimizer + backward step: only encoder[4] updates at n_last=1.
    m_b = ToyBackbone()
    trainable_b = ex2_partial_unfreeze(m_b, n_last=1)
    enc_b = list(m_b.encoder.children())
    frozen_w = enc_b[0].weight.detach().clone()
    frozen_w_2 = enc_b[2].weight.detach().clone()
    head_w = m_b.fc.weight.detach().clone()
    trainable_w = enc_b[4].weight.detach().clone()
    opt = t.optim.Adam(trainable_b, lr=1e-2)
    x = t.randn(4, 10)
    y = m_b(x)
    # Target shape (4,) for 5-way CE.
    loss = F.cross_entropy(y, t.tensor([0, 1, 2, 3]))
    loss.backward()
    opt.step()
    assert t.equal(enc_b[0].weight, frozen_w), 'encoder[0] mutated despite being frozen'
    assert t.equal(enc_b[2].weight, frozen_w_2), 'encoder[2] mutated despite being frozen'
    assert t.equal(m_b.fc.weight, head_w), 'fc head mutated despite being frozen'
    assert not t.equal(enc_b[4].weight, trainable_w), 'encoder[4] should have been updated'
    _dd_passed.add('ex2')
    print("ex2 ✓")

_test_ex2()
````

<details><summary>Solution</summary>

```python
def ex2_partial_unfreeze(model, n_last: int):
    for p in model.parameters():
        p.requires_grad = False
    children = list(model.encoder.children())
    for layer in children[-n_last:]:
        for p in layer.parameters():
            p.requires_grad = True
    return [p for p in model.parameters() if p.requires_grad]
```

**Why `list(model.encoder.children())[-n_last:]` and not `model.encoder[-n_last:]`.** Both work for `nn.Sequential`, but `.children()` is the generic interface across any `nn.Module` container — it'd work for a custom encoder built from named submodules too. The slice is the portable form.

**Why ReLU in the slice is harmless.** `nn.ReLU.parameters()` yields nothing, so `for p in relu.parameters(): p.requires_grad = True` is a no-op. The slice can include any mix of param-bearing and parameter-less modules without affecting correctness.

**Contrast with ex1.** Ex1 also replaces `model.fc` with a fresh head — new modules default to `requires_grad=True`, so the head trains. Here we LEAVE the original head in place and DON'T unfreeze it. Use case: continue using the original head (e.g. dimensionality is fine) but fine-tune deeper into the network. Common when transferring within-domain (same label space, different distribution).

**Generalization to ResNet.** Replace `model.encoder` with `model.layer4` (the last residual stage) and you have the standard 'fine-tune the last block' recipe used in countless downstream-task papers.
