In a BlockGroup, block 0 maps `in_feats -> out_feats` with `first_stride`, and the remaining `n_blocks - 1` blocks are all `ResBlock(out_feats, out_feats, first_stride=1)` - identity-shaped, so they preserve channels and resolution.

## Faded exercise 1

### Complete the BlockGroup builder

Block 0 is already created for you. Complete the function so the **remaining `n_blocks - 1` blocks** are appended as shape-preserving `ResBlock`s, then wrap the list in `nn.Sequential`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch.nn as nn

class ResBlock(nn.Module):
    def __init__(self, in_feats, out_feats, first_stride=1):
        super().__init__()
        self.in_feats = in_feats
        self.out_feats = out_feats
        self.first_stride = first_stride
        self.proj = nn.Conv2d(in_feats, out_feats, kernel_size=1, stride=first_stride)
    def forward(self, x):
        return self.proj(x)

def make_block_group(in_feats, out_feats, n_blocks, first_stride):
    blocks = [ResBlock(in_feats, out_feats, first_stride=first_stride)]
    for _ in range(n_blocks - 1):
        blocks.append(ResBlock(out_feats, out_feats, first_stride=1))
    return nn.Sequential(*blocks)

def _test():
    g = make_block_group(in_feats=10, out_feats=20, n_blocks=4, first_stride=2)
    assert len(g) == 4
    # block 0 is the shape changer
    assert g[0].in_feats == 10 and g[0].out_feats == 20 and g[0].first_stride == 2
    # remaining blocks are identity-shaped
    for i in range(1, 4):
        assert g[i].in_feats == 20, i
        assert g[i].out_feats == 20, i
        assert g[i].first_stride == 1, i
    # single-block edge case
    g1 = make_block_group(in_feats=3, out_feats=8, n_blocks=1, first_stride=1)
    assert len(g1) == 1
    assert g1[0].in_feats == 3 and g1[0].out_feats == 8

try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
