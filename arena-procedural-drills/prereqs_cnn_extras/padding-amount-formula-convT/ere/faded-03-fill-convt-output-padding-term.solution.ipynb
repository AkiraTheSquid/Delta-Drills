The complete ConvTranspose length formula adds `output_padding`: `(H_in - 1)*S - 2*P + K + OP`. Unlike `padding`, `output_padding` ADDS rows, used to disambiguate sizes when stride > 1.

## Faded exercise 3

Complete `convT_outlen_full` so it includes the output_padding term. Fill in the full formula.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn as nn

t.manual_seed(8)

def convT_outlen_full(h_in, k, s, p, op):
    out = (h_in - 1) * s - 2 * p + k + op
    return out

print(convT_outlen_full(7, 3, 2, 1, 1))


def _test():
    t.manual_seed(88)
    for h_in, k, s, p, op in [(7, 3, 2, 1, 1), (5, 4, 2, 1, 0), (6, 3, 2, 0, 1)]:
        pred = convT_outlen_full(h_in, k, s, p, op)
        layer = nn.ConvTranspose2d(1, 1, kernel_size=k, stride=s, padding=p, output_padding=op)
        actual = layer(t.randn(1, 1, h_in, h_in)).shape[-1]
        assert pred == actual, (h_in, k, s, p, op, pred, actual)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
