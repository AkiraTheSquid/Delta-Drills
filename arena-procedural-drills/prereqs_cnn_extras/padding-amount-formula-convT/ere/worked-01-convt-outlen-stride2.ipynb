## Concept

For `nn.ConvTranspose2d`, the spatial output length is `H_out = (H_in - 1) * S - 2 * P + K`. Unlike regular convolution, the `padding` argument SUBTRACTS from the output: each unit of padding peels one row off each side.

## Worked solution

We implement `convT_outlen(h_in, k, s, p)` returning `(h_in - 1) * s - 2 * p + k`. Reading the formula: `(h_in - 1) * s + k` is the no-padding output, and `-2 * p` removes one row per side per unit of padding. To validate, we build a real `nn.ConvTranspose2d` with the same kernel, stride, and padding, push a square input of the chosen size through it, and compare its actual spatial output to our formula. We seed, run a stride-2 example, and print both the predicted and actual sizes to confirm they match.

```
import torch as t
import torch.nn as nn

t.manual_seed(0)

def convT_outlen(h_in, k, s, p):
    return (h_in - 1) * s - 2 * p + k

h_in, k, s, p = 8, 4, 2, 1
pred = convT_outlen(h_in, k, s, p)
layer = nn.ConvTranspose2d(1, 1, kernel_size=k, stride=s, padding=p)
actual = layer(t.randn(1, 1, h_in, h_in)).shape[-1]
print('predicted:', pred, 'actual:', actual)
```
