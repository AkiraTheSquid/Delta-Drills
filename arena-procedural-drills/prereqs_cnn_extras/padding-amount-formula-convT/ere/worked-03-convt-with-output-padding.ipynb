## Concept

The full ConvTranspose length formula adds an `output_padding` term: `H_out = (H_in - 1) * S - 2 * P + K + OP`. `output_padding` is used to disambiguate the output size when stride > 1, and it ADDS rows (unlike `padding`, which subtracts).

## Worked solution

We extend the formula to `convT_outlen(h_in, k, s, p, op)` returning `(h_in - 1) * s - 2 * p + k + op`. The `+ op` term resolves the size ambiguity introduced by stride: several input sizes can map to the same intermediate, and `output_padding` nudges the final size up by `op` rows. We choose a stride-2 configuration with `output_padding=1`, predict the size, build the matching module, and compare. We seed, run, and print predicted vs actual, confirming the `+ op` term is correct.

```
import torch as t
import torch.nn as nn

t.manual_seed(2)

def convT_outlen(h_in, k, s, p, op):
    return (h_in - 1) * s - 2 * p + k + op

h_in, k, s, p, op = 7, 3, 2, 1, 1
pred = convT_outlen(h_in, k, s, p, op)
layer = nn.ConvTranspose2d(1, 1, kernel_size=k, stride=s, padding=p, output_padding=op)
actual = layer(t.randn(1, 1, h_in, h_in)).shape[-1]
print('predicted:', pred, 'actual:', actual)
```
