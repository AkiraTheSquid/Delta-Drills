## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

With stride 1, `H_out = (H_in - 1) - 2P + K`. Choosing `P = (K - 1) / 2` for an odd kernel makes the output equal the input, because the `+ (K - 1)` growth is exactly cancelled by the `- 2P` shrink. This is the same-padding trick for transposed convs.

## Worked solution

We pick an odd kernel `K = 5`, stride 1, and padding `P = (K - 1) // 2 = 2`. The formula gives `H_out = (H_in - 1) - 4 + 5 = H_in`, so the output should equal the input length. We compute the prediction, then build the matching `nn.ConvTranspose2d` and verify the actual output length equals the input length. We seed, run, and print the input length alongside the predicted and actual outputs, all of which agree, demonstrating the cancellation.

```
import torch as t
import torch.nn as nn

t.manual_seed(1)

def convT_outlen(h_in, k, s, p):
    return (h_in - 1) * s - 2 * p + k

h_in, k, s = 10, 5, 1
p = (k - 1) // 2  # same-padding for odd kernel, stride 1
pred = convT_outlen(h_in, k, s, p)
layer = nn.ConvTranspose2d(2, 2, kernel_size=k, stride=s, padding=p)
actual = layer(t.randn(1, 2, h_in, h_in)).shape[-1]
print('h_in:', h_in, 'predicted:', pred, 'actual:', actual)
```
