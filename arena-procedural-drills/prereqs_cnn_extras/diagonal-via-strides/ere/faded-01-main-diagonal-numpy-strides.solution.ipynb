## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

For a C-contiguous `(N, N)` NumPy array, the row stride is `N` elements and the column stride is `1` element, but `np.lib.stride_tricks.as_strided` works in **bytes**. A diagonal step is one row down plus one column right, so its byte stride is `(N + 1) * itemsize`. The diagonal has length `N`.

### Faded — main diagonal of an (N, N) array with NumPy as_strided

Complete `numpy_main_diagonal(a)`. Given a C-contiguous square array `a` of shape `(N, N)`, return a 1-D length-`N` no-copy view aliasing the main diagonal using `np.lib.stride_tricks.as_strided`.

The shape `(N,)` and the call to `as_strided` are written for you. You must supply the **byte stride** for the single diagonal axis. Remember NumPy strides are measured in bytes, so multiply the element step by `a.itemsize`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def numpy_main_diagonal(a):
    N = a.shape[0]
    strides = ((N + 1) * a.itemsize,)
    return np.lib.stride_tricks.as_strided(a, shape=(N,), strides=strides)


np.random.seed(0)
N = 5
a = np.arange(N * N).reshape(N, N)
d = numpy_main_diagonal(a)
print("diagonal:", d.tolist())


def _test():
    np.random.seed(0)
    for N in (3, 5, 8):
        a = np.arange(N * N).reshape(N, N).astype(np.int64)
        d = numpy_main_diagonal(a)
        ref = np.diagonal(a).copy()
        assert d.shape == (N,), d.shape
        assert np.array_equal(d, ref), (d, ref)
        # no-copy: shares memory with a
        assert np.shares_memory(d, a)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
