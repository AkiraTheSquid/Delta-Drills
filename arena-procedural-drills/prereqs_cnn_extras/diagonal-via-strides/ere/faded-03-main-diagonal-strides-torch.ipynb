## Concept

For a contiguous row-major `(N, N)` torch tensor, the row stride is `N` and the column stride is `1`. A single step along the main diagonal advances one row down and one column right, so the diagonal-step stride is `N + 1`. The diagonal is a length-`N` no-copy view, verifiable with `.data_ptr()`.
