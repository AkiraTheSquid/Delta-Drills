BatchNorm's affine step `y = gamma * x_hat + beta` is per-channel. In NumPy, the `(C,)` params must be reshaped to `(1, C, 1, 1)` so they broadcast over batch and spatial axes of a `(B, C, H, W)` array — a raw `(C,)` would wrongly align with the last axis `W`.
