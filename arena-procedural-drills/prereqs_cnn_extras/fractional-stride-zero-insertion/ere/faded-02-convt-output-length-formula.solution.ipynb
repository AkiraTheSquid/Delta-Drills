The no-padding ConvTranspose2d output spatial length is `(H_in - 1) * s + k`, where `H_in` is the input size, `s` is the stride, and `k` is the kernel size. This combines the dilation step `(H_in - 1) * s + 1` and the kernel overhang `k - 1` from each side of the dilated input.

## Faded exercise 2

Implement `convT_outlen(h_in, k, s)` that returns the output spatial length of a no-padding `nn.ConvTranspose2d` with kernel size `k` and stride `s` applied to an input of length `h_in`. Use the formula `(h_in - 1) * s + k`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def convT_outlen(h_in: int, k: int, s: int) -> int:
    return (h_in - 1) * s + k


def _test():
    import torch as t
    import torch.nn as nn
    cases = [(5, 3, 1), (5, 4, 2), (3, 3, 3), (1, 5, 4)]
    for h_in, k, s in cases:
        predicted = convT_outlen(h_in, k, s)
        x = t.randn(1, 1, h_in, h_in)
        conv = nn.ConvTranspose2d(1, 1, kernel_size=k, stride=s, bias=False)
        with t.no_grad():
            out = conv(x)
        assert out.shape[2] == predicted, f"h_in={h_in},k={k},s={s}: predicted {predicted}, got {out.shape[2]}"


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
