## Concept

In one dimension, inserting `s - 1` zeros between each pair of adjacent elements transforms an input of length `L` into an output of length `(L - 1) * s + 1`. The implementation allocates a zeros array of the target length and places the original elements at every `s`-th position using slice assignment.

## Faded exercise 3

Implement `zero_insert_1d(x, s)` for a 1-D tensor `x` of length `L`. Compute `L_out = (L - 1) * s + 1`, allocate `y = t.zeros(L_out, dtype=x.dtype)`, assign `y[::s] = x`, and return `y`.

**Fill in:** Compute L_out as (L-1)*s+1, allocate y as t.zeros(L_out, dtype=x.dtype), and assign the input into y at every s-th position with y[::s] = x.

````
import torch as t
from torch import Tensor

def zero_insert_1d(x: Tensor, s: int) -> Tensor:
    L = x.shape[0]
    L_out = (L - 1) * s + 1
    y = t.zeros(L_out, dtype=x.dtype)
    y[::s] = x
    return y


def _test():
    import torch as t
    x = t.tensor([1.0, 2.0, 3.0])
    y2 = zero_insert_1d(x, 2)
    assert y2.tolist() == [1.0, 0.0, 2.0, 0.0, 3.0]
    y3 = zero_insert_1d(x, 3)
    assert y3.shape[0] == 7
    assert y3[0].item() == 1.0 and y3[3].item() == 2.0 and y3[6].item() == 3.0
    # stride 1: unchanged
    y1 = zero_insert_1d(x, 1)
    assert t.equal(y1, x)
    # single element: unchanged for any stride
    x1 = t.tensor([5.0])
    assert zero_insert_1d(x1, 4).tolist() == [5.0]


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
