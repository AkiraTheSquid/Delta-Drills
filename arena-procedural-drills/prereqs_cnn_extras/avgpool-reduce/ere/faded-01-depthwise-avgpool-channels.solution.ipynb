## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

Average pooling is just a mean over a factored axis, and that axis need not be spatial. Here we pool over CHANNELS: factor `C -> (c g)` where `g` is the group size, then mean-reduce `g`. This collapses every group of `g` consecutive channels into one averaged channel, yielding `(B, C // g, H, W)`.

### Faded — average pool over the channel axis

Implement `avgpool_channels(x, g)`. Given `x: (B, C, H, W)` with `C % g == 0`, return a `(B, C // g, H, W)` tensor whose entries are the mean of each non-overlapping group of `g` consecutive channels (spatial positions untouched).

The surrounding setup and the verification are filled in. You must complete the single `einops.reduce` call that factors the channel axis and means over the inner group.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import einops
from torch import Tensor

t.manual_seed(0)

def avgpool_channels(x: Tensor, g: int) -> Tensor:
    return einops.reduce(x, 'b (c g) h w -> b c h w', 'mean', g=g)

x = t.randn(2, 6, 3, 3)
out = avgpool_channels(x, 2)
print('out shape:', tuple(out.shape))


def _test():
    import torch as t
    t.manual_seed(0)
    x = t.randn(2, 6, 3, 3)
    out = avgpool_channels(x, 2)
    assert out.shape == (2, 3, 3, 3), out.shape
    # independent ground truth: reshape into groups and mean over group axis
    ref = x.reshape(2, 3, 2, 3, 3).mean(dim=2)
    assert t.allclose(out, ref, atol=1e-6)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
