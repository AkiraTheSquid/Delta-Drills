For a square non-overlapping window of size `p`, factor both spatial axes by the SAME `p`: `(h p1)` and `(w p2)` with `p1 = p2 = p`, then mean-reduce the inner factors. This reproduces `nn.AvgPool2d(p)` / `F.avg_pool2d(x, p)` and is the canonical avgpool-as-reduce pattern.

### Faded — 3x3 square average pool

Implement `avgpool_square(x, p)`. Given `x: (B, C, H, W)` with `H % p == 0` and `W % p == 0`, return `(B, C, H // p, W // p)` where each entry is the mean of a `p x p` block.

The function signature, exercise, and `F.avg_pool2d` check are filled in. Complete the einops pattern STRING that factors both spatial axes for a square window.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn.functional as F
import einops
from torch import Tensor

t.manual_seed(0)

def avgpool_square(x: Tensor, p: int) -> Tensor:
    pattern = 'b c (h p1) (w p2) -> b c h w'
    return einops.reduce(x, pattern, 'mean', p1=p, p2=p)

x = t.randn(2, 3, 9, 9)
out = avgpool_square(x, 3)
print('out shape:', tuple(out.shape))


def _test():
    import torch as t
    import torch.nn.functional as F
    t.manual_seed(0)
    x = t.randn(2, 3, 9, 9)
    out = avgpool_square(x, 3)
    assert out.shape == (2, 3, 3, 3), out.shape
    ref = F.avg_pool2d(x, kernel_size=3)
    assert t.allclose(out, ref, atol=1e-6)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
