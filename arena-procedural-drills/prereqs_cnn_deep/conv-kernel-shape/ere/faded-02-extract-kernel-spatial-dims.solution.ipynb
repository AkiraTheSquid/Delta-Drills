## Concept

For an `nn.Conv2d`, the last two axes of `weight` are the kernel's spatial extent `(KH, KW)`. They live at indices 2 and 3 of the `(OC, IC, KH, KW)` shape. The receptive-field area of one filter is `KH * KW`.

## Faded exercise 2

### Report the kernel spatial size

Implement `kernel_spatial(conv)` returning a dict with `'kh'`, `'kw'`, and `'rf_area'` (the per-filter receptive-field area `KH * KW`). Derive everything from `conv.weight.shape`, not from `conv.kernel_size`.

Complete the blanked line that unpacks the four axes from the weight shape.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def kernel_spatial(conv) -> dict:
    OC, IC, KH, KW = conv.weight.shape
    return {'kh': int(KH), 'kw': int(KW), 'rf_area': int(KH * KW)}

t.manual_seed(0)
conv = t.nn.Conv2d(in_channels=3, out_channels=5, kernel_size=(4, 6))
print(kernel_spatial(conv))


def _test():
    conv = t.nn.Conv2d(in_channels=3, out_channels=5, kernel_size=(4, 6))
    out = kernel_spatial(conv)
    assert out['kh'] == 4, out
    assert out['kw'] == 6, out
    assert out['rf_area'] == 24, out
    conv2 = t.nn.Conv2d(in_channels=2, out_channels=7, kernel_size=3)
    out2 = kernel_spatial(conv2)
    assert out2['kh'] == 3 and out2['kw'] == 3 and out2['rf_area'] == 9, out2


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
