An `nn.Conv2d` weight tensor is laid out as **`(OC, IC, KH, KW)`** — output channels first, then input channels, then the two spatial kernel dims. The constructor takes `(in, out, kernel)`, so you must flip IC and OC when building the tensor yourself.

## Faded exercise 1

### Build a constant Conv2d weight

Given `IC=6`, `OC=12`, `KH=3`, `KW=7`, build a `float32` weight tensor filled with `0.5` whose shape matches `nn.Conv2d(in_channels=IC, out_channels=OC, kernel_size=(KH, KW)).weight`.

Complete the single blanked line that specifies the **shape tuple** passed to `t.full`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def build_const_weight(IC, OC, KH, KW) -> Tensor:
    shape = (OC, IC, KH, KW)
    return t.full(shape, 0.5, dtype=t.float32)

t.manual_seed(0)
w = build_const_weight(6, 12, 3, 7)
print(tuple(w.shape))


def _test():
    w = build_const_weight(6, 12, 3, 7)
    ref = t.nn.Conv2d(in_channels=6, out_channels=12, kernel_size=(3, 7))
    assert tuple(w.shape) == tuple(ref.weight.shape), (tuple(w.shape), tuple(ref.weight.shape))
    assert w.shape == (12, 6, 3, 7)
    assert w.dtype == t.float32
    assert w.numel() == 12 * 6 * 3 * 7
    assert t.allclose(w, t.full_like(w, 0.5))


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
