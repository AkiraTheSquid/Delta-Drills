With padding `P`, the conv sees an effectively longer input of `L + 2P`, so the output length is `(L + 2P - K) // S + 1`. Padding adds windows at the start and end; stride still subsamples them. Predicting the shape means plugging into the full formula, not just dividing by the stride.

### Faded — padded strided conv1d output length

Implement `padded_conv1d_outlen(l_in, k, s, p)` returning the output length of a 1-D convolution with kernel `k`, stride `s`, and padding `p`. Complete the single returned expression, then the test cross-checks against a real `F.conv1d` call with matching padding.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch.nn.functional as F

def padded_conv1d_outlen(l_in, k, s, p):
    out = (l_in + 2 * p - k) // s + 1
    return out


def _test():
    t.manual_seed(0)
    cases = [(20, 3, 2, 1), (31, 5, 3, 2), (16, 4, 2, 0), (50, 7, 4, 3)]
    for l_in, k, s, p in cases:
        x = t.randn(1, 2, l_in)
        w = t.randn(3, 2, k)
        y = F.conv1d(x, w, stride=s, padding=p)
        assert padded_conv1d_outlen(l_in, k, s, p) == y.shape[-1], (l_in, k, s, p, y.shape[-1])


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
