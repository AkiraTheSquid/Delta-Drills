## Concept

Converting a `ConvTranspose2d` weight `(IC, OC, KH, KW)` into `Conv2d` layout `(OC, IC, KH, KW)` is a single einops rearrange that swaps the first two axes and leaves the spatial axes alone. The correct pattern is `'i o kh kw -> o i kh kw'`.

## Faded exercise 1

### Faded — channel-axis swap with einops

Implement `convT_to_conv_swap(w_convT)`. Given a transposed-conv kernel of shape `(IC, OC, KH, KW)`, return the conv-layout kernel of shape `(OC, IC, KH, KW)`. The spatial axes must stay in their original order. Complete the single blanked `rearrange` call.

**Fill in:** the einops rearrange that swaps the in/out channel axes while preserving the spatial axes

````
def convT_to_conv_swap(w_convT: Tensor) -> Tensor:
    out = rearrange(w_convT, 'i o kh kw -> o i kh kw')
    return out


def _test():
    t.manual_seed(0)
    w = t.randn(2, 7, 3, 3)        # (IC=2, OC=7, KH=3, KW=3)
    got = convT_to_conv_swap(w)
    ref = w.transpose(0, 1)
    assert tuple(got.shape) == (7, 2, 3, 3), tuple(got.shape)
    assert t.allclose(got, ref), 'channel axes not swapped correctly'
    # spatial axes must be untouched (no flip)
    assert t.equal(got[0, 0], w[0, 0]), 'spatial axes were altered'


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
