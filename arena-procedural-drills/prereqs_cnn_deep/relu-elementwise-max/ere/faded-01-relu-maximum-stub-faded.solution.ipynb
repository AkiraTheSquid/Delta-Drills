## Concept

ReLU is `max(x, 0)` applied per element. In PyTorch, `torch.maximum(a, b)` computes the elementwise maximum of two tensors. To implement ReLU from first principles, pass the input tensor and a scalar zero tensor: `torch.maximum(x, torch.tensor(0.0))`. This reveals the structural definition of ReLU rather than hiding it behind `F.relu`.

## Faded exercise 1

Implement `my_relu(x: Tensor) -> Tensor` using `t.maximum` and a scalar zero tensor.

Do **not** use `F.relu`, `x.clamp`, or `x * (x > 0)`. The goal is to use the explicit `t.maximum` form.

**Fill in:** Return the elementwise maximum of x and a zero tensor using t.maximum.

````
import torch as t

def my_relu(x: t.Tensor) -> t.Tensor:
    return t.maximum(x, t.tensor(0.0))


def _test():
    import torch as t
    x = t.tensor([-3.0, -1.0, 0.0, 1.0, 4.0])
    y = my_relu(x)
    expected = t.tensor([0.0, 0.0, 0.0, 1.0, 4.0])
    assert t.allclose(y, expected), f'Expected {expected}, got {y}'
    # Negative inputs must be zero
    assert (y[x < 0] == 0).all()
    # Positive inputs must be unchanged
    assert t.allclose(y[x > 0], x[x > 0])


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
