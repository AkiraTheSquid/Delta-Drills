## Concept

The backward pass of ReLU multiplies the incoming gradient by a binary mask: 1 where the forward input was positive, 0 where it was negative or zero. This is equivalent to `grad_out * (x > 0).float()`. Neurons with negative pre-activations receive zero gradient and cannot be updated — the source of the "dying ReLU" problem.

## Faded exercise 2

Implement `relu_backward(grad_out: Tensor, x: Tensor) -> Tensor` which computes the gradient of the ReLU operation.

The function should multiply `grad_out` by 1 where `x > 0` and by 0 elsewhere. Use `(x > 0).float()` for the mask.

**Fill in:** Multiply grad_out by the binary mask (x > 0).float() to zero out gradients where x was non-positive.

````
import torch as t

def relu_backward(grad_out: t.Tensor, x: t.Tensor) -> t.Tensor:
    return grad_out * (x > 0).float()


def _test():
    import torch as t
    x = t.tensor([-2.0, 0.0, 1.5, 3.0, -0.1])
    grad_out = t.tensor([1.0, 1.0, 1.0, 2.0, 0.5])
    result = relu_backward(grad_out, x)
    # Negative x positions should have zero gradient
    assert result[0].item() == 0.0
    assert result[1].item() == 0.0
    assert result[4].item() == 0.0
    # Positive x positions should pass gradient through
    assert t.isclose(result[2], t.tensor(1.0))
    assert t.isclose(result[3], t.tensor(2.0))


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
