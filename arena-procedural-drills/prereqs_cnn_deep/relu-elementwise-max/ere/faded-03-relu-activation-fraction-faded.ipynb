## Concept

When analyzing a network layer, it is useful to measure what fraction of ReLU units are active (output > 0). A low fraction may indicate dying ReLUs. The activation fraction is simply `(relu_output > 0).float().mean()`. Applying `t.maximum` then computing this statistic combines forward computation with diagnostic analysis.

def relu_stats(x: t.Tensor):
