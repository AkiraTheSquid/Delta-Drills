The `'b c (h h2) (w w2) -> b c h w'` pattern decomposes the spatial axes into block coordinates `(h, w)` and within-block coordinates `(h2, w2)`. Reducing with `'mean'` and `h2=2, w2=2` averages each non-overlapping `2x2` block, performing average pooling without calling `F.avg_pool2d`.
