`einops.repeat` with a new named axis on the right side — `'c h w -> b c h w'` — broadcasts the image across `b` copies. This is the standard way to produce a fake batch for testing a model on a single image, without using `unsqueeze` and `expand`.

## Faded exercise 1

Complete `add_batch_axis`. Use `repeat` to broadcast a single `(c, h, w)` image to a batch of `b` identical copies.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from einops import repeat
import torch as t

def add_batch_axis(img, b):
    """img: (c, h, w) -> (b, c, h, w)"""
    return repeat(img, 'c h w -> b c h w', b=b)


def _test():
    import torch as t
    t.manual_seed(0)
    c, h, w = 3, 8, 8
    img = t.randn(c, h, w)
    for b in [1, 4, 8]:
        out = add_batch_axis(img, b)
        assert out.shape == (b, c, h, w)
        # All copies identical
        assert t.allclose(out[0], img)
        assert t.allclose(out[-1], img)
        # No copy — stride-0 on batch axis
        assert out.data_ptr() == img.data_ptr()


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
