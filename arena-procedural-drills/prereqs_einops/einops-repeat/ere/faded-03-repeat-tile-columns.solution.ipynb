## Concept

The `(r w)` composition on the output side of `repeat` tiles the column axis: the full original row appears `r` times end-to-end. Because `r` is the outer index (varies slower) and `w` the inner (varies faster), elements 0 through w-1 form the first tile, elements w through 2w-1 the second tile, etc.

## Faded exercise 3

Complete `tile_columns`. Use `repeat` with `(r w)` composition to tile each row `r` times horizontally.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from einops import repeat
import torch as t

def tile_columns(x, r):
    """x: (h, w) -> (h, r*w), each row tiled r times horizontally."""
    return repeat(x, 'h w -> h (r w)', r=r)


def _test():
    import torch as t
    t.manual_seed(6)
    h, w = 4, 6
    x = t.randn(h, w)
    for r in [2, 3]:
        out = tile_columns(x, r)
        assert out.shape == (h, r * w)
        for tile_idx in range(r):
            assert t.allclose(out[:, tile_idx * w:(tile_idx + 1) * w], x)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
