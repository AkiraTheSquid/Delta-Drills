## Concept

The `(h r)` composition on the output side of `repeat` stretches the height axis: each source row appears `r` times consecutively. This is nearest-neighbor vertical upsampling — `r` is the inner factor, varying fastest, so row 0 occupies output positions 0 through r-1 as a block.

## Faded exercise 2

Complete `stretch_rows`. Use `repeat` with `(h r)` composition to stretch each row `r` times.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from einops import repeat
import torch as t

def stretch_rows(x, r):
    """x: (h, w) -> (h*r, w), each row repeated r times consecutively."""
    return repeat(x, 'h w -> (h r) w', r=r)


def _test():
    import torch as t
    t.manual_seed(4)
    h, w = 5, 7
    x = t.randn(h, w)
    for r in [2, 3, 5]:
        out = stretch_rows(x, r)
        assert out.shape == (h * r, w)
        for i in range(h):
            for offset in range(r):
                assert t.allclose(out[i * r + offset], x[i])


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
