## Concept

A symmetric binary op (like `t.add`) has the same gradient rule for both inputs, so one `back_fn` object can be aliased under both `(fwd, 0)` and `(fwd, 1)`. Both lookups must then return the IDENTICAL object (`is`), not just equal functions.

## Faded exercise 2

`register_both(BFL, fwd_fn, back_fn)` should register the single `back_fn` under BOTH `argnum=0` and `argnum=1` for `fwd_fn`. The `BackwardFuncLookup` class is complete; fill in the body of `register_both` so both keys point at the same object.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
class BackwardFuncLookup:
    def __init__(self):
        self.back_funcs = {}
    def add_back_func(self, forward_fn, argnum, back_fn):
        self.back_funcs[(forward_fn, argnum)] = back_fn
    def get_back_func(self, forward_fn, argnum):
        key = (forward_fn, argnum)
        if key not in self.back_funcs:
            raise KeyError(f'No back_fn for ({forward_fn!r}, argnum={argnum}).')
        return self.back_funcs[key]


def register_both(BFL, fwd_fn, back_fn):
    BFL.add_back_func(fwd_fn, 0, back_fn)
    BFL.add_back_func(fwd_fn, 1, back_fn)
    return BFL


def _test():
    def add_back(grad_out, out, x, y):
        return grad_out
    BFL = BackwardFuncLookup()
    ret = register_both(BFL, t.add, add_back)
    # returns the same lookup
    assert ret is BFL
    # both argnums resolve to the IDENTICAL object
    f0 = BFL.get_back_func(t.add, 0)
    f1 = BFL.get_back_func(t.add, 1)
    assert f0 is add_back
    assert f1 is add_back
    assert f0 is f1
    # exactly two keys were created
    assert len(BFL.back_funcs) == 2


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
