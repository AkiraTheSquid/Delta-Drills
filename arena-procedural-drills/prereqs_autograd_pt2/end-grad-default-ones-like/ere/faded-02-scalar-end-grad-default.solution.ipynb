Implement `resolve_scalar_end_grad(end_node, end_grad)` where `end_node.array` has shape `()`. If `end_grad is None`, return `t.ones_like(end_node.array)`. Otherwise assert shapes match and return `end_grad.array`. The returned tensor must have shape `()`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

class MiniTensor:
    def __init__(self, array):
        self.array = array

def resolve_scalar_end_grad(end_node: MiniTensor, end_grad):
    if end_grad is None:
        return t.ones_like(end_node.array)
    assert end_grad.array.shape == end_node.array.shape, (
        f'end_grad shape {tuple(end_grad.array.shape)} != '
        f'end_node shape {tuple(end_node.array.shape)}'
    )
    return end_grad.array


def _test():
    import torch as t
    class MiniTensor:
        def __init__(self, array):
            self.array = array
    scalar_node = MiniTensor(t.tensor(2.5))
    seed = resolve_scalar_end_grad(scalar_node, None)
    assert seed.shape == ()
    assert seed.item() == 1.0
    # explicit 0-D end_grad
    explicit = MiniTensor(t.tensor(0.5))
    result = resolve_scalar_end_grad(scalar_node, explicit)
    assert result.item() == 0.5
    assert result.shape == ()
    # shape mismatch: provide (1,) for a () node
    bad = MiniTensor(t.tensor([1.0]))
    try:
        resolve_scalar_end_grad(scalar_node, bad)
        assert False
    except AssertionError:
        pass


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
