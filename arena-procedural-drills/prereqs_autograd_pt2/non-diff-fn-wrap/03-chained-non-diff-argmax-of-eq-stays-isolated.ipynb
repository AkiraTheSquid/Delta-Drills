````python
import numpy as np
import torch as t
from torch import Tensor
import einops
from einops import rearrange, reduce, repeat

t.manual_seed(0)
np.random.seed(0)
# === manual autograd primitives — shared across all drills in this folder ===
from dataclasses import dataclass, field
from typing import Any, Callable, Optional

grad_tracking_enabled = True

@dataclass
class Recipe:
    func: Optional[Callable] = None
    args: tuple = ()
    kwargs: dict = field(default_factory=dict)
    parents: dict = field(default_factory=dict)

class MiniTensor:
    def __init__(self, array, requires_grad: bool = False, recipe=None):
        self.array = array
        self.requires_grad = requires_grad
        self.recipe = recipe
    def __repr__(self):
        return f'MiniTensor({self.array!r}, requires_grad={self.requires_grad})'
````

## Non-diff op chained into another non-diff op — graph stays isolated

Ex1 wrapped a single non-diff op (`eq`); ex2 confirmed a non-diff branch is skipped during topo walk from a diff end-node. The deepening move chains TWO non-diff ops in series:

```
mask = eq(a, b)            # non-diff: requires_grad=False, recipe=None
idx  = argmax(mask)        # non-diff: same — operates on a non-tracked input
```

The interesting consequence: BOTH outputs have `recipe=None`. The graph walk from `idx` cannot reach `mask`, `a`, or `b` — it stops at `idx` itself (graph = `[idx]`). Each non-diff output is its own isolated leaf in the autograd-graph view, regardless of how many non-diff ops produced it.

**Why this is the right behaviour.** A chain of non-diff ops produces a tensor that cannot receive a meaningful grad — there's no chain rule that would let one flow. Treating it as a leaf in the graph is correct: the reverse pass simply has no work to do at it.

**Contrast with ex2.** Ex2 isolated ONE non-diff branch on a graph whose main path was differentiable. Here, EVERY op is non-diff — the entire subgraph collapses into independent single-node graphs.

Provided in the stub: `sorted_computational_graph(tensor)` walks the recipe-parents graph and stops at any `recipe=None` node.

Test inputs build (verbatim — your `wrap_forward_fn` must produce these structurally):
- `a`, `b`: tracked MiniTensor leaves.
- `mask = eq_wrap(a, b)` — first non-diff op. Recipe must be None.
- `idx  = argmax_wrap(mask)` — second non-diff op, OPERATES ON a non-tracked input. Recipe must also be None.
- `c = add_wrap(a, b)`, `d = mul_wrap(c, c)` — pure-diff side branch as a control.

The test verifies:
1. Both `mask.recipe is None` and `idx.recipe is None`.
2. Walking from `idx`: only `[idx]` is found (length 1 — chain stays isolated). `mask`, `a`, `b` are NOT in the walk.
3. Walking from `mask`: only `[mask]` (length 1).
4. Walking from `d`: includes `[d, c, a, b]` (the diff side); EXCLUDES `mask` and `idx`.
5. `mask.requires_grad is False` and `idx.requires_grad is False`.

````python
def wrap_forward_fn(fwd_fn, is_differentiable: bool = True):
    """Three-gate wrapper. Recipe only when requires_grad=True."""
    raise NotImplementedError()


def sorted_computational_graph(tensor):
    """Provided: reverse-topo sort, stops at recipe=None."""
    result = []
    perm = set()
    def visit(cur):
        if id(cur) in perm:
            return
        perm.add(id(cur))
        if cur.recipe is not None:
            for p in cur.recipe.parents.values():
                visit(p)
        result.append(cur)
    visit(tensor)
    return result[::-1]


def _test_ex3():
    globals()['grad_tracking_enabled'] = True
    add_wrap = wrap_forward_fn(t.add)
    mul_wrap = wrap_forward_fn(t.multiply)
    eq_wrap = wrap_forward_fn(t.eq, is_differentiable=False)
    # argmax: torch op that returns a long tensor. is_differentiable=False.
    # Note: argmax doesn't accept bool inputs in modern torch, so we cast first.
    argmax_wrap = wrap_forward_fn(lambda z: t.argmax(z.float()), is_differentiable=False)

    a = MiniTensor(t.tensor([1.0, 2.0, 3.0, 5.0]), requires_grad=True)
    b = MiniTensor(t.tensor([1.0, 0.0, 3.0, 4.0]), requires_grad=True)

    # --- non-diff branch 1: mask = eq(a, b) ---
    mask = eq_wrap(a, b)
    assert mask.recipe is None, 'mask: first non-diff must have recipe=None'
    assert mask.requires_grad is False
    # Forward value should be correct (boolean tensor of equality).
    assert (mask.array == (a.array == b.array)).all()

    # --- non-diff branch 2: idx = argmax(mask). mask is non-tracked. ---
    idx = argmax_wrap(mask)
    assert idx.recipe is None, 'idx: second non-diff (chained) must also have recipe=None'
    assert idx.requires_grad is False
    # Forward value: argmax of [T, F, T, F] cast to bool int = first True position.
    # In torch, argmax over bool returns the first max; for [1,0,1,0] -> 0.
    assert isinstance(idx.array, t.Tensor)

    # --- diff side branch (control): c = add(a, b), d = mul(c, c) ---
    c = add_wrap(a, b)
    d = mul_wrap(c, c)
    assert c.recipe is not None and c.recipe.func is t.add
    assert d.recipe is not None and d.recipe.func is t.multiply
    assert c.requires_grad is True and d.requires_grad is True

    # === walking from idx: only [idx] reachable (chain isolated) ===
    idx_graph = sorted_computational_graph(idx)
    idx_ids = {id(n) for n in idx_graph}
    assert idx_ids == {id(idx)}, (
        f'walk from chained non-diff output should yield only itself; got {idx_ids} '
        f'(expected just {{id(idx)}})'
    )
    assert id(mask) not in idx_ids, 'mask must not be reachable from idx (recipe=None breaks chain)'
    assert id(a) not in idx_ids and id(b) not in idx_ids

    # === walking from mask: only [mask] reachable ===
    mask_graph = sorted_computational_graph(mask)
    mask_ids = {id(n) for n in mask_graph}
    assert mask_ids == {id(mask)}, f'mask walk wrong: {mask_ids}'

    # === walking from d (diff side): d, c, a, b reachable; mask + idx NOT ===
    d_graph = sorted_computational_graph(d)
    d_ids = {id(n) for n in d_graph}
    assert id(d) in d_ids and id(c) in d_ids and id(a) in d_ids and id(b) in d_ids, (
        f'diff side walk missing nodes; got {d_ids}'
    )
    assert id(mask) not in d_ids, 'mask must not appear in d-walk'
    assert id(idx) not in d_ids, 'idx must not appear in d-walk'

    # === confirm reverse-topo ordering on d-walk ===
    # d first, leaves last; parents come AFTER children in the result list.
    assert d_graph[0] is d, f'd must be first in reverse-topo, got {type(d_graph[0]).__name__}'
    # c must precede a and b.
    d_list_ids = [id(n) for n in d_graph]
    assert d_list_ids.index(id(c)) < d_list_ids.index(id(a))
    assert d_list_ids.index(id(c)) < d_list_ids.index(id(b))

    # === structural property: non-diff chain length = 2, but each output is an island ===
    # Even though idx semantically depends on mask which depends on a,b, the AUTOGRAD GRAPH
    # has zero edges among them. Counted as 3 disconnected singletons.
    all_ids = idx_ids | mask_ids | d_ids
    assert id(idx) in all_ids and id(mask) in all_ids
    # but idx_ids and mask_ids and d_ids are pairwise disjoint:
    assert idx_ids.isdisjoint(mask_ids)
    assert idx_ids.isdisjoint(d_ids)
    assert mask_ids.isdisjoint(d_ids)
    _dd_passed.add('ex3')
    print("ex3 ✓")

_test_ex3()
````

```python
def wrap_forward_fn(fwd_fn, is_differentiable: bool = True):
    def tensor_func(*args, **kwargs):
        raw_args = tuple(a.array if isinstance(a, MiniTensor) else a for a in args)
        out_arr = fwd_fn(*raw_args, **kwargs)
        global_on = globals().get('grad_tracking_enabled', True)
        any_tracked = any(isinstance(a, MiniTensor) and a.requires_grad for a in args)
        requires_grad = bool(global_on and is_differentiable and any_tracked)
        out = MiniTensor(out_arr, requires_grad=requires_grad)
        if requires_grad:
            parents = {i: a for i, a in enumerate(args) if isinstance(a, MiniTensor)}
            out.recipe = Recipe(func=fwd_fn, args=raw_args, kwargs=kwargs, parents=parents)
        return out
    return tensor_func
```

**Recipe-None cascades.** When `eq_wrap(a, b)` produces `mask.requires_grad=False`, the subsequent `argmax_wrap(mask)` call sees `mask` as a non-tracked input. Even if `argmax` had been `is_differentiable=True`, the three-gate AND would still give `requires_grad=False` (no tracked input). Chaining preserves the non-tracking property.

**Per-non-diff-output is a graph singleton.** From the autograd graph's point of view, each non-diff output is its own isolated 1-node DAG. The semantic dependency `idx <- mask <- a, b` is INVISIBLE to the topo walker because there are no recipes encoding those edges. This is correct: no gradient flows along those edges, so the reverse pass has no work to do.

**Why this is a feature, not a limitation.** A user calling `loss.backward()` on a diff loss `d` that incidentally shares leaves with a non-diff `idx` doesn't want grad to flow through `idx`. The recipe=None mechanism makes that the default, with no conditional logic in the reverse pass.
