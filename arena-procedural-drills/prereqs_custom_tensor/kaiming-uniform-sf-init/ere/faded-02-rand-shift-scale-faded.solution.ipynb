A `t.rand(...)` call produces values in `[0, 1)`. To obtain a symmetric range `(-sf, +sf)`, apply a two-step transformation: multiply by 2 and subtract 1 to get `[-1, 1)`, then multiply by `sf` to scale to `(-sf, +sf)`.

## Faded exercise 2

The scale factor `sf` is already computed. Fill in the expression that transforms the raw uniform `[0,1)` samples into the target `(-sf, +sf)` range.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def kaiming_uniform_sf(in_features: int, out_features: int, generator: t.Generator) -> t.Tensor:
    sf = in_features ** -0.5
    raw = t.rand(in_features, out_features, generator=generator)
    return (raw * 2 - 1) * sf


def _test():
    import torch as t

    g = t.Generator()
    g.manual_seed(1)
    w = kaiming_uniform_sf(16, 4, g)
    sf = 16 ** -0.5
    assert w.shape == (16, 4)
    assert w.min().item() >= -sf - 1e-6
    assert w.max().item() <=  sf + 1e-6
    # mean should be near 0
    assert abs(w.mean().item()) < sf * 0.3


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
