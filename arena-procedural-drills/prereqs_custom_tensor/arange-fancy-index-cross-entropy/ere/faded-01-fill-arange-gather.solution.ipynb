Pairing `t.arange(B)` with a `(B,)` label tensor extracts one element per row via positional zipping. The blanked line is the gather itself — the heart of the `arange` fancy-index idiom.

## Faded exercise 1

Implement `pick_target_logits(logits, target)` returning a `(B,)` tensor where entry `i` is `logits[i, target[i]]`. The function is written for you except the single advanced-indexing line. Complete it without any Python loop and without `F.cross_entropy`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def pick_target_logits(logits, target):
    B = logits.shape[0]
    picked = logits[t.arange(B), target]
    return picked

t.manual_seed(0)
logits = t.randn(4, 6)
target = t.tensor([5, 0, 3, 2])
print(pick_target_logits(logits, target))


def _test():
    t.manual_seed(1)
    logits = t.randn(7, 5)
    target = t.randint(0, 5, (7,))
    out = pick_target_logits(logits, target)
    assert out.shape == (7,), out.shape
    expected = t.tensor([logits[i, target[i]] for i in range(7)])
    assert t.allclose(out, expected), (out, expected)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
