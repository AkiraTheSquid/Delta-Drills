Stable cross-entropy is `logsumexp(logits, dim=-1) - logits[arange(B), target]`, meaned over the batch. The logsumexp term is the stable replacement for `log(sum(exp(row)))`.

## Faded exercise 1

Complete `cross_entropy(logits, target)` returning the mean stable cross-entropy. The target-logit pick and the mean are given; fill in the logsumexp first term.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def cross_entropy(logits, target):
    lse = t.logsumexp(logits, dim=-1)
    B = logits.shape[0]
    picked = logits[t.arange(B), target]
    return (lse - picked).mean()


def _test():
    t.manual_seed(21)
    logits = t.randn(8, 6)
    target = t.randint(0, 6, (8,))
    got = cross_entropy(logits, target)
    # independent ground truth: torch's own cross_entropy (mean reduction)
    import torch.nn.functional as F
    expected = F.cross_entropy(logits, target)
    assert got.shape == ()
    assert t.allclose(got, expected, atol=1e-5)
    # stability: a huge logit must not overflow
    big = t.tensor([[3000.0, 0.0, 0.0]])
    assert t.isfinite(cross_entropy(big, t.tensor([0]))).item()


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
