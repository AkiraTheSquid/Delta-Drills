## Concept

`parameters()` must yield direct Parameters AND recurse into every submodule's `parameters()`. The recursion `yield from m.parameters()` is what makes nested children's tensors discoverable from the root.

## Faded exercise 2

Complete `Module.parameters()`. The direct-parameter yield is given; fill in the loop that recurses into submodules so nested parameters are also yielded.

**Fill in:** the loop yielding from each submodule's parameters recursively

````
import numpy as np

class MiniTensor:
    def __init__(self, array, requires_grad=False):
        self.array = np.asarray(array, dtype=np.float64)
        self.requires_grad = requires_grad

class Parameter(MiniTensor):
    def __init__(self, array, requires_grad=True):
        super().__init__(array, requires_grad=requires_grad)

class Module:
    def __init__(self):
        object.__setattr__(self, '_parameters', {})
        object.__setattr__(self, '_modules', {})
    def __setattr__(self, name, value):
        if isinstance(value, Parameter):
            self._parameters[name] = value
        elif isinstance(value, Module):
            self._modules[name] = value
        object.__setattr__(self, name, value)
    def parameters(self):
        for p in self._parameters.values():
            yield p
        for m in self._modules.values():
            yield from m.parameters()


def _test():
    class Leaf(Module):
        def __init__(self):
            super().__init__()
            self.w = Parameter(np.zeros(2))
    class Mid(Module):
        def __init__(self):
            super().__init__()
            self.leaf = Leaf()
            self.bias = Parameter(np.zeros(1))
    class Root(Module):
        def __init__(self):
            super().__init__()
            self.mid = Mid()
    root = Root()
    params = list(root.parameters())
    # independent count: Mid.bias + Leaf.w = 2 params, found via 2-level recursion
    assert len(params) == 2
    sizes = sorted(p.array.size for p in params)
    assert sizes == [1, 2]


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
