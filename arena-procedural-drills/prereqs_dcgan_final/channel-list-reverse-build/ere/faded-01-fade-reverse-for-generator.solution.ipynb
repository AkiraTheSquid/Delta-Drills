## Concept

Both networks read one `hidden_channels` list. The discriminator pairs it in given order; the generator must **reverse** it first (`channels[::-1]`) before pairing, so its upsampling blocks mirror the discriminator's downsampling blocks. Forgetting the reverse leaves the generator entering at the wrong (smallest) channel width.

## Faded exercise 1

### Build both halves from one channel list

Complete `build_both_halves(hidden_channels)`. The discriminator pairs are already provided. You must produce the **generator** channel list by slice-reversing the shared `hidden_channels`, then the function will zip it into consecutive pairs. Use the slice form `[::-1]`, not `reversed(...)`, so the symmetry is explicit.

**Fill in:** the generator channel list, formed by slice-reversing hidden_channels

````
def build_both_halves(hidden_channels):
    disc_pairs = list(zip(hidden_channels[:-1], hidden_channels[1:]))
    gen_channels = hidden_channels[::-1]  # slice-reverse the shared list
    gen_pairs = list(zip(gen_channels[:-1], gen_channels[1:]))
    return {'disc_pairs': disc_pairs, 'gen_pairs': gen_pairs}


def _test():
    hidden = [16, 32, 64, 128, 256]
    out = build_both_halves(hidden)
    expected_disc = list(zip(hidden[:-1], hidden[1:]))
    rev = hidden[::-1]
    expected_gen = list(zip(rev[:-1], rev[1:]))
    assert out['disc_pairs'] == expected_disc, out['disc_pairs']
    assert out['gen_pairs'] == expected_gen, out['gen_pairs']
    # mirror invariant: gen pair (a,b) == swapped reversed disc pair
    mirror = [(c, d) for (d, c) in reversed(out['disc_pairs'])]
    assert out['gen_pairs'] == mirror, (out['gen_pairs'], mirror)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
