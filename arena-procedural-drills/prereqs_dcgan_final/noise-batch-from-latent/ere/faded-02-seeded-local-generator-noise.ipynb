## Concept

Passing a seeded `torch.Generator` to `t.randn` makes noise draws reproducible without touching the global random state. The same generator seed always yields the same batch, making runs debuggable and comparable across experiments.

```python
import torch as t

def reproducible_noise(batch_size: int, latent_dim: int, seed: int) -> t.Tensor:
    g = t.Generator()
    g.manual_seed(seed)
    return t.randn(batch_size, latent_dim, generator=g)

# --- run it ---
a = reproducible_noise(4, 50, seed=13)
b = reproducible_noise(4, 50, seed=13)
print(f'a == b: {t.allclose(a, b)}')  # True
```
