import torch
import torch.nn as nn

def count_linear_conv(model: nn.Module) -> dict:
    counts = {'linear': 0, 'conv': 0, 'other': 0}
    for m in model.modules():
        if isinstance(m, nn.Linear):
            counts['linear'] += 1
        elif isinstance(m, (nn.Conv2d, nn.ConvTranspose2d)):
            counts['conv'] += 1
        else:
            counts['other'] += 1
    return counts

def _test():
    net = nn.Sequential(
        nn.Conv2d(3, 8, 3),
        nn.ReLU(),
        nn.Flatten(),
        nn.Linear(8*6*6, 64),
        nn.ReLU(),
        nn.Linear(64, 10),
        nn.ConvTranspose2d(1, 1, 3),
    )
    c = count_linear_conv(net)
    assert c['linear'] == 2, f"expected 2 linear, got {c['linear']}"
    assert c['conv'] == 2, f"expected 2 conv, got {c['conv']}"
    assert c['other'] > 0, "other should include Sequential, ReLU, Flatten"
