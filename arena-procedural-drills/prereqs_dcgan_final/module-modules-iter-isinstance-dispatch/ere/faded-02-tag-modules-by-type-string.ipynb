import torch
import torch.nn as nn

def tag_typed_layers(model: nn.Module) -> list:
    result = []
    for idx, m in enumerate(model.modules()):
        if isinstance(m, nn.Conv2d):
            result.append((idx, 'conv2d'))
        elif isinstance(m, nn.Linear):
            result.append((idx, 'linear'))
        elif isinstance(m, nn.BatchNorm2d):
            result.append((idx, 'batchnorm2d'))
        elif isinstance(m, (nn.ReLU, nn.LeakyReLU)):
            result.append((idx, 'activation'))
        # else: skip
    return result

def _test():
    net = nn.Sequential(
        nn.Conv2d(1, 4, 3),
        nn.BatchNorm2d(4),
        nn.ReLU(),
        nn.Flatten(),
        nn.Linear(4*26*26, 10),
    )
    tags = tag_typed_layers(net)
    labels = [lbl for _, lbl in tags]
    assert 'conv2d' in labels
    assert 'batchnorm2d' in labels
    assert 'activation' in labels
    assert 'linear' in labels
