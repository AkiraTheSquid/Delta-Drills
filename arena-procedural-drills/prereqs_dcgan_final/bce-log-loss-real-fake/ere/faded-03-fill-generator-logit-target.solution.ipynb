The non-saturating generator loss is `BCE_with_logits(D_logits(fake), 1)`: G wants the discriminator to call its fakes REAL, so the target is all ones even though the inputs are fakes. Using the logit form keeps the computation numerically stable via the fused softplus kernel.

## Faded exercise 3

Implement `generator_loss(d_logits_fake)` returning the scalar non-saturating generator loss from raw logits. The `F.binary_cross_entropy_with_logits` call is provided. You must complete the **target tensor** that encodes 'G wants D to think these fakes are real'.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch.nn.functional as F

def generator_loss(d_logits_fake: t.Tensor) -> t.Tensor:
    target = t.ones_like(d_logits_fake)
    return F.binary_cross_entropy_with_logits(d_logits_fake, target)

t.manual_seed(0)
d_logits_fake = t.randn(8) * 2.0
print(round(generator_loss(d_logits_fake).item(), 6))


def _test():
    import torch.nn.functional as F
    t.manual_seed(3)
    logits = t.randn(16) * 2.5
    expected = F.softplus(-logits).mean()
    got = generator_loss(logits)
    assert got.shape == t.Size([]), f'expected scalar, got shape {got.shape}'
    assert t.allclose(got, expected, atol=1e-5), f'{got.item()} vs {expected.item()}'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
