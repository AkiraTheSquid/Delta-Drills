## BCE-with-logits — numerically stable D loss — deepening

Ex1 used `F.binary_cross_entropy(probs, targets)` — the version that expects post-sigmoid probabilities. In production GAN code you should feed LOGITS to `F.binary_cross_entropy_with_logits` instead — it fuses `sigmoid + bce` internally with the log-sum-exp trick:

```python
loss_real = F.binary_cross_entropy_with_logits(D_logits_real, t.ones_like(D_logits_real))
loss_fake = F.binary_cross_entropy_with_logits(D_logits_fake, t.zeros_like(D_logits_fake))
loss_D = loss_real + loss_fake
```

**Why this matters.** When D becomes very confident on a batch (logits at ±20), the sigmoid output is `1 - 1e-9` or `1e-9`, and `log(1 - p)` or `log(p)` plus float32 rounding crash to `inf` (or `nan`) in the bare `binary_cross_entropy` path. The fused form sidesteps that — see `softplus(x) = log(1 + exp(x))` for the exact algebra.

**Numerical equivalence.** For non-extreme logits (in `[-10, +10]`-ish), the logits form and the probs form agree to single-precision.

2. `d_logits_fake` are D's raw logits on fakes (same shape, same values range).
3. Build target tensors:
   - `real_t = t.ones_like(d_logits_real)`
   - `fake_t = t.zeros_like(d_logits_fake)`
4. Compute `loss_real = F.binary_cross_entropy_with_logits(d_logits_real, real_t)`.
5. Compute `loss_fake = F.binary_cross_entropy_with_logits(d_logits_fake, fake_t)`.
6. Return `loss_real + loss_fake` — a scalar tensor.

Returns a scalar tensor.

````python
def ex2_discriminator_loss_logits(d_logits_real: Tensor, d_logits_fake: Tensor) -> Tensor:
    """BCE-with-logits(D_real, 1) + BCE-with-logits(D_fake, 0)."""
    raise NotImplementedError()


def _test_ex2():
    import torch.nn.functional as F
    import math

    # --- invariant 1: matches probs form on moderate logits ---
    t.manual_seed(0)
    logits_real = t.randn(16) * 2.0     # in roughly (-6, 6)
    logits_fake = t.randn(16) * 2.0
    ours = ex2_discriminator_loss_logits(logits_real, logits_fake)
    # Compare to the bare-probs form on the same data
    probs_real = t.sigmoid(logits_real)
    probs_fake = t.sigmoid(logits_fake)
    ref = F.binary_cross_entropy(probs_real, t.ones_like(probs_real)) + \
          F.binary_cross_entropy(probs_fake, t.zeros_like(probs_fake))
    assert t.allclose(ours, ref, atol=1e-5), (
        f'must match probs form at moderate logits; ours={ours.item()}, ref={ref.item()}'
    )

    # --- invariant 2: STABLE at extreme logits (±50) — does not overflow ---
    extreme_real = t.tensor([50.0, 50.0, 50.0, 50.0])     # D is overwhelmingly sure 'real'
    extreme_fake = t.tensor([-50.0, -50.0, -50.0, -50.0]) # D is overwhelmingly sure 'fake'
    loss_stable = ex2_discriminator_loss_logits(extreme_real, extreme_fake)
    assert t.isfinite(loss_stable).item(), (
        f'extreme logits must NOT produce inf/nan with BCE-with-logits; got {loss_stable.item()}'
    )
    # Loss should be very small (near zero — D is correct on every example).
    assert loss_stable.item() < 1e-15, (
        f'D correct at extreme logits should give ~0 loss, got {loss_stable.item()}'
    )

    # --- invariant 3: D MAXIMALLY WRONG at extreme logits → finite, large loss ---
    wrong_real = t.tensor([-50.0, -50.0])   # D says 'fake' for reals
    wrong_fake = t.tensor([50.0, 50.0])     # D says 'real' for fakes
    loss_wrong = ex2_discriminator_loss_logits(wrong_real, wrong_fake)
    assert t.isfinite(loss_wrong).item(), f'must be finite, got {loss_wrong.item()}'
    # For extreme wrong logits, per-sample loss ~= |logit|, summed over 2 batches of 2 -> ~100
    assert loss_wrong.item() > 50.0, f'wrong-extreme loss should be large, got {loss_wrong.item()}'

    # --- invariant 4: matches torch's bce-with-logits witness exactly ---
    rnd_r = t.randn(8)
    rnd_f = t.randn(8)
    got = ex2_discriminator_loss_logits(rnd_r, rnd_f)
    ref2 = F.binary_cross_entropy_with_logits(rnd_r, t.ones_like(rnd_r)) + \
           F.binary_cross_entropy_with_logits(rnd_f, t.zeros_like(rnd_f))
    assert t.allclose(got, ref2, atol=1e-6), f'must match F.binary_cross_entropy_with_logits exactly'

    # --- invariant 5: zero logits -> 2 * log(2) (coin-flip case) ---
    zero_r = t.zeros(8)
    zero_f = t.zeros(8)
    loss_coin = ex2_discriminator_loss_logits(zero_r, zero_f)
    expected_coin = 2 * math.log(2)
    assert abs(loss_coin.item() - expected_coin) < 1e-5, (
        f'logits=0 → coin-flip loss 2*log(2) = {expected_coin:.4f}, got {loss_coin.item()}'
    )
    _dd_passed.add('ex2')
    print("ex2 ✓")

_test_ex2()
````

```python
def ex2_discriminator_loss_logits(d_logits_real: Tensor, d_logits_fake: Tensor) -> Tensor:
    import torch.nn.functional as F
    loss_real = F.binary_cross_entropy_with_logits(d_logits_real, t.ones_like(d_logits_real))
    loss_fake = F.binary_cross_entropy_with_logits(d_logits_fake, t.zeros_like(d_logits_fake))
    return loss_real + loss_fake
```

**Why the fused form is stable.** `bce_with_logits(x, 1) = -log(sigmoid(x)) = softplus(-x)`, and `softplus` is implemented as `log1p(exp(-|x|)) + max(x, 0)` — the absolute-value trick guarantees `exp` always sees a non-positive argument. So `softplus(50) ≈ 50` (computed exactly) while `-log(sigmoid(50)) = -log(1 - 1e-22)` underflows to `inf` in float32.

**When to use which.** Always prefer `binary_cross_entropy_with_logits` in production. The plain `binary_cross_entropy` is occasionally useful for plotting losses against PREDICTED probabilities (interpretability), but never inside the training loop.

**Gradient is also stable.** `d/dx bce_with_logits(x, target) = sigmoid(x) - target`, computed without ever materializing `1 - sigmoid(x)` near 1.
