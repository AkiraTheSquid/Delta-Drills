## Concept

After applying the DCGAN `N(0, 0.02)` init, an audit counts how many scalar weights were resampled: the sum of `m.weight.numel()` over every Conv2d / ConvTranspose2d submodule. The init call is written; you compute and accumulate the count.

## Faded exercise 3

### Complete the counter

The loop already walks `model.modules()` and re-inits each Conv2d / ConvTranspose2d weight from `N(0, 0.02)`. Complete the blanked line so `total` accumulates the number of scalar parameters in the weight tensor that was just resampled. Return `total`.

**Fill in:** Adds the number of scalar elements in the just-resampled conv weight tensor (m.weight.numel()) to the running total.

````
def reinit_and_count(model: nn.Module) -> int:
    t.manual_seed(0)
    total = 0
    for m in model.modules():
        if isinstance(m, (nn.Conv2d, nn.ConvTranspose2d)):
            nn.init.normal_(m.weight, 0.0, 0.02)
            added = m.weight.numel()
            total += added
    return total

net = nn.Sequential(nn.Conv2d(8, 16, 3), nn.ConvTranspose2d(16, 4, 4), nn.Linear(4, 2))
result = reinit_and_count(net)


def _test():
    expected = 16 * 8 * 3 * 3 + 16 * 4 * 4 * 4  # 1152 + 1024 = 2176
    assert result == expected, f'expected {expected}, got {result}'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
