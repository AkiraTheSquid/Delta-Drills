The DCGAN init handles Conv/ConvTranspose weights with `N(0, 0.02)` and BatchNorm2d affine parameters separately: `weight` (gamma) from `N(1.0, 0.02)` and `bias` (beta) set to 0. Here the conv branch is written for you; you complete the BatchNorm branch.

## Faded exercise 1

### Complete the BatchNorm branch

The `init_fn` already initializes Conv2d / ConvTranspose2d weights from `N(0, 0.02)`. Fill in the BatchNorm2d branch so that the affine `weight` is drawn from `N(1.0, 0.02)` and the `bias` is set to the constant 0.0. Then `model.apply(init_fn)` walks the model.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def dcgan_init(model: nn.Module) -> nn.Module:
    def init_fn(m):
        if isinstance(m, (nn.Conv2d, nn.ConvTranspose2d)):
            nn.init.normal_(m.weight, 0.0, 0.02)
        elif isinstance(m, nn.BatchNorm2d):
            nn.init.normal_(m.weight, 1.0, 0.02)
            nn.init.constant_(m.bias, 0.0)
    model.apply(init_fn)
    return model

t.manual_seed(0)
net = nn.Sequential(nn.Conv2d(8, 16, 3), nn.BatchNorm2d(16))
dcgan_init(net)


def _test():
    bn = net[1]
    # bias must be exactly zero
    assert t.equal(bn.bias.detach(), t.zeros_like(bn.bias)), 'BN bias must be all zeros'
    # weight mean should be near 1.0 (N(1.0, 0.02))
    assert abs(bn.weight.mean().item() - 1.0) < 0.05, 'BN weight mean should be ~1.0'
    # conv weight std near 0.02
    assert abs(net[0].weight.std().item() - 0.02) < 0.01, 'conv std should be ~0.02'


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
