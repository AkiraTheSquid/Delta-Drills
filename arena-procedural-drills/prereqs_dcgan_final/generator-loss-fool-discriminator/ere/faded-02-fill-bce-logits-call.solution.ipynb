`F.binary_cross_entropy_with_logits` fuses sigmoid and BCE for numerical stability. With an all-ones target it gives the non-saturating generator loss directly from D's logits.

## Faded exercise 2

Implement `generator_loss_logits(d_logits)`. `d_logits` are D's pre-sigmoid outputs on the fakes, shape `(B,)`. Build the all-ones target and return the stable BCE-with-logits loss. Complete the blanked functional call.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t
import torch.nn.functional as F

t.manual_seed(4)

def generator_loss_logits(d_logits):
    targets = t.ones_like(d_logits)
    loss = F.binary_cross_entropy_with_logits(d_logits, targets)
    return loss

print(float(generator_loss_logits(t.tensor([0.0, 1.0, -1.0]))))


def _test():
    logits = t.tensor([-2.0, -0.5, 0.0, 1.5, 3.0])
    out = generator_loss_logits(logits)
    # independent ground truth: softplus(-logit) = log(1+exp(-logit)), then mean
    expected = t.log1p(t.exp(-logits)).mean()
    assert t.allclose(out, expected, atol=1e-5), (out, expected)
    # stays finite for very large positive logit
    big = generator_loss_logits(t.full((4,), 50.0))
    assert t.isfinite(big).all()


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
