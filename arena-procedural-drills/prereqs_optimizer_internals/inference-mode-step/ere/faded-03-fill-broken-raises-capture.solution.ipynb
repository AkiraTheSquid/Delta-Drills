## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

An undecorated `step` doing `p -= lr * p.grad` on a leaf raises a RuntimeError. Capturing it in a `try/except RuntimeError` lets you inspect the characteristic leaf-in-place message; the fix is to add the decorator.

## Faded exercise 3

`BrokenSGD.step` lacks the decorator and raises on the in-place leaf update. Implement `demonstrate_broken_raises(opt)` that calls `opt.step()` inside a `try/except RuntimeError` and returns the exception message (empty string if it somehow does not raise). Complete the blanked except clause.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

t.manual_seed(5)

class BrokenSGD:
    def __init__(self, params, lr):
        self.params = list(params)
        self.lr = lr
    def step(self):
        for p in self.params:
            if p.grad is not None:
                p -= self.lr * p.grad

def demonstrate_broken_raises(opt):
    try:
        opt.step()
        return ''
    except RuntimeError as e:
        return str(e)

w = t.tensor([2.0], requires_grad=True)
((w - 1) ** 2).backward()
print(bool(demonstrate_broken_raises(BrokenSGD([w], 0.1))))


def _test():
    w = t.tensor([2.0], requires_grad=True)
    ((w - 1) ** 2).backward()
    msg = demonstrate_broken_raises(BrokenSGD([w], 0.1))
    # a non-empty error message must have been captured (the leaf-in-place RuntimeError)
    assert isinstance(msg, str) and len(msg) > 0, repr(msg)
    assert 'leaf' in msg.lower() or 'inplace' in msg.lower() or 'in-place' in msg.lower(), msg


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
