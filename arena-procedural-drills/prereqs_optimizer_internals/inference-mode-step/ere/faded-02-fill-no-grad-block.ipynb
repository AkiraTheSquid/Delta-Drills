## Concept

Wrapping the step body in `with t.no_grad():` disables autograd for that scope, permitting the in-place leaf update without a decorator. It is behaviorally equivalent to `@t.inference_mode()` for this pattern.

## Faded exercise 2

Implement `ContextSGD`. The `step` method must wrap the in-place update loop in a `no_grad` context (no decorator). Complete the blanked context-manager line that opens the no-grad scope.

**Fill in:** the 'with t.no_grad():' context manager wrapping the in-place update loop

````
import torch as t

t.manual_seed(4)

class ContextSGD:
    def __init__(self, params, lr):
        self.params = list(params)
        self.lr = lr

    def step(self):
        raise NotImplementedError()  # TODO: the 'with t.no_grad():' context manager wrapping the in-place update loop
        for p in self.params:
            if p.grad is not None:
                p -= self.lr * p.grad

    def zero_grad(self):
        for p in self.params:
            p.grad = None

w = t.tensor([4.0], requires_grad=True)
opt = ContextSGD([w], 0.1)
((w - 1) ** 2).backward()
opt.step()
print(float(w))


def _test():
    w = t.tensor([4.0], requires_grad=True)
    opt = ContextSGD([w], 0.1)
    ((w - 1) ** 2).backward()
    opt.step()
    # independent closed form: 4 - 0.1*(2*(4-1)) = 3.4
    assert abs(float(w) - 3.4) < 1e-5, float(w)
    # step must not be a decorated function (context-manager form, no __wrapped__)
    assert not hasattr(ContextSGD.step, '__wrapped__')


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
