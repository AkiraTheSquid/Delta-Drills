For a dataset of size `N` and batch size `B`, a loader with `drop_last=True` yields `N // B` full batches and discards the remainder, while `drop_last=False` yields `ceil(N / B)` batches — the last one possibly partial. Counting the batches actually produced confirms the floor-vs-ceil rule.

Implement `count_batches(N, B)`. Build `TensorDataset(t.arange(N).float())` and wrap it twice with `shuffle=False`: once `drop_last=True`, once `drop_last=False`. **Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
from torch.utils.data import TensorDataset, DataLoader


def count_batches(N, B):
    ds = TensorDataset(t.arange(N).float())
    n_kept = sum(1 for _ in DataLoader(ds, batch_size=B, shuffle=False, drop_last=True))
    n_all = sum(1 for _ in DataLoader(ds, batch_size=B, shuffle=False, drop_last=False))
    return {'kept': n_kept, 'all': n_all}


def _test():
    import math
    for N, B in [(23, 5), (20, 4), (1, 8), (37, 10)]:
        out = count_batches(N, B)
        assert out['kept'] == N // B, (N, B, out)
        assert out['all'] == math.ceil(N / B), (N, B, out)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
