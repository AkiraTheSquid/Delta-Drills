## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

After accumulating the second-moment estimate `v`, Adam computes a per-coordinate step scale as `1 / (sqrt(v) + eps)`. This scale is small for parameters with large gradient history (large `v`) and large for parameters with small gradient history (small `v`). The actual update is `param -= lr * m_hat * step_scale`, where `m_hat` is the bias-corrected first moment.

## Faded exercise 3

Implement `compute_step_scales(v_tensor, eps)` that takes a 1-D `v` tensor (the second-moment buffer) and returns a dict with `'step_scale'` (tensor), `'argmax_scale_idx'` (int index of largest scale = smallest v entry), and `'argmin_scale_idx'` (int index of smallest scale = largest v entry).

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def compute_step_scales(v_tensor, eps: float) -> dict:
    step_scale = 1.0 / (v_tensor.sqrt() + eps)
    argmax_scale_idx = v_tensor.argmin().item()
    argmin_scale_idx = v_tensor.argmax().item()
    return {
        'step_scale': step_scale,
        'argmax_scale_idx': argmax_scale_idx,
        'argmin_scale_idx': argmin_scale_idx,
    }


def _test():
    import torch as t
    v = t.tensor([0.0001, 1.0, 100.0])
    eps = 1e-8
    result = compute_step_scales(v, eps)
    expected_scale = 1.0 / (v.sqrt() + eps)
    assert t.allclose(result['step_scale'], expected_scale, atol=1e-6)
    assert result['argmax_scale_idx'] == 0  # smallest v => largest scale
    assert result['argmin_scale_idx'] == 2  # largest v => smallest scale


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
