L2 weight decay augments the gradient with `lmda * theta`, returning a new tensor. The guard `if lmda != 0` bypasses the fold when decay is off, returning the original gradient object untouched.

## Faded exercise 1

### Fold decay into the gradient

Implement `apply_weight_decay(theta, g, lmda)`: when `lmda` is nonzero, return the gradient augmented with the L2 term (a new tensor); otherwise return `g` unchanged. Complete the fold expression assigned to `g`.

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
import torch as t

def apply_weight_decay(theta, g, lmda):
    if lmda != 0:
        g = g + lmda * theta
    return g

print(apply_weight_decay(t.tensor([2.0]), t.tensor([1.0]), 0.5).tolist())


def _test():
    theta = t.tensor([1.0, -2.0, 3.0])
    g = t.tensor([0.5, 0.5, 0.5])
    out = apply_weight_decay(theta, g, 0.1)
    # independent ground truth computed by hand
    expected = t.tensor([0.5 + 0.1 * 1.0, 0.5 + 0.1 * -2.0, 0.5 + 0.1 * 3.0])
    assert t.allclose(out, expected), out
    # bypass path returns the SAME object, not a copy
    bypass = apply_weight_decay(theta, g, 0.0)
    assert bypass is g
    # original g not mutated by the fold
    assert t.allclose(g, t.tensor([0.5, 0.5, 0.5]))


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
