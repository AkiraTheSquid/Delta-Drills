## Concept

Because the fold adds `lmda * theta`, the augmented gradient gains a component with the SAME sign as the weight. After a gradient-descent step `theta - lr * g_aug`, that component shrinks the weight toward zero — positive weights decrease, negative weights increase. This is the regularizing effect of L2 decay.

## Worked solution

We show one decayed SGD step shrinks the magnitude of a weight.

1. We start from a parameter `theta` and a gradient `g`, then fold in decay: `g_aug = g + lmda * theta`.
2. We take a plain SGD step `theta_new = theta - lr * g_aug`.
3. With a tiny base gradient, the decay term dominates and the absolute value of each weight drops compared to the start.

We print the before/after magnitudes to confirm the weights moved toward zero.

```
import torch as t

def decayed_step(theta, g, lr, lmda):
    g_aug = g + lmda * theta
    return theta - lr * g_aug

theta = t.tensor([4.0, -4.0])
g = t.tensor([0.0, 0.0])  # pure decay, no data gradient
theta_new = decayed_step(theta, g, lr=0.1, lmda=0.5)
print('before |theta|:', theta.abs().tolist())
print('after  |theta|:', theta_new.abs().tolist())
print('shrank:', bool((theta_new.abs() < theta.abs()).all()))
```
