## Concept

Writing `mat[i, :] *= -1` negates every element in row i in-place. The slice expression `mat[i, :]` returns a view; multiplying in-place via `*=` modifies the underlying storage without rebinding `mat`. This is equivalent to calling `mat[i].neg_()` but uses the slice-view pattern explicitly.

## Faded exercise 1

Given a (4, 3) matrix, negate row 2 in-place using the slice-view `*= -1` pattern. Return the modified matrix.

1. Extract row 2 as a view.
2. Apply the in-place negation through the view.

The blank step is the in-place negation of the row view.

````
import torch as t

t.manual_seed(5)

mat = t.arange(12, dtype=t.float32).reshape(4, 3)
row2_before = mat[2].clone()
row_view = mat[2, :]
row_view *= -1

print('Row 2 before:', row2_before.tolist())
print('Row 2 after :', mat[2].tolist())
print('Other rows unchanged (row 0):', mat[0].tolist())


def _test():
    import torch as t

    mat_ref = t.arange(12, dtype=t.float32).reshape(4, 3)
    expected_row2 = -mat_ref[2].clone()
    expected_row0 = mat_ref[0].clone()

    assert t.allclose(mat[2], expected_row2), f'row2={mat[2]} expected={expected_row2}'
    assert t.allclose(mat[0], expected_row0), 'row0 should be unchanged'
    assert t.allclose(mat[1], mat_ref[1]), 'row1 should be unchanged'
    assert t.allclose(mat[3], mat_ref[3]), 'row3 should be unchanged'


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
