## Concept

Slice views work for multi-dimensional indexing too: `mat[r_start:r_end, c_start:c_end]` returns a 2-D view of a rectangular block. Assigning a tensor of the right shape into this block mutates the underlying storage. This is how you implement in-place copy of one region into another without allocating new memory.

## Worked solution

**Step 1 — Create a 5×5 matrix of zeros.** We'll write a 2×2 block into its center.

**Step 2 — Create the block to insert.** A small 2×2 tensor with recognizable values.

**Step 3 — Assign the block via slice view.** `mat[1:3, 2:4] = block` writes the four values directly into the center of `mat`.

**Step 4 — Verify.** The center block should match; everything outside should be 0.

```
import torch as t

t.manual_seed(0)

mat = t.zeros(5, 5)
block = t.tensor([[7.0, 8.0], [9.0, 10.0]])

print('Before:\n', mat)
mat[1:3, 2:4] = block
print('After:\n', mat)

assert t.allclose(mat[1:3, 2:4], block)
assert mat[0, :].sum().item() == 0.0, 'row 0 should still be zeros'
assert mat[3, :].sum().item() == 0.0, 'row 3 should still be zeros'
print('Block inserted correctly. Surrounding zeros preserved.')
```
