## Concept

You can copy values between columns of the same matrix using two slice views: `mat[:, dst] = mat[:, src].clone()`. The `.clone()` on the source creates a detached copy before the assignment, preventing aliasing issues when src and dst overlap (though for distinct indices, it's not strictly necessary). Without `.clone()`, this is still correct for distinct columns.
