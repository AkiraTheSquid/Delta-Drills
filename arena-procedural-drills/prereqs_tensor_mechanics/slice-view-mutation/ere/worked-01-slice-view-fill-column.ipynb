## Concept

In PyTorch, `mat[:, k]` returns a 1-D view that shares storage with the original matrix. Assigning into this view — `mat[:, k] = value` — writes directly into the original matrix's memory. No copy is made; the data pointer of `mat` is unchanged before and after the assignment.

## Worked solution

**Step 1 — Create a 4×4 matrix.** We fill it with recognizable values so we can verify which entries changed.

**Step 2 — Extract the second column as a view.** `mat[:, 1]` has shape (4,). We verify it shares storage using `data_ptr()`.

**Step 3 — Write through the view.** Assigning `col[:] = 99.0` modifies all four entries in column 1 of `mat`. The original `mat` object is unchanged; only its data changed.

**Step 4 — Print the before/after matrix.** Column 1 should now be all 99s; all other entries should be untouched.

```
import torch as t

t.manual_seed(0)

mat = t.arange(16, dtype=t.float32).reshape(4, 4)
print('Before:\n', mat)

col = mat[:, 1]  # view of column 1
print('col shares storage with mat?', col.storage().data_ptr() == mat.storage().data_ptr())  # True (same storage, different offset)

col[:] = 99.0  # write through the view

print('After:\n', mat)
print('Column 1 is all 99s:', (mat[:, 1] == 99.0).all().item())
print('Other columns unchanged (spot check col 0):', mat[:, 0].tolist())
```
