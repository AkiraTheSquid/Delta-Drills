## Concept

You can confirm that a slice is a view (sharing storage) by comparing `data_ptr()` of the slice against the original tensor. For basic indexing like `mat[:, k]` or `mat[i, :]`, PyTorch always returns a view. For advanced indexing (e.g., `mat[[0, 2], :]`), PyTorch returns a copy. Understanding this difference prevents subtle bugs when you expect mutations to propagate.

## Faded exercise 3

Given a matrix, extract a row slice and a diagonal view, and check whether each shares storage with the original.

1. `row_view = mat[1, :]`
2. `diag_view = mat.diagonal()`
3. Check storage aliasing for each.

The blank step is computing the aliasing boolean for `diag_view`.

**Fill in:** Compute diag_alias by comparing diag_view.storage().data_ptr() to mat.storage().data_ptr().

````
import torch as t

t.manual_seed(0)

mat = t.arange(9, dtype=t.float32).reshape(3, 3)
row_view = mat[1, :]
diag_view = mat.diagonal()

row_alias = (row_view.storage().data_ptr() == mat.storage().data_ptr())
diag_alias = (diag_view.storage().data_ptr() == mat.storage().data_ptr())

print('row_view is alias:', row_alias)
print('diag_view is alias:', diag_alias)
diag_view[:] = 0.0
print('mat after zeroing diagonal:\n', mat)


def _test():
    import torch as t

    mat_ref = t.arange(9, dtype=t.float32).reshape(3, 3)
    diag_ref = mat_ref.diagonal()
    assert diag_alias == (diag_ref.storage().data_ptr() == mat_ref.storage().data_ptr()), f'diag_alias={diag_alias}'
    assert diag_alias == True, 'diagonal() should return a view (same storage)'
    assert row_alias == True, 'row slice should return a view'


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
