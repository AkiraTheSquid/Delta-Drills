To extract every `KxK` patch from a contiguous `(H, W)` image with `as_strided`, the view has shape `(H-K+1, W-K+1, K, K)` and stride `(sH, sW, sH, sW)`: the outer two strides move the patch origin one row/column, the inner two walk within a patch. This is the core im2col primitive behind `Conv2d`.

### Faded — im2col patch extraction via as_strided

Implement `image_patches(img, K)` for a contiguous `(H, W)` image. Return a single zero-copy view of all `KxK` patches with stride-1 in both spatial dims. The stride tuple `(sH, sW, sH, sW)` and the `as_strided` call are already written. **You must complete the `size` tuple.**

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def image_patches(img: Tensor, K: int) -> Tensor:
    H, W = img.shape
    out_H, out_W = H - K + 1, W - K + 1
    sH, sW = img.stride()
    size = (out_H, out_W, K, K)
    return img.as_strided(size=size, stride=(sH, sW, sH, sW))


def _test():
    t.manual_seed(0)
    img = t.arange(36, dtype=t.float32).reshape(6, 6)
    K = 3
    patches = image_patches(img, K)
    out_H, out_W = 6 - K + 1, 6 - K + 1
    assert patches.shape == (out_H, out_W, K, K), patches.shape
    for i in range(out_H):
        for j in range(out_W):
            assert t.equal(patches[i, j], img[i:i + K, j:j + K]), (i, j)


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
