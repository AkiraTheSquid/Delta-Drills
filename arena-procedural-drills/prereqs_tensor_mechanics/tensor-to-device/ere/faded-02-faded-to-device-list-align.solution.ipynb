## Concept

When working with a list of tensors that may be on different devices, `.to(device)` is idempotent: calling it on a tensor already on the target device returns the same object without copying. A list comprehension `[t.to(device) for t in tensors]` efficiently aligns all tensors in one line.

**Fill in:** Return `[x.to(device) for x in tensors]`.

````
import torch as t

def align_tensors(tensors: list, device: str) -> list:
    return [x.to(device) for x in tensors]

tensors = [t.randn(2), t.randn(3, 3), t.randn(1)]
result = align_tensors(tensors, 'cpu')
for r in result:
    print(r.device, r.shape)


def _test():
    import torch as t
    tensors = [t.randn(2), t.randn(3), t.randn(4)]
    result = align_tensors(tensors, 'cpu')
    assert isinstance(result, list), 'must return a list'
    assert len(result) == 3
    for r, orig in zip(result, tensors):
        assert str(r.device) == 'cpu'
        assert r.shape == orig.shape
    # Idempotence: already-cpu tensors should be the SAME object
    for r, orig in zip(result, tensors):
        assert r is orig, 'idempotent .to() should return same object'


try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
