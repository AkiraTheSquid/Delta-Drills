## Concept

`.to(device=..., dtype=...)` moves a tensor to a device and casts its dtype in a single operation. This is more efficient than two separate `.to()` calls. It is the standard pattern for converting float64 NumPy data to float32 PyTorch tensors on the target device.

**Fill in:** Pass `device=device` and `dtype=t.float32` as keyword arguments to `.to()`.

````
import torch as t

def cast_and_move(x: t.Tensor, device: str) -> t.Tensor:
    """Move x to device and cast to float32, in one .to() call."""
    return x.to(device=device, dtype=t.float32)

x64 = t.tensor([1.0, 2.0, 3.0], dtype=t.float64)
result = cast_and_move(x64, 'cpu')
print('dtype:', result.dtype)    # float32
print('device:', result.device)  # cpu


def _test():
    import torch as t
    x = t.tensor([1.0, 2.0, 3.0], dtype=t.float64)
    r = cast_and_move(x, 'cpu')
    assert r.dtype == t.float32, f'expected float32, got {r.dtype}'
    assert str(r.device) == 'cpu'
    assert t.allclose(r, x.float())


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
