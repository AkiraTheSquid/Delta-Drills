## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

A `(shape, stride)` pair (in elements) is contiguous iff the strides match the reversed-prefix-product formula `expected[-1] = 1`, `expected[k] = expected[k+1] * shape[k+1]`, with one exception: any axis of size 1 is a free dimension whose stride is irrelevant and is skipped in the comparison.

## Faded exercise 3

Implement `is_contiguous_from_meta(shape, stride)` returning `True` iff a tensor with that metadata would be contiguous, without building any tensor. The edge-case handling and the loop are given; complete the **expected-stride comparison for non-size-1 axes** (skip axes of size 1).

**Your task:** complete the one blanked step in the code cell below. The surrounding code, function signatures, and variable names are given — work out the missing expression yourself, then run the test.

````
def is_contiguous_from_meta(shape, stride):
    if len(shape) != len(stride):
        return False
    if len(shape) == 0:
        return True
    if any(s == 0 for s in shape):
        return True
    expected_stride = 1
    for k in range(len(shape) - 1, -1, -1):
        if shape[k] != 1:
            if stride[k] != expected_stride:
                return False
        expected_stride *= shape[k]
    return True

print(is_contiguous_from_meta((2, 3, 4), (12, 4, 1)))
print(is_contiguous_from_meta((2, 3, 4), (12, 1, 4)))


def _test():
    cases = [
        ((2, 3, 4), (12, 4, 1), True),
        ((2, 3, 4), (12, 1, 4), False),
        ((2, 3, 4), (1, 4, 12), False),
        ((4, 1, 5), (5, 99, 1), True),
        ((4, 1, 5), (5, 1, 1), True),
        ((), (), True),
        ((2, 3), (3, 1, 1), False),
    ]
    for shape, stride, want in cases:
        got = is_contiguous_from_meta(shape, stride)
        assert got == want, (shape, stride, got, want)
    for shape, stride in [((2, 3, 4), (12, 4, 1)), ((4, 1, 5), (5, 99, 1))]:
        real = t.zeros(shape).as_strided(shape, stride).is_contiguous()
        assert is_contiguous_from_meta(shape, stride) == real, (shape, stride)


try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
