`transpose` returns a non-contiguous view, and `view` rejects non-contiguous inputs. The fix is to call `.contiguous()` to copy the data into a fresh row-major buffer that matches the transposed logical shape, *then* `view`.

def flatten_after_transpose(x):

    out = x.transpose(-1, -2).contiguous().view(B, W * H)
