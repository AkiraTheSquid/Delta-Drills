## Concept

For a contiguous row-major tensor, `stride[-1] = 1` and every more-outer axis's stride is the product of all sizes to its right. The clean way to compute this is a right-to-left walk that records the current running product as the stride, then multiplies it by the axis size.

## Faded exercise 1

Implement `contiguous_strides(shape)` so it returns the tuple of strides a contiguous tensor of that `shape` would have, using a single right-to-left pass. The loop skeleton is given; you must fill in the **update of the running product** after each stride is recorded.

**Fill in:** Multiply the running product by the current axis size so the next (more-outer) stride is the product of all sizes to its right.

````
def contiguous_strides(shape: tuple) -> tuple:
    strides = []
    running = 1
    for dim in reversed(shape):
        strides.append(running)
        running *= dim
    return tuple(reversed(strides))

shape = (3, 1, 6, 2)
result = contiguous_strides(shape)
print(result)


def _test():
    for shape in [(3, 1, 6, 2), (2, 5, 3, 4), (7,), (4, 4)]:
        expected = tuple(t.zeros(shape).stride())
        got = contiguous_strides(shape)
        assert got == expected, (shape, got, expected)


try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
