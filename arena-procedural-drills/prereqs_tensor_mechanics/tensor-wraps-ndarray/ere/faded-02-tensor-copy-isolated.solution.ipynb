## Concept

`t.tensor(arr)` copies the data into fresh storage, producing a tensor independent of the source ndarray. A later mutation of the array does not affect the copy.

## Faded exercise 2

Complete `copy_wrap(arr)` to produce an independent copy of the ndarray. Fill in the copying torch factory call.

**Fill in:** wrap arr with the copying (independent) torch factory

````
import numpy as np


def copy_wrap(arr):
    copied = t.tensor(arr)
    return copied


arr = np.array([7.0, 8.0], dtype=np.float32)
c = copy_wrap(arr)

def _test():
    import numpy as np
    arr = np.array([7.0, 8.0], dtype=np.float32)
    c = copy_wrap(arr)
    before = c[0].item()
    arr[0] = -5.0
    after = c[0].item()
    assert before == 7.0, 'initial value must match source'
    assert after == 7.0, 'copy must be independent of the source mutation'

try:
    _test()
    _dd_passed.add('faded2')
    print('[Delta Drills] faded2 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
