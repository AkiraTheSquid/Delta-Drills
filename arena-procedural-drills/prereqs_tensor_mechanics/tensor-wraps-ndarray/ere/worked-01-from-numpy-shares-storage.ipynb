## Concept

`t.from_numpy(arr)` wraps a NumPy array WITHOUT copying — the resulting tensor and the ndarray share the same memory buffer. Mutating one is visible in the other. This is the cheap, aliasing path for getting numpy data into torch.

## Worked solution

We build a float32 ndarray and wrap it with `t.from_numpy`. We then mutate the ndarray in place (`arr[0] = 99`) and read the tensor's first element: it reflects the change, because both views point at the same storage. We print the tensor before and after the mutation to show the alias. This is why `from_numpy` is fast — zero copy — but also why an in-place edit to either side leaks across.

```
import numpy as np


arr = np.array([1.0, 2.0, 3.0], dtype=np.float32)
wrapped = t.from_numpy(arr)
before = wrapped[0].item()
arr[0] = 99.0
after = wrapped[0].item()
print('tensor[0] before mutation:', before)
print('tensor[0] after  mutation:', after)
print('shares storage:', after != before)
```
