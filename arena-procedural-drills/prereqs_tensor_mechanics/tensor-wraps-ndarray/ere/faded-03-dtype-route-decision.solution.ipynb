## Concept

The routing policy aliases via `from_numpy` only when the array's natural torch dtype already matches the target; otherwise it copies-and-casts with `t.tensor(arr, dtype=target)`. The decision hinges on comparing the natural dtype to the target.

## Faded exercise 3

The cast branch is filled. Complete the condition that decides whether the array can be aliased (its natural torch dtype matches the target).

**Fill in:** the condition that is true when arr's natural torch dtype equals target_dtype

````
import numpy as np


def wrap(arr, target_dtype):
    natural = t.from_numpy(arr).dtype
    can_alias = (natural == target_dtype)
    if can_alias:
        return t.from_numpy(arr), True
    return t.tensor(arr, dtype=target_dtype), False


arr = np.array([1.0, 2.0], dtype=np.float32)
same = wrap(arr, t.float32)
cast = wrap(arr, t.float64)

def _test():
    import numpy as np
    arr = np.array([1.0, 2.0], dtype=np.float32)
    tensor_same, aliased_same = wrap(arr, t.float32)
    tensor_cast, aliased_cast = wrap(arr, t.float64)
    assert aliased_same is True, 'matching dtype must alias'
    assert aliased_cast is False, 'mismatched dtype must copy-and-cast'
    assert tensor_cast.dtype == t.float64, 'cast must produce target dtype'
    # alias tracks mutation; cast does not
    arr[0] = 100.0
    assert tensor_same[0].item() == 100.0
    assert tensor_cast[0].item() == 1.0

try:
    _test()
    _dd_passed.add('faded3')
    print('[Delta Drills] faded3 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
