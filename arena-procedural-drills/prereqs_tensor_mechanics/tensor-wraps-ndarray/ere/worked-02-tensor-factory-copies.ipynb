## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

`t.tensor(arr)` always COPIES the data into a fresh buffer, so the resulting tensor is independent of the source ndarray. Mutating the ndarray afterward does not affect the tensor. This is the safe, isolating path — the opposite of `from_numpy`.

## Worked solution

We wrap a float32 ndarray with `t.tensor(arr)`, recording the first element. We then mutate `arr[0]`. Reading the tensor shows the original value, unchanged, because `t.tensor` copied at construction. We print both copies to contrast with `from_numpy`: here the tensor is a snapshot, immune to later edits of the source.

```
import numpy as np


arr = np.array([4.0, 5.0, 6.0], dtype=np.float32)
copied = t.tensor(arr)
before = copied[0].item()
arr[0] = -1.0
after = copied[0].item()
print('tensor[0] before mutation:', before)
print('tensor[0] after  mutation:', after)
print('is independent copy:', after == before)
```
