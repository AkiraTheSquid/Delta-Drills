## Concept

A small policy chooses the wrapper: if the ndarray's natural torch dtype already equals the target, use `t.from_numpy` (aliased, no copy); otherwise use `t.tensor(arr, dtype=target)` to copy-and-cast. The natural dtype can be probed via `t.from_numpy(arr).dtype` without committing.

## Worked solution

We define `wrap(arr, target)` that checks whether the array's natural torch dtype matches `target`. For a float32 array targeting `t.float32`, the dtypes match, so we alias via `from_numpy` (no copy). For the same array targeting `t.float64`, they differ, so we copy-and-cast with `t.tensor`. We report whether each result aliases the source by mutating `arr` and checking. We print the aliasing flag for both the matching and mismatching targets.

```
import numpy as np


def wrap(arr, target_dtype):
    natural = t.from_numpy(arr).dtype
    if natural == target_dtype:
        return t.from_numpy(arr), True
    return t.tensor(arr, dtype=target_dtype), False


arr = np.array([1.0, 2.0], dtype=np.float32)
same, aliased_same = wrap(arr, t.float32)
cast, aliased_cast = wrap(arr, t.float64)
print('float32 target aliased:', aliased_same)
print('float64 target aliased:', aliased_cast)
print('cast dtype:', cast.dtype)
```
