## Concept

_First time on this topic? Run the **Setup** cell above and skim it: every class and helper mentioned below is defined there. You don't need to have done any other drill first._

`t.stack` introduces a NEW axis, so stacking N vectors of shape `(D,)` along `dim=0` yields a `(N, D)` matrix where each input becomes one row. `t.cat` would instead require an existing axis to join along. Choose stack when you are assembling separate vectors into rows or columns of a fresh matrix.

## Worked solution

We have three `(4,)` vectors. `t.stack(vectors, dim=0)` adds a leading axis of size 3, producing a `(3, 4)` matrix whose row i is the i-th vector. Had we used `dim=1`, the new axis would be inserted second, giving `(4, 3)` — each input becoming a column. We print the stacked shape and the matrix to confirm each vector occupies one row. The key contrast: stack changed the rank from 1 to 2, which cat cannot do.

```
vectors = [t.tensor([1.0, 2.0, 3.0, 4.0]),
           t.tensor([5.0, 6.0, 7.0, 8.0]),
           t.tensor([9.0, 10.0, 11.0, 12.0])]
rows = t.stack(vectors, dim=0)
cols = t.stack(vectors, dim=1)
print('stack dim=0 shape:', tuple(rows.shape))
print('stack dim=1 shape:', tuple(cols.shape))
print('row 0:', rows[0].tolist())
```
