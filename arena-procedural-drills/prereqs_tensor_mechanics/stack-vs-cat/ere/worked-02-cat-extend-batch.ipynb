## Setup

```
import numpy as np
import torch as t
from torch import Tensor
import einops
from einops import rearrange, reduce, repeat

t.manual_seed(0)
np.random.seed(0)
```

## Concept

`t.cat` joins tensors along an EXISTING axis, keeping the rank the same but growing that axis. Concatenating two `(B, D)` batches along `dim=0` gives `(2B, D)`. Use cat (not stack) when you are appending more elements to a dimension that already exists.

## Worked solution

We have two `(3, 5)` mini-batches. `t.cat([a, b], dim=0)` stacks them end-to-end along the existing batch axis, producing `(6, 5)` — the rank stays 2 and the batch dimension doubles. Concatenating along `dim=1` instead would widen the feature axis to `(3, 10)`. We print both results' shapes to show cat extends whichever existing axis you name, without adding a new one. This is the complement of stack: same rank, longer axis.

```
t.manual_seed(0)
a = t.randn(3, 5)
b = t.randn(3, 5)
batch_cat = t.cat([a, b], dim=0)
feat_cat = t.cat([a, b], dim=1)
print('cat dim=0 shape:', tuple(batch_cat.shape))
print('cat dim=1 shape:', tuple(feat_cat.shape))
print('first 3 rows equal a:', t.equal(batch_cat[:3], a))
```
