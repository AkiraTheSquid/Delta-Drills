## Concept

`t.stack` inserts a new axis, raising the rank by one. Stacking N vectors of shape `(D,)` along `dim=0` yields `(N, D)`, with each vector as a row.

## Faded exercise 1

Three `(4,)` vectors are given. Complete the line that stacks them into a `(3, 4)` matrix with each vector as a row.

**Fill in:** stack the vectors along a new leading axis (dim=0)

````
vectors = [t.tensor([1.0, 2.0, 3.0, 4.0]),
           t.tensor([5.0, 6.0, 7.0, 8.0]),
           t.tensor([9.0, 10.0, 11.0, 12.0])]
rows = t.stack(vectors, dim=0)

def _test():
    vectors = [t.tensor([1.0, 2.0, 3.0, 4.0]),
               t.tensor([5.0, 6.0, 7.0, 8.0]),
               t.tensor([9.0, 10.0, 11.0, 12.0])]
    assert tuple(rows.shape) == (3, 4), 'stack must add a leading axis -> (3,4)'
    assert t.equal(rows[0], vectors[0]), 'row 0 must be the first vector'
    assert t.equal(rows[2], vectors[2]), 'row 2 must be the third vector'

try:
    _test()
    _dd_passed.add('faded1')
    print('[Delta Drills] faded1 passed.')
except AssertionError as _e:
    print('Test failed:', _e)
````
